```python
/**
 * What is Database Management System (DBMS):
 * > In our application, whenever we make applications, we need to store
 *   data. So, all this data, we store it under some system, especially
 *   in database system.
 * > Database is nothing but a disk, the hard disk or SSD in your
 *   computer, in that we store the data in a particular way. 
 * > So, the systems that work to store this data, called as database.
 *   i.e., location to store your data.
 * > But these databases have a system called as database management
 *   system, under which we have:
 *   - databases
 *   - database configuration, 
 *   - database management,
 *   - how to create database
 *   - how to modify database
 *   - how to configure its access
 *   All these things comes under database management.
*/
```

```python
/**
 * What does a DBMS do?
 * 1. Database Creation
 *    - CREATE DATABASE my_app;
 *
 * 2. Data Storage
 *    - SQL → Tables (rows + columns)
 *    - NoSQL → Documents/Key-Value/Graph
 *
 * 3. Data Modification
 *    - INSERT → Add data
 *    - UPDATE → Modify data
 *    - DELETE → Remove data
 *
 * 4. Access Control (Security)
 *    - GRANT / REVOKE permissions
 *
 * 5. Configuration
 *    - Memory settings, connections, caching
 *
 * 6. Concurrency Management
 *    - Multiple users can access data safely
*/
```

```python
/**
 * Why Databases?
 * 1. Data Organization
 *    - Structured storage (tables instead of messy files)
 *
 * 2. Fast Retrieval (Indexing)
 *    - Index = shortcut to find data quickly
 *
 * 3. Data Consistency
 *    - Ensures accurate and reliable data
 *
 * 4. Concurrency
 *    - Multiple users can read/write safely
 *
 * 5. Scalability
 *    - Can handle growth (100 → millions of users)
 *
 * 6. Security
 *    - Authentication + Authorization
*/
```

```python
/**
 * Types of Databases:
 * 1. Relational Databases (SQL)
 *    - Example: PostgreSQL, MySQL
 *    - Structure: Tables (rows + columns)
 *    - Strong schema, ACID compliant
 *
 * 2. Document Databases
 *    - Example: MongoDB
 *    - JSON-like flexible structure
 *
 * 3. Graph Databases
 *    - Example: Neo4j
 *    - Stores relationships (nodes + edges)
 *
 * 4. Key-Value Databases
 *    - Example: Redis
 *    - Simple key → value storage
 *    - Very fast
 *
 * 5. Vector Databases
 *    - Example: Pinecone, Weaviate
 *    - Used in AI (stores embeddings)
*/
```

```python
/**
 * NoSQL vs SQL Databases:
 * SQL:
 * - Fixed schema
 * - Strong consistency
 * - Supports joins
 *
 * NoSQL:
 * - Flexible schema
 * - Scales horizontally
 * - Faster writes
*/
```

```python
/**
 * Why NoSQL is faster for write-heavy workload?
 * 1. No strict schema → faster inserts
 * 2. No joins → less computation
 * 3. Horizontal scaling → distributes load
 * 4. Eventual consistency → faster than strict consistency
*/
```

```python
/**
 * Database Architecture:
 * > Database Architecure defines how a DBMS is internally structured and how different
 *   components work together to:
 *   - store data
 *   - process queries
 *   - manage transactions
 *   - ensure reliability
 * 
 *   +-----------------------------------+
 *   |              DBMS                 |
 *   +-----------------------------------+
 *   +-----------+---------+-------------+
 *   |   Query   | Storage | Transaction |
 *   | Processor | Manager |  Manager    |
 *   +-----------+---------+-------------+
 *   +----------------+------------------+
 *   | Buffer Manager | Recovery Manager |
 *   +----------------+------------------+
 *   +------------+-----------+----------+
 *   | Data Files | Idx Files | Log Files|
 *   +------------+-----------+----------+
 *   +-----------------------------------+
 *   |       Physical Storage (Disk)     |
 *   +-----------------------------------+
 * 
 * > When you run: SELECT * FROM users:
 *   1. Query Processor understands SQL
 *   2. Transaction Manager ensures safety
 *   3. Buffer Manager checks RAM
 *   4. Storage Manager reads disk if needed
 *   5. Data is returned
*/

/**
 * Explanation:
 * 1. Query Processor:
 *    > Role: Understand and execute SQL queries
 *    > It has 3 main parts:
 *      a. Parser:
 *         - Checks syntax (is SQL valid?)
 *         - Converts SQL -> internal format
 *      b. Optimizer
 *         - Chooses best way to run query
 *         - Ex: Use index? or full table scan?
 *      c. Executor
 *         - Actually runs the query
 *    > Example:
 *      SELECT * FROM users WHERE id = 1;
 *      Optimizer uses index instead of scanning full table
 *
 *  2. Storage Manager:
 *     > Role: Handles how data is stored in disk
 *     > Responsibilites:
 *       - Read/Write data from disk
 *       - Manage file structure
 *       - Organize tables and indexes
 *     > Think:
 *       - Query Processor = brain
 *       - Storage Manager = hands (does actual work)
 *
 *  3. Transaction Manager:
 *     > Role: Ensures safe execution of operations
 *     > A Transaction = group of operations treated as ONE unit
 *     > Example: 
 *       Transfer ₹100:
 *       - Deduct from A
 *       - Add to B
 *       If something fails -> rollback everything 
 *     > Handles ACID Properties:
 *       a. Atomicity  : All or nothing
 *       b. Consistency: Data remains valid
 *       c. Isolation  : Transactions don't interfere
 *       d. Durability : Changes are permanent
 *
 *  4. Buffer Manager:
 *     > Role: Manages Memory (RAM)
 *     > Why needed?
 *       - Disk is slow
 *       - RAM is fast
 *     > What it does?
 *       - Stores frequently used data in memory
 *       - Reduces disk access
 *     > Example:
 *       - First query  -> disk read
 *       - Second query -> RAM (faster)
 *
 *  5. Recovery Manager
 *     > Role: Handles crashes and failures
 *     > Uses: Log files
 *     > Example:
 *       - System crashes during transaction
 *       - Recovery Manager:
 *           → Undo incomplete changes
 *           → Redo committed changes
 *     > Ensures:
 *       - No data loss
 *       - Database Consistency
 *
 *  6. Files in Database:
 *     a. Data Files: Store actual table data
 *     b. Index files: Store pointers for fast lookup
 *     c. Log Files  : Record all changes (important for recovery)
 * 
 *  7. Physical Storage:
 *     > Actual Hardware:
 *       - HDD
 *       - SSD
 *     > Data is stored in: Pages (usually 4KB or 8KB)
 *     > Important:
 *       - DB never reads 1 row at a time
 *       - It reads entire page (block)
*/
```

```python
/**
 * Let's Setup Postgres Database in Docker:
 * 
 * 1. Run Postgres Container
 *    docker run -d --name postgres-container \
 *      -e POSTGRES_USER=admin \
 *      -e POSTGRES_PASSWORD=secretpassword \
 *      -e POSTGRES_DB=mydatabase \
 *      -p 5432:5432 \
 *      -v $(pwd)/postgres_data:/var/lib/postgresql/data \
 *      postgres:latest
 * 
 *    Explanation:
 *    > -v stands for "volume"
 *    > It is used to:
 *      Mount (connect) a folder from your host machine to a folder inside the docker container
 *    > -v $(pwd)/postgres_data:/var/lib/postgresql/data
 *      a. -v <host_path>:<container_path>
 *      b. $(pwd)/postgres_data
 *         - This is the host machine path (your computer)
 *         - $(pwd) : "print working directory" gives current folder path
 *         - Example: If you are in: /home/mohammad/project then:
 *           $(pwd)/postgres_data = /home/mohammad/project/postgres_data
 *      c. /var/lib/postgresql/data
 *         - This is the CONTAINER path
 *         - This is where PostgreSQL stores:
 *           - databases
 *           - tables
 *           - indexes
 *           - logs
 *         - Think of it as: "Postgres internal storage folder"
 *      d. Any data written by Postgres gets stored in your local folder.
 *    
 *    Note: Instead of -v $(pwd)/postgres_data:/var/lib/postgresql/data, you can use:
 *         -v postgres_data:/var/lib/postgresql/data
 * 
 *          ./postgres_data   <----->   /var/lib/postgresql/data
 *           ( Host Machine)               (Docker Container)
 * 
 * 
 * 2. Enter inside Postgres Container:
 *    docker exec -it postgres-container psql -U admin -d mydatabase
 *
 *    Explanation:
 *    a. docker exec -it
 *       - Executes command inside running container
 *       - -i → interactive mode
 *       - -t → terminal (TTY)
 *    b. postgres-container: Name of running container
 *    c. psql              : PostgreSQL CLI tool
 *    d. -U admin          : Username
 *    e. -d mydatabase     : Database name
 *
 *
 * 3. Run Basic Queries:
 *    a. Check PostgreSQL version:
 *       > SELECT version();
 * 
 *    b. Show all tables:
 *       > \dt
 * 
 *    c. Create table:
 *       CREATE TABLE users (
 *         id SERIAL PRIMARY KEY,
 *         name VARCHAR(100),
 *         email VARCHAR(100)
 *       );
 *
 *    d. Insert data:
 *       INSERT INTO users (name, email)
 *       VALUES ('Mohammad', 'mohammad@example.com');
 *   
 *    e. Read data:
 *       SELECT * FROM users;
 *   
 *    f. Update data:
 *       UPDATE users
 *       SET name = 'Mohd'
 *       WHERE id = 1;
 *   
 *    g. Delete data:
 *       DELETE FROM users WHERE id = 1;
 *
 *
 * 4. EXIT FROM PSQL: 
 *    > \q
 *
 *
 * FINAL FLOW
 * > You / Application
 *        ↓
 * > docker exec + psql
 *        ↓
 * > PostgreSQL DB (inside container)
 *        ↓
 * > Data stored in volume (host machine)
 */
```
