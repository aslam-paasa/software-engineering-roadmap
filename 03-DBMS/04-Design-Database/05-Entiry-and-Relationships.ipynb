```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 1. WHAT IS AN ENTITY? — Real-world objects you store data about
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ AN ENTITY is a real-world object or concept about which you        │
 * │ store data. It represents something that exists independently      │
 * │ in your system.                                                    │
 * │                                                                    │
 * │ HOW TO IDENTIFY AN ENTITY:                                         │
 * │   Ask yourself: "I want to store information about this thing."    │
 * │   If you can answer with a noun, it's likely an entity.            │
 * │                                                                    │
 * │ EXAMPLES:                                                          │
 * │                                                                    │
 * │   ┌─────────────────────────────────────────────────────────┐      │
 * │   │ Entity        │ Why it's an entity                      │      │
 * │   ├───────────────┼─────────────────────────────────────────┤      │
 * │   │ Student       │ A person who learns                     │      │
 * │   │ Course        │ A learning offering                     │      │
 * │   │ Payment       │ A financial transaction                 │      │
 * │   │ Instructor    │ A person who teaches                    │      │
 * │   │ Affiliate     │ A person who refers customers           │      │
 * │   │ Enrollment    │ A student joining a course              │      │
 * │   └───────────────┴─────────────────────────────────────────┘      │
 * │                                                                    │
 * │ KEY PROPERTY:                                                      │
 * │   Entities exist INDEPENDENTLY in the real world.                  │
 * │   A student exists whether they pay or not.                        │
 * │   A course exists whether someone enrolls or not.                  │
 * │   A payment exists because an enrollment exists.                   │
 * │                                                                    │
 * │ IN DATABASE DESIGN:                                                │
 * │   Each entity typically becomes its OWN TABLE.                     │
 * │   This is the foundation of relational modeling.                   │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 2. WHAT IS AN ATTRIBUTE? — Properties that describe entities
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ AN ATTRIBUTE is a property or characteristic of an entity.         │
 * │                                                                    │
 * │ Attributes describe the entity. They do NOT represent separate     │
 * │ real-world objects on their own.                                   │
 * │                                                                    │
 * │ EXAMPLES:                                                          │
 * │                                                                    │
 * │   ┌─────────────────────────────────────────────────────────┐      │
 * │   │ Entity      │ Attributes                                │      │
 * │   ├─────────────┼───────────────────────────────────────────┤      │
 * │   │ Student     │ student_id, name, email, phone,           │      │
 * │   │             │ date_of_birth, created_at                 │      │
 * │   ├─────────────┼───────────────────────────────────────────┤      │
 * │   │ Course      │ course_id, course_code, title,            │      │
 * │   │             │ price, currency, instructor_id, is_active │      │
 * │   ├─────────────┼───────────────────────────────────────────┤      │
 * │   │ Payment     │ payment_id, amount, currency, status,     │      │
 * │   │             │ paid_at, created_at                       │      │
 * │   ├─────────────┼───────────────────────────────────────────┤      │
 * │   │ Enrollment  │ enrollment_id, student_id, course_id,     │      │
 * │   │             │ enrolled_at, status                       │      │
 * │   └─────────────┴───────────────────────────────────────────┘      │
 * │                                                                    │
 * │ KEY PROPERTY:                                                      │
 * │   Attributes are properties OF the entity.                         │
 * │   They don't exist on their own.                                   │
 * │   "email" is not a thing — it's a property of a student.           │
 * │   "price" is not a thing — it's a property of a course.            │
 * │                                                                    │
 * │ IN DATABASE DESIGN:                                                │
 * │   Attributes become COLUMNS in the entity's table.                 │
 * │   Each column stores one attribute for each row.                   │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 3. WHAT IS A TABLE? — Collection of entity instances
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌──────────────────────────────────────────────────────────────────────┐
 * │ A TABLE is a structured collection of rows representing              │
 * │ instances of an entity.                                              │
 * │                                                                      │
 * │ If Student is an entity:                                             │
 * │   → The students table stores all student records                    │
 * │   → Each row = one student                                           │
 * │   → Each column = one attribute of student                           │
 * │                                                                      │
 * │ VISUAL REPRESENTATION:                                               │
 * │                                                                      │
 * │   ┌─────────────────────────────────────────────────────────┐        │
 * │   │                    students TABLE                       │        │
 * │   ├───────────┬───────────────────┬─────────────────────────┤        │
 * │   │ student_id│ email             │ full_name               │        │
 * │   ├───────────┼───────────────────┼─────────────────────────┤        │
 * │   │ 1         │ asha@gmail.com    │ Asha Sharma             │  ← Row │
 * │   │ 2         │ ravi@gmail.com    │ Ravi Kumar              │  ← Row │
 * │   │ 3         │ neha@gmail.com    │ Neha Singh              │  ← Row │
 * │   └───────────┴───────────────────┴─────────────────────────┘        │
 * │         ↑                           ↑                                │
 * │      Column                      Column                              │
 * │   (attribute)                  (attribute)                           │
 * │                                                                      │
 * │ RELATIONSHIP TO ENTITY:                                              │
 * │   → Table name = entity name (plural usually)                        │
 * │   → Table structure = entity definition                              │
 * │   → Table content = instances of that entity                         │
 * └──────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 4. WHAT IS A ROW? — One instance of an entity
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ A ROW represents ONE instance of an entity.                        │
 * │                                                                    │
 * │ If the entity is Student:                                          │
 * │   → One row = ONE student                                          │
 * │   → Not multiple students                                          │
 * │   → Not part of a student                                          │
 * │   → Exactly ONE student                                            │
 * │                                                                    │
 * │ THIS ALIGNS WITH THE MENTAL MODEL:                                 │
 * │   "One row should represent ONE FACT."                             │
 * │                                                                    │
 * │ CORRECT:                                                           │
 * │   ┌─────────────────────────────────────────────────────────┐      │
 * │   │ students table                                          │      │
 * │   ├───────────┬───────────────────┬─────────────────────────┤      │
 * │   │ student_id│ email             │ full_name               │      │
 * │   ├───────────┼───────────────────┼─────────────────────────┤      │
 * │   │ 1         │ asha@gmail.com    │ Asha Sharma             │      │
 * │   │ 2         │ ravi@gmail.com    │ Ravi Kumar              │      │
 * │   └───────────┴───────────────────┴─────────────────────────┘      │
 * │   Each row = one student                                           │
 * │                                                                    │
 * │ INCORRECT (spreadsheet thinking):                                  │
 * │   ┌─────────────────────────────────────────────────────────┐      │
 * │   │ raw_learning_events table                               │      │
 * │   ├───────────┬───────────────────┬─────────────────────────┤      │
 * │   │ student   │ course            │ instructor              │      │
 * │   ├───────────┼───────────────────┼─────────────────────────┤      │
 * │   │ Asha      │ SQL Basics        │ Raj                     │      │
 * │   │ Asha      │ SQL Basics        │ Raj                     │      │
 * │   └───────────┴───────────────────┴─────────────────────────┘      │
 * │   One row = mix of student + course + instructor                   │
 * │   Multiple rows for same student                                   │
 * │   Violates "one fact per row"                                      │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 5. ENTITIES IN A COACHING CENTER — Practical example
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ SCENARIO: Designing a database for a coaching center.              │
 * │                                                                    │
 * │ STEP 1: Identify the entities                                      │
 * │                                                                    │
 * │   ┌─────────────────────────────────────────────────────────┐      │
 * │   │ Entity          │ Why it's needed                       │      │
 * │   ├─────────────────┼───────────────────────────────────────┤      │
 * │   │ Student         │ Who is learning?                      │      │
 * │   │ Course          │ What are they learning?               │      │
 * │   │ Instructor      │ Who teaches?                          │      │
 * │   │ Enrollment      │ Which student took which course?      │      │
 * │   │ Payment         │ How did they pay?                     │      │
 * │   └─────────────────┴───────────────────────────────────────┘      │
 * │                                                                    │
 * │ STEP 2: Define attributes for each entity                          │
 * │                                                                    │
 * │   Student:                                                         │
 * │     → student_id (unique identifier)                               │
 * │     → email (contact)                                              │
 * │     → full_name (identity)                                         │
 * │     → phone (contact)                                              │
 * │     → date_of_birth (demographic)                                  │
 * │     → created_at (audit)                                           │
 * │                                                                    │
 * │   Course:                                                          │
 * │     → course_id (unique identifier)                                │
 * │     → course_code (business identifier)                            │
 * │     → title (what it's called)                                     │
 * │     → description (what it covers)                                 │
 * │     → instructor_id (who teaches it)                               │
 * │     → price (how much it costs)                                    │
 * │     → is_active (still offered)                                    │
 * │                                                                    │
 * │   Enrollment:                                                      │
 * │     → enrollment_id (unique identifier)                            │
 * │     → student_id (which student)                                   │
 * │     → course_id (which course)                                     │
 * │     → enrolled_at (when they joined)                               │
 * │     → status (active/completed/dropped)                            │
 * │                                                                    │
 * │   Payment:                                                         │
 * │     → payment_id (unique identifier)                               │
 * │     → enrollment_id (which enrollment)                             │
 * │     → amount (how much)                                            │
 * │     → status (paid/pending/refunded)                               │
 * │     → paid_at (when completed)                                     │
 * │                                                                    │
 * │ STEP 3: Create tables from entities                                │
 * │                                                                    │
 * │   CREATE TABLE students (...);                                     │
 * │   CREATE TABLE courses (...);                                      │
 * │   CREATE TABLE enrollments (...);                                  │
 * │   CREATE TABLE payments (...);                                     │
 * │                                                                    │
 * │ KEY PRINCIPLE:                                                     │
 * │   Each entity gets its own table.                                  │
 * │   Do NOT mix student attributes, course attributes, and            │
 * │   payment attributes into one large flat table.                    │
 * │                                                                    │
 * │ This violates entity boundaries.                                   │
 * └────────────────────────────────────────────────────────────────────┘
 */
```
