```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * SAMPLE TABLES — For illustrating relationships
 * ─────────────────────────────────────────────────────────────────────
 *
 * students (student_id, email, full_name)
 * instructors (instructor_id, email, full_name)
 * courses (course_id, course_code, title, instructor_id)
 * student_profiles (profile_id, student_id, country, graduation_year)
 * enrollments (enrollment_id, student_id, course_id, enrolled_at)
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 1. RELATIONSHIP — How entities connect via keys
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ A RELATIONSHIP defines how one entity in your system is            │
 * │ connected to another. It exists when:                              │
 * │                                                                    │
 * │   → A PRIMARY KEY (PK) uniquely identifies a row in one table      │
 * │   → A FOREIGN KEY (FK) stores that primary key value in            │
 * │     another table                                                  │
 * │                                                                    │
 * │   That FK creates the CONNECTION.                                  │
 * │                                                                    │
 * │ EXAMPLE 1: Course → Instructor                                     │
 * │                                                                    │
 * │   instructors table:                                               │
 * │   ┌───────────────┬─────────────────┐                              │
 * │   │ instructor_id │ full_name       │                              │
 * │   │ (PRIMARY KEY) │                 │                              │
 * │   ├───────────────┼─────────────────┤                              │
 * │   │ 1             │ Raj             │                              │
 * │   │ 2             │ Sara            │                              │
 * │   └───────────────┴─────────────────┘                              │
 * │                         ↑                                          │
 * │                         │                                          │
 * │   courses table:        │ (FOREIGN KEY)                            │
 * │   ┌───────────┬─────────────┬────────────────┐                     │
 * │   │ course_id │ title       │ instructor_id │                      │  
 * │   ├───────────┼─────────────┼────────────────┤                     │
 * │   │ 1         │ SQL Basics  │ 1              │ ← refers to Raj     │
 * │   │ 2         │ DSA Mastery │ 1              │ ← refers to Raj     │
 * │   │ 3         │ LLD Design  │ 2              │ ← refers to Sara    │
 * │   └───────────┴─────────────┴────────────────┘                     │
 * │                                                                    │
 * │   instructor_id is a FOREIGN KEY.                                  │
 * │   It stores the primary key from the instructors table.            │
 * │   This creates the relationship: "This course belongs to           │
 * │   this instructor."                                                │
 * │                                                                    │
 * │ EXAMPLE 2: Student Profile → Student                               │
 * │                                                                    │
 * │   students table:                                                  │
 * │   ┌────────────┬─────────────────┬──────────────┐                  │
 * │   │ student_id │ email           │ full_name    │                  │
 * │   │ (PK)       │                 │              │                  │
 * │   ├────────────┼─────────────────┼──────────────┤                  │
 * │   │ 1          │ asha@gmail.com  │ Asha         │                  │
 * │   │ 2          │ ravi@gmail.com  │ Ravi         │                  │
 * │   └────────────┴─────────────────┴──────────────┘                  │
 * │                         ↑                                          │
 * │                         │ (FOREIGN KEY)                            │
 * │   student_profiles:     │                                          │
 * │   ┌────────────┬────────────┬─────────────────┐                    │
 * │   │ profile_id │ student_id │ country         │                    │
 * │   ├────────────┼────────────┼─────────────────┤                    │
 * │   │ 1          │ 1          │ India           │ ← Asha's profile   │
 * │   │ 2          │ 2          │ USA             │ ← Ravi's profile   │
 * │   └────────────┴────────────┴─────────────────┘                    │
 * │                                                                    │
 * │   This means: "This profile belongs to that student."              │
 * │                                                                    │
 * │ IMPORTANT CLARIFICATION:                                           │
 * │   → A PRIMARY KEY uniquely identifies a row                        │
 * │   → There can be other unique columns (candidate keys)             │ 
 * │   → But there is only ONE primary key per table                    │
 * │   → That primary key is what other tables should reference         │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 2. CARDINALITY — One-to-One, One-to-Many, Many-to-Many
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ CARDINALITY answers one question:                                  │
 * │   "For one row on the left, how many rows can exist on the right?" │
 * │                                                                    │
 * │ THREE TYPES:                                                       │
 * │                                                                    │
 * │ 1. ONE-TO-ONE (1:1)                                                │
 * │    ┌──────────────────────────────────────────────────────────┐    │
 * │    │ One A maps to at most ONE B.                             │    │
 * │    │ One student → ONE profile (or zero)                      │    │
 * │    │                                                          │    │
 * │    │ students ────────── student_profiles                     │    │
 * │    │   1                  1 (at most)                         │    │
 * │    │                                                          │    │
 * │    │ ENFORCEMENT:                                             │    │
 * │    │   The foreign key must also be UNIQUE.                   │    │
 * │    │                                                          │    │
 * │    │   CREATE TABLE student_profiles (                        │    │
 * │    │     student_id INT UNIQUE,  ← this enforces 1:1          │    │
 * │    │     FOREIGN KEY (student_id) REFERENCES students(id)     │    │
 * │    │   );                                                     │    │
 * │    │                                                          │    │
 * │    │ Without UNIQUE, it becomes ONE-TO-MANY.                  │    │
 * │    └──────────────────────────────────────────────────────────┘    │
 * │                                                                    │
 * │ 2. ONE-TO-MANY (1:N)                                               │
 * │    ┌─────────────────────────────────────────────────────────┐     │
 * │    │ One A maps to MANY B. Each B maps back to exactly one A.│     │
 * │    │                                                         │     │
 * │    │ Instructor ────────── Courses                           │     │
 * │    │     1                      N                            │     │
 * │    │                                                         │     │
 * │    │ One instructor can teach MANY courses.                  │     │
 * │    │ Each course has ONE instructor.                         │     │
 * │    │                                                         │     │
 * │    │ RULE OF THUMB:                                          │     │
 * │    │   The foreign key goes on the side that can REPEAT.     │     │
 * │    │   Since many courses can have the same instructor,      │     │
 * │    │   instructor_id is stored in the courses table.         │     │
 * │    │                                                         │     │
 * │    │   courses                                               │     │
 * │    │   ┌───────────┬─────────────┬────────────────┐          │     │
 * │    │   │ course_id │ title       │ instructor_id │           │     │
 * │    │   ├───────────┼─────────────┼────────────────┤          │     │
 * │    │   │ 1         │ SQL Basics  │ 1              │          │     │
 * │    │   │ 2         │ DSA Mastery │ 1              │ ← repeats│     │
 * │    │   │ 3         │ LLD Design  │ 2              │          │     │
 * │    │   └───────────┴─────────────┴────────────────┘          │     │
 * │    └─────────────────────────────────────────────────────────┘     │
 * │                                                                    │
 * │ 3. MANY-TO-MANY (N:M)                                              │
 * │    ┌─────────────────────────────────────────────────────────┐     │
 * │    │ Many A relate to MANY B.                                │     │
 * │    │                                                         │     │
 * │    │ Students ────────── Courses                             │     │
 * │    │     N                       M                           │     │
 * │    │                                                         │     │
 * │    │ One student can enroll in MANY courses.                 │     │
 * │    │ One course can have MANY students.                      │     │
 * │    │                                                         │     │
 * │    │ WRONG APPROACH (common beginner mistake):               │     │
 * │    │                                                         │     │
 * │    │   students                                              │     │
 * │    │   ┌────────────┬─────────────────────────┐              │     │
 * │    │   │ student_id │ courses_enrolled        │              │     │
 * │    │   ├────────────┼─────────────────────────┤              │     │
 * │    │   │ 1          │ "1,3,5"                 │ ← LIST!      │     │
 * │    │   └────────────┴─────────────────────────┘              │     │
 * │    │                                                         │     │
 * │    │   Storing a list:                                       │     │
 * │    │     → Breaks First Normal Form                          │     │
 * │    │     → Breaks indexing                                   │     │
 * │    │     → Breaks filtering                                  │     │
 * │    │     → Breaks joins                                      │     │
 * │    │     → Violates single source of truth                   │     │
 * │    │     → Causes update chaos                               │     │
 * │    │                                                         │     │
 * │    │ CORRECT APPROACH: Bridge (Junction) Table               │     │
 * │    │                                                         │     │
 * │    │   enrollments (bridge table)                            │     │
 * │    │   ┌────────────┬───────────┬─────────────────────┐      │     │
 * │    │   │ enrollment │ student_id│ course_id           │      │     │
 * │    │   ├────────────┼───────────┼─────────────────────┤      │     │
 * │    │   │ 1          │ 1         │ 1                   │      │     │
 * │    │   │ 2          │ 1         │ 3                   │      │     │
 * │    │   │ 3          │ 2         │ 1                   │      │     │
 * │    │   └────────────┴───────────┴─────────────────────┘      │     │
 * │    │                                                         │     │
 * │    │   Each row = ONE FACT: "This student enrolled in        │     │
 * │    │   this course."                                         │     │
 * │    │                                                         │     │
 * │    │   Now N:M becomes two 1:N relationships:                │     │
 * │    │     → students → enrollments (1:N)                      │     │
 * │    │     → courses → enrollments (1:N)                       │     │
 * │    │                                                         │     │
 * │    │ PRODUCTION TIP:                                         │     │
 * │    │   N:M relationships are ALWAYS modeled using a third    │     │
 * │    │   table. Never compromise on this. One extra table never│     │
 * │    │   harms you. Wrong data modeling will.                  │     │
 * │    └─────────────────────────────────────────────────────────┘     │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 3. OPTIONALITY — Mandatory vs. Optional relationships
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ OPTIONALITY answers: Is the relationship MANDATORY or can it be    │
 * │ missing?                                                           │
 * │                                                                    │
 * │ This is controlled by NULL vs NOT NULL on the foreign key.         │
 * │                                                                    │
 * │ MANDATORY RELATIONSHIP (NOT NULL):                                 │
 * │                                                                    │
 * │   Example: Every course must have an instructor.                   │
 * │                                                                    │
 * │   CREATE TABLE courses (                                           │
 * │     ...                                                            │
 * │     instructor_id INT NOT NULL,  ← MANDATORY!                      │
 * │     FOREIGN KEY (instructor_id) REFERENCES instructors(id)         │
 * │   );                                                               │
 * │                                                                    │
 * │   This means:                                                      │
 * │     → A course cannot exist without an instructor                  │
 * │     → The relationship is REQUIRED                                 │
 * │     → Attempting to insert a course with NULL instructor_id        │ 
 * │       will FAIL                                                    │
 * │                                                                    │
 * │ OPTIONAL RELATIONSHIP (NULL allowed):                              │
 * │                                                                    │
 * │   Example: A student may or may not have a profile.                │
 * │                                                                    │
 * │   CREATE TABLE student_profiles (                                  │
 * │     ...                                                            │
 * │     student_id INT,  ← NULL allowed (optional)                     │
 * │     FOREIGN KEY (student_id) REFERENCES students(id)               │
 * │   );                                                               │
 * │                                                                    │
 * │   This means:                                                      │
 * │     → Some students have profiles                                  │
 * │     → Some students do NOT have profiles                           │
 * │     → NULL indicates "no profile exists"                           │
 * │                                                                    │
 * │ VISUAL COMPARISON:                                                 │
 * │                                                                    │
 * │   MANDATORY (NOT NULL):                                            │
 * │   ┌──────────────────────────────────────────────────────────┐     │
 * │   │ courses table                                            │     │
 * │   ├───────────┬─────────────┬────────────────────────────────┤     │
 * │   │ course_id │ title       │ instructor_id                  │     │
 * │   ├───────────┼─────────────┼────────────────────────────────┤     │
 * │   │ 1         │ SQL Basics  │ 1                              │     │
 * │   │ 2         │ DSA Mastery │ 1                              │     │
 * │   │ 3         │ NULL        │ ← NOT ALLOWED! Error!          │     │
 * │   └───────────┴─────────────┴────────────────────────────────┘     │
 * │                                                                    │
 * │   OPTIONAL (NULL allowed):                                         │
 * │   ┌─────────────────────────────────────────────────────────┐      │
 * │   │ student_profiles table                                  │      │
 * │   ├────────────┬────────────┬─────────────────┐             │      │
 * │   │ profile_id │ student_id │ country         │             │      │
 * │   ├────────────┼────────────┼─────────────────┤             │      │
 * │   │ 1          │ 1          │ India           │             │      │
 * │   │ 2          │ NULL       │ USA             │ ← ALLOWED   │      │
 * │   └────────────┴────────────┴─────────────────┘             │      │
 * │   (NULL means this profile is not linked to any student)    │      │
 * │                                                                    │
 * │ PRODUCTION RULE:                                                   │
 * │   → Use NOT NULL when the relationship is BUSINESS REQUIRED        │
 * │   → Allow NULL when the relationship is OPTIONAL                   │
 * │   → The database enforces this at the storage layer                │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 4. PUTTING IT ALL TOGETHER — Complete schema example
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌───────────────────────────────────────────────────────────────────────┐
 * │ ENTITIES AND THEIR RELATIONSHIPS:                                     │
 * │                                                                       │
 * │   ┌─────────────┐                                                     │
 * │   │ instructors │                                                     │
 * │   │ (1)         │                                                     │
 * │   └──────┬──────┘                                                     │
 * │          │ 1:N (mandatory)                                            │
 * │          ▼                                                            │
 * │   ┌─────────────┐     N:M      ┌─────────────┐                        │
 * │   │  courses    │◄────────────►│  students   │                        │
 * │   │ (N)         │              │ (N)         │                        │
 * │   └──────┬──────┘              └─────────────┘                        │
 * │          │                         │                                  │
 * │          │                         │ 1:1 (optional)                   │
 * │          │                         ▼                                  │
 * │          │                  ┌─────────────┐                           │
 * │          │                  │  profiles   │                           │
 * │          │                  └─────────────┘                           │
 * │          │                                                            │
 * │          │                    ┌─────────────┐                         │
 * │          └───────────────────►│ enrollments │                         │
 * │                               │ (bridge)    │                         │
 * │                               └─────────────┘                         │
 * │                                                                       │
 * │ COMPLETE DDL:                                                         │
 * │                                                                       │
 * │   CREATE TABLE instructors (                                          │
 * │     instructor_id INT AUTO_INCREMENT PRIMARY KEY,                     │
 * │     email VARCHAR(255) NOT NULL UNIQUE,                               │
 * │     full_name VARCHAR(100) NOT NULL                                   │
 * │   );                                                                  │
 * │                                                                       │
 * │   CREATE TABLE students (                                             │
 * │     student_id INT AUTO_INCREMENT PRIMARY KEY,                        │
 * │     email VARCHAR(255) NOT NULL UNIQUE,                               │
 * │     full_name VARCHAR(100) NOT NULL                                   │
 * │   );                                                                  │
 * │                                                                       │
 * │   CREATE TABLE student_profiles (                                     │
 * │     profile_id INT AUTO_INCREMENT PRIMARY KEY,                        │
 * │     student_id INT UNIQUE,  -- 1:1 optional                           │    
 * │     country VARCHAR(100),                                             │
 * │     graduation_year INT,                                              │
 * │     FOREIGN KEY (student_id) REFERENCES students(student_id)          │
 * │   );                                                                  │
 * │                                                                       │
 * │   CREATE TABLE courses (                                              │
 * │     course_id INT AUTO_INCREMENT PRIMARY KEY,                         │
 * │     course_code VARCHAR(30) NOT NULL UNIQUE,                          │
 * │     title VARCHAR(200) NOT NULL,                                      │
 * │     instructor_id INT NOT NULL,  -- 1:N mandatory                     │
 * │     price DECIMAL(10,2) NOT NULL CHECK (price >= 0),                  │
 * │     FOREIGN KEY (instructor_id) REFERENCES instructors(instructor_id) │
 * │   );                                                                  │
 * │                                                                       │
 * │   CREATE TABLE enrollments (                                          │
 * │     enrollment_id INT AUTO_INCREMENT PRIMARY KEY,                     │
 * │     student_id INT NOT NULL,                                          │
 * │     course_id INT NOT NULL,                                           │
 * │     enrolled_at DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,          │
 * │     FOREIGN KEY (student_id) REFERENCES students(student_id),         │
 * │     FOREIGN KEY (course_id) REFERENCES courses(course_id),            │
 * │     UNIQUE(student_id, course_id)  -- prevents duplicates             │
 * │   );                                                                  │
 * └───────────────────────────────────────────────────────────────────────┘
 */
```
