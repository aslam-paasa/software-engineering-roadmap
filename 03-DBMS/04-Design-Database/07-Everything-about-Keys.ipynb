```python
/**
 * UNDERSTANDING PRIMARY, FOREIGN, COMPOSITE, SURROGATE, AND NATURAL KEYS
 * ─────────────────────────────────────────────────────────────────────
 *
 * TABLE OF CONTENTS:
 * 1. PRIMARY KEY ....................... Uniquely identifies each row
 * 2. CANDIDATE KEY ..................... All columns that could be primary key
 * 3. UNIQUE CONSTRAINT ................. Enforces uniqueness, not identity
 * 4. FOREIGN KEY ....................... Enforces valid relationships
 * 5. COMPOSITE KEY ..................... Multiple columns as primary key
 * 6. SURROGATE vs NATURAL KEY .......... Generated vs. business-meaningful
 * 7. WHY INDUSTRY PREFERS SURROGATE .... Stability, performance, isolation
 * ─────────────────────────────────────────────────────────────────────
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * SAMPLE TABLES — For illustrating key concepts
 * ─────────────────────────────────────────────────────────────────────
 *
 * students (student_id, email, name)
 * courses (course_id, course_code, title)
 * enrollments (student_id, course_id, enrolled_at)
 * payments (payment_id, student_id, amount)
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 1. PRIMARY KEY — Uniquely identifies each row
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ A PRIMARY KEY uniquely identifies each row in a table.             │
 * │                                                                    │
 * │ RULES FOR PRIMARY KEY:                                             │
 * │   → Must be UNIQUE (no duplicate values)                           │
 * │   → Must be NOT NULL (every row must have a value)                 │
 * │   → Only ONE primary key per table                                 │
 * │                                                                    │
 * │ EXAMPLE:                                                           │
 * │                                                                    │
 * │   CREATE TABLE students (                                          │
 * │     student_id INT AUTO_INCREMENT PRIMARY KEY,  ← PK               │
 * │     email VARCHAR(255) NOT NULL,                                   │
 * │     full_name VARCHAR(100) NOT NULL                                │
 * │   );                                                               │
 * │                                                                    │
 * │   ┌────────────┬─────────────────┬──────────────┐                  │
 * │   │ student_id │ email           │ full_name    │                  │
 * │   │ (PK)       │                 │              │                  │
 * │   ├────────────┼─────────────────┼──────────────┤                  │
 * │   │ 1          │ asha@gmail.com  │ Asha         │                  │
 * │   │ 2          │ ravi@gmail.com  │ Ravi         │                  │
 * │   │ 3          │ neha@gmail.com  │ Neha         │                  │
 * │   └────────────┴─────────────────┴──────────────┘                  │
 * │                                                                    │
 * │ WHY PRIMARY KEY MATTERS:                                           │
 * │   → Uniquely identifies every row                                  │
 * │   → Enables reliable UPDATE and DELETE operations                  │
 * │   → Foundation for relationships (foreign keys)                    │
 * │   → Used in JOIN operations                                        │
 * │   → Automatically creates an index for fast lookups                │
 * │                                                                    │
 * │ Even if two students share the same name, their student_id         │
 * │ will be different. That's the guarantee of primary keys.           │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 2. CANDIDATE KEY — All columns that could be primary key
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ A CANDIDATE KEY is any column (or combination of columns) that     │
 * │ can uniquely identify a row.                                       │
 * │                                                                    │
 * │ From all candidate keys, ONE is chosen as the PRIMARY KEY.         │
 * │ The remaining candidate keys become ALTERNATE KEYS.                │
 * │                                                                    │
 * │ EXAMPLE:                                                           │
 * │                                                                    │
 * │   students table:                                                  │
 * │   ┌────────────────────────────────────────────────────────┐       │
 * │   │ Column         │ Can it be a candidate key?            │       │
 * │   ├────────────────┼───────────────────────────────────────┤       │
 * │   │ student_id     │ YES (unique, never changes)           │       │
 * │   │ email          │ YES (unique in business domain)       │       │
 * │   │ full_name      │ NO (duplicates possible)              │       │
 * │   │ phone          │ YES (if unique)                       │       │
 * │   └────────────────┴───────────────────────────────────────┘       │
 * │                                                                    │
 * │ Candidate keys: student_id, email, phone (if present)              │
 * │                                                                    │
 * │ CHOOSING THE PRIMARY KEY:                                          │
 * │   → Usually choose the most stable candidate                       │
 * │   → Usually choose the smallest (performance)                      │
 * │   → Usually choose a surrogate key (student_id) over natural key   │
 * │                                                                    │
 * │   PRIMARY KEY = student_id                                         │
 * │   ALTERNATE KEYS = email (with UNIQUE constraint)                  │
 * │                                                                    │
 * │   CREATE TABLE students (                                          │
 * │     student_id INT AUTO_INCREMENT PRIMARY KEY,  ← chosen           │
 * │     email VARCHAR(255) NOT NULL UNIQUE,         ← alternate        │
 * │     full_name VARCHAR(100) NOT NULL                                │
 * │   );                                                               │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 3. UNIQUE CONSTRAINT — Enforces uniqueness, not identity
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ A UNIQUE CONSTRAINT ensures that no duplicate values exist in      │
 * │ a column. Unlike PRIMARY KEY:                                      │
 * │   → A table can have MULTIPLE UNIQUE constraints                   │
 * │   → UNIQUE columns CAN be NULL (depends on database)               │
 * │                                                                    │
 * │ EXAMPLE:                                                           │
 * │                                                                    │
 * │   ALTER TABLE students                                             │
 * │   ADD CONSTRAINT unique_email UNIQUE (email);                      │
 * │                                                                    │
 * │   ┌────────────┬─────────────────┬──────────────┐                  │
 * │   │ student_id │ email           │ full_name    │                  │
 * │   ├────────────┼─────────────────┼──────────────┤                  │
 * │   │ 1          │ asha@gmail.com  │ Asha         │                  │
 * │   │ 2          │ ravi@gmail.com  │ Ravi         │                  │
 * │   │ 3          │ asha@gmail.com  │ ← REJECTED!  │                  │
 * │   └────────────┴─────────────────┴──────────────┘                  │
 * │                                                                    │
 * │ WHY UNIQUE CONSTRAINTS MATTER:                                     │
 * │   → Enforce business rules (no duplicate emails)                   │
 * │   → Prevent data corruption at the database level                  │
 * │   → Protect against application bugs                               │
 * │                                                                    │
 * │ IMPORTANT DISTINCTION:                                             │
 * │   PRIMARY KEY = identity (how we refer to a row)                   │
 * │   UNIQUE = business rule (no duplicate emails)                     │
 * │                                                                    │
 * │ A UNIQUE constraint does NOT make a column the primary key.        │
 * │ They serve different purposes.                                     │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 4. FOREIGN KEY — Enforces valid relationships
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌──────────────────────────────────────────────────────────────────────────┐
 * │ A FOREIGN KEY enforces valid relationships between tables.               │
 * │ It ensures REFERENTIAL INTEGRITY.                                        │
 * │                                                                          │
 * │ EXAMPLE:                                                                 │
 * │                                                                          │
 * │   CREATE TABLE payments (                                                │
 * │     payment_id INT AUTO_INCREMENT PRIMARY KEY,                           │
 * │     student_id INT NOT NULL,                                             │
 * │     amount DECIMAL(10,2) NOT NULL,                                       │
 * │     FOREIGN KEY (student_id) REFERENCES students(student_id)             │
 * │   );                                                                     │
 * │                                                                          │
 * │ WHAT THIS GUARANTEES:                                                    │
 * │                                                                          │
 * │   ┌──────────────────────────────────────────────────────────────────┐   │
 * │   │ Action                            │ Result                       │   │
 * │   ├───────────────────────────────────┼──────────────────────────────┤   │
 * │   │ Insert payment with student_id=1  │ OK (if student exists)       │   │
 * │   │ Insert payment with student_id=99 │ ERROR! Student doesn't exist │   |
 * │   │ Delete student who has payments   │ RESTRICT or CASCADE          │   │
 * │   └───────────────────────────────────┴──────────────────────────────┘   │
 * │                                                                          │
 * │ WHY FOREIGN KEYS MATTER:                                                 │
 * │   → Prevents ORPHAN records (payments without students)                  │
 * │   → Database ENFORCES relationships, doesn't just imply them             │
 * │   → Makes JOIN operations reliable                                       │
 * │   → Protects data integrity automatically                                │
 * │                                                                          │
 * │ IMPORTANT:                                                               │
 * │   A foreign key does NOT imply uniqueness.                               │
 * │   Multiple payments can refer to the same student.                       │
 * │   That's 1:N cardinality.                                                │
 * └──────────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 5. COMPOSITE KEY — Multiple columns as primary key
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌───────────────────────────────────────────────────────────────────────┐
 * │ Sometimes a single column is not enough to ensure uniqueness.         │
 * │ A COMPOSITE KEY combines multiple columns to form a unique key.       │
 * │                                                                       │
 * │ EXAMPLE: enrollments table                                            │
 * │                                                                       │
 * │   CREATE TABLE enrollments (                                          │
 * │     student_id INT NOT NULL,                                          │
 * │     course_id INT NOT NULL,                                           │
 * │     enrolled_at DATETIME NOT NULL,                                    │
 * │     PRIMARY KEY (student_id, course_id)   ← composite key             │
 * │   );                                                                  │
 * │                                                                       │
 * │   ┌────────────┬───────────┬─────────────────────┐                    │
 * │   │ student_id │ course_id │ enrolled_at         │                    │
 * │   ├────────────┼───────────┼─────────────────────┤                    │
 * │   │ 1          │ 101       │ 2026-01-01 10:00:00 │                    │
 * │   │ 1          │ 102       │ 2026-01-02 11:00:00 │ ← OK (diff course) │
 * │   │ 1          │ 101       │ 2026-01-03 12:00:00 │ ← REJECTED!        │
 * │   │            │           │ (duplicate combo)   │                    │
 * │   └────────────┴───────────┴─────────────────────┘                    │
 * │                                                                       │
 * │ WHAT THIS ENSURES:                                                    │
 * │   → A student can enroll in many courses                              │
 * │   → A course can have many students                                   │
 * │   → But the SAME student cannot enroll in the SAME course twice       │
 * │                                                                       │
 * │ WHEN TO USE COMPOSITE KEYS:                                           │
 * │   → Junction/bridge tables for N:M relationships                      │
 * │   → When the combination of columns uniquely identifies a row         │
 * │   → When no single column is unique                                   │
 * │                                                                       │
 * │ COMPOSITE KEYS CAN ALSO BE FOREIGN KEYS:                              │
 * │   The composite key can be referenced by another table,               │
 * │   but that's less common and adds complexity.                         │
 * └───────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 6. SURROGATE vs NATURAL KEY — Generated vs. business-meaningful
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ This is one of the most important architectural decisions.         │
 * │                                                                    │
 * │ NATURAL KEY:                                                       │
 * │   A key that has BUSINESS MEANING.                                 │
 * │   It exists in the real world.                                     │
 * │                                                                    │
 * │   Examples:                                                        │
 * │     → email                                                        │
 * │     → phone number                                                 │
 * │     → PAN number                                                   │
 * │     → passport number                                              │
 * │     → ISBN for books                                               │
 * │                                                                    │
 * │   CREATE TABLE students (                                          │
 * │     email VARCHAR(255) PRIMARY KEY,  ← natural key                 │ 
 * │     full_name VARCHAR(100) NOT NULL                                │
 * │   );                                                               │
 * │                                                                    │
 * │   PROS:                                                            │
 * │     → No extra column needed                                       │
 * │     → Business meaning is clear                                    │
 * │     → No need for JOIN to get natural identifier                   │
 * │                                                                    │
 * │   CONS:                                                            │
 * │     → Can CHANGE (email changes, phone changes)                    │
 * │     → If primary key changes, all foreign keys must update         │
 * │     → Cascading complexity                                         │
 * │     → Usually larger (VARCHAR vs INT) → slower indexes             │
 * │     → Not always available (user may not have email?)              │
 * │                                                                    │
 * │ SURROGATE KEY:                                                     │
 * │   A generated identifier with NO BUSINESS MEANING.                 │
 * │   It exists only for database purposes.                            │
 * │                                                                    │
 * │   Examples:                                                        │
 * │     → student_id (auto-increment)                                  │
 * │     → order_id                                                     │
 * │     → UUID                                                         │
 * │                                                                    │
 * │   CREATE TABLE students (                                          │
 * │     student_id INT AUTO_INCREMENT PRIMARY KEY,  ← surrogate        │ 
 * │     email VARCHAR(255) NOT NULL UNIQUE,                            │
 * │     full_name VARCHAR(100) NOT NULL                                │
 * │   );                                                               │
 * │                                                                    │
 * │   PROS:                                                            │
 * │     → STABLE (never changes)                                       │
 * │     → Small (INT/BIGINT) → faster indexes, efficient joins         │
 * │     → ISOLATED from business changes                               │
 * │     → Can be generated without business knowledge                  │
 * │                                                                    │
 * │   CONS:                                                            │
 * │     → Extra column needed                                          │
 * │     → No business meaning                                          │
 * │     → May need JOIN to get natural identifier                      │
 * │                                                                    │
 * │ THE CRITICAL DIFFERENCE:                                           │
 * │   Natural keys have meaning in the business domain.                │
 * │   Surrogate keys have meaning only in the database.                │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 7. WHY INDUSTRY PREFERS SURROGATE KEYS FOR JOINS
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ In production systems, surrogate keys are the industry standard.   │
 * │                                                                    │
 * │ REASON 1: STABILITY                                                │
 * │   Once assigned, a surrogate key NEVER changes.                    │
 * │                                                                    │
 * │   Natural key problem:                                             │
 * │     → Student changes email: 'asha@gmail.com' → 'asha.sharma@...'  │
 * │     → Email is primary key                                         │
 * │     → Need to update enrollments, payments, all foreign tables     │
 * │     → Massive update cascade                                       │
 * │     → Risk of partial updates, lock contention                     │
 * │                                                                    │
 * │   Surrogate key solution:                                          │
 * │     → student_id = 1 never changes                                 │
 * │     → Email column updated in ONE place                            │
 * │     → Foreign keys unaffected                                      │
 * │                                                                    │
 * │ REASON 2: PERFORMANCE                                              │
 * │   Surrogate keys are usually numeric (INT, BIGINT):                │
 * │     → 4 or 8 bytes                                                 │
 * │     → Smaller than VARCHAR(255) email                              │
 * │     → Faster indexing                                              │
 * │     → More entries per B+ tree page                                │
 * │     → Efficient JOINs                                              │
 * │                                                                    │
 * │ REASON 3: ISOLATION FROM BUSINESS CHANGES                          │
 * │   Business rules evolve:                                           │
 * │     → Email format changes                                         │
 * │     → Phone number format changes                                  │
 * │     → Company merges, employee IDs change                          │
 * │                                                                    │
 * │   Surrogate keys are immune to these changes.                      │
 * │   Natural keys break when business rules change.                   │
 * │                                                                    │
 * │ REASON 4: UNIQUENESS GUARANTEE                                     │
 * │   Surrogate keys are guaranteed unique by the database.            │
 * │   Natural keys may not be truly unique:                            │
 * │     → Two students could have the same name                        │
 * │     → Two students could theoretically share email? (family)       │
 * │     → Phone numbers can be reassigned                              │
 * │                                                                    │
 * │ IMPORTANT CLARIFICATION:                                           │
 * │   Surrogate keys do NOT prevent duplicates by themselves.          │
 * │                                                                    │
 * │   If you only have:                                                │
 * │     student_id (PK)                                                │
 * │     email                                                          │
 * │                                                                    │
 * │   Without UNIQUE(email), you can still store duplicate emails.     │
 * │                                                                    │
 * │   THE COMPLETE PATTERN:                                            │
 * │     → Surrogate key for IDENTITY (PRIMARY KEY)                     │
 * │     → UNIQUE constraint for BUSINESS RULES                         │
 * │                                                                    │
 * │   CREATE TABLE students (                                          │
 * │     student_id INT AUTO_INCREMENT PRIMARY KEY,  ← identity         │
 * │     email VARCHAR(255) NOT NULL UNIQUE,         ← business rule    │
 * │     full_name VARCHAR(100) NOT NULL                                │ 
 * │   );                                                               │
 * │                                                                    │
 * │   Both are needed. They serve different purposes.                  │
 * └────────────────────────────────────────────────────────────────────┘
 */
```
