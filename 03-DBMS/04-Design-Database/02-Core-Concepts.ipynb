```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * SAMPLE DOMAIN — Learning Platform (for illustrating concepts)
 * ─────────────────────────────────────────────────────────────────────
 *
 * ENTITIES WE IDENTIFY:
 *   → Student: A person who learns
 *   → Course: What students learn
 *   → Instructor: Who teaches courses
 *   → Payment: How students pay
 *   → Enrollment: The act of a student joining a course
 *
 * THE PROBLEM WITH THE FLAT TABLE:
 *   All these concepts were mixed into one raw_learning_events table.
 *   That's the root of duplication, inconsistency, and confusion.
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 1. WHY CORE CONCEPTS MATTER — Preventing chaos before it starts
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ MOST DATABASE CHAOS DOES NOT HAPPEN BECAUSE DEVELOPERS WRITE       │
 * │ BAD SQL. IT HAPPENS BECAUSE THEY NEVER PAUSE TO THINK ABOUT        │
 * │ THE CORE CONCEPTS THAT DEFINE HOW DATA SHOULD BE ORGANIZED.        │
 * │                                                                    │
 * │ THESE CONCEPTS ARE NOT ACADEMIC THEORY.                            │
 * │ THEY ARE PRACTICAL TOOLS THAT PREVENT:                             │
 * │   → Duplication                                                    │
 * │   → Inconsistency                                                  │
 * │   → Scaling disasters                                              │
 * │   → Data anomalies                                                 │
 * │   → Orphan records                                                 │
 * │   → Business rule violations                                       │
 * │                                                                    │
 * │ When you ignore these concepts:                                    │
 * │   → Entities become mixed                                          │
 * │   → Attributes become duplicated                                   │
 * │   → Relationships become unclear                                   │
 * │   → Cardinality becomes accidental                                 │
 * │   → Constraints become missing                                     │
 * │                                                                    │
 * │ The result is predictable: DATA CHAOS.                             │
 * │                                                                    │
 * │ When you apply these principles:                                   │
 * │   → Each entity has its own table                                  │
 * │   → Each attribute belongs in one place                            │
 * │   → Relationships are enforced                                     │
 * │   → Cardinality is intentional                                     │
 * │   → Constraints protect the system                                 │
 * │                                                                    │
 * │ The database scales naturally.                                     │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 2. NORMALIZATION — Reduce duplication, prevent anomalies
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ NORMALIZATION is the process of designing tables that reduce       │
 * │ duplication and prevent data anomalies.                            │
 * │                                                                    │
 * │ WHAT NORMALIZATION DOES:                                           │
 * │   → Separates repeated information into logical entities           │
 * │   → Ensures each fact is stored ONCE and only once                 │
 * │   → Prevents update anomalies (updating some rows, not others)     │
 * │   → Prevents insertion anomalies (can't insert without other data) │
 * │   → Prevents deletion anomalies (losing data when deleting)        │
 * │                                                                    │
 * │ THE FLAT TABLE ANTI-PATTERN:                                       │
 * │                                                                    │
 * │   raw_learning_events stores:                                      │
 * │     → Student details repeated on every enrollment                 │
 * │     → Course details repeated on every enrollment                  │
 * │     → Instructor details repeated on every enrollment              │
 * │                                                                    │
 * │   Result:                                                          │
 * │     → Update course title → must update 10,000 rows                │
 * │     → Student name inconsistency → Asha vs Asha Sharma             │
 * │     → Delete a student → lose enrollment history                   │
 * │                                                                    │
 * │ NORMALIZED DESIGN:                                                 │
 * │                                                                    │
 * │   students(id, email, name)          → one row per student         │
 * │   courses(id, code, title, price)    → one row per course          │
 * │   instructors(id, email, name)       → one row per instructor      │
 * │   enrollments(id, student_id, course_id, enrolled_at)              │
 * │   payments(id, enrollment_id, amount, status, paid_at)             │
 * │                                                                    │
 * │   Result:                                                          │
 * │     → Update course title → ONE row                                │
 * │     → Student name consistent across all enrollments               │
 * │     → Delete student safely → enrollments remain (history kept)    │
 * │                                                                    │
 * │ NORMALIZATION IS NOT "SPLITTING TABLES RANDOMLY."                  │
 * │ IT IS DESIGNING TABLES SO THAT EACH FACT IS STORED ONCE.           │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 3. ENTITIES — Real-world things you store data about
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ AN ENTITY is a real-world thing you store data about.              │
 * │                                                                    │
 * │ IN OUR LEARNING PLATFORM:                                          │
 * │                                                                    │
 * │   ┌─────────────────────────────────────────────────────────┐      │
 * │   │ Entity      │ What it represents                        │      │
 * │   ├─────────────┼───────────────────────────────────────────┤      │
 * │   │ Student     │ A person who learns                       │      │
 * │   │ Course      │ A learning offering                       │      │
 * │   │ Instructor  │ A person who teaches                      │      │
 * │   │ Payment     │ A financial transaction                   │      │
 * │   │ Enrollment  │ The act of a student joining a course     │      │
 * │   └─────────────┴───────────────────────────────────────────┘      │
 * │                                                                    │
 * │ KEY INSIGHT:                                                       │
 * │   Each entity exists INDEPENDENTLY in the real world.              │
 * │   → A student exists whether they pay or not.                      │
 * │   → A course exists whether someone enrolls or not.                │
 * │   → A payment exists because an enrollment exists.                 │
 * │                                                                    │
 * │ WHAT HAPPENS WHEN YOU MIX ENTITIES?                                │
 * │                                                                    │
 * │   In raw_learning_events:                                          │
 * │     → Student, Course, Instructor, Payment all in ONE ROW          │
 * │     → What does a row REPRESENT? A purchase event?                 │
 * │     → But then:                                                    │
 * │         • Student data is repeated                                 │
 * │         • Course data is repeated                                  │
 * │         • No clear separation of concerns                          │
 * │                                                                    │
 * │ GOOD DATABASE DESIGN BEGINS BY IDENTIFYING ENTITIES CORRECTLY.     │
 * │ Each entity gets its own table.                                    │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 4. ATTRIBUTES — Properties that describe entities
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ AN ATTRIBUTE is a property of an entity.                           │
 * │                                                                    │
 * │ ATTRIBUTES PER ENTITY:                                             │
 * │                                                                    │
 * │   ┌─────────────────────────────────────────────────────────┐      │
 * │   │ Entity      │ Attributes                                │      │
 * │   ├─────────────┼───────────────────────────────────────────┤      │
 * │   │ Student     │ id, email, name, created_at               │      │
 * │   │ Course      │ id, code, title, listed_price, currency   │      │
 * │   │ Instructor  │ id, email, name, joined_at                │      │
 * │   │ Enrollment  │ id, student_id, course_id, enrolled_at    │      │
 * │   │ Payment     │ id, enrollment_id, amount, status, paid_at│      │
 * │   └─────────────┴───────────────────────────────────────────┘      │
 * │                                                                    │
 * │ KEY INSIGHT:                                                       │
 * │   Attributes describe entities. They do NOT define relationships   │
 * │   between entities.                                                │
 * │                                                                    │
 * │   ✅✅ email is an attribute of Student                           │
 * │   ✅✅ title is an attribute of Course                            │
 * │   ✅✅ amount is an attribute of Payment                          │
 * │                                                                    │
 * │   ❌❌ email should NOT be stored in Enrollment                   │
 * │   ❌❌ course_title should NOT be stored in Payment               │
 * │   ❌❌ instructor_name should NOT be stored in Course             │
 * │                                                                    │
 * │ ATTRIBUTE VALIDITY:                                                │
 * │   → email must be unique (student identifier)                      │
 * │   → amount must be > 0 (payment validation)                        │
 * │   → enrolled_at must be timestamp (when event occurred)            │
 * │                                                                    │
 * │ This is where structure starts to enforce CORRECTNESS.             │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 5. RELATIONSHIPS — How entities interact
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ RELATIONSHIPS define how entities interact with each other.        │
 * │                                                                    │
 * │ IN OUR LEARNING PLATFORM:                                          │
 * │                                                                    │
 * │   ┌──────────────────────────────────────────────────────────┐     │
 * │   │ Relationship              │ How they connect             │     │
 * │   ├───────────────────────────┼──────────────────────────────┤     │
 * │   │ Student → Enrollment      │ A student enrolls in courses │     │
 * │   │ Course → Enrollment       │ A course has students        │     │
 * │   │ Enrollment → Payment      │ An enrollment has a payment  │     │
 * │   │ Course → Instructor       │ A course is taught by someone│     │
 * │   └───────────────────────────┴──────────────────────────────┘     │
 * │                                                                    │
 * │ WITHOUT EXPLICIT RELATIONSHIPS:                                    │
 * │   → raw_learning_events stores everything in one table             │
 * │   → Relationships are IMPLIED but not ENFORCED                     │
 * │   → The database cannot validate if a student exists               │
 * │   → The database cannot prevent orphan payments                    │
 * │   → The database cannot ensure course references are valid         │
 * │                                                                    │
 * │ WITH EXPLICIT RELATIONSHIPS (Foreign Keys):                        │
 * │                                                                    │
 * │   enrollments:                                                     │
 * │     student_id REFERENCES students(id)    → ensures student exists │
 * │     course_id REFERENCES courses(id)      → ensures course exists  │
 * │                                                                    │
 * │   payments:                                                        │
 * │     enrollment_id REFERENCES enrollments(id) → ensures enrollment  │
 * │                                                                    │
 * │   courses:                                                         │
 * │     instructor_id REFERENCES instructors(id) → ensures instructor  │
 * │                                                                    │
 * │ WHEN RELATIONSHIPS ARE EXPLICIT:                                   │
 * │   → The database becomes SELF-VALIDATING                           │
 * │   → You cannot create an enrollment for a non-existent student     │
 * │   → You cannot create a payment for a non-existent enrollment      │
 * │   → Data integrity is guaranteed                                   │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 6. CARDINALITY — One-to-One, One-to-Many, Many-to-Many
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ CARDINALITY answers one simple but critical question:              │
 * │   "HOW MANY?"                                                      │
 * │                                                                    │
 * │ TYPES OF CARDINALITY:                                              │
 * │                                                                    │
 * │   1. ONE-TO-ONE (1:1)                                              │
 * │      → One student has ONE profile                                 │
 * │      → One enrollment has ONE payment (in simple model)            │
 * │                                                                    │
 * │   2. ONE-TO-MANY (1:N)                                             │
 * │      → One student can have MANY enrollments                       │
 * │      → One course can have MANY enrollments                        │
 * │      → One instructor can teach MANY courses                       │
 * │                                                                    │
 * │   3. MANY-TO-MANY (M:N)                                            │
 * │      → Many students can enroll in MANY courses                    │
 * │      → This is implemented via a junction table (enrollments)      │
 * │                                                                    │
 * │ VISUAL REPRESENTATION:                                             │
 * │                                                                    │
 * │   ONE-TO-MANY:                                                     │
 * │   ┌─────────────┐          ┌─────────────┐                         │
 * │   │ Student     │ 1      N │ Enrollment  │                         │
 * │   │ (one)       │────────► │ (many)      │                         │
 * │   └─────────────┘          └─────────────┘                         │
 * │                                                                    │
 * │   MANY-TO-MANY (via junction):                                     │
 * │   ┌─────────────┐     ┌─────────────┐     ┌─────────────┐          │
 * │   │ Student     │ M   │ Enrollment  │ N   │ Course      │          │
 * │   └─────────────┘────►└─────────────┘◄────└─────────────┘          │
 * │                                                                    │
 * │ WHY CARDINALITY MATTERS:                                           │
 * │   → Without cardinality, you allow duplicate enrollments           │
 * │   → Without cardinality, you allow a student to enroll twice       │
 * │   → Without cardinality, you allow inconsistent business logic     │
 * │                                                                    │
 * │ IN THE FLAT TABLE:                                                 │
 * │   → No explicit cardinality definition                             │
 * │   → Dev enrolled in LLD301 twice → allowed!                        │
 * │   → Business rule violated                                         │
 * │                                                                    │
 * │ CARDINALITY PROTECTS YOUR DATA MODEL FROM AMBIGUITY.               │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 7. CONSTRAINTS — Rules that reject bad data
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ CONSTRAINTS are rules that reject bad data.                        │
 * │ THEY ARE NOT OPTIONAL. THEY ARE ESSENTIAL.                         │
 * │                                                                    │
 * │ TYPES OF CONSTRAINTS:                                              │
 * │                                                                    │
 * │   ┌──────────────────────────────────────────────────────────┐     │
 * │   │ Constraint Type  │ What it does                          │     │
 * │   ├──────────────────┼───────────────────────────────────────┤     │
 * │   │ PRIMARY KEY      │ Uniquely identifies each row          │     │
 * │   │ FOREIGN KEY      │ Ensures relationships are valid       │     │
 * │   │ UNIQUE           │ Prevents duplicate values             │     │
 * │   │ NOT NULL         │ Ensures a value is provided           │     │
 * │   │ CHECK            │ Validates value against condition     │     │
 * │   │ DEFAULT          │ Provides fallback value               │     │
 * │   └──────────────────┴───────────────────────────────────────┘     │
 * │                                                                    │
 * │ IN THE FLAT TABLE:                                                 │
 * │                                                                    │
 * │   raw_learning_events has ALMOST NO CONSTRAINTS:                   │
 * │     → No UNIQUE on email → Asha duplicate allowed                  │
 * │     → No UNIQUE on (student, course) → duplicate enrollment allowed│
 * │     → No CHECK on amount → negative payments possible              │
 * │     → No FOREIGN KEY → orphan records possible                     │
 * │     → No NOT NULL on important fields → incomplete data            │
 * │                                                                    │
 * │   The database accepts EVERYTHING, including mistakes.             │
 * │                                                                    │
 * │ IN A WELL-DESIGNED DATABASE:                                       │
 * │                                                                    │
 * │   CREATE TABLE students (                                          │
 * │     id SERIAL PRIMARY KEY,                                         │
 * │     email VARCHAR(255) UNIQUE NOT NULL,    ← prevents duplicates   │
 * │     name VARCHAR(100) NOT NULL             ← ensures completeness  │ 
 * │   );                                                               │
 * │                                                                    │
 * │   CREATE TABLE enrollments (                                       │
 * │     id SERIAL PRIMARY KEY,                                         │
 * │     student_id INT NOT NULL REFERENCES students(id),               │
 * │     course_id INT NOT NULL REFERENCES courses(id),                 │
 * │     enrolled_at TIMESTAMP NOT NULL DEFAULT NOW(),                  │
 * │     UNIQUE(student_id, course_id)         ← prevents duplicates    │
 * │   );                                                               │
 * │                                                                    │
 * │   CREATE TABLE payments (                                          │
 * │     id SERIAL PRIMARY KEY,                                         │
 * │     enrollment_id INT NOT NULL REFERENCES enrollments(id),         │
 * │     amount DECIMAL(10,2) NOT NULL CHECK(amount > 0), ← validates   │
 * │     status VARCHAR(20) NOT NULL CHECK(status IN ('PAID', ...)),    │
 * │     paid_at TIMESTAMP                                              │
 * │   );                                                               │
 * │                                                                    │
 * │ A WELL-DESIGNED SYSTEM REJECTS BAD DATA AUTOMATICALLY.             │
 * │ No application logic needed. The database enforces correctness.    │
 * └────────────────────────────────────────────────────────────────────┘
 */
```

```python
/**
 * ─────────────────────────────────────────────────────────────────────
 * 8. WHY THIS MATTERS — From chaos to clarity
 * ─────────────────────────────────────────────────────────────────────
 *
 * ┌────────────────────────────────────────────────────────────────────┐
 * │ BEFORE (Flat Table - raw_learning_events):                         │
 * │                                                                    │
 * │   ┌──────────────────────────────────────────────────────────┐     │
 * │   │ Problems:                                                |     │
 * │   │   X Entities mixed (student, course, instructor all one) │     │
 * │   │   X Attributes duplicated (student name repeated)        │     │
 * │   │   X Relationships implied, not enforced                  │     │
 * │   │   X Cardinality accidental (duplicate enrollments)       │     │
 * │   │   X Constraints missing (no uniqueness, no foreign keys) │     │
 * │   │                                                          │     │
 * │   │ Result:                                                  │     │
 * │   │   → Inconsistent student names                           │     │
 * │   │   → Course updates require 10,000 row changes            │     │
 * │   │   → Duplicate enrollments possible                       │     │
 * │   │   → Orphan payments possible                             │     │
 * │   │   → Data chaos                                           │     │
 * │   └──────────────────────────────────────────────────────────┘     │
 * │                                                                    │
 * │ AFTER (Normalized Design):                                         │
 * │                                                                    │
 * │   ┌──────────────────────────────────────────────────────────┐     │
 * │   │ Solutions:                                               │     │
 * │   │   - Entities separated (students, courses, instructors)  │     │
 * │   │   - Attributes stored once (email in one place)          │     │
 * │   │   - Relationships enforced via foreign keys              │     │
 * │   │   - Cardinality intentional (UNIQUE constraints)         │     │
 * │   │   - Constraints protect data (CHECK, NOT NULL)           │     │
 * │   │                                                          │     │
 * │   │ Result:                                                  │     │
 * │   │   → Student names consistent                             │     │
 * │   │   → Course updates affect ONE row                        │     │
 * │   │   → Duplicate enrollments prevented                      │     │
 * │   │   → Orphan payments impossible                           │     │
 * │   │   → Data integrity guaranteed                            │     │
 * │   └──────────────────────────────────────────────────────────┘     │
 * │                                                                    │
 * │ THE DATABASE SCALES NATURALLY WHEN THESE PRINCIPLES ARE APPLIED.   │
 * └────────────────────────────────────────────────────────────────────┘
 */
```
