```python
/**
 * JavaScript Dates:
*/

/**
 * 1. Creating Dates:
 *    - JavaScript uses the Date object to work with dates and times.
 *    - Note: Months are 0-indexed (January = 0, December = 11)
*/
let now = new Date();                         // Current date and time
let date1 = new Date("2025-05-15");           // From ISO date string
let date2 = new Date(2025, 4, 15);            // Year, Month (0-based), Day → May 15, 2025
let date3 = new Date(2025, 4, 15, 12, 30, 0); // + hours, minutes, seconds
let date4 = new Date(0);                      // Unix epoch → Jan 1, 1970 UTC
let date5 = new Date(1000 * 60 * 60);         // 1 hour after epoch


/**
 * 2. Date Internals: Timestamps
 *    Dates are stored as milliseconds since Unix Epoch: Jan 1, 1970 UTC
*/
let timestamp = Date.now();  // Current timestamp in ms
let d1 = new Date(timestamp); // Convert back to Date


/**
 * 3. Get Methods:
 *    - getFullYear()       - Year (4-digit)
 *    - getMonth()          - Month (0–11)
 *    - getDate()           - Day of the month (1–31)
 *    - getDay()            - Day of the week (0–6, Sun = 0)
 *    - getHours()          - Hour (0–23)
 *    - getMinutes()        - Minutes (0–59)
 *    - getSeconds()        - Seconds (0–59)
 *    - getMilliseconds()   - Milliseconds (0–999)
 *    - getTime()           - Timestamp (ms since epoch)
 *    - getTimezoneOffset() - Minutes offset from UTC
 *    - toISOString()       - ISO 8601 string in UTC
 *    - toDateString()      - Human-readable date string
 *    - toTimeString()      - Human-readable time string
 *    - toLocaleString()    - Local date and time string
*/

let d2 = new Date("2025-05-15T10:30:00");
d2.getFullYear();      // 2025
d2.getMonth();         // 4 (May)
d2.getDate();          // 15
d2.getDay();           // 4 (Thursday)


/**
 * 4. Set Methods:
 *    - setFullYear(y)       - Set year
 *    - setMonth(m)          - Set month (0–11)
 *    - setDate(d) 	         - Set day of month
 *    - setHours(h) 	     - Set hour
 *    - setMinutes(m) 	     - Set minutes
 *    - setSeconds(s) 	     - Set seconds
 *    - setMilliseconds(ms)  - Set milliseconds
 *    - setTime(ms) 	     - Set time from timestamp
*/

let d3 = new Date();
d3.setFullYear(2026);
d3.setMonth(11); // December
d3.setDate(25);  // 25th


/**
 * 5. Parsing Dates
 *    - Date.parse(str): Parses date string → timestamp
 *    - new Date(str)  : Also parses ISO-like date strings
 *    - Best to use ISO 8601 format (YYYY-MM-DD or YYYY-MM-DDTHH:mm:ssZ) 
 *      for cross-browser compatibility.
*/

Date.parse("2025-05-15");  // Returns timestamp
new Date("2025-05-15");    // Valid date object


/**
 * 6. Formatting Dates
 *    - toDateString() 	     - Returns just the date part
 *    - toTimeString() 	     - Returns just the time part
 *    - toISOString() 	     - Returns UTC time in ISO format
 *    - toLocaleDateString() - Returns date in local format
 *    - toLocaleTimeString() - Returns time in local format
 *    - toLocaleString()     - Full localized date and time
*/

let d4 = new Date();
d4.toISOString();        // "2025-05-15T12:30:00.000Z"
d4.toLocaleDateString(); // e.g., "5/15/2025" (US)


/**
 * 7. Date Comparison:
 *    - You can directly compare Date objects using > < ===, but using 
 *      .getTime() is safer for equality.
*/
let d5 = new Date("2025-05-15");
let d6 = new Date("2025-05-20");

d5 > d6                        // false
d5.getTime() === d6.getTime(); // false


/**
 * 8. Date Arithmetic:
 *    You can add/subtract using timestamps:
*/
let d7 = new Date();
let tomorrow = new Date(d7.getTime() + 24 * 60 * 60 * 1000);     // add 1 day
let lastWeek = new Date(d7.getTime() - 7 * 24 * 60 * 60 * 1000); // subtract 7 days

/* Or manipulate with .setDate(), .setMonth() etc.: */
let d8 = new Date("2025-05-15");
d8.setDate(d8.getDate() + 5); // Adds 5 days


/**
 * 9. Working with Timezones:
 *    JavaScript Date is based on local time, but you can get UTC 
 *    components:
*/

d9.getUTCFullYear()
d9.getUTCHours()

/* Note: To convert between timezones, use libraries like Luxon, date-fns-tz, or moment-timezone. */



/**
 * 10. International Formatting:
*/
let formatter = new Intl.DateTimeFormat("en-US", {
    dateStyle: "full",
    timeStyle: "short",
    timeZone: "Asia/Kolkata"
  });
  
formatter.format(new Date());
// Example: "Thursday, May 15, 2025 at 6:00 PM"


/**
 * Common Use Cases:
 * 1. Get current timestamp     - Date.now()
 * 2. Convert timestamp to date - new Date(ms)
 * 3. Format to ISO string      - date.toISOString()
 * 4. Add days to a date        - date.setDate(date.getDate() + n)
 * 5. Compare two dates         - date1.getTime() > date2.getTime()
 * 6. Parse ISO string          - new Date("YYYY-MM-DD")
 * 7. Format to local date/time - date.toLocaleString("en-IN", options)
*/
```

```python
/**
 * PROBLEM 1: Get current date and time
 * - Input:  (none)
 * - Output: Current date and time as a Date object
*/

const now1 = new Date();
console.log(now1);

// Output: Current date/time (e.g., 2026-09-01T10:30:00.000Z)


/**
 * PROBLEM 2: Get current timestamp in milliseconds
 * - Input:  (none)
 * - Output: 1725180000000 (milliseconds since Jan 1, 1970)
*/

const timestamp2 = Date.now();
console.log(timestamp2);

// Output: 1725180000000 (varies)


/**
 * PROBLEM 3: Create a specific date (May 15, 2025)
 * - Input:  Year: 2025, Month: May, Day: 15
 * - Output: Date object for 2025-05-15
*/

const date3 = new Date(2025, 4, 15); // Month is 0-indexed (May = 4)
console.log(date3);

// Output: 2025-05-15T00:00:00.000Z (local midnight)


/**
 * PROBLEM 4: Create a date from ISO string
 * - Input:  "2025-05-15T10:30:00"
 * - Output: Date object
*/

const date4 = new Date("2025-05-15T10:30:00");
console.log(date4);

// Output: 2025-05-15T10:30:00.000Z


/**
 * PROBLEM 5: Get year, month, and day from a Date
 * - Input:  new Date("2025-05-15")
 * - Output: Year: 2025, Month: 4 (May), Day: 15
*/

const date5 = new Date("2025-05-15");
console.log("Year:", date5.getFullYear());
console.log("Month:", date5.getMonth());
console.log("Day:", date5.getDate());

// Output: Year: 2025, Month: 4, Day: 15


/**
 * PROBLEM 6: Get day of the week (0 = Sunday)
 * - Input:  new Date("2025-05-15")
 * - Output: 4 (Thursday)
*/

const date6 = new Date("2025-05-15");
const dayOfWeek6 = date6.getDay();
const days6 = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"];
console.log(days6[dayOfWeek6]);

// Output: "Thursday"


/**
 * PROBLEM 7: Get hours, minutes, seconds from a Date
 * - Input:  new Date("2025-05-15T10:30:45")
 * - Output: Hours: 10, Minutes: 30, Seconds: 45
*/

const date7 = new Date("2025-05-15T10:30:45");
console.log("Hours:", date7.getHours());
console.log("Minutes:", date7.getMinutes());
console.log("Seconds:", date7.getSeconds());

// Output: Hours: 10, Minutes: 30, Seconds: 45


/**
 * PROBLEM 8: Format date as "DD/MM/YYYY"
 * - Input:  new Date("2025-05-15")
 * - Output: "15/05/2025"
*/

const date8 = new Date("2025-05-15");
const day8 = String(date8.getDate()).padStart(2, "0");
const month8 = String(date8.getMonth() + 1).padStart(2, "0");
const year8 = date8.getFullYear();
console.log(`${day8}/${month8}/${year8}`);

// Output: "15/05/2025"


/**
 * PROBLEM 9: Format date as "MM/DD/YYYY"
 * - Input:  new Date("2025-05-15")
 * - Output: "05/15/2025"
*/

const date9 = new Date("2025-05-15");
const month9 = String(date9.getMonth() + 1).padStart(2, "0");
const day9 = String(date9.getDate()).padStart(2, "0");
const year9 = date9.getFullYear();
console.log(`${month9}/${day9}/${year9}`);

// Output: "05/15/2025"


/**
 * PROBLEM 10: Format date as "YYYY-MM-DD"
 * - Input:  new Date("2025-05-15")
 * - Output: "2025-05-15"
*/

const date10 = new Date("2025-05-15");
const year10 = date10.getFullYear();
const month10 = String(date10.getMonth() + 1).padStart(2, "0");
const day10 = String(date10.getDate()).padStart(2, "0");
console.log(`${year10}-${month10}-${day10}`);

// Output: "2025-05-15"


/**
 * PROBLEM 11: Convert Date to ISO string
 * - Input:  new Date("2025-05-15T10:30:00")
 * - Output: "2025-05-15T10:30:00.000Z"
*/

const date11 = new Date("2025-05-15T10:30:00");
console.log(date11.toISOString());

// Output: "2025-05-15T10:30:00.000Z"


/**
 * PROBLEM 12: Format date using toLocaleDateString
 * - Input:  new Date("2025-05-15")
 * - Output: "5/15/2025" (en-US format)
*/

const date12 = new Date("2025-05-15");
console.log(date12.toLocaleDateString("en-US"));

// Output: "5/15/2025"


/**
 * PROBLEM 13: Format date using toLocaleString with options
 * - Input:  new Date("2025-05-15T10:30:00")
 * - Output: "May 15, 2025 at 10:30 AM"
*/

const date13 = new Date("2025-05-15T10:30:00");
console.log(
  date13.toLocaleString("en-US", {
    dateStyle: "medium",
    timeStyle: "short",
  })
);

// Output: "May 15, 2025, 10:30 AM"


/**
 * PROBLEM 14: Add days to a date
 * - Input:  new Date("2025-05-15"), daysToAdd = 7
 * - Output: 2025-05-22
*/

const date14 = new Date("2025-05-15");
date14.setDate(date14.getDate() + 7);
console.log(date14.toISOString().split("T")[0]);

// Output: "2025-05-22"


/**
 * PROBLEM 15: Subtract days from a date
 * - Input:  new Date("2025-05-15"), daysToSubtract = 10
 * - Output: 2025-05-05
*/

const date15 = new Date("2025-05-15");
date15.setDate(date15.getDate() - 10);
console.log(date15.toISOString().split("T")[0]);

// Output: "2025-05-05"


/**
 * PROBLEM 16: Add months to a date
 * - Input:  new Date("2025-05-15"), monthsToAdd = 3
 * - Output: 2025-08-15
*/

const date16 = new Date("2025-05-15");
date16.setMonth(date16.getMonth() + 3);
console.log(date16.toISOString().split("T")[0]);

// Output: "2025-08-15"


/**
 * PROBLEM 17: Add years to a date
 * - Input:  new Date("2025-05-15"), yearsToAdd = 2
 * - Output: 2027-05-15
*/

const date17 = new Date("2025-05-15");
date17.setFullYear(date17.getFullYear() + 2);
console.log(date17.toISOString().split("T")[0]);

// Output: "2027-05-15"


/**
 * PROBLEM 18: Compare two dates (which is earlier?)
 * - Input:  "2025-05-15" and "2025-05-20"
 * - Output: "2025-05-15 is earlier"
*/

const date18a = new Date("2025-05-15");
const date18b = new Date("2025-05-20");
if (date18a.getTime() < date18b.getTime()) {
  console.log("2025-05-15 is earlier");
} else {
  console.log("2025-05-20 is earlier");
}

// Output: "2025-05-15 is earlier"


/**
 * PROBLEM 19: Check if two dates are the same day
 * - Input:  "2025-05-15T10:30:00" and "2025-05-15T22:45:00"
 * - Output: true
*/

const date19a = new Date("2025-05-15T10:30:00");
const date19b = new Date("2025-05-15T22:45:00");
const isSameDay19 =
  date19a.getFullYear() === date19b.getFullYear() &&
  date19a.getMonth() === date19b.getMonth() &&
  date19a.getDate() === date19b.getDate();
console.log(isSameDay19);

// Output: true


/**
 * PROBLEM 20: Check if a date is in the past
 * - Input:  new Date("2020-01-01")
 * - Output: true
*/

const date20 = new Date("2020-01-01");
const isPast20 = date20.getTime() < Date.now();
console.log(isPast20);

// Output: true


/**
 * PROBLEM 21: Check if a date is in the future
 * - Input:  new Date("2030-01-01")
 * - Output: true
*/

const date21 = new Date("2030-01-01");
const isFuture21 = date21.getTime() > Date.now();
console.log(isFuture21);

// Output: true


/**
 * PROBLEM 22: Calculate difference in days between two dates
 * - Input:  "2025-05-15" and "2025-05-25"
 * - Output: 10
*/

const date22a = new Date("2025-05-15");
const date22b = new Date("2025-05-25");
const diffMs22 = Math.abs(date22b.getTime() - date22a.getTime());
const diffDays22 = Math.floor(diffMs22 / (1000 * 60 * 60 * 24));
console.log(diffDays22);

// Output: 10


/**
 * PROBLEM 23: Calculate difference in hours between two dates
 * - Input:  "2025-05-15T10:00:00" and "2025-05-15T15:30:00"
 * - Output: 5.5
*/

const date23a = new Date("2025-05-15T10:00:00");
const date23b = new Date("2025-05-15T15:30:00");
const diffMs23 = Math.abs(date23b.getTime() - date23a.getTime());
const diffHours23 = diffMs23 / (1000 * 60 * 60);
console.log(diffHours23);

// Output: 5.5


/**
 * PROBLEM 24: Calculate age from birthdate
 * - Input:  birthdate = "1995-06-15"
 * - Output: 30 (as of 2026)
*/

const birthdate24 = new Date("1995-06-15");
const today24 = new Date();
let age24 = today24.getFullYear() - birthdate24.getFullYear();
const monthDiff24 = today24.getMonth() - birthdate24.getMonth();
if (monthDiff24 < 0 || (monthDiff24 === 0 && today24.getDate() < birthdate24.getDate())) {
  age24--;
}
console.log(age24);

// Output: 31 (varies based on current date)


/**
 * PROBLEM 25: Check if a year is a leap year
 * - Input:  2024
 * - Output: true
*/

const year25 = 2024;
const isLeapYear25 = (year) => {
  return (year % 4 === 0 && year % 100 !== 0) || year % 400 === 0;
};
console.log(isLeapYear25(year25));

// Output: true


/**
 * PROBLEM 26: Get the number of days in a month
 * - Input:  Year: 2025, Month: May (index 4)
 * - Output: 31
*/

const year26 = 2025;
const month26 = 4; // May (0-indexed)
// Day 0 of next month = last day of current month
const daysInMonth26 = new Date(year26, month26 + 1, 0).getDate();
console.log(daysInMonth26);

// Output: 31


/**
 * PROBLEM 27: Get the start and end of the current week (Sunday to Saturday)
 * - Input:  new Date("2025-05-15") (Thursday)
 * - Output: Start: 2025-05-11 (Sunday), End: 2025-05-17 (Saturday)
*/

const date27 = new Date("2025-05-15");
const day27 = date27.getDay(); // 4 (Thursday)
const startDate27 = new Date(date27);
startDate27.setDate(date27.getDate() - day27); // Go back to Sunday
const endDate27 = new Date(date27);
endDate27.setDate(date27.getDate() + (6 - day27)); // Go forward to Saturday
console.log("Start:", startDate27.toISOString().split("T")[0]);
console.log("End:", endDate27.toISOString().split("T")[0]);

// Output: Start: "2025-05-11", End: "2025-05-17"


/**
 * PROBLEM 28: Get the start and end of the current month
 * - Input:  new Date("2025-05-15")
 * - Output: Start: 2025-05-01, End: 2025-05-31
*/

const date28 = new Date("2025-05-15");
const startDate28 = new Date(date28.getFullYear(), date28.getMonth(), 1);
const endDate28 = new Date(date28.getFullYear(), date28.getMonth() + 1, 0);
console.log("Start:", startDate28.toISOString().split("T")[0]);
console.log("End:", endDate28.toISOString().split("T")[0]);

// Output: Start: "2025-05-01", End: "2025-05-31"


/**
 * PROBLEM 29: Format date using Intl.DateTimeFormat
 * - Input:  new Date("2025-05-15T15:30:00")
 * - Output: "Thursday, May 15, 2025 at 3:30 PM"
*/

const date29 = new Date("2025-05-15T15:30:00");
const formatter29 = new Intl.DateTimeFormat("en-US", {
  dateStyle: "full",
  timeStyle: "short",
});
console.log(formatter29.format(date29));

// Output: "Thursday, May 15, 2025 at 3:30 PM"


/**
 * PROBLEM 30: Convert timestamp to human-readable date
 * - Input:  1725180000000 (timestamp in ms)
 * - Output: "2024-09-01" (or local equivalent)
*/

const timestamp30 = 1725180000000;
const date30 = new Date(timestamp30);
const year30 = date30.getFullYear();
const month30 = String(date30.getMonth() + 1).padStart(2, "0");
const day30 = String(date30.getDate()).padStart(2, "0");
console.log(`${year30}-${month30}-${day30}`);

// Output: "2024-09-01"
```
