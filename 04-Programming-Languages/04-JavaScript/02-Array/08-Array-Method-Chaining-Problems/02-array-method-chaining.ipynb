```python
/**
 * PROBLEM 1: Filter words starting with vowel, then get their lengths
 * - Input:  ['apple', 'banana', 'orange', 'grape', 'umbrella']
 * - Output: [5, 6, 8]
*/

const arr1 = ['apple', 'banana', 'orange', 'grape', 'umbrella'];
const result1 = arr1
  .filter((w) => 'aeiouAEIOU'.includes(w[0]))
  .map((w) => w.length);
console.log(result1);

// Output: [5, 6, 8]


/**
 * PROBLEM 2: Map to square, filter squares > 20, then sort descending
 * - Input:  [2, 5, 3, 6, 4]
 * - Output: [36, 25]
*/

const arr2 = [2, 5, 3, 6, 4];
const result2 = arr2
  .map((n) => n * n)
  .filter((n) => n > 20)
  .sort((a, b) => b - a);
console.log(result2);

// Output: [36, 25]


/**
 * PROBLEM 3: Filter strings with length > 3, sort alphabetically, then join
 * - Input:  ['dog', 'elephant', 'cat', 'banana', 'ant']
 * - Output: "banana,elephant"
*/

const arr3 = ['dog', 'elephant', 'cat', 'banana', 'ant'];
const result3 = arr3
  .filter((s) => s.length > 3)
  .sort()
  .join(',');
console.log(result3);

// Output: "banana,elephant"


/**
 * PROBLEM 4: Extract ages, filter adults, sort descending, then get first
 * - Input:  [{name: 'A', age: 17}, {name: 'B', age: 25}, {name: 'C', age: 30}, {name: 'D', age: 16}]
 * - Output: 30
*/

const arr4 = [{ name: 'A', age: 17 }, { name: 'B', age: 25 }, { name: 'C', age: 30 }, { name: 'D', age: 16 }];
const result4 = arr4
  .map((u) => u.age)
  .filter((age) => age >= 18)
  .sort((a, b) => b - a)[0];
console.log(result4);

// Output: 30


/**
 * PROBLEM 5: Filter numbers divisible by 3, map to cube, then sum
 * - Input:  [1, 3, 5, 6, 9, 10]
 * - Output: 972 (27 + 216 + 729)
*/

const arr5 = [1, 3, 5, 6, 9, 10];
const result5 = arr5
  .filter((n) => n % 3 === 0)
  .map((n) => n ** 3)
  .reduce((sum, n) => sum + n, 0);
console.log(result5);

// Output: 972


/**
 * PROBLEM 6: Map to extract city, filter unique, then sort
 * - Input:  [{name: 'A', city: 'Delhi'}, {name: 'B', city: 'Mumbai'}, {name: 'C', city: 'Delhi'}, {name: 'D', city: 'Bangalore'}]
 * - Output: ['Bangalore', 'Delhi', 'Mumbai']
*/

const arr6 = [
  { name: 'A', city: 'Delhi' },
  { name: 'B', city: 'Mumbai' },
  { name: 'C', city: 'Delhi' },
  { name: 'D', city: 'Bangalore' },
];
const result6 = [...new Set(arr6.map((u) => u.city))].sort();
console.log(result6);

// Output: ['Bangalore', 'Delhi', 'Mumbai']


/**
 * PROBLEM 7: Filter out falsy values, map to strings, then join with space
 * - Input:  [0, 'hello', false, 'world', null, 'js', undefined, '!']
 * - Output: "hello world js !"
*/

const arr7 = [0, 'hello', false, 'world', null, 'js', undefined, '!'];
const result7 = arr7
  .filter(Boolean)
  .map(String)
  .join(' ');
console.log(result7);

// Output: "hello world js !"


/**
 * PROBLEM 8: Map to extract price, filter > 50, then find average
 * - Input:  [{price: 30}, {price: 80}, {price: 60}, {price: 20}, {price: 100}]
 * - Output: 80 ( (80 + 60 + 100) / 3 )
*/

const arr8 = [{ price: 30 }, { price: 80 }, { price: 60 }, { price: 20 }, { price: 100 }];
const filteredPrices8 = arr8.map((p) => p.price).filter((p) => p > 50);
const result8 = filteredPrices8.reduce((sum, p) => sum + p, 0) / filteredPrices8.length;
console.log(result8);

// Output: 80


/**
 * PROBLEM 9: Reverse array, filter evens, then double
 * - Input:  [1, 2, 3, 4, 5, 6]
 * - Output: [12, 8, 4]
*/

const arr9 = [1, 2, 3, 4, 5, 6];
const result9 = arr9
  .reverse()
  .filter((n) => n % 2 === 0)
  .map((n) => n * 2);
console.log(result9);

// Output: [12, 8, 4]


/**
 * PROBLEM 10: Flatten, filter > 5, sort descending, then take first 2
 * - Input:  [[1, 6], [3, 8], [2, 9, 4]]
 * - Output: [9, 8]
*/

const arr10 = [[1, 6], [3, 8], [2, 9, 4]];
const result10 = arr10
  .flat()
  .filter((n) => n > 5)
  .sort((a, b) => b - a)
  .slice(0, 2);
console.log(result10);

// Output: [9, 8]


/**
 * PROBLEM 11: Split string, filter words length > 2, sort by length, join
 * - Input:  "a quick brown fox jumps over"
 * - Output: "fox over jumps quick brown"
*/

const str11 = "a quick brown fox jumps over";
const result11 = str11
  .split(' ')
  .filter((w) => w.length > 2)
  .sort((a, b) => a.length - b.length)
  .join(' ');
console.log(result11);

// Output: "fox over jumps quick brown"


/**
 * PROBLEM 12: Map to add index, filter even index+value sum, extract values
 * - Input:  [10, 20, 30, 40, 50]
 * - Output: [10, 30, 50] (index 0: 10+0=even, index 2: 30+2=even, index 4: 50+4=even)
*/

const arr12 = [10, 20, 30, 40, 50];
const result12 = arr12
  .map((val, i) => ({ val, sum: val + i }))
  .filter((item) => item.sum % 2 === 0)
  .map((item) => item.val);
console.log(result12);

// Output: [10, 30, 50]


/**
 * PROBLEM 13: Extract names, filter length > 3, sort reverse alphabetical, join
 * - Input:  [{name: 'Bob'}, {name: 'Alice'}, {name: 'Charlie'}, {name: 'Jo'}]
 * - Output: "Charlie,Bob,Alice"
*/

const arr13 = [{ name: 'Bob' }, { name: 'Alice' }, { name: 'Charlie' }, { name: 'Jo' }];
const result13 = arr13
  .map((u) => u.name)
  .filter((n) => n.length > 3)
  .sort((a, b) => b.localeCompare(a))
  .join(',');
console.log(result13);

// Output: "Charlie,Bob,Alice"


/**
 * PROBLEM 14: Filter negative, map to absolute, then sum
 * - Input:  [-5, 3, -2, 8, -1]
 * - Output: 8 (5 + 2 + 1)
*/

const arr14 = [-5, 3, -2, 8, -1];
const result14 = arr14
  .filter((n) => n < 0)
  .map((n) => Math.abs(n))
  .reduce((sum, n) => sum + n, 0);
console.log(result14);

// Output: 8


/**
 * PROBLEM 15: Map to uppercase, filter containing 'A', sort, join with '-'
 * - Input:  ['apple', 'banana', 'apricot', 'cherry', 'avocado']
 * - Output: "APPLE-APRICOT-AVOCADO"
*/

const arr15 = ['apple', 'banana', 'apricot', 'cherry', 'avocado'];
const result15 = arr15
  .map((s) => s.toUpperCase())
  .filter((s) => s.includes('A'))
  .sort()
  .join('-');
console.log(result15);

// Output: "APPLE-APRICOT-AVOCADO"


/**
 * PROBLEM 16: Filter odd, map to square root (of square), then reduce to max
 * - Input:  [1, 4, 9, 16, 25]
 * - Output: 5 (sqrt(25))
*/

const arr16 = [1, 4, 9, 16, 25];
const result16 = arr16
  .filter((n) => n % 2 !== 0)
  .map((n) => Math.sqrt(n))
  .reduce((max, n) => Math.max(max, n), -Infinity);
console.log(result16);

// Output: 5


/**
 * PROBLEM 17: Map to extract scores, filter even scores, sort ascending, take first 2
 * - Input:  [{name: 'A', score: 45}, {name: 'B', score: 92}, {name: 'C', score: 70}, {name: 'D', score: 88}]
 * - Output: [70, 88]
*/

const arr17 = [
  { name: 'A', score: 45 },
  { name: 'B', score: 92 },
  { name: 'C', score: 70 },
  { name: 'D', score: 88 },
];
const result17 = arr17
  .map((s) => s.score)
  .filter((s) => s % 2 === 0)
  .sort((a, b) => a - b)
  .slice(0, 2);
console.log(result17);

// Output: [70, 88]


/**
 * PROBLEM 18: Concat two arrays, dedupe, filter > 25, sort descending
 * - Input:  [10, 30, 50] and [20, 30, 40, 60]
 * - Output: [60, 50, 40, 30]
*/

const arr18a = [10, 30, 50];
const arr18b = [20, 30, 40, 60];
const result18 = [...new Set([...arr18a, ...arr18b])]
  .filter((n) => n > 25)
  .sort((a, b) => b - a);
console.log(result18);

// Output: [60, 50, 40, 30]


/**
 * PROBLEM 19: Map to extract names, filter containing 'e', sort by length, then join
 * - Input:  [{name: 'Alice'}, {name: 'Bob'}, {name: 'Eve'}, {name: 'Charlie'}]
 * - Output: "Eve,Alice,Charlie"
*/

const arr19 = [{ name: 'Alice' }, { name: 'Bob' }, { name: 'Eve' }, { name: 'Charlie' }];
const result19 = arr19
  .map((u) => u.name)
  .filter((n) => n.toLowerCase().includes('e'))
  .sort((a, b) => a.length - b.length)
  .join(',');
console.log(result19);

// Output: "Eve,Alice,Charlie"


/**
 * PROBLEM 20: Map to numbers, filter NaN, then sum
 * - Input:  ["10", "20", "abc", "30", "xyz"]
 * - Output: 60
*/

const arr20 = ["10", "20", "abc", "30", "xyz"];
const result20 = arr20
  .map(Number)
  .filter((n) => !Number.isNaN(n))
  .reduce((sum, n) => sum + n, 0);
console.log(result20);

// Output: 60


/**
 * PROBLEM 21: Filter users with age > 20, map to names, sort descending, join
 * - Input:  [{name: 'A', age: 25}, {name: 'B', age: 18}, {name: 'C', age: 30}, {name: 'D', age: 22}]
 * - Output: "C,D,A"
*/

const arr21 = [
  { name: 'A', age: 25 },
  { name: 'B', age: 18 },
  { name: 'C', age: 30 },
  { name: 'D', age: 22 },
];
const result21 = arr21
  .filter((u) => u.age > 20)
  .map((u) => u.name)
  .sort((a, b) => b.localeCompare(a))
  .join(',');
console.log(result21);

// Output: "C,D,A"


/**
 * PROBLEM 22: Map to double, filter > 10, sort ascending, slice last 2, sum
 * - Input:  [3, 7, 1, 8, 5, 2]
 * - Output: 26 (14 + 12... wait: doubles: 6,14,2,16,10,4. Filter >10: 14,16. Sort: 14,16. Last 2: 14,16. Sum: 30)
*/

const arr22 = [3, 7, 1, 8, 5, 2];
const result22 = arr22
  .map((n) => n * 2)
  .filter((n) => n > 10)
  .sort((a, b) => a - b)
  .slice(-2)
  .reduce((sum, n) => sum + n, 0);
console.log(result22);

// Output: 30


/**
 * PROBLEM 23: Split string, map to lengths, filter > 3, then find max
 * - Input:  "the quick brown fox jumps over lazy dog"
 * - Output: 5 (jumps)
*/

const str23 = "the quick brown fox jumps over lazy dog";
const result23 = str23
  .split(' ')
  .map((w) => w.length)
  .filter((len) => len > 3)
  .reduce((max, len) => Math.max(max, len), -Infinity);
console.log(result23);

// Output: 5


/**
 * PROBLEM 24: Flatten, map to square, filter < 50, then sort ascending
 * - Input:  [[2, 4], [6, 8], [1, 3]]
 * - Output: [1, 4, 9, 16, 36] (squares: 4,16,36,64,1,9. Filter <50: 4,16,36,1,9. Sort: 1,4,9,16,36)
*/

const arr24 = [[2, 4], [6, 8], [1, 3]];
const result24 = arr24
  .flat()
  .map((n) => n * n)
  .filter((n) => n < 50)
  .sort((a, b) => a - b);
console.log(result24);

// Output: [1, 4, 9, 16, 36]


/**
 * PROBLEM 25: Filter objects with price < 100, map to add tax, then sum
 * - Input:  [{price: 50}, {price: 200}, {price: 80}, {price: 30}]
 * - Output: 176 ( (50+50*0.1) + (80+80*0.1) + (30+30*0.1) = 55 + 88 + 33 )
*/

const arr25 = [{ price: 50 }, { price: 200 }, { price: 80 }, { price: 30 }];
const result25 = arr25
  .filter((p) => p.price < 100)
  .map((p) => p.price * 1.1)
  .reduce((sum, p) => sum + p, 0);
console.log(result25);

// Output: 176


/**
 * PROBLEM 26: Map to extract first char, filter vowels, sort, join
 * - Input:  ['apple', 'mango', 'avocado', 'banana', 'apricot']
 * - Output: "a,a,a"
*/

const arr26 = ['apple', 'mango', 'avocado', 'banana', 'apricot'];
const result26 = arr26
  .map((s) => s[0])
  .filter((c) => 'aeiouAEIOU'.includes(c))
  .sort()
  .join(',');
console.log(result26);

// Output: "a,a,a"


/**
 * PROBLEM 27: Filter numbers, map to check even/odd, then count evens
 * - Input:  [1, 'a', 2, true, 3, 'b', 4, 5, null, 6]
 * - Output: 3 (2, 4, 6 are even)
*/

const arr27 = [1, 'a', 2, true, 3, 'b', 4, 5, null, 6];
const result27 = arr27
  .filter((n) => typeof n === 'number')
  .map((n) => (n % 2 === 0 ? 'even' : 'odd'))
  .filter((s) => s === 'even').length;
console.log(result27);

// Output: 3


/**
 * PROBLEM 28: Map to extract usernames, filter length > 4, sort, then reverse
 * - Input:  [{user: 'alice'}, {user: 'bob'}, {user: 'charlie'}, {user: 'dee'}, {user: 'eve'}]
 * - Output: ['charlie', 'alice']
*/

const arr28 = [{ user: 'alice' }, { user: 'bob' }, { user: 'charlie' }, { user: 'dee' }, { user: 'eve' }];
const result28 = arr28
  .map((u) => u.user)
  .filter((u) => u.length > 4)
  .sort()
  .reverse();
console.log(result28);

// Output: ['charlie', 'alice']


/**
 * PROBLEM 29: Filter active, map to age, sort descending, then reduce to sum
 * - Input:  [{name: 'A', age: 30, active: true}, {name: 'B', age: 20, active: false}, {name: 'C', age: 25, active: true}]
 * - Output: 55 (30 + 25)
*/

const arr29 = [
  { name: 'A', age: 30, active: true },
  { name: 'B', age: 20, active: false },
  { name: 'C', age: 25, active: true },
];
const result29 = arr29
  .filter((u) => u.active)
  .map((u) => u.age)
  .sort((a, b) => b - a)
  .reduce((sum, age) => sum + age, 0);
console.log(result29);

// Output: 55


/**
 * PROBLEM 30: Full pipeline — filter, map, sort, slice, reduce
 * - Input:  [{product: 'A', price: 500}, {product: 'B', price: 200}, {product: 'C', price: 800}, {product: 'D', price: 100}, {product: 'E', price: 300}]
 * - Steps:  Filter price > 150, extract prices, sort descending, take top 3, sum them
 * - Output: 1600 (800 + 500 + 300)
*/

const arr30 = [
  { product: 'A', price: 500 },
  { product: 'B', price: 200 },
  { product: 'C', price: 800 },
  { product: 'D', price: 100 },
  { product: 'E', price: 300 },
];
const result30 = arr30
  .filter((p) => p.price > 150)
  .map((p) => p.price)
  .sort((a, b) => b - a)
  .slice(0, 3)
  .reduce((sum, price) => sum + price, 0);
console.log(result30);

// Output: 1600
```
