```python
/**
 * PROBLEM 1: Filter even numbers, then double them
 * - Input:  [1, 2, 3, 4, 5, 6]
 * - Output: [4, 8, 12]
*/

const arr1 = [1, 2, 3, 4, 5, 6];
const result1 = arr1.filter((n) => n % 2 === 0).map((n) => n * 2);
console.log(result1);

// Output: [4, 8, 12]


/**
 * PROBLEM 2: Filter odd numbers, then square them
 * - Input:  [1, 2, 3, 4, 5]
 * - Output: [1, 9, 25]
*/

const arr2 = [1, 2, 3, 4, 5];
const result2 = arr2.filter((n) => n % 2 !== 0).map((n) => n * n);
console.log(result2);

// Output: [1, 9, 25]


/**
 * PROBLEM 3: Map to uppercase, then filter strings longer than 3
 * - Input:  ['hi', 'hello', 'js', 'world']
 * - Output: ['HELLO', 'WORLD']
*/

const arr3 = ['hi', 'hello', 'js', 'world'];
const result3 = arr3.map((s) => s.toUpperCase()).filter((s) => s.length > 3);
console.log(result3);

// Output: ['HELLO', 'WORLD']


/**
 * PROBLEM 4: Sort numbers ascending, then get first 3
 * - Input:  [5, 2, 8, 1, 9, 3]
 * - Output: [1, 2, 3]
*/

const arr4 = [5, 2, 8, 1, 9, 3];
const result4 = arr4.sort((a, b) => a - b).slice(0, 3);
console.log(result4);

// Output: [1, 2, 3]


/**
 * PROBLEM 5: Sort descending, then get top 2
 * - Input:  [10, 50, 30, 20, 40]
 * - Output: [50, 40]
*/

const arr5 = [10, 50, 30, 20, 40];
const result5 = arr5.sort((a, b) => b - a).slice(0, 2);
console.log(result5);

// Output: [50, 40]


/**
 * PROBLEM 6: Extract names, then filter names starting with 'A'
 * - Input:  [{name: 'Alice'}, {name: 'Bob'}, {name: 'Anna'}]
 * - Output: ['Alice', 'Anna']
*/

const arr6 = [{ name: 'Alice' }, { name: 'Bob' }, { name: 'Anna' }];
const result6 = arr6.map((u) => u.name).filter((name) => name.startsWith('A'));
console.log(result6);

// Output: ['Alice', 'Anna']


/**
 * PROBLEM 7: Filter active users, then extract their names
 * - Input:  [{name: 'A', active: true}, {name: 'B', active: false}, {name: 'C', active: true}]
 * - Output: ['A', 'C']
*/

const arr7 = [{ name: 'A', active: true }, { name: 'B', active: false }, { name: 'C', active: true }];
const result7 = arr7.filter((u) => u.active).map((u) => u.name);
console.log(result7);

// Output: ['A', 'C']


/**
 * PROBLEM 8: Filter passing students (score >= 40), then extract names
 * - Input:  [{name: 'A', score: 85}, {name: 'B', score: 35}, {name: 'C', score: 60}]
 * - Output: ['A', 'C']
*/

const arr8 = [{ name: 'A', score: 85 }, { name: 'B', score: 35 }, { name: 'C', score: 60 }];
const result8 = arr8.filter((s) => s.score >= 40).map((s) => s.name);
console.log(result8);

// Output: ['A', 'C']


/**
 * PROBLEM 9: Filter prices > 100, then sum them
 * - Input:  [50, 120, 80, 200, 30]
 * - Output: 320
*/

const arr9 = [50, 120, 80, 200, 30];
const result9 = arr9.filter((p) => p > 100).reduce((sum, p) => sum + p, 0);
console.log(result9);

// Output: 320


/**
 * PROBLEM 10: Map to double, then sum all
 * - Input:  [1, 2, 3, 4]
 * - Output: 20
*/

const arr10 = [1, 2, 3, 4];
const result10 = arr10.map((n) => n * 2).reduce((sum, n) => sum + n, 0);
console.log(result10);

// Output: 20


/**
 * PROBLEM 11: Filter even, map to square, then sum
 * - Input:  [1, 2, 3, 4, 5, 6]
 * - Output: 56 (4 + 16 + 36)
*/

const arr11 = [1, 2, 3, 4, 5, 6];
const result11 = arr11
  .filter((n) => n % 2 === 0)
  .map((n) => n * n)
  .reduce((sum, n) => sum + n, 0);
console.log(result11);

// Output: 56


/**
 * PROBLEM 12: Flatten nested array, then filter > 2
 * - Input:  [[1, 2], [3, 4], [5, 6]]
 * - Output: [3, 4, 5, 6]
*/

const arr12 = [[1, 2], [3, 4], [5, 6]];
const result12 = arr12.flat().filter((n) => n > 2);
console.log(result12);

// Output: [3, 4, 5, 6]


/**
 * PROBLEM 13: Flatten, then sort descending
 * - Input:  [[3, 1], [6, 2], [5, 4]]
 * - Output: [6, 5, 4, 3, 2, 1]
*/

const arr13 = [[3, 1], [6, 2], [5, 4]];
const result13 = arr13.flat().sort((a, b) => b - a);
console.log(result13);

// Output: [6, 5, 4, 3, 2, 1]


/**
 * PROBLEM 14: Sort by score descending, then extract names
 * - Input:  [{name: 'A', score: 85}, {name: 'B', score: 92}, {name: 'C', score: 70}]
 * - Output: ['B', 'A', 'C']
*/

const arr14 = [{ name: 'A', score: 85 }, { name: 'B', score: 92 }, { name: 'C', score: 70 }];
const result14 = arr14.sort((a, b) => b.score - a.score).map((s) => s.name);
console.log(result14);

// Output: ['B', 'A', 'C']


/**
 * PROBLEM 15: Sort by age ascending, then get top 2 names
 * - Input:  [{name: 'A', age: 30}, {name: 'B', age: 20}, {name: 'C', age: 25}]
 * - Output: ['B', 'C']
*/

const arr15 = [{ name: 'A', age: 30 }, { name: 'B', age: 20 }, { name: 'C', age: 25 }];
const result15 = arr15.sort((a, b) => a.age - b.age).slice(0, 2).map((u) => u.name);
console.log(result15);

// Output: ['B', 'C']


/**
 * PROBLEM 16: Map to lengths, then filter lengths > 3
 * - Input:  ['hi', 'hello', 'js', 'world']
 * - Output: [5, 5]
*/

const arr16 = ['hi', 'hello', 'js', 'world'];
const result16 = arr16.map((s) => s.length).filter((len) => len > 3);
console.log(result16);

// Output: [5, 5]


/**
 * PROBLEM 17: Split sentence to words, filter words > 3 chars, then uppercase
 * - Input:  "the quick brown fox jumps"
 * - Output: ['QUICK', 'BROWN', 'JUMPS']
*/

const str17 = "the quick brown fox jumps";
const result17 = str17
  .split(" ")
  .filter((w) => w.length > 3)
  .map((w) => w.toUpperCase());
console.log(result17);

// Output: ['QUICK', 'BROWN', 'JUMPS']


/**
 * PROBLEM 18: Split string, reverse words, join back
 * - Input:  "hello world javascript"
 * - Output: "javascript world hello"
*/

const str18 = "hello world javascript";
const result18 = str18.split(" ").reverse().join(" ");
console.log(result18);

// Output: "javascript world hello"


/**
 * PROBLEM 19: Map to numbers, filter > 10, then sum
 * - Input:  ["5", "15", "8", "20", "3"]
 * - Output: 35 (15 + 20)
*/

const arr19 = ["5", "15", "8", "20", "3"];
const result19 = arr19
  .map(Number)
  .filter((n) => n > 10)
  .reduce((sum, n) => sum + n, 0);
console.log(result19);

// Output: 35


/**
 * PROBLEM 20: Map to extract ages, filter adults (>= 18), then find average
 * - Input:  [{name: 'A', age: 25}, {name: 'B', age: 15}, {name: 'C', age: 30}]
 * - Output: 27.5
*/

const arr20 = [{ name: 'A', age: 25 }, { name: 'B', age: 15 }, { name: 'C', age: 30 }];
const adultAges20 = arr20.map((u) => u.age).filter((age) => age >= 18);
const avg20 = adultAges20.reduce((sum, age) => sum + age, 0) / adultAges20.length;
console.log(avg20);

// Output: 27.5


/**
 * PROBLEM 21: Filter positive numbers, sort ascending, then get first 2
 * - Input:  [-3, 5, -1, 8, 2, -6]
 * - Output: [2, 5]
*/

const arr21 = [-3, 5, -1, 8, 2, -6];
const result21 = arr21
  .filter((n) => n > 0)
  .sort((a, b) => a - b)
  .slice(0, 2);
console.log(result21);

// Output: [2, 5]


/**
 * PROBLEM 22: Remove duplicates, then sort ascending
 * - Input:  [3, 1, 3, 2, 1, 4, 2, 5]
 * - Output: [1, 2, 3, 4, 5]
*/

const arr22 = [3, 1, 3, 2, 1, 4, 2, 5];
const result22 = [...new Set(arr22)].sort((a, b) => a - b);
console.log(result22);

// Output: [1, 2, 3, 4, 5]


/**
 * PROBLEM 23: Map to lowercase, filter unique, then sort alphabetically
 * - Input:  ['Banana', 'apple', 'banana', 'Apple', 'cherry']
 * - Output: ['apple', 'banana', 'cherry']
*/

const arr23 = ['Banana', 'apple', 'banana', 'Apple', 'cherry'];
const result23 = [...new Set(arr23.map((s) => s.toLowerCase()))].sort();
console.log(result23);

// Output: ['apple', 'banana', 'cherry']


/**
 * PROBLEM 24: Filter products in stock, map to prices, then sum
 * - Input:  [{price: 100, inStock: true}, {price: 200, inStock: false}, {price: 50, inStock: true}]
 * - Output: 150
*/

const arr24 = [
  { price: 100, inStock: true },
  { price: 200, inStock: false },
  { price: 50, inStock: true },
];
const result24 = arr24
  .filter((p) => p.inStock)
  .map((p) => p.price)
  .reduce((sum, price) => sum + price, 0);
console.log(result24);

// Output: 150


/**
 * PROBLEM 25: Map to extract scores, filter > 50, then find max
 * - Input:  [{name: 'A', score: 85}, {name: 'B', score: 30}, {name: 'C', score: 70}]
 * - Output: 85
*/

const arr25 = [{ name: 'A', score: 85 }, { name: 'B', score: 30 }, { name: 'C', score: 70 }];
const result25 = arr25
  .map((s) => s.score)
  .filter((score) => score > 50)
  .reduce((max, score) => Math.max(max, score), -Infinity);
console.log(result25);

// Output: 85


/**
 * PROBLEM 26: Reverse string, split to chars, filter vowels, join back
 * - Input:  "hello world"
 * - Output: "oole" (reversed: "dlrow olleh", vowels: o, o, e)
*/

const str26 = "hello world";
const result26 = str26
  .split("")
  .reverse()
  .filter((c) => "aeiouAEIOU".includes(c))
  .join("");
console.log(result26);

// Output: "oole"


/**
 * PROBLEM 27: Flatten deeply nested array, filter even, then multiply all
 * - Input:  [[1, 2], [3, [4, 5]], [6]]
 * - Output: 48 (2 * 4 * 6)
*/

const arr27 = [[1, 2], [3, [4, 5]], [6]];
const flat27 = arr27.flat(Infinity);
const result27 = flat27
  .filter((n) => n % 2 === 0)
  .reduce((prod, n) => prod * n, 1);
console.log(result27);

// Output: 48


/**
 * PROBLEM 28: Group by city, then get city names with > 1 user
 * - Input:  [{name: 'A', city: 'Delhi'}, {name: 'B', city: 'Mumbai'}, {name: 'C', city: 'Delhi'}]
 * - Output: ['Delhi']
*/

const arr28 = [
  { name: 'A', city: 'Delhi' },
  { name: 'B', city: 'Mumbai' },
  { name: 'C', city: 'Delhi' },
];
const grouped28 = arr28.reduce((acc, u) => {
  (acc[u.city] = acc[u.city] || []).push(u);
  return acc;
}, {});
const result28 = Object.entries(grouped28)
  .filter(([city, users]) => users.length > 1)
  .map(([city]) => city);
console.log(result28);

// Output: ['Delhi']


/**
 * PROBLEM 29: Map to add 10, filter > 15, then reduce to product
 * - Input:  [1, 5, 8, 3, 10]
 * - Output: 390 (20 * 18... wait: 5+10=15(not >15), 8+10=18(>15), 10+10=20(>15). 18*20=360)
*/

const arr29 = [1, 5, 8, 3, 10];
const result29 = arr29
  .map((n) => n + 10)
  .filter((n) => n > 15)
  .reduce((prod, n) => prod * n, 1);
console.log(result29);

// Output: 360


/**
 * PROBLEM 30: Full pipeline — filter, map, sort, slice, reduce
 * - Input:  [{name: 'A', score: 45}, {name: 'B', score: 85}, {name: 'C', score: 70}, {name: 'D', score: 90}]
 * - Steps:  Filter score >= 50, extract scores, sort descending, take top 2, sum them
 * - Output: 160 (90 + 70)
*/

const arr30 = [
  { name: 'A', score: 45 },
  { name: 'B', score: 85 },
  { name: 'C', score: 70 },
  { name: 'D', score: 90 },
];
const result30 = arr30
  .filter((s) => s.score >= 50)
  .map((s) => s.score)
  .sort((a, b) => b - a)
  .slice(0, 2)
  .reduce((sum, score) => sum + score, 0);
console.log(result30);

// Output: 175 (90 + 85 = 175)
```
