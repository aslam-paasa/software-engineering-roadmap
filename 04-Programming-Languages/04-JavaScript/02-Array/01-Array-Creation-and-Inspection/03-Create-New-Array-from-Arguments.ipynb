```typescript

/**
 * Problem with Array() Constructor:
 * > Array() constructor confusing hai jab single number argument milta hai.
*/

/**
 * Solution: Array.of()
 * > Array.of() har argument ko as an element treat karta hai, chahe single 
 *   ho ya multiple.
 * > Syntax: Array.of(element1, element2, ...)
*/
```

```typescript
/**
 * Example 1: Single Number - No Confusion
 * > ✅ Array.of() - Consistent behavior
*/
const arr1 = Array.of(5);       // Creates [5]
const arr2 = Array.of(5, 10);   // Creates [5, 10]
const arr3 = Array.of(1, 2, 3); // Creates [1, 2, 3]

console.log("arr1:", arr1); // [5]
console.log("arr2:", arr2); // [5, 10] 
console.log("arr3:", arr3); // [1, 2, 3]
```

```typescript
/**
 * Example 2: Mixed Data Types
*/

/* ✅ Different data types ke saath */
const mixedArray = Array.of("Hello", 42, true, null, [1, 2]);
console.log("Mixed array:", mixedArray); /* ['Hello', 42, true, null, [1, 2]] */

/* Empty array */
const empty = Array.of();
console.log("Empty array:", empty); /* [] */
```

```typescript
/**
 * Example-3: Array.of() with Rest Params (...args)
 * > ...args inside a function = rest parameter
 * > It collects ALL remaining arguments into a plain Array
 * > So if ...args is already an array, why combine it with Array.of()? 
 *   Two distinct scenarios:
*/
function demo(...args) {
    console.log(args);                // already a plain Array
    console.log(Array.isArray(args)); // true
}
demo(1, 2, 3);
```

```typescript
/**
 * Scenario-1: You want ONE array containing everything as a single unit
 * > Array.of(...args) = spread args as individual elements into Array.of()
 * > Same as writing Array.of(1, 2, 3) — gives you [1, 2, 3]
*/
function collectAll(...args) {
    return Array.of(...args);
}

collectAll(10, 20, 30); // [10, 20, 30]
```

```typescript
/**
 * Scenario-2: Wrapping a single unknown value safely
 * > input could be a single value OR already an array
*/

function normalize(input) {
    return Array.isArray(input) ? input : Array.of(input);
}

/* API response scenarios */
normalize("active");            // ["active"]
normalize(["active", "pending"]) // ["active","pending"] — untouched
normalize(42);                  // [42]  ← safe even for numbers
normalize(1);                   // [1]   ← not 1 empty slot!
```

```typescript
/**
 * Key distinction to remember:
*/
function example(...args) {
    /* a. args is already [ 1, 2, 3 ] — a plain Array */
    console.log(args);

    /* b. spreads back out → Array.of(1, 2, 3) → [1, 2, 3] */
    /*    (identical result for plain Arrays)              */
    console.log(Array.of(...args));

    /* c. THIS is where it matters                */
    /*    returns a NumberList, not a plain Array */
    console.log(NumberList.of(...args));
}
```

```typescript
/**
 * Example 1: E-commerce product prices 
*/

const singleProduct = Array.of(499);                    // Single product price
const multipleProducts = Array.of(299, 499, 799, 1299); // Multiple prices

console.log("Single product price:", singleProduct);       // [499]
console.log("Multiple product prices:", multipleProducts); // [299, 499, 799, 1299]


/* Compare with Array() constructor */
const confusingArray = new Array(499);           // ❌ Creates 499 empty slots!
console.log("Confusing array:", confusingArray); // [empty × 499]
```

```typescript
/**
 * Example 1: E-Commerce - Safe Price List Creation
*/

// Product catalog with safe price arrays
const PRICE_TIERS = {
    budget: Array.of(199, 299, 399),
    mid: Array.of(499, 699, 899),
    premium: Array.of(1299, 1999, 2999),
    luxury: Array.of(9999),           // ← single item, safely [9999] not 9999 empty slots
};

function getPricesInRange(tier, max) {
    return PRICE_TIERS[tier].filter(price => price <= max);
}

console.log(getPricesInRange("mid", 700));  // [499, 699]
console.log(getPricesInRange("luxury", 20000)); // [9999]
```

```typescript
/**
 * Example 3: Building dynamic query conditions (backend/ORM usage)
*/

// When building SQL-like queries dynamically — args can be 1 or many
function buildWhereClause(field, ...values) {
    const valueArray = Array.of(...values); // guaranteed array, no length confusion

    if (valueArray.length === 1) {
        return `${field} = ${valueArray[0]}`;
    }
    return `${field} IN (${valueArray.join(", ")})`;
}

console.log(buildWhereClause("status", "active"));              // status = active
console.log(buildWhereClause("status", "active", "pending"));   // status IN (active, pending)
console.log(buildWhereClause("price", 499));                     // price = 499  ✅ no bug
```

```typescript
/**
 * Array() vs Array.of() Comparison:
*/

console.log("🔍 Array() vs Array.of() Comparison:");

/* Single number */
console.log("new Array(5):", new Array(5));         // [empty × 5]
console.log("Array.of(5):", Array.of(5));           // [5]

/* Multiple numbers */
console.log("new Array(5, 10):", new Array(5, 10)); // [5, 10]
console.log("Array.of(5, 10):", Array.of(5, 10));   // [5, 10]

/* Empty */
console.log("new Array():", new Array());          // []
console.log("Array.of():", Array.of());            // []
```
