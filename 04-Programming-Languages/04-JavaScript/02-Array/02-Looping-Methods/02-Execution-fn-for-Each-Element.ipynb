```typescript
/**
 * Step 1: forEach() - Simple Execution Har Element Par
 * > forEach() har array element ke liye ek function execute karta hai. 
 * > Kuch return nahi karta.
 * 
 * > Syntax: 
 *    array.forEach(function(currentValue, index, array) {
 *        // code to execute
 *    });
*/
```

```typescript
/**
 * Example 1: Basic forEach - Console Log
*/

const fruits = ["Apple", "Banana", "Orange", "Mango"];

/* Har fruit ko print karo */
console.log("Fruits List:");
fruits.forEach(function(fruit, index) {
    console.log(`${index + 1}. ${fruit}`);
});

/* Output:   */
/* 1. Apple  */
/* 2. Banana */
/* 3. Orange */
/* 4. Mango  */
```

```typescript
/**
 * Example 2: Real World - Shopping Cart
*/

const cartItems = [
    { name: "Laptop", price: 50000, quantity: 1 },
    { name: "Mouse", price: 1500, quantity: 2 },
    { name: "Keyboard", price: 3000, quantity: 1 }
];

let totalAmount = 0;

/* Har item ka cost calculate karo */
console.log("Shopping Cart Breakdown:");
cartItems.forEach(function(item) {
    const itemTotal = item.price * item.quantity;
    totalAmount += itemTotal;
    console.log(`${item.name} - ₹${item.price} × ${item.quantity} = ₹${itemTotal}`);
});

console.log(`Total Amount: ₹${totalAmount}`);
```

```typescript
/**
 * PROBLEM 13: Flatten a nested array (1 level)
 * - Input:  [1, [2, 3], [4, 5]]
 * - Output: [1, 2, 3, 4, 5]
*/

const arr13 = [1, [2, 3], [4, 5]];
const flattened13: number[] = [];
arr13.forEach((elem) => {
  if (Array.isArray(elem)) {
    elem.forEach((inner) => flattened13.push(inner));
  } else {
    flattened13.push(elem);
  }
});
console.log(flattened13);

// Output: [1, 2, 3, 4, 5]
```

```typescript
/**
 * PROBLEM 30: Custom forEach polyfill (with thisArg)
 * - THE most asked question in interviews
 * - Re-implement Array.prototype.forEach from scratch
 * - Signature: arr.forEach(callback, thisArg)
 * - callback receives: (currentValue, index, array)
*/

declare global {
  interface Array<T> {
    myForEach(callback: (value: T, index: number, array: T[]) => void, thisArg?: any): void;
  }
}

Array.prototype.myForEach = function <T>(this: T[], callback: (value: T, index: number, array: T[]) => void, thisArg?: any): void {
  for (let i = 0; i < this.length; i++) {
    // Skip holes in sparse arrays (like native forEach does)
    if (i in this) {
      // call() sets 'this' inside callback to thisArg
      callback.call(thisArg, this[i], i, this);
    }
  }
};

// ---- TEST 1: Basic iteration ----
const testArr1 = [10, 20, 30];
testArr1.myForEach((elem, index) => {
  console.log(`Index ${index}: ${elem}`);
});
// Output: Index 0: 10, Index 1: 20, Index 2: 30

// ---- TEST 2: With thisArg ----
const obj = { multiplier: 10 };
const testArr2 = [1, 2, 3];
testArr2.myForEach(function (elem) {
  console.log(elem * this.multiplier);
}, obj);
// Output: 10 20 30 (each on new line)

// ---- TEST 3: Sum using polyfill ----
const testArr3 = [1, 2, 3, 4, 5];
let polyfillSum = 0;
testArr3.myForEach((elem) => {
  polyfillSum += elem;
});
console.log(polyfillSum);
// Output: 15

// ---- TEST 4: Modifying original array (edge case) ----
const testArr4 = [1, 2, 3];
testArr4.myForEach((elem, index, arr) => {
  arr[index] = elem * 2;
});
console.log(testArr4);
// Output: [2, 4, 6]

// ---- TEST 5: Sparse array (has holes) ----
const sparseArr = [1, , 3];
sparseArr.myForEach((elem) => {
  console.log(elem);
});
// Output: 1 3 (skips the hole at index 1, like native forEach)
```
