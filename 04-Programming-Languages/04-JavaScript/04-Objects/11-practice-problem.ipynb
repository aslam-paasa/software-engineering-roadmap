```python
/**
 * PROBLEM 1: Create an object and access properties
 * - Input:  { name: "John", age: 30 }
 * - Output: "John", 30
*/

const obj1 = { name: "John", age: 30 };
console.log(obj1.name);   // dot notation
console.log(obj1["age"]); // bracket notation

// Output: "John", 30


/**
 * PROBLEM 2: Add a new property to an existing object
 * - Input:  { name: "John" }, add key "age" with value 30
 * - Output: { name: "John", age: 30 }
*/

const obj2 = { name: "John" };
obj2.age = 30;
console.log(obj2);

// Output: { name: "John", age: 30 }


/**
 * PROBLEM 3: Delete a property from an object
 * - Input:  { name: "John", age: 30 }, delete "age"
 * - Output: { name: "John" }
*/

const obj3 = { name: "John", age: 30 };
delete obj3.age;
console.log(obj3);

// Output: { name: "John" }


/**
 * PROBLEM 4: Check if a key exists in an object
 * - Input:  { name: "John", age: 30 }, key = "age"
 * - Output: true
*/

const obj4 = { name: "John", age: 30 };
console.log("age" in obj4);

// Output: true


/**
 * PROBLEM 5: Get all keys of an object
 * - Input:  { name: "John", age: 30, city: "Delhi" }
 * - Output: ["name", "age", "city"]
*/

const obj5 = { name: "John", age: 30, city: "Delhi" };
console.log(Object.keys(obj5));

// Output: ["name", "age", "city"]


/**
 * PROBLEM 6: Get all values of an object
 * - Input:  { name: "John", age: 30, city: "Delhi" }
 * - Output: ["John", 30, "Delhi"]
*/

const obj6 = { name: "John", age: 30, city: "Delhi" };
console.log(Object.values(obj6));

// Output: ["John", 30, "Delhi"]


/**
 * PROBLEM 7: Get all key-value pairs of an object
 * - Input:  { name: "John", age: 30 }
 * - Output: [["name", "John"], ["age", 30]]
*/

const obj7 = { name: "John", age: 30 };
console.log(Object.entries(obj7));

// Output: [["name", "John"], ["age", 30]]


/**
 * PROBLEM 8: Convert array of key-value pairs back to object
 * - Input:  [["name", "John"], ["age", 30]]
 * - Output: { name: "John", age: 30 }
*/

const entries8 = [["name", "John"], ["age", 30]];
console.log(Object.fromEntries(entries8));

// Output: { name: "John", age: 30 }


/**
 * PROBLEM 9: Iterate over object properties using for...in
 * - Input:  { name: "John", age: 30 }
 * - Output: "name: John", "age: 30"
*/

const obj9 = { name: "John", age: 30 };
for (let key in obj9) {
  console.log(`${key}: ${obj9[key]}`);
}

// Output: "name: John", "age: 30"


/**
 * PROBLEM 10: Iterate over object using Object.entries
 * - Input:  { name: "John", age: 30 }
 * - Output: "name → John", "age → 30"
*/

const obj10 = { name: "John", age: 30 };
Object.entries(obj10).forEach(([key, value]) => {
  console.log(`${key} → ${value}`);
});

// Output: "name → John", "age → 30"


/**
 * PROBLEM 11: Shallow clone an object using spread operator
 * - Input:  { name: "John", age: 30 }
 * - Output: { name: "John", age: 30 } (new reference)
*/

const original11 = { name: "John", age: 30 };
const clone11 = { ...original11 };
console.log(clone11);
console.log(clone11 === original11);

// Output: { name: "John", age: 30 }, false


/**
 * PROBLEM 12: Shallow clone using Object.assign
 * - Input:  { name: "John", age: 30 }
 * - Output: { name: "John", age: 30 } (new reference)
*/

const original12 = { name: "John", age: 30 };
const clone12 = Object.assign({}, original12);
console.log(clone12);
console.log(clone12 === original12);

// Output: { name: "John", age: 30 }, false


/**
 * PROBLEM 13: Merge two objects (spread operator, right-most wins)
 * - Input:  { a: 1, b: 2 } and { b: 3, c: 4 }
 * - Output: { a: 1, b: 3, c: 4 }
*/

const obj13a = { a: 1, b: 2 };
const obj13b = { b: 3, c: 4 };
const merged13 = { ...obj13a, ...obj13b };
console.log(merged13);

// Output: { a: 1, b: 3, c: 4 }


/**
 * PROBLEM 14: Merge two objects using Object.assign
 * - Input:  { a: 1 } and { b: 2 }
 * - Output: { a: 1, b: 2 }
*/

const obj14a = { a: 1 };
const obj14b = { b: 2 };
const merged14 = Object.assign({}, obj14a, obj14b);
console.log(merged14);

// Output: { a: 1, b: 2 }


/**
 * PROBLEM 15: Destructure object properties into variables
 * - Input:  { name: "John", age: 30, city: "Delhi" }
 * - Output: "John", 30, "Delhi"
*/

const obj15 = { name: "John", age: 30, city: "Delhi" };
const { name, age, city } = obj15;
console.log(name, age, city);

// Output: "John" 30 "Delhi"


/**
 * PROBLEM 16: Destructure with renamed variables
 * - Input:  { name: "John", age: 30 }
 * - Output: "John", 30 (as variables n and a)
*/

const obj16 = { name: "John", age: 30 };
const { name: n, age: a } = obj16;
console.log(n, a);

// Output: "John" 30


/**
 * PROBLEM 17: Destructure with default values
 * - Input:  { name: "John" }, age defaults to 25
 * - Output: "John", 25
*/

const obj17 = { name: "John" };
const { name: name17, age: age17 = 25 } = obj17;
console.log(name17, age17);

// Output: "John" 25


/**
 * PROBLEM 18: Destructure nested objects
 * - Input:  { user: { name: "John", address: { city: "Delhi" } } }
 * - Output: "John", "Delhi"
*/

const obj18 = { user: { name: "John", address: { city: "Delhi" } } };
const { user: { name: name18, address: { city: city18 } } } = obj18;
console.log(name18, city18);

// Output: "John" "Delhi"


/**
 * PROBLEM 19: Count the number of properties in an object
 * - Input:  { name: "John", age: 30, city: "Delhi" }
 * - Output: 3
*/

const obj19 = { name: "John", age: 30, city: "Delhi" };
console.log(Object.keys(obj19).length);

// Output: 3


/**
 * PROBLEM 20: Deep clone an object using JSON
 * - Input:  { name: "John", address: { city: "Delhi" } }
 * - Output: Deep clone (new nested references)
*/

const original20 = { name: "John", address: { city: "Delhi" } };
const deepClone20 = JSON.parse(JSON.stringify(original20));
deepClone20.address.city = "Mumbai";
console.log(original20.address.city);
console.log(deepClone20.address.city);

// Output: "Delhi" (original unchanged), "Mumbai"


/**
 * PROBLEM 21: Demonstrate shallow copy pitfall with nested objects
 * - Input:  { name: "John", address: { city: "Delhi" } }
 * - Output: "Mumbai", "Mumbai" (both changed — shallow copy shares reference)
*/

const original21 = { name: "John", address: { city: "Delhi" } };
const shallowClone21 = { ...original21 };
shallowClone21.address.city = "Mumbai";
console.log(original21.address.city);
console.log(shallowClone21.address.city);

// Output: "Mumbai", "Mumbai" (shallow copy shares nested reference)


/**
 * PROBLEM 22: Freeze an object (prevent modifications)
 * - Input:  { name: "John" }, try to change name to "Jane"
 * - Output: { name: "John" } (unchanged, mutation silently fails in non-strict mode)
*/

const obj22 = { name: "John" };
Object.freeze(obj22);
obj22.name = "Jane";
console.log(obj22);

// Output: { name: "John" }


/**
 * PROBLEM 23: Seal an object (allow value changes, no add/delete)
 * - Input:  { name: "John", age: 30 }
 * - Output: { name: "Jane", age: 30 } (can modify, but can't add/delete)
*/

const obj23 = { name: "John", age: 30 };
Object.seal(obj23);
obj23.name = "Jane";
obj23.city = "Delhi"; // won't be added
delete obj23.age;      // won't be deleted
console.log(obj23);

// Output: { name: "Jane", age: 30 }


/**
 * PROBLEM 24: Check if an object is frozen
 * - Input:  Object.freeze({ name: "John" })
 * - Output: true
*/

const obj24 = Object.freeze({ name: "John" });
console.log(Object.isFrozen(obj24));

// Output: true


/**
 * PROBLEM 25: Swap key and value of an object
 * - Input:  { a: 1, b: 2, c: 3 }
 * - Output: { 1: "a", 2: "b", 3: "c" }
*/

const obj25 = { a: 1, b: 2, c: 3 };
const swapped25 = Object.fromEntries(
  Object.entries(obj25).map(([key, value]) => [value, key])
);
console.log(swapped25);

// Output: { '1': 'a', '2': 'b', '3': 'c' }


/**
 * PROBLEM 26: Filter object properties by value
 * - Input:  { a: 1, b: 2, c: 3, d: 4 }, keep values > 2
 * - Output: { c: 3, d: 4 }
*/

const obj26 = { a: 1, b: 2, c: 3, d: 4 };
const filtered26 = Object.fromEntries(
  Object.entries(obj26).filter(([key, value]) => value > 2)
);
console.log(filtered26);

// Output: { c: 3, d: 4 }


/**
 * PROBLEM 27: Map over object values and transform them
 * - Input:  { a: 1, b: 2, c: 3 }, double each value
 * - Output: { a: 2, b: 4, c: 6 }
*/

const obj27 = { a: 1, b: 2, c: 3 };
const doubled27 = Object.fromEntries(
  Object.entries(obj27).map(([key, value]) => [key, value * 2])
);
console.log(doubled27);

// Output: { a: 2, b: 4, c: 6 }


/**
 * PROBLEM 28: Deep merge two objects (recursively)
 * - Input:  { a: 1, b: { x: 1, y: 2 } } and { b: { y: 3, z: 4 }, c: 5 }
 * - Output: { a: 1, b: { x: 1, y: 3, z: 4 }, c: 5 }
*/

const deepMerge28 = (obj1, obj2) => {
  const result = { ...obj1 };
  for (let key in obj2) {
    if (typeof obj2[key] === "object" && obj2[key] !== null && !Array.isArray(obj2[key])) {
      result[key] = deepMerge28(result[key] || {}, obj2[key]);
    } else {
      result[key] = obj2[key];
    }
  }
  return result;
};
const obj28a = { a: 1, b: { x: 1, y: 2 } };
const obj28b = { b: { y: 3, z: 4 }, c: 5 };
console.log(deepMerge28(obj28a, obj28b));

// Output: { a: 1, b: { x: 1, y: 3, z: 4 }, c: 5 }


/**
 * PROBLEM 29: Check if two objects are deeply equal
 * - Input:  { a: 1, b: { c: 2 } } and { a: 1, b: { c: 2 } }
 * - Output: true
*/

const deepEqual29 = (obj1, obj2) => {
  if (obj1 === obj2) return true;
  if (typeof obj1 !== "object" || typeof obj2 !== "object" || obj1 === null || obj2 === null) {
    return false;
  }
  const keys1 = Object.keys(obj1);
  const keys2 = Object.keys(obj2);
  if (keys1.length !== keys2.length) return false;
  return keys1.every((key) => deepEqual29(obj1[key], obj2[key]));
};
const obj29a = { a: 1, b: { c: 2 } };
const obj29b = { a: 1, b: { c: 2 } };
console.log(deepEqual29(obj29a, obj29b));

// Output: true


/**
 * PROBLEM 30: Group array of objects by a property
 * - Input:  [{ name: "A", city: "Delhi" }, { name: "B", city: "Mumbai" }, { name: "C", city: "Delhi" }]
 * - Output: { Delhi: [{...A}, {...C}], Mumbai: [{...B}] }
*/

const arr30 = [
  { name: "A", city: "Delhi" },
  { name: "B", city: "Mumbai" },
  { name: "C", city: "Delhi" },
];
const grouped30 = arr30.reduce((acc, elem) => {
  if (!acc[elem.city]) {
    acc[elem.city] = [];
  }
  acc[elem.city].push(elem);
  return acc;
}, {});
console.log(grouped30);

// Output: { Delhi: [{name: "A", city: "Delhi"}, {name: "C", city: "Delhi"}], Mumbai: [{name: "B", city: "Mumbai"}] }
```
