```python
/**
 * JavaScript Numbers
*/

/**
 * 1. Basics of Numbers in JS:
 *    - JS has only one number type: Number
 *    - Numbers can be:
 *      > Integers: 42
 *      > Floats(decimals): 3.14
 *      > Scientific Notation: 1.5e3 (i.e, 1500)
*/

let num1 = 10;
let num2 = 3.14;
let sci = 2.5e6;


/**
 * 2. Special Number Values:
 *    a. Infinity: Larger than any other number
 *    b. -Infinity: Smaller than any other number
 *    c. NaN      : "Not-a-Number" (invalid number result)
 *    d. Number.MAX_VALUE: Largest possible number
 *    e. Number.MIN_VALUE: Smallest possible number > 0
 *    f. Number.MAX_SAFE_INTEGER:  2^53 - 1  (largest safe integer)
 *    g. Number.MIN_SAFE_INTEGER: -(2^53 - 1)
*/

console.log(1 / 0);       // Infinity
console.log(-1 / 0);      // -Infinity
console.log("abc" * 2);   // NaN


/**
 * 3. Number Conversion
 *    a. Number(value)    : Converts to number
 *    b. parseInt(value)  : Parses integer from string
 *    c. parseFloat(value): Parses float from string
*/

Number("42");         // 42
parseInt("42px");     // 42
parseFloat("3.14abc") // 3.14
    + "123";               // 123


/**
 * 4. Checking Numbers
*/

isNaN(NaN);                             // true
Number.isNaN(NaN);                      // true
Number.isNaN("abc");                    // false

isFinite(100);                          // true
Number.isFinite(100);                   // true
Number.isFinite("100");                 // false

Number.isInteger(42);                   // true
Number.isSafeInteger(9007199254740991); // true


/**
 * Number Methods:
 * 1. toFixed(n)
 * 2. toPrecision(n)
 * 3. toExponential(n)
 * 4. toString(base)
 * 5. valueof()
*/

/* 1. toFixed(n): Returns a string with n digits after the decimal */
let n1 = 3.14159;
n1.toFixed(2);  // "3.14"


/* 2. toPrecision(n): Returns a string with a total of n significant digits */
let n2 = 123.456;
n2.toPrecision(4); // "123.5"


/* 3. toExponential(n): Converts to exponential notation */
let n3 = 12345;
n3.toExponential(2); // "1.23e+4"


/* 4. toString(base): Converts number to string in given base (2 to 36) */
(255).toString(2);  // "11111111"
(255).toString(16); // "ff"


/* 5. valueOf(): Returns the primitive value of a number object */
let n = new Number(5);
n.valueOf(); // 5


/**
 * Math Methods (commonly used):
*/
Math.round(4.7);      // 5
Math.floor(4.7);      // 4
Math.ceil(4.1);       // 5
Math.trunc(4.9);      // 4
Math.abs(-7);         // 7
Math.sqrt(16);        // 4
Math.pow(2, 3);       // 8
Math.random();        // 0 <= x < 1
Math.min(1, 3, -5);   // -5
Math.max(1, 3, -5);   // 3


/**
 * BigInt (for large integers):
 * > Used when numbers exceed 2^53 - 1
*/
const big = 1234567890123456789012345678901234567890n;
typeof big; // "bigint"


/**
 * Common Issues with Numbers:
 * > Floating Point Precision: 
 *   0.1 + 0.2 === 0.3; // false
 * 
 * > Use Number.EPSILON to compare safely: 
 *   Math.abs(0.1 + 0.2 - 0.3) < Number.EPSILON; // true
*/


/**
 * Useful Constants:
*/
Number.EPSILON             // Smallest difference between 1 and next number
Number.MAX_SAFE_INTEGER    // 9007199254740991
Number.MIN_SAFE_INTEGER    // -9007199254740991
Number.POSITIVE_INFINITY
Number.NEGATIVE_INFINITY
Number.NaN



/**
 * Summary:
 * 1. Check NaN                : isNaN(), Number.isNaN()
 * 2. Check finite number      : Number.isFinite()
 * 3. Convert to Number        : Number(), +value(), parseInt()
 * 4. Fix decimal places       : .toFixed(n)
 * 5. Format significant digits: .toPrecision(n)
 * 6. Convert to base          : .toString(base)
 * 7. Random Number            : Math.random()
 * 8. Handle large integers    : BigInt
*/
```

```python
/**
 * PROBLEM 1: Convert string to number
 * - Input:  "42"
 * - Output: 42 (number type)
*/

const str1 = "42";
const num1 = Number(str1);
console.log(typeof num1, num1);

// Output: "number" 42


/**
 * PROBLEM 2: Convert string to number using unary plus
 * - Input:  "123"
 * - Output: 123 (number type)
*/

const str2 = "123";
const num2 = +str2;
console.log(typeof num2, num2);

// Output: "number" 123


/**
 * PROBLEM 3: Parse integer from string (ignores trailing non-digits)
 * - Input:  "42px"
 * - Output: 42
*/

const str3 = "42px";
const num3 = parseInt(str3, 10);
console.log(num3);

// Output: 42


/**
 * PROBLEM 4: Parse float from string
 * - Input:  "3.14abc"
 * - Output: 3.14
*/

const str4 = "3.14abc";
const num4 = parseFloat(str4);
console.log(num4);

// Output: 3.14


/**
 * PROBLEM 5: Format number to 2 decimal places
 * - Input:  3.14159
 * - Output: "3.14"
*/

const num5 = 3.14159;
console.log(num5.toFixed(2));

// Output: "3.14"


/**
 * PROBLEM 6: Format number to 4 significant digits
 * - Input:  123.456
 * - Output: "123.5"
*/

const num6 = 123.456;
console.log(num6.toPrecision(4));

// Output: "123.5"


/**
 * PROBLEM 7: Convert number to binary string
 * - Input:  255
 * - Output: "11111111"
*/

const num7 = 255;
console.log(num7.toString(2));

// Output: "11111111"


/**
 * PROBLEM 8: Convert number to hex string
 * - Input:  255
 * - Output: "ff"
*/

const num8 = 255;
console.log(num8.toString(16));

// Output: "ff"


/**
 * PROBLEM 9: Convert binary string back to number
 * - Input:  "11111111"
 * - Output: 255
*/

const binStr9 = "11111111";
console.log(parseInt(binStr9, 2));

// Output: 255


/**
 * PROBLEM 10: Check if a value is NaN
 * - Input:  "abc" * 2
 * - Output: true
*/

const val10 = "abc" * 2;
console.log(Number.isNaN(val10));

// Output: true


/**
 * PROBLEM 11: Check if a value is a finite number
 * - Input:  100
 * - Output: true
*/

const val11 = 100;
console.log(Number.isFinite(val11));

// Output: true


/**
 * PROBLEM 12: Check if a value is an integer
 * - Input:  42
 * - Output: true
*/

const val12 = 42;
console.log(Number.isInteger(val12));

// Output: true


/**
 * PROBLEM 13: Round a number to nearest integer
 * - Input:  4.7
 * - Output: 5
*/

const num13 = 4.7;
console.log(Math.round(num13));

// Output: 5


/**
 * PROBLEM 14: Round down a number
 * - Input:  4.9
 * - Output: 4
*/

const num14 = 4.9;
console.log(Math.floor(num14));

// Output: 4


/**
 * PROBLEM 15: Round up a number
 * - Input:  4.1
 * - Output: 5
*/

const num15 = 4.1;
console.log(Math.ceil(num15));

// Output: 5


/**
 * PROBLEM 16: Get absolute value
 * - Input:  -7
 * - Output: 7
*/

const num16 = -7;
console.log(Math.abs(num16));

// Output: 7


/**
 * PROBLEM 17: Generate random number between 0 and 1
 * - Input:  (none)
 * - Output: e.g., 0.54321...
*/

const rand17 = Math.random();
console.log(rand17);

// Output: e.g., 0.54321... (0 <= x < 1)


/**
 * PROBLEM 18: Generate random integer between min and max (inclusive)
 * - Input:  min = 1, max = 10
 * - Output: e.g., 7
*/

const min18 = 1;
const max18 = 10;
const randInt18 = Math.floor(Math.random() * (max18 - min18 + 1)) + min18;
console.log(randInt18);

// Output: e.g., 7 (random 1-10)


/**
 * PROBLEM 19: Find the maximum of a list of numbers
 * - Input:  [3, 7, 2, 9, 1]
 * - Output: 9
*/

const nums19 = [3, 7, 2, 9, 1];
console.log(Math.max(...nums19));

// Output: 9


/**
 * PROBLEM 20: Find the minimum of a list of numbers
 * - Input:  [3, 7, 2, 9, 1]
 * - Output: 1
*/

const nums20 = [3, 7, 2, 9, 1];
console.log(Math.min(...nums20));

// Output: 1


/**
 * PROBLEM 21: Calculate the power of a number
 * - Input:  base = 2, exponent = 10
 * - Output: 1024
*/

const base21 = 2;
const exp21 = 10;
console.log(Math.pow(base21, exp21));

// Output: 1024


/**
 * PROBLEM 22: Calculate the square root
 * - Input:  144
 * - Output: 12
*/

const num22 = 144;
console.log(Math.sqrt(num22));

// Output: 12


/**
 * PROBLEM 23: Fix floating point precision error
 * - Input:  0.1 + 0.2
 * - Output: 0.3 (true, not 0.30000000000000004)
*/

const sum23 = 0.1 + 0.2;
const isClose23 = Math.abs(sum23 - 0.3) < Number.EPSILON;
console.log(isClose23);

// Output: true


/**
 * PROBLEM 24: Clamp a number between min and max
 * - Input:  num = 15, min = 1, max = 10
 * - Output: 10
*/

const clamp24 = (num, min, max) => Math.max(min, Math.min(num, max));
console.log(clamp24(15, 1, 10));

// Output: 10


/**
 * PROBLEM 25: Format number with commas as thousands separator
 * - Input:  1234567
 * - Output: "1,234,567"
*/

const num25 = 1234567;
console.log(num25.toLocaleString("en-US"));

// Output: "1,234,567"


/**
 * PROBLEM 26: Format number as Indian currency (₹)
 * - Input:  125000.5
 * - Output: "₹1,25,000.50"
*/

const num26 = 125000.5;
console.log(
  num26.toLocaleString("en-IN", {
    style: "currency",
    currency: "INR",
  })
);

// Output: "₹1,25,000.50"


/**
 * PROBLEM 27: Remove decimal places without rounding (truncate)
 * - Input:  4.9
 * - Output: 4
*/

const num27 = 4.9;
console.log(Math.trunc(num27));

// Output: 4


/**
 * PROBLEM 28: Check if a number is safe integer
 * - Input:  9007199254740992
 * - Output: false
*/

const num28 = 9007199254740992; // MAX_SAFE_INTEGER + 1
console.log(Number.isSafeInteger(num28));

// Output: false


/**
 * PROBLEM 29: Add two large integers using BigInt
 * - Input:  123456789012345678901n and 987654321098765432109n
 * - Output: 1111111110111111111010n
*/

const big1 = 123456789012345678901n;
const big2 = 987654321098765432109n;
const sum29 = big1 + big2;
console.log(sum29);

// Output: 1111111110111111111010n


/**
 * PROBLEM 30: Convert a number to exponential notation
 * - Input:  12345
 * - Output: "1.23e+4"
*/

const num30 = 12345;
console.log(num30.toExponential(2));

// Output: "1.23e+4"
```
