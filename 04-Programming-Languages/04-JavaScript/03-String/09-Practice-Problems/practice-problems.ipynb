```python
/**
 * PROBLEM 1: Convert string to uppercase
 * - Input:  "hello world"
 * - Output: "HELLO WORLD"
*/

const str1 = "hello world";
console.log(str1.toUpperCase());

// Output: "HELLO WORLD"


/**
 * PROBLEM 2: Convert string to lowercase
 * - Input:  "HELLO WORLD"
 * - Output: "hello world"
*/

const str2 = "HELLO WORLD";
console.log(str2.toLowerCase());

// Output: "hello world"


/**
 * PROBLEM 3: Get the length of a string
 * - Input:  "hello"
 * - Output: 5
*/

const str3 = "hello";
console.log(str3.length);

// Output: 5


/**
 * PROBLEM 4: Get character at a specific index
 * - Input:  "hello", index = 1
 * - Output: "e"
*/

const str4 = "hello";
console.log(str4.charAt(1));

// Output: "e"


/**
 * PROBLEM 5: Get character code at a specific index
 * - Input:  "ABC", index = 0
 * - Output: 65 (ASCII code for 'A')
*/

const str5 = "ABC";
console.log(str5.charCodeAt(0));

// Output: 65


/**
 * PROBLEM 6: Get the first character of a string
 * - Input:  "hello"
 * - Output: "h"
*/

const str6 = "hello";
console.log(str6.charAt(0));

// Output: "h"


/**
 * PROBLEM 7: Get the last character of a string
 * - Input:  "hello"
 * - Output: "o"
*/

const str7 = "hello";
console.log(str7.charAt(str7.length - 1));

// Output: "o"


/**
 * PROBLEM 8: Capitalize the first letter of a string
 * - Input:  "hello"
 * - Output: "Hello"
*/

const str8 = "hello";
const cap8 = str8.charAt(0).toUpperCase() + str8.slice(1);
console.log(cap8);

// Output: "Hello"


/**
 * PROBLEM 9: Capitalize the first letter of each word
 * - Input:  "hello world javascript"
 * - Output: "Hello World Javascript"
*/

const str9 = "hello world javascript";
const capWords9 = str9
  .split(" ")
  .map((word) => word.charAt(0).toUpperCase() + word.slice(1))
  .join(" ");
console.log(capWords9);

// Output: "Hello World Javascript"


/**
 * PROBLEM 10: Reverse a string
 * - Input:  "hello"
 * - Output: "olleh"
*/

const str10 = "hello";
const reversed10 = str10.split("").reverse().join("");
console.log(reversed10);

// Output: "olleh"


/**
 * PROBLEM 11: Check if a string is a palindrome
 * - Input:  "madam"
 * - Output: true
*/

const str11 = "madam";
const isPalindrome11 = str11 === str11.split("").reverse().join("");
console.log(isPalindrome11);

// Output: true


/**
 * PROBLEM 12: Check if a string is a palindrome (ignoring case and spaces)
 * - Input:  "A man a plan a canal Panama"
 * - Output: true
*/

const str12 = "A man a plan a canal Panama";
const cleanStr12 = str12.toLowerCase().replace(/\s/g, "");
const isPalindrome12 = cleanStr12 === cleanStr12.split("").reverse().join("");
console.log(isPalindrome12);

// Output: true


/**
 * PROBLEM 13: Count the number of vowels in a string
 * - Input:  "javascript"
 * - Output: 3 (a, a, i)
*/

const str13 = "javascript";
const vowels13 = "aeiouAEIOU";
let count13 = 0;
for (let char of str13) {
  if (vowels13.includes(char)) count13++;
}
console.log(count13);

// Output: 3


/**
 * PROBLEM 14: Count the number of consonants in a string
 * - Input:  "javascript"
 * - Output: 7
*/

const str14 = "javascript";
const vowels14 = "aeiouAEIOU";
let count14 = 0;
for (let char of str14) {
  if (char.match(/[a-z]/i) && !vowels14.includes(char)) count14++;
}
console.log(count14);

// Output: 7


/**
 * PROBLEM 15: Remove all whitespace from a string
 * - Input:  "  hello   world  "
 * - Output: "helloworld"
*/

const str15 = "  hello   world  ";
console.log(str15.replace(/\s/g, ""));

// Output: "helloworld"


/**
 * PROBLEM 16: Trim leading and trailing whitespace
 * - Input:  "  hello world  "
 * - Output: "hello world"
*/

const str16 = "  hello world  ";
console.log(str16.trim());

// Output: "hello world"


/**
 * PROBLEM 17: Check if a string contains a substring
 * - Input:  "hello world", substring = "world"
 * - Output: true
*/

const str17 = "hello world";
console.log(str17.includes("world"));

// Output: true


/**
 * PROBLEM 18: Find the index of a substring
 * - Input:  "hello world", substring = "world"
 * - Output: 6
*/

const str18 = "hello world";
console.log(str18.indexOf("world"));

// Output: 6


/**
 * PROBLEM 19: Find the last index of a character
 * - Input:  "hello world", char = "o"
 * - Output: 7
*/

const str19 = "hello world";
console.log(str19.lastIndexOf("o"));

// Output: 7


/**
 * PROBLEM 20: Replace all occurrences of a substring
 * - Input:  "hello world hello", replace "hello" with "hi"
 * - Output: "hi world hi"
*/

const str20 = "hello world hello";
console.log(str20.replaceAll("hello", "hi"));

// Output: "hi world hi"


/**
 * PROBLEM 21: Count the frequency of each character in a string
 * - Input:  "hello"
 * - Output: { h: 1, e: 1, l: 2, o: 1 }
*/

const str21 = "hello";
const freq21 = {};
for (let char of str21) {
  freq21[char] = (freq21[char] || 0) + 1;
}
console.log(freq21);

// Output: { h: 1, e: 1, l: 2, o: 1 }


/**
 * PROBLEM 22: Split a string into an array of words
 * - Input:  "hello world javascript"
 * - Output: ["hello", "world", "javascript"]
*/

const str22 = "hello world javascript";
console.log(str22.split(" "));

// Output: ["hello", "world", "javascript"]


/**
 * PROBLEM 23: Join an array of strings into a single string
 * - Input:  ["hello", "world", "javascript"]
 * - Output: "hello-world-javascript"
*/

const arr23 = ["hello", "world", "javascript"];
console.log(arr23.join("-"));

// Output: "hello-world-javascript"


/**
 * PROBLEM 24: Extract a substring using slice
 * - Input:  "hello world", start = 0, end = 5
 * - Output: "hello"
*/

const str24 = "hello world";
console.log(str24.slice(0, 5));

// Output: "hello"


/**
 * PROBLEM 25: Extract a substring using substring
 * - Input:  "hello world", start = 6
 * - Output: "world"
*/

const str25 = "hello world";
console.log(str25.substring(6));

// Output: "world"


/**
 * PROBLEM 26: Repeat a string n times
 * - Input:  "abc", n = 3
 * - Output: "abcabcabc"
*/

const str26 = "abc";
console.log(str26.repeat(3));

// Output: "abcabcabc"


/**
 * PROBLEM 27: Check if a string starts with a specific substring
 * - Input:  "hello world", substring = "hello"
 * - Output: true
*/

const str27 = "hello world";
console.log(str27.startsWith("hello"));

// Output: true


/**
 * PROBLEM 28: Check if a string ends with a specific substring
 * - Input:  "hello world", substring = "world"
 * - Output: true
*/

const str28 = "hello world";
console.log(str28.endsWith("world"));

// Output: true


/**
 * PROBLEM 29: Count the number of words in a sentence
 * - Input:  "The quick brown fox jumps"
 * - Output: 5
*/

const str29 = "The quick brown fox jumps";
const words29 = str29.trim().split(/\s+/);
console.log(words29.length);

// Output: 5


/**
 * PROBLEM 30: Truncate a string and add "..." if it exceeds a max length
 * - Input:  "hello world javascript", maxLength = 10
 * - Output: "hello w..."
*/

const str30 = "hello world javascript";
const maxLength30 = 10;
const truncated30 =
  str30.length > maxLength30
    ? str30.slice(0, maxLength30 - 3) + "..."
    : str30;
console.log(truncated30);

// Output: "hello w..."
```
