```python
# 1. What is String?
#    - A string is a sequence of characters.
#    - Concretely, it's a value of the built-in type "str", and you create one by 
#      wrapping text in quotes:

product_name = "Wireless Headphones"
customer_name = "Aarav Mehta"
order_status = "shipped"

print(product_name)
print(customer_name)
print(order_status)
print(type(product_name))
```

```python
# 2. Single Quotes, Double Quotes, and What's Inside:
#    - Python let's you write a string with single quotes or double quotes.
#    - Both produce the exact same "str" object:

title_a = "Wireless Headphones"
title_b = 'Wireless Headphones'

print(title_a == title_b)
print(type(title_a))
print(type(title_b))

# Note:
# - There is no "double-quote string" type and "single-quote string" type.
# - The quote characters are just the boundary markers Python uses to figure out
#   where the string starts and end.

# Why two choices?
# - Because text often contains a quote character, and you want the freedom to pick the
#   boundary that keeps the value readable. 
# - If your string has an apostrophe in it, wrap it in double quotes so the apostrophe
#   doesn't end the string early:

review = "It's the best laptop I've ever owned."
print(review)

# If your string has double quotes inside it, wrap it in single quotes:
product_blurb = 'Now featuring the new "Pro" mode for power users.'
print(product_blurb)

# If a string contains both single and double quotes, you have a few options. 
# The simplest is to escape one of them with a backslash:

mixed = "He said \"It's on sale\" and walked off."
print(mixed)
```

```python
# 3. Counting Characters with len()
#    - To find out how many characters a string has, pass it to the built-in len()
#    - len() counts every character, including spaces and punctuation.
#    - len() is a built-in function, not a method on the string. 
#    - So, you call it as len(s), not s.len(). 

product_name = "Wireless Headphones"
email = "aarav@example.com"
empty = ""

print(len(product_name))
print(len(email))
print(len(empty))
```

```python
# Ex: Checking whether some text the customer typed is short enough or long enough 
#     to be valid:

search_query = "wireless"

if len(search_query) < 3:
    print("Search must be at least 3 characters")
else:
    print("Searching for: " + search_query)
```

```python
# 4. Turning Other Values into Strings with str():

quantity = 3
price = 29.99
in_stock = True

quantity_text = str(quantity)
price_text = str(price)
in_stock_text = str(in_stock)

print(quantity_text)
print(price_text)
print(in_stock_text)
print(type(quantity_text))
```

```python
# 5. Concatenation: Gluing Strings with +

first_name = "Aarav"
last_name = "Mehta"

full_name = first_name + " " + last_name
print(full_name)
```

```python
# Example: 
product = "Wireless Headphones"
status = "shipped"

confirmation = "Your order for " + product + " has been " + status + "."
print(confirmation)
```

```python
# Big Rule: 
# - Both operands must be strings. 
# - Python won't quietly convert a number into text for you. This breaks:

# What's wrong with this code?
# - quantity is an int, not a str. 
# - The + operator doesn't know how to combine those two types, so it raises TypeError.
# - The fix is to 
quantity = 3
message = "You have " + quantity + " items in your cart."
print(message)
```

```python
# Fix: Convert the integer to a string first with str():
# - str(quantity) produces "3", which + is happy to glue onto the surrounding text.

quantity = 3
message = "You have " + str(quantity) + " items in your cart."
print(message)
```

```python
# Another Way: Let print itself handle the mixed types.
# - print() accepts any number of arguments of any type, converts each one to its 
#   string form, and joins them with a space.
# - No str() needed, no + needed. 
quantity = 3
print("You have", quantity, "items in your cart.")
```

```python
# 5. Repetition with *:
#    - Multiplying a string by an integer repeats it that many times. 
#    - The result is a new string:

divider = "-" * 30
print(divider)

separator = "= " * 10
print(separator)
```

```python
# Example: If you multiply by zero or a negative number, you get an empty string:

nothing = "abc" * 0
also_nothing = "abc" * -5

print(repr(nothing))
print(repr(also_nothing))
print(len(nothing))
print(len(also_nothing))
```

```python
# Example: A nice combination of + and * for visual receipts:

title = "ORDER SUMMARY"
print("=" * 40)
print(" " * 13 + title)
print("=" * 40)
```

```python
# 6. Membership: in and not in
#    - To check whether a string contains a smaller string, use in. 
#    - The result is a boolean:

product_name = "Wireless Bluetooth Headphones"

print("Bluetooth" in product_name)
print("USB" in product_name)
print("phones" in product_name)
print("Wireless" not in product_name)
```

```python
# Example: in works against single characters too, since a single character is just a 
#          string of length one:

email = "aarav@example.com"

print("@" in email)
print("." in email)
print("?" in email)
```

```python
# Example: Quick category filter for a search box
# - Here in does the substring check directly, with no methods needed:

search_query = "Headphones"
product_titles = [
    "Wireless Bluetooth Headphones",
    "Mechanical Keyboard",
    "Noise Cancelling Headphones",
    "USB-C Charging Cable",
]

for title in product_titles:
    if search_query in title:
        print("Match:", title)
```

```python
# Example: not in is just the inverse. 
# - It's clearer than writing not (x in y):
# - Use not in when the natural way to read the condition is "X isn't in Y."

email = "aarav@example.com"

if "@" not in email:
    print("Invalid email address")
else:
    print("Looks like an email")
```

```python
# 7. Indexing: Reaching into a String by Position
#    - A string is a sequence, and like any sequence in Python, you can grab one 
#      character at a time by writing the position in square brackets. 
#    - Position counts from zero on the left:

product = "laptop"

print(product[0])
print(product[1])
print(product[2])
print(product[5])
```

```python
# Example: 
# - You can also count from the right using negative indices. 
# - -1 is the last character, -2 is the second-to-last, and so on:

product = "laptop"

print(product[-1])
print(product[-2])
print(product[-6])

# Here's the index map for "laptop":
# a. Positive Index:
#    0   1   2   3   4   5
#    ↓   ↓   ↓   ↓   ↓   ↓
#    l   a   p   t   o   p
#
# b. Negative Index:
#   -6  -5  -4  -3  -2  -1
#    ↓   ↓   ↓   ↓   ↓   ↓
#    l   a   p   t   o   p
#
# c. So:
#    l →  0 / -6
#    a →  1 / -5
#    p →  2 / -4
#    t →  3 / -3
#    o →  4 / -2
#    p →  5 / -1
```

```python
# Example: 
# - The result of indexing a string is always another string, of length one. 
# - There's no separate "char" type in Python:

product = "laptop"

first = product[0]
print(first)
print(type(first))
print(len(first))
```

```python
# Example: If you ask for an index that isn't there, you get an IndexError:

product = "laptop"
print(product[10])
```

```python
# Example:
# Suppose you have a product code "AB" and you want to access the 4th character using
# index 3. But the string contains only 2 characters. What will happen if you directly
# access product_code[3]?

product_code = "AB"

# Check if the string has at least 4 characters
if len(product_code) >= 4:
    category_code = product_code[3]
    print("Category:", category_code)
else:
    print("Code too short")
```

```python
# 8. Iterating Through a String with for
#    - Because a string is a sequence, you can walk through its characters one at a 
#      time with a for loop. 
#    - The loop variable holds each character in turn:

product = "laptop"

for char in product:
    print(char)
```

```python
# Problem:
# Suppose you have an email address and you want to count how many @ signs it contains.
# How can you check every character without manually managing the index?

# Solution:
email = "aarav@example.com"
at_count = 0

# Loop through each character in the string
for c in email:
    if c == "@":
        at_count = at_count + 1

print("Number of @ signs:", at_count)
```

```python
# Problem:
# Suppose you have a product review and you want to count how many exclamation marks (!)
# it contains. Can you do this without creating a counter and writing an explicit if 
# statement?

# Solution:
review = "Great laptop! Battery lasts forever."

exclamations = sum(c == "!" for c in review)

print("Exclamation marks:", exclamations)
```

```python
# Problem:
# Suppose you have a product name "laptop" and you want to print both the index and 
# the character at that index. How can you get the character and its position together
# without manually using range(len(product))?

# Solution: Enumerate()
# - enumerate() automatically gives you two values during each iteration:
#   - index → position of the character
#   - char  → actual character

product = "laptop"
for index, char in enumerate(product):
    print(index, char)
```

```python
# 9. The Empty String and Truthiness
#    - "" is a perfectly valid string, just one with zero characters. 
#    - You'll create empty strings as starting points for accumulators, as default values
#      when you have nothing to show, and as the result of operations that simply 
#      produce no characters.

greeting = ""
print(repr(greeting))
print(len(greeting))
print(type(greeting))
```

```python
# Problem:
# Suppose you have a customer's shipping address stored in a variable. 
# You want to check whether an address is available before displaying it. 
# If the address is empty, you should show "No address on file".
# How can you use Python's truthiness to perform this check?

# Solution:
customer_address = ""

if customer_address:
    print("Shipping to:", customer_address)
else:
    print("No address on file")
```

```python
# Problem:
# Suppose a user is searching for a product, but they haven't entered anything in the
# search box. You want to display a message asking them to enter a search term.
# How can you use Python's not operator to check whether the string is empty?

# Solution:
search_query = ""

if not search_query:
    print("Please enter a search term")
```

```python
# Problem:
# Suppose a customer address contains only spaces, like " ". 
# You want to check whether the address is actually empty.
# Will Python consider a string containing only spaces as empty?

# Solution:
# The string " " contains 3 space characters. Therefore, it is not an empty string.
# - ""      → 0 characters → False
# - " "     → 1 character  → True
# - "   "   → 3 characters → True
# - "abc"   → 3 characters → True
# - maybe_address = "   "
# So, if maybe_address: is True because the string contains characters, even though 
# those characters are only spaces.

if maybe_address:
    print("Has an address (technically)")

print("Length:", len(maybe_address))

# Fix: If you want to treat whitespace-only strings as empty (zero chars), use strip():
# - strip() removes whitespace from the beginning and end of the string. 
# - So " ".strip() becomes "", which is considered False.

maybe_address = "   "

if maybe_address.strip():
    print("Has an address")
else:
    print("No address provided")
```

```python
# 10. IMMUTABILITY:
#     - Once a string exists, you can't change a character inside it. 
#     - Trying to assign into a position raises an error:

product = "laptop"
product[0] = "L"
```

```python
# Fix: To "change" a string, you build a new string and re-bind the name:
product = "laptop"
product = "L" + product[1:]
print(product)
```
