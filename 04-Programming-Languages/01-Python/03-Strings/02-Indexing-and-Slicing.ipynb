```python
# Slicing: 
# - Slicing is how you pull a piece out of a string by position. 
# - Syntax: s[start:stop:step] 
```

```python
# 1. The Basic Slice: s[start:stop]
#    - A slice carves out a substring by giving Python two indices. 
#    - The first character of the result is at start. 
#    - The slice stops just before stop, never at it. 
#    - That's called a half-open interval, and once you internalize it, every other
#      slicing rule clicks into place. 

product = "wireless"
print(product[0:4])
print(product[2:6])
print(product[4:8])
```

```python
# Problem:
# Suppose you have the string "laptop" and you want to extract the characters from 
# index 1 up to, but not including, index 4. What will laptop[1:4] return?

# Solution:
# - The start index is included, but the end index is excluded.
# - For laptop:
#   Index:     0   1   2   3   4   5
#   Character: l   a   p   t   o   p
#                  ↑           ↑
#                start        end

laptop = "laptop"
print(laptop[1:4])

# Cost:
# - It creates a new string object containing the sliced characters.
# - If the slice contains k characters:
#   - TC: O(k)
#   - SC: O(k)
```

```python
# 2. Defaults: Omitting start, stop, or Both
#    - Each piece of a slice has a default. 
#    - Leave out start and Python uses 0. 
#    - Leave out stop and Python uses the length of the string 
#     (which means "go all the way to the end"). 
#    - Leave out both and you get the whole string.

category = "electronics"

print(category[:5])
print(category[5:])
print(category[:])
```
