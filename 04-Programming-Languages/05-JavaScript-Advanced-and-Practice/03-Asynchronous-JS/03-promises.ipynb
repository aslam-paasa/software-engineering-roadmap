```python
/**
 * Approach-1: Callback Hell (Worst Approach)
 * - Don't use setTimeout() + callback function together for async operations.
 * - It's a bad practice.
*/

function stepOne(callback) {
    setTimeout(() => {
        console.log("Step one done!");
        callback();
    }, 1000);
}

function stepTwo(callback) {
    setTimeout(() => {
        console.log("Step two done!");
        callback();
    }, 1000);
}

function stepThree(callback) {
    setTimeout(() => {
        console.log("Step three done!");
        callback();
    }, 1000);
}

/**
 * Callback Hell:
 */
stepOne(() => {
    stepTwo(() => {
        stepThree(() => {
            console.log("All steps completed!");
        });
    });
});


/**
 * Approach-2: Promises (Better Approach)
 * - A Promise reqresents the eventual completion or failure of an async operation.
 * - A Promise is like a contract that says:
 *   a. Success case: Fulfilled
 *      - When the task completes successfully
 *      - We will return the result
 *   b. Error case: Rejected
 *      - If the task fails to complete
 *      - We will return an error message
 *   c. Waiting period: Pending
 *      - Until we get the result/error
 *      - We will wait
 * - In simple words: 
 *   "I promise that I'll give you the result if task succeeds,
 *    an error if it fails, and until then please wait"
*/

/**
 * Benefits:
 * 1. Readability
 * 2. Error Handling
 * 3. Code Organization
 * 4. Method Chaining
*/


/**
 * Why we don't need to pass callback function as an argument in Promise?
 * 1. Callback approach:
 *    - A callback is a function passed an argument to be invoked after an async
 *      operation completes.
 *    - Nesting multiple async operations lead to callback hell - deeply nested,
 *      hard-to-read code.
 *    - We have to explicitly pass the callback function as an argument
 *    - Example: stepOne(callback)
 * 
 * 2. Promise approach:
 *    - Promise itself provides the .then() method
 *    - We can simply write: promiseOne().then(...)   [chaining]
 *    - Promise automatically handles when to execute the next step
 * 
 * In simple words: 
 * - Promise is a modern way to handle callbacks without manually passing them.
 * - Promise handles everything for us!
*/

function promiseOne() {
    return new Promise((resolve, reject) => {
        setTimeout(() => {
            console.log("Step one done!");
            resolve();
        }, 1000);
    });
}

function promiseTwo() {
    return new Promise((resolve, reject) => {
        setTimeout(() => {
            console.log("Step two done!");
            resolve();
        }, 1000);
    });
}


function promiseThree() {
    return new Promise((resolve, reject) => {
        setTimeout(() => {
            console.log("Step three done!");
            resolve();
        }, 1000);
    });
}

/**
 * Chaining Promises:
*/
promiseOne()
    .then(promiseTwo)
    .then(promiseThree)
    .then(() => {
        console.log("All steps completed!");
    });
```

```python
/**
 * PROBLEM 1: Create a basic resolved promise
 * - Input:  Promise.resolve(42)
 * - Output: 42
*/

const promise1 = Promise.resolve(42);
promise1.then((val) => console.log(val));

// Output: 42


/**
 * PROBLEM 2: Create a basic rejected promise
 * - Input:  Promise.reject("Error!")
 * - Output: "Error!"
*/

const promise2 = Promise.reject("Error!");
promise2.catch((err) => console.log(err));

// Output: "Error!"


/**
 * PROBLEM 3: Create a promise from scratch (resolve after 1 second)
 * - Input:  new Promise((resolve) => setTimeout(() => resolve("Done"), 1000))
 * - Output: "Done" (after 1 second)
*/

const promise3 = new Promise((resolve) => {
  setTimeout(() => resolve("Done"), 1000);
});
promise3.then((val) => console.log(val));

// Output: "Done" (after 1 second)


/**
 * PROBLEM 4: Create a promise that resolves or rejects based on condition
 * - Input:  checkAge(20)
 * - Output: "Adult"
*/

function checkAge(age) {
  return new Promise((resolve, reject) => {
    if (age >= 18) {
      resolve("Adult");
    } else {
      reject("Minor");
    }
  });
}
checkAge(20)
  .then((val) => console.log(val))
  .catch((err) => console.log(err));

// Output: "Adult"


/**
 * PROBLEM 5: Promise with .then(), .catch(), and .finally()
 * - Input:  fetchData(true)
 * - Output: "Data: Hello", then "Cleanup"
*/

function fetchData(success) {
  return new Promise((resolve, reject) => {
    if (success) {
      resolve("Hello");
    } else {
      reject("Network error");
    }
  });
}
fetchData(true)
  .then((data) => console.log(`Data: ${data}`))
  .catch((err) => console.log(err))
  .finally(() => console.log("Cleanup"));

// Output: "Data: Hello", "Cleanup"


/**
 * PROBLEM 6: Promise chaining — pass value through chain
 * - Input:  Promise.resolve(5).then(add 10).then multiply 2
 * - Output: 30
*/

Promise.resolve(5)
  .then((val) => val + 10)
  .then((val) => val * 2)
  .then((val) => console.log(val));

// Output: 30


/**
 * PROBLEM 7: Promise chaining with async operations
 * - Input:  fetch user → fetch posts → count posts
 * - Output: 3
*/

function fetchUser() {
  return Promise.resolve({ id: 1, name: "John" });
}
function fetchPosts(userId) {
  return Promise.resolve([
    { id: 1, userId: 1 },
    { id: 2, userId: 1 },
    { id: 3, userId: 1 },
  ]);
}
fetchUser()
  .then((user) => fetchPosts(user.id))
  .then((posts) => console.log(posts.length));

// Output: 3


/**
 * PROBLEM 8: Error handling in promise chain
 * - Input:  Promise.resolve(10) → throw error → catch
 * - Output: "Caught: Something went wrong"
*/

Promise.resolve(10)
  .then((val) => {
    throw new Error("Something went wrong");
  })
  .catch((err) => console.log(`Caught: ${err.message}`));

// Output: "Caught: Something went wrong"


/**
 * PROBLEM 9: Catch only catches errors from above
 * - Input:  Promise.resolve(1) → then → catch → then (error happens after catch)
 * - Output: 2, "Uncaught (no catch after)"
*/

Promise.resolve(1)
  .then((val) => val + 1)
  .catch((err) => console.log("This won't run"))
  .then((val) => {
    throw new Error("Late error");
  })
  .catch((err) => console.log(`Late catch: ${err.message}`));

// Output: "Late catch: Late error"


/**
 * PROBLEM 10: Promise.all — wait for all to resolve
 * - Input:  [Promise.resolve(1), Promise.resolve(2), Promise.resolve(3)]
 * - Output: [1, 2, 3]
*/

const promises10 = [Promise.resolve(1), Promise.resolve(2), Promise.resolve(3)];
Promise.all(promises10).then((results) => console.log(results));

// Output: [1, 2, 3]


/**
 * PROBLEM 11: Promise.all — one rejects, all fails
 * - Input:  [Promise.resolve(1), Promise.reject("Error"), Promise.resolve(3)]
 * - Output: "Error" (rejects immediately)
*/

const promises11 = [
  Promise.resolve(1),
  Promise.reject("Error"),
  Promise.resolve(3),
];
Promise.all(promises11)
  .then((results) => console.log(results))
  .catch((err) => console.log(err));

// Output: "Error"


/**
 * PROBLEM 12: Promise.race — first to settle wins
 * - Input:  [resolve after 1s, resolve after 0.5s]
 * - Output: "Fast" (the 0.5s one)
*/

const slow12 = new Promise((resolve) => setTimeout(() => resolve("Slow"), 1000));
const fast12 = new Promise((resolve) => setTimeout(() => resolve("Fast"), 500));
Promise.race([slow12, fast12]).then((val) => console.log(val));

// Output: "Fast" (after 500ms)


/**
 * PROBLEM 13: Promise.allSettled — wait for all, regardless of resolve/reject
 * - Input:  [Promise.resolve(1), Promise.reject("Err"), Promise.resolve(3)]
 * - Output: [{status: "fulfilled", value: 1}, {status: "rejected", reason: "Err"}, {status: "fulfilled", value: 3}]
*/

const promises13 = [
  Promise.resolve(1),
  Promise.reject("Err"),
  Promise.resolve(3),
];
Promise.allSettled(promises13).then((results) => console.log(results));

// Output: [{status: "fulfilled", value: 1}, {status: "rejected", reason: "Err"}, {status: "fulfilled", value: 3}]


/**
 * PROBLEM 14: Promise.any — first to RESOLVE wins (ignores rejections)
 * - Input:  [Promise.reject("Err1"), Promise.resolve("Success"), Promise.reject("Err2")]
 * - Output: "Success"
*/

const promises14 = [
  Promise.reject("Err1"),
  Promise.resolve("Success"),
  Promise.reject("Err2"),
];
Promise.any(promises14).then((val) => console.log(val));

// Output: "Success"


/**
 * PROBLEM 15: Convert callback to promise (promisify)
 * - Input:  callback-based function → promise-based
 * - Output: "Data received"
*/

function promisify(fn) {
  return function (...args) {
    return new Promise((resolve, reject) => {
      fn(...args, (err, data) => {
        if (err) reject(err);
        else resolve(data);
      });
    });
  };
}
// Mock callback function
function loadData(cb) {
  setTimeout(() => cb(null, "Data received"), 100);
}
const loadDataPromise = promisify(loadData);
loadDataPromise().then((data) => console.log(data));

// Output: "Data received"


/**
 * PROBLEM 16: Sequential execution using async/await
 * - Input:  [fn1, fn2, fn3] executed one by one
 * - Output: "Step 1", "Step 2", "Step 3" (in order, with delays)
*/

async function runSequentially() {
  const step1 = await new Promise((r) => setTimeout(() => r("Step 1"), 100));
  console.log(step1);
  const step2 = await new Promise((r) => setTimeout(() => r("Step 2"), 100));
  console.log(step2);
  const step3 = await new Promise((r) => setTimeout(() => r("Step 3"), 100));
  console.log(step3);
}
runSequentially();

// Output: "Step 1", "Step 2", "Step 3" (each 100ms apart)


/**
 * PROBLEM 17: Parallel execution using Promise.all + async/await
 * - Input:  3 async functions executed in parallel
 * - Output: ["A", "B", "C"]
*/

async function runParallel() {
  const results = await Promise.all([
    Promise.resolve("A"),
    Promise.resolve("B"),
    Promise.resolve("C"),
  ]);
  console.log(results);
}
runParallel();

// Output: ["A", "B", "C"]


/**
 * PROBLEM 18: Error handling with try/catch in async/await
 * - Input:  async function that throws
 * - Output: "Error caught: Fail!"
*/

async function riskyOperation() {
  try {
    const result = await Promise.reject("Fail!");
    console.log(result);
  } catch (err) {
    console.log(`Error caught: ${err}`);
  }
}
riskyOperation();

// Output: "Error caught: Fail!"


/**
 * PROBLEM 19: Fetch API simulation with promises
 * - Input:  mockFetch("/users/1")
 * - Output: { id: 1, name: "John" }
*/

function mockFetch(url) {
  return new Promise((resolve, reject) => {
    setTimeout(() => {
      if (url.includes("/users/1")) {
        resolve({ id: 1, name: "John" });
      } else {
        reject("404 Not Found");
      }
    }, 200);
  });
}
mockFetch("/users/1")
  .then((data) => console.log(data))
  .catch((err) => console.log(err));

// Output: { id: 1, name: "John" }


/**
 * PROBLEM 20: Retry a failed promise N times
 * - Input:  retry(unstableFn, 3, 100)
 * - Output: "Success on attempt 3"
*/

function retry(fn, maxRetries, delay) {
  return new Promise((resolve, reject) => {
    let attempts = 0;
    function attempt() {
      attempts++;
      fn()
        .then(resolve)
        .catch((err) => {
          if (attempts >= maxRetries) {
            reject(`Failed after ${maxRetries} attempts: ${err}`);
          } else {
            setTimeout(attempt, delay);
          }
        });
    }
    attempt();
  });
}
let count20 = 0;
const unstableFn = () =>
  new Promise((resolve, reject) => {
    count20++;
    if (count20 < 3) reject("Not yet");
    else resolve("Success on attempt 3");
  });
retry(unstableFn, 5, 100).then((val) => console.log(val));

// Output: "Success on attempt 3"


/**
 * PROBLEM 21: Promise-based delay/sleep function
 * - Input:  sleep(500).then(() => "Woke up")
 * - Output: "Woke up" (after 500ms)
*/

function sleep(ms) {
  return new Promise((resolve) => setTimeout(resolve, ms));
}
sleep(500).then(() => console.log("Woke up"));

// Output: "Woke up" (after 500ms)


/**
 * PROBLEM 22: Execute promises sequentially with reduce
 * - Input:  [fn1, fn2, fn3]
 * - Output: 6 (1+2+3 accumulated)
*/

const tasks22 = [
  () => Promise.resolve(1),
  () => Promise.resolve(2),
  () => Promise.resolve(3),
];
tasks22
  .reduce((acc, task) => acc.then((sum) => task().then((val) => sum + val)), Promise.resolve(0))
  .then((total) => console.log(total));

// Output: 6


/**
 * PROBLEM 23: Promise pool — run N promises at a time
 * - Input:  [p1, p2, p3, p4, p5], concurrency = 2
 * - Output: [1, 2, 3, 4, 5] (runs 2 at a time, resolves all)
*/

async function promisePool(tasks, concurrency) {
  const results = [];
  let index = 0;
  async function worker() {
    while (index < tasks.length) {
      const currentIndex = index++;
      results[currentIndex] = await tasks[currentIndex]();
    }
  }
  const workers = Array.from({ length: concurrency }, () => worker());
  await Promise.all(workers);
  return results;
}
const tasks23 = [
  () => Promise.resolve(1),
  () => Promise.resolve(2),
  () => Promise.resolve(3),
  () => Promise.resolve(4),
  () => Promise.resolve(5),
];
promisePool(tasks23, 2).then((r) => console.log(r));

// Output: [1, 2, 3, 4, 5]


/**
 * PROBLEM 24: Timeout a promise — reject if not resolved in time
 * - Input:  timeout(slowPromise, 100)
 * - Output: "Timeout"
*/

function timeout(promise, ms) {
  const timeoutPromise = new Promise((_, reject) =>
    setTimeout(() => reject("Timeout"), ms)
  );
  return Promise.race([promise, timeoutPromise]);
}
const slowPromise24 = new Promise((r) => setTimeout(() => r("Done"), 1000));
timeout(slowPromise24, 100)
  .then((val) => console.log(val))
  .catch((err) => console.log(err));

// Output: "Timeout"


/**
 * PROBLEM 25: Chain promises dynamically from an array
 * - Input:  [fn1, fn2, fn3] chained dynamically
 * - Output: "Step 3"
*/

const steps25 = [
  () => Promise.resolve("Step 1"),
  (prev) => Promise.resolve("Step 2"),
  (prev) => Promise.resolve("Step 3"),
];
let chain25 = Promise.resolve("");
steps25.forEach((step) => {
  chain25 = chain25.then((prev) => step(prev));
});
chain25.then((result) => console.log(result));

// Output: "Step 3"


/**
 * PROBLEM 26: Resolve array of mixed values and promises
 * - Input:  [1, Promise.resolve(2), 3, Promise.resolve(4)]
 * - Output: [1, 2, 3, 4]
*/

const mixed26 = [1, Promise.resolve(2), 3, Promise.resolve(4)];
Promise.all(mixed26).then((results) => console.log(results));

// Output: [1, 2, 3, 4]


/**
 * PROBLEM 27: Fetch with fallback URL
 * - Input:  fetchWithFallback("/primary", "/fallback")
 * - Output: "From fallback" (primary fails)
*/

function fetchWithFallback(primaryUrl, fallbackUrl) {
  function mockFetchUrl(url) {
    if (url === "/primary") return Promise.reject("Primary failed");
    return Promise.resolve(`From ${url}`);
  }
  return mockFetchUrl(primaryUrl).catch(() => mockFetchUrl(fallbackUrl));
}
fetchWithFallback("/primary", "/fallback").then((data) => console.log(data));

// Output: "From /fallback"


/**
 * PROBLEM 28: Queue promises — execute one after another automatically
 * - Input:  queue.enqueue(fn1), queue.enqueue(fn2)
 * - Output: "Task 1 done", "Task 2 done"
*/

function createPromiseQueue() {
  let queue = Promise.resolve();
  return {
    enqueue(fn) {
      queue = queue.then(fn).catch((err) => console.log(err));
      return queue;
    },
  };
}
const queue28 = createPromiseQueue();
queue28.enqueue(() => Promise.resolve("Task 1 done").then((r) => console.log(r)));
queue28.enqueue(() => Promise.resolve("Task 2 done").then((r) => console.log(r)));

// Output: "Task 1 done", "Task 2 done"


/**
 * PROBLEM 29: Memoize async function (cache promise results)
 * - Input:  memoizedFetch("/users/1") called twice
 * - Output: "Fetched", { id: 1 } (cached on 2nd call)
*/

function memoizeAsync(fn) {
  const cache = {};
  return function (...args) {
    const key = JSON.stringify(args);
    if (!cache[key]) {
      console.log("Fetching...");
      cache[key] = fn(...args).catch((err) => {
        delete cache[key];
        throw err;
      });
    } else {
      console.log("From cache");
    }
    return cache[key];
  };
}
const memoFetch = memoizeAsync(function (url) {
  return Promise.resolve({ id: 1 });
});
memoFetch("/users/1").then((data) => console.log(data));
memoFetch("/users/1").then((data) => console.log(data));

// Output: "Fetching...", { id: 1 }, "From cache", { id: 1 }


/**
 * PROBLEM 30: Custom Promise implementation (simplified — resolve/reject/then)
 * - THE most asked question in interviews
 * - Implement a basic MyPromise with: resolve, reject, then, catch
 * - Supports: async resolution, chaining, error propagation
*/

class MyPromise {
  constructor(executor) {
    this.state = "pending";
    this.value = undefined;
    this.reason = undefined;
    this.onFulfilledCallbacks = [];
    this.onRejectedCallbacks = [];

    const resolve = (value) => {
      if (this.state === "pending") {
        this.state = "fulfilled";
        this.value = value;
        this.onFulfilledCallbacks.forEach((cb) => cb(this.value));
      }
    };

    const reject = (reason) => {
      if (this.state === "pending") {
        this.state = "rejected";
        this.reason = reason;
        this.onRejectedCallbacks.forEach((cb) => cb(this.reason));
      }
    };

    try {
      executor(resolve, reject);
    } catch (err) {
      reject(err);
    }
  }

  then(onFulfilled) {
    return new MyPromise((resolve, reject) => {
      const handleFulfilled = (value) => {
        try {
          const result = onFulfilled ? onFulfilled(value) : value;
          resolve(result);
        } catch (err) {
          reject(err);
        }
      };

      if (this.state === "fulfilled") {
        handleFulfilled(this.value);
      } else if (this.state === "pending") {
        this.onFulfilledCallbacks.push(handleFulfilled);
      }
    });
  }

  catch(onRejected) {
    return new MyPromise((resolve, reject) => {
      const handleRejected = (reason) => {
        try {
          const result = onRejected ? onRejected(reason) : reason;
          resolve(result);
        } catch (err) {
          reject(err);
        }
      };

      if (this.state === "rejected") {
        handleRejected(this.reason);
      } else if (this.state === "pending") {
        this.onRejectedCallbacks.push(handleRejected);
      }
    });
  }
}

// ---- TEST 1: Basic resolve ----
const p1 = new MyPromise((resolve) => {
  setTimeout(() => resolve("Hello"), 100);
});
p1.then((val) => console.log(val));
// Output: "Hello"

// ---- TEST 2: Chaining ----
const p2 = new MyPromise((resolve) => resolve(5));
p2.then((val) => val + 10).then((val) => console.log(val));
// Output: 15

// ---- TEST 3: Catch ----
const p3 = new MyPromise((_, reject) => reject("Error!"));
p3.catch((err) => console.log(err));
// Output: "Error!"

// ---- TEST 4: Async resolve ----
const p4 = new MyPromise((resolve) => {
  setTimeout(() => resolve("Async done"), 200);
});
p4.then((val) => console.log(val));
// Output: "Async done"
```
