```typescript
/**
 * Lexical Scope:
 * 
 * - Lexical Scope ka matlab hai: "Function apne variables kahan se 
 *   milenge, yeh decide hota hai uske LIKHE HUE code ke basis pe 
 *   (jahan wo physically likha gaya hai), na ki uske call hone ke 
 *   basis pe."
 * - Agar ek function doosre function ke andar likha hai, toh andar 
 *   wala function bahar wale function ke variables access kar sakta hai.
 */

/**
 * Q. What will be the output and why?
 */
function outer() {
    let name = "Aslam"; // outer function ka variable

    function inner() {
        // inner function apne parent (outer) ka 'name' access kar raha hai
        console.log(name); 
    }

    inner(); // inner function ko call kiya
}

outer(); // Output: "Aslam"

/**
 * Explanation:
 * 1. 'inner' function 'outer' ke andar physically likha gaya hai.
 * 2. Isliye 'inner' ko apne lexical environment mein 'name' variable 
 *    mil jata hai. Yahi Lexical Scope hai.
 */
```

```typescript
/**
 * Closure:
 * 
 * - Closure ek aisa function hai jo apne outer function ke variables ko 
 *   "yaad" rakhta hai, EVEN AFTER the outer function has finished 
 *   executing (return ho chuka hai).
 * - Simple words: Function ne apne birthplace ki cheezein apne saath 
 *   bandh (bind) kar li hain.
 */

/**
 * Q. What will be the output and why?
 */
function createCounter() {
    let count = 0; // Outer function ka private variable

    return function () {
        count++; // Outer function ka 'count' access kar rahe hain
        console.log(count);
    };
}

const counter = createCounter(); // createCounter() ne function return kiya aur gayab ho gaya
counter(); // Output?
counter(); // Output?

/**
 * Explanation:
 * 1. createCounter() call hote hi usne ek function return kiya aur 
 *    apna kaam khatam kar diya. Call stack se wo hata ho gaya.
 * 2. Lekin 'counter' variable ab bhi 'count' ko access kar pa raha hai!
 * 3. Kyun? Kyunki return hua function ek "Closure" ban gaya hai. Usne 
 *    apne lexical scope ka 'count' variable apne saath bandh liya hai.
 * 4. Pehli call pe count 1 ho gaya. Doosri call pe wahi closure wapas 
 *    usi 'count' ko 1 se 2 par le gaya.
 * 
 * Output:
 * 1
 * 2
 */
```

```typescript
/**
 * Data Hiding (Private Variables):
 * 
 * - Closure ka sabse bada use case hai variables ko private banana.
 * - JS mein class se pehle private variables banane ka ek hi tarika 
 *   tha: Closures.
 */

/**
 * Q. Can we access 'secret' directly?
 */
function createBankAccount() {
    let balance = 1000; // Private variable, nobody can touch this directly

    return {
        deposit: function (amount) {
            balance += amount;
            console.log("Balance:", balance);
        },
        withdraw: function (amount) {
            if (amount <= balance) {
                balance -= amount;
                console.log("Balance:", balance);
            } else {
                console.log("Insufficient funds!");
            }
        }
    };
}

const myAccount = createBankAccount();

myAccount.deposit(500);   // Output: Balance: 1500
myAccount.withdraw(2000); // Output: Insufficient funds!
myAccount.withdraw(500);  // Output: Balance: 1000

console.log(myAccount.balance); // Output: undefined (Cannot access directly!)

/**
 * Explanation:
 * 1. 'balance' variable ko hum direct access nahi kar sakte.
 * 2. Sirf un functions ke through access kar sakte hain jo closure 
 *    bana rahe hain (deposit, withdraw).
 * 3. Is tarah humne apne data ko protect kar liya unauthorized access se.
 */
```

```typescript
/**
 * The 'var' vs 'let' Loop Problem:
 * 
 * Interviewer ka sabse favourite sawaal!
 */

/**
 * Q. What will be the output?
 */
for (var i = 0; i < 3; i++) {
    setTimeout(function () {
        console.log(i);
    }, 1000);
}

/**
 * Explanation:
 * 1. var function-scoped (ya global) hota hai. Loop mein ek hi 'i' 
 *    variable baar-baar update hota hai.
 * 2. setTimeout ke andar wala function ek closure hai jo 'i' ka 
 *    reference yaad rakhta hai.
 * 3. Jab 1 second baad callbacks chale, tab tak loop khatam ho chuka 
 *    hai aur 'i' ki value 3 ho chuki hai.
 * 4. Saare callbacks usi 3 ko print karenge.
 * 
 * Output:
 * 3
 * 3
 * 3
 */

// FIX 1: Use 'let' instead of 'var'
for (let j = 0; j < 3; j++) {
    setTimeout(function () {
        console.log(j);
    }, 1000);
}
/**
 * Explanation:
 * 1. let block-scoped hota hai. Har loop iteration ek NAYA block scope 
 *    banata hai.
 * 2. Har iteration ka apna alag 'j' variable hota hai memory mein.
 * 3. Closure us specific iteration ke 'j' ko yaad rakhta hai.
 * 
 * Output:
 * 0
 * 1
 * 2
 */

// FIX 2: Use IIFE (Immediately Invoked Function Expression) - The old way
for (var k = 0; k < 3; k++) {
    (function (innerK) {
        setTimeout(function () {
            console.log(innerK);
        }, 1000);
    })(k);
}
/**
 * Explanation:
 * 1. IIFE ek naya function scope banata hai har iteration mein.
 * 2. Hum 'k' ki value 'innerK' mein pass kar rahe hain, jo us block 
 *    ke liye private ho jati hai.
 * 
 * Output:
 * 0
 * 1
 * 2
 */
```

```typescript
/**
 * Garbage Collection & Memory Leaks:
 * 
 * - JS engine memory manage karta hai. Jab koi variable use nahi ho 
 *   raha, toh usko memory se hata deta hai (Garbage Collection).
 * - Lekin Closure ki wajah se kabhi-kabhi variables memory mein 
 *   reh jate hain, even if hume unki zaroorat nahi.
 */

/**
 * Q. Will 'largeData' be garbage collected?
 */
function outer() {
    let largeData = new Array(1000000).fill('*'); // Bahut bada array

    return function inner() {
        console.log("Inner function called");
        // Notice: 'largeData' use nahi ho raha 'inner' mein!
    };
}

const fn = outer();
fn();

/**
 * Explanation:
 * 1. 'inner' function ek closure hai. Wo apne lexical scope ko yaad 
 *    rakhta hai.
 * 2. Modern JS engines (V8) bahut smart hain. Wo dekhte hain ki 'inner' 
 *    function ke andar 'largeData' use ho raha hai ya nahi.
 * 3. Kyunki 'largeData' use nahi ho raha, smart engine usko garbage 
 *    collect kar deta hai memory bachane ke liye.
 * 4. Lekin agar 'inner' function mein 'largeData' use hota, toh wo 
 *    memory mein tab tak rakha jata jab tak 'fn' exist karta.
 */
```

```typescript
/**
 * 1. Basic Closure
Explanation:

outer() call hote hi usne inner function return kar diya. outer ka execution khatam ho gaya.
Lekin inner ek closure hai, isliye count variable ko apne saath bandh (bind) kar laya.
Jab counterFunc() pehli baar chala, count 0 se 1 ho gaya.
Doosri baar chale, toh usi closure ke count (jo ab 1 tha) ko 2 par le gaya. count reset nahi hua.

*/

function outer() {
    let count = 0;
    function inner() {
        count++;
        console.log(count);
    }
    return inner;
}

const counterFunc = outer();
counterFunc(); // Output?
counterFunc(); // Output?
```

```typescript
/**
 * 3. Block Scope + Closure
*/

function createFunctions() {
    let funcs = [];
    for (let i = 0; i < 3; i++) {
        funcs.push(function() {
            console.log(i);
        });
    }
    return funcs;
}

const myFuncs = createFunctions();
myFuncs[0](); // Output?
myFuncs[1](); // Output?
myFuncs[2](); // Output?
```

```typescript
/**
 * 4. Parameter Binding (Tricky)
*/

function createAdder(base) {
    return function (num) {
        return base + num;
    };
}

const add5 = createAdder(5);
console.log(add5(10)); // Output?
```

```typescript
/**
 * 5. Multiple Instances (Independent Closures)
*/

function counter() {
    let count = 0;
    return function () {
        count++;
        return count;
    };
}

const counter1 = counter();
const counter2 = counter();

console.log(counter1()); // Output?
console.log(counter1()); // Output?
console.log(counter2()); // Output?
console.log(counter1()); // Output?
```

```typescript
/**
 * 6. Shadowing in Closures:
*/

let x = 10;

function outer() {
    let x = 20;
    function inner() {
        let x = 30;
        console.log(x);
    }
    inner();
}

outer();
// Output?
```

```typescript
/**
 * 7. Function Hoisting inside Closure
*/

var x = 1;

function foo() {
    var x = 2;
    function bar() {
        console.log(x); 
    }
    x = 3;
    return bar;
}

var baz = foo();
baz(); // Output?
```

```typescript
/**
 * 9. Object Methods & Closures
*/

function createAccount(initialBalance) {
    let balance = initialBalance;

    return {
        deposit: function(amount) {
            balance += amount;
        },
        getBalance: function() {
            return balance;
        }
    };
}

const acc1 = createAccount(100);
acc1.deposit(50);
console.log(acc1.getBalance()); // Output?
```

```typescript
/**
 * 10. The Tricky Return
*/

function trickyClosure() {
    let funcs = [];
    for (var i = 0; i < 3; i++) {
        funcs[i] = function() {
            return i;
        };
    }
    i = 10; // Notice this line!
    return funcs;
}

const myFuncs = trickyClosure();
console.log(myFuncs[0]()); // Output?
console.log(myFuncs[1]()); // Output?
console.log(myFuncs[2]()); // Output?
```

```typescript
/**
 * 11. Hoisting and Closures
*/

var a = 1;

function outer() {
    var b = 2;
    function inner() {
        console.log(a); 
        console.log(b); 
    }
    return inner;
}

var innerFunc = outer();
a = 10;
b = 20; // Global scope mein 'b' nahi hai, ye error dega agar 'var' nahi lagao
// but for the sake of question, assume it runs
innerFunc();
```
