```typescript
/**
 * The Module System: Import/Export
 * - Named imports and exports provide a convenient and modular way to
 *   organize and reuse code in JavaScript applications.
 * - They allow developer to selectively import and export only the code
 *   that they need, which can help to reduce the size of the codebase
 *   and improve maintainability. 
*/

export const add = (a, b) => a + b;
export const square = (a) => a * a;
```

```typescript
import { add, square } from './utils';

console.log(add(1, 2));
console.log(square(5));
```

```typescript
/**
 * The Module System: default export
 * - Default exports provide a convenient way to export a single, often
 *   used entity from a module without requiring other modules to use
 *   a specific name or syntax to import it.
 * - The key difference between named and default import/export is that
 *   named exports allow multiple functions to be exported from a module,
 *   whereas default exports only allow one function to be designated
 *   as the default export and need to be used without parentheses.
*/

const square = (a) => a * a;

export default square;
```

```typescript
import square from './utils';
```

```typescript
/**
 * Import/Export with rename:
 * - 'add' is defined in a module '/utils.js' and export using the 
 *   'export' keyword with a renamed identifier using the 'as' keyword.
 * - The renamed 'sum' is exported instead of the original 'add', so
 *   that can be imported with the new name using the 'import' statement
 *   in another module.
 * - In the '/app.js' module, the 'import' keyword is used to import
 *   the 'sum' from the 'utils.js', but it is given a new name 'xyz'
 *   using the 'as' keyword.
 * - This allows the function to be used in the importing module with
 *   the new name while still referring to the original function defined
 *   in the exporting module.
 * 
 * - Renaming imports and exports can be helpful in several scenarios,
 *   such as:
 *   1. Avoiding naming conflicts
 *   2. Simplifying the code
 *   3. Adapting to renaming conventions:
*/

const add = (x, y) => x + y;

export { add as sum };
```

```typescript
import { sum as xyz } from './utils';

console.log(xyz(2, 5));
```
