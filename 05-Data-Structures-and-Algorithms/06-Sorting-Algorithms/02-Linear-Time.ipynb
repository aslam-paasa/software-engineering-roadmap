```python
# 1. Counting Sort [Low Priority]

def counting_sort(arr):
    min_val = min(arr)
    max_val = max(arr)
    count_range = max_val - min_val + 1
    count = [0] * count_range

    # Phase 1: Count occurrences
    for num in arr:
        count[num - min_val] += 1

    # Phase 2: Cumulative counts
    for i in range(1, count_range):
        count[i] += count[i - 1]

    # Phase 3: Build output
    output = [0] * len(arr)
    for i in range(len(arr) - 1, -1, -1):
        output[count[arr[i] - min_val] - 1] = arr[i]
        count[arr[i] - min_val] -= 1

    # Copy back
    for i in range(len(arr)):
        arr[i] = output[i]

    return arr
```

```python
# 2. Bucket Sort [High Priority]

def bucket_sort(arr):
    n = len(arr)
    min_val = min(arr)
    max_val = max(arr)
    k = max(1, n // 2)
    bucket_size = max(1, (max_val - min_val) // k + 1)

    # Create buckets
    buckets = [[] for _ in range(k)]

    # Distribute elements
    for num in arr:
        idx = min(k - 1, (num - min_val) // bucket_size)
        buckets[idx].append(num)

    # Sort each bucket (insertion sort)
    result = []
    for bucket in buckets:
        for i in range(1, len(bucket)):
            key = bucket[i]
            j = i - 1
            while j >= 0 and bucket[j] > key:
                bucket[j + 1] = bucket[j]
                j -= 1
            bucket[j + 1] = key
        result.extend(bucket)

    # Copy back
    for i in range(n):
        arr[i] = result[i]

    return arr
```

```python
# 3. Radix Sort

def radix_sort(arr):
    max_val = max(arr)
    exp = 1

    while max_val // exp > 0:
        n = len(arr)
        count = [0] * 10
        output = [0] * n

        # Count digits
        for num in arr:
            digit = (num // exp) % 10
            count[digit] += 1

        # Make cumulative
        for i in range(1, 10):
            count[i] += count[i - 1]

        # Build output
        for i in range(n - 1, -1, -1):
            digit = (arr[i] // exp) % 10
            output[count[digit] - 1] = arr[i]
            count[digit] -= 1

        # Copy back
        for i in range(n):
            arr[i] = output[i]

        exp *= 10

    return arr
```
