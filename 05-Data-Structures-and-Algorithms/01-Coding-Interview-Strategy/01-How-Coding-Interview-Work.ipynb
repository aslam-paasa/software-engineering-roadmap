```python
# Round-1: Phone Screen
# - Phone screen last 30-60 min conduced over shared editor like Hackerrank, Notepad.
# - Problems tend to be leetcode easy to medium.
```

```python
# Round-2: Onsite (Virtual or In-Person)
# - This is a block of 4-6 interviews.
# - Each round lasts 45-60 min and covers a different area.
#   - Round 1 : 45 min - Data structures and algorithms
#   - Round 2 : 45 min - Data structures and algorithms
#   - Round 3 : 45 min - System design (senior+)
#   - Round 4 : 45 min - Behavioral / culture fit
#   - Round 5 : 45 min - Coding or domain-specific
# - Each interviewer submits independent feedback, and hiring manager reviews the
#   full set of rounds together.
```

```python
# Round-3: LLD
# - Here, we share our screen and write code in real IDE.
# - This round checks whether a candidate can write code that compiles, passes
#   test cases, and get debugged on the spot.
# - Instead of a single algorithm, we build a small working component from a loose
#   specification, like an in-memory cache, a parking lot, or a splitwise-style 
#   expense tracker.
# - This round run longer, often 90-120 min.
```

```python
# Round-4: Take Home Assignment
# - Interviewer gives us a project (usually 2-6 hrs of work) to complete on our own time.
# - We might build a small API, implement a feature, or solve a more complex algo.
# - We typically get 3-7 days to submit.
# - Companies expect:
#   - cleaner code
#   - better test coverage
#   - thoughtful architecture.
```
