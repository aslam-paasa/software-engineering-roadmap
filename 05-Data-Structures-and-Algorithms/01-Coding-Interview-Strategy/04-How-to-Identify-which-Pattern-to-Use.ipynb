```python
/**
 * How to Identify Which Pattern to Use:
 * - Pattern ka concept samajhna ek cheez hai (jaise sliding window ya binary search 
 *   kaise kaam karta hai). 
 * - Lekin naya problem dekh ke turant pehchan lena ki "yaha konsa pattern lagega" - 
 *   ye ek alag hi skill hai. 
 * - Interview mein log yahi pe atak jaate hai - unko pattern aata hai, but connect 
 *   nahi kar paate problem ke sath.
 * - Is guide mein hum ek REPEATABLE tareeka banayenge - 3 signals use karke problem ko
 *   1-2 pattern tak narrow down karenge:
 *   a. Input Type  (data kis form mein hai)
 *   b. Constraints (n kitna bada hai)
 *   c. Keywords    (problem statement mein kaunse words hai)
 * 
 * 
 * SIGNAL 1: INPUT TYPE (Sabse pehla aur sabse strong hint)
 * - Jo data structure input mein diya hai, wahi sabse bada clue hai.
 * - Har data structure ke sath kuch fixed patterns baar baar aate hai.
 * - Isse hi lambi list of patterns turant chhoti ho jaati hai.
 * 
 *  +----------------+--------------------------------------------------------------------+
 *  | Input Type     |  Common Patterns                                                   |
 *  +----------------+--------------------------------------------------------------------+
 *  | Sorted Array   |  Binary Search, Two Pointers                                       |
 *  | Unsorted Array |  Hash Map, Sort+Two Pointers, Sliding Window, Prefix Sum           |
 *  | String         |  Sliding Window, Two Pointers, Hash Map, Trie                      |
 *  | Linked List    |  Two Pointers (fast/slow), In-place Reversal                       |
 *  | Binary Tree    |  DFS (recursion), BFS (level order), Path-based traversal          |
 *  | Graph          |  BFS, DFS, Union Find, Topological Sort, Shortest Path algorithms  |
 *  | Matrix/Grid    |  BFS/DFS (flood fill), DP                                          |
 *  | Intervals      |  Sort + Greedy, Merge/Sweep line                                   |
 *  +----------------+--------------------------------------------------------------------+ 
 *   
 * Note: 
 * - Ye table 100% fixed rule nahi hai. 
 * - Kabhi kabhi 2-3 pattern ek sath bhi mix hote hai, ya problem is table se bahar
 *   bhi ho sakta hai. 
 * - Lekin starting filter ke liye ye kaafi kaam aata hai.
 * 
 * 
 * SIGNAL 2: CONSTRAINTS (n dekh ke pata chal jaata hai)
 * - "n" (input size) bata deta hai ki kaunsi complexity time limit mein PASS hogi. 
 * - Isse bahut saare patterns turant reject ho jaate hai bina try kiye.
 * - General rule: Modern computer roughly 10^8 (10 crore) simple operations per second
 *   kar sakta hai. Isi hisaab se:
 * 
 *   +-----------------+----------------------+-----------------------------------------------------+
 *   | n ki value      |  Max Complexity OK   |  Typical Approach                                   |
 *   | ----------------|--------------------- |-----------------------------------------------------+
 *   | n <= 12         |  O(2^n), O(n!)       |  Backtracking, bitmask DP, brute force              |
 *   | n <= 20-25      |  O(2^n) with pruning |  Backtracking + memo, meet in the middle            |
 *   | n <= 500        |  O(n^3)              |  Triple nested loop, Floyd-Warshall                 |
 *   | n <= 5,000      |  O(n^2)              |  DP with 2D table, nested loops                     |
 *   | n <= 100,000    |  O(n log n)          |  Sorting, binary search, heap, merge sort           |
 *   | n <= 10,000,000 |  O(n)                |  Hash map, prefix sum, sliding window, two pointers |
 *   | n > 10^8        |  O(log n) or O(1)    |  Math formula, binary search on answer              |
 *   +-----------------+----------------------+-----------------------------------------------------+
 * 
 * Example: 
 * - Agar problem mein "n <= 10^5" likha hai, to turant samajh jao ki O(n^2) wala 
 *   brute force nested loop nahi chalega.
 * - Seedha O(n log n) ya O(n) wale pattern (sorting, binary search, hashmap, 
 *   sliding window) ki taraf jao.
 * 
 * 
 * SIGNAL 3: KEYWORDS (Problem statement ke words)
 * - Problem ke andar kuch phrases hote hai jo directly kisi pattern ki taraf point 
 *   karte hai:
 * 
 *   +---------------------------------------+-------------------------------------------------+
 *   | Keyword/Phrase                        | Pattern                                         |
 *   +---------------------------------------+-------------------------------------------------+
 *   | 1. "Contiguous subarray"              | Sliding Window / Prefix Sum                     |
 *   | 2. "Subarray sum"                     | Prefix Sum + Hash Map                           |
 *   | 3. "Shortest path"                    | BFS (unweighted) / Dijkstra (weighted)          |
 *   | 4. "All combinations" / "all subsets" | Backtracking                                    |
 *   | 5. "All permutations"                 | Backtracking                                    |
 *   | 6. "Number of ways"                   | Dynamic Programming                             |
 *   | 7. "Minimum/maximum steps"            | BFS or DP                                       |
 *   | 8. "Can you reach" / "is it possible" | BFS, DFS, or DP                                 |
 *   | 9. "Sorted array" + "find target"     | Binary Search                                   |
 *   | 10. "Merge intervals"/"overlapping"   | Sort + Greedy                                   |
 *   | 11. "Connected components"            | Union Find or DFS                               |
 *   | 12. "Cycle detection"                 | DFS (directed graph) or Union Find (undirected) |
 *   | 13. "kth largest/smallest"            | Heap (priority queue)                           |
 *   | 14. "Top k" / "k most frequent"       | Heap or Bucket Sort                             |
 *   | 15. "Palindrome"                      | Two Pointers or DP                              |
 *   | 16. "Parentheses"/"matching pairs"    | Stack                                           |
 *   | 17. "Next greater/smaller element"    | Monotonic Stack                                 |
 *   | 18. "Longest increasing subsequence"  | DP or Binary Search + Patience Sorting          |
 *   | 19. "In-place"                        | Two Pointers or swap-based                      |
 *   +---------------------------------------+-------------------------------------------------+
 * 
 * - Yaad rakho: 
 *   - Ek keyword se 100% confirm nahi hota, sirf options kam ho jaate hai. 
 *   - Jab teeno signals (input type + constraints + keyword) ek sath dekhte ho, 
 *     tab final answer clear ho jaata hai.
 * 
 * PATTERN SELECTION FLOWCHART:
 * - Ek naya problem aaya? Ye steps follow karo:
 *   1. Input type kya hai?
 *      - Array/String, 
 *      - Tree, 
 *      - Graph, 
 *      - Linked List, ya Intervals?
 *       (Isse 4-5 candidate pattern bachte hai)
 * 
 *   2. Operation/question kya puch raha hai?
 *      - Target dhundna? 
 *      - Pair/Triplet? 
 *      - Subarray? 
 *      - All combinations?
 *      -  Optimal value (min/max)? 
 *      - kth element? 
 *      - Shortest path?
 *      - Connectivity? 
 *      - Cycle detection? 
 *      - Ordering?
 *       (Isse 2-3 candidate bachte hai)
 * 
 *   3. Constraints + keywords use karke FINAL ek pattern chuno.
 * 
 * - Is tarah "20 patterns mein se konsa?" wala confusing sawaal, chhote chhote simple
 *   sawalon mein badal jaata hai:
 *   - Input type        -> 4-5 candidates
 *   - Operation         -> 2-3 candidates
 *   - Constraints+words -> 1 strong final candidate
 * 
 * 
 * PRACTICE MEIN KAISE LAGTA HAI - 5 EXAMPLES
 * 
 *   Example 1: Maximum Subarray Sum
 *   Problem: Array diya hai, sabse bada sum wala contiguous subarray
 *   dhundo.
 *     - Input type: Array
 *     - Keyword: "contiguous subarray" -> sliding window ya prefix
 *       sum sochte hai, lekin numbers negative bhi ho sakte hai, isse
 *       simple sliding window fail ho jaata hai.
 *     - Constraint: n ~10^5, isliye O(n) ya O(n log n) chahiye.
 *     - FINAL PATTERN: Kadane's Algorithm (1D Dynamic Programming).
 *       "contiguous subarray" + "largest sum" ye combo seedha
 *       Kadane's ki taraf point karta hai.
 * 
 *   Example 2: Number of Islands
 *   Problem: 2D grid mein '1' (land) aur '0' (water) hai. Islands
 *   count karo (adjacent lands connected hoke island banate hai).
 *     - Input type: Matrix/Grid -> BFS ya DFS
 *     - Keyword: "connected", "adjacent", "count the number of"
 *       -> connected components count karna hai
 *     - Constraint: grid 300x300 tak, O(m*n) easily chalega
 *     - FINAL PATTERN: DFS/BFS flood fill. Har unvisited '1' se
 *       traversal start karo, jo bhi connect ho wo visited mark
 *       karo. Har naya traversal = ek naya island.
 * 
 *   Example 3: Course Schedule
 *   Problem: n courses hai, kuch courses ke prerequisites hai. Pata
 *   karo saare courses complete ho sakte hai ya nahi.
 *     - Input type: Directed Graph (course = node, prerequisite =
 *       edge)
 *     - Keyword: "prerequisites" -> ordering; "can you finish all"
 *       -> possible hai ya nahi
 *     - Constraint: n <= 2000, edges <= 5000 -> O(V+E) chalega
 *     - FINAL PATTERN: Directed graph mein Cycle Detection. Agar
 *       cycle hai to complete karna impossible hai. DFS ya Kahn's
 *       algorithm (topological sort) se cycle check karo.
 * 
 *   Example 4: Find Median from Data Stream
 *   Problem: Numbers ek ek karke aa rahe hai, hume har baar current
 *   median chahiye.
 *     - Input type: Stream of numbers (koi fixed array nahi)
 *     - Keyword: "median" -> beech ka element chahiye, wo bhi
 *       dynamically badhte data mein
 *     - Constraint: ~50,000 calls, har operation fast hona chahiye
 *     - FINAL PATTERN: Do Heaps - ek Max-Heap (lower half ke liye)
 *       aur ek Min-Heap (upper half ke liye). Dono ka top milake
 *       median milta hai.
 * 
 *   Example 5: Word Break
 *   Problem: String s aur dictionary of words di hai. Check karo ki
 *   s ko dictionary words se break kiya ja sakta hai ya nahi.
 *     - Input type: String
 *     - Keyword: "can be segmented" -> "is it possible" -> DP ka
 *       strong hint (ek choice, agla choice affect karta hai)
 *     - Constraint: string length <= 300, O(n^2) chalega
 *     - FINAL PATTERN: Dynamic Programming. dp[i] = "kya pehle i
 *       characters break ho sakte hai?" Har position pe pichle
 *       saare possible words check karo.
 * 
 * 
 * COMPLETE CHEAT SHEET:
 * 
 *   Problem Characteristic              Pattern           Complexity
 *   ---------------------------------------------------------------
 *   Sorted array mein pair dhundna      Two Pointers       O(n)
 *   Unsorted array mein pair dhundna    Hash Map           O(n)
 *   Fixed size contiguous subarray      Sliding Window     O(n)
 *                                        (fixed)
 *   Condition wali contiguous subarray  Sliding Window     O(n)
 *                                        (variable)
 *   Subarray sum equal to k             Prefix Sum +       O(n)
 *                                        Hash Map
 *   Sorted array mein target dhundna    Binary Search      O(log n)
 *   Linked list mein cycle detect       Fast & Slow        O(n)
 *                                        Pointers
 *   Tree ka level-order traversal       BFS with Queue     O(n)
 *   Tree path problems                  DFS (recursion)    O(n)
 *   Shortest path (unweighted)          BFS                O(V+E)
 *   Shortest path (weighted, no neg)    Dijkstra           O(E log V)
 *   Connected components                Union Find/DFS     O(V+E)
 *   Dependency wali task ordering       Topological Sort   O(V+E)
 *   Saare subsets/combinations          Backtracking       O(2^n)
 *   Saare permutations                  Backtracking       O(n!)
 *   Choices ke sath optimal value       Dynamic            Varies
 *                                        Programming
 *   kth largest/smallest                Heap               O(n log k)
 *   Next greater/smaller element        Monotonic Stack    O(n)
 *   Overlapping intervals               Sort + Greedy      O(n log n)
 *   String pattern matching             Trie/Hash Map      Varies
 * 
 * 
 *   --------------------------------------------------------------
 *   JAB EK SE ZYADA PATTERN FIT LAGE, TO KYA KARE?
 *   --------------------------------------------------------------
 * 
 *   1) Constraints se pehle eliminate karo (sabse sasta filter):
 *      - n bada hai (~10^5 ya usse zyada) -> O(n^2) bahar
 *      - n chhota hai (<=20) -> brute force / backtracking wapas
 *        list mein aa sakta hai
 * 
 *   2) Problem kya optimize kar raha hai, wo dekho:
 *      - Global optimum (min/max/total ways) -> Dynamic Programming
 *      - Local decisions jo global answer bana de -> Greedy
 *      - Kam se kam steps/levels -> BFS
 * 
 *   3) Output ka type dekho:
 *      - "All possible results" chahiye -> Backtracking
 *      - "Count of ways" ya "best value" -> Dynamic Programming
 *      - "Is it possible" -> BFS, DFS, ya kabhi kabhi DP
 * 
 *   4) Simple se start karo, phir improve karo:
 *      Agar confusion abhi bhi hai, to ek slow-but-correct solution
 *      pehle likho. Baad mein use optimize karna, ekdum shuru se
 *      best solution sochne se aasan hota hai.
 * 
 * 
 *   --------------------------------------------------------------
 *   FINAL NOTE
 *   --------------------------------------------------------------
 *   Ye pattern-recognition tareeka zyadatar interview problems ke
 *   liye kaam karega, lekin har problem is table mein fit nahi
 *   hoga. Kuch problems bilkul naye lagenge jo pehle dekhe hue kisi
 *   pattern se match nahi karenge - unke liye scratch se sochna
 *   padega. (Wo agla topic hai.)
 * ====================================================================
*/
```
