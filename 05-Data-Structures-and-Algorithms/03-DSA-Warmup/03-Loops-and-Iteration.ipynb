```python
# Problem 1 - Print Numbers from 1 to N b
```

```python
class Solution:
    def printNumbers(self, n: int) -> list[int]:
        # Count upward and store each value in order
        result = []
        for i in range(1, n + 1):
            result.append(i)
        return result
```

```python
# Problem 2 - Multiplication Table of a Number
```

```python
class Solution:
    def multiplicationTable(self, n: int) -> list[int]:
        result = [0] * 10
        # Index i - 1 holds the product n * i
        for i in range(1, 11):
            result[i - 1] = n * i
        return result
```

```python
class Solution:
    def multiplicationTable(self, n: int) -> list[int]:
        result = [0] * 10
        total = 0
        # Add n each step so total equals n * i
        for i in range(1, 11):
            total += n
            result[i - 1] = total
        return result
```

```python
# Problem 3 - Sum of Even or Odd Numbers
```

```python
class Solution:
    def sum_by_parity(self, n: int, parity: str) -> int:
        # Even numbers leave remainder 0, odd numbers leave remainder 1
        target = 0 if parity == "even" else 1
        total = 0
        for i in range(1, n + 1):
            if i % 2 == target:
                total += i
        return total
```

```python
class Solution:
    def sum_by_parity(self, n: int, parity: str) -> int:
        # Even numbers start at 2, odd numbers start at 1
        start = 2 if parity == "even" else 1
        total = 0
        for i in range(start, n + 1, 2):
            total += i
        return total
```

```python
class Solution:
    def sum_by_parity(self, n: int, parity: str) -> int:
        if parity == "even":
            # k even numbers exist from 1 to n, and their sum is k * (k + 1)
            k = n // 2
            return k * (k + 1)
        # k odd numbers exist from 1 to n, and their sum is k * k
        k = (n + 1) // 2
        return k * k
```

```python
# Problem 4 - Count Digits in an Integer
```

```python
class Solution:
    def countDigits(self, n: int) -> int:
        # Zero is written as a single symbol, so it has 1 digit
        if n == 0:
            return 1
        # Work on the absolute value so the loop counts the same digits
        n = abs(n)
        count = 0
        # Each division by 10 removes one digit
        while n != 0:
            n //= 10
            count += 1
        return count
```

```python
class Solution:
    def countDigits(self, n: int) -> int:
        # The string of the absolute value contains only digit characters
        return len(str(abs(n)))
```

```python
import math

class Solution:
    def countDigits(self, n: int) -> int:
        # Zero has no logarithm, so handle it directly
        if n == 0:
            return 1
        return int(math.floor(math.log10(abs(n)))) + 1
```

```python
# Problem 5 - Sum of All Divisors of a Number
```

```python
class Solution:
    def sumOfDivisors(self, n: int) -> int:
        # Python integers grow as needed, so there is no overflow to manage
        total = 0
        for i in range(1, n + 1):
            if n % i == 0:
                total += i
        return total
```

```python
class Solution:
    def sumOfDivisors(self, n: int) -> int:
        # Each divisor i below the square root pairs with n // i above it
        total = 0
        i = 1
        while i * i <= n:
            if n % i == 0:
                total += i
                pair = n // i
                # Add the partner only when it is a different value
                if pair != i:
                    total += pair
            i += 1
        return total
```
