```python
# Problem 1 - Even or Odd
```

```python
class Solution:
    def evenOrOdd(self, n: int) -> str:
        # Compare the remainder against 0 so negatives are handled correctly
        if n % 2 != 0:
            return "Odd"
        return "Even"
```

```python
class Solution:
    def evenOrOdd(self, n: int) -> str:
        # The lowest bit is 1 for odd numbers and 0 for even numbers
        if n & 1:
            return "Odd"
        return "Even"
```

```python
# Problem 2 - Positive, Negative or Zero
```

```python
class Solution:
    def classify(self, n: int) -> str:
        if n > 0:
            return "Positive"
        elif n < 0:
            return "Negative"
        else:
            # Neither greater nor less than 0, so it must be 0
            return "Zero"
```

```python
# Problem 3 - Leap Year Check
```

```python
class Solution:
    def isLeapYear(self, year: int) -> bool:
        # Divisible by 4, and either not a century year or divisible by 400
        return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)
```

```python
# Problem 4 - Grade from Marks
```

```python
class Solution:
    def grade(self, marks: int) -> str:
        # Check the highest band first so each test needs only one threshold
        if marks >= 90:
            return "A"
        elif marks >= 80:
            return "B"
        elif marks >= 70:
            return "C"
        elif marks >= 60:
            return "D"
        else:
            return "F"
```

```python
# Problem 5 - Valid Triangle from Three Sides
```

```python
class Solution:
    def isValidTriangle(self, a: float, b: float, c: float) -> bool:
        # Every pair of sides must sum to strictly more than the third
        return a + b > c and a + c > b and b + c > a
```

```python
# Problem 6 - Vowel or Consonant
```

```python
class Solution:
    def vowelOrConsonant(self, ch: str) -> str:
        # Normalize to lowercase so case does not matter
        c = ch.lower()
        if c in ("a", "e", "i", "o", "u"):
            return "Vowel"
        return "Consonant"
```

```python
# Problem 7 - Check If Three Points are Collinear
```

```python
class Solution:
    def areCollinear(self, x1: int, y1: int, x2: int, y2: int, x3: int, y3: int) -> bool:
        # Cross product of the two direction vectors from point 1
        cross = (y2 - y1) * (x3 - x1) - (y3 - y1) * (x2 - x1)
        return cross == 0
```
