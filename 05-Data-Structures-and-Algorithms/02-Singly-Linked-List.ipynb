```python
class Node:
    def __init__(self, information, next=None):
        self.data = information
        self.next = next
```

```python
# ============================================================
# Singly Linked List
# ============================================================

class SinglyLinkedList:

    def __init__(self, head=None):
        # 'head' stores the first node of the linked list.
        self.head = head


    # ========================================================
    # Insert a Node at the End
    # ========================================================

    def insertAtEnd(self):

        # Create a new node with value 30.
        new_node = Node(30)


        # ----------------------------------------------------
        # Case 1: Linked List is Empty
        # ----------------------------------------------------

        # If there is no first node, make the new node the first node.
        if self.head is None:
            self.head = new_node
            return


        # ----------------------------------------------------
        # Case 2: Linked List is Not Empty
        # ----------------------------------------------------

        # Start from the first node.
        temp = self.head

        # Move forward until we reach the last node.
        # The last node is the node whose 'next' is None.
        while temp.next is not None:
            temp = temp.next

        # Connect the new node after the last node.
        temp.next = new_node
```
