```python
# Approach-1: Selection Sort (Brute Force)

class Solution:
    def sortArray(self, nums: list[int]) -> list[int]:
        n = len(nums)
        # For each position, find the minimum in the remaining unsorted portion
        for i in range(n - 1):
            min_idx = i
            for j in range(i + 1, n):
                if nums[j] < nums[min_idx]:
                    min_idx = j
            # Swap the minimum into position i
            nums[i], nums[min_idx] = nums[min_idx], nums[i]
        return nums
```

```python
# Approach-2: Merge Sort

class Solution:
    def sortArray(self, nums: list[int]) -> list[int]:
        if len(nums) <= 1:
            return nums

        self._merge_sort(nums, 0, len(nums) - 1)
        return nums

    def _merge_sort(self, nums: list[int], left: int, right: int) -> None:
        if left >= right:
            return

        mid = left + (right - left) // 2
        self._merge_sort(nums, left, mid)
        self._merge_sort(nums, mid + 1, right)
        self._merge(nums, left, mid, right)

    def _merge(self, nums: list[int], left: int, mid: int, right: int) -> None:
        # Create temp array for the merged result
        temp = []
        i, j = left, mid + 1

        # Compare elements from both halves and pick the smaller one
        while i <= mid and j <= right:
            if nums[i] <= nums[j]:
                temp.append(nums[i])
                i += 1
            else:
                temp.append(nums[j])
                j += 1

        # Copy remaining elements from left half
        while i <= mid:
            temp.append(nums[i])
            i += 1

        # Copy remaining elements from right half
        while j <= right:
            temp.append(nums[j])
            j += 1

        # Copy merged result back to original array
        for p in range(len(temp)):
            nums[left + p] = temp[p]
```

```python
# Approach-3: Randomized Quick Sort

class Solution:
    def sortArray(self, nums: list[int]) -> list[int]:
        self._quick_sort(nums, 0, len(nums) - 1)
        return nums

    def _quick_sort(self, nums: list[int], left: int, right: int) -> None:
        if left >= right:
            return

        # Random pivot to avoid worst-case O(n^2)
        pivot_idx = random.randint(left, right)
        pivot_val = nums[pivot_idx]

        # Move pivot to end
        nums[pivot_idx], nums[right] = nums[right], nums[pivot_idx]

        # Partition: elements <= pivot go to the left side
        store_idx = left
        for i in range(left, right):
            if nums[i] <= pivot_val:
                nums[i], nums[store_idx] = nums[store_idx], nums[i]
                store_idx += 1
        # Place pivot in its final position
        nums[store_idx], nums[right] = nums[right], nums[store_idx]

        # Recursively sort left and right partitions
        self._quick_sort(nums, left, store_idx - 1)
        self._quick_sort(nums, store_idx + 1, right)
```
