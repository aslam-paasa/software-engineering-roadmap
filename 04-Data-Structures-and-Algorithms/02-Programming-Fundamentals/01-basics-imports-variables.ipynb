```python
# 8. MODULAR ARITHMETIC (% operator)
#    Python mein % ka result HAMESHA positive divisor ke saath positive hota hai

print(-7 % 3)   
# Output: 2  
# - kuch languages mein yeh -1 hota hai, Python mein nahi!)
# - Matlab: tumhe ((n % m) + m) % m wala trick karne ki zaroorat nahi Python mein.

# Bahut saare problems bolte hain "answer ko 10^9 + 7 se mod karke do":
MOD = 10**9 + 7
values = [123, 456, 789]
running_result = 0
for val in values:
    running_result = (running_result + val) % MOD
    # HAR step pe mod lagao, sirf end mein nahi -> warna number bahut bada ho sakta hai
```

```python
# 9. BITWISE OPERATORS (bit manipulation ke liye)
#    - XOR trick: array mein ek number sirf ek baar hai, baaki sab 2 baar hain -> use dhundo

nums2 = [4, 1, 2, 1, 2]
single = 0
for num in nums2:
    single ^= num   # XOR se duplicate cancel ho jaate hain, jo bacha wahi answer hai
# single = 4

i = 3
mask = 0
is_set = (mask & (1 << i)) != 0     # bit i set hai ya nahi check karo
mask = mask | (1 << i)              # bit i ko ON (set) karo
mask = mask & ~(1 << i)             # bit i ko OFF (clear) karo
mask = mask ^ (1 << i)              # bit i ko toggle karo (on->off, off->on)

# Table samajhne ke liye:
# &  (AND)   -> masking, check karna ki bit set hai ya nahi
# |  (OR)    -> bit set karna
# ^  (XOR)   -> unique element dhundna, toggle karna
# ~  (NOT)   -> bit ulta karna
# << (left)  -> 2 ki power se multiply (1 << n = 2^n)
# >> (right) -> 2 ki power se divide

print(bin(10))            # '0b1010' -> binary string
print((10).bit_length())  # 4 -> kitne bits chahiye is number ko store karne ke liye

# NOTE: Python mein "unsigned right shift" (>>>) operator nahi hota kyunki Python
# ke integers ki fixed bit-width nahi hoti. Agar 32-bit unsigned simulate karna ho
# (jaise "reverse bits" wale problem mein), to & 0xFFFFFFFF se mask kar do.
```

```python
# 10. CONTROL FLOW (loops)

nums3 = [10, 20, 30]

# Sirf index chahiye:
for i in range(len(nums3)):
    pass

# Ulta (reverse) traversal - DP aur stack problems mein common:
for i in range(len(nums3) - 1, -1, -1):
    pass

# Sirf value chahiye:
for num in nums3:
    pass

# Index AUR value dono chahiye -> enumerate() use karo, yeh sabse clean tarika hai
for i, num in enumerate(nums3):
    pass

# enumerate ko custom starting index bhi de sakte ho:
for i, num in enumerate(nums3, start=1):
    pass   # i ab 1 se shuru hoga

# jab termination condition index-based na ho, tab while use karo
lo, hi = 0, 9
while lo < hi:
    lo += 1
```
