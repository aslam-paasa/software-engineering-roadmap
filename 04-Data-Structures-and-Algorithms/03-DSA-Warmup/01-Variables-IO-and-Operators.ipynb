```python
# Problem 1 - Swap two numbers
```

```python
# Approach-1: Using a Temporary Variable
class Solution:
    def swap(self, a: int, b: int) -> list[int]:
        # Save a so it is not lost when overwritten
        temp = a
        a = b
        b = temp
        return [a, b]
```

```python
# Approach 2 - Without a Temporary Variable
class Solution:
    def swap(self, a: int, b: int) -> list[int]:
        # Encode the sum, then peel each value back out
        a = a + b
        b = a - b
        a = a - b
        return [a, b]
```

```python
# Problem 2 - Temperature Converter
```

```python
# Approach-1: Apply the Conversion Formula
class Solution:
    def convert(self, temp: float, scale: str) -> float:
        if scale == "C":
            # Celsius to Fahrenheit: scale the degrees, then add the offset
            result = temp * 9 / 5 + 32
        else:
            # Fahrenheit to Celsius: remove the offset, then scale back
            result = (temp - 32) * 5 / 9
        # Round to 2 decimal places
        return round(result, 2)
```

```python
# Problem 3 - Simple Interest and Compound Interest
```

```python
# Approach 1 - Apply the Interest Formula

class Solution:
    def interest(self, principal: float, rate: float, time: float) -> list[float]:
        years = int(time)

        # Simple interest grows on the principal only
        simple = principal * rate * time / 100.0

        # Compound factor: multiply (1 + rate/100) once per year
        factor = 1.0
        for _ in range(years):
            factor = factor * (1.0 + rate / 100.0)
        compound = principal * factor - principal

        return [self.round2(simple), self.round2(compound)]

    def round2(self, value: float) -> float:
        return round(value * 100.0 + 1e-9) / 100.0
```

```python
# Problem-3: Convert Seconds to Hours, Minutes, Seconds
```

```python
# Approach-1: Divide and Take the Remainder

class Solution:
    def seconds_to_hms(self, totalSeconds: int) -> list[int]:
        # Whole hours, then carry the leftover forward
        hours = totalSeconds // 3600
        remaining = totalSeconds % 3600

        # Split the leftover into minutes and seconds
        minutes = remaining // 60
        seconds = remaining % 60

        return [hours, minutes, seconds]
```

```python
# Problem 4 - Absolute Value Without Built-In
```

```python
# Approach-1: Conditional Negation

class Solution:
    def absolute_value(self, n: int) -> int:
        # Negate only when n is negative
        if n < 0:
            return -n
        return n
```

```python
# Approach-2: Bit Manipulation

class Solution:
    def absolute_value(self, n: int) -> int:
        # Python ints are unbounded, but for 32-bit values n >> 31
        # still yields -1 for negatives and 0 otherwise, so the
        # same formula works across the test range.
        mask = n >> 31
        return (n ^ mask) - mask
```

```python
# Problem 5 - Quotient and Remainder of a Division
```

```python
# Approach-1: Integer Division and Modulo

class Solution:
    def divide(self, dividend: int, divisor: int) -> list[int]:
        # // performs integer division, keeping only the whole part
        quotient = dividend // divisor
        # % gives the leftover
        remainder = dividend % divisor
        return [quotient, remainder]
```

```python
# Problem 6 - Area and Perimeter of Shapes
```

```python
# Approach-1: Branch by Shape and Apply Formulas

class Solution:
    PI = 3.14159

    # Babylonian method for square root
    def _sqrt(self, x: float) -> float:
        if x == 0:
            return 0.0
        guess = x
        for _ in range(50):
            guess = (guess + x / guess) / 2.0
        return guess

    def _round2(self, value: float) -> float:
        return round(value, 2)

    def area_and_perimeter(self, shape: str, dims: list[float]) -> list[float]:
        area = 0.0
        perimeter = 0.0

        if shape == "rectangle":
            length, width = dims[0], dims[1]
            area = length * width
            perimeter = 2 * (length + width)
        elif shape == "square":
            side = dims[0]
            area = side * side
            perimeter = 4 * side
        elif shape == "circle":
            r = dims[0]
            area = self.PI * r * r
            perimeter = 2 * self.PI * r
        elif shape == "triangle":
            a, b, c = dims[0], dims[1], dims[2]
            perimeter = a + b + c
            s = perimeter / 2.0
            area = self._sqrt(s * (s - a) * (s - b) * (s - c))

        return [self._round2(area), self._round2(perimeter)]
```
