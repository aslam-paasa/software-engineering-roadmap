```python
# ============================================================
#  POLYMORPHISM — Gen-Z Hinglish Guide (super easy, chill vibe)
# ============================================================
#
#  Polymorphism allows the same method name or interface to exhibit
#  different behaviors depending on the object that is invoking it.
#
#  Term Greek se aaya hai — "many forms" (bahut saare roop). Programming
#  mein yeh humein generic, extensible, aur reusable code likhne deta
#  hai, jabki actual behavior runtime ya compile-time pe decide hota
#  hai object ke actual type ke basis pe.
#
#  Simple mein:
#  > Polymorphism lets you call the same method on different objects,
#  > and have each object respond in its own way.
#
#  Tum code likhte ho jo ek "common type" ko target karta hai, par
#  actual behavior "concrete implementation" decide karti hai.
#
#  ------------------------------------------------------------
#  Real-World Analogy — Universal Remote Control
#  ------------------------------------------------------------
#  Ek universal remote soch:
#  - Buttons wahi hain: powerOn(), volumeUp(), mute().
#  - Par device pe depend karta hai: TV, Air Conditioner, ya
#    Projector — har button alag action perform karti hai.
#
#  User ke liye interface (remote) kabhi change nahi hota. Par
#  internally har device same signal ko alag tarike se interpret
#  karta hai.
#
#  Yahi hai POLYMORPHISM IN ACTION. Same interface triggers different
#  behaviors depending on the receiver (device type).
#
# ============================================================
#
#  WHY POLYMORPHISM MATTERS (kyun zaroori hai)
#  --------------------------------------------
#  Chaar concrete benefits:
#  - Loose coupling encourage karta hai: Tum abstractions (interfaces
#    ya base classes) ke saath interact karte ho, specific
#    implementations se nahi.
#  - Flexibility enhance karta hai: Naye behaviors bina existing code
#    ko modify kiye introduce kar sakte ho (Open/Closed Principle).
#  - Scalability promote karta hai: System aur features support karne
#    ke liye grow kar sakta hai with minimal impact on existing code.
#  - Extensibility enable karta hai: Naye implementations "plug in"
#    kar sakte ho bina core business logic ko chhuede.
#
# ============================================================
#
#  HOW POLYMORPHISM WORKS (kaise kaam karta hai)
#  ---------------------------------------------
#  OOP mein polymorphism do forms mein aata hai: compile-time (program
#  run hone se pehle decide) aur runtime (program chalte waqt decide).
#  Dono same method name ko alag behave karne dete hain, par mechanism
#  fundamentally alag hai.
#
#  ------------------------------------------------------------
#  1. Compile-time Polymorphism (Method Overloading)
#  ------------------------------------------------------------
#  Compile-time polymorphism (method overloading) tab hota hai jab
#  same class mein same naam ke multiple methods hote hain par
#  different parameter lists ke saath.
#
#  Compiler decide karta hai kaunsa version call karna hai arguments
#  ki sankhya, types, ya order ke basis pe. Decision program run
#  hone se pehle hota hai.
#
#  Example:
#  # Python natively method overloading support nahi karta.
#  # Same naam ke multiple methods define karoge toh sirf last wala
#  # survive karega. Standard workaround: default args ya *args.
#
#  class Calculator:
#      def add(self, *args):
#          return sum(args)
#
#  calc = Calculator()
#  print(calc.add(2, 3))        # 5
#  print(calc.add(2.5, 3.5))    # 6.0
#  print(calc.add(1, 2, 3))     # 6
#
#  Compiler arguments ke basis pe resolve karta hai kaunsa add() call
#  karna hai. Do ints -> add(int, int). Do doubles -> add(double,
#  double). Teeno ints -> add(int, int, int). Koi runtime decision
#  nahi chahiye.
#
#  ------------------------------------------------------------
#  2. Runtime Polymorphism (Method Overriding / Dynamic Dispatch)
#  ------------------------------------------------------------
#  Runtime polymorphism zyada powerful aur important form hai. Yeh
#  tab hota hai jab child class parent class mein defined method ko
#  OVERRIDE karti hai, aur kaunsa version call karna hai yeh decision
#  RUNTIME pe hota hai object ke actual type ke basis pe — reference
#  ke declared type ke basis pe nahi.
#
#  Example:
#  Maan lo tum notification system bana rahe ho — email, SMS, push
#  support karna hai.
#
#  class Notification:
#      def __init__(self, recipient: str, message: str):
#          self._recipient = recipient
#          self._message = message
#
#      def send(self):
#          print(f"Sending generic notification to {self._recipient}")
#
#  class EmailNotification(Notification):
#      def __init__(self, recipient, message, subject):
#          super().__init__(recipient, message)
#          self._subject = subject
#
#      def send(self):
#          print(f"Sending EMAIL to {self._recipient} | Subject: {self._subject}")
#
#  class SMSNotification(Notification):
#      def __init__(self, recipient, message, phone_number):
#          super().__init__(recipient, message)
#          self._phone_number = phone_number
#
#      def send(self):
#          print(f"Sending SMS to {self._phone_number} | Message: {self._message}")
#
#  class PushNotification(Notification):
#      def __init__(self, recipient, message, device_token):
#          super().__init__(recipient, message)
#          self._device_token = device_token
#
#      def send(self):
#          print(f"Sending PUSH to device {self._device_token[:8]}"
#                f"... | Alert: {self._message}")
#
#  if __name__ == "__main__":
#      notifications = [
#          EmailNotification("alice@example.com", "Your order shipped!", "Order Update"),
#          SMSNotification("Bob", "Code: 482910", "+1-555-0123"),
#          PushNotification("Charlie", "New message", "d8a3f4b2c1e5a9b7"),
#      ]
#
#      for n in notifications:
#          n.send()
#
#  Key baat: list ke har element ek Notification reference ke roop mein
#  store hai, par runtime sahi child class ka send() call karta hai.
#  Variable type kehta hai "Notification". Behavior kehta hai
#  "EmailNotification", "SMSNotification", ya "PushNotification".
#  Yahi hai runtime polymorphism (dynamic dispatch).
#
#  ------------------------------------------------------------
#  3. Polymorphism with Interfaces vs Abstract Classes
#  ------------------------------------------------------------
#  Dono interfaces aur abstract classes polymorphism enable karte
#  hain. Notification example mein tum Notification ko ya toh abstract
#  class ya interface bana sakte ho. Polymorphic behavior (base
#  reference pe send() call karna aur child ka version chalna) dono
#  mein same kaam karta hai. Toh kya use karein?
#
#  | Aspect          | Interface                                   | Abstract Class                                  |
#  |-----------------|---------------------------------------------|-------------------------------------------------|
#  | Relationship    | "can do" (capability)                       | "is a" (family)                                 |
#  | Shared behavior | None (contract only)                       | Yes (concrete methods + fields)                 |
#  | Multiple        | Class bahut saare implement kar sakti hai   | Class sirf ek extend kar sakti hai              |
#  | When to use     | Unrelated classes share a capability        | Related classes share logic                     |
#  | Example         | Sendable implemented by Email, Invoice, ... | Notification extended by Email, SMS, Push        |
#
#  INTERFACE use karo jab: implementing classes fundamentally
#  alag hain par ek capability share karti hain. Email, Invoice, aur
#  Report mein structurally kuch common nahi, par sab send() kar
#  sakte hain. Interface us contract ko define karta hai bina shared
#  hierarchy force kiye.
#
#  ABSTRACT CLASS use karo jab: implementing classes ek family hain
#  with shared logic. Sab notifications ko same format_header()
#  method chahiye, same recipient aur message fields, same constructor
#  pattern. Abstract class yeh sab deta hai — plus abstract send()
#  jo har child alag tarike se implement karta hai.
#
#  Practice mein bahut designs dono use karte hain. Abstract
#  Notification class shared fields/formatting deti hai, jabki
#  Sendable interface kisi bhi cheez ko mark karta hai jo bheji ja
#  sakti hai (notifications, reports, alerts).
#
# ============================================================
#
#  SUMMARY (Bhai, yeh yaad rakhna)
#  --------------------------------
#  - Polymorphism = "many forms." Same method naam, alag objects,
#    alag behavior. Common type pe code likho, actual behavior
#    concrete implementation decide kare.
#  - Analogy: Universal remote. Buttons same, par TV/AC/Projector
#    alag respond karte hain.
#  - Fayde: Loose coupling, flexibility (Open/Closed), scalability,
#    extensibility (plug-in naye implementations bina core code
#    chhede).
#  - Do forms: (1) Compile-time (Method Overloading) — same naam,
#    different params, compiler decide karta hai. (2) Runtime
#    (Method Overriding) — child parent ka method override karta
#    hai, runtime pe actual object type decide karta hai.
#  - Python mein overloading natively nahi hai (last method survive
#    karta hai) — *args/default args se workaround.
#  - Runtime polymorphism: list mein Notification references, par
#    har object ka apna send() chalta hai (dynamic dispatch).
#  - Interface vs Abstract Class: Interface = "can do" capability,
#    unrelated classes. Abstract Class = "is-a" family, shared logic.
#    Dono polymorphism enable karte hain.
#
#  Simple line: Polymorphism = "ek method naam, alag-alag objects apna
#  apna response karte hain." Isse code generic aur extensible banta
#  hai — naya type add karo, existing code bina chhede.
#
# ============================================================
```

```python
# ============================================================
#  POLYMORPHISM — Beginner-Friendly Hinglish Guide
# ============================================================
#
#  Polymorphism ek aisi cheez hai jisme ek hi method ka naam alag-
#  alag objects pe alag-alag tarike se kaam karta hai.
#
#  English mein "poly" matlab bahut, "morph" matlab roop. Matlab
#  ek hi cheez ke bahut saare roop. Programming mein — ek hi method
#  ka naam, par alag object pe call karne par alag result.
#
#  Bilkul simple bataun toh:
#  > Tum ek hi button dabate ho (jaise remote pe "power"), par TV
#  > band hota hai, AC chalta hai, projector off hota hai. Button
#  > wahi, kaam alag.
#
#  ------------------------------------------------------------
#  Real-Life Example — Universal Remote
#  ------------------------------------------------------------
#  Ek universal remote socho. Isme buttons hain:
#    - powerOn()
#    - volumeUp()
#    - mute()
#
#  Ab yeh remote kaunse device ko control kar raha hai, uske hisaab
#  se kaam alag hota hai:
#    - TV pe: TV on/off hota hai
#    - AC pe: AC on/off hota hai
#    - Projector pe: Projector on/off hota hai
#
#  Remote ke buttons same hain. Par har device unhe apne tareeke se
#  samajhta hai. User ko farak nahi padta — woh bas button dabata
#  hai. Yahi polymorphism hai — same interface, alag-alag behavior.
#
# ============================================================
#
#  POLYMORPHISM KYUN ZAROORI HAI?
#  ------------------------------
#  (Bina iske code bahut messy ho jaata)
#
#  1. Code reusable banta hai
#     Tum ek hi method ko bahut saare objects pe use kar sakte ho.
#     Har object ke liye alag method likhne ki zaroorat nahi.
#
#  2. Naya type add karna aasaan
#     Naya device aaya remote mein? Bas uska apna powerOn() likh do.
#     Baaki code kuch nahi badalna. Existing code todhne ki zaroorat
#     nahi. (Isko Open/Closed Principle kehte hain.)
#
#  3. Code clean aur simple rehta hai
#     Tumhe har object ka type check karke if-else likhne nahi padte.
#     Bas ek method call karo, sahi version apne aap chal jaata hai.
#
#  4. Loose coupling
#     Tumhara code kisi specific class pe depend nahi karta. Woh
#     ek "common type" pe depend karta hai. Isse classes ke beech
#     tight linkage nahi banti — ek change doosre ko todhta nahi.
#
# ============================================================
#
#  POLYMORPHISM KE 2 TYPES
#  -----------------------
#  OOP mein polymorphism 2 tareeke se aata hai. Dono ka matlab same
#  (ek naam, alag kaam), par decide karne ka tareeka alag hai.
#
#  ------------------------------------------------------------
#  Type 1: Compile-time Polymorphism (Method Overloading)
#  ------------------------------------------------------------
#  Isme ek hi class ke andar same naam ke bahut saare methods hote
#  hain, par unke parameters alag hote hain.
#
#  Program chalne se PEHLE hi compiler decide kar leta hai ki kaunsa
#  method call karna hai — arguments dekh kar.
#
#  Example socho — ek Calculator class jisme add() method hai:
#    - add(2, 3) -> 2 numbers jod do
#    - add(1, 2, 3) -> 3 numbers jod do
#
#  Compiler dekhta hai tumne kitne arguments diye. Uske hisaab se
#  sahi wala add() chal jaata hai. Tumhe kuch nahi batana padta.
#
#  Dhyan rakhna: Python mein asal method overloading nahi hota.
#  Agar tum same naam ke 2 method likhoge, toh sirf last wala bachega.
#  Iska solution hai *args ya default arguments use karna.
#
#  class Calculator:
#      def add(self, *args):
#          return sum(args)
#
#  calc = Calculator()
#  print(calc.add(2, 3))        # 5
#  print(calc.add(1, 2, 3))     # 6
#
#  Yahan *args ka matlab — jitne bhi numbers do, sabko jod do.
#  Ek hi method, alag-alag number of arguments, sab kaam.
#
#  ------------------------------------------------------------
#  Type 2: Runtime Polymorphism (Method Overriding)
#  ------------------------------------------------------------
#  Yeh zyada important aur powerful hai. Isme child class apne
#  parent class ke method ko "override" (dobara likh) deti hai — apna
#  khud ka version deti hai.
#
#  Aur kaunsa version chalega, yeh PROGRAM CHALTE WAQT (runtime pe)
#  decide hota hai — object ka actual type dekh kar. Reference type
#  dekh kar nahi.
#
#  Example — Notification System:
#  Ek Notification class hai (parent). Iska send() method hai.
#  3 child classes hain — EmailNotification, SMSNotification,
#  PushNotification. Sab apna send() khud likhte hain.
#
#  class Notification:
#      def __init__(self, recipient, message):
#          self._recipient = recipient
#          self._message = message
#
#      def send(self):
#          print(f"Sending notification to {self._recipient}")
#
#  class EmailNotification(Notification):
#      def __init__(self, recipient, message, subject):
#          super().__init__(recipient, message)
#          self._subject = subject
#
#      def send(self):   # Override kiya parent ka method
#          print(f"EMAIL to {self._recipient} | Subject: {self._subject}")
#
#  class SMSNotification(Notification):
#      def __init__(self, recipient, message, phone_number):
#          super().__init__(recipient, message)
#          self._phone_number = phone_number
#
#      def send(self):   # Override kiya parent ka method
#          print(f"SMS to {self._phone_number} | Message: {self._message}")
#
#  class PushNotification(Notification):
#      def __init__(self, recipient, message, device_token):
#          super().__init__(recipient, message)
#          self._device_token = device_token
#
#      def send(self):   # Override kiya parent ka method
#          print(f"PUSH to device {self._device_token[:8]}... | Alert: {self._message}")
#
#  Ab dekho magic. Hum ek list banate hain jisme teeno type ke
#  notifications hain. Sabko hum Notification samajh ke rakhte hain.
#
#  notifications = [
#      EmailNotification("alice@example.com", "Order shipped!", "Order Update"),
#      SMSNotification("Bob", "Code: 482910", "+1-555-0123"),
#      PushNotification("Charlie", "New message", "d8a3f4b2c1e5a9b7"),
#  ]
#
#  for n in notifications:
#      n.send()
#
#  Output:
#  -------
#  EMAIL to alice@example.com | Subject: Order Update
#  SMS to +1-555-0123 | Message: Code: 482910
#  PUSH to device d8a3f4b2... | Alert: New message
#
#  Dekho kya hua — list mein har item ek Notification reference hai.
#  Par jab n.send() call hota hai, toh har object ka APNA send()
#  chalta hai. EmailNotification ka send() email bhejta hai,
#  SMSNotification ka send() SMS bhejta hai.
#
#  Variable type kehta hai "Notification". Par asli behavior object ke
#  actual type se aata hai. Yahi hai RUNTIME POLYMORPHISM — program
#  chalte waqt decide hota hai kaunsa send() chalana hai.
#
#  Sabse badi baat: hume koi if-else likhna nahi pada. Nahi likhna
#  pada ki "agar email hai toh yeh bhejo, agar SMS hai toh woh bhejo".
#  Bas loop chalaya, send() call kiya. Har object ne apna kaam kiya.
#  Isliye code clean rehta hai.
#
# ============================================================
#
#  INTERFACE vs ABSTRACT CLASS — Dono polymorphism mein kaam aate hain
#  ------------------------------------------------------------------
#  Tum Notification ko ya toh interface bana sakte ho ya abstract
#  class. Polymorphic behavior dono mein same kaam karega — send()
#  call karo, child ka version chalega. Toh kya use karein?
#
#  INTERFACE use karo jab:
#  - Classes mein kuch common nahi, bas ek capability share karti hain.
#  - Jaise Email, Invoice, Report — inme structurally kuch common nahi,
#    par sab "send" ho sakte hain. Interface sirf yeh contract banata
#    hai ki "send() method hona chahiye", shared logic nahi deta.
#  - Ek class bahut saare interfaces implement kar sakti hai.
#
#  ABSTRACT CLASS use karo jab:
#  - Classes ek "family" hain with shared logic.
#  - Jaise saari notifications ko recipient, message fields chahiye,
#    aur format_header() method chahiye. Abstract class yeh sab ek
#    jagah pe de deti hai, plus send() ko abstract chhod deti hai jo
#    har child apne tareeke se likhega.
#  - Ek class sirf ek abstract class extend kar sakti hai.
#
#  Asli projects mein dono saath use hote hain. Abstract Notification
#  class shared fields aur formatting deti hai. Aur ek Sendable
#  interface mark karta hai ki yeh cheez bheji ja sakti hai (notifications,
#  reports, alerts — sab).
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  - Polymorphism = ek hi naam ke method, alag object pe alag kaam.
#    "Many forms" — ek method ke bahut roop.
#  - Example: Remote ke buttons same, par TV/AC/Projector alag
#    respond karte hain.
#  - Fayde: code reusable, naya type add karna aasaan, code clean
#    (kam if-else), loose coupling.
#  - 2 types:
#    (1) Compile-time (Method Overloading) — same naam, alag params.
#        Compiler pehle decide karta hai. Python mein directly nahi
#        hota, *args se karte hain.
#    (2) Runtime (Method Overriding) — child parent ka method dobara
#        likhta hai. Runtime pe object ka actual type decide karta hai.
#        Yeh zyada powerful hai.
#  - Runtime polymorphism mein: reference type parent ka, behavior
#    child ka. Bina if-else ke sahi method apne aap chalta hai.
#  - Interface = "can-do" capability, unrelated classes ke liye.
#    Abstract class = "is-a" family, shared logic ke liye. Dono
#    polymorphism enable karte hain.
#
#  Ek line mein: Polymorphism = "ek hi method ko call karo, har object
#  apne tareeke se jawab dega." Isse code simple, flexible, aur
#  reusable banta hai.
#
# ============================================================
```
