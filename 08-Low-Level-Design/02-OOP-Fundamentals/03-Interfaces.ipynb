```python
# ============================================================
#  INTERFACES — Gen-Z Hinglish Guide (super easy, chill vibe)
# ============================================================
#
#  Soch bhai — tum ek app bana rahe ho. Tumhe chahiye ki system
#  flexible ho, test ho sake, aur ek component ka change dusre ko
#  na tod de. Yahi pe INTERFACES enter karte hain.
#
#  Interface ka funda simple hai: yeh batata hai ki "kya" karna hai,
#  "kaise" karna hai nahi. Matlab — kaam ka contract fix, execution
#  tumhari marzi.
#
#  Is separation ki wajah se system ke alag-alag parts ek dusre ke
#  internal details jaane bina, "well-defined contracts" ke through
#  saath kaam kar sakte hain.
#
# ============================================================
#
#  1. INTERFACE kya hota hai?
#  ---------------------------
#  Basic level pe interface ek "CONTRACT" hai — ek list of methods
#  jo koi bhi class implement karti hai, usko provide KARNA hi
#  padta hai. Interface batata hai kaun-se behaviors chahiye, par
#  unka detail har implementation pe chhod deta hai.
#
#  Simple mein:
#  > Interface batata hai "WHAT" (kya karna hai)
#  > Class batati hai "HOW" (kaise karna hai)
#
#  ------------------------------------------------------------
#  Real-World Analogy — Remote Control
#  ------------------------------------------------------------
#  Ek remote lo. Uspe standard buttons hain:
#    - play()
#    - pause()
#    - volumeUp()
#    - powerOff()
#
#  Remote use karne wale ko farak nahi padta ki yeh TV control kar
#  raha hai, soundbar, ya projector — sab wohi commands samajhte
#  hain.
#
#  Remote = INTERFACE (contract: yeh 4 buttons honge)
#  TV / SoundBar / Projector = IMPLEMENTATIONS (har device ka apna
#  internal logic, par buttons wahi).
#
#  Har device pe play() dabane pe alag action ho sakta hai, par
#  contract same rehta hai.
#
# ============================================================
#
#  2. Interfaces ke Key Properties
#  --------------------------------
#  Interface sirf method declarations nahi — yeh flexible software
#  design ka foundation hain.
#
#  (a) Behavior define karta hai, implementation dictate nahi
#      Interface sirf yeh declare karta hai ki kaun-se operations
#      chahiye. Yeh nahi batata ki woh kaise honge. Implementer
#      ko poori azadi milti hai apna logic likhne ki, bas contract
#      hona chahiye.
#
#  (b) Polymorphism enable karta hai
#      Alag-alag classes same interface ko alag tarike se implement
#      kar sakti hain. Isse tumhara code multiple implementations ke
#      saath "interchangeably" (ek ko dusre ki jagah use karke) kaam
#      kar sakta hai. Faida? Naya variant add karna easy.
#
#  (c) Decoupling promote karta hai
#      Jo code interfaces pe depend karta hai, woh concrete (actual)
#      classes ke changes se surakshit rehta hai. Isse code:
#        - Extend karna easy (naya implementation bina purana code
#          change kiye add karo)
#        - Test karna easy (unit tests mein interface mock kar lo)
#        - Maintain karna easy (ek change se kahin aur side effects
#          kam)
#
#  ------------------------------------------------------------
#  Example: Payment Gateway
#  ------------------------------------------------------------
#  Maan lo CheckoutService ko payment process karna hai. providers
#  ho sakte hain: Stripe, Razorpay, PayPal.
#
#  Agar CheckoutService directly StripePayment pe depend karega,
#  toh provider change karna = CheckoutService ka code tod ke
#  likhna.
#
#  Isliye ek PaymentGateway INTERFACE banate hain. Stripe, Razorpay,
#  PayPal — sab usko implement karte hain. Ab CheckoutService sirf
#  PaymentGateway pe depend karta hai. Koi bhi provider chaho, bina
#  CheckoutService chhede plug kar do. Clean.
#
#  (Yeh dependency injection + programming to interface wala funda
#   hai — neeche code mein dekho.)
#
# ============================================================
#
#  3. Code Example: Payment Gateway Interface
#  -------------------------------------------
#  Tum payment module bana rahe ho — Stripe, Razorpay, PayPal.
#  Tumhe business logic kisi specific provider pe depend nahi karni.
#  Bas ek common tarika chahiye payment initiate karne ka.
#
#  ------------------------------------------------------------
#  (a) Interface Define Karna
#  ------------------------------------------------------------
#  PaymentGateway interface — ek hi operation: initiate_payment()
#  with an amount.
#
#  from abc import ABC, abstractmethod
#
#  class PaymentGateway(ABC):
#      @abstractmethod
#      def initiate_payment(self, amount):
#          pass
#
#  Ab yeh contract hai: har payment gateway ko initiate_payment()
#  method dena hi padega. Par kaise process karenge, woh interface
#  nahi batata — har provider ki marzi.
#
#  ------------------------------------------------------------
#  (b) Interface Implement Karna
#  ------------------------------------------------------------
#  Do classes banate hain — StripePayment aur RazorpayPayment. Dono
#  PaymentGateway ko implement karti hain. Dono ka internal logic
#  alag, par dono contract poori karte hain.
#
#  class StripePayment(PaymentGateway):
#      def initiate_payment(self, amount):
#          print(f"Processing payment via Stripe: ${amount}")
#
#  class RazorpayPayment(PaymentGateway):
#      def initiate_payment(self, amount):
#          print(f"Processing payment via Razorpay: ₹{amount}")
#
#  StripePayment Stripe API se baat karega, RazorpayPayment
#  Razorpay API se. Par interface guarantee deta hai ki dono ke
#  paas initiate_payment() hai. Iska matlab jo bhi code
#  PaymentGateway pe depend karta hai, dono ke saath chalega.
#
#  ------------------------------------------------------------
#  (c) Programming to the Interface (Asli magic yahan)
#  ------------------------------------------------------------
#  CheckoutService ko StripePayment ya RazorpayPayment pe depend
#  karne ke bajaye, hum PaymentGateway INTERFACE pe depend karate
#  hain. Isko pata hi nahi kiunse provider ka use ho raha hai.
#
#  class CheckoutService:
#      def __init__(self, payment_gateway):
#          self.payment_gateway = payment_gateway
#
#      def set_payment_gateway(self, payment_gateway):
#          self.payment_gateway = payment_gateway
#
#      def checkout(self, amount):
#          self.payment_gateway.initiate_payment(amount)
#
#  Constructor dekho — yeh PaymentGateway leta hai, StripePayment
#  nahi. Ek decision ne service ko har specific provider se
#  "decouple" kar diya.
#
#  checkout() method jo bhi gateway inject ki gayi hai, usi pe
#  initiate_payment() call karta hai. Woh Stripe ho sakti hai,
#  Razorpay ho sakti hai, test ke liye koi mock ho sakti hai, ya
#  koi provider jo abhi tak bana hi nahi hai.
#
#  Is pattern ko DEPENDENCY INJECTION kehte hain — class apni
#  dependencies khud nahi banati, bahar se milti hai. Aur yeh
#  sirf tab possible hai jab dependency ek INTERFACE type ki ho,
#  concrete class ki nahi.
#
#  ------------------------------------------------------------
#  (d) Runtime Flexibility (Provider switch on the fly)
#  ------------------------------------------------------------
#  Sab kuch wire karne ka time. Runtime pe tum decide karte ho
#  kaunsa implementation inject karna hai — aur beech mein bhi
#  swap kar sakte ho.
#
#  if __name__ == "__main__":
#      stripe_gateway = StripePayment()
#      checkout_service = CheckoutService(stripe_gateway)
#      checkout_service.checkout(120.50)
#      # Output: Processing payment via Stripe: $120.5
#
#      # Switch to Razorpay
#      razorpay_gateway = RazorpayPayment()
#      checkout_service.set_payment_gateway(razorpay_gateway)
#      checkout_service.checkout(150.50)
#      # Output: Processing payment via Razorpay: ₹150.5
#
#  Output:
#  -------
#  Processing payment via Stripe: $120.5
#  Processing payment via Razorpay: ₹150.5
#
#  Dekho — CheckoutService ke code mein kuch nahi badla dono
#  calls ke beech. Bas jo implementation "plug" kiya, wahi
#  change hua. Yahi hai "programming to interfaces" ki power:
#  calling code implementation details se poori tarah insulated.
#
# ============================================================
#
#  SUMMARY (Bhai, yeh yaad rakhna)
#  --------------------------------
#  - Interface = contract. Batata hai "kya" karna hai, "kaise"
#    nahi. Classes details handle karti hain.
#  - Analogy: Remote (interface) vs TV/SoundBar/Projector
#    (implementations). Buttons wahi, internals alag.
#  - Interface define karta hai behavior, implementation dictate
#    nahi karta.
#  - Polymorphism: ek interface, multiple implementations, code
#    interchangeably chalta hai.
#  - Decoupling: interface pe depend karo, toh concrete class
#    change hone se code nahi tootta.
#  - Dependency Injection: class dependencies khud nahi banati,
#    bahar se milti hain — interface type ki hone ke karan hi yeh
#    possible hai.
#  - Runtime pe provider swap karo bina service code chhede.
#
#  Simple line: Interface = "rule book" kiunse kaam honge. Kaam
#  kaise hoga, woh implementation pe chhod do. Isse code flexible,
#  testable, aur loosely coupled rehta hai. Easy.
#
# ============================================================
```
