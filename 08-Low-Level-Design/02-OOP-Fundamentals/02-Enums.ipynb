```python
# ENUMS:
# - Socho tum ek e-commerce platform bana rahe ho aur har order ka status track 
#   karna hai. 
# - Orders alag-alag states se guzarte hain: 
#   - PLACED, 
#   - CONFIRMED, 
#   - SHIPPED, 
#   - DELIVERED, 
#   - CANCELLED 
# - In states ko code mein kaise represent karo?
#
# - Tum strings use kar sakte ho: 
#   - "PLACED", 
#   - "SHIPPED", 
#   - "DELIVERED".
# - Par agar koi galti se "Shiped" (ek 'P' kam) likh de? 
# - Compiler catch nahi karega aur code chup-chaap runtime pe fail ho jaayega.
#
# - Ya phir integers use kar lo: 
#   - 1 = placed, 
#   - 2 = shipped, 
#   - 3 = delivered.
# - Ab tumhare code mein "magic numbers" bhar jaayenge. 
# - "if (status == 2)" ka kya matlab hai? 
# - Kisi ko samajh nahi aayega.
# - Yahi problem ENUMS solve karte hain.
```

```python
# What is ENUM?
# - Enum (enumeration ka short form) ek special data type hai jo ek fixed set 
#   of named constants define karta hai. 
# - Strings ya integers ke khilaaf, enums "type-safe" hote hain — matlab compiler
#   ensure karta hai ki tum sirf wahi values use kar sako jo tumhare defined
#   set mein actually exist karti hain.
#
# - Enum ensure karta hai ki ek variable sirf ek pre-defined set ke valid options
#   mein se hi value le — aur kuch nahi.
#
# - Rule of thumb: Agar ek value sirf pre-defined options mein se ek hi ho sakti hai,\
#   toh enum use karo.
```

```python
# Why Use Enums? (Enums kyun use karein?)
# - Enum tumhe kya deta hai, uski value samajhne ke liye alternative dekho. 
# - Bina enum ke tum order statuses strings se represent karoge:
#   status = "PENDING"

# Kahin aur codebase mein...
if status == "PNDING":   # Typo! Yeh condition kabhi true nahi hogi
    process_order()

# Explanation:
# - Yeh code bina kisi warning ke compile ho jaata hai. 
# - "PNDING" ek perfectly valid string hai compiler ke liye. 
# - Bug tabhi saamne aata hai jab customer complaint karta hai ki uska order process
#   hi nahi ho raha.
# - Enums is poori category ke bugs ko khatam kar dete hain. 
# - Jab tum OrderStatus ko enum ke roop mein define karte ho with values jaise
#   PENDING, CONFIRMED, SHIPPED, DELIVERED — toh compiler ko exactly pata hota hai
#   kaunsi values valid hain.
```

```python
# Key Advantages of ENUM: (plain constants/strings ke muqaable):
# a. "Magic values" se bachaav: 
#    - Code mein scattered strings ("PENDING") ya integers (3) nahi padte.
#
# b. Readability behtar: 
#    - OrderStatus.SHIPPED ka matlab 3 se kahin zyada clear hai.
#
# c. Compiler checks: 
#    - Compiler enum usage validate karta hai, typos aur invalid assignments 
#      jaldi pakad mein aate hain.
#
# d. IDE support: 
#    - Most IDEs enum values ke liye auto-completion aur refactoring tools dete hain.
#
# e. Bugs kam: 
#    - Tum galti se koi random string ya number assign nahi kar sakte jo enum mein
#      nahi hai.
```

```python
# Example Enums:
# - Enums un categories ya states ke liye perfect hain jo rarely change hote hain:
#   - Order States (PENDING, IN_PROGRESS, COMPLETED)
#   - User Roles (ADMIN, CUSTOMER, DRIVER)
#   - Vehicle Types (CAR, BIKE, TRUCK)
#   - Directions (NORTH, SOUTH, EAST, WEST)
# - Raw strings ki jagah enums use karke tum system ko samajhna aasaan aur galat use
#   karna mushkil bana dete ho.
```

```python
# ENUM Types:
# 1. Simple Enum
# 2. Enums with Properties and Methods
```

```python
# 1. Simple Enum
#    - Sabse basic enum ek single type ke andar named constants ki list define karta hai.
#    - E-commerce system mein order status model karte hain:

from enum import Enum

# Simple Enum
class OrderStatus(Enum):
    PLACED = "PLACED"
    CONFIRMED = "CONFIRMED"
    SHIPPED = "SHIPPED"
    DELIVERED = "DELIVERED"
    CANCELLED = "CANCELLED"

# Yeh enum ek "finite set of valid states" define karta hai jo ek order mein ho 
# sakte hain. Aur kuch allowed nahi.

status = OrderStatus.SHIPPED
if status == OrderStatus.SHIPPED:
    print("Your package is on the way!")

# Simple enums tab kaam aate hain jab tumhe sirf named constants ki list chahiye. 
# Par kya hagar har constant ke saath extra data bhi hona chahiye?
```

```python
# 2. Enums with Properties and Methods
#    - Enums sirf constants ko naam nahi dete. 
#    - Bahut saari languages mein, har enum value extra data bhi hold kar sakti hai
#      aur behavior (methods) bhi define kar sakti hai. 
#    - Isse domain concepts model karne ke liye bahut powerful bante hain.

# - Ek Coin enum dekhte hain jo U.S. coins aur unki denomination represent karta hai. 
# - Har coin ka naam (PENNY, NICKEL, DIME, QUARTER) aur value in cents (1, 5, 10, 25)
#   hai. 
# - Coin values ke liye alag lookup table maintain karne ke bajaye, tum value directly
#   enum mein embed kar sakte ho.

from enum import Enum

class Coin(Enum):
    PENNY = 1
    NICKEL = 5
    DIME = 10
    QUARTER = 25

    def __init__(self, value):
        self.coin_value = value

    def get_value(self):
        return self.coin_value

total = Coin.DIME.get_value() + Coin.QUARTER.get_value()  # 35
print(total)

# - Yeh alag arrays ya lookup maps maintain karne se kahin zyada elegant aur safe hai. 
# - Data us constant ke saath hi rehta hai jisse belong karta hai — isse koi risk nahi
#   ki value aur naam sync se bahar ho jaayein.
```

```python
# SUMMARY (Key Takeaways)
# a. Enum = special data type jo fixed set of named constants define karta hai. 
#    Type-safe hota hai — compiler sirf valid values allow karta hai.
#
# b. Strings/integers use karne ki jagah enum use karo jab ek variable sirf pre-defined
#    options mein se ek hi value le sakta ho.
#
# c. Fayde: magic values nahi, readability behtar, compiler checks, IDE support, 
#    bugs kam.
#
# d. Simple Enum: sirf named constants ki list (jaise OrderStatus).
# 
# e. Enum with Properties/Methods: 
#    - har value extra data aur methods hold kar sakti hai 
#     (jaise Coin with value aur get_value()).
#
# f. Data constant ke saath hi rehta hai, isliye naam aur value kabhi out of sync 
#    nahi hote.

# Simple yaad rakho:
# - Agar koi value sirf kuch fixed options mein se ek ho sakti hai
#   (order status, user role, direction, coin type), toh uske liye
#   enum banao — strings ya magic numbers se mat chalo.
# - Enum = named + type-safe + compiler-verified constants.
```
