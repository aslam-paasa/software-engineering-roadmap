```python
# ============================================================
#  INHERITANCE — Gen-Z Hinglish Guide (super easy, chill vibe)
# ============================================================
#
#  Inheritance ek class (subclass / child class) ko doosri class
#  (superclass / parent class) ke properties aur behaviors inherit
#  karne deta hai.
#
#  Simple mein:
#  > Inheritance code reuse enable karta hai — common logic ek
#  > baar base class mein likho, phir derived classes mein extend
#  > ya specialize karo.
#
#  Isse software cleaner, modular, aur maintainable banta hai.
#
#  ------------------------------------------------------------
#  Real-World Analogy — Web App ka User System
#  ------------------------------------------------------------
#  Ek web app mein User system soch:
#  - Base User class: common attributes (username, email) aur
#    methods (login(), logout()).
#  - Specialized roles: Admin, Customer, Vendor — sab User se
#    inherit karte hain, par apna role-specific behavior add karte
#    hain.
#      * Admin -> deleteUser()
#      * Customer -> browseProducts()
#      * Vendor -> addProduct()
#
#  Saare specialized users common data/behavior User class se
#  inherit karte hain, par apni functionality extend kar sakte hain.
#
# ============================================================
#
#  1. INHERITANCE kyun zaroori hai?
#  --------------------------------
#  Inheritance ke chaar main benefits:
#
#  (a) Code Reusability
#      Yeh DRY (Don't Repeat Yourself) principle follow karta hai.
#      Common logic ek baar parent class mein likha jaata hai aur
#      sab subclasses mein share hota hai — redundancy kam.
#
#  (b) Logical Hierarchy
#      Yeh clear aur intuitive hierarchy banata hai jo real-world
#      "is-a" relationships model karta hai — jaise ElectricCar
#      "is a" Car, ya Admin "is a" User.
#
#  (c) Ease of Maintenance
#      Agar shared logic mein bug mila ya change chahiye, sirf
#      superclass mein fix karo. Sab subclasses automatically fix
#      inherit kar lengi.
#
#  (d) Polymorphism
#      Inheritance polymorphism ki prerequisite hai — alag-alag
#      subclasses ke objects ko superclass ke object ki tarah treat
#      kar sakte ho.
#
# ============================================================
#
#  2. Inheritance kaise kaam karta hai?
#  -----------------------------------
#  Jab ek class doosri se inherit karti hai:
#  - Subclass superclass ke saare non-private fields aur methods
#    inherit karti hai.
#  - Subclass inherited methods ko OVERRIDE kar sakti hai (apna
#    alag implementation de sakti hai).
#  - Subclass superclass ko EXTEND bhi kar sakti hai — naye fields
#    aur methods add kar sakti hai.
#
#  Isse reuse aur customization dono milte hain.
#
#  ------------------------------------------------------------
#  Example: Vehicle Hierarchy
#  ------------------------------------------------------------
#  Vehicle (base class) mein make, model, year attributes hain,
#  aur start_engine(), stop_engine(), display_info() methods.
#  ElectricCar aur GasCar Vehicle se inherit karte hain aur apna
#  specific behavior add karte hain.
#
#  class Vehicle:
#      def __init__(self, make: str, model: str, year: int):
#          self._make = make
#          self._model = model
#          self._year = year
#
#      def start_engine(self):
#          print("Engine started")
#
#      def stop_engine(self):
#          print("Engine stopped")
#
#      def display_info(self):
#          print(f"{self._year} {self._make} {self._model}")
#
#  Yeh Vehicle class basic attributes aur common behaviors define
#  karti hai jo har car mein honge.
#
#  Ab specialized vehicle types banate hain:
#
#  class ElectricCar(Vehicle):
#      def __init__(self, make, model, year, battery_capacity):
#          super().__init__(make, model, year)   # Parent constructor call
#          self._battery_capacity = battery_capacity
#
#      def charge_battery(self):
#          print(f"Charging {self._battery_capacity}kWh battery")
#
#  class GasCar(Vehicle):
#      def __init__(self, make, model, year, fuel_tank_size):
#          super().__init__(make, model, year)
#          self._fuel_tank_size = fuel_tank_size
#
#      def fill_tank(self):
#          print(f"Filling {self._fuel_tank_size}L fuel tank")
#
#  Dhyan do: ElectricCar aur GasCar dono ne Vehicle ke __init__ ko
#  super() se call kiya (common data set karne ke liye), phir apna
#  specific field (battery/tank) add kiya. Yahi reuse + extend hai.
#
# ============================================================
#
#  3. Types of Inheritance (Inheritance ke patterns)
#  -------------------------------------------------
#  Saare inheritance hierarchies same nahi hote. Kuch common
#  patterns hain, har ek ka structure aur trade-off alag.
#
#  (a) Single Inheritance
#      Simplest form — ek child class ek parent class ko extend
#      karti hai. ElectricCar extends Vehicle wala relationship.
#      Sabse common pattern, sabhi major languages support karte
#      hain.
#
#  (b) Multi-level Inheritance
#      Jab child class khud ek parent ban jaati hai. Example:
#      Vehicle -> Car -> ElectricCar. Har level pe zyada
#      specialization aati hai. Moderate depth mein theek hai, par
#      deep chains (5+ levels) fragile aur samajhna mushkil ho
#      jaate hain.
#
#  (c) Hierarchical Inheritance
#      Jab multiple child classes same parent ko extend karti hain.
#      Humara vehicle example — ElectricCar aur GasCar dono Vehicle
#      ko extend karte hain. Bahut common aur natural.
#
#  (d) Multiple Inheritance
#      Jab ek child class ek se zyada parents ko extend karti hai.
#      Yahan complication aati hai. Sirf C++ aur Python directly
#      support karte hain. Java, C#, TypeScript nahi karte.
#      Reason? THE DIAMOND PROBLEM.
#
#      Diamond Problem: Soch ElectricCar dono Vehicle aur Machine
#      ko extend karta hai. Dono (Vehicle aur Machine) mein start()
#      method hai. Jab tum electricCar.start() call karoge, konsa
#      version chalega? Vehicle wala? Machine wala? Dono?
#
#      - C++: "virtual inheritance" se handle karta hai — complex
#        aur error-prone.
#      - Python: "Method Resolution Order (MRO)" se handle karta
#        hai — ek well-defined algorithm (C3 linearization) jo
#        decide karta hai kaunse parent ka method priority lega.
#      - Java/C#: Problem se bachne ke liye sirf single class
#        inheritance allow karte hain. Multiple interfaces implement
#        kar sakte ho, par sirf ek class extend kar sakte ho.
#
# ============================================================
#
#  4. Kab Inheritance use karein? (Practical Checklist)
#  -----------------------------------------------------
#  Inheritance powerful hai, par intentionally use karo — sirf jab
#  yeh real-world relationship truly model karta ho. Galat decision
#  early design mein code ko change/test/reason karne mein mushkil
#  bana deta hai.
#
#  USE inheritance jab:
#  - Clear "is-a" relationship ho (Dog is an Animal, Car is a
#    Vehicle). Agar "X is a Y" naturally nahi bol sakte, toh
#    inheritance galat tool hai — composition dekho.
#  - Parent class common behavior/data define karta hai jo children
#    share karenge. (Sab vehicles mein start_engine() hai, toh
#    parent mein rakho — har type mein duplicate nahi karna.)
#  - Child class parent se expected behavior ko violate na kare.
#    Agar koi Vehicle reference ElectricCar pe point karta hai,
#    toh har Vehicle method waise hi kaam kare jaisa expected.
#  - Code reuse chahiye through shared logic/structure, aur
#    hierarchy shallow ho (2-3 levels max).
#
#  AVOID inheritance jab:
#  - Relationship "has-a" ya "uses-a" ho ("is-a" nahi). Car has
#    an Engine (Car Engine nahi hai). Printer uses a Logger
#    (Printer Logger nahi hai).
#  - Multiple sources se behaviors dynamically combine karne
#    hoon. Inheritance tumhe compile-time pe ek parent pe lock
#    kar deta hai, composition components mix-match karne deta hai.
#  - Runtime pe behaviors swap karne ki flexibility chahiye.
#    Composition se implementations inject kar sakte ho
#    (FileLogger <-> ConsoleLogger swap). Inheritance mein parent
#    fix hai.
#  - Child aur parent internals ke beech tight coupling avoid
#    karna hai. Parent class ka change poore hierarchy mein ripple
#    karta hai — large codebases mein risky.
#
#  Rule of thumb: Doubt ho toh composition se shuru karo. Genuine
#  "is-a" hierarchy emerge hone par baad mein inheritance me
#  refactor kar sakte ho. Ulta direction (deep inheritance tree
#  ko composition mein untangle karna) bahut zyada mushkil hai.
#
# ============================================================
#
#  5. Practical Example: Notification System
#  -----------------------------------------
#  Inheritance ko ek alag domain pe apply karte hain — taaki dikhe
#  yeh sirf vehicles tak limited nahi. Ek notification system bana
#  rahe ho jo alag channels se messages bhej sakta hai: email, SMS,
#  push notifications.
#
#  Sab notification types common properties share karte hain:
#  recipient, message, timestamp. Sabko format_header() method
#  chahiye jo consistent header format deta hai. Par send() method
#  har channel ke liye alag kaam karta hai:
#    - Email: subject line chahiye
#    - SMS: 160 character limit hai
#    - Push: device token aur priority level chahiye
#
#  from datetime import datetime
#
#  class Notification:
#      def __init__(self, recipient: str, message: str):
#          self._recipient = recipient
#          self._message = message
#          self._timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
#
#      def format_header(self) -> str:
#          return f"[{self._timestamp}] To: {self._recipient}"
#
#      def send(self):
#          print(self.format_header())
#          print(f"Message: {self._message}")
#
#  class EmailNotification(Notification):
#      def __init__(self, recipient, message, subject):
#          super().__init__(recipient, message)
#          self._subject = subject
#
#      def send(self):
#          print(self.format_header())
#          print(f"Subject: {self._subject}")
#          print(f"Body: {self._message}")
#          print("Status: Email delivered")
#
#  class SMSNotification(Notification):
#      MAX_LENGTH = 160
#
#      def __init__(self, recipient, message, phone_number):
#          super().__init__(recipient, message)
#          self._phone_number = phone_number
#
#      def send(self):
#          print(self.format_header())
#          print(f"Phone: {self._phone_number}")
#          sms_body = (self._message[:self.MAX_LENGTH - 3] + "..."
#                      if len(self._message) > self.MAX_LENGTH
#                      else self._message)
#          print(f"SMS: {sms_body}")
#          print(f"Status: SMS sent ({len(sms_body)}/{self.MAX_LENGTH} chars)")
#
#  class PushNotification(Notification):
#      def __init__(self, recipient, message, device_token, priority):
#          super().__init__(recipient, message)
#          self._device_token = device_token
#          self._priority = priority
#
#      def send(self):
#          print(self.format_header())
#          print(f"Device: {self._device_token[:8]}...")
#          print(f"Priority: {self._priority}")
#          print(f"Alert: {self._message}")
#          print("Status: Push notification delivered")
#
#  if __name__ == "__main__":
#      email = EmailNotification(
#          "alice@example.com", "Your order has been shipped!", "Order Update")
#      email.send()
#
#      print()
#
#      sms = SMSNotification(
#          "Bob", "Your verification code is 482910.", "+1-555-0123")
#      sms.send()
#
#      print()
#
#      push = PushNotification(
#          "Charlie", "New message from Alice", "d8a3f4b2c1e5a9b7", "high")
#      push.send()
#
#  ------------------------------------------------------------
#  Why This Design Works (Design kyun sahi hai)
#  ------------------------------------------------------------
#  - Shared logic ek baar likha: recipient, message, timestamp
#    fields Notification mein defined hain. format_header() teeno
#    notification types mein inherit hota hai — consistent header
#    format. Timestamp format change karna ho toh ek method badlo.
#  - Har child channel-specific complexity encapsulate karta hai:
#    SMSNotification 160-char limit handle karta hai,
#    PushNotification device tokens aur priority manage karta hai,
#    EmailNotification subject line add karta hai. Yeh details
#    parent ya ek dusre mein leak nahi hoti.
#  - Naya channel add karna simple: Slack chahiye? SlackNotification
#    banao, webhookUrl field add karo, send() override karo. Kisi
#    existing code ko chhued nahi — open for extension, closed for
#    modification.
#
# ============================================================
#
#  SUMMARY (Bhai, yeh yaad rakhna)
#  --------------------------------
#  - Inheritance = code reuse. Common logic parent mein likho,
#    children inherit karte hain. "Is-a" relationship model karta
#    hai (ElectricCar is a Vehicle).
#  - Benefits: DRY/code reuse, logical hierarchy, easy maintenance
#    (ek jagah fix, sab jagah apply), polymorphism ki prerequisite.
#  - Subclass non-private fields/methods inherit karta hai, methods
#    override kar sakta hai, naye fields/methods add kar sakta hai.
#  - Types: Single (1 child <- 1 parent), Multi-level (chain),
#    Hierarchical (multiple children <- 1 parent), Multiple (1 child
#    <- multiple parents — diamond problem).
#  - Diamond problem: Multiple inheritance mein conflict jab dono
#    parents mein same method. C++ = virtual inheritance, Python =
#    MRO (C3 linearization), Java/C# = allow hi nahi karte.
#  - Use jab: clear "is-a" relationship, shared behavior, shallow
#    hierarchy. Avoid jab: "has-a"/"uses-a" relationship, runtime
#    flexibility chahiye, tight coupling avoid karna ho.
#  - Doubt ho toh composition se shuru karo — baad mein inheritance
#    me refactor easy, ulta bahut mushkil.
#
#  Simple line: Inheritance = "common cheezein parent mein rakho,
#  children apna specific behavior add karein." Sirf tab use karo
#  jab sach mein "is-a" relationship ho, warna composition better.
#
# ============================================================
```
