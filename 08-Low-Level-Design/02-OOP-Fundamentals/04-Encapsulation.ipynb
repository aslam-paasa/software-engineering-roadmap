```python
# ============================================================
#  ENCAPSULATION — Gen-Z Hinglish Guide (super easy, chill vibe)
# ============================================================
#
#  Encapsulation OOP ke chaar foundational principles mein se ek hai.
#  Iska matlab hai: related data (variables) aur unka behavior
#  (methods) jo us data pe kaam karte hain, unko ek single unit
#  (usually ek class) mein group karna, aur class ke internal
#  details tak direct access rok dena.
#
#  Simple mein:
#  > Encapsulation = Data hiding + Controlled access
#
#  ------------------------------------------------------------
#  Real-World Analogy — Bank Account (Vault + ATM)
#  ------------------------------------------------------------
#  Soch ek bank account ek "vault" hai bank ke andar. Tum seedha
#  vault mein jaake numbers nahi badalte, na?
#
#  Tum usse interact karte ho ek "well-defined interface" ke through
#  — yani ATM. ATM limited par specific operations deta hai:
#    - deposit()
#    - withdraw()
#    - checkBalance()
#
#  User ---interacts via public methods---> BankAccount
#    (balance private hai, sirf methods se access)
#
#  Tum direct bank ke internal data ko access ya modify nahi kar
#  sakte. Bank data kaise store karti hai, interest kaise lagati
#  hai, transactions kaise validate karta hai — yeh sab badal
#  sakta hai, par tum ATM kaise use karte ho, woh nahi badalta.
#
#  Yahi hai Encapsulation in action: internal complexity chhupao,
#  sirf zaroori cheez expose karo.
#
#  Well-encapsulated design mein external code ko yeh nahi pata
#  hota ki kuch "kaise" hota hai. Usse sirf yeh pata hota hai ki
#  "kya" ho sakta hai.
#
# ============================================================
#
#  1. ENCAPSULATION kyun zaroori hai?
#  ----------------------------------
#  Encapsulation sirf data protection nahi — yeh robust, secure,
#  aur maintain systems design karne ke baare mein hai.
#
#  (a) Data Hiding
#      Sensitive data (jaise bank balance ya password) directly
#      expose nahi hona chahiye. Encapsulation is data ko private
#      rakhta hai, aur sirf controlled methods se accessible banata
#      hai.
#
#  (b) Controlled Access and Validation
#      Yeh ensure karta hai ki data sirf "controlled, predictable"
#      tarike se modify ho. Example: deposit/withdraw mein input
#      validate karke invalid amounts ko rook sakte ho (negative
#      deposit? reject!).
#
#  (c) Improved Maintainability
#      Internal details hidden hone ki wajah se, tum implementation
#      change kar sakte ho (data kaise store hota hai, validation
#      kaise hoti hai) bina us code ko todhe jo ispe depend karta
#      hai.
#
#  (d) Security and Stability
#      External tampering rokne se, inconsistent ya invalid system
#      states ka risk kam ho jaata hai.
#
# ============================================================
#
#  2. Encapsulation kaise achieve hota hai?
#  ----------------------------------------
#  Encapsulation mainly do language features se implement hota
#  hai: (1) Access Modifiers jo visibility control karte hain, aur
#  (2) Getters/Setters jo private data tak controlled access dete
#  hain.
#
#  ------------------------------------------------------------
#  (a) Access Modifiers
#  ------------------------------------------------------------
#  Yeh keywords hain jo control karte hain ki code ke kaunse
#  parts class ke fields/methods ko dekh aur interact kar sakte
#  hain. Teen common types:
#
#  - private: Sirf same class ke andar accessible. Data chhupane
#    ka primary tool.
#  - protected: Same class aur uske subclasses mein accessible.
#    Jab child class ko parent ka data chahiye tab useful.
#  - public: Kahin se bhi accessible. Yeh controlled interface ke
#    liye use hota hai.
#
#  General rule: Sab kuch default mein "private" rakho, phir
#  selectively jo zaroori ho woh "public" expose karo.
#
#  Minimal example:
#
#  class Product:
#      def __init__(self, name: str, price: float):
#          self.__name = name      # Name-mangled (private by convention)
#          self.__price = price    # Name-mangled (private by convention)
#
#      def get_name(self) -> str:    # Koi bhi name padh sakta hai
#          return self.__name
#
#      def get_price(self) -> float: # Koi bhi price padh sakta hai
#          return self.__price
#
#  (Python mein __ (double underscore) se name mangling hoti hai,
#   jisse variable directly access karna mushkil ho jaata hai.)
#
#  ------------------------------------------------------------
#  (b) Getters and Setters
#  ------------------------------------------------------------
#  Yeh public methods hain jo private attributes tak controlled,
#  indirect access dete hain.
#
#  - Getter (e.g., get_balance()): Attribute tak read-only access.
#  - Setter (e.g., set_price()): Attribute modify karne deta hai,
#    usually with validation logic.
#
#  Example jahan setter invalid data ko object mein aane se rokta
#  hai (Python @property decorator ka use):
#
#  class Product:
#      def __init__(self, name: str, price: float):
#          self.__name = name
#          self.price = price    # Property setter use hota hai validation ke liye
#
#      @property
#      def name(self) -> str:
#          return self.__name
#
#      @property
#      def price(self) -> float:
#          return self.__price
#
#      @price.setter
#      def price(self, value: float) -> None:
#          if value < 0:
#              raise ValueError("Price cannot be negative")
#          self.__price = value
#
#  Dekho — price negative set karna chaho toh error aata hai.
#  Validation andar built-in hai. Data galat a hi nahi sakta.
#
# ============================================================
#
#  3. Encapsulation in Practice: BankAccount
#  -----------------------------------------
#  Ab ek complete encapsulated class dekhte hain — proper
#  validation, controlled access, aur business rules ke saath.
#  BankAccount class balance ko private rakhti hai, aur sirf
#  deposit() aur withdraw() se modification allow karti hai (har
#  ek apne rules enforce karte hain).
#
#  class BankAccount:
#      def __init__(self, account_holder: str):
#          self.__account_holder = account_holder
#          self.__balance = 0.0
#
#      def deposit(self, amount: float) -> None:
#          if amount <= 0:
#              raise ValueError("Deposit amount must be positive")
#          self.__balance += amount
#
#      def withdraw(self, amount: float) -> None:
#          if amount <= 0:
#              raise ValueError("Withdrawal amount must be positive")
#          if amount > self.__balance:
#              raise ValueError("Insufficient funds")
#          self.__balance -= amount
#
#      @property
#      def balance(self) -> float:
#          return self.__balance
#
#      @property
#      def account_holder(self) -> str:
#          return self.__account_holder
#
#  Kya ho raha hai dhyan se dekho:
#  - balance private hai (__balance), toh koi external class isse
#    directly access ya modify nahi kar sakta.
#  - deposit() aur withdraw() public entry points hain jo state
#    update karne se pehle user input validate karte hain.
#      * deposit: amount positive hona chahiye
#      * withdraw: amount positive hona chahiye AND balance se
#        zyada nahi hona chahiye (insufficient funds check)
#  - balance property read-only access deti hai bina underlying
#    variable ko expose kiye ya external code ko change karne dete.
#
#  Isse account hamesha valid state mein rehta hai aur business
#  rules controlled interfaces ke through enforce hote hain.
#
# ============================================================
#
#  SUMMARY (Bhai, yeh yaad rakhna)
#  --------------------------------
#  - Encapsulation = Data hiding + Controlled access. Data aur
#    behavior ko class mein group karo, internal details chhupao.
#  - Analogy: Bank vault (internal data) + ATM (public interface).
#    Tum vault mein nahi jaate, ATM se interact karte ho.
#  - Kyun zaroori? Data hiding, controlled access + validation,
#    maintainability, security/stability.
#  - Access Modifiers: private (sirf class), protected (class +
#    subclasses), public (kahin se bhi). Default private rakho.
#  - Getters/Setters: private data tak controlled, indirect access.
#    Getter = read-only, Setter = modify with validation.
#  - Python mein __ (double underscore) = name mangling (private
#    by convention), @property = getter, @x.setter = setter.
#  - BankAccount example: balance private, deposit/withdraw
#    validate karte hain, balance property read-only access deti
#    hai. Account hamesha valid state mein.
#
#  Simple line: Encapsulation = "internal chhupao, sirf zaroori
#  cheez expose karo, aur data ko controlled tareeke se hi badalne
#  do." Isse code robust, secure, aur maintainable banta hai. Easy.
#
# ============================================================
```
