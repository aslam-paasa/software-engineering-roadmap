```python
# CLASSES AND OBJECTS:
# - Har object-oriented system ki shuruaat ek basic sawaal se hoti hai:
#   "Main real-world entities ko code mein kaise represent karoon?"
# - Iska jawab hain — Classes aur Objects. 
# - Yeh dono milkar OOP ka foundation banate hain. 
# - Java, Python, C++, C#, Go, TypeScript — sabhi OOP languages is concept ko use 
#   karti hain code ko real-world entities ke around organize aur structure karne ke
#   liye.
```

```python
# What is Class?
# - Class ek "blueprint", "template", ya "recipe" hai objects banane ke liye. 
# - Yeh define karti hai ki:
#   a. ek object mein kya hoga (uska data) aur 
#   b. woh kya kar sakega (uska behavior).
# - Class khud ek object nahi hai — yeh ek template hai jisse bahut saare objects 
#   bante hain, jin ka structure same hota hai par state (data) alag-alag hota hai.
```

```python
# Real-World Analogy (Cake ki Recipe):
# a. Ingredients  (flour, sugar, eggs)  -> fields / attributes (variables)
# b. Instructions (mix, bake, decorate) -> methods / functions (operations)
#
# - Recipe khud cake nahi banati — bas batati hai cake kaise banega.
# - Jab tum recipe follow kar ke cake bana lete ho, toh tumne ek OBJECT create
#   kar liya.
#
# Code mein: 
# - recipe = class definition, 
# - har cake = ek object with apna flavor, frosting, size.
#
# Class ki Key Characteristics:
# - Related data (attributes) aur actions (methods) ko ek saath group karti hai.
# - Attributes define karti hai -> object ka state/data.
# - Methods (class ke andar functions) define karti hai -> object ka behavior / actions
#   jo woh kar sakta hai.
```

```python
# Example: Car Class Blueprint
# - Ek simple Car class banate hain, with:
#   a. attributes (brand, model, speed) aur 
#   b. methods (accelerate, display_status)

class Car:
    # Constructor
    def __init__(self, brand, model):
        # Attributes (convention: underscore = private)
        self._brand = brand
        self._model = model
        self._speed = 0

    # Method to accelerate
    def accelerate(self, increment):
        self._speed += increment

    # Method to display info
    def display_status(self):
        print(f"{self._brand} is running at {self._speed} km/h.")

# Note:
# - Yeh Car class define karti hai ki har car object kaisa dikhega (brand, model, speed)
#   aur kya kar sakta hai (accelerate, display).
# - Par class akele sirf source code mein ek definition hai. 
# - Kuch useful karne ke liye tumhe usse OBJECTS banana padte hain.
```

```python
# What are Objects?
# - Object ek class ka "instance" hai. 
# - Yeh woh actual cheez hai jiske saath tum interact kar sakte ho, data store 
#   kar sakte ho, aur methods ko invoke (call) kar sakte ho.
#
# - Object create karte waqt tum basically keh rahe ho:
#  "Is blueprint (class) ko lo aur usse ek actual cheez (object) banao."
#
# - Har object ko class mein defined data ka apna khud ka copy milta hai. 
# - Sab objects ka structure aur behavior same hota hai, par har object apne aap mein
#   independent hota hai — dusre objects se alag.
```

```python
# Example: Car class se Objects Banana
corolla = Car("Toyota", "Corolla")
mustang = Car("Ford", "Mustang")

corolla.accelerate(20)
mustang.accelerate(40)

# Displaying status of each car
corolla.display_status()
print("-----------------")
mustang.display_status()


# Output:
# - Toyota Corolla is running at 20 km/h.
# - -----------------
# - Ford Mustang is running at 40 km/h.

# Note:
# - Dhyan do: corolla aur mustang dono Car objects hain, par dono ka state bilkul 
#   independent hai. 
# - Jab humne corolla.accelerate(20) call kiya, sirf Corolla ki speed badhi. 
# - Mustang ki speed tab tak 0 hi rahi jab tak humne explicitly accelerate(40) 
#   call karke usse 40 nahi kiya.
```

```python
# SUMMARY (Key Takeaways)
# - Class = blueprint/template/recipe. 
# - Yeh batati hai object kaisa hoga aur kya karega. 
# - Class khud object nahi hai.
#
# - Object = class ka instance. 
# - Actual cheez jise tum use kar sakte ho. 
# - Har object ka apna independent state/data hota hai.
 
# - Attributes (variables) = object ka data/state 
#   (e.g., brand, model, speed).
# - Methods (class ke functions) = object ka behavior/actions
#   (e.g., accelerate, display_status).
# - Constructor = special method (jaise __init__) jo object bante waqt uske 
#   initial values set karta hai.
# - Ek class se bahut saare independent objects ban sakte hain — sabka structure same, 
#   par state alag-alag.

# Simple yaad rakho:
# - Recipe (class) batati hai cake kaise banana hai.
# - Cake (object) woh actual cheez hai jo recipe follow karke banti hai. 
#   Har cake ka flavor/apne alag features ho sakte hain.
```
