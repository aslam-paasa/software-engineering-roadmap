```python
# Structural Design Patterns — GameForge Studio Part 2

# - Pichli notebook mein humne Creational Patterns se GameForge Studio  ke OBJECTS 
#   banaye (Player, Enemy, World...). Ab is notebook mein hum 7 Structural Patterns 
#   seekhenge — inka focus hai "OBJECTS EK DOOSRE SE KAISE JUDENGE / ORGANIZE HONGE", 
#   taaki poora SYSTEM flexible aur maintainable rahe.

# Hum GameForge Studio ki Part 2 journey karenge: "Game World Assemble karna". 
# Har pattern PICHLE pattern se NATURAL EXTENSION hoga.

# Journey Map:
# 1. 🔌 Adapter — Ek THIRD-PARTY Physics Engine ka MISMATCHED interface fix karte hain
# 2. 🌉 Bridge — Enemy TYPE aur Renderer TYPE ko ALAG karte hain (class explosion se bachte hain)
# 3. 🌳 Composite — Poori Game Scene ko ek TREE STRUCTURE mein organize karte hain
# 4. ☕ Decorator — Player par PowerUps ko DYNAMICALLY wrap karte hain
# 5. 🎭 Facade — Saare complex subsystems ko EK SIMPLE interface ke peeche chhupate hain
# 6. 🪶 Flyweight — Hazaaron trees render karte hain bina MEMORY explode kiye
# 7. 🕵️ Proxy — Save-file ACCESS ko CONTROL karte hain
```

```python
# STRUCTURAL DESIGN PATTERNS — YE CATEGORY KYA HAI?
# - Design Patterns 3 categories mein baante jaate hain:
#   1. CREATIONAL -> "Object KAISE BANEGA?"
#   2. STRUCTURAL -> "Objects EK DOOSRE SE KAISE JUDENGE?" (ISI PAR ye notebook hai)
#   3. BEHAVIORAL -> "Objects EK DOOSRE SE KAISE BAAT KARENGE?"

# STRUCTURAL PATTERNS ka MAIN KAAM hai: 
# - CLASSES aur OBJECTS ko EK DOOSRE ke saath COMBINE/ORGANIZE karna, taaki BADI, 
#   FLEXIBLE STRUCTURES ban sakein -- BINA underlying code ko COMPLEX banaye.

# YE ZAROORAT KYUN PADI?
# Jab system BADA hota hai, to ye SAWAAL aate hain:
# - Do INCOMPATIBLE interfaces ko kaise MILAO? (Adapter)
# - Ek CLASS ki MULTIPLE VARIATIONS ko kaise MANAGE karo bina class explosion ke? 
#   (Bridge)
# - INDIVIDUAL objects aur GROUPS OF objects ko SAME TAREEKE se kaise treat karo? 
#   (Composite)
# - Object mein NAYA BEHAVIOR kaise ADD karo bina uski CLASS change kiye? (Decorator)
# - Ek COMPLEX SYSTEM ko SIMPLE interface ke peeche kaise CHHUPAO? (Facade)
# - Hazaaron SIMILAR objects ko MEMORY-EFFICIENT kaise banao? (Flyweight)
# - Kisi object ke ACCESS ko kaise CONTROL karo? (Proxy)

#  PATTERN      KIS SAWAAL KA JAWAB DETA HAI?
#  ---------------------------------------------------
#  1. Adapter   -> "2 INCOMPATIBLE interfaces ko kaise jodu?"
#  2. Bridge    -> "Abstraction aur Implementation ko ALAG kaise rakhu?"
#  3. Composite -> "Tree/hierarchy ko UNIFORM tareeke se kaise treat karu?"
#  4. Decorator -> "Object mein NAYA behavior RUNTIME par kaise add karu?"
#  5. Facade    -> "Complex subsystem ko SIMPLE kaise dikhaun?"
#  6. Flyweight -> "Bahut saare SIMILAR objects MEMORY-EFFICIENT kaise banau?"
#  7. Proxy     -> "Object ke ACCESS ko kaise CONTROL karu?"

# IMPORTANT INSIGHT:
# - Jaise Creational Patterns EK DOOSRE ko SUPPORT karte hain, waise hi Structural 
#   Patterns bhi SAATH MEIN kaam karte hain. Chalo ab GameForge Studio ki Part 2 
#   journey shuru karte hain -- "Game World Assemble karna"!
```

```python
# ADAPTER DESIGN PATTERN:
# - Adapter ek STRUCTURAL design pattern hai jo DO INCOMPATIBLE interfaces ke BEECH 
#   "BRIDGE" (pul) ka kaam karta hai, taaki wo EK SAATH kaam kar sakein -- BINA unka 
#   ORIGINAL code CHHUE.

# SOCHO EK EXAMPLE:
# - Tum India se ghoome ghoome USA gaye ho, aur tumhare paas Indian plug (round pins) 
#   wala charger hai.
# - USA ke socket mein Indian plug SEEDHA FIT NAHI hota (flat pins hote hain). 
# - Iske liye tum ek "TRAVEL ADAPTER" use karte ho -- wo Indian plug ko EK SIDE se
#   accept karta hai, aur USA socket mein DOOSRI SIDE se FIT ho jaata hai. 
# - NA tumhara charger badla, NA USA ka socket -- bas BEECH mein ek ADAPTER aa gaya!

# SOFTWARE MEIN BHI YAHI HOTA HAI:
# - Socho tumhare paas ek PURANI "LegacyPrinter" class hai jiska method hai 
#   `print_document()`. Lekin tumhara NAYA system `Printer` interface EXPECT karta hai
#   jiska method hai `print()`. 
# - Naam ALAG hai, isliye NAYA system SEEDHA LegacyPrinter USE NAHI kar sakta.

# Adapter isko solve karta hai:
# - Ek "PrinterAdapter" class banao jo NAYE system ka `Printer` interface IMPLEMENT 
#   karti hai (matlab `print()` method deti hai)
# - ANDAR SE, jab `print()` call ho, wo PURANE `LegacyPrinter.print_document()` ko 
#   CALL kar de

# MAIN COMPONENTS:
# 1. Target Interface
#    - Jo interface CLIENT EXPECT karta hai
#    - Ex: Printer
# 2. Adaptee
#    - EXISTING class jiska interface INCOMPATIBLE hai 
#    - Ex: LegacyPrinter
# 3. Adapter
#    - It acts as a BRIDGE class
#    - Target interface IMPLEMENT karti hai, ANDAR Adaptee ko WRAP karke uska kaam 
#      FORWARD karti hai
# 4. Client
#    - Sirf Target interface se kaam  karta hai, use Adapter/Adaptee ki DETAILS
#      pata NAHI hoti

# KAISE KAAM KARTA HAI (4 STEPS):
# 1. Client, Adapter ke Target-interface method ko call karta hai
# 2. Adapter, request ko Adaptee ke format mein TRANSLATE karta hai
# 3. Adaptee ACTUAL kaam karta hai
# 4. Client ko result milta hai -- use Adapter/Adaptee ka pata TAK NAHI chalta

# ADAPTER KE 4 TAREEKE:
# 1. Class Adapter (Inheritance-based)
#    - Multiple inheritance support karne wali languages (C++) mein use hota hai
# 2. Object Adapter (Composition-based)
#    - Adaptee ka OBJECT ANDAR STORE karta hai 
#    - Python/Java/C# mein MOST COMMON
# 3. Two-way Adapter
#    - DONO directions mein kaam karta hai (Target <-> Adaptee)
# 4. Interface Adapter (Default Adapter)
#    - Bade interfaces ke liye DEFAULT implementations deta hai

# REAL LIFE EXAMPLES:
# - File Format Converters: CSV/JSON/XML ko app ke format mein convert karna
# - Device Drivers: Operating System aur Hardware ke beech
# - Database Connectors: Alag alag DB engines ko SAME interface se access karna
# - Language Converters: Chinese-to-English, English-to-Hindi

# FAYDE (ADVANTAGES):
# - EXISTING code REUSE hota hai bina modify kiye
# - Classes apne CORE LOGIC par focused rehti hain (adaptation
#   ALAG mein isolated hai)
# - Naye interfaces INTERCHANGEABLE adapters se support ho sakte hain
# - System, IMPLEMENTATIONS se DECOUPLE ho jaata hai -- swap karna easy

# NUKSAAN (DISADVANTAGES):
# - Extra INDIRECTION se code samajhna THODA COMPLEX ho sakta hai
# - Thoda PERFORMANCE overhead (extra function calls) ho sakta hai
# - Bahut saare adapters ban jaayein to MAINTENANCE badh jaata hai
# - Bahut saare ALAG interfaces handle karne ke liye MULTIPLE
#   adapters chahiye ho sakte hain
```

```python
# Pehle dekho: BINA Adapter Pattern ke (Problem Code)
# - Chalo pehle ye dekhte hain ki Adapter Pattern KE BINA log purani 
#   `LegacyPrinter` class ko naye system mein use karne ki koshish karte hain, 
#   aur usme kya dikkat aati hai.
```

```python
# BINA ADAPTER KE — PROBLEM DEMO
# - Yahan "LegacyPrinter" ek PURANI class hai jiska method "print_document()" hai. 
# - Naya system EXPECT karta hai ki har printer ke paas "print()" method ho 
#   (jaise "Printer" interface mein defined hai). 
# - Dekho bina Adapter ke kya dikkat aati hai.

from abc import ABC, abstractmethod

# ---- Naya system ka EXPECTED interface ----
class Printer(ABC):
    @abstractmethod
    def print(self):
        pass


# ---- PURANI, INCOMPATIBLE class (isko hum MODIFY NAHI kar sakte -- legacy/third-party hai) ----
class LegacyPrinter:
    def print_document(self):
        print("Legacy Printer is printing a document.")


# ---- Client code jo "Printer" interface EXPECT karta hai ----
def client_code_bad(printer: Printer):
    printer.print()   # Client SIRF "print()" jaanta hai


# Client ye try karta hai:
legacy = LegacyPrinter()
try:
    client_code_bad(legacy)   # LegacyPrinter ke paas "print()" method HAI HI NAHI!
except AttributeError as e:
    print(f"❌ ERROR aaya: {e}")
```

```python
# ISME PROBLEM KYA THI?
# 1. INTERFACE MISMATCH: 
#    - Client "print()" call karna chahta hai, lekin LegacyPrinter ke paas SIRF 
#      "print_document()" hai.
#    - Upar dekha -- direct call karne se ERROR aa gaya!
#
# 2. LEGACY CODE MODIFY NAHI KAR SAKTE: 
#    - LegacyPrinter shayad ek THIRD-PARTY LIBRARY hai ya BAHUT PURANA, HEAVILY-USED 
#      code hai jise hum SEEDHA EDIT NAHI kar sakte (ya karna RISKY hai -- kahin aur
#     break ho sakta hai).
# 
# 2b. AGAR CLIENT CODE MEIN if-else DAAL DEIN (jaisa neeche dikhaya hai), to bhi 
#     problem hai:
#
#       if isinstance(printer, LegacyPrinter):
#           printer.print_document()
#       else:
#           printer.print()
#
#    - Ye TIGHT COUPLING create karta hai -- CLIENT ko HAR ALAG TYPE ke printer ke 
#      baare mein PATA hona padega, aur NAYA printer type aane par CLIENT CODE MODIFY
#      karna padega.
#
# 3. NOT SCALABLE: Agar kal EK AUR legacy system (jaise "OldFaxMachine" jiska method
#    "send_fax()" hai) add karna ho, to WAHI PROBLEM DOBARA aayegi.
#
# - YE SAB PROBLEMS "Adapter Design Pattern" SOLVE KARTA HAI! Neeche BEST APPROACH 
#   dekhte hain -- ek ADAPTER class banayenge jo LegacyPrinter ko "Printer" interface
#   ke SAATH COMPATIBLE bana degi, bina LegacyPrinter ko CHHUE.
```

```python
# BEST APPROACH — Adapter Pattern se solve karte hain
```

```python
# 1. TARGET INTERFACE (Printer)
#    Ye wo interface hai jo CLIENT EXPECT karta hai. Isme sirf ek abstract method hai:
#    - print().

class Printer(ABC):
    @abstractmethod
    def print(self):
        pass
```

```python
# 2. ADAPTEE - LegacyPrinter (Existing, Incompatible class)
#    Ye wahi PURANI class hai jise hum MODIFY NAHI kar sakte (ya nahi karna chahte). 
#    Iska method "print_document()" hai -- naam ALAG hai Target interface ke "print()"
#    se.

class LegacyPrinter:
    def print_document(self):
        print("Legacy Printer is printing a document.")
```

```python
# 3. ADAPTER - PrinterAdapter (Bridge)
#    Ye class Target interface (Printer) ko IMPLEMENT karti hai -- isliye Client isko 
#    "Printer" hi samjhega. Lekin ANDAR SE, jab "print()" call ho, ye LegacyPrinter ke
#    "print_document()" ko FORWARD kar deti hai.

#    YE HI "OBJECT ADAPTER" APPROACH hai -- COMPOSITION use kiya 
#    (LegacyPrinter ka OBJECT andar STORE kiya), INHERITANCE nahi.

class PrinterAdapter(Printer):
    def __init__(self, legacy_printer: LegacyPrinter):
        self.legacy_printer = legacy_printer   # Adaptee ka reference STORE kiya

    def print(self):
        # Client ka "print()" call, ANDAR LegacyPrinter ke "print_document()" ko FORWARD ho raha hai
        self.legacy_printer.print_document()
```

```python
# 4. CLIENT CODE - Sirf Target Interface se kaam karta hai
#    Client ko SIRF "Printer" interface pata hai -- use ye BILKUL NAHI PATA ki ANDAR 
#    "LegacyPrinter" hai. Adapter ne SAB CHHUPA diya!

def client_code(printer: Printer):
    printer.print()


# Ab LegacyPrinter ko Adapter mein WRAP karke Client ko dete hain
legacy_printer = LegacyPrinter()
adapter = PrinterAdapter(legacy_printer)

client_code(adapter)   # Ab koi ERROR nahi aayega!

print("\n✅ Adapter Pattern se PURANI LegacyPrinter class, NAYE system ke saath SEAMLESSLY kaam kar rahi hai -- bina uska code CHHUE.")
```

```python
# Problem Statement:
# - Ek Payment Gateway Integration System banao using Adapter Design Pattern
#   (jaise document ke "Different implementations" section mein "major API changes" 
#   ka example diya gaya tha).

# - GameForge Studio apne naye system mein PaymentProcessor interface use karta hai
#   jiska method pay(amount) hai. Lekin company ne ek PURANA, THIRD-PARTY 
#   OldBillingSystem licence kiya hai jiska method make_payment(rupees) hai — 
#   interface MISMATCH hai!

# Tumhe ye banana hai (Printer example jaisi hi structure use karo):
# 1. PaymentProcessor (Target Interface) — abstract method pay(amount)
# 2. OldBillingSystem (Adaptee) — method make_payment(rupees) jo print kare 
#    "OldBillingSystem: ₹{rupees} charged."
# 3. BillingAdapter (Adapter) — PaymentProcessor implement kare, andar OldBillingSystem
#    ka object store kare, pay(amount) mein make_payment(amount) ko FORWARD kare
# 4. Client code likho jo checkout(processor: PaymentProcessor, amount) function 
#    banaye, aur BillingAdapter use karke OldBillingSystem se payment process kare
```

```python
# SOLUTION EXPLANATION
# - Yahan bilkul Printer/LegacyPrinter/PrinterAdapter jaisi hi structure use ki hai, 
#   bas ab "Payment" bana rahe hain instead of "Printing".
#   1. PaymentProcessor    -> Target Interface
#   2. OldBillingSystem    -> Adaptee (existing, incompatible)
#   3. BillingAdapter      -> Adapter (bridge)
#   4. Neeche client code hai jo sab kuch jodta hai

from abc import ABC, abstractmethod

# ---- 1. Target Interface ----
class PaymentProcessor(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


# ---- 2. Adaptee ----
class OldBillingSystem:
    def make_payment(self, rupees):
        print(f"OldBillingSystem: ₹{rupees} charged.")


# ---- 3. Adapter ----
class BillingAdapter(PaymentProcessor):
    def __init__(self, old_billing_system: OldBillingSystem):
        self.old_billing_system = old_billing_system

    def pay(self, amount):
        self.old_billing_system.make_payment(amount)


# ---- 4. Client Code ----
def checkout(processor: PaymentProcessor, amount):
    processor.pay(amount)


old_system = OldBillingSystem()
billing_adapter = BillingAdapter(old_system)

checkout(billing_adapter, 1499)
```

## 🔌 Chapter 1: Adapter Pattern — Third-Party Physics Engine jodna

**Problem:** GameForge Studio apna khud ka `PhysicsBody` interface use karta hai
(`apply_force()`, `get_position()`). Lekin ab team ne ek **THIRD-PARTY Physics Engine**
LICENSE kiya hai jiska interface BILKUL ALAG hai (`push(x, y)`, `location()`) — GameForge
ka code isko SEEDHA use NAHI kar sakta, interfaces MATCH nahi karte!

**Solution:** Adapter Pattern se ek "TRANSLATOR" class banate hain jo third-party engine ko
GameForge ke interface jaisa DIKHATI hai.

```python
"""
ADAPTER PATTERN — Physics Engine Integration
--------------------------------------------------
Target             -> PhysicsBody (GameForge ka EXPECTED interface)
Adaptee              -> ThirdPartyPhysicsEngine (INCOMPATIBLE, EXISTING library)
Adapter                -> PhysicsEngineAdapter (translator, Target ko implement karta hai,
                         andar Adaptee ko WRAP karta hai)
"""

from abc import ABC, abstractmethod

# ---- Target Interface (jo GameForge EXPECT karta hai) ----
class PhysicsBody(ABC):
    @abstractmethod
    def apply_force(self, fx, fy):
        pass

    @abstractmethod
    def get_position(self):
        pass


# ---- Adaptee (THIRD-PARTY library, INCOMPATIBLE interface -- humne isko CHHED nahi sakte) ----
class ThirdPartyPhysicsEngine:
    def __init__(self):
        self.x, self.y = 0, 0

    def push(self, dx, dy):   # naam aur signature ALAG hai "apply_force" se
        print(f"⚙️ [ThirdParty Engine] push({dx}, {dy}) called")
        self.x += dx
        self.y += dy

    def location(self):        # naam ALAG hai "get_position" se
        return (self.x, self.y)


# ---- Adapter (TRANSLATOR) ----
class PhysicsEngineAdapter(PhysicsBody):
    def __init__(self, third_party_engine: ThirdPartyPhysicsEngine):
        self.engine = third_party_engine

    def apply_force(self, fx, fy):
        # GameForge ka "apply_force" call, ANDAR third-party ka "push" ko FORWARD kar diya
        self.engine.push(fx, fy)

    def get_position(self):
        # GameForge ka "get_position" call, ANDAR third-party ka "location" ko FORWARD kar diya
        return self.engine.location()
```

```python
"""
TEST: GameForge code, apne "PhysicsBody" interface se hi kaam karta hai
------------------------------------------------------------------------------
Client (GameForge) ko ye pata NAHI hai ki ANDAR third-party engine hai --
use bas apna FAMILIAR interface (apply_force, get_position) dikh raha hai.
"""

def move_game_object(body: PhysicsBody):
    body.apply_force(5, 2)
    print(f"📍 Naya position: {body.get_position()}")


third_party_engine = ThirdPartyPhysicsEngine()
adapted_body = PhysicsEngineAdapter(third_party_engine)

move_game_object(adapted_body)   # GameForge code ko third-party engine ka pata bhi nahi!

print("\n✅ Adapter Pattern se ek INCOMPATIBLE third-party library, GameForge ke SYSTEM mein SEAMLESSLY fit ho gayi.")
```

```python
# BRIDGE DESIGN PATTERN:
# - Bridge ek STRUCTURAL design pattern hai jo ek "ABSTRACTION" ko uski "IMPLEMENTATION"
#   se ALAG (SEPARATE) karta hai, taaki DONO INDEPENDENTLY (ek doosre se bina asar liye)
#   BADAL/EXTEND ho sakein.

# Bridge Pattern mein 2 HISSE hote hain:
#  1. ABSTRACTION       -> "KYA karna hai" (high-level concept)
#  2. IMPLEMENTATION      -> "KAISE karna hai" (low-level details)

# SOCHO EK EXAMPLE:
# - Ek Remote Control system hai jo DIFFERENT devices (TV, Light) CONTROL kar sakta hai.
# - Agar hum HAR remote-device COMBINATION ke liye ALAG CLASS banayein (TVRemote, 
#   LightRemote, AdvancedTVRemote, AdvancedLightRemote...), to COMBINATIONS EXPONENTIALLY
#   badh jaayengi (Decorator/Composite wali "class explosion" jaisi problem)!

# Bridge isko solve karta hai:
# - "Remote" (ABSTRACTION) ANDAR EK "Device" (IMPLEMENTATION) ka REFERENCE rakhta hai
# - Remote apne actions (turn_on, turn_off) ko Device ko DELEGATE (forward) kar deta hai
# - Naya Remote TYPE ya naya Device TYPE, DONO INDEPENDENTLY add kar sakte hain -- BINA
#   EK DOOSRE ko CHHUE

# MAIN COMPONENTS:
# 1. Abstraction
#    - CORE class jo Implementer ka REFERENCE rakhti hai 
#    - Ex: Vehicle
# 2. Refined Abstraction
#    - Abstraction ko EXTEND karti hai, FINER details ADD karti hai 
#    - Ex: Car, Bike
# 3. Implementer
#    - Interface jo Implementation classes DEFINE karta hai 
#    - Ex: Workshop
# 4. Concrete Implementation
#    - Implementer ka ACTUAL kaam karne wali class 
#    - Ex: Produce, Assemble

# KAISE KAAM KARTA HAI:
# 1. Abstraction ke ANDAR Implementer ka REFERENCE hota hai
# 2. Concrete Implementations ACTUAL behavior dete hain
# 3. Client SIRF Abstraction se INTERACT karta hai, Implementation se SEEDHA NAHI

# REAL LIFE EXAMPLES:
# - Remote Control aur Devices: EK universal remote MULTIPLE devices (TV, Radio, AC) 
#   control kar sakta hai
# - Shapes aur Colors: EK Shape (Circle/Rectangle) DIFFERENT Color implementations 
#   (Red/Blue) use kar sakta hai

# FAYDE (ADVANTAGES):
# - ABSTRACTION aur IMPLEMENTATION ke beech TIGHT COUPLING KAM hoti hai
# - DONO ko INDEPENDENTLY EXTEND kiya ja sakta hai
# - System ki MAINTAINABILITY aur SCALABILITY IMPROVE hoti hai

# NUKSAAN (DISADVANTAGES):
# - EXTRA interfaces/classes ki wajah se INITIAL DESIGN thodi COMPLEX ho jaati hai
# - SIMPLE applications ke liye ZAROORAT SE ZYADA ho sakta hai
# - BEGINNERS ke liye code SAMAJHNA THODA MUSHKIL ho sakta hai
```

```python
# Pehle dekho: BINA Bridge Pattern ke (Problem Code)
# - Chalo pehle ye dekhte hain ki Bridge Pattern KE BINA log alag-alag 
#   Vehicle × Workshop task ke liye classes kaise banate hain (jaise document ke 
#   "Without Bridge Design Pattern" section mein `Bus` example diya gaya tha), aur 
#   usme kya dikkat aati hai.
```

```python
# BINA BRIDGE KE — PROBLEM DEMO (SUBCLASSING SE)
# - Yahan HAR "Vehicle + Workshop-task" COMBINATION ke liye ALAG SUBCLASS bana rahe 
#   hain -- jaise document mein "Bus" ke liye "ProduceBus" aur "AssembleBus" dikhaye
#   gaye the.

class CarBad:
    pass


class BikeBad:
    pass


# HAR vehicle ke liye, HAR workshop-task ke liye ALAG SUBCLASS!
class ProduceCarBad(CarBad):
    def work(self):
        print("Car Produced", end="")


class AssembleCarBad(CarBad):
    def work(self):
        print(" And Assembled.")


class ProduceBikeBad(BikeBad):
    def work(self):
        print("Bike Produced", end="")


class AssembleBikeBad(BikeBad):
    def work(self):
        print(" And Assembled.")


# Client ko HAR combination ke liye ALAG classes yaad rakhni padti hain
ProduceCarBad().work()
AssembleCarBad().work()
print()
ProduceBikeBad().work()
AssembleBikeBad().work()
```

```python
# ISME PROBLEM KYA THI?
# 1. CLASS EXPLOSION: 
#    - Sirf 2 vehicles (Car, Bike) aur 2 workshop tasks (Produce, Assemble) ke liye 
#      BHI 4 classes bani (ProduceCarBad, AssembleCarBad, ProduceBikeBad,
#      AssembleBikeBad).
#    - Agar EK NAYA vehicle (Bus) aur EK NAYA task (QualityCheck) add ho, to 
#      COMBINATIONS 3x3 = 9 ho jaayenge -- YE PATTERN EXPONENTIALLY BADHTA hai.
#
# 2. TIGHT COUPLING: 
#    - "ProduceCarBad" SEEDHA "CarBad" ko EXTEND karta hai -- Vehicle aur Workshop-task
#      PERMANENTLY BOUND hain.
#    - Agar "Produce" LOGIC change karni ho (jaise SABHI vehicles ke liye), to HAR 
#      "ProduceXBad" class mein JAAKAR update karna padega.
# 
# 3. HARD TO MAINTAIN: 
#    - Document mein bataya gaya jaisa -- agar "Bus" class change karni ho, to 
#      "ProduceBus" aur "AssembleBus" BHI change karne pad sakte hain, aur agar 
#      change WORKSHOP-SPECIFIC hai, to "Bike" ki classes BHI CHHUNI PAD SAKTI HAIN!
# 
# 4. NOT INDEPENDENT: 
#    - Vehicle TYPES aur Workshop TASKS INDEPENDENTLY BADLE NAHI ja sakte -- DONO EK 
#      DOOSRE se TIGHTLY JUDE hain is INHERITANCE-BASED approach mein.
# 
# - YE SAB PROBLEMS "Bridge Design Pattern" SOLVE KARTA HAI!
```

```python
# BEST APPROACH — Bridge Pattern se solve karte hain
```

```python
# 1. IMPLEMENTER (Workshop)
#    Ye IMPLEMENTATION ka ALAG HIERARCHY hai. Isme SIRF EK abstract method hai: work().
#    Ye Vehicle (Abstraction) se BILKUL INDEPENDENT hai.

from abc import ABC, abstractmethod

class Workshop(ABC):
    @abstractmethod
    def work(self):
        pass
```

```python
# 2. CONCRETE IMPLEMENTATIONS (Produce, Assemble)
#    Ye ACTUAL workshop-task classes hain. 
#    In classes ko ye BILKUL PATA NAHI ki inhe "Car" use karega ya "Bike" -- ye BAS
#    apna KAAM karti hain.

class Produce(Workshop):
    def work(self):
        print("Produced", end="")


class Assemble(Workshop):
    def work(self):
        print(" And", end="")
        print(" Assembled.")
```

```python
# 3. ABSTRACTION (Vehicle)
#    Ye ABSTRACTION ka ALAG HIERARCHY hai. 
#    Isme DO Workshop (Implementer) REFERENCES store hote hain -- YE HI "BRIDGE" HAI!
#    Vehicle KHUD "work" NAHI karta, ye kaam Workshop OBJECTS ko DELEGATE kar deta hai.

class Vehicle(ABC):
    def __init__(self, workshop1: Workshop, workshop2: Workshop):
        self.workshop1 = workshop1   # <-- YE "BRIDGE" (composition) HAI
        self.workshop2 = workshop2

    @abstractmethod
    def manufacture(self):
        pass
```

```python
# 4. REFINED ABSTRACTIONS (Car, Bike)
#    Ye Vehicle ko EXTEND karti hain, aur apna "manufacture()" method DEFINE karti hain
#    -- ANDAR SE workshop1 aur workshop2 (jo KOI BHI Workshop implementation ho sakti 
#    hai) ko "work()" call kar DELEGATE kar deti hain.

class Car(Vehicle):
    def manufacture(self):
        print("Car", end=" ")
        self.workshop1.work()
        self.workshop2.work()


class Bike(Vehicle):
    def manufacture(self):
        print("Bike", end=" ")
        self.workshop1.work()
        self.workshop2.work()
```

```python
# 5. CLIENT CODE - Vehicle aur Workshop, DONO ka MIX-AND-MATCH karta hai
#    Ab dekho -- Car aur Bike, DONO ko SAME "Produce"/"Assemble" implementations ke 
#    saath USE kar sakte hain, BINA "ProduceCar", "ProduceBike" jaisi ALAG classes 
#    banaye!

vehicle1 = Car(Produce(), Assemble())
vehicle1.manufacture()

print()

vehicle2 = Bike(Produce(), Assemble())
vehicle2.manufacture()

print("\n\n✅ Bridge Pattern se Car aur Bike (Abstraction) SAME Produce/Assemble (Implementation) ko REUSE kar rahe hain -- naye combinations ke liye NAYI classes NAHI banani padin.")
```

```python
# Problem Statement:
# - Ek Universal Remote Control System banao using Bridge Design Pattern 
#   (jaise document ke intro mein "Remote Control aur Devices" example diya gaya tha).

# Ek RemoteControl (Abstraction) DIFFERENT Devices (Implementation) — jaise TV aur 
# Light — ko control kar sakta hai, bina HAR combination ke liye ALAG class banaye.

# Tumhe ye banana hai (Vehicle/Workshop example jaisi hi structure use karo):
# 1. Device (Implementer interface) — abstract methods turn_on() aur turn_off()
# 2. Concrete Implementations: 
#    - TV (print "📺 TV is ON" / "📺 TV is OFF"), 
#    - Light (print "💡 Light is ON" / "💡 Light is OFF")
# 3. RemoteControl (Abstraction) — constructor mein device: 
#    - Device STORE kare; abstract method toggle_power()
# 4. BasicRemote (Refined Abstraction) — 
#    - toggle_power() mein device.turn_on() call kare
# 5. Client code likho jo:
#    - Ek BasicRemote banaye jo TV ko control kare, aur toggle_power() call kare
#    - Ek AUR BasicRemote banaye jo Light ko control kare, aur toggle_power() call kare
#    - Dikhao ki SAME BasicRemote class, DIFFERENT devices ke saath kaam kar rahi hai
```

```python
# SOLUTION EXPLANATION
# - Yahan bilkul Vehicle/Workshop jaisi hi structure use ki hai, bas ab "Remote Control
#   System" bana rahe hain instead of "Vehicle Manufacturing".
#   1. Device        -> Implementer
#   2. TV, Light     -> Concrete Implementations
#   3. RemoteControl -> Abstraction
#   4. BasicRemote   -> Refined Abstraction
#   5. Neeche client code hai jo SAME remote class, ALAG devices ke saath use karta hai

from abc import ABC, abstractmethod

# ---- 1. Implementer ----
class Device(ABC):
    @abstractmethod
    def turn_on(self):
        pass

    @abstractmethod
    def turn_off(self):
        pass


# ---- 2. Concrete Implementations ----
class TV(Device):
    def turn_on(self):
        print("📺 TV is ON")

    def turn_off(self):
        print("📺 TV is OFF")


class Light(Device):
    def turn_on(self):
        print("💡 Light is ON")

    def turn_off(self):
        print("💡 Light is OFF")


# ---- 3. Abstraction ----
class RemoteControl(ABC):
    def __init__(self, device: Device):
        self.device = device   # <-- BRIDGE

    @abstractmethod
    def toggle_power(self):
        pass


# ---- 4. Refined Abstraction ----
class BasicRemote(RemoteControl):
    def toggle_power(self):
        self.device.turn_on()


# ---- 5. Client Code ----
tv_remote = BasicRemote(TV())
tv_remote.toggle_power()

light_remote = BasicRemote(Light())
light_remote.toggle_power()

print("\n✅ SAME 'BasicRemote' class ne DONO alag devices (TV, Light) control kiye -- koi 'TVRemote', 'LightRemote' banane ki zaroorat nahi padi.")
```

## 🌉 Chapter 2: Bridge Pattern — Enemy TYPE aur Renderer TYPE ko ALAG karna

**Extension of Adapter:** Adapter ne EK MISMATCHED interface fix kiya. Ab GameForge Studio
ko game **2D mode** aur **3D mode** DONO mein render karna hai. Agar hum HAR Enemy TYPE
(`Goblin`, `Dragon`) ke liye HAR Renderer TYPE (`2D`, `3D`) ki ALAG SUBCLASS banayein —
`Goblin2D`, `Goblin3D`, `Dragon2D`, `Dragon3D` — to ye Decorator wale "class explosion"
jaisi problem ban jaayegi!

**Solution:** Bridge Pattern se Enemy (ABSTRACTION) aur Renderer (IMPLEMENTATION) ko **ALAG
HIERARCHIES** mein rakhte hain, aur EK DOOSRE se "BRIDGE" (jodte) hain.

```python
"""
BRIDGE PATTERN — Enemy (Abstraction) + Renderer (Implementation)
-------------------------------------------------------------------------
Abstraction        -> Enemy (Renderer ka REFERENCE rakhta hai)
Refined Abstraction  -> GoblinEnemy, DragonEnemy
Implementor            -> Renderer (interface)
Concrete Implementor     -> Renderer2D, Renderer3D
"""

# ---- Implementor (rendering ka ALAG hierarchy) ----
class Renderer(ABC):
    @abstractmethod
    def render(self, shape_name):
        pass


# ---- Concrete Implementors ----
class Renderer2D(Renderer):
    def render(self, shape_name):
        print(f"🖼️ [2D Renderer] Drawing flat sprite for {shape_name}")


class Renderer3D(Renderer):
    def render(self, shape_name):
        print(f"🧊 [3D Renderer] Rendering 3D mesh for {shape_name}")


# ---- Abstraction (Enemy ka ALAG hierarchy, Renderer ka REFERENCE rakhta hai) ----
class Enemy(ABC):
    def __init__(self, renderer: Renderer):
        self.renderer = renderer   # <-- YE HI "BRIDGE" HAI

    @abstractmethod
    def display(self):
        pass


# ---- Refined Abstractions ----
class GoblinEnemy(Enemy):
    def display(self):
        self.renderer.render("Goblin")


class DragonEnemy(Enemy):
    def display(self):
        self.renderer.render("Dragon")
```

```python
"""
TEST: Enemy TYPE aur Renderer TYPE ko MIX-AND-MATCH kar sakte hain
------------------------------------------------------------------------
Sirf 4 classes (Renderer2D/3D, GoblinEnemy/DragonEnemy) se HAR
COMBINATION mil gaya -- naye "Goblin2D", "Goblin3D" jaise subclasses
BANANE ki ZAROORAT NAHI padi!
"""

goblin_2d = GoblinEnemy(Renderer2D())
goblin_2d.display()

dragon_3d = DragonEnemy(Renderer3D())
dragon_3d.display()

# Runtime par renderer BADAL bhi sakte hain!
goblin_2d.renderer = Renderer3D()
print("\nPlayer ne 3D mode ON kar diya:")
goblin_2d.display()

print("\n✅ Bridge Pattern se Enemy TYPE (Goblin/Dragon) aur Rendering TYPE (2D/3D) INDEPENDENTLY badal sakte hain -- class explosion se bach gaye.")
```

## 🌳 Chapter 3: Composite Pattern — Poori Scene ko TREE mein organize karna

**Extension of Bridge:** Ab humare paas ALAG-ALAG renderable enemies hain. Lekin ek REAL
game LEVEL mein SIRF enemies nahi hote — poori "Scene" ek **TREE STRUCTURE** hoti hai: ek
`Forest` group mein `Trees` (individual objects) aur `EnemyGroup` (jisme aur enemies hain)
DONO ho sakte hain. Agar hum INDIVIDUAL objects aur GROUPS ko ALAG-ALAG TAREEKE se treat
karein, to code MESSY ho jaayega (har jagah "agar ye group hai to loop karo, warna direct
render karo" jaisi if-else lagegi).

**Solution:** Composite Pattern se INDIVIDUAL objects aur GROUPS ko **SAME INTERFACE** se
UNIFORMLY treat karte hain.

```python
"""
COMPOSITE PATTERN — Scene Tree
------------------------------------
Component     -> SceneNode (common interface -- render())
Leaf            -> GameObjectLeaf (individual object, e.g. ek tree, ek enemy)
Composite         -> SceneGroup (GROUP jisme MULTIPLE SceneNodes ho sakte hain -- LEAF ya GROUP dono)
"""

# ---- Component (common interface) ----
class SceneNode(ABC):
    @abstractmethod
    def render(self, indent=0):
        pass


# ---- Leaf ----
class GameObjectLeaf(SceneNode):
    def __init__(self, name):
        self.name = name

    def render(self, indent=0):
        print("  " * indent + f"🎯 {self.name}")


# ---- Composite ----
class SceneGroup(SceneNode):
    def __init__(self, name):
        self.name = name
        self.children = []   # ismein SceneNode (Leaf YA Group) DONO ja sakte hain

    def add(self, node: SceneNode):
        self.children.append(node)

    def render(self, indent=0):
        print("  " * indent + f"📁 {self.name}/")
        for child in self.children:
            child.render(indent + 1)   # RECURSIVELY render -- CHILD leaf ho ya group, FARAK NAHI PADTA!
```

```python
"""
TEST: Poori Forest World Scene banate hain -- GROUPS ke andar GROUPS bhi ho sakte hain
------------------------------------------------------------------------------------------
"""

forest_scene = SceneGroup("Forest World")

trees = SceneGroup("Trees")
trees.add(GameObjectLeaf("Oak Tree #1"))
trees.add(GameObjectLeaf("Oak Tree #2"))

enemies_group = SceneGroup("Enemies")
enemies_group.add(GameObjectLeaf("Goblin #1"))
enemies_group.add(GameObjectLeaf("Goblin #2"))

forest_scene.add(trees)
forest_scene.add(enemies_group)
forest_scene.add(GameObjectLeaf("Treasure Chest"))   # ek DIRECT leaf bhi top-level par

forest_scene.render()

print("\n✅ Composite Pattern se hum PORE Scene TREE ko SIRF EK '.render()' call se PROCESS kar rahe hain -- chahe wo LEAF ho ya NESTED GROUP.")
```

## ☕ Chapter 4: Decorator Pattern — Player par PowerUps wrap karna

**Extension of Composite:** Composite ne humein bataya ki MULTIPLE objects ko TREE mein
kaise organize karein. Ab humein ek `Player` (jo pichli Creational notebook mein Builder
se bana tha) par **RUNTIME PowerUps** (Speed Boost, Shield) add karne hain — bina Player ki
CLASS ko chhue, aur JITNE CHAHO utne PowerUps COMBINE kar sakte hain.

**Solution:** Decorator Pattern se PowerUps ko Player ke UPAR "WRAP" karte hain.

```python
"""
DECORATOR PATTERN — Player PowerUps
------------------------------------------
Component      -> PlayerStats (interface)
ConcreteComponent -> BasePlayerStats
Decorator          -> PowerUpDecorator (abstract wrapper)
ConcreteDecorators   -> SpeedBoostDecorator, ShieldDecorator
"""

# ---- Component ----
class PlayerStats(ABC):
    @abstractmethod
    def get_speed(self):
        pass

    @abstractmethod
    def get_defense(self):
        pass

    @abstractmethod
    def describe(self):
        pass


# ---- Concrete Component ----
class BasePlayerStats(PlayerStats):
    def get_speed(self):
        return 10

    def get_defense(self):
        return 5

    def describe(self):
        return "Base Player"


# ---- Decorator (abstract) ----
class PowerUpDecorator(PlayerStats):
    def __init__(self, wrapped_stats: PlayerStats):
        self.wrapped_stats = wrapped_stats

    def get_speed(self):
        return self.wrapped_stats.get_speed()

    def get_defense(self):
        return self.wrapped_stats.get_defense()

    def describe(self):
        return self.wrapped_stats.describe()


# ---- Concrete Decorators ----
class SpeedBoostDecorator(PowerUpDecorator):
    def get_speed(self):
        return self.wrapped_stats.get_speed() + 5   # +5 speed

    def describe(self):
        return self.wrapped_stats.describe() + " + Speed Boost"


class ShieldDecorator(PowerUpDecorator):
    def get_defense(self):
        return self.wrapped_stats.get_defense() + 8   # +8 defense

    def describe(self):
        return self.wrapped_stats.describe() + " + Shield"
```

```python
"""
TEST: Player par RUNTIME par PowerUps COMBINE karte hain
----------------------------------------------------------------
"""

player_stats = BasePlayerStats()
print(f"{player_stats.describe()} -> Speed: {player_stats.get_speed()}, Defense: {player_stats.get_defense()}")

boosted_player = SpeedBoostDecorator(ShieldDecorator(BasePlayerStats()))
print(f"\n{boosted_player.describe()} -> Speed: {boosted_player.get_speed()}, Defense: {boosted_player.get_defense()}")

print("\n✅ Decorator Pattern se Player par PowerUps DYNAMICALLY combine hue -- BasePlayerStats ki CLASS bilkul NAHI chhui.")
```

## 🎭 Chapter 5: Facade Pattern — Poore GameEngine ko SIMPLE banana

**Extension of everything so far:** Ab tak humne Physics (Adapter), Renderer (Bridge), Scene
(Composite), aur PlayerStats (Decorator) — SAB ALAG ALAG subsystems bana liye. Lekin agar
`main.py` mein HAR game start karne wale ko in SAARE subsystems ko MANUALLY setup karna pade
(physics engine banao, adapter wrap karo, renderer chuno, scene banao...), to ye BAHUT
COMPLICATED aur ERROR-PRONE hoga.

**Solution:** Facade Pattern se ek **SIMPLE `GameEngineFacade`** banate hain jo in SAARE
subsystems ko ANDAR HI ANDAR SETUP kar deta hai — client ko SIRF `start_level()` call karna
hai.

```python
"""
FACADE PATTERN — GameEngineFacade
----------------------------------------
Facade ek SIMPLE interface deta hai jo ANDAR SE humare BANAYE HUE
saare subsystems (Physics Adapter, Bridge Renderer, Composite
Scene, Decorator Stats) ko COORDINATE karta hai.
"""

class GameEngineFacade:
    def __init__(self):
        # ANDAR HI ANDAR saare complex subsystems SETUP ho rahe hain
        self.physics = PhysicsEngineAdapter(ThirdPartyPhysicsEngine())
        self.renderer = Renderer2D()
        self.scene = SceneGroup("Level 1")
        self.player_stats = BasePlayerStats()

    def start_level(self):
        print("🎮 [Facade] Game Engine start ho raha hai...")

        # Physics setup (Adapter ke through)
        self.physics.apply_force(2, 0)

        # Scene populate karo (Composite ke through)
        goblin = GoblinEnemy(self.renderer)
        self.scene.add(GameObjectLeaf("Ground"))
        self.scene.add(GameObjectLeaf("Trees"))

        # Player stats boost karo (Decorator ke through)
        self.player_stats = SpeedBoostDecorator(self.player_stats)

        print("\n📍 Player Physics Position:", self.physics.get_position())
        print("\n🌳 Scene:")
        self.scene.render()
        print(f"\n🏃 Player: {self.player_stats.describe()} (Speed: {self.player_stats.get_speed()})")

        goblin.display()
```

```python
"""
TEST: Client ko SIRF Facade use karna hai -- ANDAR ki complexity ka pata NAHI
------------------------------------------------------------------------------------
"""

engine = GameEngineFacade()
engine.start_level()   # ek hi call mein Physics + Renderer + Scene + PlayerStats sab kaam kar gaye!

print("\n✅ Facade Pattern se Client ko SIRF '.start_level()' call karna pada -- Adapter/Bridge/Composite/Decorator ki DETAILS bilkul CHHUPI rahin.")
```

## 🪶 Chapter 6: Flyweight Pattern — Hazaaron Trees MEMORY-EFFICIENT render karna

**Extension of Composite:** Chapter 3 mein humne `Trees` group banaya tha. Lekin agar Forest
World mein **10,000 trees** hon, aur HAR Tree object apna khud ka **HEAVY mesh + texture
data** (jo MB mein ho sakta hai) STORE kare, to MEMORY EXPLODE ho jaayegi — jabki SAARE Oak
Trees ka MESH/TEXTURE SAME hai, sirf unki **POSITION** ALAG hai!

**Solution:** Flyweight Pattern se HEAVY, SHARED data (mesh, texture) ko ek baar STORE
karte hain aur SAARE trees mein SHARE karte hain — sirf UNIQUE data (position) HAR tree ke
paas alag rakhte hain.

```python
"""
FLYWEIGHT PATTERN — Tree Rendering
----------------------------------------
Flyweight (Intrinsic/Shared State) -> TreeType (mesh + texture -- HEAVY, SHARED)
Context (Extrinsic/Unique State)     -> Tree (position -- HALKA, UNIQUE)
Flyweight Factory                       -> TreeTypeFactory (SAME TreeType ko REUSE karta hai)
"""

# ---- Flyweight (HEAVY, SHARED data) ----
class TreeType:
    def __init__(self, name, mesh, texture):
        print(f"⏳ Naya TreeType '{name}' load ho raha hai (HEAVY mesh+texture, SIRF EK BAAR)...")
        self.name = name
        self.mesh = mesh          # socho ye MB ka data hai
        self.texture = texture    # socho ye bhi MB ka data hai

    def render(self, x, y):
        print(f"🌳 {self.name} (mesh={self.mesh}, texture={self.texture}) render at ({x}, {y})")


# ---- Flyweight Factory (SAME TreeType ko REUSE karta hai) ----
class TreeTypeFactory:
    _tree_types = {}

    @classmethod
    def get_tree_type(cls, name, mesh, texture):
        key = (name, mesh, texture)
        if key not in cls._tree_types:
            cls._tree_types[key] = TreeType(name, mesh, texture)
        else:
            print(f"♻️  TreeType '{name}' pehle se hai -- REUSE ho raha hai")
        return cls._tree_types[key]


# ---- Context (HALKA, UNIQUE data -- SIRF position) ----
class Tree:
    def __init__(self, x, y, tree_type: TreeType):
        self.x = x
        self.y = y
        self.tree_type = tree_type   # SHARED reference, apna COPY NAHI

    def render(self):
        self.tree_type.render(self.x, self.y)
```

```python
"""
TEST: 6 trees banate hain, lekin sirf 2 TreeType objects actually create honge
----------------------------------------------------------------------------------
"""

forest = []
for i in range(6):
    tree_type = TreeTypeFactory.get_tree_type("Oak", "oak_mesh.obj", "oak_texture.png")
    forest.append(Tree(x=i * 10, y=0, tree_type=tree_type))

# Ek DIFFERENT tree type bhi add karte hain
pine_type = TreeTypeFactory.get_tree_type("Pine", "pine_mesh.obj", "pine_texture.png")
forest.append(Tree(x=100, y=0, tree_type=pine_type))

print()
for tree in forest:
    tree.render()

print(f"\nTotal Tree (context) objects: {len(forest)}")
print(f"Total TreeType (HEAVY, shared) objects actually banaye gaye: {len(TreeTypeFactory._tree_types)}")

print("\n✅ Flyweight Pattern se 7 trees ke liye sirf 2 HEAVY TreeType objects bane -- MEMORY bahut saved hui!")
```

## 🕵️ Chapter 7: Proxy Pattern — Save-File ACCESS control karna

**Extension of Facade:** `GameEngineFacade` ne humein game START karna EASY bana diya. Ab
game ko **Save File** mein PROGRESS likhni hai — lekin sirf **LOGGED-IN, VERIFIED users**
ko hi save karne dena hai. Agar `SaveFile` class DIRECTLY expose kar di, to KOI BHI code
BINA CHECK ke seedha `.write()` call kar sakta hai — SECURITY RISK!

**Solution:** Proxy Pattern se `SaveFile` ke AAGE ek "GATEKEEPER" (Proxy) rakhte hain jo
ACCESS CHECK karke hi REAL object ko call allow karta hai.

```python
"""
PROXY PATTERN — Protected Save File Access
--------------------------------------------------
Subject       -> SaveFile (common interface)
RealSubject     -> RealSaveFile (ACTUAL file writing logic)
Proxy             -> SaveFileProxy (ACCESS CONTROL check karta hai, phir REAL object ko FORWARD karta hai)

NOTE: Ye Lazy Loading (Creational) ke Virtual Proxy se MILTA-JULTA
structure hai, lekin FOCUS ALAG hai -- yahan hum PERFORMANCE ke
liye nahi, SECURITY/ACCESS CONTROL ke liye proxy laga rahe hain.
"""

# ---- Subject ----
class SaveFile(ABC):
    @abstractmethod
    def write(self, user_role, data):
        pass


# ---- Real Subject ----
class RealSaveFile(SaveFile):
    def write(self, user_role, data):
        print(f"💾 Data SAVE ho gaya: '{data}'")


# ---- Proxy (Access Control) ----
class SaveFileProxy(SaveFile):
    def __init__(self):
        self.real_save_file = RealSaveFile()

    def write(self, user_role, data):
        if user_role != "verified_player":
            print(f"⛔ ACCESS DENIED! Role '{user_role}' ko save karne ki PERMISSION nahi hai.")
            return
        print(f"✅ Access GRANTED for role '{user_role}'")
        self.real_save_file.write(user_role, data)   # CHECK pass hua, ab REAL object ko forward
```

```python
"""
TEST: Proxy se hoke SAVE karne ki koshish karte hain
--------------------------------------------------------
"""

save_proxy = SaveFileProxy()

print("Guest user save karne ki koshish karta hai:")
save_proxy.write("guest", "Level 5 progress")

print("\nVerified player save karta hai:")
save_proxy.write("verified_player", "Level 5 progress")

print("\n✅ Proxy Pattern se 'RealSaveFile' KABHI DIRECT expose nahi hua -- HAR request ACCESS CHECK se GUZARTI hai.")
```

## 📋 Summary — Kaunsa Pattern Kab Use Karo?

| # | Pattern | Kis Sawaal Ka Jawab Deta Hai | GameForge Example |
|---|---------|-------------------------------|---------------------|
| 1 | **Adapter** | 2 incompatible interfaces ko kaise jodu? | Third-party Physics Engine |
| 2 | **Bridge** | Type aur Implementation ko alag kaise rakhu? | Enemy (Goblin/Dragon) × Renderer (2D/3D) |
| 3 | **Composite** | Tree/hierarchy ko uniform kaise treat karu? | Scene Graph (Groups + Leaves) |
| 4 | **Decorator** | Object mein runtime behavior kaise add karu? | Player PowerUps |
| 5 | **Facade** | Complex subsystem ko simple kaise dikhaun? | `GameEngineFacade` |
| 6 | **Flyweight** | Bahut saare similar objects memory-efficient kaise? | Trees (shared mesh/texture) |
| 7 | **Proxy** | Object ke access ko kaise control karu? | Save File permission check |
