```python
# Creational Design Patterns — GameForge Studio Story
# - Is notebook mein hum "saare 7 Creational Design Patterns" ek hi "connected story" 
#   ke through seekhenge — hum ek Game Engine, "GameForge Studio", dheere dheere BANATE
#   jaayenge, aur har naye problem ke liye ek NAYA pattern use karenge.
# 
# - Important: Ye patterns ISOLATED NAHI hain — har pattern PICHLE pattern ka 
#   NATURAL EXTENSION hai. Isse tumhe pata chalega ki REAL PROJECTS mein ye patterns
#   EK SAATH kaise kaam karte hain.

# Journey Map
# 1. Builder: 
#    - Ek complex `Player` character step-by-step banate hain
#
# 2. Prototype: 
#    - Us character ko TEMPLATE bana kar FAST clone karte hain 
#      (naye NPCs spawn karne ke liye)
#
# 3. Factory Method:
#    - Clone se kaam nahi chalega agar bilkul ALAG TYPE ka enemy chahiye — Factory 
#      decide karegi
#
# 4. Abstract Factory: 
#    - Ab poore "World Theme" ke RELATED objects (enemy + weapon) EK SAATH consistent 
#      banane hain
#
# 5. Singleton:
#    - Poore game ko ek hi `GameManager` control kare
#
# 6. Object Pool: 
#    - Bullets/particles ko baar baar create/destroy karne ke bajaye REUSE karo
#
# 7. Lazy Loading:
#    - Heavy textures/assets ko SIRF zaroorat par load karo
```

```python
# CREATIONAL DESIGN PATTERNS — YE CATEGORY KYA HAI?
# 
# - Design Patterns 3 BADI categories mein baante jaate hain:
#   1. CREATIONAL   -> "Object KAISE BANEGA?"
#   2. STRUCTURAL   -> "Objects EK DOOSRE SE KAISE JUDENGE?"
#   3. BEHAVIORAL   -> "Objects EK DOOSRE SE KAISE BAAT KARENGE?"
#
# - CREATIONAL PATTERNS ka SIRF EK KAAM hai: 
#   OBJECT CREATION ko SMART, FLEXIBLE, aur CONTROLLED banana -- seedha 
#   "new ClassName()" likhne ke bajaye.
```

```python
# YE ZAROORAT KYUN PADI?
# - Agar hum HAMESHA seedha constructor call karte rahein 
#   (Player("warrior", "sword", ...)), to:
#   > Constructor mein bahut saare PARAMETERS ho sakte hain (messy)
#   > HAR object SCRATCH SE banega, chahe wo EXPENSIVE ho ya REPEATED ho
#   > Client ko EXACT CONCRETE CLASS pata honi padegi (tight coupling)
#   > Object ki EXACT COUNT ya LIFECYCLE control karna mushkil hoga

# Creational Patterns in SAB problems ko ALAG ALAG TAREEKON se solve karte hain -- 
# har pattern ek DIFFERENT "creation problem" ka JAWAB hai:

# PATTERN              KIS SAWAL KA JAWAB DETA HAI?
# ------------------------------------------------------------
# 1. Builder           -> "Complex object STEP BY STEP kaise banau?"
# 2. Prototype         -> "Existing object se NAYA object FAST kaise banau?"
# 3. Factory Method    -> "RUNTIME par decide karna hai ki KAUNSA TYPE banega?"
# 4. Abstract Factory  -> "RELATED objects KA GROUP consistent kaise banau?"
# 5. Singleton         -> "Sirf EK HI instance kaise guarantee karu?"
# 6. Object Pool       -> "EXPENSIVE objects ko REUSE kaise karu?"
# 7. Lazy Loading      -> "Object ko SIRF zaroorat par kaise banau?"

# IMPORTANT INSIGHT:
# - Real projects mein ye patterns ISOLATED use NAHI hote -- ek SYSTEM mein AKSAR 
#   MULTIPLE creational patterns EK SAATH, EK DOOSRE KO SUPPORT karte hue use hote hain.
# - Isi cheez ko samajhne ke liye hum ab "GameForge Studio" banayenge -- dekhoge kaise
#   ek problem doosre pattern ki zaroorat KHUD PAIDA karta hai!
```

```python
# Chapter 1: Builder Pattern
# - Socho tumhe ek COMPLEX object banana hai — jaise ek "Computer".
# - Ek computer mein CPU, RAM, Storage, GPU, jaise bahut saare parts hote hain.
#
# - Agar hum ek hi CONSTRUCTOR mein sab kuch daal denge, jaise:
#   Computer(cpu, ram, storage, gpu, motherboard, cabinet, ...)
#
# - To ye constructor bahut lamba aur confusing ho jaayega. 
# - Isko "Telescoping Constructor Problem" kehte hain — matlab constructor mein 
#   parameters ka dher lag jaana.

# BUILDER PATTERN iska solution hai:
# -> Object ko STEP-BY-STEP banate hain, ek ek part karke.
# -> Construction logic ko ek ALAG "Builder" class mein daal dete hain.
# -> Isse SAME construction process use karke DIFFERENT variations
#    ke objects bana sakte hain (jaise Gaming PC, Office PC, etc.)

# MAIN PARTS (COMPONENTS):
# 1. Product: 
#    - Wo final complex object jo banana hai 
#    - Ex: Computer
# 2. Builder: 
#    - Ek interface/abstract class jisme part banane ke methods define hote hain 
#      (buildCPU, buildRAM, etc.)
# 3. Concrete Builder
#    - Builder ko implement karta hai, actual logic likhta hai ki har part kaise banega
#      (Gaming ke liye alag, Office ke liye alag)
# 4. Director
#    - Ye decide karta hai ki steps KIS ORDER mein call honge. 
#    - Isse client ko order yaad nahi rakhna padta.
# 5. Client
#    - Jo Builder aur Director use karke final product (object) prapt karta hai.

# REAL LIFE EXAMPLE:
# - SQL Query Builder (select(), where(), orderBy() -> chain karke query banti hai)
# - UI Form Builder (button add karo, text field add karo, style do)
# - Document/PDF Builder (heading do, paragraph do, image do)

# FAYDE (ADVANTAGES):
# - Bade constructors se chutkara milta hai
# - Alag alag variations ka object same process se ban sakta hai
# - Code readable aur maintainable rehta hai

# NUKSAAN (DISADVANTAGES):
# - Extra classes banani padti hain (Builder, ConcreteBuilder, Director)
# - Chhote/simple objects ke liye ye overkill ho sakta hai
```

```python
# Pehle dekho: BINA Builder Pattern ke (Problem Code)
# - Chalo pehle ye dekhte hain ki Builder Pattern KE BINA log `Computer` object kaise
#   banate hain, aur usme kya dikkat aati hai.
```

```python
# BINA BUILDER PATTERN KE — PROBLEM DEMO
# - Yahan Computer class ka CONSTRUCTOR mein SEEDHA saare parameters daal diye hain 
#   (jise "Telescoping Constructor" kehte hain).
# - Dekho neeche ye kaise use hota hai aur iska kya nuksaan hai.

class ComputerBad:
    def __init__(self, cpu, ram, storage, gpu=None, motherboard=None,
                 cabinet=None, cooling=None, wifi=None):
        self.cpu = cpu
        self.ram = ram
        self.storage = storage
        self.gpu = gpu
        self.motherboard = motherboard
        self.cabinet = cabinet
        self.cooling = cooling
        self.wifi = wifi

    def display_info(self):
        print(f"CPU: {self.cpu}, RAM: {self.ram}, Storage: {self.storage}, "
              f"GPU: {self.gpu}, Cooling: {self.cooling}, WiFi: {self.wifi}")


# Client ko YAAD rakhna padega ki kaunsa argument KIS position par hai!
pc1 = ComputerBad("Gaming CPU", "16GB DDR4", "1TB SSD", "RTX 4070", None, None, "Liquid Cooling", True)
pc1.display_info()

# Agar galti se order mix ho gaya (jaise cooling aur wifi ki jagah badal di)...
pc2 = ComputerBad("Office CPU", "8GB DDR4", "512GB SSD", None, None, None, True, "Liquid Cooling")
pc2.display_info()   # Yahan wifi=True daala tha, lekin ab "Cooling: True" print ho raha hai — BUG!
```

```python
# ISME PROBLEM KYA THI?
# 1. TELESCOPING CONSTRUCTOR PROBLEM: 
#    - Jaise jaise features (parts) badhte hain, constructor mein parameters ka 
#      DHER lag jaata hai.
#    - 3 params se 8 params ho gaye — agar 15 optional parts ho gaye to constructor 
#      bilkul UNREADABLE ho jaayega.
#
# 2. ORDER YAAD RAKHNA PADTA HAI: 
#    - Client ko exactly pata hona chahiye ki 7th position par "cooling" hai aur
#      8th par "wifi" — upar pc2 mein maine GALTI se inko SWAP kar diya, aur code
#      CRASH nahi hua, bas GALAT DATA store ho gaya 
#      (SILENT BUG — sabse khatarnak type ka bug, kyunki error bhi nahi dikhta!).
#
# 3. OPTIONAL FIELDS KE LIYE "None, None, None" likhna padta hai — ye bahut CONFUSING
#    aur error-prone hai, khaas kar jab bahut saare optional parts ho.
# 
# 4. HARD TO EXTEND: 
#    - Naya part (jaise "RGB Lighting") add karna ho to EXISTING constructor ki 
#      signature change karni padegi, aur jahan jahan Computer banaya gaya hai wahan
#      bhi dekhna padega.
# 
# YE SAB PROBLEMS "Builder Design Pattern" SOLVE KARTA HAI!
```

```python
# BEST APPROACH — Builder Design Pattern se solve karte hain
# - Neeche wahi `Computer` object step-by-step, ek alag "Builder class" ke through 
#   banayenge.
# - Har part apne naam se set hoga (build_cpu(), build_ram()...), order ki tension nahi,
#   aur naya part add karna bhi easy hoga.
```

```python
# 1. PRODUCT CLASS — `Computer`
#    - Ye wo FINAL OBJECT hai jise hum step by step banayenge.
#    - Isme sirf attributes (cpu, ram, storage) aur unhe set/display karne ke methods
#      hain. 
#    - Isko khud se "build" karna nahi aata — uske liye Builder class use hogi.

class Computer:
    def __init__(self):
        self.cpu = None       # abhi khaali hai, Builder isko set karega
        self.ram = None
        self.storage = None

    def set_cpu(self, cpu):
        self.cpu = cpu

    def set_ram(self, ram):
        self.ram = ram

    def set_storage(self, storage):
        self.storage = storage

    def display_info(self):
        print(f"Computer Configuration:\nCPU: {self.cpu}\nRAM: {self.ram}\nStorage: {self.storage}\n")
```

```python
# 2. BUILDER INTERFACE (Abstract Interface)
#    - Ye ek BLUEPRINT hai. 
#    - Isme sirf method names diye hain (abstract), inki body khaali hai. 
#    - Har Concrete Builder ko in methods ko apne hisaab se implement karna padega.
#
#    - @abstractmethod use karte hain taaki koi is class ka object seedha na bana sake,
#      aur har child class in methods ko zaroor implement kare.

from abc import ABC, abstractmethod

class Builder(ABC):
    @abstractmethod
    def build_cpu(self):
        pass

    @abstractmethod
    def build_ram(self):
        pass

    @abstractmethod
    def build_storage(self):
        pass

    @abstractmethod
    def get_result(self):
        pass
```

```python
# 3. CONCRETE BUILDER (GamingComputerBuilder)
#    - Ye Builder ko IMPLEMENT karta hai aur batata hai ki "Gaming Computer" ke liye 
#      har part KAISA hoga.
#    - Agar hum "OfficeComputerBuilder" banate, to usme CPU/RAM/Storage ki values ALAG
#      hoti — lekin process (build_cpu -> build_ram -> build_storage) SAME rehta.
#    - Yahi Builder Pattern ki power hai!

class GamingComputerBuilder(Builder):
    def __init__(self):
        self.computer = Computer()   # naya khaali Computer object banaya

    def build_cpu(self):
        self.computer.set_cpu("Gaming CPU (High Performance)")

    def build_ram(self):
        self.computer.set_ram("16GB DDR4")

    def build_storage(self):
        self.computer.set_storage("1TB SSD")

    def get_result(self):
        return self.computer   # final ready computer wapas do
```

```python
# 4. DIRECTOR
#    - Director ka kaam hai STEPS ka SAHI ORDER decide karna.
#    - Client ko ye yaad nahi rakhna padta ki pehle CPU banega ya RAM — Director khud
#      construct() method mein sahi sequence follow karta hai.
#
#    - Isse client sirf itna kehta hai: "Bhai, mujhe computer bana ke do" aur Director
#      andar hi andar sab steps sambhal leta hai.

class ComputerDirector:
    def construct(self, builder):
        builder.build_cpu()
        builder.build_ram()
        builder.build_storage()
```

```python
# 5. CLIENT CODE (Sab Kuch use karke final Computer banate hai)
#    - Client teen kaam karta hai:
#      1. Ek Concrete Builder object banata hai (yahan GamingComputerBuilder)
#      2. Director ko builder de deta hai 
#      3. Director se construct() karwa ke, final product (Computer) le leta hai
#    - Isse client ko Computer ke andar ke internal steps ki details nahi jaannni 
#      padti — sab kuch encapsulated hai!

gaming_builder = GamingComputerBuilder()
director = ComputerDirector()

director.construct(gaming_builder)             # step by step build karwaya
gaming_computer = gaming_builder.get_result()  # final object mila

gaming_computer.display_info()
```

```python
# Problem Statement:
# - Ek Pizza Builder System banao using Builder Design Pattern.
# - Ek Pizza ke ye parts hote hain:
#   - size     (e.g. "Small", "Medium", "Large")
#   - crust    (e.g. "Thin Crust", "Cheese Burst")
#   - toppings (e.g. "Mushroom, Capsicum, Olives")

# Tumhe ye classes banani hain (Computer example jaisi hi structure use karo):
# 1. Pizza (Product) — jisme size, crust, toppings attributes ho, aur ek display_info()
#    method ho.
#
# 2. PizzaBuilder (Builder interface) — abstract methods: 
#    - build_size(), build_crust(), build_toppings(), get_result()
# 
# 3. VegPizzaBuilder (Concrete Builder) — Veg Pizza banaye: 
#    - Size = "Medium", 
#    - Crust = "Cheese Burst", 
#    - Toppings = "Mushroom, Capsicum, Olives"
#
# 4. PizzaDirector — jo sahi order mein steps call kare (size -> crust -> toppings)
#
# 5. Client code likho jo VegPizzaBuilder aur PizzaDirector use karke ek pizza banaye 
#    aur display_info() print kare
```

```python
# SOLUTION EXPLANATION
# - Yahan bilkul Computer wale example jaisi hi structure use ki hai, bas ab "Pizza" 
#   bana rahe hain instead of "Computer".
#   1. Pizza            -> Product
#   2. PizzaBuilder     -> Builder (abstract)
#   3. VegPizzaBuilder  -> Concrete Builder
#   4. PizzaDirector    -> Director
#   5. Neeche client code hai jo sab kuch jodta hai

from abc import ABC, abstractmethod

# ---- 1. Product ----
class Pizza:
    def __init__(self):
        self.size = None
        self.crust = None
        self.toppings = None

    def set_size(self, size):
        self.size = size

    def set_crust(self, crust):
        self.crust = crust

    def set_toppings(self, toppings):
        self.toppings = toppings

    def display_info(self):
        print(f"Pizza Order:\nSize: {self.size}\nCrust: {self.crust}\nToppings: {self.toppings}\n")


# ---- 2. Builder (Abstract) ----
class PizzaBuilder(ABC):
    @abstractmethod
    def build_size(self):
        pass

    @abstractmethod
    def build_crust(self):
        pass

    @abstractmethod
    def build_toppings(self):
        pass

    @abstractmethod
    def get_result(self):
        pass


# ---- 3. Concrete Builder ----
class VegPizzaBuilder(PizzaBuilder):
    def __init__(self):
        self.pizza = Pizza()

    def build_size(self):
        self.pizza.set_size("Medium")

    def build_crust(self):
        self.pizza.set_crust("Cheese Burst")

    def build_toppings(self):
        self.pizza.set_toppings("Mushroom, Capsicum, Olives")

    def get_result(self):
        return self.pizza


# ---- 4. Director ----
class PizzaDirector:
    def construct(self, builder):
        builder.build_size()
        builder.build_crust()
        builder.build_toppings()


# ---- 5. Client ----
veg_builder = VegPizzaBuilder()
pizza_director = PizzaDirector()

pizza_director.construct(veg_builder)
veg_pizza = veg_builder.get_result()

veg_pizza.display_info()
```

```python
# Problem: 
# - GameForge Studio mein ek `Player` character banana hai. 
# - Player ke bahut saare PARTS hain — class, weapon, armor, skills. 
# - Agar ye sab ek CONSTRUCTOR mein daalein, to constructor bahut MESSY ho jaayega
#   (telescoping constructor problem).

# Solution: Builder Pattern se STEP BY STEP, PART BY PART character banate hain.
```

```python
# BUILDER PATTERN — Player Character
# 1. Product         -> Player               (final complex object)
# 2. Builder         -> PlayerBuilder        (abstract steps)
# 3. ConcreteBuilder -> WarriorPlayerBuilder (Warrior ke specific parts)
# 4. Director        -> PlayerDirector       (steps ka sahi order)

from abc import ABC, abstractmethod

# ---- Product ----
class Player:
    def __init__(self):
        self.character_class = None
        self.weapon = None
        self.armor = None
        self.skills = []

    def display(self):
        print(f"🎮 Player -> Class: {self.character_class}, Weapon: {self.weapon}, "
              f"Armor: {self.armor}, Skills: {self.skills}")


# ---- Builder (abstract) ----
class PlayerBuilder(ABC):
    @abstractmethod
    def build_class(self): pass
    @abstractmethod
    def build_weapon(self): pass
    @abstractmethod
    def build_armor(self): pass
    @abstractmethod
    def build_skills(self): pass
    @abstractmethod
    def get_result(self): pass


# ---- Concrete Builder 1: Kratos ----
class KratosPlayerBuilder(PlayerBuilder):

    def __init__(self):
        self.player = Player()

    def build_class(self):
        self.player.character_class = "God of War"

    def build_weapon(self):
        self.player.weapon = "Leviathan Axe"

    def build_armor(self):
        self.player.armor = "Spartan Armor"

    def build_skills(self):
        self.player.skills = [
            "Axe Throw",
            "Spartan Rage",
            "Shield Bash"
        ]

    def get_result(self):
        return self.player


# ---- Concrete Builder 2: Freya ----
class FreyaPlayerBuilder(PlayerBuilder):

    def __init__(self):
        self.player = Player()

    def build_class(self):
        self.player.character_class = "Valkyrie"

    def build_weapon(self):
        self.player.weapon = "Bows & Magic"

    def build_armor(self):
        self.player.armor = "Valkyrie Armor"

    def build_skills(self):
        self.player.skills = [
            "Nature Magic",
            "Valkyrie Magic",
            "Healing"
        ]

    def get_result(self):
        return self.player



# ---- Director ----
class PlayerDirector:
    def construct(self, builder: PlayerBuilder):
        builder.build_class()
        builder.build_weapon()
        builder.build_armor()
        builder.build_skills()
```

```python
# =========================================================
# TEST: Create Kratos
# =========================================================

kratos_builder = KratosPlayerBuilder()

director = PlayerDirector()
director.construct(kratos_builder)

kratos = kratos_builder.get_result()

kratos.display()


# =========================================================
# TEST: Create Freya
# =========================================================

freya_builder = FreyaPlayerBuilder()

director.construct(freya_builder)

freya = freya_builder.get_result()

freya.display()
```

```python
# 2. Prototype Design Pattern:
#    - Prototype ek CREATIONAL design pattern hai jo use hota hai jab object banana
#      TIME-CONSUMING ya COSTLY operation ho. Iske bajaye ki HAR BAAR scratch se naya
#      object banaya jaaye, hum EXISTING object ki EK COPY (CLONE) bana lete hain.

# SOCHO EK EXAMPLE:
# - Ek Game mein character banane ke liye graphics load karni padti hain, skills setup
#   karni padti hain, configurations load karni padti hain -- ye SAB TIME LETA HAI.
# - Agar 100 similar enemies (jaise "Zombie") spawn karne hain, to HAR BAAR SCRATCH SE
#   naya zombie banana bahut SLOW hoga. Iske bajaye:
#   - EK "prototype zombie" banao (ek baar heavy setup karo)
#   - Baaki sab zombies ko is prototype se CLONE karo (FAST!)
#   - Zaroorat ho to clone ki properties thoda modify kar do (jaise position, health)

# KAISE KAAM KARTA HAI (MECHANISM):
# 1. Ek "Prototype" interface banate hain jisme ek clone() method hota hai.
# 2. Concrete classes is interface ko implement karti hain aur apna clone() method
#    define karti hain (khud ki copy kaise banani hai).
# 3. Client sirf prototype.clone() call karta hai -- use pata nahi hona chahiye ki
#    object ANDAR SE kaise bana tha.

# MAIN COMPONENTS:
# 1. Prototype Interface
#    - clone() method declare karta hai
#    - Ex: Shape
# 2. Concrete Prototype
#    - Actual class jo apna clone() logic implement karti hai
#    - Ex: Circle
# 3. Client
#    - Prototype ka clone() call karke naya object leta hai 
#    - Ex: ShapeClient
# 4. Clone Method
#    - Define karta hai object COPY kaise hoga 
#     (shallow copy ya deep copy)

# REAL LIFE EXAMPLES:
# - Document/Template Systems: Existing template clone karke naya document banate hain
#   (Google Docs "Make a copy" jaisa)
# - Game Engines: Complex characters/environment objects ko re-initialize karne ke 
#   bajaye CLONE karte hain (performance ke liye)

# FAYDE (ADVANTAGES):
# - Costly object initialization se BACHTE hain (clone() fast hota hai)
# - Bahut saari subclasses banane ki zaroorat nahi -- variations clone karke banao
# - Runtime par dynamically naye object types add/remove kar sakte ho

# NUKSAAN (DISADVANTAGES):
# - Complex objects (jinke andar nested objects hon) ko clone karna MUSHKIL ho sakta hai
# - DEEP COPY implement karna complicated ho sakta hai
# - Agar sambhal ke na kiya jaaye to SHARED STATE ka bug aa sakta hai
#   (do clones EK HI internal list/object share kar rahe hon)
# - Har class ko apna clone() logic SAHI se implement karna padta hai
```

```python
# Pehle dekho: BINA Prototype Pattern ke (Problem Code)
# - Chalo pehle ye dekhte hain ki Prototype Pattern KE BINA log Game Characters 
#   kaise banate hain, aur usme kya dikkat aati hai.
```

```python
# BINA PROTOTYPE PATTERN KE — PROBLEM DEMO
# - Yahan GameCharacterBad ka __init__ EXPENSIVE hai -- graphics load karna, skills
#   setup karna, configuration load karna, sabko SIMULATE kar rahe hain "time.sleep()"
#   se (real app mein ye disk/database/network calls hoti).
#
# - Har baar jab naya character chahiye, HAMESHA SCRATCH SE ye POORA heavy process 
#   chalta hai -- chahe character 99% SAME ho!

import time

class GameCharacterBad:
    def __init__(self, name, character_class, weapon, level):
        print(f"⏳ '{name}' character bana rahe hain (heavy setup)...")
        self._load_graphics()        # expensive
        self._load_skills()          # expensive
        self._load_configurations()  # expensive
        self.name = name
        self.character_class = character_class
        self.weapon = weapon
        self.level = level

    def _load_graphics(self):
        time.sleep(0.3)   # graphics load hone ka simulation

    def _load_skills(self):
        time.sleep(0.3)   # skills setup hone ka simulation

    def _load_configurations(self):
        time.sleep(0.3)   # config load hone ka simulation

    def display(self):
        print(f"Character: {self.name}, Class: {self.character_class}, "
              f"Weapon: {self.weapon}, Level: {self.level}")


start = time.time()

# 3 similar "Warrior" characters banane hain -- lekin HAR BAAR
# poora heavy setup (graphics+skills+config) SCRATCH SE chal raha hai!
warrior1 = GameCharacterBad("Warrior_1", "Warrior", "Sword", 1)
warrior2 = GameCharacterBad("Warrior_2", "Warrior", "Sword", 1)
warrior3 = GameCharacterBad("Warrior_3", "Warrior", "Sword", 1)

end = time.time()

warrior1.display()
warrior2.display()
warrior3.display()
print(f"\n⏱️ Total time liya: {end - start:.2f} seconds")
```

```python
# ISME PROBLEM KYA THI?
# 1. REPEATED EXPENSIVE SETUP: 
#    - Teeno characters LAGBHAG IDENTICAL hain (sab "Warrior" class, "Sword" weapon, 
#      level 1) -- phir bhi HAR BAAR graphics load karna, skills setup karna, aur
#      config load karna PURA SCRATCH SE ho raha hai. Upar dekha, ~0.9 second se zyada
#      time laga sirf 3 characters ke liye!
#
# 2. RESOURCE WASTE: 
#    - Agar 1000 similar "Warrior" enemies spawn karne hon (jaise ek game level mein),
#      to ye APPROACH game ko BAHUT SLOW kar degi -- performance ki dikkat hogi.
#
# 3. DUPLICATE INITIALIZATION LOGIC: 
#    - Har baar constructor mein SAME parameters (name ke alawa) baar baar likhne padte
#      hain -- agar kal "Warrior" ki DEFAULT weapon change karni ho, to HAR JAGAH jahan
#      "GameCharacterBad(...)" call hua hai, wahan update karna padega.
#
# 4. NO WAY TO REUSE AN EXISTING OBJECT: 
#    - Agar hamare paas ek PEHLE SE BANA HUA achha configured character ho, to bhi hum
#      use REUSE nahi kar sakte -- naya banana hi padega SCRATCH SE.
#
# YE SAB PROBLEMS "Prototype Design Pattern" SOLVE KARTA HAI!
```

```python
# BEST APPROACH — Prototype Pattern se solve karte hain (Game Character)
# - Ab hum SIRF EK BAAR heavy setup karenge (prototype banate waqt), aur baaki 
#   characters ko FAST clone() karenge -- naye characters banane mein 
#   graphics/skills/config phir se load NAHI karni padegi.
```

```python
# PROTOTYPE APPROACH — GameCharacter
# - Yahan clone() method Python ke built-in "copy" module se object ki COPY bana raha
#   hai -- EXPENSIVE __init__ dobara call NAHI hota, bas EXISTING object ki memory copy
#   ho jaati hai (bahut FAST).

from copy import copy

class GameCharacter:
    def __init__(self, name, character_class, weapon, level):
        print(f"⏳ '{name}' PROTOTYPE bana rahe hain (heavy setup, SIRF EK BAAR)...")
        self._load_graphics()
        self._load_skills()
        self._load_configurations()
        self.name = name
        self.character_class = character_class
        self.weapon = weapon
        self.level = level

    def _load_graphics(self):
        time.sleep(0.3)

    def _load_skills(self):
        time.sleep(0.3)

    def _load_configurations(self):
        time.sleep(0.3)

    def clone(self):
        # copy() se naya object banta hai bina __init__ dobara chalaye --
        # isliye graphics/skills/config phir se load NAHI hote (FAST!)
        return copy(self)

    def display(self):
        print(f"Character: {self.name}, Class: {self.character_class}, "
              f"Weapon: {self.weapon}, Level: {self.level}")


start = time.time()

# Ek baar HEAVY setup karke "prototype" banaya
warrior_prototype = GameCharacter("Warrior_Prototype", "Warrior", "Sword", 1)

# Ab baaki characters ko sirf CLONE kar rahe hain -- FAST!
warrior1 = warrior_prototype.clone()
warrior1.name = "Warrior_1"

warrior2 = warrior_prototype.clone()
warrior2.name = "Warrior_2"

warrior3 = warrior_prototype.clone()
warrior3.name = "Warrior_3"

end = time.time()

warrior1.display()
warrior2.display()
warrior3.display()
print(f"\n⏱️ Total time liya: {end - start:.2f} seconds  (pehle wale se KAAFI kam!)")
```

```python
# Problem: Ab document ka original example dekhte hain — Shape / Circle / ShapeClient
# - Ye wahi structured example hai jo doc mein diya gaya tha (drawing application ke 
#   liye) — isse components ka standard naming (Prototype Interface, Concrete Prototype,
#   Client) clearly samajh aayega.
```

```python
# 1. PROTOTYPE INTERFACE (Shape)
#    - Ye ABSTRACT INTERFACE hai jo har shape ko follow karna hai. 
#    - Isme DO abstract methods hain: clone() aur draw().

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def clone(self):
        """Apni ek copy banao"""
        pass

    @abstractmethod
    def draw(self):
        """Shape ko draw karo"""
        pass
```

```python
# 2. CONCRETE PROTOTYPE (Circle)
#    - Circle, Shape interface ko IMPLEMENT karta hai. Iska clone() method EXPLICITLY
#      batata hai ki naya Circle object kaise banega (same "color" ke saath).

class Circle(Shape):
    def __init__(self, color):
        self.color = color

    def clone(self):
        # Yahan hum khud NAYA Circle object bana rahe hain SAME color ke saath
        return Circle(self.color)

    def draw(self):
        print(f"Drawing a {self.color} circle.")
```

```python
# 3. CLIENT (ShapeClient)
#    - Client ke paas ek "prototype" (shape) ka reference hai. 
#    - Jab bhi naya shape chahiye, wo bas prototype.clone() call karta hai -- use pata
#      nahi ki ANDAR clone() kaise implement hua hai.

class ShapeClient:
    def __init__(self, shape_prototype):
        self.shape_prototype = shape_prototype

    def create_shape(self):
        return self.shape_prototype.clone()
```

```python
# MAIN CODE — Sab kuch jodte hain
# 1. Ek concrete prototype banate hain (ek red circle)
# 2. Us prototype ko ShapeClient ko de dete hain
# 3. Client se naya shape "create" karwate hain (jo asal mein PROTOTYPE ka CLONE hai)
# 4. Naye shape ko draw karte hain

# Prototype banaya (ek red circle)
circle_prototype = Circle("red")

# Client ko prototype diya
client = ShapeClient(circle_prototype)

# Prototype se naya shape banaya (clone)
red_circle = client.create_shape()
red_circle.draw()

# Verify karte hain ki dono ALAG objects hain (independent copies)
print("\nKya prototype aur clone SAME object hain?", circle_prototype is red_circle)
print("Kya dono ka color SAME hai?", circle_prototype.color == red_circle.color)
```

```python
# Problem Statement:
# - Ek Document Template System banao using Prototype Design Pattern 
#   (jaise doc ke "Real Life Software Examples" section mein bataya gaya tha — 
#   Google Docs jaisa "Make a copy" feature).
# - Har Document ka title, content, aur formatting (jaise "Bold, 12pt") hota
#   hai. 
# - Naya document banane ke liye Company ek EXISTING TEMPLATE ko CLONE karti hai, phir
#   sirf title aur content change karti hai — formatting same rehta hai.

# Tumhe ye banana hai (Shape/Circle example jaisi hi structure use karo):
# 1. Document (Prototype interface/abstract class) —
#    - abstract methods clone() aur display()
#
# 2. ReportTemplate (Concrete Prototype) — jisme title, content, formatting attributes
#    hon, aur clone() method jo naya ReportTemplate object return kare same properties
#    ke saath
#
# 3. DocumentClient — jo ek prototype le kar create_document() method se naya document
#    banaye
# 
# 4. Client code likho jo:
#  - Ek ReportTemplate prototype banaye 
#    (title="Untitled Report", content="", formatting="Bold, 12pt")
#  - DocumentClient use karke 2 naye documents clone kare
#  - Dono clones ke title aur content ALAG-ALAG set kare 
#    (jaise "Sales Report" aur "HR Report")
#  - Verify kare ki dono clones ka formatting SAME hai lekin title/content ALAG hai
```

```python
# SOLUTION EXPLANATION
# - Yahan bilkul Shape/Circle/ShapeClient jaisi hi structure use ki hai, bas ab 
#   "Documents" bana rahe hain instead of "Shapes".
#   1. Document          -> Prototype Interface
#   2. ReportTemplate     -> Concrete Prototype
#   3. DocumentClient      -> Client
#   4. Neeche client code hai jo sab kuch jodta hai

from abc import ABC, abstractmethod

# ---- 1. Prototype Interface ----
class Document(ABC):
    @abstractmethod
    def clone(self):
        pass

    @abstractmethod
    def display(self):
        pass


# ---- 2. Concrete Prototype ----
class ReportTemplate(Document):
    def __init__(self, title, content, formatting):
        self.title = title
        self.content = content
        self.formatting = formatting

    def clone(self):
        # Naya ReportTemplate object banate hain SAME properties ke saath
        return ReportTemplate(self.title, self.content, self.formatting)

    def display(self):
        print(f"Title: {self.title}\nContent: {self.content}\nFormatting: {self.formatting}\n")


# ---- 3. Client ----
class DocumentClient:
    def __init__(self, document_prototype):
        self.document_prototype = document_prototype

    def create_document(self):
        return self.document_prototype.clone()


# ---- 4. Client Code ----
# Prototype banaya -- ek default, blank template
template_prototype = ReportTemplate("Untitled Report", "", "Bold, 12pt")

client = DocumentClient(template_prototype)

# Pehla document clone karke banaya
sales_report = client.create_document()
sales_report.title = "Sales Report"
sales_report.content = "Q3 sales grew by 15%."

# Doosra document clone karke banaya
hr_report = client.create_document()
hr_report.title = "HR Report"
hr_report.content = "New hiring plan for next quarter."

sales_report.display()
hr_report.display()

# Verify karte hain
print("Formatting same hai dono mein?", sales_report.formatting == hr_report.formatting)
print("Title alag hai dono mein?", sales_report.title != hr_report.title)
print("Kya dono ALAG objects hain?", sales_report is not hr_report)
```

```python
# Prototype Pattern — `hero` ko TEMPLATE bana kar Fast Clone karna

# Extension of Builder: 
# - Humne abhi `Builder` se `hero` naam ka EK detailed Player banaya.
# - Ab game mein 50 SIMILAR "Warrior NPCs" spawn karne hain — agar HAR baar Builder
#   se SCRATCH banayein, to slow hoga (Director + Builder steps baar baar chalenge).

# Solution: `hero` ko ek PROTOTYPE bana kar, use CLONE karke naye NPCs FAST banate hain.
```

```python
# PROTOTYPE PATTERN — Player ko clone() ki capability do
# - Humne PICHLE chapter ka Player class hi use kiya -- bas ab ek clone() method 
#   ADD kar rahe hain. 
# - Ye dikhata hai ki Builder se bana object AAGE Prototype ke roop mein REUSE ho 
#   sakta hai.

# BUILDER PATTERN — Player Character
# 1. Product         -> Player
# 2. Builder         -> PlayerBuilder
# 3. ConcreteBuilder -> KratosPlayerBuilder / FreyaPlayerBuilder
# 4. Director        -> PlayerDirector


from abc import ABC, abstractmethod
from copy import deepcopy


# 1. Product
class Player:

    def __init__(self):
        self.character_class = None
        self.weapon = None
        self.armor = None
        self.skills = []

    def display(self):
        print(
            f"🎮 Player -> "
            f"Class: {self.character_class}, "
            f"Weapon: {self.weapon}, "
            f"Armor: {self.armor}, "
            f"Skills: {self.skills}"
        )


# 2. Builder (Abstract)
class PlayerBuilder(ABC):

    @abstractmethod
    def build_class(self):
        pass

    @abstractmethod
    def build_weapon(self):
        pass

    @abstractmethod
    def build_armor(self):
        pass

    @abstractmethod
    def build_skills(self):
        pass

    @abstractmethod
    def get_result(self):
        pass


# 3. Concrete Builder 1: Kratos
class KratosPlayerBuilder(PlayerBuilder):

    def __init__(self):
        self.player = Player()

    def build_class(self):
        self.player.character_class = "God of War"

    def build_weapon(self):
        self.player.weapon = "Leviathan Axe"

    def build_armor(self):
        self.player.armor = "Spartan Armor"

    def build_skills(self):
        self.player.skills = [
            "Axe Throw",
            "Spartan Rage",
            "Shield Bash"
        ]

    def get_result(self):
        return self.player


# 4. Concrete Builder 2: Freya
class FreyaPlayerBuilder(PlayerBuilder):

    def __init__(self):
        self.player = Player()

    def build_class(self):
        self.player.character_class = "Valkyrie"

    def build_weapon(self):
        self.player.weapon = "Bows & Magic"

    def build_armor(self):
        self.player.armor = "Valkyrie Armor"

    def build_skills(self):
        self.player.skills = [
            "Nature Magic",
            "Valkyrie Magic",
            "Healing"
        ]

    def get_result(self):
        return self.player


# 5. Director
class PlayerDirector:

    def construct(self, builder: PlayerBuilder):

        builder.build_class()
        builder.build_weapon()
        builder.build_armor()
        builder.build_skills()


# BUILDER TEST
kratos_builder = KratosPlayerBuilder()

director = PlayerDirector()
director.construct(kratos_builder)

hero = kratos_builder.get_result()


print("========== ORIGINAL HERO ==========")

hero.display()


# PROTOTYPE PATTERN
# - PROTOTYPE PATTERN — Player ko clone() ki capability do
# - Humne PICHLE chapter ka Player class hi use kiya.
# - Bas ab Player mein clone() method ADD kar rahe hain.
# - Isse Builder se bana hua "hero" AAGE Prototype ke
#   roop mein REUSE ho sakta hai.
#
# - deepcopy() use kar rahe hain taaki skills list bhi
#   INDEPENDENT copy ho.


class Player(Player):

    def clone(self):

        # deepcopy():
        # - Original object ki completely independent copy banata hai.
        # - Isliye skills list bhi shared nahi hogi.

        return deepcopy(self)


# hero ko Player (clone() wale naye version) mein convert karte hain
hero.__class__ = Player


# ============================================================
# PROTOTYPE TEST
# ============================================================
# - TEST: hero (prototype) se 3 NPCs FAST clone karte hain
# - Dekho: Builder/Director dobara CALL nahi karna pada.
# - Bas: clone() + properties change


# ---- NPC 1 ----
npc1 = hero.clone()
npc1.character_class = "Warrior NPC #1"

# ---- NPC 2 ----
npc2 = hero.clone()
npc2.character_class = "Warrior NPC #2"

# ---- NPC 3 ----
npc3 = hero.clone()
npc3.character_class = "Warrior NPC #3"


# ============================================================
# DISPLAY NPCs
# ============================================================

print("\n========== CLONED NPCs ==========")

for npc in [npc1, npc2, npc3]:
    npc.display()


# ============================================================
# VERIFY: Independent Skills
# ============================================================

npc1.skills.append("Battle Cry")


print("\n========== VERIFY INDEPENDENT COPY ==========")

print("npc1 ki skills:", npc1.skills)
print("npc2 ki skills:", npc2.skills)


# ============================================================
# VERIFY: Different Objects
# ============================================================

print(
    "\nKya npc1 aur npc2 ALAG objects hain?",
    npc1 is not npc2
)

print(
    "Kya npc1 ki skills aur npc2 ki skills ALAG hain?",
    npc1.skills is not npc2.skills
)


print(
    "\n✅ Prototype Pattern se hum SAME TYPE ke "
    "NPCs FAST bana rahe hain, bina Builder dobara chalaye."
)
```

```python
# Chapter 3: FACTORY METHOD DESIGN PATTERN:
# - Factory Method ek CREATIONAL design pattern hai jo object banane (create karne) ka
#   kaam ek ALAG method ko de deta hai, seedha "new ClassName()" karne ke bajaye.

# MAIN IDEA:
# - Ek "interface/method" define karo jo object banaye, lekin ye DECIDE karna ki 
#   EXACT kaunsa object banega, wo SUBCLASSES par chhod do.

# SOCHO EK EXAMPLE:
# - Ek Notification System hai jisme 3 tarah ke notifications bhej sakte hain: 
#   Email, SMS, Push.
# - Agar hum client code mein hi likh dein:
#   if type == "email": obj = EmailNotification()
#   elif type == "sms": obj = SMSNotification()
#   elif type == "push": obj = PushNotification()
# - To ye TIGHT COUPLING create karta hai — client ko HAR product class ke baare mein
#   pata hona chahiye, aur naya notification type (jaise WhatsApp) add karne ke liye
#   CLIENT CODE ko modify karna padega. 
# - Ye "Open/Closed Principle" todta hai (code extend hona chahiye, modify nahi).

# FACTORY METHOD isko solve karta hai:
# - Ek Factory (Creator) class/interface banate hain jisme "create_product()" jaisa 
#   method hota hai
# - Har Concrete Factory apna khud ka object banati hai
# - Client sirf Factory se object maangta hai, use ye pata nahi hota ki andar kaunsi
#   class use ho rahi hai

# MAIN COMPONENTS:
# 1. Product
#    - Interface/abstract class jo sab objects follow karte hain
#    - Ex: Notification
# 2. Concrete Product
#    - Actual classes (EmailNotification, SMSNotification)
# 3. Creator (Factory)
#    - Interface jisme factory method declare hota hai
#    - Ex: NotificationFactory
# 4. Concrete Creator
#    - Har concrete creator apna specific product banata hai
#    - Ex: EmailNotificationFactory

# REAL LIFE EXAMPLES:
# - Web Browsers: content type ke hisaab se sahi renderer/plugin banate hain
# - Payment Gateways (Stripe, PayPal): user selection se sahi payment processor object
#   banate hain
# - Game Development: level/environment ke hisaab se enemy/NPC spawn karte hain

# FAYDE (ADVANTAGES):
# - Object creation logic encapsulate ho jaati hai
# - Loose coupling — client sirf abstraction (interface) par depend karta hai
# - Naya product type add karna easy hai (existing code modify nahi karna padta)

# NUKSAAN (DISADVANTAGES):
# - Extra classes ban jaati hain (har product ke liye ek factory)
# - Chhote/simple applications ke liye zaroorat se zyada complexity ho sakti hai
```

```python
# PRODUCT INTERFACE (Notification)
# - Ye common INTERFACE hai jise har notification type follow karega.
# - Isme sirf ek abstract method hai: send()

from abc import ABC, abstractmethod

class Notification(ABC):
    @abstractmethod
    def send(self, message):
        pass
```

```python
# CONCRETE PRODUCTS
# - Ye actual notification classes hain jo Notification interface ko implement karti hain.
# - Har class apne tareeke se message "send" karti hai.

class EmailNotification(Notification):
    def send(self, message):
        print(f"📧 Sending EMAIL notification: {message}")


class SMSNotification(Notification):
    def send(self, message):
        print(f"📱 Sending SMS notification: {message}")


class PushNotification(Notification):
    def send(self, message):
        print(f"🔔 Sending PUSH notification: {message}")
```

```python
# CREATOR INTERFACE (Factory)
# - Ye FACTORY interface hai. Isme "create_notification()" naam ka abstract method hai
#   -- yahi hamara FACTORY METHOD hai.
# - Note: Isme ek "notify()" method bhi hai jo factory method ko CALL karta hai lekin
#   khud ye nahi jaanta ki kaunsa concrete product banega -- ye subclass decide karegi!

class NotificationFactory(ABC):
    @abstractmethod
    def create_notification(self):
        pass   # <-- Ye hi FACTORY METHOD hai

    def notify(self, message):
        # Factory method ko call karke product object mil raha hai
        notification = self.create_notification()
        notification.send(message)
```

```python
# CONCRETE CREATORS (Concrete Factories)
# - Har Concrete Factory apna KHUD KA specific product return karti hai.
# - Client ko ye pata nahi hona chahiye ki andar "EmailNotification()" banaya jaa raha
#   hai -- bas factory se maang lo!

class EmailNotificationFactory(NotificationFactory):
    def create_notification(self):
        return EmailNotification()


class SMSNotificationFactory(NotificationFactory):
    def create_notification(self):
        return SMSNotification()


class PushNotificationFactory(NotificationFactory):
    def create_notification(self):
        return PushNotification()
```

```python
# CLIENT CODE
# - Ab client sirf ITNA jaanta hai: "Mujhe ek NotificationFactory do, main uska notify()
#   call kar dunga." 
# - Client ko ye MATLAB NAHI ki andar EmailNotification banega ya SMSNotification -- 
#   YE HI FACTORY METHOD PATTERN KI TAAKAT HAI!
# - Agar kal ek "WhatsAppNotificationFactory" add karni ho, to EXISTING code CHHEDNE ki
#   zaroorat nahi -- bas NAYI class banao!

def send_user_notification(factory: NotificationFactory, message: str):
    factory.notify(message)


# User ne "Email" choose kiya
send_user_notification(EmailNotificationFactory(), "Your order has been shipped!")

# User ne "SMS" choose kiya
send_user_notification(SMSNotificationFactory(), "Your OTP is 4521")

# User ne "Push" choose kiya
send_user_notification(PushNotificationFactory(), "You have a new follower!")
```

```python
# Problem Statement:
# - Ek Payment Gateway System banao using Factory Method Design Pattern.
# - Ek E-commerce app mein user teen tarah se payment kar sakta hai:
#   - Credit Card
#   - UPI
#   - PayPal
#
# - Har payment method ka apna pay(amount) tareeka hota hai (message alag hoga).
# - Tumhe ye banana hai (Notification example jaisi hi structure use karo):
#   1. PaymentMethod (Product interface)
#      - abstract method pay(amount)
#   2. Concrete Products: 
#      - CreditCardPayment, UPIPayment, PayPalPayment — har ek apna print message de 
#      - (e.g. "💳 Paid ₹{amount} using Credit Card")
#   3. PaymentFactory (Creator interface) — 
#      - abstract method create_payment_method(), aur ek concrete method 
#        process_payment(amount) jo factory method call karke pay() chalaye
#   4. Concrete Creators: 
#      - CreditCardPaymentFactory, UPIPaymentFactory, PayPalPaymentFactory
#   5. Client code likho jo teeno factories use karke process_payment(amount) call 
#      kare
```

```python
# SOLUTION EXPLANATION
# - Yahan bilkul Notification System jaisi hi structure use ki hai, bas ab "Payment" 
#   bana rahe hain instead of "Notification".
#   1. PaymentMethod           -> Product (interface)
#   2. CreditCardPayment,
#      UPIPayment,
#      PayPalPayment           -> Concrete Products
#   3. PaymentFactory          -> Creator (Factory interface)
#   4. CreditCardPaymentFactory,
#      UPIPaymentFactory,
#      PayPalPaymentFactory    -> Concrete Creators
#   5. Neeche client code hai jo sab kuch jodta hai

from abc import ABC, abstractmethod

# ---- 1. Product ----
class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


# ---- 2. Concrete Products ----
class CreditCardPayment(PaymentMethod):
    def pay(self, amount):
        print(f"💳 Paid ₹{amount} using Credit Card")


class UPIPayment(PaymentMethod):
    def pay(self, amount):
        print(f"📲 Paid ₹{amount} using UPI")


class PayPalPayment(PaymentMethod):
    def pay(self, amount):
        print(f"🅿️ Paid ₹{amount} using PayPal")


# ---- 3. Creator (Factory Interface) ----
class PaymentFactory(ABC):
    @abstractmethod
    def create_payment_method(self):
        pass   # <-- Factory method

    def process_payment(self, amount):
        payment_method = self.create_payment_method()
        payment_method.pay(amount)


# ---- 4. Concrete Creators ----
class CreditCardPaymentFactory(PaymentFactory):
    def create_payment_method(self):
        return CreditCardPayment()


class UPIPaymentFactory(PaymentFactory):
    def create_payment_method(self):
        return UPIPayment()


class PayPalPaymentFactory(PaymentFactory):
    def create_payment_method(self):
        return PayPalPayment()


# ---- 5. Client Code ----
def checkout(factory: PaymentFactory, amount):
    factory.process_payment(amount)


checkout(CreditCardPaymentFactory(), 1500)
checkout(UPIPaymentFactory(), 799)
checkout(PayPalPaymentFactory(), 2999)
```

```python
# Factory Method Pattern — ALAG TYPE ke enemies banana

# Extension of Prototype: 
# - Clone karna tab kaam karta hai jab enemy SAME TYPE ka ho (sirf stats/naam alag). 
# - Lekin ab GameForge Studio mein game LEVEL ke hisaab se BILKUL ALAG TYPE ke enemies
#   chahiye — kabhi `Goblin`, kabhi `Dragon`. 
# - Prototype clone yahan kaam NAHI karega, kyunki ye ALAG CLASSES hain, sirf STATS 
#   ka farak nahi.

# Solution: 
# - Factory Method se DECIDE karte hain RUNTIME par ki KAUNSI CLASS banegi.
```

```python
# FACTORY METHOD PATTERN — Enemy Spawning
# 1. Product             -> Enemy (interface)
# 2. Concrete Products   -> Goblin, Dragon
# 3. Creator             -> EnemyFactory (factory method: create_enemy())
# 4. Concrete Creators   -> GoblinFactory, DragonFactory

# ---- Product ----
class Enemy(ABC):
    @abstractmethod
    def attack(self):
        pass


# ---- Concrete Products ----
class Goblin(Enemy):
    def attack(self):
        print("👺 Goblin attacks with a small dagger!")


class Dragon(Enemy):
    def attack(self):
        print("🐉 Dragon breathes FIRE!")


# ---- Creator (Factory) ----
class EnemyFactory(ABC):
    @abstractmethod
    def create_enemy(self):
        pass   # <-- Factory Method

    def spawn(self):
        enemy = self.create_enemy()
        enemy.attack()


# ---- Concrete Creators ----
class GoblinFactory(EnemyFactory):
    def create_enemy(self):
        return Goblin()


class DragonFactory(EnemyFactory):
    def create_enemy(self):
        return Dragon()
```

```python
# TEST: Level ke hisaab se ALAG enemy spawn karte hain

def spawn_enemy_for_level(level):
    if level < 5:
        factory = GoblinFactory()
    else:
        factory = DragonFactory()
    factory.spawn()


print("Level 2 mein enemy:")
spawn_enemy_for_level(2)

print("\nLevel 8 mein enemy:")
spawn_enemy_for_level(8)

print("\n✅ Factory Method se RUNTIME par decide hua ki kaunsa Enemy TYPE banega -- client ko if-else nahi likhna pada seedha classes ke saath.")
```

```python
# Chapter 4: Abstract Factory Pattern
# - Abstract Factory ek CREATIONAL design pattern hai jise "FACTORY OF FACTORIES" bhi
#   kaha jaata hai. 
# - Ye pattern RELATED OBJECTS KE "FAMILIES" (groups) banane mein help karta hai, 
#   without knowing unki exact concrete classes.

# FACTORY METHOD vs ABSTRACT FACTORY -- FARAK KYA HAI?
# - Factory Method: EK product banata hai (jaise sirf Notification)
# - Abstract Factory: RELATED PRODUCTS KA POORA GROUP/FAMILY banata hai 
#   (jaise Car + CarSpecification, dono EK SAATH, EK HI region ke consistent)

# SOCHO EK EXAMPLE:
# - Ek car company North America aur Europe dono ke liye cars banati hai. 
# - Har region ki apni requirements hain:
#   - North America -> Sedan car + North America Safety Specification
#   - Europe        -> Hatchback car + Europe Emission Specification

# IMPORTANT: 
# - Car aur Specification HAMESHA SAME REGION KI hone chahiye -- 
#   North America ki Sedan ke saath Europe ki specification LAGANA GALAT hoga! 
# - Abstract Factory ye "consistency" guarantee karta hai.

# MAIN COMPONENTS:
# 1. Abstract Factory
#    - Interface jisme related products banane ke methods hote hain 
#    - Ex: CarFactory mein create_car() aur create_specification())
# 2. Concrete Factory
#    - Har concrete factory EK FAMILY ke saare related products banati hai
#    - Ex: NorthAmericaCarFactory Sedan + NA Spec dono banayegi)
# 3. Abstract Product
#    - Har product type ka common interface
#    - Ex: Car, CarSpecification
# 4. Concrete Product
#    - Actual product classes 
#    - (Sedan, Hatchback, NorthAmericaSpecification, EuropeSpecification)
# 5. ClientSirf 
#    - Abstract Factory aur Abstract Product ke through kaam karta hai, 
#      concrete classes ko kabhi seedha nahi jaanta

# REAL LIFE EXAMPLES:
# - UI Theme Systems: 
#   - Light Theme Factory (Light Button + Light TextBox) 
#      vs 
#   - Dark Theme Factory (Dark Button + Dark TextBox)
#
# - Cloud Platforms: 
#   - AWSFactory (S3 Storage + EC2 Compute) 
#      vs 
#   - AzureFactory (Blob Storage + VM Compute)
#
# - Database Systems: 
#   SQLFactory vs NoSQLFactory 
#   (dono apne related connection + query objects banate hain)

# FAYDE (ADVANTAGES):
# - Related objects ke beech CONSISTENCY guarantee hoti hai (mismatch nahi hota)
# - Client concrete classes se LOOSELY COUPLED rehta hai
# - Naya "family" (jaise "Asia" region) add karna easy hai -- bas naya factory banao

# NUKSAAN (DISADVANTAGES):
# - Bahut saari classes/interfaces ban jaati hain
# - Chhote projects ke liye zaroorat se zyada complex ho sakta hai
# - Naya PRODUCT TYPE add karna (jaise "Engine") mushkil hai -- sabhi factories ke 
#   abstract interface ko modify karna padega
```

```python
# Pehle dekho: BINA Abstract Factory Pattern ke (Problem Code)
# - Chalo pehle ye dekhte hain ki Abstract Factory Pattern KE BINA log alag-alag 
#   "regions" ke liye cars kaise banate hain, aur usme kya dikkat aati hai.
```

```python
# BINA ABSTRACT FACTORY KE — PROBLEM DEMO
# - Yahan Client KHUD if-else se decide kar raha hai ki kaunsi car banani hai aur 
#   kaunsi specification. 
# - Dekho neeche kya dikkat hai.

class SedanBad:
    def assemble(self):
        print("Assembling Sedan car.")


class HatchbackBad:
    def assemble(self):
        print("Assembling Hatchback car.")


class NorthAmericaSpecBad:
    def display(self):
        print("North America Spec: Safety features compliant with local regulations.")


class EuropeSpecBad:
    def display(self):
        print("Europe Spec: Fuel efficiency and emissions compliant with EU standards.")


# Client KHUD region ke hisaab se if-else likh raha hai -- YE GALAT TAREEKA HAI
def build_car_bad(region):
    if region == "north_america":
        car = SedanBad()
        spec = NorthAmericaSpecBad()
    elif region == "europe":
        car = HatchbackBad()
        spec = EuropeSpecBad()
    else:
        raise ValueError("Unknown region")
    return car, spec


car, spec = build_car_bad("north_america")
car.assemble()
spec.display()

# Ab dekho -- kisi developer ne GALTI se MISMATCH kar diya:
mismatched_car = HatchbackBad()          # Europe ki car
mismatched_spec = NorthAmericaSpecBad()  # lekin North America ki spec laga di!
mismatched_car.assemble()
mismatched_spec.display()   # GALAT combination -- lekin code CRASH nahi hua!
```

```python
# ISME PROBLEM KYA THI?
# 1. CONSISTENCY GUARANTEE NAHI HAI: 
#    - Upar maine jaan-boojh kar "HatchbackBad" (Europe ki car) ke saath 
#      "NorthAmericaSpecBad" (North America ki spec) laga di -- aur code ne KOI ERROR
#      NAHI diya! 
#    - Ye SILENT BUG hai jo production mein bahut khatarnak ho sakta hai 
#      (jaise galat safety compliance laga dena).
#
# 2. IF-ELSE LOGIC HAR JAGAH PHAIL JAATI HAI: 
#    - Agar 10 jagah cars banani hain, to har jagah "if region == ..." 
#      likhna padega -- agar kal koi naya region add ho (jaise "asia"), 
#      to SAB JAGAH ye if-else DHOONDH KE update karna padega.
# 
# 3. TIGHT COUPLING: 
#    - Client function ko SedanBad, HatchbackBad, NorthAmericaSpecBad, 
#      EuropeSpecBad -- SABKO seedha jaanna padta hai. 
#    - Naya region add karne ke liye is function ko MODIFY karna padega 
#      (Open/Closed Principle violate hota hai).
#
# 4. HARD TO SCALE: 
#    - Jaise jaise regions aur product types badhte hain (car, spec, engine, tyres...),
#      if-else conditions ka combination EXPONENTIALLY complex ho jaata hai.
#
# YE SAB PROBLEMS "Abstract Factory Design Pattern" SOLVE KARTA HAI!
```

```python
# BEST APPROACH — Abstract Factory Pattern se solve karte hain
# - Har region ki apni Concrete Factory hogi jo related products (Car + Specification)
#   EK SAATH, GUARANTEED CONSISTENT banayegi.
```

```python
# ABSTRACT PRODUCTS (Car aur CarSpecification)
# - Ye COMMON INTERFACES hain jo har product family follow karegi. 
# - "Car" ka interface sabhi car types (Sedan, Hatchback) follow karenge, aur 
#   "CarSpecification" ka interface sabhi specification types follow karenge.

from abc import ABC, abstractmethod

class Car(ABC):
    @abstractmethod
    def assemble(self):
        pass


class CarSpecification(ABC):
    @abstractmethod
    def display(self):
        pass
```

```python
# CONCRETE PRODUCTS (Sedan, Hatchback, NorthAmericaSpecification, EuropeSpecification)
# - Ye ACTUAL classes hain jo Abstract Products ko implement karti hain.
# - Har class apna specific behavior define karti hai.

class Sedan(Car):
    def assemble(self):
        print("Assembling Sedan car.")


class Hatchback(Car):
    def assemble(self):
        print("Assembling Hatchback car.")


class NorthAmericaSpecification(CarSpecification):
    def display(self):
        print("North America Car Specification: Safety features compliant with local regulations.")


class EuropeSpecification(CarSpecification):
    def display(self):
        print("Europe Car Specification: Fuel efficiency and emissions compliant with EU standards.")
```

```python
# ABSTRACT FACTORY (Interface) - CarFactory
# - Ye "FACTORY OF FACTORIES" ka MAIN INTERFACE hai. Isme DO methods hain -- ek car 
#   banane ke liye, ek specification banane ke liye.
# - Har Concrete Factory in DONO methods ko implement karegi taaki RELATED products 
#   EK SAATH ban sakein.

class CarFactory(ABC):
    @abstractmethod
    def create_car(self):
        pass

    @abstractmethod
    def create_specification(self):
        pass
```

```python
# CONCRETE FACTORIES (NorthAmericaCarFactory aur EuropeCarFactory)
# - Har Concrete Factory EK REGION ke liye RELATED products (car + specification) 
#   banati hai. 
# - Isse GUARANTEE milta hai ki jo bhi factory use karoge, uski car aur spec HAMESHA 
#   MATCHING honge -- mismatch HO HI NAHI SAKTA!

class NorthAmericaCarFactory(CarFactory):
    def create_car(self):
        return Sedan()

    def create_specification(self):
        return NorthAmericaSpecification()


class EuropeCarFactory(CarFactory):
    def create_car(self):
        return Hatchback()

    def create_specification(self):
        return EuropeSpecification()
```

```python
# CLIENT CODE (Sirf Abstract Factory use karke cars banate hain)
# - Client ko SIRF itna pata hai ki "factory.create_car()" aur 
#   "factory.create_specification()" call karna hai. 
# - Use ye BILKUL pata nahi ki andar "Sedan" banega ya "Hatchback" -- aur MISMATCH
#   hona IMPOSSIBLE hai kyunki EK factory sirf APNE region ke products hi banati hai.
#
# - Naya region (jaise "Asia") add karna ho to bas ek "AsiaCarFactory" banao -- 
#   EXISTING code CHHEDNE ki zaroorat NAHI!

def build_car(factory: CarFactory):
    car = factory.create_car()
    spec = factory.create_specification()
    car.assemble()
    spec.display()
    print()


# North America ke liye
na_factory = NorthAmericaCarFactory()
build_car(na_factory)

# Europe ke liye
eu_factory = EuropeCarFactory()
build_car(eu_factory)
```

```python
# Problem Statement:
# - Ek UI Theme System banao using Abstract Factory Design Pattern 
#   (jaise document ke intro mein bataya gaya tha).
# - App Light Theme aur Dark Theme support karta hai. Har theme ke apne Button aur 
#   TextBox components hain, aur ek theme ke components DUSRI theme ke saath MIX nahi
#   hone chahiye (warna UI inconsistent dikhegi — jaise Light Button ke saath 
#   Dark TextBox).
#
# Tumhe ye banana hai (Car example jaisi hi structure use karo):
# 1. Abstract Products: 
#    - Button aur TextBox (dono abstract classes, render() method ho)
#
# 2. Concrete Products: 
#    - LightButton, LightTextBox (message: e.g. "Rendering Light Button ☀️") aur
#      DarkButton, DarkTextBox (message: e.g. "Rendering Dark Button 🌙")
#
# 3. Abstract Factory: 
#    - UIFactory — abstract methods create_button() aur create_textbox()
#
# 4. Concrete Factories: 
#    - LightThemeFactory aur DarkThemeFactory
# 
# 5. Client code likho jo render_ui(factory) function banaye — jo factory se button 
#    aur textbox banaye aur render kare. Dono themes ke liye test karo.
```

```python
# SOLUTION EXPLANATION
# - Yahan bilkul Car example jaisi hi structure use ki hai, bas ab "UI Components" bana
#   rahe hain instead of "Cars".
#   1. Button, TextBox                        -> Abstract Products
#   2. LightButton, LightTextBox,
#      DarkButton, DarkTextBox                -> Concrete Products
#   3. UIFactory                                -> Abstract Factory
#   4. LightThemeFactory, DarkThemeFactory      -> Concrete Factories
#   5. Neeche client code hai jo sab kuch jodta hai

from abc import ABC, abstractmethod

# ---- 1. Abstract Products ----
class Button(ABC):
    @abstractmethod
    def render(self):
        pass


class TextBox(ABC):
    @abstractmethod
    def render(self):
        pass


# ---- 2. Concrete Products (Light Theme) ----
class LightButton(Button):
    def render(self):
        print("Rendering Light Button ☀️")


class LightTextBox(TextBox):
    def render(self):
        print("Rendering Light TextBox ☀️")


# ---- 2. Concrete Products (Dark Theme) ----
class DarkButton(Button):
    def render(self):
        print("Rendering Dark Button 🌙")


class DarkTextBox(TextBox):
    def render(self):
        print("Rendering Dark TextBox 🌙")


# ---- 3. Abstract Factory ----
class UIFactory(ABC):
    @abstractmethod
    def create_button(self):
        pass

    @abstractmethod
    def create_textbox(self):
        pass


# ---- 4. Concrete Factories ----
class LightThemeFactory(UIFactory):
    def create_button(self):
        return LightButton()

    def create_textbox(self):
        return LightTextBox()


class DarkThemeFactory(UIFactory):
    def create_button(self):
        return DarkButton()

    def create_textbox(self):
        return DarkTextBox()


# ---- 5. Client Code ----
def render_ui(factory: UIFactory):
    button = factory.create_button()
    textbox = factory.create_textbox()
    button.render()
    textbox.render()
    print()


print("---- Light Theme ----")
render_ui(LightThemeFactory())

print("---- Dark Theme ----")
render_ui(DarkThemeFactory())
```

```python
# Abstract Factory Pattern — Poore "World Theme" ke RELATED objects

# Extension of Factory Method:
# - Factory Method ek EK product (sirf Enemy) banata hai. 
# - Ab GameForge Studio mein game MULTIPLE WORLD THEMES support karta hai — 
#   "Forest World" aur "Ice World". 
# - Har world ka apna MATCHING Enemy + MATCHING Weapon hona chahiye — Forest ka
#   Goblin, Ice World ki Frost Sword ke saath MIX NAHI hona chahiye (inconsistency bug)!

# Solution: 
# - Abstract Factory se RELATED PRODUCTS KA GROUP (Enemy + Weapon), EK SAATH, 
#   GUARANTEED CONSISTENT banate hain.
```

```python
# ABSTRACT FACTORY PATTERN — World Theme System
# - Abstract Products  -> Enemy (already bana hai), Weapon (naya)
# - Concrete Products  -> Goblin/Dragon (Enemy), + FrostSword/FireSword (Weapon)
# - Abstract Factory   -> WorldFactory (create_enemy() + create_weapon())
# - Concrete Factories -> ForestWorldFactory, IceWorldFactory

# ---- Abstract Product (naya) ----
class Weapon(ABC):
    @abstractmethod
    def show(self):
        pass


# ---- Concrete Products (naye) ----
class WoodenClub(Weapon):
    def show(self):
        print("🪵 Wooden Club (Forest World ka weapon)")


class FrostBlade(Weapon):
    def show(self):
        print("❄️ Frost Blade (Ice World ka weapon)")


# ---- Concrete Products (Enemy ke liye, Ice World ka naya enemy) ----
class IceGolem(Enemy):
    def attack(self):
        print("🧊 Ice Golem smashes with frozen fists!")


# ---- Abstract Factory ----
class WorldFactory(ABC):
    @abstractmethod
    def create_enemy(self) -> Enemy:
        pass

    @abstractmethod
    def create_weapon(self) -> Weapon:
        pass


# ---- Concrete Factories ----
class ForestWorldFactory(WorldFactory):
    def create_enemy(self):
        return Goblin()          # Chapter 3 ka Goblin REUSE hua!

    def create_weapon(self):
        return WoodenClub()


class IceWorldFactory(WorldFactory):
    def create_enemy(self):
        return IceGolem()

    def create_weapon(self):
        return FrostBlade()
```

```python
# TEST: Har world ka CONSISTENT set milta hai (mismatch IMPOSSIBLE hai)

def load_world(factory: WorldFactory):
    enemy = factory.create_enemy()
    weapon = factory.create_weapon()
    enemy.attack()
    weapon.show()
    print()


print("--- Forest World load ho raha hai ---")
load_world(ForestWorldFactory())

print("--- Ice World load ho raha hai ---")
load_world(IceWorldFactory())

print("✅ Abstract Factory se GUARANTEE mila ki Forest ka Goblin sirf Wooden Club ke saath aayega, IceGolem sirf Frost Blade ke saath -- mismatch ho hi NAHI sakta.")
```

```python
# Chapter 5 - Singleton Pattern
# - Singleton ek CREATIONAL design pattern hai jo guarantee karta hai ki
#   kisi class ka SIRF EK HI OBJECT (instance) poore program mein bane —
#   aur us ek object ko access karne ka ek GLOBAL POINT bhi milta hai.

# SOCHO EK EXAMPLE:
# - Tumhare app mein ek "Database Connection Manager" hai. Agar har baar
#   naya object banoge to har baar naya database connection khulega —
#   ye resources (memory, network) waste karega aur bugs bhi la sakta hai
#   (jaise do alag connections mein data mismatch).

# Solution:
# - Singleton isko solve karta hai:
#   - Pehli baar object banta hai
#   - Agli baar koi bhi "naya object mango" bole, to SAME purana
#     object hi wapas milta hai (naya nahi banta)

# KAISE KAAM KARTA HAI (MECHANISM):
# 1. Constructor ko PRIVATE (ya restricted) bana dete hain, taaki
#    bahar se koi seedha "naya object" na bana sake.
# 2. Class ke andar hi ek STATIC variable rakhte hain jisme instance
#    store hoga.
# 3. Ek STATIC method (jaise get_instance()) banate hain jo check
#    karta hai:
#    - Agar instance already bana hai -> wahi return karo
#    - Agar nahi bana -> naya banao, store karo, phir return karo

# REAL LIFE EXAMPLE:
# - Database Connection Pool / Manager
# - Logger class (poore app mein ek hi Logger object use hota hai)
# - Configuration Manager (app settings ek hi jagah se load ho)
# - Thread Pool Manager
# - Cache Manager

# FAYDE (ADVANTAGES):
# - Memory aur resources save hote hain (ek hi object baar baar use hota hai)
# - Global access point milta hai (kahin se bhi same object mil jaata hai)
# - Consistency banti hai (sab log same data/state use karte hain)

# NUKSAAN (DISADVANTAGES):
# - Global state hone se testing mushkil ho sakti hai
# - Multi-threading mein agar sambhal ke na banaya jaaye to bug aa
#   sakta hai (do threads ek saath do alag instance bana sakte hain)
# - Overuse karne se code tightly coupled ho jaata hai (dependency
#   chhupi rehti hai, jo debugging mushkil banata hai)
```

```python
# 1. Sabse Simple Singleton — "__new__" method use karke:
#    Python mein jab bhi hum "ClassName()" likhte hain, andar do steps hote hain:
#    1. __new__()  -> naya object banata hai (memory allocate karta hai)
#    2. __init__() -> us object ko initialize karta hai (values set karta hai)
#
#   Hum __new__() ko OVERRIDE karke check karte hain:
#   - Agar class variable "_instance" mein pehle se object hai -> wahi do
#   - Agar nahi hai -> naya banao, store karo, phir do
#
# Isse chahe kitni baar bhi "Logger()" call karo, HAMESHA same object milega.
```

```python
class Logger:
    _instance = None   # class-level variable jisme single instance store hoga

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            print("Naya Logger object ban raha hai...")
            cls._instance = super().__new__(cls)
        else:
            print("Purana Logger object hi wapas mil raha hai...")
        return cls._instance

    def __init__(self):
        # Note: __init__ HAR BAAR call hota hai (chahe naya object bane ya na bane)
        # Isliye important cheezein sirf pehli baar set karo agar zaroorat ho
        self.logs = getattr(self, "logs", [])

    def log(self, message):
        self.logs.append(message)
        print(f"LOG: {message}")
```

```python
# TEST KARTE HAIN — Kya dono object SAME hain?

logger1 = Logger()
logger1.log("Pehla message")

logger2 = Logger()
logger2.log("Doosra message")

print("\nlogger1 id:", id(logger1))
print("logger2 id:", id(logger2))
print("Kya dono same object hain?", logger1 is logger2)

print("\nlogger1 ke saare logs (kyunki dono ek hi object hain):")
print(logger1.logs)
```

```python
# 2. Thread-Safe Singleton - Multiple threads mein safe kaise banaye

# Problem Statement: 
# - Agar do THREADS EK SAATH pehli baar "get_instance()" call karein, 
#   to ho sakta hai dono ko lage "instance abhi nahi bana" aur
#   DONO apna apna naya object bana lein — isse Singleton TOOT jaata hai!

# Solution: 
# - Ek LOCK use karte hain (threading.Lock). Lock ka matlab hai
#   "ek waqt mein sirf EK thread hi ye code chala sakta hai". 
# - Doosre threads ko wait karna padta hai jab tak pehla thread lock
#   release na kare.

# Isse guarantee milta hai ki instance sirf EK BAAR hi banega, chahe
# kitne bhi threads ek saath try karein.
```

```python
import threading

class DatabaseConnection:
    _instance = None
    _lock = threading.Lock()   # lock banaya taaki ek waqt mein ek hi thread andar jaaye

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            with cls._lock:                # lock lo
                if cls._instance is None:  # double-check (kisi aur thread ne pehle na bana diya ho)
                    print("DatabaseConnection ka naya object ban raha hai...")
                    cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self):
        self.connection_id = getattr(self, "connection_id", "DB-CONN-001")

    def query(self, sql):
        print(f"[{self.connection_id}] Running query: {sql}")
```

```python
# TEST: Multiple threads se try karte hain

results = []

def create_connection():
    conn = DatabaseConnection()
    results.append(id(conn))

threads = [threading.Thread(target=create_connection) for _ in range(5)]

for t in threads:
    t.start()
for t in threads:
    t.join()

print("\nSaare threads ke object ids:", results)
print("Kya sab same hain?", len(set(results)) == 1)
```

```python
# Problem Statement:
# - Ek Printer Spooler System banao using Singleton Design Pattern.
# - Real life mein agar ek office mein ek hi printer hai, to sabhi employees
#   ke print jobs ek hi "queue" (spooler) mein jaani chahiye — do alag
#   spooler objects nahi hone chahiye, warna print jobs kho sakti hain
#   ya conflict ho sakta hai.

# Tumhe ye banana hai:
# 1. PrinterSpooler class banao jo Singleton ho 
#    (koi bhi tareeka use kar sakte ho — __new__, decorator, etc.)
# 2. Isme ek list print_queue honi chahiye jisme jobs store hon
# 3. add_job(document_name) method banao — jo document ka naam queue mein add kare
# 4. print_all() method banao — jo queue ke saare documents print kare 
#    (print statement se) aur phir queue khaali kar de
# 5. Client code likho jaha tum do alag "spooler" variables banao 
#    (spooler1 = PrinterSpooler() aur spooler2 = PrinterSpooler()), dono se alag alag 
#    jobs add karo, aur verify karo ki dono variables actually SAME object hain 
#    (spooler1 is spooler2)
```

```python
# SOLUTION EXPLANATION
# Yahan __new__ wala classic tareeka use kiya hai (jaise Logger example mein tha), 
# taaki concept clear rahe.
# 1. PrinterSpooler class ke andar _instance variable hai
# 2. __new__ mein check karte hain ki instance pehle se bana hai ya nahi
# 3. __init__ mein print_queue list sirf pehli baar initialize karte hain
# 4. add_job() aur print_all() normal instance methods hain

class PrinterSpooler:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            print("PrinterSpooler ka naya object ban raha hai...")
            cls._instance = super().__new__(cls)
        else:
            print("PrinterSpooler ka purana object mil raha hai...")
        return cls._instance

    def __init__(self):
        # getattr trick: agar pehle se queue bani hai to usko mat todo
        if not hasattr(self, "print_queue"):
            self.print_queue = []

    def add_job(self, document_name):
        self.print_queue.append(document_name)
        print(f"Job added: {document_name}")

    def print_all(self):
        print("\n--- Printing All Jobs ---")
        for doc in self.print_queue:
            print(f"Printing: {doc}")
        self.print_queue = []   # queue khaali kar do
        print("--- Saari jobs print ho gayi, queue khaali hai ---\n")


# ---- Client Code ----
spooler1 = PrinterSpooler()
spooler1.add_job("Resume.pdf")
spooler1.add_job("Invoice.docx")

spooler2 = PrinterSpooler()
spooler2.add_job("Report.xlsx")

print("\nspooler1 is spooler2:", spooler1 is spooler2)
print("spooler2 ki queue (dono jobs dikhengi kyunki same object hai):", spooler2.print_queue)

spooler2.print_all()
```

```python
# Chapter 5: Singleton Pattern — Ek Central `GameManager`
# Extension of everything so far: 
# - Abhi tak humne Players, NPCs, Enemies, Weapons  banaye — lekin in SAB ko TRACK kaun
#   karega? Score, current level, current world theme — ye sab
#   GLOBAL STATE hai. 
# - Agar `GameManager` ki MULTIPLE copies ban gayin (galti se), to score ALAG ALAG jagah
#  track hoga aur INCONSISTENT ho jaayega.

# Solution: 
# - Singleton Pattern se guarantee karte hain ki poore game mein `GameManager` ka
#   SIRF EK HI instance ho.
```

```python
# SINGLETON PATTERN — GameManager
# - __new__ method OVERRIDE karke guarantee karte hain ki GameManager() jitni baar bhi 
#   call ho, SAME object mile.

class GameManager:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            print("🎮 GameManager ka PEHLA (aur AKHRI) instance ban raha hai...")
            cls._instance = super().__new__(cls)
            cls._instance.score = 0
            cls._instance.current_world = None
        return cls._instance

    def add_score(self, points):
        self.score += points

    def set_world(self, world_name):
        self.current_world = world_name

    def status(self):
        print(f"📊 GameManager Status -> Score: {self.score}, World: {self.current_world}")
```

```python
# TEST: Chahe kitni baar bhi GameManager() bulao, SAME object milega

gm1 = GameManager()
gm1.set_world("Forest World")
gm1.add_score(50)

gm2 = GameManager()   # naya banane ki koshish -- lekin PURANA hi milega
gm2.add_score(30)

print("gm1 is gm2:", gm1 is gm2)
gm1.status()   # dono ka score EK SAATH update hua, kyunki ye SAME object hai

print("\n✅ Singleton Pattern se poore game mein score/world EK HI JAGAH SE track ho raha hai -- koi confusion nahi.")
```

```python
# Chapter 6: Object Pool Pattern — Reusable Bullets
# - Extension of Singleton: Singleton ne humein bataya "SIRF EK object rakho aur SHARE 
#   karo".
# - Lekin ab Player ki `Bullet` fire karni hai — game mein HAZAARON bullets fire hoti hain,
#   aur Singleton yahan kaam NAHI karega (hamein EK NAHI, MULTIPLE REUSABLE bullets 
#   chahiye)! Agar har bullet ke liye naya object banao aur turant destroy karo, to 
#   Garbage Collector par load badh jaayega aur game LAG karega.

# Solution: Object Pool se LIMITED, REUSABLE bullets ka POOL maintain karte hain.
```

```python
# OBJECT POOL PATTERN — Bullet Pool
# - Bullets ko baar baar create/destroy karne ke bajaye, EK POOL maintain karte hain jahan
#   se bullets BORROW aur RETURN hote hain.

class Bullet:
    _counter = 0

    def __init__(self):
        Bullet._counter += 1
        self.id = Bullet._counter
        print(f"   🔫 Bullet #{self.id} banaya gaya")

    def fire(self, target):
        print(f"💥 Bullet #{self.id} fired at {target}!")


class BulletPool:
    def __init__(self, max_size=3):
        self.max_size = max_size
        self.available = []
        self.in_use = []

    def borrow_bullet(self):
        if self.available:
            bullet = self.available.pop()
        elif len(self.in_use) < self.max_size:
            bullet = Bullet()
        else:
            raise Exception("⚠️ Bullet pool full hai! Wait karo.")
        self.in_use.append(bullet)
        return bullet

    def return_bullet(self, bullet):
        self.in_use.remove(bullet)
        self.available.append(bullet)
```

```python
# TEST: 5 baar fire karte hain lekin sirf FEW bullets ACTUALLY bante hain

bullet_pool = BulletPool(max_size=2)

for shot in range(1, 6):
    print(f"\nShot #{shot}:")
    bullet = bullet_pool.borrow_bullet()
    bullet.fire("Goblin")
    bullet_pool.return_bullet(bullet)   # fire hote hi wapas pool mein

print(f"\nTotal Bullet objects banaye gaye: {Bullet._counter} (5 shots fire hue, lekin max_size=2 hi bane!)")

print("\n✅ Object Pool Pattern se hazaaron bullets fire hone par bhi sirf LIMITED objects banenge -- performance behtar.")
```

```python
# Chapter 7: Lazy Loading Pattern — Textures sirf zaroorat par load karo
# Extension of Abstract Factory (World Themes): 
# - Chapter 4 mein humne `ForestWorldFactory` aur `IceWorldFactory` banayi. 
# - Lekin har world ke saath HIGH-RESOLUTION TEXTURES judi hoti hain 
#   (bahut MEMORY leti hain). 
# - Agar game START hote hi SAARE worlds ki textures load kar lein, to game bahut SLOW 
#   start hogi — jabki player shayad SIRF Forest World hi khelega!

# Solution: Lazy Loading se textures SIRF TAB load karte hain jab player us WORLD mein
# ACTUALLY ENTER kare.
```

```python
# LAZY LOADING PATTERN (Virtual Proxy) — World Textures
# - WorldTextureProxy ek PLACEHOLDER hai -- real (heavy) textures SIRF PEHLI BAAR 
#  "get_textures()" call hone par load hoti hain.

import time

class WorldTextures:
    def __init__(self, world_name):
        print(f"⏳ '{world_name}' ki HIGH-RES textures load ho rahi hain (EXPENSIVE)...")
        time.sleep(0.4)
        self.world_name = world_name
        self.textures = [f"{world_name}_ground.png", f"{world_name}_sky.png", f"{world_name}_trees.png"]

    def get_textures(self):
        return self.textures


class WorldTextureProxy:
    def __init__(self, world_name):
        self.world_name = world_name
        self._real_textures = None   # abhi tak LOAD NAHI hui

    def get_textures(self):
        if self._real_textures is None:
            print(f"♻️  Player pehli baar '{self.world_name}' mein ENTER kar raha hai -- ab load hogi")
            self._real_textures = WorldTextures(self.world_name)
        else:
            print(f"✅ '{self.world_name}' ki textures pehle se load hain -- REUSE ho rahi hain")
        return self._real_textures.get_textures()
```

```python
# TEST: Game start hota hai, lekin textures TABHI load hoti hain jab player world mein 
#       jaata hai

forest_textures = WorldTextureProxy("Forest World")
ice_textures = WorldTextureProxy("Ice World")

print("🎮 Game START ho gaya!")
print("(Abhi tak KISI world ki textures load NAHI hui -- fast startup!)\n")

print("Player 'Forest World' mein enter karta hai:")
print(forest_textures.get_textures())

print("\nPlayer WAPAS 'Forest World' mein aata hai:")
print(forest_textures.get_textures())   # ab REUSE hoga, load nahi

print("\n(Player kabhi 'Ice World' mein gaya hi nahi -- uski textures AAJ TAK load NAHI hui!)")

print("\n✅ Lazy Loading se GameForge Studio ka STARTUP FAST raha, aur sirf ACTUALLY USED worlds ki textures load hui.")
```

## 📋 Summary — Kaunsa Pattern Kab Use Karo?

| # | Pattern | Kis Sawaal Ka Jawab Deta Hai | GameForge Example |
|---|---------|-------------------------------|---------------------|
| 1 | **Builder** | Complex object step-by-step kaise banau? | `Player` character banaya |
| 2 | **Prototype** | Existing object se naya FAST kaise banau? | `hero` ko clone karke NPCs |
| 3 | **Factory Method** | Runtime par TYPE decide karna hai? | Level ke hisaab se Goblin/Dragon |
| 4 | **Abstract Factory** | Related objects ka GROUP consistent chahiye? | Forest World ka Enemy+Weapon |
| 5 | **Singleton** | Sirf EK instance guarantee chahiye? | `GameManager` |
| 6 | **Object Pool** | Expensive objects REUSE karne hain? | `Bullet` pool |
| 7 | **Lazy Loading** | Object sirf zaroorat par banana hai? | World textures |
