## ❌ Pehle dekho: BINA Facade Pattern ke (Problem Code)

Chalo pehle ye dekhte hain ki Facade Pattern KE BINA log ek Hotel ke Restaurant System se kaise
interact karte hain (jaise document ke Hotel/Restaurant example mein bataya gaya), aur usme kya
dikkat aati hai.

```python
"""
BINA FACADE KE — PROBLEM DEMO
------------------------------------
Yahan Client (customer) ko SEEDHA HAR restaurant SUBSYSTEM CLASS
ke baare mein pata hona padta hai, aur unhe MANUALLY, EK EK karke
call karna padta hai.
"""

class VegMenuBad:
    def __str__(self):
        return "Veg Menu: Paneer Tikka, Dal Makhani, Naan"


class NonVegMenuBad:
    def __str__(self):
        return "Non-Veg Menu: Butter Chicken, Mutton Curry"


class BothMenuBad:
    def __str__(self):
        return "Mixed Menu: Veg + Non-Veg combo platters"


class VegRestaurantBad:
    def get_menu(self):
        return VegMenuBad()


class NonVegRestaurantBad:
    def get_menu(self):
        return NonVegMenuBad()


class VegNonBothRestaurantBad:
    def get_menu(self):
        return BothMenuBad()


# Client ko SEEDHA teeno restaurant classes ke baare mein PATA hona padta hai!
print("Customer ko VEG menu chahiye:")
veg_restaurant = VegRestaurantBad()
print(veg_restaurant.get_menu())

print("\nCustomer ko NON-VEG menu chahiye:")
non_veg_restaurant = NonVegRestaurantBad()
print(non_veg_restaurant.get_menu())

print("\nCustomer ko MIXED menu chahiye:")
both_restaurant = VegNonBothRestaurantBad()
print(both_restaurant.get_menu())
```

```python
"""
ISME PROBLEM KYA THI?
-------------------------
1. CLIENT KO SUBSYSTEM KI DETAILS PATA HONI PADTI HAIN: Customer
   (client) ko YE PATA hona chahiye ki "VegRestaurantBad",
   "NonVegRestaurantBad", "VegNonBothRestaurantBad" -- TEENO
   ALAG ALAG classes hain, aur KIS KAAM ke liye KAUNSI class
   INSTANTIATE karni hai. REAL LIFE mein customer sirf "मुझे veg
   khana hai" bolta hai, use restaurant ki INTERNAL classes se
   MATLAB NAHI hota!

2. TIGHT COUPLING: Client code SEEDHA teeno restaurant classes
   IMPORT aur USE kar raha hai. Agar kal koi RESTAURANT class
   ka naam ya LOGIC badle, to Client code mein BHI CHANGES
   karne padenge.

3. REPEATED, SCATTERED LOGIC: Agar 10 jagah customer se "kaunsa
   menu chahiye" poochna ho, to HAR JAGAH ye SAME "if veg then
   VegRestaurantBad(), elif non-veg then ..." jaisi LOGIC
   REPEAT karni padegi.

4. NO SINGLE ENTRY POINT: Poore "Hotel System" ka koi EK,
   UNIFIED tareeka NAHI hai INTERACT karne ka -- client ko
   HAR BAAR SUBSYSTEM ke ANDAR JHAANKNA padta hai.

YE SAB PROBLEMS "Facade Design Pattern" SOLVE KARTA HAI!
Neeche BEST APPROACH dekhte hain -- ek "HotelKeeper" (Facade)
banayenge jo customer ke liye SAB KUCH COORDINATE karega.
"""
print("Chalo ab BEST APPROACH (Facade Pattern) dekhte hain...")
```

## ✅ Ab dekho: BEST APPROACH — Facade Pattern se solve karte hain

### 1️⃣ Subsystem Classes — Menus aur Restaurants

```python
"""
SUBSYSTEM CLASSES (Menus + Restaurants)
--------------------------------------------
Ye COMPLEX SUBSYSTEM hai -- multiple restaurant classes, jinhe
Facade (HotelKeeper) COORDINATE karegi. Inhe FACADE ka BILKUL
PATA NAHI hai -- ye BAS apna KAAM karti hain.
"""

class VegMenu:
    def __str__(self):
        return "Veg Menu: Paneer Tikka, Dal Makhani, Naan"


class NonVegMenu:
    def __str__(self):
        return "Non-Veg Menu: Butter Chicken, Mutton Curry"


class Both:
    def __str__(self):
        return "Mixed Menu: Veg + Non-Veg combo platters"


class VegRestaurant:
    def get_menus(self):
        return VegMenu()


class NonVegRestaurant:
    def get_menus(self):
        return NonVegMenu()


class VegNonBothRestaurant:
    def get_menus(self):
        return Both()
```

### 2️⃣ Facade Interface — `HotelKeeper` (abstract)

```python
"""
FACADE INTERFACE (HotelKeeper)
------------------------------------
Ye HIGH-LEVEL interface hai jo CLIENT ko expose hoga. Client ko
SIRF ye 3 SIMPLE methods pata hone chahiye -- ANDAR ki
complexity ka MATLAB NAHI.
"""

from abc import ABC, abstractmethod

class HotelKeeper(ABC):
    @abstractmethod
    def get_veg_menu(self):
        pass

    @abstractmethod
    def get_non_veg_menu(self):
        pass

    @abstractmethod
    def get_veg_non_menu(self):
        pass
```

### 3️⃣ Facade Implementation — `HotelKeeperImplementation`

```python
"""
FACADE IMPLEMENTATION (HotelKeeperImplementation)
--------------------------------------------------------
Ye class HotelKeeper interface IMPLEMENT karti hai. ANDAR HI
ANDAR ye teeno Restaurant SUBSYSTEM classes ko INSTANTIATE
karke, unse Menu MAANGTI hai, aur CLIENT ko RETURN kar deti hai.

CLIENT ko YE PATA HI NAHI CHALTA ki ANDAR "VegRestaurant",
"NonVegRestaurant" jaisi classes maujood hain!
"""

class HotelKeeperImplementation(HotelKeeper):
    def get_veg_menu(self):
        v = VegRestaurant()
        return v.get_menus()

    def get_non_veg_menu(self):
        nv = NonVegRestaurant()
        return nv.get_menus()

    def get_veg_non_menu(self):
        b = VegNonBothRestaurant()
        return b.get_menus()
```

### 4️⃣ Client — Sirf Facade (`HotelKeeper`) se baat karta hai

```python
"""
CLIENT CODE
---------------
Ab Customer (client) ko SIRF "HotelKeeperImplementation" se baat
karni hai -- SIMPLE, HIGH-LEVEL methods. Use SUBSYSTEM ke ANDAR
kya ho raha hai, ISKA BILKUL PATA NAHI.
"""

keeper = HotelKeeperImplementation()

print("Customer keeper se VEG menu maangta hai:")
veg_menu = keeper.get_veg_menu()
print(veg_menu)

print("\nCustomer keeper se NON-VEG menu maangta hai:")
non_veg_menu = keeper.get_non_veg_menu()
print(non_veg_menu)

print("\nCustomer keeper se MIXED menu maangta hai:")
both_menu = keeper.get_veg_non_menu()
print(both_menu)

print("\n✅ Facade Pattern se Customer ko SIRF 'HotelKeeper' se baat karni padi -- 3 alag Restaurant classes ki DETAILS bilkul CHHUPI rahin.")
```

## ✅ Solution — Practice Question ka Coding Solution

```python
"""
SOLUTION EXPLANATION
-----------------------
Yahan bilkul HotelKeeperImplementation jaisi hi structure use ki
hai, bas ab "Home Theater Devices" bana rahe hain instead of
"Restaurants".

1. Projector, SoundSystem, StreamingPlayer -> Subsystem Classes
2. HomeTheaterFacade                          -> Facade
3. Neeche client code hai jo sirf EK call se movie start karta hai
"""

# ---- 1. Subsystem Classes ----
class Projector:
    def turn_on(self):
        print("📽️ Projector ON ho gaya.")


class SoundSystem:
    def set_volume(self, level):
        print(f"🔊 Sound System volume set kiya gaya: {level}")


class StreamingPlayer:
    def play(self, movie_name):
        print(f"▶️ '{movie_name}' PLAY ho raha hai...")


# ---- 2. Facade ----
class HomeTheaterFacade:
    def __init__(self):
        self.projector = Projector()
        self.sound_system = SoundSystem()
        self.streaming_player = StreamingPlayer()

    def watch_movie(self, movie_name):
        print(f"🎬 '{movie_name}' dekhne ke liye SETUP ho raha hai...\n")
        self.projector.turn_on()
        self.sound_system.set_volume(15)
        self.streaming_player.play(movie_name)
        print("\n✅ Enjoy your movie!")


# ---- 3. Client Code ----
home_theater = HomeTheaterFacade()
home_theater.watch_movie("Inception")
```
