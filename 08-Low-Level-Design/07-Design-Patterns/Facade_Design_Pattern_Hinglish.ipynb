# 🏨 Facade Design Pattern — Hinglish Beginner Guide

**Facade Design Pattern** ek **structural design pattern** hai jo ek **complex subsystem** ke 
upar ek **simple aur unified interface** provide karta hai. Ye system ki internal complexity 
ko **hide** kar deta hai, jisse use karna aur maintain karna aasan ho jata hai.

> **Simple bhasha mein:** Socho tum ek **hotel** mein khana order karna chahte ho. Hotel mein 
> alag-alag restaurants hain — Veg, Non-Veg, aur Mixed. Tumhe har restaurant ka menu, staff, 
> kitchen process — kuch bhi directly nahi jaanna. Tum sirf **Hotel Keeper (Manager)** se baat 
> karte ho — "Mujhe Veg menu chahiye" — aur wo andar hi andar sahi restaurant se menu la deta hai.
> Yahi **Facade Pattern** ka core idea hai — **"Ek simple counter, complexity peeche chhupi hui."**

## 🏠 Real-Life Analogy (Samjhne ke liye)

**Home Automation System** — sabse common example:

Ghar mein alag-alag devices hain — **Lights**, **AC**, **Security Cameras**, **Music System**. 
Agar tumhe "Movie Night Mode" on karna ho, to manually:
- Lights dim karni padengi
- AC 22°C pe set karna padega
- Cameras ko "night mode" pe daalna padega
- Music system pe playlist start karni padegi

Har device ka apna alag remote/app hoga — ye **bahut complex** ho jayega!

**Facade Pattern isko solve karta hai** — ek single **"Home Control Panel"** (Facade) banate hain 
jisme bas ek button hota hai: **"Movie Night Mode"**. Ye button click karte hi Facade **andar hi 
andar** saare devices (Lights, AC, Camera, Music) ko coordinate kar deta hai. User ko in devices 
ke internal working se koi matlab nahi!

**Doosre real-world examples:**
- 🏨 Hotel Keeper (Veg/Non-Veg/Mixed restaurants ko coordinate karta hai)
- 🎥 Video Streaming Platforms (encoding + buffering + playback ek simple "Play" button ke peeche)
- 🏦 Banking Systems (balance check/transfer ek simple API ke peeche, complex backend chhupa hota hai)
- 💻 Compiler (Scanner + Parser + BytecodeStream ko ek simple `compile()` method ke peeche)


## ❌ Step 1: Problem — BINA Facade Pattern ke (Bad Approach)

Chalo pehle dekhte hain agar **Client (customer)** khud har restaurant ke saath **directly** 
interact kare, to kya dikkat aati hai.


```python
"""
BAD APPROACH — Client seedha har Subsystem class ke saath directly deal karta hai
========================================================================================
Idea: Customer khud VegRestaurant, NonVegRestaurant, VegNonBothRestaurant
classes ko jaanta hai aur unse SEEDHA interact karta hai.
"""

class VegMenu:
    def __str__(self):
        return "Veg Menu: Paneer Tikka, Dal Makhani, Roti"

class NonVegMenu:
    def __str__(self):
        return "Non-Veg Menu: Butter Chicken, Fish Curry"

class BothMenu:
    def __str__(self):
        return "Mixed Menu: Veg + Non-Veg combo"


class VegRestaurant:
    def get_menus(self):
        # yaha bahut sara internal setup ho sakta hai (kitchen check, stock check, etc.)
        return VegMenu()

class NonVegRestaurant:
    def get_menus(self):
        return NonVegMenu()

class VegNonBothRestaurant:
    def get_menus(self):
        return BothMenu()


# ❌ Client (Customer) ko HAR restaurant class ka naam aur method pata hona chahiye!
def customer_orders_food(choice):
    if choice == "veg":
        restaurant = VegRestaurant()
        menu = restaurant.get_menus()
    elif choice == "non-veg":
        restaurant = NonVegRestaurant()
        menu = restaurant.get_menus()
    elif choice == "mixed":
        restaurant = VegNonBothRestaurant()
        menu = restaurant.get_menus()
    else:
        return "Invalid choice!"
    return menu


print(customer_orders_food("veg"))
print(customer_orders_food("non-veg"))

"""
❌ PROBLEMS WITH THIS BAD APPROACH:

1. CLIENT KO SUBSYSTEM KI DEEP KNOWLEDGE CHAHIYE: Customer ko pata hona
   chahiye ki VegRestaurant, NonVegRestaurant, VegNonBothRestaurant classes
   EXIST karti hain, aur unka method get_menus() hai -- ye bahut zyada
   "internal knowledge" hai ek simple customer ke liye.

2. TIGHT COUPLING: customer_orders_food() function directly saari
   restaurant classes pe depend karta hai. Agar kal ek naya restaurant
   type add ho (jaise "Vegan"), to is function ko EDIT karna padega.

3. CODE DUPLICATION: Har jagah jaha customer order karna chahta hai
   (website, mobile app, kiosk), wahi if-else logic REPEAT karna padega.

4. NO SINGLE ENTRY POINT: Complex real-world subsystems (jaise 10-15
   restaurants, alag-alag services) mein ye if-else chain BAHUT bada
   aur unmanageable ho jayega.

LESSON: Humein ek "Manager/Receptionist" jaisa object chahiye jo saari
complexity apne andar sambhal le, aur client ko sirf EK simple interface de.
Yahi Facade Pattern solve karta hai! Chalo best approach dekhte hain 👇
"""
```

```python
"""
TEXT-BASED UML CLASS DIAGRAM — Facade Design Pattern (BEST APPROACH)
================================================================================

                          +----------------------------+
                          |          Client                 |
                          |   (Customer / Main program)      |
                          +----------------------------+
                                         |
                                         | uses ONLY the facade
                                         v
                          +----------------------------+
                          |         HotelKeeper              |
                          |     (Facade Interface - ABC)      |
                          +----------------------------+
                          | + get_veg_menu()  (abstract)       |
                          | + get_non_veg_menu()  (abstract)    |
                          | + get_veg_non_menu()  (abstract)     |
                          +--------------^---------------+
                                         |
                                         | implements
                                         |
                          +--------------+---------------+
                          |    HotelKeeperImplementation      |
                          |          (Concrete Facade)          |
                          +----------------------------+
                          | + get_veg_menu()                       |
                          | + get_non_veg_menu()                    |
                          | + get_veg_non_menu()                     |
                          +----+----------+----------+---------+
                               |          |          |
              coordinates ---> |          |          | <--- coordinates
                     +---------+   +------+----+   +-+-----------------+
                     |             |           |                       |
           +---------+----+ +------+-----+ +---+----------------+
           | VegRestaurant   | | NonVegRestaurant | | VegNonBothRestaurant |
           | (Subsystem 1)    | | (Subsystem 2)    | |    (Subsystem 3)      |
           +-----------------+ +------------------+ +------------------------+
           | + get_menus()      | | + get_menus()      | | + get_menus()            |
           |   -> VegMenu()        | |   -> NonVegMenu()     | |   -> Both()                |
           +-----------------+ +------------------+ +------------------------+

    NOTE: Subsystem classes (VegRestaurant, etc.) DON'T know that HotelKeeper
    exists -- they have NO reference back to the Facade. Facade knows about
    THEM, not the other way around (one-directional dependency).


FLOW DIAGRAM — Client kaise Facade use karta hai
=========================================================

    Client              HotelKeeper (Facade)         VegRestaurant
      |                       |                            |
      |--- get_veg_menu() --->|                            |
      |                       |--- new VegRestaurant() --->|
      |                       |--- restaurant.get_menus() ->|
      |                       |<------- VegMenu() -----------|
      |<---- VegMenu() -------|                            |
      |                       |                            |

    Client ko sirf "get_veg_menu()" call karna hai -- usko VegRestaurant
    class ka NAAM tak pata nahi hai! Saari complexity Facade ke andar chhupi hai.
"""

print("UML aur flow diagram upar comment mein hai — dhyan se padho! 👆")
```

## 🧩 3 Main Components (Yaad rakhne wali cheez)

| Component | Kaam kya hai | Analogy (Hotel example) |
|---|---|---|
| **Facade** | Simple, unified interface jo Client use karta hai | `HotelKeeper` — customer sirf isi se baat karta hai |
| **Subsystem Classes** | Actual complex kaam karne wali classes; Facade ko nahi jaanti | `VegRestaurant`, `NonVegRestaurant`, `VegNonBothRestaurant` |
| **Interface (High-level operations)** | Facade ke expose kiye gaye simple methods | `get_veg_menu()`, `get_non_veg_menu()`, `get_veg_non_menu()` |

### Working — 4 Steps
1. **Client** sirf Facade (`HotelKeeper`) ke saath interact karta hai, subsystem classes se nahi
2. **Facade** andar hi andar zaroori subsystem method(s) ko call karta hai
3. **Subsystem classes** apna kaam karti hain — unhe Facade ke bare mein kuch pata nahi hota (one-way relationship)
4. **Facade** complex workflow ko ek simple, easy-to-use operation mein convert kar deta hai

> **Important Note:** Facade Pattern subsystem classes ko **completely hide nahi** karta — 
> agar koi advanced user directly `VegRestaurant` use karna chahe, to wo kar sakta hai. 
> Facade sirf **most common use-cases** ke liye ek **shortcut/convenience layer** deta hai.


## ✅ Step 2: Solution — WITH Facade Pattern (Best Approach)

Ab dekhte hain **best approach** — Facade Pattern use karke Hotel Keeper example ko 
**clean aur simple** tareeke se implement karte hain.

### Implementation Steps:
1. **Hotel Interface** banao — sab restaurants isko implement karenge
2. **Concrete Restaurant classes** banao — VegRestaurant, NonVegRestaurant, VegNonBothRestaurant
3. **Facade Interface** banao (HotelKeeper) — jo high-level operations declare kare
4. **Concrete Facade** banao (HotelKeeperImplementation) — jo andar se restaurants coordinate kare
5. **Client** — sirf HotelKeeper use kare, kisi restaurant class ka naam na jaane


```python
"""
STEP 1: HOTEL INTERFACE (Common contract for all restaurants)
====================================================================
Sabhi restaurant types is common interface ko follow karte hain.
"""

from abc import ABC, abstractmethod

class Hotel(ABC):
    @abstractmethod
    def get_menus(self):
        pass
```

```python
"""
STEP 2: SUBSYSTEM CLASSES (Concrete Restaurants + Menu classes)
======================================================================
Ye classes actual complex kaam karti hain. Inhe HotelKeeper (Facade)
ke bare mein KUCH pata nahi hai — inka Facade se koi reference nahi hai.
"""

class VegMenu:
    def __str__(self):
        return "Veg Menu: Paneer Tikka, Dal Makhani, Roti"

class NonVegMenu:
    def __str__(self):
        return "Non-Veg Menu: Butter Chicken, Fish Curry"

class BothMenu:
    def __str__(self):
        return "Mixed Menu: Veg + Non-Veg combo"


class VegRestaurant(Hotel):
    def get_menus(self):
        return VegMenu()

class NonVegRestaurant(Hotel):
    def get_menus(self):
        return NonVegMenu()

class VegNonBothRestaurant(Hotel):
    def get_menus(self):
        return BothMenu()
```

```python
"""
STEP 3: FACADE INTERFACE (HotelKeeper)
============================================
Ye high-level operations declare karta hai jo Client actually use karega.
Simple, easy-to-remember method names — customer ko kisi restaurant ka
naam yaad nahi rakhna padta.
"""

class HotelKeeper(ABC):
    @abstractmethod
    def get_veg_menu(self):
        pass

    @abstractmethod
    def get_non_veg_menu(self):
        pass

    @abstractmethod
    def get_veg_non_menu(self):
        pass
```

```python
"""
STEP 4: CONCRETE FACADE (HotelKeeperImplementation)
==========================================================
Ye asli "Manager" hai -- ye jaanta hai ki kaunse subsystem class se
kaunsa kaam karwana hai, aur Client ko sirf FINAL RESULT deta hai.
"""

class HotelKeeperImplementation(HotelKeeper):

    def get_veg_menu(self):
        restaurant = VegRestaurant()          # Facade ke andar complexity hai
        return restaurant.get_menus()          # Client ko ye sab dikhta hi nahi

    def get_non_veg_menu(self):
        restaurant = NonVegRestaurant()
        return restaurant.get_menus()

    def get_veg_non_menu(self):
        restaurant = VegNonBothRestaurant()
        return restaurant.get_menus()
```

```python
"""
STEP 5: CLIENT CODE
=========================
Client (Customer) ko sirf HotelKeeper pata hai -- VegRestaurant,
NonVegRestaurant, jaisi classes ka naam TAK nahi pata!
"""

def main():
    keeper = HotelKeeperImplementation()

    veg_menu = keeper.get_veg_menu()
    print("Customer asked for Veg Menu ->", veg_menu)

    non_veg_menu = keeper.get_non_veg_menu()
    print("Customer asked for Non-Veg Menu ->", non_veg_menu)

    both_menu = keeper.get_veg_non_menu()
    print("Customer asked for Mixed Menu ->", both_menu)


main()

"""
EXPLANATION — Ye BEST APPROACH kyun better hai bad approach se:

✅ SIMPLE CLIENT CODE: Client sirf "keeper.get_veg_menu()" jaisa simple
   call karta hai -- VegRestaurant class ka wajood tak nahi jaanta.

✅ LOOSE COUPLING: Client, HotelKeeper (abstraction) pe depend karta hai,
   concrete VegRestaurant/NonVegRestaurant classes pe nahi.

✅ CENTRALIZED COMPLEXITY: Saari coordination logic EK JAGAH
   (HotelKeeperImplementation) mein hai -- agar restaurant internals
   change ho, sirf Facade update karna padega, Client bilkul safe rahega.

✅ EASY TO EXTEND: Naya restaurant type (jaise "VeganRestaurant") add
   karna ho to Facade mein ek naya method add karo -- Client ka existing
   code TOOTEGA nahi.

Yahi Facade Pattern ki asli power hai — "complexity chhupao, simplicity dikhao"!
"""
```

## ✅ Advantages (Fayde)

| # | Advantage | Explanation |
|---|---|---|
| 1 | Simplified Interface | Client ko complex system ke bajaye simple, clean interface milta hai |
| 2 | Reduced Coupling | Client subsystem internals pe depend nahi karta, modularity badhti hai |
| 3 | Encapsulation | Subsystem mein changes hone pe bhi Client code safe rehta hai |
| 4 | Improved Maintainability | System refactor karna aasan hota hai bina client code todhe |

## ❌ Disadvantages (Nuksan)

| # | Disadvantage | Explanation |
|---|---|---|
| 1 | Extra abstraction layer | Facade khud ek naya layer hai — debugging thodi mushkil ho sakti hai |
| 2 | Reduced flexibility | Agar client ko subsystem ka specific/advanced feature chahiye, Facade limit kar sakta hai |
| 3 | Overengineering risk | Chhote/simple systems ke liye Facade unnecessary complexity add kar sakta hai |
| 4 | Performance overhead | Extra indirection (Facade ke through call jana) thoda overhead add karta hai |

## Q1. 🎬 Home Theater Facade

**Problem:** Ek Home Theater system hai jisme `Projector`, `SoundSystem`, aur `Lights` 
alag-alag subsystems hain. Ek `HomeTheaterFacade` banao jisme ek simple method 
`watch_movie()` ho jo saari devices ko sahi order mein set kare.

**Try yourself first**, phir neeche solution dekho.


```python
"""
SOLUTION — Home Theater Facade
====================================
"""

# Subsystem classes
class Projector:
    def on(self):
        print("📽️ Projector ON")
    def set_input(self, source):
        print(f"📽️ Projector input set to {source}")

class SoundSystem:
    def on(self):
        print("🔊 Sound System ON")
    def set_volume(self, level):
        print(f"🔊 Volume set to {level}")

class Lights:
    def dim(self, level):
        print(f"💡 Lights dimmed to {level}%")


# Facade
class HomeTheaterFacade:
    def __init__(self):
        self.projector = Projector()
        self.sound_system = SoundSystem()
        self.lights = Lights()

    def watch_movie(self, movie_source):
        print(f"--- Setting up movie night: {movie_source} ---")
        self.lights.dim(20)
        self.projector.on()
        self.projector.set_input(movie_source)
        self.sound_system.on()
        self.sound_system.set_volume(15)
        print("🎬 Enjoy your movie!\n")


# Client
theater = HomeTheaterFacade()
theater.watch_movie("Netflix")

"""
EXPLANATION:
Client sirf ek line likhta hai -- theater.watch_movie("Netflix") --
aur andar hi andar 3 alag devices (Projector, SoundSystem, Lights)
sahi order mein coordinate ho jate hain. Bilkul HotelKeeper jaisa hi
concept -- ek simple method, andar complex coordination.
"""
```

## Q2. 🛒 E-commerce Checkout Facade

**Problem:** Ek E-commerce checkout process mein `InventoryService`, `PaymentService`, 
aur `ShippingService` involve hote hain. Ek `OrderFacade` banao jisme ek simple method 
`place_order()` ho jo saari services ko coordinate kare.


```python
"""
SOLUTION — E-commerce Order Facade
=========================================
"""

# Subsystem classes
class InventoryService:
    def check_stock(self, item):
        print(f"📦 Checking stock for {item}... Available!")
        return True

class PaymentService:
    def process_payment(self, amount):
        print(f"💳 Processing payment of Rs.{amount}...")
        return True

class ShippingService:
    def schedule_shipping(self, item):
        print(f"🚚 Shipping scheduled for {item}")


# Facade
class OrderFacade:
    def __init__(self):
        self.inventory = InventoryService()
        self.payment = PaymentService()
        self.shipping = ShippingService()

    def place_order(self, item, amount):
        print(f"--- Placing order for {item} ---")
        if not self.inventory.check_stock(item):
            print("❌ Out of stock!")
            return
        if not self.payment.process_payment(amount):
            print("❌ Payment failed!")
            return
        self.shipping.schedule_shipping(item)
        print(f"✅ Order placed successfully for {item}!\n")


# Client
order = OrderFacade()
order.place_order("Wireless Headphones", 2999)

"""
EXPLANATION:
Client ko InventoryService, PaymentService, ShippingService — kisi ke
bhi bare mein pata nahi hai. Bas "place_order()" call karta hai, aur
OrderFacade andar hi andar saari checks aur services coordinate kar
deta hai. Agar kal koi "FraudCheckService" add karni ho, sirf Facade
update hoga, Client code bilkul safe rahega.
"""
```

## Q3. 🍜 Hotel Facade Extension — Naya Restaurant Type Add Karo

**Problem:** Document ke original Hotel example mein Veg, Non-Veg, Mixed restaurants 
the. Ab ek naya **VeganRestaurant** add karo, aur dikhao ki Facade Pattern ki wajah se 
Client code **bilkul change nahi karna padta**.


```python
"""
SOLUTION — Hotel Facade with a new Vegan Restaurant added
==================================================================
"""

from abc import ABC, abstractmethod

# Subsystem base
class Hotel(ABC):
    @abstractmethod
    def get_menus(self):
        pass

class VeganMenu:
    def __str__(self):
        return "Vegan Menu: Tofu Curry, Vegan Salad, Almond Milk Shake"

# ✅ NAYA SUBSYSTEM -- purani classes ko touch nahi kiya
class VeganRestaurant(Hotel):
    def get_menus(self):
        return VeganMenu()


# Facade interface EXTEND karte hain (sirf ek naya method add hua)
class HotelKeeper(ABC):
    @abstractmethod
    def get_vegan_menu(self):
        pass


class HotelKeeperImplementation(HotelKeeper):
    def get_vegan_menu(self):
        restaurant = VeganRestaurant()
        return restaurant.get_menus()


# Client -- bas NAYA method call kar raha hai, purana code bilkul safe hai
keeper = HotelKeeperImplementation()
print("Customer asked for Vegan Menu ->", keeper.get_vegan_menu())

"""
EXPLANATION:
Notice karo -- humne VegRestaurant, NonVegRestaurant, VegNonBothRestaurant
classes ko BILKUL TOUCH NAHI KIYA. Sirf ek nayi VeganRestaurant class banayi,
aur Facade mein ek naya method add kiya. Existing client code (jo veg/non-veg
menu maangta tha) abhi bhi bilkul waise hi kaam karega -- ye Facade Pattern
ki extensibility dikhata hai.
"""
```

## Q4. 🤔 Conceptual Question (No Code)

**Question:** Pichle notebook mein humne **Adapter Pattern** seekha tha, jisme bhi ek 
class doosri classes ke saath client ko interact karwati hai. Confusion ye hai — 
*"Facade aur Adapter dono to beech mein aake kaam karte hain na? Dono same hi hain kya?"*

### ✅ Answer:

| Aspect | Adapter Pattern | Facade Pattern |
|---|---|---|
| **Purpose** | Do **incompatible interfaces** ko compatible banana | Ek **complex subsystem** ko simple banana |
| **Number of classes involved** | Generally EK Adaptee ko translate karta hai | MULTIPLE subsystem classes ko coordinate karta hai |
| **Interface change** | Adapter ka interface Adaptee se **DIFFERENT** hota hai (naam translate hota hai) | Facade **naya, simplified** interface deta hai, kisi ek specific class ko match nahi karta |
| **Kab use karo** | Jab existing class ka method naam/format tumhare system se **match nahi karta** | Jab ek poora **subsystem bahut complex** hai aur usko simplify karna hai |
| **Example** | LegacyPrinter (`print_document()`) → Printer (`print()`) | VegRestaurant + NonVegRestaurant + VegNonBothRestaurant → HotelKeeper |

**Simple yaad rakhne ka tarika:** Adapter = **"Translator"** (do cheezon ko compatible banata hai, 
usually 1-to-1). Facade = **"Receptionist/Manager"** (bahut saari cheezon ko coordinate karke 
ek simple counter deta hai, usually many-to-1).

**Extra Insight:** Kabhi kabhi dono patterns saath mein bhi use hote hain — jaise agar Hotel mein 
koi ek restaurant ka interface bhi incompatible ho, to HotelKeeper (Facade) ke andar ek 
RestaurantAdapter (Adapter) bhi ho sakta hai!


## Q5. 🧪 Debug Karo: Ye Design Kyun Problematic Hai?

## 🎯 Quick Summary (Revision ke liye)

- **Facade Pattern** ek **complex subsystem** ke upar ek **simple, unified interface** deta hai
- 3 components: **Facade, Subsystem Classes, Interface (high-level operations)**
- Subsystem classes ko Facade ke bare mein **kuch pata nahi hota** (one-directional relationship)
- **Bad approach** mein Client ko HAR subsystem class directly manage karni padti hai (tight coupling, duplication)
- **Best approach** mein Client sirf Facade ke saath interact karta hai — saari complexity peeche chhup jati hai

### Facade vs Adapter (Quick Recall)
| Pattern | Scope | Purpose |
|---|---|---|
| **Adapter** | Usually 1 incompatible class | Interfaces ko COMPATIBLE banana |
| **Facade** | Usually MULTIPLE complex classes | Poore subsystem ko SIMPLE banana |

### All Patterns So Far (Quick Recall)
| Pattern | Type | Focus |
|---|---|---|
| **Singleton** | Creational | Sirf EK instance guarantee karna |
| **Builder** | Creational | Complex object ko STEP-BY-STEP banana |
| **Factory Method** | Creational | Object creation SUBCLASS ko delegate karna |
| **Decorator** | Structural | Object mein RUNTIME pe naya BEHAVIOR add karna |
| **Adapter** | Structural | Do INCOMPATIBLE interfaces ko compatible banana |
| **Facade** | Structural | COMPLEX subsystem ko SIMPLE interface ke peeche chhupana |
