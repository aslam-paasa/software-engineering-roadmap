> Note: Adapter, **Structural Design Pattern** hai (ab tak humne jo bhi patterns dekhe -- Builder, Singleton,
> Factory Method, etc. -- wo sab **Creational** the, jo batate hain "object KAISE banega". Structural patterns
> batate hain "objects/classes EK DOOSRE SE KAISE JUDENGE").

```python
"""
ADAPTER DESIGN PATTERN — KYA HAI?
======================================

Adapter ek STRUCTURAL design pattern hai jo do INCOMPATIBLE
interfaces ke beech ek "BRIDGE" (pul) ka kaam karta hai, taaki
wo DONO EK SAATH kaam kar sakein.

Ye SABSE ZYADA useful hai jab:
    -> Tumhe LEGACY (purana) code integrate karna ho
    -> Tumhe THIRD-PARTY library use karni ho jiska interface
       tumhare system se MATCH nahi karta

SOCHO EK EXAMPLE:
--------------------
Socho tumhare paas ek MOBILE CHARGER hai jiska pin "US TYPE"
hai, lekin tumhare ghar ka SOCKET "INDIAN TYPE" hai. Dono
INCOMPATIBLE hain -- charger seedha socket mein nahi lagega!

Isko solve karne ke liye tum ek "TRAVEL ADAPTER" use karte ho --
ye US pin ko INDIAN socket ke saath COMPATIBLE bana deta hai,
bina charger ya socket ko BADLE.

SOFTWARE mein bhi YAHI hota hai -- Adapter Pattern PURANE
(incompatible) code ko NAYE system ke saath compatible bana
deta hai, bina EXISTING code ko CHHUE.

MAIN COMPONENTS:
--------------------
1. Target Interface -> Wo interface jo CLIENT expect karta hai
                       (e.g. Printer with print())
2. Adaptee            -> EXISTING class jiska interface
                       INCOMPATIBLE hai (e.g. LegacyPrinter with
                       print_document())
3. Adapter             -> Target interface ko IMPLEMENT karta hai,
                       aur ANDAR SE Adaptee ko CALL karta hai --
                       yahi "BRIDGE" hai
4. Client               -> Sirf Target interface use karta hai,
                       use Adapter/Adaptee ke internal DETAILS
                       ka pata NAHI hota

KAISE KAAM KARTA HAI (WORKING):
------------------------------------
1. Client, Adapter ke method ko Target interface ke through call
   karta hai
2. Adapter is call ko Adaptee ke FORMAT mein CONVERT/TRANSLATE
   karta hai
3. Adaptee apna ACTUAL kaam karta hai
4. Client ko RESULT mil jaata hai -- use pata hi nahi chalta ki
   beech mein ADAPTER tha

2 MAIN IMPLEMENTATIONS:
----------------------------
1. CLASS ADAPTER (Inheritance-based) -> Adapter, Target AUR
   Adaptee DONO se INHERIT karta hai. Sirf un languages mein
   possible hai jo MULTIPLE INHERITANCE support karti hain
   (jaise C++). Python/Java mein LESS COMMON hai.
2. OBJECT ADAPTER (Composition-based) -> Adapter, Adaptee ka
   ek INSTANCE apne ANDAR STORE karta hai (composition). Ye
   ZYADA FLEXIBLE hai -- ek Adapter MULTIPLE Adaptees ke saath
   kaam kar sakta hai. Java/Python/C# mein YAHI zyada use hota hai.

REAL LIFE EXAMPLES:
-----------------------
- File Format Adapters: CSV/JSON/XML ko app ke expected format
  mein convert karna
- Device Drivers: OS aur hardware ke beech bridge
- Database Connectors: Alag alag databases ko EK COMMON interface
  ke through access karna
- Language Converters: Chinese-to-English, English-to-Hindi
- Legacy API Migration: Purani API ko NAYI API ke format mein
  wrap karna, bina purana code REWRITE kiye

FAYDE (ADVANTAGES):
------------------------
- Existing code REUSE hota hai bina MODIFY kiye
- Classes apna CORE LOGIC par FOCUSED rehti hain (adaptation
  ALAG jagah isolated hai)
- Multiple ADAPTERS se multiple interfaces support ho sakti hain
- System, IMPLEMENTATION se DECOUPLED rehta hai -- easy to swap

NUKSAAN (DISADVANTAGES):
----------------------------
- Extra LAYER add hoti hai -- code samajhna THODA mushkil ho
  sakta hai
- Har call mein EXTRA INDIRECTION hoti hai -- thoda PERFORMANCE
  overhead
- Bahut saare adapters ban jaaye to MAINTENANCE effort badhta hai
- Bahut saari INCOMPATIBLE interfaces ho to MULTIPLE adapters
  banane padte hain -- design COMPLEX ho sakta hai
"""
print("Concept samajh liya! Ab neeche code mein implement karke dekhte hain.")
```

## ❌ Pehle dekho: BINA Adapter Pattern ke (Problem Code)

Chalo pehle ye dekhte hain ki Adapter Pattern KE BINA log ek PURANE `LegacyPrinter` ko NAYE
system mein integrate karne ki koshish kaise karte hain, aur usme kya dikkat aati hai.

```python
"""
BINA ADAPTER PATTERN KE — PROBLEM DEMO
--------------------------------------------
Socho company ke paas ek PURANA "LegacyPrinter" system hai jo
"print_document()" method use karta hai. Naya system EXPECT
karta hai ki HAR printer ek "print()" method follow kare
(jaise Target interface mein hoga).

Ab client code seedha LegacyPrinter use karne ki koshish karta
hai...
"""

class LegacyPrinter:
    def print_document(self):
        print("🖨️ Legacy Printer is printing a document (OLD method name).")


class ModernPrinter:
    def print(self):
        print("🖨️ Modern Printer is printing (NEW method name).")


# Naya system EXPECT karta hai ki HAR printer ke paas "print()" ho
def client_code_bad(printer):
    printer.print()   # naya system SIRF "print()" call karna jaanta hai


modern = ModernPrinter()
client_code_bad(modern)   # ye theek chalega

legacy = LegacyPrinter()
try:
    client_code_bad(legacy)   # ye CRASH hoga -- LegacyPrinter ke paas "print()" hai hi nahi!
except AttributeError as e:
    print(f"\n❌ ERROR: {e}")
```

```python
"""
ISME PROBLEM KYA THI?
-------------------------
1. INCOMPATIBLE INTERFACE: "LegacyPrinter" ke paas "print()"
   method hai HI NAHI -- usme "print_document()" hai. Naya client
   code SIRF "print()" jaanta hai, isliye CRASH ho gaya
   (AttributeError).

2. LEGACY CODE MODIFY KARNA RISKY HAI: Ek "solution" ye ho sakta
   tha ki "LegacyPrinter" class ke ANDAR JAKE "print_document()"
   ka naam "print()" kar dein. Lekin agar ye class KAHI AUR bhi
   use ho rahi hai (jaise purane modules mein), to naam badalne
   se WAHAN sab TOOT jaayega!

3. THIRD-PARTY CODE MODIFY NAHI KAR SAKTE: Agar "LegacyPrinter"
   ek THIRD-PARTY LIBRARY se aaya ho (jiska SOURCE CODE hamare
   paas hai hi nahi), to hum use MODIFY kar hi NAHI sakte.

4. TIGHT COUPLING: Agar client code mein HAR JAGAH if-else
   likh dein ("agar legacy printer hai to print_document() call
   karo, warna print() call karo"), to ye TIGHT COUPLING create
   karega -- naya printer type add karne par YE if-else HAR JAGAH
   badalna padega.

YE SAB PROBLEMS "Adapter Design Pattern" SOLVE KARTA HAI!
Neeche BEST APPROACH dekhte hain -- LegacyPrinter ko BINA CHHUE,
ek "ADAPTER" banayenge jo use naye system ke saath COMPATIBLE
bana dega.
"""
print("Chalo ab BEST APPROACH (Adapter Pattern) dekhte hain...")
```

## ✅ Ab dekho: BEST APPROACH — Adapter Pattern se solve karte hain

Ye wahi document ka `Printer` / `LegacyPrinter` / `PrinterAdapter` example hai, step-by-step.

### 1️⃣ Target Interface — `Printer`

```python
"""
TARGET INTERFACE (Printer)
-------------------------------
Ye wo interface hai jo CLIENT expect karta hai. Client sirf
"print()" method ke baare mein jaanta hai -- use ye pata nahi
hona chahiye ki andar KAUNSA printer actually kaam kar raha hai.
"""

from abc import ABC, abstractmethod

class Printer(ABC):
    @abstractmethod
    def print(self):
        pass
```

### 2️⃣ Adaptee — `LegacyPrinter` (bina chhue)

```python
"""
ADAPTEE (LegacyPrinter)
----------------------------
Ye EXISTING class hai jiska interface INCOMPATIBLE hai. Isko
hum BILKUL BHI MODIFY NAHI kar rahe -- jaisa hai, WAISA hi
istemaal karenge. (Socho ye ek THIRD-PARTY library ka class hai
jise hum touch nahi kar sakte.)
"""

class LegacyPrinter:
    def print_document(self):
        print("🖨️ Legacy Printer is printing a document.")
```

### 3️⃣ Adapter — `PrinterAdapter` (bridge banata hai)

```python
"""
ADAPTER (PrinterAdapter)
-----------------------------
Ye class Printer interface ko IMPLEMENT karti hai (isliye Client
ise ek NORMAL Printer samajhega), lekin ANDAR SE ye "LegacyPrinter"
ka instance store karti hai aur uska "print_document()" method
CALL karti hai.

Yahi "OBJECT ADAPTER" (composition-based) approach hai -- Adapter
ke andar Adaptee ka OBJECT hai, inheritance nahi.
"""

class PrinterAdapter(Printer):
    def __init__(self, legacy_printer: LegacyPrinter):
        self.legacy_printer = legacy_printer   # composition -- adaptee ko STORE kiya

    def print(self):
        # Client "print()" call karega, hum ANDAR SE "print_document()" call kar dete hain
        self.legacy_printer.print_document()
```

### 4️⃣ Client Code

```python
"""
CLIENT CODE
---------------
Client ko SIRF "Printer" interface pata hai. Use ye BILKUL pata
nahi ki andar ACTUAL mein "LegacyPrinter" kaam kar raha hai --
Adapter ne is FARAK ko CHHUPA diya.
"""

def client_code(printer: Printer):
    printer.print()


legacy_printer = LegacyPrinter()
adapter = PrinterAdapter(legacy_printer)

# Ab client seedha "print()" call kar sakta hai -- CRASH nahi hoga!
client_code(adapter)

print("\n✅ LegacyPrinter ko BILKUL MODIFY nahi kiya, phir bhi naye system ke saath COMPATIBLE ho gaya.")
```

3. `XMLToJSONAdapter` (Adapter) — `JSONDataSource` ko implement kare, `XMLDataSource` ka instance store kare, aur `get_json_data()` mein XML string ko simple JSON-jaisi string mein convert kare (poora XML parser banane ki zaroorat nahi — bas ek simple conversion dikhao, jaise `{"name": "Alice", "age": "30"}` jaisa dictionary return karo)

## ✅ Solution — Practice Question ka Coding Solution

```python
"""
SOLUTION EXPLANATION
-----------------------
Yahan bilkul Printer/LegacyPrinter/PrinterAdapter jaisi hi
structure use ki hai, bas ab "Data Formats" adapt kar rahe hain
instead of "Printers".

1. JSONDataSource   -> Target Interface
2. XMLDataSource      -> Adaptee (bina chhue)
3. XMLToJSONAdapter    -> Adapter (bridge)
4. Neeche client code hai jo sab kuch jodta hai
"""

import re
from abc import ABC, abstractmethod

# ---- 1. Target Interface ----
class JSONDataSource(ABC):
    @abstractmethod
    def get_json_data(self):
        pass


# ---- 2. Adaptee (bina modify kiye) ----
class XMLDataSource:
    def get_xml_data(self):
        return "<data><name>Alice</name><age>30</age></data>"


# ---- 3. Adapter ----
class XMLToJSONAdapter(JSONDataSource):
    def __init__(self, xml_source: XMLDataSource):
        self.xml_source = xml_source   # composition -- adaptee store kiya

    def get_json_data(self):
        xml_string = self.xml_source.get_xml_data()
        # Simple conversion: XML tags dhoondh kar dictionary bana rahe hain
        # (real app mein xml.etree.ElementTree jaisi library use hogi)
        pairs = re.findall(r"<(\w+)>(.*?)</\1>", xml_string)
        json_data = {key: value for key, value in pairs}
        return json_data


# ---- 4. Client Code ----
def client_code(source: JSONDataSource):
    data = source.get_json_data()
    print(f"📦 JSON data mila: {data}")


xml_source = XMLDataSource()
adapter = XMLToJSONAdapter(xml_source)

client_code(adapter)

print("\n✅ XMLDataSource ko BILKUL MODIFY nahi kiya, phir bhi naya system ise JSON ki tarah use kar pa raha hai.")
```
