```python
# 1. Define a simple class with attributes and a method (the building block of every pattern)
#
# Concept / Algorithm:
#    Use __init__ to set instance attributes
#    Add a method that uses those attributes

class Dog:
    def __init__(self, name):
        self.name = name
    def bark(self):
        return f"{self.name} says Woof!"

d = Dog("Rex")
print(d.bark())
```

```python
# 2. Demonstrate encapsulation by hiding internal state behind methods
#
# Concept / Algorithm:
#    Store data in a private-by-convention attribute (underscore prefix)
#    Expose controlled access through getter/setter methods

class BankAccount:
    def __init__(self, balance):
        self._balance = balance
    def deposit(self, amount):
        self._balance += amount
    def get_balance(self):
        return self._balance

acc = BankAccount(100)
acc.deposit(50)
print(acc.get_balance())
```

```python
# 3. Demonstrate inheritance by extending a base class
#
# Concept / Algorithm:
#    Define a base class with shared behavior
#    Define a subclass that inherits and adds its own behavior

class Animal:
    def eat(self):
        return "eating"

class Cat(Animal):
    def meow(self):
        return "meow"

c = Cat()
print(c.eat(), c.meow())
```

```python
# 4. Demonstrate polymorphism by calling the same method name on different classes
#
# Concept / Algorithm:
#    Define multiple classes with a method of the same name
#    Call that method on each without checking the exact type

class Circle:
    def area(self):
        return "pi * r^2"

class Square:
    def area(self):
        return "side^2"

for shape in [Circle(), Square()]:
    print(shape.area())
```

```python
# 5. Demonstrate abstraction using an abstract base class
#
# Concept / Algorithm:
#    Import ABC and abstractmethod
#    Define an abstract method that subclasses must implement

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2

print(Square(4).area())
```

```python
# 6. Show why instantiating an abstract class directly raises an error
#
# Concept / Algorithm:
#    Define an abstract class with an unimplemented abstract method
#    Attempt to instantiate it directly and catch the TypeError

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

try:
    Shape()
except TypeError as e:
    print("Cannot instantiate:", e)
```

```python
# 7. Demonstrate composition (has-a) as an alternative to inheritance (is-a)
#
# Concept / Algorithm:
#    Give a class an instance of another class as an attribute
#    Delegate work to that inner object instead of inheriting from it

class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        self.engine = Engine()
    def start(self):
        return self.engine.start()

print(Car().start())
```

```python
# 8. Compare inheritance versus composition for code reuse
#
# Concept / Algorithm:
#    Implement the same behavior once via inheritance
#    Implement it again via composition to see the tradeoff

class FlyingInherit:
    def fly(self):
        return "flying (inherited)"

class Bird(FlyingInherit):
    pass

class FlyBehavior:
    def fly(self):
        return "flying (composed)"

class Plane:
    def __init__(self):
        self.fly_behavior = FlyBehavior()
    def fly(self):
        return self.fly_behavior.fly()

print(Bird().fly(), Plane().fly())
```

```python
# 10. Demonstrate the Single Responsibility Principle (the S in SOLID)
#
# Concept / Algorithm:
#    Split a class doing two jobs into two classes, each with one responsibility

class InvoicePrinter:
    def print_invoice(self, invoice):
        return f"Printing invoice #{invoice['id']}"

class InvoiceCalculator:
    def calculate_total(self, invoice):
        return sum(invoice["items"])

invoice = {"id": 1, "items": [10, 20]}
print(InvoiceCalculator().calculate_total(invoice))
print(InvoicePrinter().print_invoice(invoice))
```

```python
# 11. Demonstrate the Open/Closed Principle (the O in SOLID): open for extension, closed for modification
#
# Concept / Algorithm:
#    Define a base behavior interface
#    Add new behavior by creating a new subclass rather than editing existing code

class Discount:
    def apply(self, price):
        return price

class StudentDiscount(Discount):
    def apply(self, price):
        return price * 0.9

def checkout(price, discount: Discount):
    return discount.apply(price)

print(checkout(100, StudentDiscount()))
```

```python
# 12. Demonstrate the Liskov Substitution Principle (the L in SOLID): subclasses should be usable wherever the base class is expected
#
# Concept / Algorithm:
#    Define a base class and a subclass that honors the same contract
#    Show the subclass can be substituted without breaking calling code

class Bird:
    def move(self):
        return "moving"

class Sparrow(Bird):
    def move(self):
        return "flying"

def release(bird: Bird):
    return bird.move()

print(release(Sparrow()))
```

```python
# 13. Demonstrate the Interface Segregation Principle (the I in SOLID): prefer small, focused interfaces
#
# Concept / Algorithm:
#    Split a large interface into smaller, role-specific ones
#    Implement only the interfaces a class actually needs

from abc import ABC, abstractmethod

class Printable(ABC):
    @abstractmethod
    def print_doc(self): pass

class Scannable(ABC):
    @abstractmethod
    def scan_doc(self): pass

class SimplePrinter(Printable):
    def print_doc(self):
        return "printing"

print(SimplePrinter().print_doc())
```

```python
# 14. Demonstrate the Dependency Inversion Principle (the D in SOLID): depend on abstractions, not concrete classes
#
# Concept / Algorithm:
#    Define an abstract interface
#    Have a high-level class depend on the interface rather than a specific implementation

from abc import ABC, abstractmethod

class MessageSender(ABC):
    @abstractmethod
    def send(self, msg): pass

class EmailSender(MessageSender):
    def send(self, msg):
        return f"Email: {msg}"

class Notifier:
    def __init__(self, sender: MessageSender):
        self.sender = sender
    def notify(self, msg):
        return self.sender.send(msg)

print(Notifier(EmailSender()).notify("Hello"))
```

```python
# 15. Demonstrate tight coupling and why it makes code harder to change
#
# Concept / Algorithm:
#    Hardcode a concrete dependency inside a class
#    Note that swapping the dependency requires editing this class directly

class MySQLDatabase:
    def save(self, data):
        return f"Saved {data} to MySQL"

class UserService:
    def __init__(self):
        self.db = MySQLDatabase()  # tightly coupled
    def save_user(self, user):
        return self.db.save(user)

print(UserService().save_user("Alice"))
```

```python
# 16. Refactor tightly coupled code into loosely coupled code using dependency injection
#
# Concept / Algorithm:
#    Accept the dependency through the constructor instead of creating it internally
#    Now any compatible dependency can be swapped in

class MySQLDatabase:
    def save(self, data):
        return f"Saved {data} to MySQL"

class UserService:
    def __init__(self, db):
        self.db = db  # loosely coupled
    def save_user(self, user):
        return self.db.save(user)

print(UserService(MySQLDatabase()).save_user("Alice"))
```

```python
# 42. Implement the Singleton pattern using __new__ to guarantee only one instance
#
# Concept / Algorithm:
#    Override __new__ to check if an instance already exists
#    Return the existing instance instead of creating a new one

class Singleton:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

a, b = Singleton(), Singleton()
print(a is b)
```

```python
# 46. Implement a thread-safe Singleton using a lock to prevent race conditions
#
# Concept / Algorithm:
#    Import threading
#    Acquire a lock before checking/creating the instance to avoid two threads creating two instances

import threading

class ThreadSafeSingleton:
    _instance = None
    _lock = threading.Lock()
    def __new__(cls):
        with cls._lock:
            if cls._instance is None:
                cls._instance = super().__new__(cls)
        return cls._instance

print(ThreadSafeSingleton() is ThreadSafeSingleton())
```

```python
# 47. Implement Singleton using a decorator instead of overriding __new__
#
# Concept / Algorithm:
#    Write a decorator that wraps a class and caches its single instance

def singleton(cls):
    instances = {}
    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]
    return get_instance

@singleton
class AppSettings:
    def __init__(self):
        self.theme = "light"

s1 = AppSettings()
s2 = AppSettings()
print(s1 is s2)
```

```python
# 48. Implement Singleton using a metaclass (the more advanced, framework-style approach)
#
# Concept / Algorithm:
#    Define a metaclass overriding __call__
#    Have the metaclass return the cached instance on repeat instantiation

class SingletonMeta(type):
    _instances = {}
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Cache(metaclass=SingletonMeta):
    def __init__(self):
        self.data = {}

print(Cache() is Cache())
```

```python
# 55. Compare a Singleton class with a simple global variable and discuss why Singleton is often preferred
#
# Concept / Algorithm:
#    Implement the same shared counter both ways
#    Note the Singleton version offers encapsulation and controlled access

global_counter = 0
def increment_global():
    global global_counter
    global_counter += 1

class CounterSingleton:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.count = 0
        return cls._instance
    def increment(self):
        self.count += 1

increment_global()
CounterSingleton().increment()
print(global_counter, CounterSingleton().count)
```

```python
# 57. Implement a Singleton registry that stores and retrieves named singletons of different types
#
# Concept / Algorithm:
#    Use a dictionary keyed by class to store one instance per registered type

class Registry:
    _instances = {}
    @classmethod
    def get(cls, klass):
        if klass not in cls._instances:
            cls._instances[klass] = klass()
        return cls._instances[klass]

class ServiceA: pass
class ServiceB: pass

print(Registry.get(ServiceA) is Registry.get(ServiceA))
print(Registry.get(ServiceA) is Registry.get(ServiceB))
```

```python
# 58. Discuss and demonstrate a downside of Singleton: it can make unit testing harder due to shared global state
#
# Concept / Algorithm:
#    Show that state set in one test can leak into another via the shared Singleton instance
#    Add a reset method as a common workaround

class Settings:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.value = None
        return cls._instance
    @classmethod
    def reset(cls):
        cls._instance = None

Settings().value = "test1"
print(Settings().value)
Settings.reset()
print(Settings().value)
```

```python
# 71. Implement Factory Method as an abstract method that subclasses override to control object creation
#
# Concept / Algorithm:
#    Define an abstract creator class with an abstract factory method
#    Let each concrete creator subclass decide which product to build

from abc import ABC, abstractmethod

class Product(ABC):
    @abstractmethod
    def use(self): pass

class ConcreteProductA(Product):
    def use(self): return "Using product A"

class Creator(ABC):
    @abstractmethod
    def factory_method(self) -> Product: pass
    def operate(self):
        product = self.factory_method()
        return product.use()

class CreatorA(Creator):
    def factory_method(self):
        return ConcreteProductA()

print(CreatorA().operate())
```

```python
# 102. Implement a basic Builder that constructs a complex object step by step
#
# Concept / Algorithm:
#    Define a builder with chainable methods that each configure one part
#    Provide a build() method returning the finished object

class PizzaBuilder:
    def __init__(self):
        self.toppings = []
    def add_topping(self, t):
        self.toppings.append(t)
        return self
    def build(self):
        return {"toppings": self.toppings}

pizza = PizzaBuilder().add_topping("cheese").add_topping("olives").build()
print(pizza)
```

```python
# 103. Apply Builder to construct an HTTP request with optional headers and body
#
# Concept / Algorithm:
#    Provide chainable methods for method, header, and body
#    Assemble the final request dictionary with build()

class RequestBuilder:
    def __init__(self):
        self.method = "GET"
        self.headers = {}
        self.body = None
    def set_method(self, m):
        self.method = m
        return self
    def add_header(self, k, v):
        self.headers[k] = v
        return self
    def set_body(self, b):
        self.body = b
        return self
    def build(self):
        return {"method": self.method, "headers": self.headers, "body": self.body}

req = RequestBuilder().set_method("POST").add_header("Content-Type", "json").build()
print(req)
```

```python
# 190. Apply Composite combined with Visitor-like traversal to compute the maximum depth of a nested structure
#
# Concept / Algorithm:
#    Define a common depth() method on both leaf and composite nodes
#    Have composite nodes return one plus the maximum depth of their children

class Leaf:
    def depth(self):
        return 1

class Node:
    def __init__(self):
        self.children = []
    def add(self, c):
        self.children.append(c)
    def depth(self):
        if not self.children:
            return 1
        return 1 + max(c.depth() for c in self.children)

leaf_node = Node()
leaf_node.add(Leaf())
root = Node()
root.add(leaf_node)
print(root.depth())
```

```python
# 198. Apply Decorator to add timing instrumentation to any function without modifying it
#
# Concept / Algorithm:
#    Write a decorator measuring elapsed time around the wrapped function

import time

def timed(func):
    def wrapper(*a, **k):
        start = time.perf_counter()
        result = func(*a, **k)
        print(f"{func.__name__} took {time.perf_counter()-start:.6f}s")
        return result
    return wrapper

@timed
def work():
    return sum(range(10000))

work()
```

```python
# 199. Apply Decorator to add caching behavior to an expensive function
#
# Concept / Algorithm:
#    Write a decorator storing previous results keyed by arguments
#    Return the cached result on repeat calls instead of recomputing

def cache(func):
    store = {}
    def wrapper(n):
        if n not in store:
            store[n] = func(n)
        return store[n]
    return wrapper

@cache
def square(n):
    print("computing")
    return n * n

print(square(4))
print(square(4))
```

```python
# 202. Apply Decorator to add a retry mechanism around an unreliable function
#
# Concept / Algorithm:
#    Write a decorator retrying the wrapped function a fixed number of times on failure

def retry(times):
    def decorator(func):
        def wrapper(*a, **k):
            for _ in range(times):
                try:
                    return func(*a, **k)
                except Exception:
                    continue
            raise Exception("all retries failed")
        return wrapper
    return decorator

attempts = {"n": 0}
@retry(3)
def flaky():
    attempts["n"] += 1
    if attempts["n"] < 2:
        raise ValueError("fail")
    return "success"

print(flaky())
```

```python
# 208. Apply Decorator to add rate-limiting behavior around an API call function
#
# Concept / Algorithm:
#    Write a decorator tracking call counts and rejecting calls beyond a limit

def rate_limited(max_calls):
    def decorator(func):
        state = {"count": 0}
        def wrapper(*a, **k):
            if state["count"] >= max_calls:
                raise Exception("rate limit exceeded")
            state["count"] += 1
            return func(*a, **k)
        return wrapper
    return decorator

@rate_limited(2)
def call_api():
    return "ok"

print(call_api())
print(call_api())
try:
    call_api()
except Exception as e:
    print(e)
```

```python
# 229. Implement a basic Flyweight sharing common data across many similar objects
#
# Concept / Algorithm:
#    Cache shared immutable data keyed by its defining attributes
#    Reuse the same instance for identical attribute sets instead of duplicating it

category_cache = {}

def get_category(name):
    if name not in category_cache:
        category_cache[name] = {"name": name}
    return category_cache[name]

a = get_category("Electronics")
b = get_category("Electronics")
print(a is b)
```

```python
# 248. Apply a Rate-Limiting Proxy that restricts how often the real object's method can be called
#
# Concept / Algorithm:
#    Wrap the real object
#    Track call timestamps and reject calls beyond an allowed rate

import time

class RealAPI:
    def call(self):
        return "response"

class RateLimitedProxy:
    def __init__(self, real, max_calls_per_sec):
        self.real, self.max_calls = real, max_calls_per_sec
        self.calls = []
    def call(self):
        now = time.time()
        self.calls = [t for t in self.calls if now - t < 1]
        if len(self.calls) >= self.max_calls:
            raise Exception("rate limit exceeded")
        self.calls.append(now)
        return self.real.call()

proxy = RateLimitedProxy(RealAPI(), 2)
print(proxy.call())
print(proxy.call())
try:
    proxy.call()
except Exception as e:
    print(e)
```

```python
# 252. Implement a basic Chain of Responsibility where each handler decides to process or pass along a request
#
# Concept / Algorithm:
#    Define handlers each checking a condition
#    Chain them so each is tried in order until one handles the request

def check_a(r): return None if r.get("a") else "missing a"
def check_b(r): return None if r.get("b") else "missing b"

def run_chain(r, handlers):
    for h in handlers:
        error = h(r)
        if error: return error
    return "all checks passed"

print(run_chain({"a": 1}, [check_a, check_b]))
```

```python
# 256. Apply Chain of Responsibility to a logging system where each handler processes messages of a specific severity or higher
#
# Concept / Algorithm:
#    Define handlers for INFO, WARNING, and ERROR that each pass the message along if they cannot fully handle it

class Logger:
    def __init__(self, level, next_logger=None):
        self.level, self.next_logger = level, next_logger
    def log(self, message_level, message):
        result = []
        if message_level >= self.level:
            result.append(f"[{self.level}] {message}")
        if self.next_logger:
            result += self.next_logger.log(message_level, message)
        return result

chain = Logger(1, Logger(2, Logger(3)))
print(chain.log(2, "warning occurred"))
```

```python
# 259. Apply Chain of Responsibility to an ATM cash dispenser breaking a withdrawal into bill denominations
#
# Concept / Algorithm:
#    Chain handlers for $100, $50, $20, and $10 bills, each dispensing what it can and passing the remainder along

class BillHandler:
    def __init__(self, denomination, next_handler=None):
        self.denomination, self.next_handler = denomination, next_handler
    def dispense(self, amount):
        count = amount // self.denomination
        remainder = amount % self.denomination
        result = [f"{count}x ${self.denomination}"] if count else []
        if remainder and self.next_handler:
            result += self.next_handler.dispense(remainder)
        return result

chain = BillHandler(100, BillHandler(50, BillHandler(20)))
print(chain.dispense(170))
```

```python
# 274. Implement a basic custom Iterator class supporting for-loop iteration
#
# Concept / Algorithm:
#    Implement __iter__ returning self and __next__ producing the next item
#    Raise StopIteration when exhausted

class CountUp:
    def __init__(self, limit):
        self.limit, self.current = limit, 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.current >= self.limit:
            raise StopIteration
        self.current += 1
        return self.current

for n in CountUp(3):
    print(n)
```

```python
# 278. Apply Iterator to traverse a tree structure depth-first without exposing the tree's internal shape
#
# Concept / Algorithm:
#    Implement __iter__ using a generator that recursively yields nodes

class TreeNode:
    def __init__(self, value, children=None):
        self.value = value
        self.children = children or []
    def __iter__(self):
        yield self.value
        for child in self.children:
            yield from child

tree = TreeNode(1, [TreeNode(2), TreeNode(3, [TreeNode(4)])])
print(list(tree))
```

```python
# 279. Apply Iterator to paginate through a large dataset lazily, one page at a time
#
# Concept / Algorithm:
#    Implement __next__ to fetch and return the next page only when requested

class PagedResults:
    def __init__(self, data, page_size):
        self.data, self.page_size, self.index = data, page_size, 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index >= len(self.data):
            raise StopIteration
        page = self.data[self.index:self.index+self.page_size]
        self.index += self.page_size
        return page

for page in PagedResults(list(range(7)), 3):
    print(page)
```

```python
# 282. Implement a custom Iterator that supports early termination via a stop condition
#
# Concept / Algorithm:
#    Have __next__ raise StopIteration as soon as a stopping condition is met, even if more data technically remains

class StopAtIterator:
    def __init__(self, data, stop_value):
        self.data, self.stop_value, self.index = data, stop_value, 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index >= len(self.data) or self.data[self.index] == self.stop_value:
            raise StopIteration
        val = self.data[self.index]
        self.index += 1
        return val

print(list(StopAtIterator([1,2,3,99,5], 99)))
```

```python
# 297. Implement a Memento with a dedicated Caretaker class managing the history of snapshots
#
# Concept / Algorithm:
#    Define a Caretaker holding a stack of Mementos separate from the originator object

class Originator:
    def __init__(self):
        self.state = ""
    def save(self):
        return self.state
    def restore(self, memento):
        self.state = memento

class Caretaker:
    def __init__(self):
        self.history = []
    def backup(self, originator):
        self.history.append(originator.save())
    def undo(self, originator):
        if self.history:
            originator.restore(self.history.pop())

orig = Originator()
caretaker = Caretaker()
orig.state = "v1"
caretaker.backup(orig)
orig.state = "v2"
caretaker.undo(orig)
print(orig.state)
```

```python
# 299. Apply Memento to support redo alongside undo by keeping two stacks
#
# Concept / Algorithm:
#    Move undone snapshots into a redo stack
#    Restore from the redo stack when redo is requested

class Editor:
    def __init__(self):
        self.text = ""
        self.undo_stack, self.redo_stack = [], []
    def type(self, chars):
        self.undo_stack.append(self.text)
        self.text += chars
        self.redo_stack.clear()
    def undo(self):
        if self.undo_stack:
            self.redo_stack.append(self.text)
            self.text = self.undo_stack.pop()
    def redo(self):
        if self.redo_stack:
            self.undo_stack.append(self.text)
            self.text = self.redo_stack.pop()

ed = Editor()
ed.type("A")
ed.type("B")
ed.undo()
ed.redo()
print(ed.text)
```

```python
# 335. Implement a basic Template Method defining the skeleton of an algorithm while letting subclasses fill in steps
#
# Concept / Algorithm:
#    Define a base class with a fixed method calling several steps in order
#    Let subclasses override individual steps without changing the overall sequence

class DataProcessor:
    def process(self):
        data = self.load()
        cleaned = self.clean(data)
        return self.save(cleaned)
    def load(self): raise NotImplementedError
    def clean(self, data): return data
    def save(self, data): raise NotImplementedError

class CSVProcessor(DataProcessor):
    def load(self): return "raw csv data"
    def save(self, data): return f"saved: {data}"

print(CSVProcessor().process())
```

```python
# 346. Apply Visitor to compute different operations (area and perimeter) on the same shape hierarchy without changing the shapes
#
# Concept / Algorithm:
#    Define two separate visitors implementing different operations
#    Reuse the same shape classes for both

class Circle:
    def __init__(self, r): self.r = r
    def accept(self, visitor): return visitor.visit_circle(self)

class AreaVisitor:
    def visit_circle(self, c): return 3.14 * c.r ** 2

class PerimeterVisitor:
    def visit_circle(self, c): return 2 * 3.14 * c.r

c = Circle(5)
print(c.accept(AreaVisitor()))
print(c.accept(PerimeterVisitor()))
```

```python
# 347. Apply Visitor to a Composite file/folder tree to compute total size without modifying File or Folder classes
#
# Concept / Algorithm:
#    Define a SizeVisitor with visit methods for both File and Folder
#    Have Folder's visit method recursively call accept on its children

class File:
    def __init__(self, size): self.size = size
    def accept(self, visitor): return visitor.visit_file(self)

class Folder:
    def __init__(self):
        self.children = []
    def add(self, c): self.children.append(c)
    def accept(self, visitor): return visitor.visit_folder(self)

class SizeVisitor:
    def visit_file(self, f): return f.size
    def visit_folder(self, folder):
        return sum(c.accept(self) for c in folder.children)

root = Folder()
root.add(File(100))
root.add(File(200))
print(root.accept(SizeVisitor()))
```

```python
# 351. Apply Visitor to an abstract syntax tree (AST) to evaluate simple arithmetic expressions
#
# Concept / Algorithm:
#    Define Number and Add node classes with accept()
#    Define an EvaluateVisitor that computes the result by visiting each node

class Number:
    def __init__(self, value): self.value = value
    def accept(self, visitor): return visitor.visit_number(self)

class Add:
    def __init__(self, left, right): self.left, self.right = left, right
    def accept(self, visitor): return visitor.visit_add(self)

class EvaluateVisitor:
    def visit_number(self, n): return n.value
    def visit_add(self, a): return a.left.accept(self) + a.right.accept(self)

expr = Add(Number(2), Number(3))
print(expr.accept(EvaluateVisitor()))
```

```python
# 385. Combine Singleton and Factory Method: a single shared registry that creates and caches objects by type
#
# Concept / Algorithm:
#    Show a Singleton-style registry acting as a factory, ensuring one shared instance per requested type

class Registry:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.cache = {}
        return cls._instance
    def get_or_create(self, key, factory_fn):
        if key not in self.cache:
            self.cache[key] = factory_fn()
        return self.cache[key]

reg = Registry()
obj1 = reg.get_or_create("db", lambda: {"connection": "open"})
obj2 = reg.get_or_create("db", lambda: {"connection": "open"})
print(obj1 is obj2)
```

```python
# 387. Combine Observer and Singleton: a single shared event bus that many parts of an app subscribe to
#
# Concept / Algorithm:
#    Show a Singleton EventBus so all subscribers and publishers share the exact same instance automatically

class EventBus:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.subscribers = []
        return cls._instance
    def subscribe(self, fn):
        self.subscribers.append(fn)
    def publish(self, event):
        for fn in self.subscribers:
            fn(event)

EventBus().subscribe(lambda e: print("Handler A:", e))
EventBus().publish("something happened")
```

```python
# 389. Combine Composite and Visitor: use a Visitor to compute a total price across a Composite shopping cart tree
#
# Concept / Algorithm:
#    Reuse the Composite cart structure from earlier
#    Add a PriceVisitor to compute totals without modifying the cart classes

class Product:
    def __init__(self, price): self.price = price
    def accept(self, visitor): return visitor.visit_product(self)

class Bundle:
    def __init__(self):
        self.items = []
    def add(self, item): self.items.append(item)
    def accept(self, visitor): return visitor.visit_bundle(self)

class PriceVisitor:
    def visit_product(self, p): return p.price
    def visit_bundle(self, b): return sum(i.accept(self) for i in b.items)

cart = Bundle()
cart.add(Product(10))
cart.add(Product(20))
print(cart.accept(PriceVisitor()))
```

```python
# 390. Combine Command and Memento: implement full undo support where each command stores its own memento
#
# Concept / Algorithm:
#    Have each command capture a memento of the affected state before executing
#    Use that memento to implement undo cleanly

class SetValueCommand:
    def __init__(self, obj, key, new_value):
        self.obj, self.key, self.new_value = obj, key, new_value
        self.previous_value = None
    def execute(self):
        self.previous_value = self.obj.get(self.key)  # memento captured here
        self.obj[self.key] = self.new_value
    def undo(self):
        self.obj[self.key] = self.previous_value

data = {"status": "draft"}
cmd = SetValueCommand(data, "status", "published")
cmd.execute()
cmd.undo()
print(data["status"])
```

```python
# 421. Final capstone: design a tiny notification system combining Singleton (one dispatcher), Observer (subscribers), and Strategy (delivery channel)
#
# Concept / Algorithm:
#    Bring together three patterns learned throughout this notebook into one small, working, realistic system

class Dispatcher:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.subscribers = []
        return cls._instance
    def subscribe(self, fn):
        self.subscribers.append(fn)
    def dispatch(self, message, channel_strategy):
        formatted = channel_strategy.format(message)
        for fn in self.subscribers:
            fn(formatted)

class EmailStyle:
    def format(self, msg): return f"[Email] {msg}"

Dispatcher().subscribe(lambda m: print("Received:", m))
Dispatcher().dispatch("Your order shipped", EmailStyle())
```

```python
# 422. Final capstone: design a tiny document system combining Builder (constructing sections), Composite (nesting sections), and Visitor (word count)
#
# Concept / Algorithm:
#    Bring together three structural/creational/behavioral patterns into one small working document system

class Section:
    def __init__(self, text=""):
        self.text, self.children = text, []
    def add(self, child):
        self.children.append(child)
        return self
    def accept(self, visitor):
        return visitor.visit(self)

class SectionBuilder:
    def __init__(self):
        self.section = Section()
    def with_text(self, text):
        self.section.text = text
        return self
    def add_child(self, child):
        self.section.add(child)
        return self
    def build(self):
        return self.section

class WordCountVisitor:
    def visit(self, section):
        count = len(section.text.split())
        return count + sum(self.visit(c) for c in section.children)

child = SectionBuilder().with_text("A short child section").build()
root = SectionBuilder().with_text("Main document body").add_child(child).build()
print(root.accept(WordCountVisitor()))
```

```python
# 424. Final capstone: design a tiny e-commerce checkout combining Chain of Responsibility (validation), Command (place order), and Memento (rollback)
#
# Concept / Algorithm:
#    Bring together Chain of Responsibility, Command, and Memento into one small checkout flow

def check_cart_not_empty(order):
    return None if order["items"] else "Cart is empty"

def check_payment_valid(order):
    return None if order.get("payment") else "No payment method"

def run_chain(order, handlers):
    for h in handlers:
        err = h(order)
        if err: return err
    return None

class PlaceOrderCommand:
    def __init__(self, order):
        self.order = order
        self.snapshot = None
    def execute(self):
        self.snapshot = dict(self.order)
        self.order["status"] = "placed"
    def undo(self):
        self.order.clear()
        self.order.update(self.snapshot)

order = {"items": ["book"], "payment": "card"}
error = run_chain(order, [check_cart_not_empty, check_payment_valid])
if not error:
    cmd = PlaceOrderCommand(order)
    cmd.execute()
print(order)
```

```python
# 446. Build a small combined example: a Builder producing a Composite tree of UI elements, then rendered with a Visitor
#
# Concept / Algorithm:
#    Combine Builder, Composite, and Visitor to build then process a small UI tree

class Widget:
    def __init__(self, name):
        self.name, self.children = name, []
    def add(self, child):
        self.children.append(child)
        return self
    def accept(self, visitor):
        return visitor.visit(self)

class UIBuilder:
    def __init__(self, name):
        self.widget = Widget(name)
    def add_child(self, child):
        self.widget.add(child)
        return self
    def build(self):
        return self.widget

class RenderVisitor:
    def visit(self, widget):
        rendered = [widget.name]
        for c in widget.children:
            rendered.append(c.accept(self))
        return rendered

child = UIBuilder("Button").build()
root = UIBuilder("Panel").add_child(child).build()
print(root.accept(RenderVisitor()))
```

```python
# 452. Final capstone: build a tiny undo-capable settings panel combining Memento (snapshot), Observer (live preview), and Singleton (shared settings)
#
# Concept / Algorithm:
#    Bring together Memento, Observer, and Singleton into a small settings management system

class Settings:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.data = {"theme": "light"}
            cls._instance.observers = []
        return cls._instance
    def subscribe(self, fn):
        self.observers.append(fn)
    def set(self, key, value):
        self.data[key] = value
        for fn in self.observers:
            fn(self.data)
    def snapshot(self):
        return dict(self.data)
    def restore(self, snap):
        self.data = snap

Settings().subscribe(lambda d: print("Preview updated:", d))
backup = Settings().snapshot()
Settings().set("theme", "dark")
Settings().restore(backup)
print(Settings().data)
```
