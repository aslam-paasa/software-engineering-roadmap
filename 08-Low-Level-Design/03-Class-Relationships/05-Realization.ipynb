```python
# ============================================================
#  REALIZATION (Implementation) — Aasan Hinglish Guide
# ============================================================
#
#  Socho tum ek payment system bana rahe ho. Tumhare paas alag-alag
#  payment methods hain: Credit Card, PayPal, Bank Transfer. Har
#  method kaam alag tarike se karta hai, par sabka "contract" same
#  hai — ek amount lo aur result do.
#
#  Is "contract poora karna" wale relationship ko code mein kaise
#  model karte hain? Yahi pe REALIZATION aata hai.
#
#  Simple mein:
#  > Realization ek "implements" (contract poora karna) wala
#  > relationship hai. Jahan ek class ek interface (ya abstract class)
#  > ke saath jo wada kiya hai, use poora karti hai.
#
# ============================================================
#
#  1. REALIZATION kya hai?
#  -----------------------
#  Realization ek "contract poora karne" ka relationship hai. Isko
#  ek wade ki tarah samjho:
#    - Interface wada karta hai: "Yeh methods zaroor honge."
#    - Class woh wada poora karti hai: "Main yeh methods dunga."
#
#  Iska kaam kaise hota hai:
#    - Interface batata hai KIYA karna hai (contract).
#    - Class batati hai KYA karna hai (fulfillment).
#    - Implement karne wali class ko interface ke saare methods
#      provide karne hi padte hain.
#    - Multiple classes usi interface ko alag-alag tarike se implement
#      kar sakti hain.
#
#  ------------------------------------------------------------
#  Real-Life Example — Job Description
#  ------------------------------------------------------------
#  Job description (interface) batati hai ki kaam karne wale ke
#  paas kaun-kaun se skills aur responsibilities honi chahiye.
#  Alag-alag employees (classes) us job ko apne tareeke se kar
#  sakte hain, par sabko job description ki requirements puri karni
#  padti hai.
#
# ============================================================
#
#  2. UML mein kaise dikhta hai?
#  ------------------------------
#  UML mein Realization ko ek "dashed line" (tooti hui line) aur
#  "hollow triangle" (khali triangle) se dikhate hain jo interface
#  ki taraf point karta hai.
#
#  Bird ..< Bird realizes Flyable
#  (dashed line + khali triangle)
#
#  Yaad rakho (Inheritance vs Realization):
#  - Inheritance: Solid line (---) + khali triangle. (Direct descent,
#    "is-a" relationship.)
#  - Realization: Tooti hui line (..) + khali triangle. (Contract
#    fulfillment, "implements" relationship.)
#
#  ------------------------------------------------------------
#  Multiple Interfaces
#  ------------------------------------------------------------
#  Ek class multiple interfaces implement kar sakti hai. Ek hi class
#  ke paas bahut saari capabilities aa jaati hain. Yeh single
#  inheritance se zyada flexible hai.
#
#  Example: FileHandler class teen interfaces implement karta hai:
#    - Readable (padh sakta hai)
#    - Writable (likh sakta hai)
#    - Closeable (band kar sakta hai)
#
#  Ek method jise sirf padhna hai, woh sirf Readable accept karega.
#  Use farak nahi padta ki woh FileHandler hai ya NetworkHandler.
#  Bas woh Readable capability chahiye.
#
# ============================================================
#
#  3. Code Example — Flyable Interface
#  -------------------------------------
#  Bird, Airplane, aur Drone — teeno bilkul alag classes hain. Par
#  sab Flyable interface ko implement karte hain. Har class ka apna
#  internal state aur behavior hai, par sab usi contract ko poora
#  karte hain.
#
#  from abc import ABC, abstractmethod
#
#  class Flyable(ABC):
#      @abstractmethod
#      def fly(self):
#          pass
#
#      @abstractmethod
#      def get_flight_info(self):
#          pass
#
#  class Bird(Flyable):
#      def __init__(self, species, wing_span):
#          self.species = species
#          self.wing_span = wing_span
#
#      def fly(self):
#          print(f"{self.species} flaps its wings and takes off.")
#
#      def get_flight_info(self):
#          return f"{self.species} (wingspan: {self.wing_span}m)"
#
#  class Airplane(Flyable):
#      def __init__(self, model, max_altitude):
#          self.model = model
#          self.max_altitude = max_altitude
#
#      def fly(self):
#          print(f"{self.model} engines roar as it accelerates.")
#
#      def get_flight_info(self):
#          return f"{self.model} (max altitude: {self.max_altitude}ft)"
#
#  # Use karte hain:
#  flying_things = [
#      Bird("Eagle", 2.3),
#      Airplane("Boeing 737", 41000),
#  ]
#
#  for flyer in flying_things:
#      print(flyer.get_flight_info())
#      flyer.fly()
#
#  ------------------------------------------------------------
#  Is code mein 4 zaroori baatein dhyan do:
#  ------------------------------------------------------------
#  1. Interface sirf contract deta hai, code nahi. Flyable sirf
#     fly() aur get_flight_info() declare karta hai. Har class apna
#     version khud likhti hai. (Inheritance mein parent ka code free
#     mein milta tha, yahan nahi.)
#
#  2. Teen classes bilkul unrelated hain. Bird, Airplane, Drone mein
#     kuch common nahi hai — na parent, na fields. Sirf Flyable
#     contract common hai.
#
#  3. Calling code sirf interface pe depend karta hai. main method
#     Flyable list pe loop chalata hai. Use farak nahi padta bird
#     hai ya airplane. Code same hai.
#
#  4. Naya flying thing add karna easy. Helicopter add karna ho toh
#     bas ek nayi class banao jo Flyable implement kare, list mein
#     daalo. Existing code change nahi karna. (Open/Closed Principle)
#
# ============================================================
#
#  4. Realization vs Inheritance (Farq samjho)
#  --------------------------------------------
#  Dono hierarchical relationship banate hain, par dono ka kaam alag.
#
#  INHERITANCE = "Identity" (Pehchaan) model karta hai.
#  "A Dog IS an Animal."
#  Child parent ke sab kuch inherit karta hai (state + behavior).
#  Jab true "is-a" relationship ho tab use karo.
#  Example: Dog, Cat, Bird — sab Animals hain. Unka common identity
#  hai (Animal).
#
#  REALIZATION = "Capability" (Kshamta) model karta hai.
#  "A Bird CAN fly, and so can an Airplane."
#  Implementing classes share karte hain kya kar sakte hain, na ki
#  kya hain. Bird aur Airplane mein aur kuch common nahi hai.
#  Jab unrelated classes ek capability share karein tab use karo.
#  Example: Flyable, Serializable, Comparable.
#
#  ------------------------------------------------------------
#  Kab Inheritance use karein?
#  ------------------------------------------------------------
#  - True "is-a" relationship ho (Dog is an Animal).
#  - Related classes ke beech implementation code share karna ho.
#  - Child classes parent ki specializations hon.
#  - State (fields) inherit karna ho.
#
#  ------------------------------------------------------------
#  Kab Realization (Interface) use karein?
#  ------------------------------------------------------------
#  - Unrelated classes ek capability share karein (Flyable, etc.).
#  - Multiple inheritance of behavior chahiye (ek class bahut saare
#    interfaces implement kar sakti hai).
#  - Maximum flexibility aur loose coupling chahiye.
#  - Shared implementation se zyada contract matter karta ho.
#
#  Asli projects mein dono saath use hote hain. Jaise Car class
#  Vehicle ko extend karti hai (common vehicle behavior share karne
#  ke liye) aur Drivable, Insurable interfaces implement karti hai
#  (alag-alag capabilities ke liye).
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  - Realization = "implements" relationship. Class ek interface ke
#    contract ko poora karti hai.
#  - Interface batata hai "kiya" karna hai, class "kaise" karna hai.
#  - UML: tooti hui line (..) + khali triangle.
#  - Inheritance (solid line) = "is-a" (identity). Realization
#    (dashed line) = "can-do" (capability).
#  - Ek class multiple interfaces implement kar sakti hai (multiple
#    capabilities). Inheritance mein ek hi parent.
#  - Naya implementation add karna easy (Open/Closed), unrelated
#    classes ko jodne ka best tareeka.
#  - Asli code mein dono milte hain: Inheritance common behavior ke
#    liye, Interfaces capabilities ke liye.
#
#  Ek line mein: Realization = "Main us wade ko poora karta hoon jo
#  interface ne kiya tha. Hum relation mein nahi, bas same kaam
#  kar sakte hain."
#
# ============================================================
```
