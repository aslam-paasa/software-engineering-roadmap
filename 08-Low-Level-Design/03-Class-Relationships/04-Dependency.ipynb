```python
# ============================================================
#  DEPENDENCY (Part-1) — Sabse Aasan Hinglish Guide
# ============================================================
#
#  Ab tak humne Association, Aggregation, Composition seekha — jahan
#  objects ke beech permanent (long-term) connection hota tha.
#
#  Par kya hota hai jab ek class ko doosri class sirf ek chote se
#  kaam ke liye chahiye? Jaise hi kaam ho jaaye, doosri class ko
#  bhula de?
#
#  Yahi DEPENDENCY hai. Yeh classes ke beech sabse weak (hal ka)
#  relationship hai. Isme koi permanent connection nahi hota.
#
# ============================================================
#
#  1. DEPENDENCY kya hai? (Simple mein)
#  -------------------------------------
#  Dependency tab hoti hai jab ek class doosri class pe kisi kaam
#  ke liye bharosa karti hai, par uska reference hamesha ke liye
#  apne paas nahi rakhti. Bas kaam ho gaya, rishta khatam.
#
#  Simple mein:
#  > "Main tumhe ek kaam ke liye use karta hoon. Kaam hote hi
#  >  tum ja sakte ho. Main tumhe apne paas nahi rakhunga."
#
#  Yeh usually tab hota hai jab:
#    - Ek class doosri class ko method parameter ke roop mein accept
#      karti hai.
#    - Ek class doosri class ko method ke andar instantiate (new)
#      karti hai.
#    - Ek class doosri class ka object method se return karti hai.
#
#  ------------------------------------------------------------
#  Real-Life Analogy — Chef aur Chaku (Knife)
#  ------------------------------------------------------------
#  Soch ek Chef khana bana raha hai.
#    - Chef chaku uthata hai sabzi kaatne ke liye.
#    - Kaatne ke baad chaku wapas rakh diya jaata hai.
#    - Chef chaku ka maalik ho sakta hai ya nahi, par woh usse
#      hamesha apne paas nahi rakhta. Sirf kaam ke waqt use karta hai.
#
#  Yehi Dependency hai. Chef chaku pe sirf khana banate waqt
#  depend karta hai. Baad mein chaku dusre kaam ke liye use ho
#  sakta hai. Permanent connection nahi.
#
#  ------------------------------------------------------------
#  Key Characteristics (Dependency ki pehchaan)
#  ------------------------------------------------------------
#  - Short-lived (kam umar): Relationship sirf method execute hone
#    tak exist karta hai. Method khatam, relationship khatam.
#  - No ownership (koi maaliki nahi): Dependent class doosri class
#    ko field (instance variable) ke roop mein store nahi karti.
#  - "Uses-a" relationship: Class doosri class ko ek kaam poora
#    karne ke liye use karti hai, par retain nahi karti.
#
# ============================================================
#
#  2. UML mein kaise dikhta hai?
#  ------------------------------
#  UML mein Dependency ko ek "dashed arrow" (tooti hui teer) se
#  dikhate hain. Yeh sabse halka notation hai — kyunki yeh sabse
#  halka relationship hai.
#
#  Printer ..> Document
#  (depends on)
#
#  Tooti hui teer ka matlab: "Main tumpe temporarily depend karta
#  hoon, par permanently nahi."
#
#  Yaad rakho:
#  - Association = sirf line (---)
#  - Aggregation = khali heera (◇)
#  - Composition = bhara hua heera (◆)
#  - Dependency = tooti hui teer (..>) [sabse halka]
#
# ============================================================
#
#  3. Code Example — Printer aur Document
#  -------------------------------------
#  Ek Printer hai jo Document ko print karta hai. Printer ko Document
#  method parameter ke roop mein milta hai. Woh use karta hai, aur
#  store nahi karta.
#
#  class Document:
#      def __init__(self, content):
#          self.content = content
#
#      def get_content(self):
#          return self.content
#
#  class Printer:
#      # Dhyan do: Printer ke andar koi 'self.document' field NAHI hai.
#      # Document sirf print() method ke parameter ke roop mein aata hai.
#      def print(self, document):
#          print("Printing:", document.get_content())
#
#  # Use karte hain:
#  doc = Document("Hello, World!")
#  printer = Printer()
#
#  printer.print(doc)
#
#  # print() method khatam hote hi, Printer ka Document se koi naata
#  # nahi rehta. Printer bhool gaya ki koi Document tha bhi ya nahi.
#
#  ------------------------------------------------------------
#  Yeh Association se kya alag hai?
#  ------------------------------------------------------------
#  Association mein: Class doosre object ka reference apne paas
#  field (instance variable) mein rakhti thi. Jaise Order ke paas
#  PaymentGateway ka field tha. Connection method ke baad bhi rehta.
#
#  Dependency mein: Class doosre object ka reference apne paas RAKHTI
#  NAHI. Document sirf print() method ke andar aaya, kaam hua, gaya.
#  Printer ke paas Document ka koi field nahi hai.
#
#  Agar Printer ne Document ko field mein store kar liya hota
#  (jaise self.last_printed = document), toh yeh Association ban
#  jaata. Kyunki connection method ke baad bhi reh jaata.
#
#  Bas yahi farq hai — Field mein store karna (Association) vs
#  Method parameter mein temporarily use karna (Dependency).
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  - Dependency = sabse weak relationship. Ek class doosri class
#    ko ek waqt ke kaam ke liye use karti hai, bina usse apne paas
#    rakhe.
#  - "Uses-a" relationship. Short-lived (method tak hi rehta).
#  - No ownership. Doosri class ko field mein store nahi karta.
#  - UML: tooti hui teer (..>) — sabse halka symbol.
#  - Code: Doosri class method parameter, local variable, ya return
#    type ke roop mein use hoti hai.
#
#  Ek line mein: Dependency = "Main tumhe ek kaam ke liye use karta
#  hoon. Kaam ho gaya, toh tum ja sakte ho. Main tumhe apne paas
#  nahi rakhunga."
#
# ============================================================
```

```python
# ============================================================
#  DEPENDENCY (Part-2) — Code mein Pehchanna — Hinglish Guide
# ============================================================
#
#  Pichle part mein humne seekha ki Dependency sabse halka rishta
#  hai — "Ek kaam karwa lo, phir bhul jaao." Koi field mein store
#  nahi karna.
#
#  Ab dekhenge ki code mein Dependency kaise dikhti hai. Iske 4
#  common roop (forms) hote hain. Ek-ek karke samajhte hain.
#
# ============================================================
#
#  1. Method Parameters ke roop mein (Sabse common)
#  -------------------------------------------------
#  Dependent class doosri class ko method parameter ke roop mein
#  accept karti hai. Method ke andar use karti hai, aur method
#  khatam hote hi doosri class ko chhod deti hai.
#
#  Example: ReportGenerator class generate() method mein DataSource
#  ko parameter ke roop mein leti hai. Data laati hai, report banati
#  hai, aur DataSource ko chhod deti hai. Store nahi karti.
#
#  class ReportGenerator:
#      # 'source' ek parameter hai. Yeh instance variable (field) nahi hai.
#      def generate(self, source):
#          data = source.fetch_all()
#          # Data se report banao...
#          return formatted_report
#
#  Pehchaan: Doosri class object method ke andar aaya, kaam hua,
#  method khatam hote hi object gayab. Class ke paas field nahi.
#
# ============================================================
#
#  2. Local Variables ke roop mein
#  -------------------------------
#  Kabhi-kabhi class doosri class ko method ke ANDAR create karti
#  hai (new karti hai), use karti hai, aur method khatam hote hi
#  discard kar deti hai. Object method ke scope se bahar nahi jaata.
#
#  Example: OrderProcessor class process() method ke andar
#  JsonFormatter banati hai. Use karti hai, aur method khatam hote
#  hi formatter gayab ho jaata hai. Koi field, koi permanent link nahi.
#
#  class OrderProcessor:
#      def process(self, order):
#          formatter = JsonFormatter()  # Andar hi bana liya
#          json = formatter.format(order)
#          # json ko external API pe bhejo...
#          # method khatam hote hi 'formatter' gayab ho jaayega
#
#  Pehchaan: Object method ke andar bana (new), use hua, aur method
#  ke scope se bahar nahi gaya. Field nahi banaya, isliye Dependency
#  hai. Agar self.formatter banaya hota, toh Association ban jaata.
#
# ============================================================
#
#  3. Return Types ke roop mein
#  ----------------------------
#  Kabhi class ka method doosri class ka object return kar deta hai.
#  Yahan bhi class us object ko apne paas field mein store nahi
#  karti. Class us type pe depend karti hai kyunki woh use create
#  karke return karti hai, par retain nahi karti.
#
#  Example: UserFactory class create_user() method mein User object
#  banati hai aur return kar deti hai. Factory ka kaam User banana
#  hai, unhe apne paas rakhna nahi.
#
#  class UserFactory:
#      def create_user(self, name, email):
#          return User(name, email)  # Banaya aur wapas kar diya
#
#  Pehchaan: Method doosri class ka object return kar raha hai.
#  Class khud us object ko apne paas field mein nahi rakh rahi.
#
# ============================================================
#
#  4. Static Method Calls ke roop mein
#  -----------------------------------
#  Class kisi object ka reference liye bina, doosri class ke static
#  methods ko call kar sakti hai. Yahan koi object reference hi
#  nahi hota — sirf class-level dependency hoti hai.
#
#  Example: PasswordService class HashUtils ke sha256() static
#  method ko call karti hai. PasswordService ke paas HashUtils ka
#  koi object nahi hai, bas uske method pe depend karti hai.
#
#  class PasswordService:
#      def verify(self, input_text, hash_value):
#          return HashUtils.sha256(input_text) == hash_value
#
#  Pehchaan: Class ke method ke andar dusri class ka naam likha hai
#  aur uska method call kiya (ClassName.method()). Koi object,
#  koi field, koi parameter nahi.
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  Code mein Dependency 4 tareeke se dikhti hai:
#
#  1. Method Parameter: Doosri class parameter ke roop mein aati.
#     Method khatam, class chhod di. (ReportGenerator -> DataSource)
#
#  2. Local Variable: Doosri class method ke andar 'new' ki gayi.
#     Method khatam, object gayab. (OrderProcessor -> JsonFormatter)
#
#  3. Return Type: Method doosri class ka object return karta hai.
#     Class usse apne paas nahi rakhti. (UserFactory -> User)
#
#  4. Static Method Call: Class doosri class ka object liye bina,
#     uske static method call karti hai. (PasswordService -> HashUtils)
#
#  Sabse zaroori baat: In saari situations mein class doosre object
#  ko apne paas field (instance variable) ke roop mein store NAHI
#  karti. Yahi inhe Dependency banata hai. Agar field mein store
#  kare, toh Association ban jaata hai.
#
#  Ek line mein: "Agar doosri class sirf method ke andar use hui
#  (parameter, local variable, return, ya static call), aur class
#  ke paas field mein store nahi hui — toh yeh Dependency hai."
#
# ============================================================
```

```python
# ============================================================
#  DEPENDENCY (Part-3) — Dependency Injection (DI) — Aasan Hinglish
# ============================================================
#
#  Asli duniya ke apps mein classes akele kaam nahi karti. Unhe
#  dusri classes ki zaroorat padti hai. Jaise:
#    - UserService ko Database chahiye data laane ke liye.
#    - NotificationService ko EmailSender chahiye message bhejne ke liye.
#
#  Par yeh dependencies kaise milni chahiye? Class apne andar hi
#  khud bana le? Ya bahar se mil jaaye?
#
#  Agar class khud bana legi, toh problem aayegi — code change karna
#  mushkil, test karna mushkil. Isliye behtar tareeka hai dependencies
#  bahar se dena. Isko hi DEPENDENCY INJECTION (DI) kehte hain.
#
# ============================================================
#
#  1. Dependency Injection (DI) kya hai?
#  -------------------------------------
#  DI ek design technique hai jahan class apne dependencies ko khud
#  nahi banati, balki bahar se receive karti hai.
#
#  Simple mein:
#  > "Class yeh nahi kehti ki 'main apna saaman khud banaunga'.
#  >  Woh kehti hai 'mujhe yeh chahiye, tum mujhe do'."
#
#  Iske 3 fayde hain:
#  - Better testing: Test ke waqt asli cheez ki jagah nakli (mock)
#    cheez daal sakte ho.
#  - Modularity: Implementation badal sakte ho bina core code chhede.
#    (Email -> SMS)
#  - Loose coupling: Class concrete cheez pe nahi, contract (interface)
#    pe depend karti hai.
#
# ============================================================
#
#  2. Problem: Khud banana (Tight Coupling)
#  -----------------------------------------
#  Maan lo ek NotificationService hai jo email bhejti hai. Simple
#  tareeke mein woh apna EmailSender khad banati hai:
#
#  class NotificationService:
#      def __init__(self):
#          self.sender = EmailSender()  # Apna dependency khud bana liya
#
#      def notify_user(self, message):
#          self.sender.send(message)
#
#  Yeh code dekhne mein theek lag raha hai, par isme 3 problems hain:
#
#  1. Switch nahi kar sakte: Agar SMS bhejna ho email ki jagah, toh
#     NotificationService class ka code change karna padega.
#  2. Test karna mushkil: Unit test mein asli email jaayega (ya fail
#     hoga). Nakli (mock) sender daalne ka koi tareeka nahi.
#  3. Open/Closed Principle break: Naya channel add karne ke liye
#     existing code change karna pad raha hai (extend nahi kar sakte).
#
#  NotificationService EmailSender ke saath tightly juda hua hai.
#  Class khud decide karti hai konsa sender use karna hai, bahar
#  koi change nahi kar sakta.
#
# ============================================================
#
#  3. Solution: Bahar se do (Inject from outside)
#  -----------------------------------------------
#  Class ko apne dependencies khud banane ke bajaye, bahar se
#  provide karo. Yahi Dependency Injection hai.
#
#  Pehle ek Sender interface banao. Phir EmailSender aur SmsSender
#  usko implement karenge. NotificationService ab EmailSender pe
#  nahi, Sender interface pe depend karega. Konsa sender use karna
#  hai, woh constructor se bahar se aayega.
#
#  from abc import ABC, abstractmethod
#
#  class Sender(ABC):
#      @abstractmethod
#      def send(self, message: str) -> None:
#          pass
#
#  class EmailSender(Sender):
#      def send(self, message: str) -> None:
#          print(f"Email: {message}")
#
#  class SmsSender(Sender):
#      def send(self, message: str) -> None:
#          print(f"SMS: {message}")
#
#  class NotificationService:
#      # Dhyan do: constructor 'Sender' interface type leta hai.
#      # Konsa sender (Email/SMS) woh bahar se aayega.
#      def __init__(self, sender: Sender):
#          self.sender = sender  # Bahar se inject hua
#
#      def notify_user(self, message: str) -> None:
#          self.sender.send(message)
#
#  # Use karte hain:
#  # Email bhejna ho toh:
#  email_service = NotificationService(EmailSender())
#  email_service.notify_user("Hello!")
#
#  # SMS bhejna ho toh (bina class change kiye):
#  sms_service = NotificationService(SmsSender())
#  sms_service.notify_user("Hello!")
#
#  Ab NotificationService Sender interface pe depend karta hai, kisi
#  concrete EmailSender pe nahi. Specific implementation bahar se
#  constructor ke through provide ki jaati hai.
#
#  Iske 3 fayde:
#  - Swappable: Production mein EmailSender do, mobile users ke liye
#    SmsSender do. NotificationService ka code change nahi karna.
#  - Easy Testing: Test ke waqt MockSender do jo actual email na bheje,
#    bas record kare. Bina side effects ke test karo.
#  - Loose Coupling: NotificationService sirf Sender interface jaanta
#    hai. Message kaise deliver hota hai, usse koi farak nahi padta.
#
# ============================================================
#
#  4. Real-World mein DI
#  ---------------------
#  Asli apps mein tum yeh manually nahi karte. Frameworks yeh kaam
#    tumhare liye karte hain. Jaise:
#    - Spring (Java)
#    - ASP.NET (C#)
#    - NestJS (TypeScript)
#
#  Yeh frameworks dependencies ko automatically resolve aur inject
#  karte hain configuration ya annotations ke basis pe. Tumhe sab
#  manually wire nahi karna padta. Par underlying concept same hai:
#  classes batati hain unhe kya chahiye, aur koi bahar wali cheez
#  woh provide karti hai.
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  - Dependency Injection (DI) = Class apni dependencies khud nahi
#    banati, bahar se receive karti hai. "Mujhe yeh chahiye, tum do."
#  - Problem (khud banane mein): Switch nahi kar sakte, test mushkil,
#    Open/Closed break. Tight coupling.
#  - Solution (bahar se do): Interface pe depend karo, concrete class
#    pe nahi. Constructor se inject karo. Swap karo, test karo, loose
#    coupling.
#  - Fayde: Better testability (mock daal sakte ho), modularity
#    (Email -> SMS swap), loose coupling (interface pe depend).
#  - Frameworks (Spring, NestJS) DI automatically karte hain.
#
#  Ek line mein: "Apne aap mat banao jo tumhe chahiye, bahar se
#  mangwao. Isse code flexible, testable, aur reusable banta hai."
#
# ============================================================
```

```python
# ============================================================
#  DEPENDENCY — Sabse Aasan Hinglish Guide
# ============================================================
#
#  Ab tak humne 3 rishte seekhe:
#  - Association: "Hum aapas mein jante hain, par apna-apna kaam."
#  - Aggregation: "Main tumhe apne dabbe mein rakhta hoon, par
#    tum akele bhi reh sakte ho." (Playlist-Song)
#  - Composition: "Main tumhe banata hoon, mere bina tumhari koi
#    zindagi nahi." (Ghar-Kamra)
#
#  Ab socho: agar ek class ko doosri class sirf ek chote se kaam
#  ke liye chahiye, jaise hi kaam ho jaaye, doosri class ko bhula
#  de. Toh kya risht hoga? Yahi DEPENDENCY hai.
#
#  Yeh sabse halka (weak) rishta hai. Koi permanent connection nahi.
#
# ============================================================
#
#  1. DEPENDENCY kya hai? (Ekdam simple mein)
#  --------------------------------------------
#  Dependency tab hoti hai jab ek class doosri class ko ek waqt
#  ke kaam ke liye use karti hai, par usse hamesha ke liye apne
#  paas nahi rakhti. Kaam ho gaya, rishta khatam.
#
#  Simple mein:
#  > "Ek kaam karwa lo, phir bhul jaao."
#
#  ------------------------------------------------------------
#  Real-Life Example — Chef aur Chaku (Knife)
#  ------------------------------------------------------------
#  Ek Chef khana bana raha hai.
#    - Chef chaku uthata hai sabzi kaatne ke liye.
#    - Kaatne ke baad chaku rakh diya.
#    - Chef chaku ko hamesha apne paas nahi rakhta. Sirf kaam ke
#      waqt use karta hai, phir wapas rakh deta hai.
#
#  Yahi Dependency hai. Chef chaku pe sirf khana banate waqt depend
#  karta hai. Baad mein chaku dusre kaam mein use ho sakta hai.
#
#  ------------------------------------------------------------
#  Pehchaan (Key Characteristics)
#  ------------------------------------------------------------
#  - Short-lived (kam umar): Rishta sirf kaam ke waqt exist karta
#    hai. Kaam khatam, rishta khatam.
#  - No ownership (koi maaliki nahi): Class doosri class ko apne
#    paas (field/variable mein) store nahi karti.
#  - "Uses-a" relationship: Use kiya, kaam karwaya, chhod diya.
#
# ============================================================
#
#  2. UML mein kaise dikhta hai?
#  ------------------------------
#  UML mein Dependency ko "tooti hui teer" (dashed arrow) se
#  dikhate hain. Yeh sabse halka symbol hai.
#
#  Printer ..> Document
#  (depends on)
#
#  Tooti hui teer ka matlab: "Main tumpe temporarily depend karta
#  hoon, par permanently nahi."
#
#  Yaad rakho (halka se strong tak):
#  - Dependency  = tooti hui teer (..>) [sabse halka]
#  - Association = sirf line (---)
#  - Aggregation = khali heera (◇)
#  - Composition = bhara hua heera (◆) [sabse strong]
#
# ============================================================
#
#  3. Code Example — Printer aur Document
#  -------------------------------------
#  Ek Printer hai jo Document ko print karta hai. Printer ko
#  Document sirf print karne ke liye chahiye. Print ho gaya, toh
#  Printer Document ko bhool jaata hai.
#
#  class Document:
#      def __init__(self, content):
#          self.content = content
#
#      def get_content(self):
#          return self.content
#
#  class Printer:
#      # Dhyan do: Printer ke andar koi 'self.document' field NAHI hai.
#      # Document sirf print() method ke parameter ke roop mein aata hai.
#      def print(self, document):
#          print("Printing:", document.get_content())
#
#  # Use karte hain:
#  doc = Document("Hello, World!")
#  printer = Printer()
#
#  printer.print(doc)
#
#  # print() method khatam hote hi, Printer ka Document se koi naata
#  # nahi rehta. Printer bhool gaya ki koi Document tha bhi ya nahi.
#
#  ------------------------------------------------------------
#  Yeh Association se kya alag hai?
#  ------------------------------------------------------------
#  Association: Class doosre object ka reference apne paas field
#  mein rakhti thi. Connection method ke baad bhi rehta tha.
#  (Jaise Order ke paas PaymentGateway ka field tha.)
#
#  Dependency: Class doosre object ka reference apne paas RAKHTI
#  NAHI. Document sirf print() method ke andar aaya, kaam hua, gaya.
#  Printer ke paas Document ka koi field nahi hai.
#
#  Agar Printer ne Document ko field mein store kar liya hota
#  (jaise self.last_printed = document), toh yeh Association ban
#  jaata. Kyunki connection method ke baad bhi reh jaata.
#
#  Bas yahi farq hai — Field mein store karna (Association) vs
#  Method parameter mein temporarily use karna (Dependency).
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  - Dependency = sabse halka rishta. "Ek kaam karwa lo, phir
#    bhul jaao."
#  - Short-lived (method tak hi rehta), no ownership (field mein
#    store nahi), "uses-a" relationship.
#  - UML: tooti hui teer (..>) — sabse halka symbol.
#  - Code: Doosri class method parameter, local variable, ya return
#    type ke roop mein use hoti hai. Field mein nahi.
#
#  Ek line mein: Dependency = "Main tumhe ek kaam ke liye use karta
#  hoon. Kaam ho gaya, toh tum ja sakte ho. Main tumhe apne paas
#  nahi rakhunga."
#
# ============================================================
```

```python
# ============================================================
#  DEPENDENCY (Part-2) — Code mein Pehchanna — Aasan Hinglish
# ============================================================
#
#  Pichle part mein seekha: Dependency = sabse halka rishta.
#  "Ek kaam karwa lo, phir bhul jaao. Apne paas mat rakhna."
#
#  Ab baat yeh hai ki code likhte waqt yeh Dependency kaise
#  dikhti hai? Iske 4 simple tareeke hain. Chalo dekhenge.
#
# ============================================================
#
#  1. Method Parameter (Sabse common)
#  ----------------------------------
#  Tum ek class ka method bana rahe ho. Us method ko input
#  (parameter) ke roop mein doosri class ka object chahiye.
#  Method ke andar use karo, method khatam hote hi object
#  bhool jaao.
#
#  Example: Ek report banane wali class hai. Usko data chahiye.
#  Tum data wali class ka object parameter mein de dete ho.
#  Report ban gaya, object ka kaam khatam.
#
#  class ReportGenerator:
#      def generate(self, source):   # 'source' parameter hai
#          data = source.fetch_all()
#          # report banao...
#          return formatted_report
#
#  Pehchaan: Object method ke andar aaya, kaam hua, gayab ho gaya.
#  Class ke paas field (self.source) nahi hai.
#
# ============================================================
#
#  2. Local Variable (Andar hi bana lo)
#  ------------------------------------
#  Kabhi tum method ke andar hi ek naya object bana leta ho
#  (new karke), use karte ho, aur method khatam hote hi woh
#  object khatam ho jaata hai. Bahar ka pata hi nahi chalta.
#
#  Example: Order process karte waqt tumhe JSON format karna
#  hai. Tum method ke andar hi JsonFormatter bana lo. Kaam ho
#  gaya, formatter gayab.
#
#  class OrderProcessor:
#      def process(self, order):
#          formatter = JsonFormatter()  # Andar hi bana liya
#          json = formatter.format(order)
#          # json bhejo...
#          # method khatam, 'formatter' gayab
#
#  Pehchaan: Object method ke andar bana, andar hi mara.
#  Field (self.formatter) nahi banaya, isliye Dependency hai.
#  Agar self.formatter banaya hota, toh Association ho jaata.
#
# ============================================================
#
#  3. Return Type (Wapas kar diya)
#  -------------------------------
#  Kabhi tumhara method doosri class ka naya object banata hai
#  aur usko wapas (return) kar deta hai. Tum apne paas nahi
#  rakhte. Tumhara kaam sirf usko banana hai.
#
#  Example: Ek Factory class hai jo User banati hai. User bana,
#  return kar diya. Factory apne paas User nahi rakhti.
#
#  class UserFactory:
#      def create_user(self, name, email):
#          return User(name, email)  # Banaya aur wapas kar diya
#
#  Pehchaan: Method doosri class ka object return kar raha hai.
#  Class khad us object ko apne paas field mein nahi rakh rahi.
#
# ============================================================
#
#  4. Static Method Call (Bina object ke)
#  --------------------------------------
#  Kabhi tum kisi class ka object banaaye bina uske static
#  method ko direct call kar sakte ho. Sirf class ka naam
#  likho aur method call karo. Koi object, koi field nahi.
#
#  Example: Password check karte waqt tum HashUtils ke sha256()
#  method ko call karte ho. Tumhare paas HashUtils ka object
#  nahi hai, bas uske method pe depend karte ho.
#
#  class PasswordService:
#      def verify(self, input_text, hash_value):
#          # HashUtils.sha256 direct class pe call kiya
#          return HashUtils.sha256(input_text) == hash_value
#
#  Pehchaan: ClassName.method() wala pattern. Koi object, koi
#  field, koi parameter nahi. Sirf class ka naam.
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  Code mein Dependency 4 tareeke se dikhti hai:
#
#  1. Parameter: Doosri class input mein aayi, kaam hua, chhooti.
#     (ReportGenerator -> DataSource)
#
#  2. Local Variable: Doosri class method ke andar bani, aur
#     method ke andar hi khatam ho gayi.
#     (OrderProcessor -> JsonFormatter)
#
#  3. Return Type: Method doosri class ka object banake wapas
#     kar diya. Apne paas nahi rakha.
#     (UserFactory -> User)
#
#  4. Static Method: Doosri class ka object banaye bina uske
#     method ko direct call kiya.
#     (PasswordService -> HashUtils)
#
#  Sabse badi baat: In sab mein class doosre object ko apne paas
#  field (self.xxx) mein store NAHI karti. Yahi inhe Dependency
#  banata hai. Agar field mein store kare, toh Association ban
#  jaata hai.
#
#  Ek line mein: "Agar doosri class sirf method ke andar use hui
#  aur class ke paas field mein nahi hui — toh yeh Dependency hai."
#
# ============================================================
```

```python
# ============================================================
#  DEPENDENCY (Part-3) — Dependency Injection (DI) — Aasan Hinglish
# ============================================================
#
#  Asli apps mein classes akele kaam nahi karti. Unhe dusri
#  classes ki zaroorat padti hai. Jaise:
#    - UserService ko Database chahiye.
#    - NotificationService ko EmailSender chahiye.
#
#  Par yeh helper class apne andar khad bana le ya bahar se mile?
#  Agar khud banayi, toh problem aayegi. Isliye behtar hai bahar
#  se mil jaaye. Isko hi DEPENDENCY INJECTION (DI) kehte hain.
#
# ============================================================
#
#  1. Dependency Injection (DI) kya hai? (Ekdam simple)
#  ----------------------------------------------------
#  Class apne helper objects ko khud NAHI banata. Woh bahar se
#  mil jaate hain. Class sirf bolta hai "mujhe yeh chahiye", aur
#  koi bahar wala usko de deta hai.
#
#  Simple Analogy:
#  > Tum restaurant mein ho. Tum khud kitchen mein jaakar khana
#  > nahi banate. Tum waiter ko bolte ho "mujhe Pizza chahiye".
#  > Waiter tumhe Pizza de deta hai. Tumne khud nahi banaya,
#  > bahar se mil gaya. Yahi DI hai.
#
# ============================================================
#
#  2. Problem: Khud banana (Galat tareeka)
#  -----------------------------------------
#  Maan lo ek NotificationService class hai jo Email bhejti hai.
#  Woh apna EmailSender khad banati hai:
#
#  class NotificationService:
#      def __init__(self):
#          self.sender = EmailSender()  # Khud bana liya!
#
#      def notify_user(self, message):
#          self.sender.send(message)
#
#  Isme 3 problems hain:
#  1. Switch nahi kar sakte: Agar SMS bhejna ho, toh class ka code
#     todna padega. (EmailSender nikal kar SmsSender likhna padega.)
#  2. Test karna mushkil: Test ke waqt asli email jaayega. Nakli
#     (mock) sender daalne ka koi tareeka nahi.
#  3. Naya feature add karna mushkil: Naya channel (Push) add karne
#     ke liye purana code change karna padega.
#
#  Problem yahi hai ki class aur helper (EmailSender) tightly jude
#  hue hain. Class decide kar rahi hai konsa helper use hoga, bahar
#  koi change nahi kar sakta.
#
# ============================================================
#
#  3. Solution: Bahar se do (Sahi tareeka - DI)
#  --------------------------------------------
#  Class ko apna helper khud banane ke bajaye, bahar se provide karo.
#
#  Pehle ek Sender interface banao. Phir EmailSender aur SmsSender
#  usko implement karenge. NotificationService ab Sender interface
#  pe depend karega. Konsa sender use karna hai, woh constructor
#  se bahar se aayega.
#
#  from abc import ABC, abstractmethod
#
#  class Sender(ABC):
#      @abstractmethod
#      def send(self, message: str) -> None:
#          pass
#
#  class EmailSender(Sender):
#      def send(self, message: str) -> None:
#          print(f"Email: {message}")
#
#  class SmsSender(Sender):
#      def send(self, message: str) -> None:
#          print(f"SMS: {message}")
#
#  class NotificationService:
#      # Dhyan do: constructor mein 'Sender' type likha hai.
#      # Konsa sender (Email/SMS) woh bahar se aayega.
#      def __init__(self, sender: Sender):
#          self.sender = sender  # Bahar se mil gaya (Injected)
#
#      def notify_user(self, message: str) -> None:
#          self.sender.send(message)
#
#  # Use karte hain:
#  # Email bhejna ho toh:
#  email_service = NotificationService(EmailSender())
#  email_service.notify_user("Hello!")
#
#  # SMS bhejna ho toh (bina class change kiye):
#  sms_service = NotificationService(SmsSender())
#  sms_service.notify_user("Hello!")
#
#  Dekho kitna simple hai. Class ka code change kiye bina hum
#  Email kaam kar sakte hain ya SMS. Bas bahar se appropriate
#  object pass kar do.
#
#  Iske 3 fayde:
#  - Swap karo: Production mein EmailSender do, mobile pe SmsSender.
#    Class ka code change nahi karna.
#  - Test karo: Test ke waqt Nakli (Mock) Sender do jo asli email
#    na bheje, bas record kare.
#  - Loose Coupling: Class sirf Sender interface jaanti hai. Email
#    jayega ya SMS, usse koi farak nahi padta.
#
# ============================================================
#
#  4. Real-World mein DI
#  ---------------------
#  Asli apps mein tum yeh manually nahi karte. Frameworks yeh kaam
#  tumhare liye karte hain. Jaise:
#    - Spring (Java)
#    - NestJS (TypeScript)
#
#  Yeh frameworks dependencies ko automatically resolve aur inject
#  karte hain. Tumhe sab manually wire nahi karna padta. Par
#  underlying concept same hai: classes batati hain unhe kya
#  chahiye, aur framework woh provide karta hai.
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  - Dependency Injection (DI) = Class apne helper objects khud
#    nahi banata, bahar se mangwata hai. (Restaurant analogy: tum
#    khana khud nahi banate, waiter se mangwate ho.)
#  - Problem (khud banane mein): Switch nahi kar sakte, test mushkil,
#    naya feature add karna mushkil. Tight coupling.
#  - Solution (bahar se do): Interface pe depend karo. Constructor
#    se bahar se object pass karo. Class ka code change nahi karna.
#  - Fayde: Swap karna easy, test karna easy (mock daal sakte ho),
#    loose coupling.
#  - Frameworks (Spring, NestJS) yeh automatically karte hain.
#
#  Ek line mein: "Apne aap mat banao jo tumhe chahiye, bahar se
#  mangwao. Isse code flexible, testable, aur reusable banta hai."
#
# ============================================================
```
