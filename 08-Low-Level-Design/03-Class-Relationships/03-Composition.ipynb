```python
# ============================================================
#  COMPOSITION — Bahut Aasan Hinglish Guide (Beginner ke liye)
# ============================================================
#
#  Pichle chapter mein humne Aggregation seekha — jahan "whole"
#  part ko contain karta tha, par part bina whole ke bhi exist kar
#  sakta tha (jaise School aur Teacher).
#
#  Par kya hota hai agar relationship itna strong ho ki part bina
#  whole ke exist hi na kar sake? Whole mar jaaye toh part bhi mar
#  jaaye?
#
#  Yahi duniya hai COMPOSITION ki. Yeh "has-a" relationship ka sabse
#  strong form hai, jahan whole apne parts ka poora maalik hota hai
#  aur unka lifecycle control karta hai.
#
#  Simple mein:
#  > "Main tumhe banata hoon, main hi tumhe khatam karta hoon.
#  >  Tum bina mere ke kuch nahi ho."
#
# ============================================================
#
#  1. COMPOSITION kya hai? (Simple mein)
#  -------------------------------------
#  Composition ek aisa relationship hai jahan "whole" class apne
#  "part" objects ko khud create karti hai, unhe manage karti hai,
#  aur jab whole destroy hota hai, toh parts bhi destroy ho jaate
#  hain. Parts ka bina whole ke koi existence nahi hota.
#
#  ------------------------------------------------------------
#  Simple Analogy — Ghar (House) aur Kamre (Rooms)
#  ------------------------------------------------------------
#  Soch ek Ghar hai aur uske Kamre hain (living room, bedroom, kitchen).
#    - Ghar ke bina kamre ka koi matlab nahi. Kamra akele exist nahi
#      kar sakta.
#    - Agar ghar gir gaya (demolish), toh saare kamre bhi gir gaye.
#    - Tum ek bedroom ko utha kar doosre ghar mein nahi le ja sakte.
#
#  Kamre ghar ke "integral parts" hain. Unka lifecycle ghar ke saath
#  juda hua hai. Ghar mara, toh kamre bhi mare. Yahi Composition hai.
#
#  ------------------------------------------------------------
#  Key Characteristics (Composition ki pehchaan)
#  ------------------------------------------------------------
#  - Strong "has-a" relationship (mera hissa hai, poori maaliki).
#  - Whole, part ka lifecycle control karta hai (create + destroy).
#  - Whole destroy hoga toh parts bhi destroy ho jaayenge.
#  - Parts doosre objects ke saath share nahi hote (exclusive).
#  - Part ka bina whole ke koi independent identity nahi hota.
#
#  Rule of thumb: Agar part bina whole ke sense hi nahi karta, toh
#  use Composition.
#
# ============================================================
#
#  2. UML mein Composition kaise dikhta hai?
#  -----------------------------------------
#  UML diagram mein Composition ko "bhare hueere" (filled diamond)
#  se dikhate hain — "whole" side pe.
#
#  Ghar ◆---- Kamra
#  (whole)       (part)
#
#  Bhara hua heera ka matlab: "poori maaliki, part mere saath hi
#  paida hua, mere saath hi mar jaayega."
#
#  (Aggregation mein khali heera tha, Composition mein bhara hua.
#   Bhara hua = zyada strong, zyada ownership.)
#
# ============================================================
#
#  3. Code Example — Order aur LineItem
#  -------------------------------------
#  Ek Order (whole) mein bahut saare LineItem (parts) hote hain.
#  Order apne LineItems ko khud banata hai (create) aur jab Order
#  delete hota hai, toh saare LineItems bhi delete ho jaate hain.
#
#  class LineItem:
#      def __init__(self, product_name, quantity, unit_price):
#          self.product_name = product_name
#          self.quantity = quantity
#          self.unit_price = unit_price
#
#      def get_subtotal(self):
#          return self.quantity * self.unit_price
#
#      def describe(self):
#          print(f"{self.product_name} x{self.quantity} "
#                f"@ ${self.unit_price:.2f} = ${self.get_subtotal():.2f}")
#
#  class Order:
#      def __init__(self, order_id):
#          self.order_id = order_id
#          self.line_items = []   # Order apne parts khad manage karta hai
#
#      def add_item(self, product, quantity, unit_price):
#          # Order andar hi LineItem CREATE kar raha hai!
#          # Bahar se pass nahi hua. Yahi Composition ki pehchaan hai.
#          self.line_items.append(LineItem(product, quantity, unit_price))
#
#      def remove_item(self, product):
#          self.line_items = [
#              item for item in self.line_items
#              if item.product_name != product
#          ]
#
#      def get_total(self):
#          return sum(item.get_subtotal() for item in self.line_items)
#
#      def print_receipt(self):
#          print(f"Order: {self.order_id}")
#          for item in self.line_items:
#              item.describe()
#          print(f"Total: ${self.get_total():.2f}")
#
#  Usage:
#  order = Order("ORD-1001")
#  order.add_item("Wireless Mouse", 2, 29.99)
#  order.add_item("USB-C Cable", 3, 9.99)
#  order.add_item("Laptop Stand", 1, 49.99)
#
#  order.print_receipt()
#
#  # Jab order delete hoga, saare LineItems bhi destroy ho jaayenge.
#  # Koi LineItem Order ke bina exist hi nahi karta.
#
#  ------------------------------------------------------------
#  Teen cheezon pe dhyan do (jo isse Composition banati hain)
#  ------------------------------------------------------------
#  1. Order apne LineItems khud banata hai. add_item() bahar se koi
#     LineItem object receive nahi karta. Woh raw data (naam, qty,
#     price) leta hai aur andar "new LineItem(...)" create karta hai.
#     (Aggregation mein parts bahar bante the aur pass hote the.)
#
#  2. LineItem ka independent existence nahi hai. System mein koi
#     aisa LineItem nahi hai jo kisi Order ke andar na ho. Koi dusri
#     class in LineItems ka reference nahi rakhti. Woh Order ke andar
#     paida hue, Order ke saath hi marenge.
#
#  3. Order delete hone se saare LineItems destroy ho jaate hain.
#     Koi orphan (ghumakkad) LineItem nahi bachta, koi dangling
#     reference nahi rehta.
#
#  Yahi true Composition hai — parts sirf whole ke context mein exist
#  karte hain, unka lifecycle poora whole control karta hai.
#
# ============================================================
#
#  4. COMPOSITION kab use karein?
#  ------------------------------
#  Jab in sawaalon ka jawab "HAAN" ho, tab Composition use karo:
#
#  - Kya part bina whole ke meaningless hai?
#    (LineItem bina Order ka koi kaam nahi. Kamra bina Ghar ke bekar.)
#    Haan? Toh Composition.
#
#  - Kya whole ko part ka lifecycle control karna chahiye?
#    (Whole parts ko create aur destroy kare.)
#    Haan? Toh Composition.
#
#  - Kya parts exclusive hain (sirf ek whole ke)?
#    (Ek LineItem doosre Order mein nahi ja sakta. Ek Kamra doosre
#    Ghar mein nahi ja sakta.)
#    Haan? Toh Composition.
#
#  - Kya relationship "is composed of" wala hai?
#    (Ghar kamro se bana hai. Order LineItems se bana hai.)
#    Haan? Toh Composition.
#
#  ------------------------------------------------------------
#  "Favor Composition over Inheritance" — GoF Design Principle
#  ------------------------------------------------------------
#  Composition, Inheritance ke preferred alternative ke roop mein
#  dekha jaata hai. Inheritance ke fayde bahut hain, par Composition
#  se system flexible banta hai.
#
#  Kyun?
#  - Tum complex behavior chhote, reusable parts se bana sakte ho.
#  - Inheritance ki tarah tight coupling aur fragility se bach jaate ho.
#  - Parts ko dynamically swap kar sakte ho behavior change karne ke
#    liye.
#
#  Example:
#  - Ek Vehicle class Engine interface ko compose karta hai.
#  - Runtime pe swap kar sakte ho: PetrolEngine, ElectricEngine, ya
#    HybridEngine.
#
#  Isse code cleaner, testable, aur decoupled banta hai.
#
# ============================================================
#
#  5. Association vs Aggregation vs Composition
#  --------------------------------------------
#  Teeno ka comparison dekhte hain — ownership, lifecycle, aur usage
#  ke hisaab se:
#
#  | Feature      | Association              | Aggregation                  | Composition                      |
#  |--------------|--------------------------|------------------------------|----------------------------------|
#  | Ownership    | None                     | Weak (has-a)                 | Strong (owns-a)                  |
#  | Lifecycle    | Independent              | Independent                  | Dependent (part dies with whole) |
#  | Coupling     | Loose                    | Moderate                     | Tight                            |
#  | Reusability  | High (parts reusable)    | Moderate (parts often reused)| Low (parts not reused outside)   |
#  | UML Symbol   | Solid Line (---)         | Hollow Diamond (◇)           | Filled Diamond (◆)               |
#  | Who creates  | Either/External          | External (passed in)         | Whole (created internally)       |
#  | Real Example | Student <-> Course       | Playlist -> Song             | Order -> LineItem                |
#
#  ------------------------------------------------------------
#  Aise samjho (Simple way to remember):
#  ------------------------------------------------------------
#  - Association: Do log aapas mein jante hain. Par apna-apna kaam.
#                 Koi maalik nahi. (Student-Course)
#
#  - Aggregation: Ek dabba (container) hai, usme cheezein hain.
#                 Par cheezein akele bhi exist kar sakti hain.
#                 Dabba mara, toh cheezein bach jaayengi.
#                 (Playlist-Song: Playlist delete, Song reh jaata.)
#
#  - Composition: Ek dabba (container) hai, usme cheezein itni
#                 tightly judi hain ki woh uske "integral part" hain.
#                 Dabba mara, toh cheezein bhi maar di jaayengi.
#                 Bina dabbe ke cheezon ka koi matlab nahi.
#                 (House-Room: Ghar gira, toh kamre bhi gaye.)
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  - Composition = strongest "has-a" relationship. Whole apne parts
#    ka poori maalik hai aur unka lifecycle control karta hai.
#  - Part bina whole ke exist hi nahi kar sakta. Whole delete ho,
#    toh parts bhi delete ho.
#  - Parts exclusive hote hain — doosre whole ke saath share nahi hote.
#  - UML: bhara hua heera (filled diamond) on the whole side.
#  - Code: Whole parts ko khud create karta hai (constructor ya
#    method ke andar "new" se), bahar se receive nahi karta.
#  - Composition inheritance se behtar hota hai flexibility ke liye.
#
#  Ek line mein: Composition = "Main tumhe banata hoon, mere bina
#  tumhari koi zindagi nahi. Main marunga toh tum bhi maroge."
#
# ============================================================
```

```python
# ============================================================
#  COMPOSITION — Sabse Aasan Hinglish Guide
# ============================================================
#
#  Pehle Association seekha: "Do cheezein connected hain, par
#  dono akele reh sakti hain."
#
#  Phir Aggregation seekha: "Ek dabba hai, usme cheezein hain.
#  Dabba khatam ho, toh cheezein bach jaayengi." (School-Teacher)
#
#  Ab socho: Agar dabba aur uski cheezein itne jude hue hon ki
#  dono ek hi ho jaayein. Dabba toote, toh cheezein bhi toot jaayein.
#  Cheez ka bina dabbe koi matlab hi na ho.
#
#  Isko hi COMPOSITION kehte hain. Chalo bilkul aasan bhasha mein
#  samajhte hain.
#
# ============================================================
#
#  1. COMPOSITION kya hai? (Ekdam simple mein)
#  --------------------------------------------
#  Composition matlab: Ek cheez (whole) doosri cheezon (parts) se
#  bani hoti hai. Aur parts itni tightly judi hoti hain ki:
#    - Part bina whole ke exist hi nahi kar sakta.
#    - Whole khatam ho, toh parts bhi khatam ho jaate hain.
#    - Part doosre whole ke saath share nahi hota.
#
#  Simple mein:
#  > "Main tumhe banata hoon. Tum bina mere ke kuch nahi ho.
#  >  Main marunga toh tum bhi maroge. Tum kisi doosre ke nahi
#  >  ho sakte."
#
#  ------------------------------------------------------------
#  Real-Life Example — Ghar aur Kamra
#  ------------------------------------------------------------
#  Ek Ghar (House) socho. Usme Kamre (Rooms) hain — bedroom,
#  kitchen, living room.
#    - Kya kamra bina ghar ke exist kar sakta hai? NAHI.
#    - Agar ghar gir gaya, toh kya kamre bhi gir jaayenge? HAAN.
#    - Kya tum ek bedroom utha kar doosre ghar mein le ja sakte ho?
#      NAHI.
#
#  Kamra ghar ka "integral part" hai. Ghar aur kamre ek saath
#  zinda hain, ek saath marte hain. Yahi Composition hai.
#
#  ------------------------------------------------------------
#  Aur ek Example — Human Body aur Heart
#  ------------------------------------------------------------
#  - Heart body ke bina exist nahi kar sakta.
#  - Body mar jaaye, toh heart bhi kaam nahi karega.
#  - Heart doosre body ke saath share nahi hota (ek hi time pe).
#
#  Body aur Heart ka relationship = Composition.
#
# ============================================================
#
#  2. UML mein kaise dikhta hai?
#  ------------------------------
#  UML mein Composition ko "bhare hueere" (filled diamond) se
#  dikhate hain — "whole" side pe.
#
#  Ghar ◆---- Kamra
#  (whole)       (part)
#
#  Bhara hua heera = "meri poori maaliki hai, part mere saath hi
#  paida hua, mere saath hi mar jaayega."
#
#  Yaad rakho:
#  - Association = sirf line (---)
#  - Aggregation = khali heera (◇)
#  - Composition = bhara hua heera (◆)  [sabse strong]
#
# ============================================================
#
#  3. Code Example — Order aur LineItem
#  -------------------------------------
#  Socho tum Amazon pe order karte ho. Ek Order (whole) mein
#  bahut saare items (parts) hote hain — Mouse, Cable, etc.
#
#  Ab baat samajho:
#  - Order khud apne items ko banata hai (create karta hai).
#  - Item bina Order ke exist nahi kar sakta. "Mouse x2" ka kya
#    matlab hai agar koi Order hi nahi hai?
#  - Order delete ho jaaye, toh saare items bhi delete ho jaayenge.
#
#  class LineItem:
#      def __init__(self, product_name, quantity, unit_price):
#          self.product_name = product_name
#          self.quantity = quantity
#          self.unit_price = unit_price
#
#      def get_subtotal(self):
#          return self.quantity * self.unit_price
#
#  class Order:
#      def __init__(self, order_id):
#          self.order_id = order_id
#          self.line_items = []
#
#      def add_item(self, product, quantity, unit_price):
#          # YAHAN DHYAN DO: Order andar hi LineItem bana raha hai!
#          # Yeh "new LineItem(...)" andar ban raha hai, bahar se
#          # pass nahi hua. Yahi Composition ki pehchaan hai.
#          self.line_items.append(LineItem(product, quantity, unit_price))
#
#      def get_total(self):
#          return sum(item.get_subtotal() for item in self.line_items)
#
#  # Use karte hain:
#  order = Order("ORD-1001")
#  order.add_item("Wireless Mouse", 2, 29.99)
#  order.add_item("USB-C Cable", 3, 9.99)
#
#  print(f"Total: ${order.get_total():.2f}")
#
#  # Jab 'order' delete hoga, toh uske LineItems bhi delete ho jaayenge.
#  # Koi LineItem Order ke bina exist nahi karta.
#
#  ------------------------------------------------------------
#  Yeh Aggregation se kya alag hai?
#  ------------------------------------------------------------
#  Aggregation mein: Parts bahar bante the, phir whole mein pass
#  hote the. (Jaise Professors bahar bane, phir Department mein
#  gaye. Department band ho, Professors bache rahe.)
#
#  Composition mein: Parts andar hi bante hain. Order apna LineItem
#  khud banata hai. Order gaya, toh LineItem bhi gaya. Bacha nahi.
#
#  Bas yahi farq hai — parts khud banane (internal create) ya bahar
#  se lana (external pass).
#
# ============================================================
#
#  4. COMPOSITION kab use karein?
#  ------------------------------
#  In sawaalon ka jawab "HAAN" ho, toh Composition use karo:
#
#  1. Kya part bina whole ke bekar hai?
#     (LineItem bina Order ka koi kaam nahi. Kamra bina Ghar ke
#     bekar hai.) -> HAAN? Composition.
#
#  2. Kya whole part ko khud create aur destroy karta hai?
#     (Order apne LineItems khud banata hai.) -> HAAN? Composition.
#
#  3. Kya part doosre whole ke saath share nahi hota?
#     (Ek Kamra doosre Ghar mein nahi ja sakta.) -> HAAN? Composition.
#
#  ------------------------------------------------------------
#  "Favor Composition over Inheritance"
#  ------------------------------------------------------------
#  Yeh ek famous principle hai: "Inheritance se Composition behtar
#  hai." Kyun?
#
#  Inheritance mein problem: Ek baad ek jo class pe class inherit
#  karte jaate ho, toh hierarchy deep ho jaata hai. Code change
#  karna mushkil ho jaata hai. Ek change parent mein karo toh saare
#  children pe asar padta hai. Risky.
#
#  Composition mein fayda: Tum chhote-chhote parts ko jod kar badi
#  cheez bana sakte ho. Jaise lego blocks. Koi block change karo,
#  toh sirf wahi impact hota hai. Baaki safe rehte hain.
#
#  Example: Ek Vehicle class banaya. Usme Engine daala.
#  - Agar Engine PetrolEngine hai, toh petrol wali gaadi.
#  - Agar Engine ElectricEngine hai, toh electric gaadi.
#  - Runtime pe engine swap kar sakte ho. Flexible!
#  - Inheritance mein yeh itna aasaan nahi tha.
#
# ============================================================
#
#  5. Association vs Aggregation vs Composition (Aasan comparison)
#  ----------------------------------------------------------------
#
#  Teeno mein kya farq hai? Ek hi example se samajho:
#
#  Association (Loosest — koi maalik nahi):
#  - Student aur Teacher. Dono aapas mein connected hain.
#  - Par koi kisi ka maalik nahi. Dono akele bhi reh sakte hain.
#  - Teacher retire ho jaaye, toh student phir bhi padhai kar lega.
#
#  Aggregation (Medium — dabba wala):
#  - Playlist aur Song. Playlist gaano ka dabba hai.
#  - Playlist delete karo, toh gaane phir bhi rehte hain. (Dusri
#    playlist mein daal sakte ho.)
#  - Song akele bhi exist kar sakta hai. Maaliki kamzor hai.
#
#  Composition (Strongest — poori maaliki):
#  - Ghar aur Kamra. Kamra ghar ke bina exist nahi kar sakta.
#  - Ghar gir gaya, toh kamre bhi gir gaye.
#  - Kamra doosre ghar mein nahi ja sakta. Maaliki poori hai.
#
#  | Cheez        | Association  | Aggregation         | Composition             |
#  |--------------|--------------|---------------------|-------------------------|
#  | Maaliki      | Nahi         | Kamzor (has-a)      | Strong (owns-a)         |
#  | Lifecycle    | Independent  | Independent         | Dependent (saath marte) |
#  | UML Symbol   | Line (---)   | Khali Heera (◇)     | Bhara Heera (◆)         |
#  | Part ka kaam | Dono akele   | Part akele reh sakta| Part bina whole bekar   |
#  | Example      | Student-Teacher | Playlist-Song    | Ghar-Kamra, Order-Item  |
#
# ============================================================
#
#  SUMMARY — YAAD RAKHNA
#  ---------------------
#  - Composition = sabse strong "has-a" relationship. Whole part
#    ka poora maalik hai.
#  - Part bina whole ke exist nahi kar sakta. Whole mare, part bhi
#    mare.
#  - Parts share nahi hote — exclusive sirf ek whole ke liye.
#  - UML: bhara hua heera (◆) on the whole side.
#  - Code: Whole parts ko khud banata hai (andar "new" se), bahar
#    se receive nahi karta.
#  - Composition > Inheritance (flexibility ke liye).
#
#  Ek line mein: Composition = "Main tumhe banata hoon, mere bina
#  tumhari koi zindagi nahi. Main marunga toh tum bhi maroge. Tum
#  kisi doosre ke nahi ho sakte."
#
#  Example: Ghar aur Kamra. Ghar gira, toh kamre bhi gaye. Kamre
#  ka bina ghar ke koi matlab nahi.
#
# ============================================================
```
