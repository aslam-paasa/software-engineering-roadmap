```python
"""
BRIDGE DESIGN PATTERN — KYA HAI?
=====================================

Bridge ek STRUCTURAL design pattern hai jo ek "ABSTRACTION" ko
uski "IMPLEMENTATION" se ALAG (SEPARATE) karta hai, taaki DONO
INDEPENDENTLY (ek doosre se bina asar liye) BADAL/EXTEND ho sakein.

Bridge Pattern mein 2 HISSE hote hain:
    1. ABSTRACTION       -> "KYA karna hai" (high-level concept)
    2. IMPLEMENTATION      -> "KAISE karna hai" (low-level details)

SOCHO EK EXAMPLE:
--------------------
Ek Remote Control system hai jo DIFFERENT devices (TV, Light)
CONTROL kar sakta hai. Agar hum HAR remote-device COMBINATION ke
liye ALAG CLASS banayein (TVRemote, LightRemote, AdvancedTVRemote,
AdvancedLightRemote...), to COMBINATIONS EXPONENTIALLY badh
jaayengi (Decorator/Composite wali "class explosion" jaisi
problem)!

Bridge isko solve karta hai:
    -> "Remote" (ABSTRACTION) ANDAR EK "Device" (IMPLEMENTATION)
       ka REFERENCE rakhta hai
    -> Remote apne actions (turn_on, turn_off) ko Device ko
       DELEGATE (forward) kar deta hai
    -> Naya Remote TYPE ya naya Device TYPE, DONO INDEPENDENTLY
       add kar sakte hain -- BINA EK DOOSRE ko CHHUE

MAIN COMPONENTS:
--------------------
1. Abstraction          -> CORE class jo Implementer ka REFERENCE
                          rakhti hai (e.g. Vehicle)
2. Refined Abstraction    -> Abstraction ko EXTEND karti hai,
                          FINER details ADD karti hai (e.g. Car, Bike)
3. Implementer               -> Interface jo Implementation
                          classes DEFINE karta hai (e.g. Workshop)
4. Concrete Implementation      -> Implementer ka ACTUAL kaam
                          karne wali class (e.g. Produce, Assemble)

KAISE KAAM KARTA HAI:
--------------------------
1. Abstraction ke ANDAR Implementer ka REFERENCE hota hai
2. Concrete Implementations ACTUAL behavior dete hain
3. Client SIRF Abstraction se INTERACT karta hai, Implementation
   se SEEDHA NAHI

REAL LIFE EXAMPLES:
-----------------------
- Remote Control aur Devices: EK universal remote MULTIPLE
  devices (TV, Radio, AC) control kar sakta hai
- Shapes aur Colors: EK Shape (Circle/Rectangle) DIFFERENT Color
  implementations (Red/Blue) use kar sakta hai

FAYDE (ADVANTAGES):
------------------------
- ABSTRACTION aur IMPLEMENTATION ke beech TIGHT COUPLING KAM
  hoti hai
- DONO ko INDEPENDENTLY EXTEND kiya ja sakta hai
- System ki MAINTAINABILITY aur SCALABILITY IMPROVE hoti hai

NUKSAAN (DISADVANTAGES):
----------------------------
- EXTRA interfaces/classes ki wajah se INITIAL DESIGN thodi
  COMPLEX ho jaati hai
- SIMPLE applications ke liye ZAROORAT SE ZYADA ho sakta hai
- BEGINNERS ke liye code SAMAJHNA THODA MUSHKIL ho sakta hai
"""
print("Concept samajh liya! Ab neeche code mein implement karke dekhte hain.")
```

## ❌ Pehle dekho: BINA Bridge Pattern ke (Problem Code)

Chalo pehle ye dekhte hain ki Bridge Pattern KE BINA log alag-alag Vehicle × Workshop task ke
liye classes kaise banate hain (jaise document ke "Without Bridge Design Pattern" section mein
`Bus` example diya gaya tha), aur usme kya dikkat aati hai.

```python
"""
BINA BRIDGE KE — PROBLEM DEMO (SUBCLASSING SE)
----------------------------------------------------
Yahan HAR "Vehicle + Workshop-task" COMBINATION ke liye ALAG
SUBCLASS bana rahe hain -- jaise document mein "Bus" ke liye
"ProduceBus" aur "AssembleBus" dikhaye gaye the.
"""

class CarBad:
    pass


class BikeBad:
    pass


# HAR vehicle ke liye, HAR workshop-task ke liye ALAG SUBCLASS!
class ProduceCarBad(CarBad):
    def work(self):
        print("Car Produced", end="")


class AssembleCarBad(CarBad):
    def work(self):
        print(" And Assembled.")


class ProduceBikeBad(BikeBad):
    def work(self):
        print("Bike Produced", end="")


class AssembleBikeBad(BikeBad):
    def work(self):
        print(" And Assembled.")


# Client ko HAR combination ke liye ALAG classes yaad rakhni padti hain
ProduceCarBad().work()
AssembleCarBad().work()
print()
ProduceBikeBad().work()
AssembleBikeBad().work()
```

```python
"""
ISME PROBLEM KYA THI?
-------------------------
1. CLASS EXPLOSION: Sirf 2 vehicles (Car, Bike) aur 2 workshop
   tasks (Produce, Assemble) ke liye BHI 4 classes bani
   (ProduceCarBad, AssembleCarBad, ProduceBikeBad, AssembleBikeBad).
   Agar EK NAYA vehicle (Bus) aur EK NAYA task (QualityCheck) add
   ho, to COMBINATIONS 3x3 = 9 ho jaayenge -- YE PATTERN
   EXPONENTIALLY BADHTA hai.

2. TIGHT COUPLING: "ProduceCarBad" SEEDHA "CarBad" ko EXTEND
   karta hai -- Vehicle aur Workshop-task PERMANENTLY BOUND hain.
   Agar "Produce" LOGIC change karni ho (jaise SABHI vehicles ke
   liye), to HAR "ProduceXBad" class mein JAAKAR update karna
   padega.

3. HARD TO MAINTAIN: Document mein bataya gaya jaisa -- agar
   "Bus" class change karni ho, to "ProduceBus" aur "AssembleBus"
   BHI change karne pad sakte hain, aur agar change WORKSHOP-SPECIFIC
   hai, to "Bike" ki classes BHI CHHUNI PAD SAKTI HAIN!

4. NOT INDEPENDENT: Vehicle TYPES aur Workshop TASKS
   INDEPENDENTLY BADLE NAHI ja sakte -- DONO EK DOOSRE se TIGHTLY
   JUDE hain is INHERITANCE-BASED approach mein.

YE SAB PROBLEMS "Bridge Design Pattern" SOLVE KARTA HAI!
Neeche BEST APPROACH dekhte hain -- Vehicle (Abstraction) aur
Workshop (Implementation) ko ALAG HIERARCHIES mein rakhenge, aur
EK DOOSRE se "BRIDGE" (compose) karenge.
"""
print("Chalo ab BEST APPROACH (Bridge Pattern) dekhte hain...")
```

## ✅ Ab dekho: BEST APPROACH — Bridge Pattern se solve karte hain

### 1️⃣ Implementer — `Workshop`

```python
"""
IMPLEMENTER (Workshop)
----------------------------
Ye IMPLEMENTATION ka ALAG HIERARCHY hai. Isme SIRF EK abstract
method hai: work(). Ye Vehicle (Abstraction) se BILKUL
INDEPENDENT hai.
"""

from abc import ABC, abstractmethod

class Workshop(ABC):
    @abstractmethod
    def work(self):
        pass
```

### 2️⃣ Concrete Implementations — `Produce`, `Assemble`

```python
"""
CONCRETE IMPLEMENTATIONS (Produce, Assemble)
----------------------------------------------------
Ye ACTUAL workshop-task classes hain. In classes ko ye BILKUL
PATA NAHI ki inhe "Car" use karega ya "Bike" -- ye BAS apna KAAM
karti hain.
"""

class Produce(Workshop):
    def work(self):
        print("Produced", end="")


class Assemble(Workshop):
    def work(self):
        print(" And", end="")
        print(" Assembled.")
```

### 3️⃣ Abstraction — `Vehicle`

```python
"""
ABSTRACTION (Vehicle)
--------------------------
Ye ABSTRACTION ka ALAG HIERARCHY hai. Isme DO Workshop
(Implementer) REFERENCES store hote hain -- YE HI "BRIDGE" HAI!
Vehicle KHUD "work" NAHI karta, ye kaam Workshop OBJECTS ko
DELEGATE kar deta hai.
"""

class Vehicle(ABC):
    def __init__(self, workshop1: Workshop, workshop2: Workshop):
        self.workshop1 = workshop1   # <-- YE "BRIDGE" (composition) HAI
        self.workshop2 = workshop2

    @abstractmethod
    def manufacture(self):
        pass
```

### 4️⃣ Refined Abstractions — `Car`, `Bike`

```python
"""
REFINED ABSTRACTIONS (Car, Bike)
--------------------------------------
Ye Vehicle ko EXTEND karti hain, aur apna "manufacture()" method
DEFINE karti hain -- ANDAR SE workshop1 aur workshop2 (jo KOI
BHI Workshop implementation ho sakti hai) ko "work()" call kar
DELEGATE kar deti hain.
"""

class Car(Vehicle):
    def manufacture(self):
        print("Car", end=" ")
        self.workshop1.work()
        self.workshop2.work()


class Bike(Vehicle):
    def manufacture(self):
        print("Bike", end=" ")
        self.workshop1.work()
        self.workshop2.work()
```

### 5️⃣ Client — Vehicle aur Workshop, DONO ko MIX-AND-MATCH karta hai

```python
"""
CLIENT CODE
---------------
Ab dekho -- Car aur Bike, DONO ko SAME "Produce"/"Assemble"
implementations ke saath USE kar sakte hain, BINA "ProduceCar",
"ProduceBike" jaisi ALAG classes banaye!
"""

vehicle1 = Car(Produce(), Assemble())
vehicle1.manufacture()

print()

vehicle2 = Bike(Produce(), Assemble())
vehicle2.manufacture()

print("\n\n✅ Bridge Pattern se Car aur Bike (Abstraction) SAME Produce/Assemble (Implementation) ko REUSE kar rahe hain -- naye combinations ke liye NAYI classes NAHI banani padin.")
```

ko control kar sakta hai, bina HAR combination ke liye ALAG class banaye.

Tumhe ye banana hai (Vehicle/Workshop example jaisi hi structure use karo):

1. `Device` (Implementer interface) — abstract methods `turn_on()` aur `turn_off()`
2. Concrete Implementations: `TV` (print `"📺 TV is ON"` / `"📺 TV is OFF"`), `Light` (print `"💡 Light is ON"` / `"💡 Light is OFF"`)
3. `RemoteControl` (Abstraction) — constructor mein `device: Device` STORE kare; abstract method `toggle_power()`
4. `BasicRemote` (Refined Abstraction) — `toggle_power()` mein `device.turn_on()` call kare
5. Client code likho jo:

   - Ek AUR `BasicRemote` banaye jo `Light` ko control kare, aur `toggle_power()` call kare
   - Dikhao ki SAME `BasicRemote` class, DIFFERENT devices ke saath kaam kar rahi hai

## ✅ Solution — Practice Question ka Coding Solution

```python
"""
SOLUTION EXPLANATION
-----------------------
Yahan bilkul Vehicle/Workshop jaisi hi structure use ki hai, bas
ab "Remote Control System" bana rahe hain instead of "Vehicle
Manufacturing".

1. Device            -> Implementer
2. TV, Light            -> Concrete Implementations
3. RemoteControl           -> Abstraction
4. BasicRemote                -> Refined Abstraction
5. Neeche client code hai jo SAME remote class, ALAG devices ke saath use karta hai
"""

from abc import ABC, abstractmethod

# ---- 1. Implementer ----
class Device(ABC):
    @abstractmethod
    def turn_on(self):
        pass

    @abstractmethod
    def turn_off(self):
        pass


# ---- 2. Concrete Implementations ----
class TV(Device):
    def turn_on(self):
        print("📺 TV is ON")

    def turn_off(self):
        print("📺 TV is OFF")


class Light(Device):
    def turn_on(self):
        print("💡 Light is ON")

    def turn_off(self):
        print("💡 Light is OFF")


# ---- 3. Abstraction ----
class RemoteControl(ABC):
    def __init__(self, device: Device):
        self.device = device   # <-- BRIDGE

    @abstractmethod
    def toggle_power(self):
        pass


# ---- 4. Refined Abstraction ----
class BasicRemote(RemoteControl):
    def toggle_power(self):
        self.device.turn_on()


# ---- 5. Client Code ----
tv_remote = BasicRemote(TV())
tv_remote.toggle_power()

light_remote = BasicRemote(Light())
light_remote.toggle_power()

print("\n✅ SAME 'BasicRemote' class ne DONO alag devices (TV, Light) control kiye -- koi 'TVRemote', 'LightRemote' banane ki zaroorat nahi padi.")
```
