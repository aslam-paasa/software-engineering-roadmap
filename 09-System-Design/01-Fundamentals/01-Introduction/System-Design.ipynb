```python
/**
 * SYSTEM DESIGN - INTRODUCTION
 * - System Design ka basic concept samajhna aur interview mein kisi bhi system design
 *   problem ko systematically approach karna seekhna.
*/
```

```python
/**
 * 1. WHAT IS SYSTEM DESIGN?
 *    - "Software system ko kis tarah structure karein taaki woh given requirements
 *       aur expected scale par properly kaam kare."
 *    - Simple words mein:
 *      > Requirements
 *      > Architecture
 *      > Components + Data Flow
 *      > Scalable + Reliable System
 *
 *    - System Design sirf boxes aur arrows draw karna nahi hai.
 *    - Hume explain karna hota hai:
 *      - System kya karega?
 *      - Data kaise flow karega?
 *      - Kaunsa component kya responsibility handle karega?
 *      - System kaha slow ho sakta hai?
 *      - System kaha fail ho sakta hai?
 *      - Scale badhne par kya problem hogi?
 *      - Humne particular technology/architecture kyun choose ki?
 *      - Us choice ka trade-off kya hai?
*/
```

```python

/**
 * 2. SMALL SYSTEM VS LARGE SYSTEM
 *    - Small application:
 *
 *       Client
 *         |
 *         v
 *       Server
 *         |
 *         v
 *       Database
 *
 *    - Agar traffic, users aur data increase ho jaye, to additional components ki 
 *      zaroorat pad sakti hai:
 *      - Load Balancer
 *      - Cache
 *      - Message Queue
 *      - Database Replicas
 *      - Partitioning / Sharding
 *      - Monitoring
 *      - Failure Handling
 *
 *    - Important:
 *      - Har system mein ye saare components add karna zaroori nahi hai.
 *      - Rule: "Use the simplest architecture that satisfies the requirements."
*/
```

```python
/**
 * 3. SYSTEM DESIGN = TRADE-OFFS
 *    - System Design mein generally ek perfect solution nahi hota.
 *    - Hume requirements aur constraints ke according trade-offs make karne padte hain.
 *
 *    Example 1: Low Latency
 *    - Agar users ko very fast response chahiye: Cache use kar sakte hain.
 *    - Ya users ke closer region mein data serve kar sakte hain.
 *
 *    Example 2: High Availability
 *    - Agar system failure ke time bhi available rehna chahiye:
 *      Redundancy + Failover use kar sakte hain.
 *
 *    Example 3: High Write Traffic
 *    - Agar ek database high write traffic handle nahi kar pa raha:
 *      Partitioning / Sharding consider kar sakte hain.
 *
 *    Important interview mindset:
 *    - "Technology se start mat karo. Requirement se start karo."
*/

/**
 * 4. DON'T START WITH TECHNOLOGY
 *    - Bad approach:
 *      "Kafka use karenge."
 *      "Redis use karenge."
 *      "Kubernetes use karenge."
 *      "Cassandra use karenge."
 *
 *    - Problem:
 *      Pehle technology decide kar di aur problem baad mein solve karne ki koshish
 *      kar rahe hain.
 *
 *    - Better approach:
 *      > Requirement: High write volume + Traffic spikes + Asynchronous processing
 *      > Then reason: Queue / Event Log useful ho sakta hai.
 *
 *    - Interview mein always explain:
 *      > Requirement
 *      > Problem
 *      > Component
 *      > Technology
*/

/**
 * 5. SYSTEM DESIGN KE MAIN QUESTIONS
 *    Kisi bhi system design problem mein ye questions systematically ask karo:
 *
 *    1. REQUIREMENTS
 *       - System ko kya karna hai?
 *       - Aur kya out of scope hai?
 *
 *    2. SCALE
 *       - Kitne: Users?
 *       - Kitne: Requests/second?
 *       - Kitne: Reads?
 *       - Kitne: Writes?
 *       - Kitne: Data?
 *
 *    3. DATA
 *       - Kya data store karna hai?
 *       - Kahan store karna hai?
 *       - Kaise access karna hai?
 *
 * 4. APIs & COMMUNICATION
 *    - Client aur server kaise communicate karenge?
 *    - Services ek doosre se kaise communicate karengi?
 *
 * 5. RELIABILITY
 *    - Agar: Server down ho jaye?
 *    - Agar: Database down ho jaye?
 *    - Agar: Network fail ho jaye?
 *    - Agar: External API fail ho jaye?
 *    > To system ka behavior kya hoga?
 *
 *
 * 6. PERFORMANCE
 *    - Required latency kya hai?
 *    - Required throughput kya hai?
 *
 * 7. CONSISTENCY
 *    - Kaunsa data immediately correct hona chahiye?
 *    - Aur kaha eventual consistency acceptable hai?
 *
 * 8. OPERATIONS
 *    - System ko: Monitor kaise karenge?
 *    - System ko: Debug kaise karenge?
 *    - System ko: Deploy kaise karenge?
 *    - System ko: Future mein evolve kaise karenge?
 *
 * 9. COST
 *    - Expected traffic aur business scale ke according architecture financially
 *      reasonable hai ya nahi?
*/

/**
 * 6. REQUIREMENTS ARCHITECTURE DECIDE KARTI HAIN
 *    - Same product ke liye different scale par different architecture ho sakti hai.
 *    - Example  : Small internal chat application    != Global messaging platform
 *    - Similarly: One-city food delivery application != Global marketplace
 *    - Isliye:
 *      > Requirements
 *          |
 *          v
 *      > Scale
 *          |
 *          v
 *      > Architecture
*/

/**
 * 7. BASIC SYSTEM DESIGN BUILDING BLOCKS
 *    - Typical system mein ye components ho sakte hain:
 *
 *       Client
 *         |
 *         v
 *       API Layer
 *         |
 *         v
 *       Load Balancer
 *         |
 *         v
 *       Application Servers
 *         |
 *         +--------> Cache
 *         |
 *         +--------> Database / Storage
 *         |
 *         +--------> Message Queue
 *         |             |
 *         |             v
 *         |           Worker
 *         |
 *         +--------> External APIs
 *         |
 *         v
 *       Observability
 *
 *
 *    - PDF ke architecture diagram mein Client, API Layer, Load Balancer, App Servers,
 *      Cache, Database/Storage, Message Queue, Worker, External APIs aur 
 *      Observability ko connected flow mein dikhaya gaya hai. 
 *      :contentReference[oaicite:1]{index=1}
*/

/**
 * 8. CLIENT
 *    - Client wo component hai jo system ko request bhejta hai.
 *    - Examples:
 *      - Web application
 *      - Mobile application
 *      - Browser
 *      - Another service
 *
 *    - Example:
 *
 *        Mobile App
 *             |
 *             v
 *        GET /users
*/

/**
 * 9. API LAYER
 *    - API layer clients aur backend system ke beech interface provide karti hai.
 *    - Example:
 *
 *        Client
 *          |
 *          v
 *         API
 *          |
 *          v
 *    Application Service
 *
 *    - API layer request receive karke appropriate backend functionality tak request
 *     pahucha sakti hai.
*/

/**
 * 10. LOAD BALANCER
 *     - Load Balancer incoming traffic ko multiple application servers ke beech 
 *       distribute karta hai.
 *     - Example:
 *
 *                 Load Balancer
 *                  /     |     \
 *                 /      |      \
 *                v       v       v
 *             Server1 Server2 Server3
 *
 *
 *     - Benefit:
 *       - Traffic distribute hota hai.
 *       - Ek server par excessive load avoid hota hai.
 *       - Multiple servers ke through availability improve ki ja sakti hai.
*/

/**
 * 11. APPLICATION SERVER
 *     - Application server business logic execute karta hai.
 *     - Responsibilities:
 *       - Business rules
 *       - Database reads
 *       - Database writes
 *       - Workflows
 *       - Request processing
 *
 *     - Example:
 *
 *        Request
 *          |
 *          v
 *        App Server
 *          |
 *          +----> Database
 *          |
 *          +----> Cache
 *          |
 *          +----> Queue
*/

/**
 * 12. CACHE
 *     - Cache frequently accessed data ko temporarily store karta hai.
 *     - Main purpose: Faster response + Database load reduce karna
 *     - Example:
 *
 *       a. Without cache:
 *           Client
 *             |
 *             v
 *           Server
 *             |
 *             v
 *           Database
 *
 *       b. With cache:
 *
 *           Client
 *             |
 *             v
 *           Server
 *             |
 *             v
 *           Cache
 *             |
 *             +---- HIT ----> Return data
 *             |
 *             +---- MISS ---> Database
 *
 *      Important trade-off:
 *      - Cache latency reduce kar sakta hai, lekin cache invalidation problem 
 *        introduce karta hai.
*/

/**
 * 13. DATABASE / STORAGE
 *     - Database ya storage system ka durable data store karta hai.
 *     - Example:
 *       - Users
 *       - Orders
 *       - Payments
 *       - Products
 *     - Cache ke opposite: Database generally durable/persistent storage provide 
 *       karta hai.
*/

/**
 *
 * ============================================================
 * 14. MESSAGE QUEUE / EVENT LOG
 * ============================================================
 *
 * Message Queue producers aur consumers ko decouple karta hai.
 *
 *
 * Producer
 *    |
 *    v
 * Queue
 *    |
 *    v
 * Consumer / Worker
 *
 *
 * Useful when:
 *
 * - Work asynchronous ho sakta hai.
 * - Traffic spikes ko buffer karna hai.
 * - Producer aur consumer ko independently scale karna hai.
 *
 *
 * Example:
 *
 * User places order
 *       |
 *       v
 * Order Service
 *       |
 *       v
 * Message Queue
 *       |
 *       v
 * Email Worker
 *
 *
 * User ko email send hone ka wait nahi karna padega.
 *
 *
 * Important trade-off:
 *
 * Asynchronous processing traffic spikes absorb kar sakta hai,
 * lekin user ko result immediately na mile.
 *
 *
 * ============================================================
 * 15. WORKER
 * ============================================================
 *
 * Worker background/asynchronous tasks process karta hai.
 *
 * Example:
 *
 * Queue
 *   |
 *   v
 * Worker
 *   |
 *   +----> Send Email
 *   +----> Generate Report
 *   +----> Process Image
 *   +----> Send Notification
 *
 *
 * ============================================================
 * 16. EXTERNAL SERVICES
 * ============================================================
 *
 * Application kabhi-kabhi third-party services par depend
 * karti hai.
 *
 * Examples:
 *
 * - Payment provider
 * - Email provider
 * - SMS provider
 * - Identity provider
 * - Analytics service
 *
 *
 * Problem:
 *
 * External service slow ya unavailable ho sakti hai.
 *
 * Isliye system design mein external dependency failure
 * ko consider karna important hai.
 *
 *
 * ============================================================
 * 17. OBSERVABILITY
 * ============================================================
 *
 * Observability ka purpose hai production system ke behavior
 * ko understand karna.
 *
 * Main components:
 *
 *     Logs
 *     Metrics
 *     Traces
 *     Alerts
 *     Dashboards
 *
 *
 * Example:
 *
 * Application
 *     |
 *     +----> Logs
 *     |
 *     +----> Metrics
 *     |
 *     +----> Traces
 *              |
 *              v
 *        Observability System
 *
 *
 * Tumhare microservices project mein jo Winston,
 * Prometheus aur Grafana jaise concepts hain, woh isi
 * observability area se related hain.
 *
 *
 * ============================================================
 * 18. SYSTEM DESIGN IS ABOUT DATA FLOW
 * ============================================================
 *
 * Sirf components list karna system design nahi hai.
 *
 * Important question:
 *
 *     "Request aur data system ke andar kaise flow karte hain?"
 *
 *
 * Example: User Feed
 *
 * User
 *   |
 *   v
 * API
 *   |
 *   v
 * Cache?
 *   |
 *   +---- HIT ----> Return feed
 *   |
 *   +---- MISS
 *          |
 *          v
 *       Database
 *
 *
 * Further questions:
 *
 * - Cache miss par kya hoga?
 * - Feed database se kaise fetch hoga?
 * - Feed read par calculate hoga?
 * - Ya write time par precompute hoga?
 * - Ya hybrid approach hogi?
 *
 * Ye actual system design decisions hain.
 *
 *
 * ============================================================
 * 19. HOW TO APPROACH A SYSTEM DESIGN PROBLEM
 * ============================================================
 *
 * PDF ke according basic approach 5 steps mein divide hota hai:
 *
 *
 * Step 1:
 *     Clarify Requirements
 *
 *        |
 *        v
 *
 * Step 2:
 *     Estimate Scale
 *
 *        |
 *        v
 *
 * Step 3:
 *     Start With Simple Architecture
 *
 *        |
 *        v
 *
 * Step 4:
 *     Identify Bottlenecks & Failure Modes
 *
 *        |
 *        v
 *
 * Step 5:
 *     Explain Trade-offs
 *
 *
 * Architecture ko zaroorat padne par revise karte hain.
 *
 *
 * ============================================================
 * STEP 1: CLARIFY REQUIREMENTS
 * ============================================================
 *
 * Architecture choose karne se pehle problem ko samjho.
 *
 * Questions:
 *
 * 1. Core features kya hain?
 *
 * 2. Users kaun hain?
 *
 * 3. Kya out of scope hai?
 *
 * 4. Expected scale kya hai?
 *
 * 5. Kaunsi operations ko low latency chahiye?
 *
 * 6. Kaunsi operations ko strong correctness chahiye?
 *
 * 7. Availability/durability expectations kya hain?
 *
 *
 * Important:
 *
 * Requirements architecture decide karti hain.
 *
 *
 * Example:
 *
 * "Users timeline dekh sakte hain"
 *
 * Ye requirement alone enough nahi hai.
 *
 * 10,000 daily users
 *        vs
 * 500 million daily users
 *
 * dono ke liye same architecture necessarily suitable nahi hogi.
 *
 * Agar 500M users ke saath 200ms latency target hai,
 * to architecture par significantly different pressure aayega.
 *
 *
 * ============================================================
 * STEP 2: ESTIMATE SCALE
 * ============================================================
 *
 * Exact calculation se zyada important hai approximate
 * numbers se design pressure identify karna.
 *
 * Estimate:
 *
 * - Read requests/second
 * - Write requests/second
 * - Storage growth
 * - Bandwidth
 * - Peak traffic
 * - Hot keys
 * - Uneven access patterns
 *
 *
 * Example:
 *
 * System A:
 *
 *     100 writes/sec
 *
 * System B:
 *
 *     1,000,000 writes/sec
 *
 * Dono systems ke bottlenecks same nahi honge.
 *
 *
 * Scale estimation ka purpose:
 *
 *     "Wrong architecture design karne se bachna."
 *
 *
 * ============================================================
 * STEP 3: START WITH SIMPLE ARCHITECTURE
 * ============================================================
 *
 * Starting architecture:
 *
 *     Client
 *       |
 *       v
 *     API Layer
 *       |
 *       v
 *     Application Service
 *       |
 *       v
 *     Database
 *
 *
 * Initially unnecessarily:
 *
 *     Kafka
 *     Redis
 *     Kubernetes
 *     Sharding
 *     Multi-region
 *
 * add mat karo.
 *
 * Pehle simplest working architecture banao.
 *
 *
 * Then requirement ke according components add karo.
 *
 *
 * ============================================================
 * WHEN TO ADD WHICH COMPONENT?
 * ============================================================
 *
 *
 * Cache:
 *
 * Add when:
 *
 * - Repeated reads database ko overload kar rahe hain.
 * - Latency high hai.
 *
 *
 * Queue:
 *
 * Add when:
 *
 * - Work asynchronous ho sakta hai.
 * - Traffic spikes ko buffer karna hai.
 *
 *
 * Replicas:
 *
 * Add when:
 *
 * - Read traffic increase ho raha hai.
 * - Availability ke liye redundancy chahiye.
 *
 *
 * Partitioning / Sharding:
 *
 * Add when:
 *
 * - Single machine data hold/serve nahi kar pa rahi.
 *
 *
 * Regional Deployment:
 *
 * Add when:
 *
 * - Users geographically far hain.
 * - Availability requirements demand karti hain.
 *
 *
 * Important:
 *
 * Real systems ek saath fully complex nahi bante.
 *
 * They evolve:
 *
 *     Simple
 *       |
 *       v
 *     More Traffic
 *       |
 *       v
 *     Bottleneck
 *       |
 *       v
 *     New Component
 *       |
 *       v
 *     New Bottleneck
 *       |
 *       v
 *     Further Improvement
 *
 *
 * ============================================================
 * STEP 4: IDENTIFY BOTTLENECKS & FAILURE MODES
 * ============================================================
 *
 * Baseline architecture banane ke baad weak points identify karo.
 *
 * Ask:
 *
 * What if database goes down?
 *
 * What if traffic suddenly becomes 10x?
 *
 * What if cache is empty?
 *
 * What if queue consumer is slow?
 *
 * What if third-party API becomes unavailable?
 *
 * What if third-party API becomes very slow?
 *
 *
 * Also ask:
 *
 * Which data can be stale?
 *
 * Which data must be immediately correct?
 *
 *
 * Example:
 *
 * Product recommendations:
 *     Slightly stale data may be acceptable.
 *
 * Bank balance:
 *     Incorrect/stale data may be unacceptable.
 *
 *
 * ============================================================
 * STEP 5: EXPLAIN TRADE-OFFS
 * ============================================================
 *
 * Har architecture decision ka benefit aur cost dono hote hain.
 *
 *
 * CACHE
 *
 * Benefit:
 *     Lower latency
 *     Lower DB load
 *
 * Cost:
 *     Cache invalidation complexity
 *
 *
 * REPLICATION
 *
 * Benefit:
 *     Better availability
 *     Better read throughput
 *
 * Cost:
 *     Consistency lag
 *
 *
 * SHARDING
 *
 * Benefit:
 *     Higher write/data capacity
 *
 * Cost:
 *     Queries and operations become more complex
 *
 *
 * ASYNCHRONOUS PROCESSING
 *
 * Benefit:
 *     Traffic spikes absorb kar sakta hai
 *
 * Cost:
 *     Result immediately visible nahi ho sakta
 *
 *
 * Interview mein sirf ye mat bolo:
 *
 *     "I will use Redis."
 *
 * Instead:
 *
 *     "Repeated reads are putting pressure on the database,
 *      so I would introduce a cache. This reduces latency
 *      and database load, but introduces cache invalidation
 *      complexity."
 *
 *
 * ============================================================
 * 20. SYSTEM DESIGN IN INTERVIEW VS REAL LIFE
 * ============================================================
 *
 * REAL LIFE:
 *
 * System design ek continuous process hai.
 *
 *     Design
 *       |
 *       v
 *     Build
 *       |
 *       v
 *     Measure
 *       |
 *       v
 *     Learn
 *       |
 *       v
 *     Revise
 *
 *
 * Real systems mein additional constraints bhi hote hain:
 *
 * - Legacy code
 * - Team ownership
 * - Budget
 * - Existing migrations
 * - Compliance
 * - Operational history
 *
 *
 * INTERVIEW:
 *
 * Usually limited time hota hai, around 45-60 minutes.
 *
 * Interview mein:
 *
 * 1. Scope clarify karo.
 * 2. Requirements samjho.
 * 3. Scale estimate karo.
 * 4. Basic architecture banao.
 * 5. Bottlenecks identify karo.
 * 6. Failure handling explain karo.
 * 7. Trade-offs explain karo.
 *
 *
 * ============================================================
 * 21. SYSTEM DESIGN INTERVIEW MINDSET
 * ============================================================
 *
 * Interviewer ko ye prove nahi karna ki:
 *
 *     "Mujhe 20 technologies pata hain."
 *
 * Instead prove karo ki:
 *
 *     "Main requirements ko architecture decisions mein
 *      convert kar sakta hoon."
 *
 *
 * Har component ke liye 3 questions:
 *
 *     WHY?
 *       |
 *       v
 *     Ye component kyun chahiye?
 *
 *     WHAT PROBLEM?
 *       |
 *       v
 *     Kaunsi problem solve karta hai?
 *
 *     WHAT TRADE-OFF?
 *       |
 *       v
 *     Iske saath new problem kya aayegi?
 *
 *
 * ============================================================
 * 22. INTERVIEW QUESTIONS & ANSWERS
 * ============================================================
 *
 *
 * Q1. What is System Design?
 *
 * Answer:
 * System Design is the process of deciding how a software
 * system should be structured so that it can satisfy its
 * functional and non-functional requirements at the expected
 * scale.
 *
 *
 * Q2. Is System Design just drawing architecture diagrams?
 *
 * Answer:
 * Nahi.
 *
 * Diagram sirf ek part hai.
 *
 * System design mein requirements, data flow, components,
 * scalability, reliability, performance, failures aur
 * trade-offs explain karne hote hain.
 *
 *
 * Q3. What is the most important starting point in System Design?
 *
 * Answer:
 * Requirements clarify karna.
 *
 * Architecture directly technology se start nahi karni chahiye.
 *
 *
 * Q4. Why should we not start with technologies?
 *
 * Answer:
 * Kyunki technology problem ke according choose honi chahiye.
 *
 * Pehle requirements identify karte hain, phir problem,
 * phir suitable component aur finally technology.
 *
 *
 * Q5. What are the main questions you ask before designing?
 *
 * Answer:
 *
 * - Core features?
 * - Users?
 * - Out of scope?
 * - Scale?
 * - Latency requirements?
 * - Correctness requirements?
 * - Availability?
 * - Durability?
 *
 *
 * Q6. Why do we estimate scale?
 *
 * Answer:
 * Scale estimation se design pressure identify hota hai.
 * 100 requests/sec aur 1 million requests/sec ke systems
 * ke bottlenecks different honge.
 *
 *
 * Q7. Why start with a simple architecture?
 *
 * Answer:
 * Unnecessary complexity avoid karne ke liye.
 * Components tab add karne chahiye jab requirements unki
 * need justify karein.
 *
 *
 * Q8. When would you introduce a cache?
 *
 * Answer:
 * Jab repeated reads database par load create kar rahe hon
 * ya latency reduce karni ho.
 *
 *
 * Q9. When would you introduce a queue?
 *
 * Answer:
 * Jab work asynchronous ho sakta ho ya traffic spikes ko
 * buffer karna ho.
 *
 *
 * Q10. What is a load balancer?
 *
 * Answer:
 * Load balancer incoming traffic ko multiple servers ke
 * beech distribute karta hai.
 *
 *
 * Q11. What is the purpose of a message queue?
 *
 * Answer:
 * Producer aur consumer ko decouple karna, asynchronous
 * processing enable karna aur traffic spikes ko buffer karna.
 *
 *
 * Q12. What is observability?
 *
 * Answer:
 * Production system ke behavior ko understand karne ke liye
 * logs, metrics, traces, alerts aur dashboards ka use
 * observability kehlata hai.
 *
 *
 * Q13. What is a bottleneck?
 *
 * Answer:
 * System ka wo component jo overall system performance ya
 * scalability ko limit kar raha ho.
 *
 * Example:
 *
 *     Very high DB load
 *          |
 *          v
 *     Database becomes bottleneck
 *
 *
 * Q14. What is a failure mode?
 *
 * Answer:
 * Aisi situation jahan system ka koi component fail,
 * slow ya unavailable ho jata hai.
 *
 * Examples:
 *
 * - Database down
 * - Cache unavailable
 * - Queue consumer slow
 * - External API unavailable
 *
 *
 * Q15. Why are trade-offs important in System Design?
 *
 * Answer:
 * Kyunki har design decision ka benefit ke saath ek cost
 * ya new problem hoti hai.
 *
 *
 * ============================================================
 * 23. MOST IMPORTANT INTERVIEW FRAMEWORK
 * ============================================================
 *
 * Kisi bhi System Design question mein ye sequence yaad rakho:
 *
 *
 *     1. REQUIREMENTS
 *            |
 *            v
 *     2. SCALE
 *            |
 *            v
 *     3. SIMPLE ARCHITECTURE
 *            |
 *            v
 *     4. BOTTLENECKS
 *            |
 *            v
 *     5. FAILURE MODES
 *            |
 *            v
 *     6. TRADE-OFFS
 *            |
 *            v
 *     7. REVISE ARCHITECTURE
 *
 *
 * ============================================================
 * 24. QUICK REVISION
 * ============================================================
 *
 * System Design
 *     -> Requirements ko scalable architecture mein convert
 *        karna.
 *
 * Requirement
 *     -> Architecture decide karti hai.
 *
 * Scale
 *     -> Users, RPS, reads, writes, storage, bandwidth.
 *
 * Load Balancer
 *     -> Traffic distribute karta hai.
 *
 * Application Server
 *     -> Business logic execute karta hai.
 *
 * Cache
 *     -> Faster reads + DB load reduction.
 *
 * Database
 *     -> Durable data storage.
 *
 * Queue
 *     -> Async processing + decoupling + buffering.
 *
 * Worker
 *     -> Background tasks process karta hai.
 *
 * External API
 *     -> Third-party dependency.
 *
 * Observability
 *     -> Logs + Metrics + Traces + Alerts + Dashboards.
 *
 * Bottleneck
 *     -> System ka limiting component.
 *
 * Failure Mode
 *     -> Component failure/slowdown scenario.
 *
 * Trade-off
 *     -> Ek benefit ke badle aane wali cost/new problem.
 *
 *
 * ============================================================
 * 25. ONE-LINE INTERVIEW ANSWER
 * ============================================================
 *
 * "System design is the process of converting functional and
 * non-functional requirements into an architecture that can
 * handle the expected scale, performance, reliability and
 * operational constraints while making explicit trade-offs."
 *
 *
 * ============================================================
 * GOLDEN RULE
 * ============================================================
 *
 * Never ask:
 *
 *     "Which technology should I use?"
 *
 * First ask:
 *
 *     "What problem am I trying to solve?"
 *
 * Then:
 *
 *     Requirement
 *          ↓
 *     Scale
 *          ↓
 *     Problem
 *          ↓
 *     Component
 *          ↓
 *     Technology
 *          ↓
 *     Trade-off
 *
 * ============================================================
 */
```
