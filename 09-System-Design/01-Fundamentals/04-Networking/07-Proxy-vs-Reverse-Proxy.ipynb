```python
/**
 * ============================================================================
 *  PROXY vs REVERSE PROXY - Complete Simple Notes (Hinglish)
 *  Beginners ke liye - Bilkul simple bhasha mein
 * ============================================================================
 *
 * ============================================================================
 *  INTRO: PROXY KYA HAI (Sabse Pehle Samjho)
 * ============================================================================
 *
 * PROXY ek HELPER SERVER hai jo BEECH MEIN BAITHTA hai aur TRAFFIC FORWARD
 * karta hai
 *
 * CONFUSION KAHAN HAI (Pareshani Kya Hai):
 *   - Do BILKUL ALAG systems dono ko "PROXY" keh sakte hain
 *   - Sabse aasaan tarika unhe alag karne ka ek hi sawaal hai:
 *
 *   "PROXY KISKI SIDE PE HAI?"
 *
 * ============================================================================
 *  FORWARD PROXY vs REVERSE PROXY (Simple Fark)
 * ============================================================================
 *
 *   - FORWARD PROXY: CLIENT ki SIDE pe hota hai
 *     * Yeh clients ko DOOSRE SERVERS tak pahunchne mein madad karta hai
 *
 *   - REVERSE PROXY: SERVER ki SIDE pe hota hai
 *     * Yeh servers ko clients se TRAFFIC RECEIVE karne mein madad karta hai
 *
 * ============================================================================
 *  EXAMPLES (Dono Ka Fark Aise Samjho)
 * ============================================================================
 *
 * FORWARD PROXY EXAMPLE:
 *   - Ek company apne employees ki web traffic ko forward proxy se guzarti hai
 *   - Taaki woh:
 *     * Unsafe sites BLOCK kar sake
 *     * Audit logs RAKH sake
 *     * Control kar sake ki employees KUNSE SERVICES access kar sakte hain
 *
 * REVERSE PROXY EXAMPLE:
 *   - Ek web service apne backend servers ke SAAMNE ek reverse proxy rakhta hai
 *   - Taaki public users sirf EK SAFE ENTRY POINT se baat karein
 *   - Reverse proxy phir kar sakta hai:
 *     * HTTPS handle karna
 *     * Requests route karna
 *     * Traffic ko servers pe spread karna
 *     * Suspicious requests BLOCK karna
 *
 * ============================================================================
 *  1. FORWARD PROXY (Client Side Ka Proxy)
 * ============================================================================
 *
 * FORWARD PROXY clients aur un UN BAHAR WALE SERVERS ke beech baithta hai
 * jin tak woh pahunchna chahte hain
 *
 * IMPORTANT - BAHAR WALE SERVER KYA DEKHTA HAI:
 *   - Bahar wala server PROXY ko dekhta hai
 *   - ORIGINAL CLIENT KO NAHI
 *   - Woh proxy ko DIRECT CALLER samajhta hai
 *
 * ============================================================================
 *  FORWARD PROXY KYA HIDE KARTA HAI
 * ============================================================================
 *
 * Yeh proxy ko CLIENT KI IP ADDRESS HIDE karne deta hai:
 *   - Company rules APPLY karna
 *   - Requests LOG karna
 *   - Common responses CACHE karna
 *   - Unsafe traffic BLOCK karna
 *
 * Diagram flow:
 *   Client Network (Client A, B, C) -> Forward Proxy -> Public Internet
 *   (Server 1, 2, 3)
 *
 * ============================================================================
 *  REQUEST FLOW (Request Ka Safar - Step by Step)
 * ============================================================================
 *
 *   1. Client forward proxy ko ek request bhejta hai
 *
 *   2. Proxy CHECK karta hai ki request ALLOWED hai ya nahi
 *
 *   3. Agar ALLOWED hai, toh proxy destination server se ek CONNECTION
 *      OPEN karta hai
 *
 *   4. Destination server PROXY ko respond karta hai
 *
 *   5. Proxy response CLIENT ko wapas bhejta hai
 *
 * ============================================================================
 *  PLAIN HTTP vs HTTPS (Dono Mein Proxy Kaise Kaam Karta Hai)
 * ============================================================================
 *
 * PLAIN HTTP:
 *   - Client ko pata ho sakta hai ki woh proxy use kar raha hai
 *   - Aur requests PROXY ko DIRECTLY bhej sakta hai
 *
 * HTTPS:
 *   - Proxy AKSAR CONNECT METHOD use karta hai
 *   - CONNECT ko aise socho ki proxy se POOCHNA: "Target site tak ek tunnel
 *     khol do"
 *   - Phir ENCRYPTED HTTPS traffic US TUNNEL se guzarta hai
 *
 * ============================================================================
 *  TLS INSPECTION (Kuch Companies Kya Karti Hain)
 * ============================================================================
 *
 * Kuch companies TLS INSPECTION bhi karti hain:
 *   - Us setup mein, proxy HTTPS traffic ko DECRYPT karta hai
 *   - Phir use CHECK karta hai
 *   - Aur destination tak ek NAYI ENCRYPTED CONNECTION banata hai
 *
 * YEH TAB HI KAAM KARTA HAI JAB:
 *   - Client device ek COMPANY-MANAGED CERTIFICATE AUTHORITY pe TRUST kare
 *
 * KYA MATLAB HAI:
 *   - Proxy ko BAHUT ZYADA VISIBILITY milti hai
 *   - Ise EK SENSITIVE TRUST POINT ki tarah treat KARNA CHAHIYE
 *
 * ============================================================================
 *  TEAMS FORWARD PROXIES KYUN USE KARTE HAIN
 * ============================================================================
 *
 * Forward proxies tab COMMON hain jab ek organization chahe ki:
 *   - Uski NETWORK CHHODNE wali TRAFFIC PE CONTROL ho
 *
 *   ┌──────────────────────────────┬──────────────────────────────────────────────────────────┐
 *   │ Use Case                     │ What the Proxy Provides                                  │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Company outbound traffic     │ Destinations ko allow ya block karta hai aur external    │
 *   │ rules                        │ access log karta hai                                     │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Privacy or IP masking        │ Client ki IP ki jagah PROXY KI IP dikhata hai            │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Content filtering            │ Malware, phishing, adult content, ya unapproved tools    │
 *   │                              │ ko block karta hai                                       │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Caching                      │ Bandwidth bachane ke liye popular responses REUSE karta  │
 *   │                              │ hai                                                      │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Developer access             │ Traffic ko ek CONTROLLED NETWORK PATH se bhejta hai      │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Compliance                   │ Network chhodne wali traffic ke liye AUDIT LOGS rakhta   │
 *   │                              │ hai                                                      │
 *   └──────────────────────────────┴──────────────────────────────────────────────────────────┘
 *
 * ============================================================================
 *  IMPORTANT - IP MASKING ko TRUE ANONYMATY MAT SAMJHO
 * ============================================================================
 *
 *   - Ek proxy destination se aapki IP HIDE kar sakta hai
 *   - Lekin PROXY OPERATOR ABHI BHI LOG KAR SAKTA HAI:
 *     * Aap KAUN ho
 *     * Aap KAHAN connect karte ho
 *     * Aap KAB connect karte ho
 *     * Kabhi kabhi aap KYA bhejte ho
 *   - Destination site aapko IN SE BHI PEHCHAAN SAKTI HAI:
 *     * Cookies
 *     * Logins
 *     * Browser fingerprints
 *     * Application tokens
 *
 * ============================================================================
 *  PROXY vs VPN (Fark Kya Hai)
 * ============================================================================
 *
 * DONO KYA KARTE HAIN:
 *   - VPN aur forward proxy DONO traffic ko DOOSRE SYSTEM se guzarte hain
 *   - Lekin woh ALAG ALAG LEVELS pe kaam karte hain
 *
 * VPN:
 *   - Usually EK DEVICE ke zyada tar ya POORE TRAFFIC ke liye ENCRYPTED TUNNEL
 *     banata hai
 *
 * PROXY:
 *   - Usually EK SPECIFIC APP, BROWSER, ya PROTOCOL ke liye traffic handle
 *     karta hai
 *
 * DONO KE LIYE IMPORTANT:
 *   - Koi bhi ABHI BHI LOGGED, BLOCKED, INSPECTED, ya MISCONFIGURED ho sakta hai
 *   - Yeh depend karta hai ki ISE KAUN CHALATA hai
 *
 * ============================================================================
 *  2. REVERSE PROXY (Server Side Ka Proxy)
 * ============================================================================
 *
 * REVERSE PROXY BACKEND SERVERS ke SAAMNE (in front of) baithta hai
 *
 * KAISE KAAM KARTA HAI:
 *   1. Clients reverse proxy se CONNECT karte hain
 *   2. Reverse proxy phir CHOOSE karta hai ki har request KOHAN BHEJNI hai
 *
 * CLIENT KO KYA DIKHTA HAI:
 *   - Client ko service EK STABLE ENDPOINT lagta hai
 *
 * PEECHE KYA HO SAKTA HAI (Behind the scenes):
 *   - Backend servers ADD ho sakte hain
 *   - REMOVE ho sakte hain
 *   - REPLACE ho sakte hain
 *   - RESTART ho sakte hain
 *   - DEPLOY ho sakte hain
 *   - Ya PRIVATE rakhe ja sakte hain
 *
 * ============================================================================
 *  REVERSE PROXY REQUEST FLOW (Step by Step)
 * ============================================================================
 *
 * Diagram flow:
 *   Public Clients (Client A, B, C) -> Reverse Proxy -> Private Backends
 *   (Backend 1, 2, 3)
 *
 *   1. Client https://api.example.com se CONNECT karta hai
 *
 *   2. DNS us naam ko EK REVERSE PROXY, LOAD BALANCER, CDN, GATEWAY, ya
 *      Kubernetes ingress controller ki taraf point karta hai
 *
 *   3. Reverse proxy handle karta hai:
 *      * Connection rules
 *      * HTTPS
 *      * Routing
 *      * Security checks
 *      * Timeouts
 *
 *   4. Yeh ek BACKEND SERVICE ya SERVER INSTANCE choose karta hai
 *
 *   5. Backend reverse proxy ko respond karta hai
 *
 *   6. Reverse proxy response CLIENT ko wapas bhejta hai
 *
 * ============================================================================
 *  REVERSE PROXIES KYUN COMMON HAIN
 * ============================================================================
 *
 * Reverse proxies isliye common hain kyunki yeh ek CONTROLLED FRONT DOOR
 * banate hain:
 *   - PUBLIC CLIENTS ke beech mein
 *   - Aur PRIVATE INFRASTRUCTURE ke beech mein
 *
 * ============================================================================
 *  WHAT REVERSE PROXIES PROVIDE (Reverse Proxy Kya Deta Hai)
 * ============================================================================
 *
 *   ┌──────────────────────────────┬──────────────────────────────────────────────────────────┐
 *   │ Capability                   │ Why It Matters                                           │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ TLS handling                 │ Certificates aur HTTPS policies ko ek jagah rakhta hai   │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Load balancing               │ Traffic ko healthy backend servers pe bhejta hai          │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Routing                      │ Host, path, header, method, ya API se backend choose     │
 *   │                              │ karta hai                                                │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Caching                      │ Backend load kam karta hai aur responses tez banata hai   │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Compression                  │ Responses ko clients ko bhejne se PEHLE CHHOTA banata     │
 *   │                              │ hai                                                      │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Web application firewall     │ Common attacks aur bad requests ko BLOCK karta hai        │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Rate limiting                │ Backends ko abuse ya sudden overload se bachata hai       │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Authentication support       │ Tokens CHECK karta hai ya user identity application ko    │
 *   │                              │ forward karta hai                                        │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Monitoring and logs          │ Logs, latency, status codes, aur trace IDs collect karta  │
 *   │                              │ hai                                                      │
 *   ├──────────────────────────────┼──────────────────────────────────────────────────────────┤
 *   │ Backend connection           │ Backend connections REUSE karta hai aur timeouts apply     │
 *   │ management                   │ karta hai                                                │
 *   └──────────────────────────────┴──────────────────────────────────────────────────────────┘
 *
 * ============================================================================
 *  EXAMPLES (Reverse Proxy Tools)
 * ============================================================================
 *
 *   - NGINX
 *   - HAProxy
 *   - Envoy
 *   - Apache httpd
 *   - Traefik
 *   - Caddy
 *   - Cloudflare
 *   - Fastly
 *   - AWS Application Load Balancer
 *   - Google Cloud Load Balancing
 *   - Azure Application Gateway
 *   - Kubernetes ingress controllers
 *
 * ============================================================================
 *  3. SIDE-BY-SIDE COMPARISON (Dono Ka Direct Comparison)
 * ============================================================================
 *
 *   ┌──────────────────┬──────────────────────────────────┬──────────────────────────────────┐
 *   │ Aspect           │ Forward Proxy                    │ Reverse Proxy                    │
 *   ├──────────────────┼──────────────────────────────────┼──────────────────────────────────┤
 *   │ Represents       │ Clients                          │ Servers or services              │
 *   │ Traffic Direction│ Leaving a client network         │ Entering a service               │
 *   │ Who Configures   │ Client, browser, OS, company     │ Service owner, platform team,    │
 *   │ It               │ network, workload platform       │ CDN, cloud provider              │
 *   │ Destination Sees │ Proxy as the client              │ Reverse proxy as the server      │
 *   │                  │                                  │ endpoint                         │
 *   │ Common Examples  │ Company proxy, Squid, browser    │ NGINX, HAProxy, Envoy, CDN,      │
 *   │                  │ proxy, outbound proxy            │ API gateway, Kubernetes ingress  │
 *   └──────────────────┴──────────────────────────────────┴──────────────────────────────────┘
 *
 * ============================================================================
 *  NAAM KYUN CONFUSING HAI (Why The Naming Is Confusing)
 * ============================================================================
 *
 *   - Dono "BEECH MEIN" hote hain - isi liye naam confusing hai
 *   - UNHE ALAG KARNE WALI CHEEZ yeh hai:
 *     "Woh KIS SIDE KO REPRESENT karte hain"
 *
 * ============================================================================
 *  4. LAYER 4 vs LAYER 7 PROXIES
 * ============================================================================
 *
 * Proxies NETWORK STACK ke ALAG ALAG LEVELS pe kaam kar sakte hain
 * - Do common hain: Layer 4 aur Layer 7
 *
 * ============================================================================
 *  LAYER 4 PROXY
 * ============================================================================
 *
 * - Layer 4 proxy TCP ya UDP CONNECTIONS ke saath kaam karta hai
 *
 * YEH KYA DEKH SAKTA HAI:
 *   - Network-level details jaise IP addresses
 *   - Ports
 *   - Aur connection OPEN hai ya CLOSED
 *
 * YEH KYA NAHI SAMAJHTA:
 *   - Usually HTTP paths
 *   - Headers
 *   - Cookies
 *   - Request bodies
 *
 * LAYER 4 KAB USEFUL HAI:
 *   - Jab traffic HTTP NAHI hai
 *   - Jab aapko SIMPLE aur FAST forwarding chahiye
 *   - Ya jab ENCRYPTED TRAFFIC ko PEECHE TAK POORA BINA DECRYPT kiye
 *     pass karna ho
 *
 * ============================================================================
 *  LAYER 7 PROXY
 * ============================================================================
 *
 * - Layer 7 proxy APPLICATION PROTOCOL samajhta hai, usually HTTP
 *
 * YEH KYA DEKH SAKTA HAI (aur decide kar sakta hai):
 *   - Hostnames
 *   - URL paths
 *   - HTTP methods
 *   - Headers
 *   - Cookies
 *   - gRPC methods
 *   - Carefully validated JWT claims
 *
 * IMPORTANT WARNING:
 *   - Layer 7 aapko ZYADA CONTROL deta hai
 *   - Lekin yeh GALT KARNE KE ZYADA TARIKE bhi banata hai
 *   - Ek baar jab proxy REQUESTS ko READ, CHANGE, RETRY, ya BUFFER karta hai
 *   - Toh yeh AB SIRF EK PIPE NAHI hai
 *   - Yeh APPLICATION KI BEHAVIOUR KA HISSA ban jaata hai
 *
 * ============================================================================
 *  5. HTTPS HANDLING AND TRUST POINTS (Suraksha Ke Bharose Ke Points)
 * ============================================================================
 *
 * Reverse proxies AKSAR service ke liye HTTPS handle karte hain
 *
 * KAISE KAAM KARTA HAI:
 *   - Client proxy ko ek HTTPS connection banata hai
 *   - Proxy phir request BACKEND ko forward karta hai:
 *     * Plain HTTP ke through, ya
 *     * Ek NAYI HTTPS connection ke through
 *
 * ============================================================================
 *  COMMON PATTERNS (Common Tarike)
 * ============================================================================
 *
 *   ┌──────────────────┬─────────────────────────────────────┬─────────────────────────────────────┐
 *   │ Pattern          │ Description                         │ Tradeoff                            │
 *   ├──────────────────┼─────────────────────────────────────┼─────────────────────────────────────┤
 *   │ TLS termination  │ Proxy backend ko bhejne se PEHLE   │ Layer 7 routing enable karta hai,   │
 *   │                  │ HTTPS DECRYPT karta hai             │ lekin proxy TRAFFIC PADH SAKTA hai  │
 *   ├──────────────────┼─────────────────────────────────────┼─────────────────────────────────────┤
 *   │ TLS              │ Proxy decrypt karta hai, CHECK      │ Backend hop ko PROTECT karta hai,   │
 *   │ re-encryption    │ karta hai, phir backend pe HTTPS    │ lekin CERTIFICATE WORK add karta hai│
 *   │                  │ OPEN karta hai                      │                                     │
 *   ├──────────────────┼─────────────────────────────────────┼─────────────────────────────────────┤
 *   │ TLS              │ Proxy ENCRYPTED TRAFFIC BINA        │ END-TO-END ENCRYPTION rakhta hai,   │
 *   │ pass-through     │ PADHE forward karta hai             │ lekin HTTP-AWARE ROUTING LIMIT       │
 *   │                  │                                     │ karta hai                           │
 *   └──────────────────┴─────────────────────────────────────┴─────────────────────────────────────┘
 *
 * ============================================================================
 *  YEH EK IMPORTANT SECURITY BOUNDARY HAI
 * ============================================================================
 *
 * Agar ek reverse proxy HTTPS DECRYPT karta hai:
 *   - Ise TRUSTED INFRASTRUCTURE HONA CHAHIYE
 *   - Yeh dekh sakta hai:
 *     * Headers
 *     * Cookies
 *     * Authorization tokens
 *     * Request bodies
 *     * Responses
 *
 * ============================================================================
 *  ASLI SAWAAL BAHUT SIMPLE HAI:
 * ============================================================================
 *
 *   "DECRYPTED TRAFFIC KO KAHAN EXIST KARNE KI IZZAT HAI?"
 *
 * ----------------------------------------------------------------------------
 *  TERMINATION ya RE-ENCRYPTION ke saath:
 * ----------------------------------------------------------------------------
 *   - Traffic THODI DER ke liye PROXY KE ANDAR DECRYPT hota hai
 *   - Yeh proxy ko deti hai:
 *     * Path se route karna
 *     * WAF rules apply karna
 *     * Headers add karna
 *   - RISK yeh hai ki ek COMPROMISED PROXY traffic PADH sakta hai
 *
 * ----------------------------------------------------------------------------
 *  PASS-THROUGH ke saath:
 * ----------------------------------------------------------------------------
 *   - Proxy ENCRYPTED BYTES bina KEY ke forward karta hai
 *   - Yeh DECRYPTED REQUEST KABHI NAHI dekhta
 *
 * PASS-THROUGH MODE MEIN KYA HOTA HAI:
 *   - Backend HTTPS handle karta hai
 *   - Backend CERTIFICATE rakhta hai
 *   - COST yeh hai ki proxy HTTP paths, headers, responses, ya bodies
 *     INSPECT NAHI kar sakta
 *   - Isliye bahut saari LAYER 7 FEATURES UNAVAILABLE ho jaati hain
 *
 * SHORT MEIN:
 *   - Jab aapko HTTP-AWARE FEATURES chahiye aur proxy pe TRUST hai
 *     -> TERMINATE ya RE-ENCRYPT karo
 *   - Jab TRAFFIC KO END TAK ENCRYPTED RAKHNA zyada important hai
 *     -> PASS-THROUGH use karo
 *
 * ============================================================================
 *  FORWARD PROXIES KE LIYE TLS INSPECTION (Zyada Sensitive Hai)
 * ============================================================================
 *
 *   - Forward proxies ke liye TLS inspection AUR BHI SENSITIVE hai
 *   - Kyunki yeh USER TRAFFIC ko affect karta hai
 *   - Yeh kuch COMPANY ENVIRONMENTS mein APPROPRIATE ho sakta hai
 *   - Lekin ise EXPLICIT, DOCUMENTED, aur CAREFULLY CONTROLLED hona chahiye
 *
 * ============================================================================
 *  6. HEADERS AND CLIENT IDENTITY (Client Ki Pehchaan)
 * ============================================================================
 *
 * Jab ek reverse proxy request forward karta hai:
 *   - Backend AKSAR PROXY KI IP ADDRESS DIRECT CALLER ke roop mein dekhta hai
 *
 * ============================================================================
 *  BACKEND KO ORIGINA CLIENT INFO KYUN CHAHIYE
 * ============================================================================
 *
 * Backend ko zaroorat ho sakti hai jaan-ne ki:
 *   - ORIGINAL CLIENT IP
 *   - ORIGINAL HOST
 *   - Aur client ne HTTPS use kiya tha ya nahi
 *
 * Agar yeh jaankari chahiye toh:
 *   - Proxy ko yeh HEADERS mein PASS karna padta hai
 *
 * ============================================================================
 *  COMMON HEADERS
 * ============================================================================
 *
 *   ┌──────────────────────┬──────────────────────────────────────────────────────────────────┐
 *   │ Header               │ Purpose                                                          │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Host                 │ Hostname jo client ne request kiya                               │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ X-Forwarded-For      │ Client aur proxy IP chain                                        │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ X-Forwarded-Proto    │ Original scheme, jaise https                                     │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Forwarded            │ Forwarded information ka standard form                           │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ X-Request-ID         │ Logs mein ek request ko follow karne ke liye ID                  │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ traceparent          │ Distributed tracing systems ke liye ID                           │
 *   └──────────────────────┴──────────────────────────────────────────────────────────────────┘
 *
 * ============================================================================
 *  SECURITY RULE (Bahut Zaroori!)
 * ============================================================================
 *
 *   - IN HEADERS KO TAB HI TRUST KARO jab woh AAPKE PROXIES SE AAYE hon
 *   - PUBLIC CLIENTS X-Forwarded-For jaise headers FAKE kar sakte hain
 *
 * EK ACHHA EDGE PROXY KYA KARTA HAI:
 *   - Incoming requests se UNTRUSTED FORWARDING HEADERS REMOVE karta hai
 *   - AUR TRAFFIC KE APPLICATION TAK PAHUNCHNE SE PEHLE CLEAN ONES LIKHTA HAI
 *
 * ============================================================================
 *  YEH GALTI KARNE SE PRODUCTION BUGS AATE HAIN (Common Bugs)
 * ============================================================================
 *
 * Bahut saare production bugs ISI SE aate hain:
 *
 *   1. Applications HTTP ke peeche EK HTTPS proxy ke saath HTTP:// redirects
 *      generate karte hain
 *
 *   2. Rate limiters REAL CLIENT IP ki jagah PROXY IP ko throttle karte hain
 *
 *   3. Audit logs SIRF LOAD BALANCER ADDRESSES record karte hain
 *
 *   4. Security checks FAKE CLIENT IP HEADERS pe trust karte hain
 *
 *   5. Absolute URLs isliye BREAK hote hain kyunki Host PRESERVE NAHI hua
 *
 * ============================================================================
 *  7. CACHING (Cache Karke Bachao)
 * ============================================================================
 *
 * Dono forward aur reverse proxies RESPONSES CACHE kar sakte hain
 * - Lekin woh ALAG REASONS se karte hain
 *
 * ----------------------------------------------------------------------------
 *  FORWARD PROXY CACHING:
 * ----------------------------------------------------------------------------
 *   - CLIENT NETWORK ke liye BANDWIDTH bachata hai
 *
 * ----------------------------------------------------------------------------
 *  REVERSE PROXY CACHING:
 * ----------------------------------------------------------------------------
 *   - BACKEND SERVICES pe LOAD KAM karta hai
 *   - AUR USERS ke liye RESPONSES TEZ banata hai
 *
 * ============================================================================
 *  CACHING KO HTTP CACHING RULES FOLLOW KARNE CHAHIYE
 * ============================================================================
 *
 * IMPORTANT SIGNALS include karte hain:
 *   - Cache-Control
 *   - ETag
 *   - Vary
 *   - Authorization
 *   - Cookies
 *   - Query parameters
 *   - Content encoding
 *
 * ============================================================================
 *  DANGEROUS MISTAKE (Sabse Khatarnak Galti)
 * ============================================================================
 *
 *   - PRIVATE ya PERSONALIZED DATA ko CACHE karna
 *   - AUR PHIR USE GALAT USER KO SERVE karna
 *
 * REVERSE PROXY CACHE KO KYA KARNA CHAHIYE:
 *   - Logged-in ya user-specific responses ke saath SAAVDHANI BARTANI CHAHIYE
 *   - Jab tak application clearly NAHI batata ki UNHE CACHE KARNA SAFE hai
 *
 * ============================================================================
 *  8. REVERSE PROXY vs LOAD BALANCER vs API GATEWAY
 * ============================================================================
 *
 * Yeh terms REAL PRODUCTS mein OVERLAP karti hain
 * - TENSION mat lo agar ek tool KAI LABELS fit karta hai
 *
 * ============================================================================
 *  COMPONENT ROLES (Har Component Ka Kaam)
 * ============================================================================
 *
 *   ┌──────────────────────┬──────────────────────────────────────────────────────────────────┐
 *   │ Component            │ Primary Role                                                     │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Reverse proxy        │ Services ke saamne khada hota hai aur requests backends pe       │
 *   │                      │ forward karta hai                                                │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Load balancer        │ Healthy backend servers ya regions choose karta hai               │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ API gateway          │ API rules add karta hai jaise auth, rate limits, quotas,         │
 *   │                      │ transformations, aur developer controls                           │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ CDN                  │ Edge locations se content CACHE aur SERVE karta hai,            │
 *   │                      │ aksar reverse proxy ki tarah kaam karke                          │
 *   ├──────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Service mesh proxy   │ Service-to-service traffic handle karta hai, mutual TLS,        │
 *   │                      │ retries, metrics, tracing, aur policy checks                     │
 *   └──────────────────────┴──────────────────────────────────────────────────────────────────┘
 *
 * ============================================================================
 *  EK SYSTEM KAI ROLES KHEL SAKTA HAI
 * ============================================================================
 *
 * Example:
 *   - Envoy ho sakta hai:
 *     * Ek reverse proxy
 *     * Ek load balancer
 *     * Ek API gateway ka hissa
 *     * Ek service mesh sidecar
 *
 *   - Cloudflare ho sakta hai:
 *     * Ek CDN
 *     * Ek WAF
 *     * Ek reverse proxy
 *     * Ek DDoS protection layer
 *
 * ============================================================================
 *  IMPORTANT - LABELS SE ZYADA BEHAVIOUR MATTER KARTA HAI
 * ============================================================================
 *
 * YEH SAWAAL POOCHO (In order):
 *
 *   1. Yeh KYA TRAFFIC dekhta hai?
 *      (What traffic does it see?)
 *
 *   2. Yeh KYA DECISIONS leta hai?
 *      (What decisions does it make?)
 *
 *   3. Yeh KYA BADALTA hai, CACHE karta hai, RETRY karta hai, ya LOG karta hai?
 *      (What does it change, cache, retry, or log?)
 *
 *   4. Jab yeh FAIL ho toh KYA HOTA HAI?
 *      (What happens when it fails?)
 *
 * ============================================================================
 *  9. NGINX REVERSE PROXY EXAMPLE (Practical Example)
 * ============================================================================
 *
 * Yahan ek chhota NGINX reverse proxy hai ek HTTP backend ke liye
 *
 * ============================================================================
 *  INSTALL AND RELOAD (Install Aur Reload Karo)
 * ============================================================================
 *
 * Pehle NGINX install karo, config valid check karo, aur service reload karo:
 *
 *   sudo apt update
 *   sudo apt install nginx
 *   sudo systemctl reload nginx
 *
 * ============================================================================
 *  BASIC REVERSE PROXY
 * ============================================================================
 *
 * Yeh server block HAR REQUEST ko ek backend pe bhejta hai
 * Aur headers SET karta hai jo backend ko samajhne mein madad karte hain
 * ki ORIGINAL CLIENT REQUEST kya thi
 *
 *   server {
 *       listen 80;
 *       server_name example.com;
 *
 *       location / {
 *           proxy_pass http://backend1.internal.example.com:8080;
 *           proxy_http_version 1.1;
 *
 *           proxy_set_header Host $host;
 *           proxy_set_header X-Real-IP $remote_addr;
 *           proxy_set_header X-Forwarded-For $proxy_add_x_forwarded_for;
 *           proxy_set_header X-Forwarded-Proto $scheme;
 *           proxy_set_header X-Request-ID $request_id;
 *
 *           proxy_connect_timeout 3s;
 *           proxy_read_timeout 30s;
 *           proxy_send_timeout 30s;
 *       }
 *   }
 *
 * ============================================================================
 *  LOAD BALANCING ACROSS BACKENDS (Traffic Baanto)
 * ============================================================================
 *
 * - Ek upstream block add karna ek proxy ko KAI BACKENDS pe TRAFFIC SPREAD
 *   karne deta hai
 * - Failure settings NGINX ko STOP karne mein madad karte hain
 *   ki traffic ko us server pe jo BAAR BAAR FAIL ho raha hai
 *
 *   upstream backend_app {
 *       least_conn;
 *
 *       server backend1.internal.example.com:8080 max_fails=3 fail_timeout=10s;
 *       server backend2.internal.example.com:8080 max_fails=3 fail_timeout=10s;
 *       server backend3.internal.example.com:8080 max_fails=3 fail_timeout=10s;
 *   }
 *
 * ============================================================================
 *  NGINX LOAD BALANCING ALGORITHMS
 * ============================================================================
 *
 *   - NGINX by default ROUND ROBIN use karta hai
 *     * Matlab yeh backend servers ko CYCLE karta hai
 *
 *   - least_conn tab useful hai jab kuch requests DOOSRON SE ZYADA DER LETE
 *     hain
 *     * Kyunki yeh naye requests us server ko bhejta hai jiske paas SABSE KAM
 *       ACTIVE CONNECTIONS hain
 *
 * ============================================================================
 *  STICKY ROUTING (Sticky Routing Ke Baare Mein)
 * ============================================================================
 *
 *   - Sticky routing POSSIBLE hai
 *   - Lekin ISE CAREFULLY USE KARO
 *   - Yeh ye FACT HIDE kar sakta hai ki backend LOCAL SERVER STATE pe
 *     depend karta hai
 *
 * ============================================================================
 *  NOTE - YEH EXAMPLE CHHOTA RAKHA GAYA HAI
 * ============================================================================
 *
 * Production configs ko yeh bhi chahiye:
 *   - HTTPS
 *   - Access logs
 *   - Request size limits
 *   - Buffering choices
 *   - Health checks ya failure policy
 *   - Compression
 *   - Security headers
 *   - Aur timeouts jo application se MATCH karein
 *
 * ============================================================================
 *  10. FAILURE MODES (Proxy Kab Fail Hota Hai)
 * ============================================================================
 *
 * Proxies systems ko ZYADA FLEXIBLE banate hain
 * - Lekin yeh HAR REQUEST KE PATH PE BHI BAITHE HOTE HAIN
 *
 * COMMON FAILURE MODES:
 *
 *   ┌──────────────────────────┬──────────────────────────────────────────────────────────────────┐
 *   │ Failure                  │ What Happens                                                     │
 *   ├──────────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Proxy outage             │ Traffic fail hota hai chahe backend servers HEALTHY hon          │
 *   ├──────────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Bad timeout              │ Slow backends RESOURCES TIE UP karte hain ya clients bahut        │
 *   │                          │ jaldi fail karte hain                                            │
 *   ├──────────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Retry storm              │ Proxy retries ek OVERLOADED BACKEND ko aur BUSY bana dete hain   │
 *   ├──────────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Buffering mismatch       │ Streaming, uploads, ya WebSockets BREAK ho jaate hain            │
 *   ├──────────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Header trust bug         │ Application FAKE CLIENT INFORMATION pe trust karta hai           │
 *   ├──────────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ TLS misconfiguration     │ HTTPS handshakes fail hote hain ya security WEAK ho jaati hai    │
 *   ├──────────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Cache mistake            │ Stale ya private data GALAT TARIKE se serve hoti hai             │
 *   ├──────────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Weak health check        │ Proxy BROKEN BACKENDS pe traffic bhejta hai                      │
 *   ├──────────────────────────┼──────────────────────────────────────────────────────────────────┤
 *   │ Connection pool          │ Proxy ENOUGH BACKEND CONNECTIONS open ya reuse NAHI kar sakta    │
 *   │ exhaustion               │                                                                  │
 *   └──────────────────────────┴──────────────────────────────────────────────────────────────────┘
 *
 * ============================================================================
 *  AI SYSTEMS KE LIYE SPECIAL WARNING
 * ============================================================================
 *
 * AI systems ke liye, IN PE SPECIAL ATTENTION do:
 *   - Long-running aur streaming requests
 *   - Token streams
 *   - File uploads
 *   - Batch jobs
 *   - Model inference calls
 *   - IN SABKO CLEAN TIMEOUT, BUFFERING, aur CANCELLATION BEHAVIOUR chahiye
 *
 * IMPORTANT WARNING:
 *   - Ek proxy jo SHORT WEB REQUESTS ke liye TUNE kiya gaya hai
 *   - Yeh IN WORKFLOWS KO LOAD KE UNDER BREAK kar sakta hai
 *
 * ============================================================================
 *  SUMMARY (Saransh)
 * ============================================================================
 *
 * 1. FORWARD PROXIES aur REVERSE PROXIES DONO BEECH MEIN HAIN
 *    - Lekin woh ALAG ALAG SIDES represent karte hain
 *
 * 2. FORWARD PROXY CLIENTS represent karta hai:
 *    - Yeh ek CLIENT NETWORK CHHODNE wali TRAFFIC ko control karta hai
 *    - Filtering, privacy, logging, compliance, aur controlled access ke liye
 *      useful hai
 *
 * 3. REVERSE PROXY SERVICES represent karta hai:
 *    - Yeh EK SYSTEM MEIN ENTER HONE WALI TRAFFIC ko control karta hai
 *    - HTTPS handling, routing, load balancing, caching, security filtering,
 *      rate limiting, monitoring, aur logs ke liye useful hai
 *
 * 4. DONO TRUST POINTS hain:
 *    - Forwarded headers ke saath CAREFUL RAHO
 *    - TLS inspection ke saath CAREFUL RAHO
 *    - Caching ke saath CAREFUL RAHO
 *    - Retries ke saath CAREFUL RAHO
 *    - Logs ke saath CAREFUL RAHO
 *
 * 5. PROXY KE BARE MEIN SOCHNE KA SABSE ACHHA TARIKA:
 *    - Yeh mat socho ki "bas requests forward karta hai"
 *    - YEH SAWAAL POOCHO:
 *      * Yeh KYA DEKH SAKTA hai?
 *      * Yeh KYA BADALTA hai?
 *      * Yeh KYA CACHE karta hai?
 *      * Yeh KYA RETRY karta hai?
 *      * Yeh KYA LOG karta hai?
 *      * Jab yeh FAIL ho toh KYA HOTA HAI?
 *
 * ============================================================================
 **/
```
