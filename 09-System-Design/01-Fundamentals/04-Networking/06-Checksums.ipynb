```python
/**
 * ============================================================================
 *  CHECKSUMS - Complete Simple Notes (Hinglish)
 *  Beginners ke liye - Bilkul simple bhasha mein
 * ============================================================================
 *
 * ============================================================================
 *  INTRO: DATA HAMESHA SAHI NAHI AATA (Problem Kya Hai)
 * ============================================================================
 *
 * Data HAMESHA waise waise NAHI aata jaise bheja gaya tha
 *
 * KYA HO SAKTA HAI:
 *   - Ek NETWORK PACKET DAMAGE ho sakta hai
 *   - Ek DISK purana (old) block return kar sakta hai
 *   - Ek DOWNLOAD aadhe raste pe RUK sakta hai
 *
 * TRICKY BAAT (Mushkil kya hai):
 *   - Bytes khud YEH ANNOUNCE NAHI karte ki "Main TOOT GAYA hoon"
 *   - Woh bas CHUP-CHAP GALAT ho jaate hain
 *
 * ============================================================================
 *  CHECKSUM KYA HAI (Simple Definition)
 * ============================================================================
 *
 * CHECKSUM ek chhota value hota hai jo ek BADE DATA se CALCULATE hota hai
 *
 * KAISE KAAM KARTA HAI:
 *   1. Pehle ek hissa (part) value CALCULATE karta hai
 *   2. Baad mein system dobara se yeh value CALCULATE karta hai
 *   3. Phir DONO RESULTS COMPARE karte hain
 *   4. Agar MATCH ho gaya = data sahi hai
 *   5. Agar MATCH NAHI hua = data BADLA hai (kharab hai)
 *
 * ============================================================================
 *  CHECKSUMS KAHAN DIKHTE HAIN (Sab Jagah Milte Hain)
 * ============================================================================
 *
 * Real systems mein CHECKSUMS HAR JAGAH use hote hain:
 *   - Network frames
 *   - TCP aur UDP packets
 *   - Storage blocks
 *   - Database pages
 *   - Backups
 *   - Container images
 *   - Replication logs
 *
 * ============================================================================
 *  IMPORTANT DESIGN SAWAAL (Kya Problem Pakadni Hai?)
 * ============================================================================
 *
 * SABSE ZAROORI SAWAAL: "Aap KYA TYPE KI PROBLEM pakadna chahte ho?"
 *
 *   - CRC ACCIDENTAL DAMAGE ke liye achha hai (galati se hua nuksan)
 *   - CRYPTOGRAPHIC HASH tab achha hai jab STRONG CONTENT FINGERPRINT chahiye
 *   - HMAC ya DIGITAL SIGNATURE tab use hota hai jab aapko yeh bhi jaanna ho
 *     ki DATA KAUN NE BANAYA (kisne produce kiya)
 *
 * ============================================================================
 *  1. CHECKSUM KYA HAI (What a Checksum Is)
 * ============================================================================
 *
 * CHECKSUM ek BADE DATA se calculate kiya gaya CHHOTA VALUE hota hai
 *
 * ----------------------------------------------------------------------------
 *  EXAMPLE (Storage System Kya Store Karta Hai):
 * ----------------------------------------------------------------------------
 *   object:    video-segment-001
 *   bytes:     8 MB
 *   checksum:  crc32c=0x7f9c2ba4
 *
 * ----------------------------------------------------------------------------
 *  KAISE CHECK HOTA HAI (Baad Mein):
 * ----------------------------------------------------------------------------
 *   - Jab object READ BACK hota hai
 *   - System crc32c DOBARA SE calculate karta hai RECEIVED bytes ke upar
 *   - Agar NAYA value ALAG hai
 *   - Toh system jaanta hai ki object KAHIN NA KAHIN BADLA hai:
 *     * Corrupted ho sakta hai
 *     * Cut short (aadha) ho sakta hai
 *     * Galat bytes se mix ho sakta hai
 *     * Ya kisi aur tarike se damaged ho sakta hai
 *
 * Diagram flow:
 *   Large Data (8 MB object) -> Checksum Function (e.g., CRC-32C) ->
 *   Compact Value (0x7f9c2ba4)
 *
 * ============================================================================
 *  CHECKSUM KYUN USEFUL HAI (Chhota Kyun Hai?)
 * ============================================================================
 *
 * Checksum useful hai kyunki yeh DATA SE BAHUT CHHOTA hota hai:
 *   - 4-byte CRC ek NETWORK FRAME ko protect kar sakta hai
 *   - 32-byte SHA-256 digest ek BAHUT BADE file (gigabytes) ko identify kar
 *     sakta hai
 *
 * ============================================================================
 *  COLLISION KYA HAI (Do Data, Ek Hi Checksum)
 * ============================================================================
 *
 * Kyunki checksum DATA SE CHHOTA hai:
 *   - Do ALAG ALAG inputs KABHI KABHI SAME CHECKSUM produce kar sakte hain
 *   - Isse kehte hain COLLISION (takrav)
 *
 * GOOD ALGORITHMS (Achhe algorithms):
 *   - Accidental missed corruption ko EXTREMELY UNLIKELY banate hain
 *   - Un problems ke liye jinhe pakadne ke liye woh DESIGN kiye gaye hain
 *
 * ============================================================================
 *  2. CHECKSUMS KYA BATA SAKTE HAIN AUR KYA NAHI
 * ============================================================================
 *
 * Checksums ek NARROW (chhoti) sawaal ka jawaab dete hain:
 *
 *   "Kya yeh bytes WOH BYTES HAIN jo maine EXPECT kiye the,
 *    IS ALGORITHM ke hisaab se?"
 *
 * ============================================================================
 *  CHECKSUMS KYA DETECT KARTE HAIN (Yeh Pakadte Hain)
 * ============================================================================
 *
 *   1. Bit flips (bits ka ulta ho jaana)
 *   2. Truncated writes (aadha data likhna)
 *   3. Partial reads (aadha data padhna)
 *   4. Wrong block reads (galat block padhna)
 *   5. Corrupted network frames (network frames kharab hona)
 *   6. Damaged files (files tootna)
 *   7. Bad disk sectors (disk ke kharab hisse)
 *   8. Memory ya DMA corruption (memory kharab hona)
 *   9. Replication ya backup transfer errors (transfer mein galti)
 *
 * ============================================================================
 *  CHECKSUMS KYA AUTOMATICALLY NAHI BATA SAKTE (Limitations)
 * ============================================================================
 *
 *   1. DATA KAUN NE BANAYA? (Who created the data?)
 *
 *   2. KYA CHECKSUM KHUD TAMPERED hua hai?
 *
 *   3. KYA yeh data AUTHORIZED hai?
 *
 *   4. KYA yeh algorithm ATTACKER KE KHILAF SECURE hai?
 *
 *   5. KYA APPLICATION-LEVEL OPERATION CORRECT hai?
 *
 * ============================================================================
 *  ATTACKER PROBLEM (Bahut Zaroori Samjho!)
 * ============================================================================
 *
 * Agar attacker FILE aur CHECKSUM DONO change kar sakta hai:
 *   - Toh ek PLAIN CHECKSUM aapko PROTECT NAHI karta
 *   - Attacker file ko modify kar sakta hai
 *   - Phir MODIFIED FILE ke liye NAYA CHECKSUM calculate kar sakta hai
 *
 * ============================================================================
 *  STRONGER TOOL KAB USE KAREIN
 * ============================================================================
 *
 * Jab LOG DATA TAMPER karne ki koshish kar rahe hon
 * - Toh EK STRONGER TOOL use karo:
 *
 *   1. HMACs
 *   2. Digital signatures
 *   3. Signed manifests
 *   4. TLS
 *   5. Package signatures
 *   6. Ya koi aur TRUSTED VERIFICATION MECHANISM
 *
 * ============================================================================
 *  3. VERIFICATION KAISE KAAM KARTA HAI
 * ============================================================================
 *
 * Verification flow SAME hai IN SAB JAGAH:
 *   - Networks mein
 *   - Files mein
 *   - Databases mein
 *   - Object storage mein
 *
 * Diagram flow:
 *   Write or read data -> Compute checksum -> Store data + checksum
 *   -> Read stored data -> Recompute checksum -> MATCH?
 *      -> Yes: Accept data
 *      -> No: Reject, retry, repair, or alert
 *
 * ============================================================================
 *  MISMATCH KE BAAD KYA HOTA HAI (Mismatch Kahan Mila?)
 * ============================================================================
 *
 *   1. NETWORK CARD: Frame DROP kar sakta hai aur higher layer pe depend
 *      kar sakta hai RETRANSMIT ke liye
 *
 *   2. TCP: Ek bad segment DISCARD kar sakta hai
 *      - Sender use PHIR SE BHEJEGA jab use pata chalega ki data MISSING hai
 *
 *   3. STORAGE ENGINE: Doosra REPLICA read kar sakta hai
 *
 *   4. OBJECT STORE: Request FAIL kar sakta hai aur ek DATA-INTEGRITY ERROR
 *      return kar sakta hai
 *
 *   5. BACKUP SYSTEM: Backup ko UNUSABLE mark kar sakta hai
 *
 *   6. DATABASE: Startup ROK sakta hai, corrupt page REPLAY karne ke bajaye
 *
 * ============================================================================
 *  CHECKSUMS vs RECOVERY (Yeh Alag Alag Kaam Hain!)
 * ============================================================================
 *
 * IMPORTANT:
 *   - CHECKSUMS corruption DETECT karte hain
 *   - Yeh KHUD SE REPAIR NAHI karte
 *
 * RECOVERY KE LIYE CHAHIYE (Alag mechanism chahiye):
 *   - Retransmission (phir se bhejna)
 *   - Another replica (doosra copy)
 *   - Backup restore (backup se wapas lana)
 *   - A repair job (theek karne wala kaam)
 *   - Manual intervention (khud theek karna)
 *
 * ============================================================================
 *  4. TYPES OF INTEGRITY CHECKS (Integrity Checks Ke Types)
 * ============================================================================
 *
 * Sab checksums SAME problem solve NAHI karte
 * - Algorithm choose karo us cheez ke hisaab se jisse aap PROTECT karna chahte ho
 *
 * ============================================================================
 *  PARITY
 * ============================================================================
 *
 * PARITY BIT kya record karta hai:
 *   - Ki bits ke group mein 1 BITS ka count EVEN hai ya ODD
 *
 * PARITY KYA KAR SAKTA HAI:
 *   - Us group mein KOI BHI SINGLE-BIT ERROR detect kar sakta hai
 *
 * PARITY KYA NAHI KAR SAKTA:
 *   - Bahut saare MULTI-BIT ERRORS detect NAHI kar sakta hai
 *   - Simple aur cheap hai
 *   - Lekin ZYADA TAR application-level integrity checks ke liye BAHUT KAMZOR hai
 *
 * ============================================================================
 *  SIMPLE ADDITIVE CHECKSUMS
 * ============================================================================
 *
 * Simple checksum kya karta hai:
 *   - Bytes ya words ko ADD karta hai
 *   - Result ka hissa store karta hai
 *
 * PROS:
 *   - FAST hain
 *   - Implement karna EASY hai
 *
 * CONS (Kamzoriyan):
 *   - WEAK hain
 *   - REORDERED BYTES miss kar sakte hain
 *   - Woh ERROR PATTERNS miss kar sakte hain jo EK DUSRE KO CANCEL karte hain
 *
 * KAB USE KARO:
 *   - Jab ek EXISTING PROTOCOL unhe REQUIRE karta hai
 *   - Ya jab FAILURE CASES BAHUT LIMITED hain
 *
 * ============================================================================
 *  CRC (Cyclic Redundancy Check)
 * ============================================================================
 *
 * CRC ka full form: "Cyclic Redundancy Check"
 * - Yeh FAST CHECKS ki ek FAMILY hai
 * - NETWORKING aur STORAGE mein BAHUT ZYADA use hoti hain
 * - Inka MATH simple addition se ZYADA involved hai
 * - Lekin PRACTICAL IDEA simple hai: CRCs ACCIDENTAL DAMAGE ke COMMON PATTERNS
 *   pakadne ke liye DESIGN kiye gaye hain
 *
 * CRCs KISME SABSE ACCHHE HAIN:
 *   - Burst errors pakadne mein (lambi galtiyan)
 *   - Bit flips mein (bits ka ulta hona)
 *   - Transmission noise mein (signal mein shor)
 *
 * COMMON EXAMPLES:
 *   - CRC-32
 *   - CRC-32C
 *   - CRC-64
 *
 * ============================================================================
 *  CRC-32C KYUN POPULAR HAI
 * ============================================================================
 *
 * CRC-32C STORAGE aur NETWORKING mein WIDELY USED hai kyunki:
 *   - Yeh COMMON ERRORS ACHHI TARAH pakadta hai
 *   - Bahut saare CPUs ise HARDWARE mein JALDI calculate kar sakte hain
 *
 * ============================================================================
 *  IMPORTANT - CRC CRYPTOGRAPHIC NAHI HAI
 * ============================================================================
 *
 * CRC ek CRYPTOGRAPHIC ALGORITHM NAHI hai
 * - Ek attacker DATA DELIBERATELY MODIFY kar sakta hai
 * - Aur NAYA CRC CALCULATE kar sakta hai
 *
 * ============================================================================
 *  CRYPTOGRAPHIC HASHES (Strong Version)
 * ============================================================================
 *
 * Ek CRYPTOGRAPHIC HASH ek FIXED-SIZE DIGEST produce karta hai
 * - Jaise ek SHA-256 value
 * - Ise aise DESIGN kiya gaya hai ki attackers REALISTICALLY:
 *     * Same digest wala DOOSRA INPUT dhoond NAHI paate
 *     * Digest se ORIGINAL INPUT tak wapas NAHI ja sakte
 *
 * ============================================================================
 *  COMMON CRYPTOGRAPHIC HASH EXAMPLES
 * ============================================================================
 *
 *   1. SHA-256
 *   2. SHA-384
 *   3. SHA-512
 *   4. BLAKE3
 *
 * ============================================================================
 *  CRYPTOGRAPHIC HASHES KE LIYE USEFUL (Kahan Kaam Aate Hain)
 * ============================================================================
 *
 *   1. Verifying downloads (downloads ki jaanch karna)
 *   2. Identifying content by digest (digest se content pehchaanna)
 *   3. Avoiding duplicate storage for immutable blobs (duplicate storage se bachna)
 *   4. Building Merkle trees
 *   5. Verifying container images and package files (jaanch karna)
 *   6. Comparing large files without reading them repeatedly (badi files compare karna)
 *
 * ============================================================================
 *  IMPORTANT - MD5 AUR SHA-1 MAT USE KARO (Security Matters Ho Toh)
 * ============================================================================
 *
 *   - MD5 aur SHA-1 tab USE NAHI KARNA CHAHIYE jab SECURITY MATTER karti hai
 *   - Kyunki attackers INKE LIYE PRACTICE MEIN COLLISIONS create kar sakte hain
 *   - Aap unhe PURANE systems mein ABHI BHI DEKH SAKTE HO
 *   - Lekin NAYE DESIGNS ko SHA-256 ya KOI MODERN ALTERNATIVE prefer karna chahiye
 *
 * ============================================================================
 *  HMAC AND DIGITAL SIGNATURES
 * ============================================================================
 *
 * Agar aapko PROVE karna hai ki:
 *   - Data EK TRUSTED PARTY se aaya hai
 *   - Aur ise MODIFY NAHI kiya gaya hai
 * - Toh ek PLAIN CHECKSUM ya HASH KAFI NAHI hai
 *
 * ============================================================================
 *  HMAC KAB USE KARO
 * ============================================================================
 *
 * HMAC use karo jab DONO SIDES ek SHARED SECRET KEY rakhte hain
 *
 * HMAC KAISE KAAM KARTA HAI:
 *   - Ek SHARED SECRET KEY use karta hai
 *   - DONO sender aur receiver KE PAAS SAME KEY hoti hai
 *   - Sender DATA aur KEY se ek TAG calculate karta hai
 *   - Phir DATA KE SAATH TAG bhejta hai
 *
 * HMAC KA FAYDA:
 *   - Kyunki HMAC verify karne ke liye WOHI KEY chahiye jo usse banane ke liye thi
 *   - KOI BHI jo HMAC verify kar sakta hai, woh use CREATE bhi kar sakta hai
 *   - Isliye HMAC DO PARTIES ke liye achha hai jo PEHLE SE EK DUSRE PE BHAROSA
 *     karte hain
 *   - Jaise ek WEBHOOK PROVIDER aur uske CALLS RECEIVE karne wala BACKEND
 *
 * HMAC KI LIMITATION:
 *   - Ek HMAC PROVE karta hai ki data KISI AISA SHARED SECRET HOLDER se aaya hai
 *   - Yeh EXACTLY YEH PROVE NAHI KARTA ki US HOLDERS MEIN SE KAUN USE BANAYA
 *
 * ============================================================================
 *  DIGITAL SIGNATURE KAB USE KARO
 * ============================================================================
 *
 * Digital signature use karo jab:
 *   - EK PARTY data ko SIGN kare
 *   - AUR BAHUT SAARI DOOSRI PARTIES ko use VERIFY karna ho
 *
 * DIGITAL SIGNATURE KAISE KAAM KARTA HAI:
 *   - Yeh ek KEY PAIR use karta hai
 *   - Signer ek PRIVATE KEY rakhta hai aur DATA ko SIGN karta hai
 *   - Baaki SAB log MATCHING PUBLIC KEY se signature VERIFY karte hain
 *   - Verifiers signatures CREATE NAHI kar sakte kyunki unhe PRIVATE KEY
 *     KABHI NAHI MILTI
 *
 * FAYDA:
 *   - Yehi cheez EK PUBLISHER ko SOFTWARE RELEASE sign karne deti hai
 *   - AUR MILLIONS OF PEOPLE usse SAFELY verify kar sakte hain
 *
 * COST (Nuksan):
 *   - KEY MANAGEMENT hai
 *   - Verifiers ko TRUST karna padta hai ki PUBLIC KEY SACHME PUBLISHER ki hai
 *
 * ============================================================================
 *  EXAMPLES (HMAC aur Signatures Ke)
 * ============================================================================
 *
 *   1. Signed package repositories (signed packages)
 *   2. Signed container images (signed containers)
 *   3. Webhook signatures (webhook ki jaanch)
 *   4. Request signatures (request ki jaanch)
 *   5. Software update manifests (software updates)
 *
 * ============================================================================
 *  IMPORTANT - HMACs AND SIGNATURES DONO CHEEZEIN EK SAATH PROVE KARTE HAIN
 * ============================================================================
 *
 *   1. DATA CHANGE NAHI hua
 *   2. Aur yeh KISI TRUSTED PERSON se aaya hai
 *
 * ============================================================================
 *  5. ALGORITHM COMPARISON (Sab Algorithms Ka Comparison)
 * ============================================================================
 *
 *   ┌───────────────────┬──────────────────────────────────┬────────────────────────────────────────┐
 *   │ Mechanism         │ Good For                         │ Not Good For                           │
 *   ├───────────────────┼──────────────────────────────────┼────────────────────────────────────────┤
 *   │ Parity            │ Very cheap single-bit error      │ Multi-bit corruption, files,           │
 *   │                   │ detection                        │ deliberate tampering                   │
 *   ├───────────────────┼──────────────────────────────────┼────────────────────────────────────────┤
 *   │ Additive checksum │ Simple protocol checks           │ Strong error detection or security     │
 *   ├───────────────────┼──────────────────────────────────┼────────────────────────────────────────┤
 *   │ CRC-32 / CRC-32C  │ Accidental corruption in frames, │ Malicious tampering                    │
 *   │                   │ blocks, files                    │                                        │
 *   ├───────────────────┼──────────────────────────────────┼────────────────────────────────────────┤
 *   │ SHA-256           │ Strong content digest and file   │ Proving who produced the digest        │
 *   │                   │ verification                     │                                        │
 *   ├───────────────────┼──────────────────────────────────┼────────────────────────────────────────┤
 *   │ HMAC              │ Integrity and authenticity with  │ Public verification without sharing    │
 *   │                   │ a shared secret                  │ the secret                             │
 *   ├───────────────────┼──────────────────────────────────┼────────────────────────────────────────┤
 *   │ Digital signature │ Proving data came from a trusted │ Very high-throughput per-packet checks │
 *   │                   │ signer                           │                                        │
 *   └───────────────────┴──────────────────────────────────┴────────────────────────────────────────┘
 *
 * ============================================================================
 *  COMMON MISTAKE (Sabse Aam Galti)
 * ============================================================================
 *
 * SABSE AAM GALTI:
 *   - Ek AISA TOOL CHUNNA jo ALAG PROBLEM SOLVE KARTA HO
 *
 * Example:
 *   - CRC DISK ya NETWORK CORRUPTION detect karne ke liye EXCELLENT hai
 *   - Lekin yeh SOFTWARE VERIFY karne ke liye GALAT TOOL hai
 *     jab woh UNTRUSTED MIRROR se aaye
 *   - Jab tak EXPECTED DIGEST ALAG SE TRUSTED na ho
 *
 * ============================================================================
 *  6. WHERE CHECKSUMS ARE USED (Checksums Kahan Use Hote Hain)
 * ============================================================================
 *
 * Checksums aur hashes SYSTEM KE HAR LAYER pe DIKHTE HAIN
 *
 * Diagram mein MAIN PLACES dikhaye gaye hain:
 *   - Networking: Ethernet FCS, TCP/UDP checksum
 *   - Storage: Disk blocks, block journals
 *   - Databases: Page checksums, replication
 *   - Object Storage: Upload integrity, restore paths
 *   - Packages: Disk SHA-256, container images
 *   - Distributed Systems: Merkle trees, replica sync
 *
 * ============================================================================
 *  NETWORKING (Network Layer Pe Checksums)
 * ============================================================================
 *
 * ETHERNET FRAMES:
 *   - Ek FRAME CHECK SEQUENCE include karte hain
 *   - Commonly ek CRC
 *   - Agar frame WIRE pe damage ho jaaye, toh RECEIVER use DISCARD kar sakta hai
 *
 * IPv4:
 *   - Apne HEADER ke liye ek CHECKSUM rakhta hai
 *
 * IPv6:
 *   - Us IP-LAYER CHECKSUM ko HATA DETA hai
 *   - Lower aur higher layers pe depend karta hai uske bajaye
 *
 * TCP aur UDP:
 *   - Inke paas bhi CHECKSUMS hote hain
 *   - Yeh COVER karte hain:
 *     * Packet header
 *     * Payload
 *     * Kuch IP information jo protocol use karta hai
 *   - Yeh checks ACCIDENTAL DAMAGE pakadte hain, ATTACKS NAHI
 *   - STRONGER PROTECTION in se aati hai:
 *     * TLS
 *     * QUIC
 *     * IPsec
 *     * Ya application-level signatures
 *
 * ============================================================================
 *  STORAGE SYSTEMS (Storage Pe Checksums)
 * ============================================================================
 *
 * Storage systems CHECKSUMS use karte hain CORRUPTION DETECT karne ke liye:
 *   - Jab data STORE hota hai
 *   - Aur jab data MOVE hota hai
 *
 *   1. Disk blocks
 *   2. Filesystem blocks
 *   3. Database pages
 *   4. Write-ahead logs
 *   5. Object storage parts
 *   6. Backup chunks
 *   7. Recovery fragments jo missing data REBUILD kar sakte hain
 *
 * ============================================================================
 *  IMPORTANT - MODERN STORAGE SYSTEMS CHECKSUMS KO METADATA MEIN RAKHTE HAIN
 * ============================================================================
 *
 *   - Yeh SIRF DATA BLOCK PE TRUST NAHI karte
 *   - Agar ek disk GALAT BLOCK return kare BINA I/O ERROR report kiye
 *   - Toh STORED CHECKSUM ABHI BHI MISMATCH PAKAD SAKTA HAI
 *
 * ============================================================================
 *  DATABASES (Database Pe Checksums)
 * ============================================================================
 *
 * Databases PAGES aur LOGS pe CHECKSUMS use karte hain DETECT karne ke liye:
 *   - Torn writes (aadha likha data)
 *   - Disk corruption (disk kharab hona)
 *   - Bad replication data (galat replicate data)
 *
 * Agar database ek page read kare aur CHECKSUM FAIL ho jaaye:
 *   - Ise PAGE VALID PRETEND NAHI KARNA CHAHIYE
 *   - System ke hisaab se woh:
 *     * Ek REPLICA se read kar sakta hai
 *     * BACKUP se restore kar sakta hai
 *     * PROCESS ROK sakta hai
 *     * Ya page ko CORRUPT mark kar sakta hai
 *
 * ============================================================================
 *  OBJECT STORAGE AND UPLOADS
 * ============================================================================
 *
 * Object stores AKSAR clients ko CHECKSUM UPLOAD KE SAATH bhejne dete hain
 * - Service APNA KHUD KA CHECKSUM calculate karta hai
 * - Agar values ALAG hain toh OBJECT KO REJECT KARTA HAI
 *
 * ============================================================================
 *  MULTIPART UPLOADS KE LIYE SAAVDHANI (Be Careful!)
 * ============================================================================
 *
 *   - Provider-specific ETAGS HAMESHA POORE OBJECT ka simple MD5 NAHI hote
 *   - INTEGRITY CHECKS ke liye PROVIDER KE DOCUMENTED CHECKSUM FIELDS use karo
 *   - UN FIELDS KO MAT USE KARO jo bas CHECKSUM JAISE DIKHTE HAIN
 *
 * ============================================================================
 *  PACKAGE AND ARTIFACT DISTRIBUTION
 * ============================================================================
 *
 * Package managers, container registries, aur release systems use karte hain:
 *   - CRYPTOGRAPHIC HASHES
 *   - Released files IDENTIFY karne ke liye
 *
 * Example:
 *   sha256sum app-linux-amd64.tar.gz
 *
 * ============================================================================
 *  IMPORTANT - DIGEST TAB HI HELP KARTA HAI JAB EXPECTED VALUE TRUSTED HO
 * ============================================================================
 *
 * Expected value KOI NA KOI AISA JAGAH SE AANI CHAHIYE JO AAP TRUST KARTE HO:
 *   - Ek signed release manifest
 *   - Package index
 *   - Transparency log
 *   - Secure website
 *
 * ============================================================================
 *  DISTRIBUTED SYSTEMS
 * ============================================================================
 *
 * Checksums aur hashes NODES ko DATA COMPARE karne mein madad karte hain
 * - BINA har byte PEHLE BHEJE
 *
 * EXAMPLES:
 *   1. Replication systems LOG SEGMENTS ke liye checksums COMPARE karte hain
 *
 *   2. Replica repair HASHES ya MERKLE TREES use kar sakta hai
 *      un ranges dhoondne ke liye jo ALAG hain
 *
 *   3. Content-addressable storage DATA KO DIGEST SE NAME karta hai
 *
 *   4. Deduplication systems IDENTICAL CHUNKS DO BAAR STORE hone se bachate hain
 *
 *   5. Backup systems VERIFY karte hain ki RESTORED DATA WOHI hai jo LIKHA gaya tha
 *
 * ============================================================================
 *  IMPORTANT - IS LAYER PE CHECKSUMS SIRF LOW-LEVEL PLUMBING NAHI HAIN
 * ============================================================================
 *
 * Yeh ek DESIGN TOOL ban jaate hain DIFFERENCES SASTE MEIN DHOONDNE ke liye
 *
 * ============================================================================
 *  7. END-TO-END INTEGRITY (End Se End Tak Suraksha)
 * ============================================================================
 *
 * EK COMMON MISTAKE:
 *   - Yeh MANNA ki EK LAYER ka checksum POORE SYSTEM KO PROTECT KARTA HAI
 *   - YEH NAHI KARTA!
 *
 * EXAMPLE:
 *   - Ek NETWORK FRAME CHECKSUM dikha sakta hai ki ek frame EK LINK pe
 *     INTACT AAYA
 *   - Lekin yeh PROVE NAHI KARTA ki:
 *     * Application ne SAHI OBJECT LIKHA
 *     * Proxy ne SAHI FILE RETURN KI
 *     * Storage ne MAHINO BAAD BLOCK INTACT RAKHA
 *
 * ============================================================================
 *  IMPORTANT DATA KE LIYE - END-TO-END INTEGRITY USE KARO
 * ============================================================================
 *
 *   1. PRODUCER pe ek CHECKSUM ya DIGEST COMPUTE karo
 *
 *   2. Ise TRUSTED METADATA ke saath STORE karo
 *
 *   3. HAR IMPORTANT BOUNDARY pe VERIFY karo:
 *      * Upload
 *      * Replication
 *      * Storage
 *      * Restore
 *      * Download
 *
 *   4. Jab verification FAIL ho toh DOOSRE COPY SE REPAIR karo
 *
 * Diagram flow:
 *   Producer (compute digest) -> Upload (verify) -> Storage (verify on
 *   write/read) -> Replication (verify chunks) -> Consumer (verify final bytes)
 *
 * ============================================================================
 *  DISTRIBUTED STORAGE SYSTEMS KYUN CARE KARTE HAIN
 * ============================================================================
 *
 * Distributed storage systems ISI LIYE CARE karte hain
 * - Kyunki corruption BAHUT JAGAH HO SAKTA HAI:
 *     1. Client memory
 *     2. Network cards
 *     3. Kernel buffers
 *     4. Disks
 *     5. SSD firmware
 *     6. Replication jobs
 *     7. Compaction
 *     8. Backup pipelines
 *     9. Restore tooling
 *
 * ============================================================================
 *  8. CHECKSUMS vs ENCRYPTION (Fark Kya Hai?)
 * ============================================================================
 *
 * Checksums aur encryption ALAG ALAG problems solve karte hain
 *
 *   - CHECKSUMS: CHANGES DETECT karte hain (badlav pakadte hain)
 *
 *   - ENCRYPTION: CONTENT ko un parties se HIDE karta hai jinke paas KEY NAHI hai
 *
 * ============================================================================
 *  MODERN SECURE PROTOCOLS DONO DETE HAIN
 * ============================================================================
 *
 * Modern secure protocols aksar DONO dete hain:
 *   - ENCRYPTION
 *   - INTEGRITY CHECKS
 *
 * Phir bhi IDEAS ALAG hain:
 *   - ENCRYPTION data ko HIDE karta hai
 *   - CHECKSUM check karta hai ki data BADLA hai ya NAHI
 *   - CHECKSUM KOI CHEEZ HIDE NAHI KARTA
 *
 * ============================================================================
 *  PRACTICE MEIN KYA USE KARO (In Practice)
 * ============================================================================
 *
 *   1. CRCs use karo FAST ACCIDENTAL CORRUPTION DETECTION ke liye
 *
 *   2. SHA-256 ya koi MODERN HASH use karo CONTENT IDENTITY ke liye
 *
 *   3. HMACs ya SIGNATURES use karo jab aapko PROVE karna ho ki data KAUN NE
 *      BANAYA
 *
 *   4. TLS ya koi SECURE PROTOCOL use karo PROTECTED COMMUNICATION ke liye
 *
 *   5. ENCRYPTION AT REST use karo jab STORED DATA PRIVATE rehna chahiye
 *
 * ============================================================================
 *  9. DESIGN CONSIDERATIONS (Design Karte Waqt Kya Sochein)
 * ============================================================================
 *
 * System mein CHECKSUMS add karna KUCH PRACTICAL SAWAAL UTHATA HAI:
 *
 *   1. Ek checksum KITNA DATA cover karna chahiye?
 *
 *   2. Checksum KAHAN rehna chahiye?
 *
 *   3. System KOAB verify karna chahiye?
 *
 *   4. Jab MATCH NAHI ho, toh KYA HONA CHAHIYE?
 *
 * YEH CHOICES decide karti hain ki checksums SACHME DATA PROTECT karte hain
 * ya bas OVERHEAD add karte hain
 *
 * ============================================================================
 *  CHOOSE THE RIGHT GRANULARITY (Sahi Size Chuno)
 * ============================================================================
 *
 * Example:
 *   - Poore 10 GB object ka checksumming aapko batayega ki KUCH GALAT hai
 *   - Lekin yeh NAHI batayega ki KONSA HISSA galat hai
 *
 *   - 4 MB CHUNKS ka checksumming ek system ko SIRF DAMAGED CHUNK RETRY ya
 *     REPAIR karne deta hai
 *
 * ============================================================================
 *  SMALLER CHUNKS KA TRADEOFF
 * ============================================================================
 *
 *   - Chhote chunks REPAIR ko ZYADA PRECISE banate hain
 *   - Lekin yeh ZYADA METADATA aur VERIFICATION ke dauraan ZYADA KAAM bhi
 *     create karte hain
 *
 * ============================================================================
 *  STORE CHECKSUMS WHERE THEY CATCH THE FAILURE (Wahan Rakho Jahan Fail Ho)
 * ============================================================================
 *
 * Agar DATA aur CHECKSUM EK HI TARIKE SE DONO DAMAGE ho jaayein:
 *   - Verification MAY NOT HELP
 *
 * Storage engines aksar checksums rakhte hain:
 *   - Page headers mein
 *   - Metadata blocks mein
 *   - Manifests mein
 *   - Separate indexes mein
 *   - Yeh depend karta hai ki KAUNSI FAILURES pakadni hain
 *
 * DESIGN SAWAAL (Simple hai):
 *   "Aap KYA FAILURES pakadne ki koshish kar rahe ho?"
 *
 * ============================================================================
 *  VERIFY ON BOTH WRITE AND READ (Dono Taraf Verify Karo)
 * ============================================================================
 *
 *   - Sirf WRITES ke dauraan verify karna UPLOD ya TRANSMISSION ERRORS pakadta hai
 *   - READS pe verify karna BIT ROT aur SILENT CORRUPTION pakadta hai
 *     jo data STORE hote waqt hua
 *
 * COLD DATA (thanda data):
 *   - Peridic checking chahiye kyunki ise MAHINO tak READ NAHI kiya ja sakta
 *
 * IMPORTANT QUOTE:
 *   "A backup that is never restored or verified is only a theory."
 *   (Woh backup jise KABHI RESTORE ya VERIFY NAHI kiya gaya, woh sirf EK
 *    SOCH hai, asli nahi!)
 *
 * ============================================================================
 *  DO NOT IGNORE MISMATCHES (Mismatch Ko Ignore Mat Karo!)
 * ============================================================================
 *
 * Ek CHECKSUM MISMATCH sirf EK WARNING NAHI hai jise LOG karke IGNORE kar dein
 * - Iska matlab hai ki BYTES WOH NAHI HAIN jo is LAYER NE EXPECT kiye the
 *
 * ============================================================================
 *  POSSIBLE RESPONSES (Kya Kar Sakte Ho)
 * ============================================================================
 *
 *   1. Transfer RETRY karo (phir se bhejo)
 *
 *   2. Doosra REPLICA read karo (alag copy padho)
 *
 *   3. Recovery fragments se data RECONSTRUCT karo
 *
 *   4. Object ko QUARANTINE karo (alag kar do)
 *
 *   5. Backup ko INVALID mark karo (awaahya ghoshit karo)
 *
 *   6. Ek OPERATOR ko ALERT karo (soochna do)
 *
 * ============================================================================
 *  IMPORTANT WARNING
 * ============================================================================
 *
 * Corrupt data ke saath CONTINUE karna AKSAR:
 *   - Ek CONTAINED DATA-INTEGRITY FAILURE ko
 *   - Ek BADE INCIDENT mein badal deta hai
 *
 * ============================================================================
 *  SUMMARY (Saransh)
 * ============================================================================
 *
 * 1. CHECKSUMS COMPACT VALUES hain jo use DETECT karne ke liye hote hain ki
 *    DATA BADLA HAI ya NAHI
 *    - Ek MATCH ka matlab hai ki KOI CHANGE DETECT NAHI hua
 *    - Yeh PROVE NAHI karta ki CHANGE IMPOSSIBLE tha
 *
 * 2. CHECKSUM KA TYPE MATTER KARTA HAI:
 *    - CRCs ACCIDENTAL CORRUPTION ke liye STRONG TOOLS hain
 *    - Lekin DELIBERATE TAMPERING ke liye NAHI
 *    - SHA-256 jaise CRYPTOGRAPHIC HASHES CONTENT IDENTITY aur FILE
 *      VERIFICATION ke liye BEHTAR hain
 *    - HMACs aur DIGITAL SIGNATURES PROOF add karte hain ki data KISI
 *      TRUSTED PERSON se aaya hai
 *
 * 3. DETECTION aur RECOVERY ALAG ALAG KAAM HAIN:
 *    - Checksums aapko batate hain ki CORRUPTION HUI
 *    - Replicas, retransmission, repair fragments, aur backups woh hain jo
 *      DATA RECOVER karte hain
 *
 * 4. USEFUL HONE KE LIYE VERIFICATION ko yahan HONA CHAHIYE:
 *    - SYSTEM BOUNDARIES pe
 *    - Aur FAILURE RATES DEKHEIN
 *    - Ek checksum sirf storage ke andar reh kar tab HI USEFUL hai jab us
 *      layer pe loss ki sambhavna ho
 *    - NAHI TOH WOH SIRF HARMLESS NOISE hai
 *
 * ============================================================================
 **/
```
