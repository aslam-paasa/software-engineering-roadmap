```python
/**
 * ============================================================================
 *  OSI MODEL - Complete Simple Notes (Hinglish)
 *  Beginners ke liye pura detail mein samjhein
 * ============================================================================
 *
 * ============================================================================
 *  INTRO: OSI Model Kya Hai Aur Kyun Zaroori Hai?
 * ============================================================================
 *
 * Jab koi engineer bolta hai "The network is broken", toh pehla sawaal hona
 * chahiye: "KIS HISSE mein problem hai?" Yeh sawaal OSI model se milta hai.
 *
 * - OSI ka full form: "Open Systems Interconnection" model
 * - Isne network communication ko 7 hisson (layers) mein baanta hai
 * - Sabse neeche: raw electrical ya radio signals (jo taar ya hawa mein
 *   chalte hain)
 * - Sabse upar: application protocols (jaise HTTP aur DNS - matlab websites
 *   chalana ya names resolve karna)
 *
 * - ISO ne yeh model 1980s mein banaya tha, taaki sab log ek jaisi baat
 *   karein jab network ki baat hoti hai (common vocabulary)
 *
 * - Aaj kal zyada tar systems literally OSI implement nahi karte
 * - Internet usually TCP/IP model se describe hota hai (sirf 4 layers)
 * - Modern protocols bhi OSI boundaries ke around blur ho jaate hain:
 *     * TLS transport aur application ke beech baithta hai
 *     * QUIC (Google ka protocol) UDP ke upar transport ka kaam karta hai
 *     * Service meshes Layer 4 aur Layer 7 dono pe kaam karte hain
 *
 * - Lekin OSI model phir bhi BAHUT USEFUL hai kyunki yeh diagnostic clarity
 *   deta hai
 * - "The network is broken" vague hai
 * - OSI isko sharp banata hai: "Kya yeh cable problem hai? Local network
 *   problem? Routing problem? TCP problem? TLS problem? Ya application
 *   problem?"
 *
 * Mnemonic (yaad rakhne ka tarika, neeche se upar):
 *   "Please Do Not Throw Sausage Pizza Away"
 *   = Physical, Data Link, Network, Transport, Session, Presentation, Application
 *
 * ============================================================================
 *  THE 7 LAYERS (Bottom to Top - Neeche se Upar)
 * ============================================================================
 *
 * Model neeche se upar jaata hai. Har layer ek kaam karti hai aur result agle
 * layer ko deti hai. Neeche ki layer mein problem hai toh upar ki layer mein
 * confusing behaviour dikhega.
 *
 * ============================================================================
 *  LAYER 1: PHYSICAL (Physical Layer - Taar, Cable, Signal)
 * ============================================================================
 *
 * Kaam: Raw bits (sirf 0 aur 1) ko ek physical medium se move karna
 *
 * Medium matlab physical cheez jisse data chalta hai:
 *   - Copper taar (electrical signals)
 *   - Fiber optic cable (light ki roshni)
 *   - Radio waves (WiFi aur mobile/cellular networks)
 *
 * Layer 1 ko HTTP, IP addresses, ports, ya files se koi matlab nahi hai
 * Iska kaam SIRF itna hai ki signal itni achhi chale ki agla layer samajh sake
 *
 * Layer 1 ki concerns (chhejain jinhe handle karta hai):
 *   1. Cable aur connector standards (kaisi taar, kaisa plug)
 *   2. Optical transceivers (jo electrical signal ko light mein aur wapas
 *      change karte hain)
 *   3. Radio frequencies aur signal modulation (kitni frequency pe bhejna hai)
 *   4. Signal strength aur noise (kitni taakat se signal hai, kitna
 *      interference hai)
 *   5. Link speed negotiation (dono sides agree karein kitni speed pe
 *      chalega)
 *   6. Bit timing aur synchronization (har bit sahi time pe pahunche)
 *
 * Kaisa dikhta hai: Host A -> Network Interface -> Copper/Fiber/Radio ->
 * Network Interface -> Host B
 *
 * Medium ki Types:
 *   ┌──────────────────┬────────────────────────────────────────┬─────────────────────────┐
 *   │ Medium           │ Kya Dekhna Chahiye (What to Watch)     │ Typical Use             │
 *   ├──────────────────┼────────────────────────────────────────┼─────────────────────────┤
 *   │ Twisted-pair     │ Cable category, length, interference,  │ Offices, racks,         │
 *   │ Ethernet         │ negotiated speed                       │ edge devices            │
 *   ├──────────────────┼────────────────────────────────────────┼─────────────────────────┤
 *   │ Fiber optic      │ Transceiver type, wavelength, light    │ Data centers, metro     │
 *   │                  │ level, connector quality               │ links, long-haul links  │
 *   ├──────────────────┼────────────────────────────────────────┼─────────────────────────┤
 *   │ WiFi             │ Signal strength, channel contention,   │ Laptops, phones, local  │
 *   │                  │ interference, roaming                  │ wireless access         │
 *   ├──────────────────┼────────────────────────────────────────┼─────────────────────────┤
 *   │ Cellular         │ Coverage, radio conditions, carrier    │ Mobile and IoT clients  │
 *   │                  │ NAT, variable latency                  │                         │
 *   └──────────────────┴────────────────────────────────────────┴─────────────────────────┘
 *
 * IMPORTANT: Agar Layer 1 fail ho rahi hai toh upar ki layers ka koi fix
 * kaam nahi karega. Kharab cable, tooti fiber, weak WiFi signal, ya overloaded
 * radio channel aise dikhte hain jaise random application instability hai -
 * jab tak aap link check nahi karte!
 *
 * ============================================================================
 *  LAYER 2: DATA LINK (Local Network - Ghar ya Office ke andar)
 * ============================================================================
 *
 * Kaam: Raw bits ko "frames" mein convert karna aur local network ke andar
 * delivery karna
 *
 * Ethernet aur WiFi common examples hain
 *
 * Layer 2 "MAC Addresses" laya hai:
 *   - MAC address ek interface ki local identity hai (hardware ID jaisa)
 *   - Usually 48-bit value hota hai jaise: 00:1A:2B:3C:4D:5E
 *   - Purane explanations kehte hain yeh permanent hardware address hai
 *   - Lekin real systems mein: yeh operating system, hypervisor, container,
 *     ya cloud platform se configure, virtualize, ya randomize ho sakta hai
 *
 * Layer 2 ki responsibilities (kaam):
 *   1. Framing data (raw bits ko structured frames mein baantna)
 *   2. MAC addresses se local addressing
 *   3. Checksums (jaise Ethernet FCS) se corrupted frames detect karna
 *   4. Shared medium pe access control (kaun baat kar sakta hai)
 *   5. Local network ke andar frames switch karna
 *
 * Frame ka structure (data ka dhancha):
 *   Preamble (7 bytes) -> Destination MAC (6 bytes) -> Source MAC (6 bytes)
 *   -> EtherType (2 bytes) -> Payload (actual data) -> FCS (4 bytes)
 *
 * Devices:
 *   - Ethernet switch: Common Layer 2 device. Seekhta hai ki kaunsa MAC
 *     address kis port pe hai, phir frames usi port pe bhejta hai.
 *   - Hub: Har port pe signal repeat karta hai (smart nahi hai, essentially
 *     Layer 1 device)
 *
 * NOTE: Bahut saare modern switches Layer 3 routing features bhi dete hain,
 * toh "switch" product class ka naam hai, strict OSI layer nahi
 *
 * Layer 2 SIRF LOCAL hai:
 *   - Wahi link ya VLAN (Virtual LAN) pe doosre device tak frame deliver
 *     kar sakta hai
 *   - Agar alag network pe jaana hai toh Layer 3 chahiye
 *
 * Important Concepts:
 *   ┌─────────────────┬──────────────────────────────────────────────┐
 *   │ Concept         │ Role (Kaam)                                  │
 *   ├─────────────────┼──────────────────────────────────────────────┤
 *   │ MAC address     │ Ek interface ki local link address           │
 *   │ Switch          │ Local network ke andar frames forward karta  │
 *   │ VLAN            │ Shared infrastructure pe local networks      │
 *   │                 │ ko separate rakhta hai                       │
 *   │ FCS             │ Corrupted Ethernet frames detect karta hai   │
 *   └─────────────────┴──────────────────────────────────────────────┘
 *
 * ============================================================================
 *  LAYER 3: NETWORK (Routing - Ek Network se Dusre Network)
 * ============================================================================
 *
 * Kaam: Traffic ko ek network se dusre network tak route karna
 *
 * Yahi woh layer hai jisse:
 *   - Mumbai ka laptop Virginia (USA) mein API server se baat kare
 *   - Ek cloud VPC ke subnet se doosre region ke database subnet tak
 *     service pahunche
 *
 * Layer 3 IP Addresses use karti hai:
 *   - IPv4 (purana, jaise 192.168.1.10)
 *   - IPv6 (naya, lamba address)
 *   - Routers destination IP address dekhte hain aur har packet ko agle
 *     hop (next stop) pe bhejte hain
 *
 * Layer 3 ki responsibilities:
 *   1. IP addressing (IP address dena)
 *   2. Routing between networks (beech ka raasta decide karna)
 *   3. Packet forwarding (packets aage bhejna)
 *   4. ICMP se network reachability aur error reporting (Ping, Traceroute)
 *   5. Packet size discovery aur fragmentation behaviour
 *
 * Flow: Client (192.168.1.10) -> [Packet to 10.0.0.5] -> Router (Routing
 * Table) -> [Forward toward destination] -> Service (10.0.0.5)
 *
 * Network A: 192.168.1.0/24  |  Network B: 10.0.0.0/24
 *
 * Layer 3 ke Important Protocols:
 *   ┌─────────────────────────┬──────────────────────────────────────────────┐
 *   │ Protocol                │ Purpose (Kaam)                               │
 *   ├─────────────────────────┼──────────────────────────────────────────────┤
 *   │ IPv4 / IPv6             │ Addressing aur packet routing                │
 *   │ ICMP / ICMPv6           │ Error reporting aur diagnostics (Ping,       │
 *   │                         │ Path MTU discovery)                          │
 *   │ ARP                     │ Local network pe IPv4 -> MAC address         │
 *   │                         │ mapping karta hai                            │
 *   │ Neighbor Discovery      │ IPv6 neighbour resolution aur router         │
 *   │                         │ discovery                                    │
 *   │ OSPF / BGP              │ Routing protocols jo reachability            │
 *   │                         │ information exchange karte hain              │
 *   └─────────────────────────┴──────────────────────────────────────────────┘
 *
 * IMPORTANT NOTE about Fragmentation:
 *   - "Layer 3 fragments packets" yeh phrase savdhani se use karo
 *   - IPv4 fragmentation support karta hai lekin ispe rely karna poor design
 *     hai
 *   - IPv6 routers transit mein packets fragment NAHI karte
 *   - Modern systems usually "Path MTU Discovery" pe depend karte hain
 *   - Matlab: path ke liye maximum packet size dhoondhna aur sensible payload
 *     sizing karna
 *
 * Layer 3 ko nahi pata ki host ke andar KONSA PROCESS data receive karega
 * - Yeh kaam Layer 4 ka hai
 *
 * MAC vs IP Address (Difference):
 *   - MAC address: local delivery ke liye ek link pe
 *   - IP address: cross-network routing ke liye
 *
 * ============================================================================
 *  LAYER 4: TRANSPORT (Process to Process - Ports, TCP, UDP)
 * ============================================================================
 *
 * Kaam: Data ko do PROCESSES (programs/apps) ke beech deliver karna
 *
 * Layer 4 PORTS add karta hai:
 *   - Port identify karta hai ki host ke andar kaunsi application hai
 *   - Web server usually port 443 pe listen karta hai
 *   - SSH usually port 22 pe listen karta hai
 *   - Client usually temporary source port use karta hai jo operating system
 *     deta hai
 *
 * Network Flow identify karne ke liye:
 *   Protocol + Source IP + Source Port + Destination IP + Destination Port
 *   = Network Flow (poori pehchaan)
 *
 * Layer 4 ki responsibilities:
 *   1. Ports use karke process-to-process delivery
 *   2. Segmentation aur reassembly (bada data chhote hisson mein todna
 *      aur wapas jodna)
 *   3. Flow control
 *   4. Congestion control (protocol pe depend karta hai)
 *
 * TCP vs UDP (2 Main Protocols):
 *   ┌────────────────────┬──────────────────────────────────┬──────────────────────────────────┐
 *   │ Feature            │ TCP                              │ UDP                              │
 *   ├────────────────────┼──────────────────────────────────┼──────────────────────────────────┤
 *   │ Connection setup   │ Handshake use karta hai          │ Koi connection setup nahi        │
 *   ├────────────────────┼──────────────────────────────────┼──────────────────────────────────┤
 *   │ Delivery model     │ Reliable, ordered byte stream    │ Best effort datagrams            │
 *   │                    │ (jab tak connection healthy hai) │ (pakka nahi)                     │
 *   ├────────────────────┼──────────────────────────────────┼──────────────────────────────────┤
 *   │ Retransmission     │ Built in (apne aap dobara bhejta │ Built in nahi                    │
 *   │                    │ hai)                             │                                  │
 *   ├────────────────────┼──────────────────────────────────┼──────────────────────────────────┤
 *   │ Ordering           │ TCP preserve karta hai           │ Guaranteed nahi                  │
 *   ├────────────────────┼──────────────────────────────────┼──────────────────────────────────┤
 *   │ Congestion control │ Built in                         │ Application ya protocol ko       │
 *   │                    │                                  │ khud handle karna padta hai      │
 *   ├────────────────────┼──────────────────────────────────┼──────────────────────────────────┤
 *   │ Common use         │ HTTP/1.1, HTTP/2, database       │ DNS, QUIC, real-time media,      │
 *   │                    │ connections, SSH                 │ some game traffic                │
 *   └────────────────────┴──────────────────────────────────┴──────────────────────────────────┘
 *
 * TCP ke baare mein IMPORTANT:
 *   - TCP guarantee karta hai ki bytes RELIABLE aur ORDERED deliver ho
 *     jab tak connection healthy hai
 *   - Lekin TCP guarantee NAHI karta ki application ka operation succeed ho
 *   - Agar connection beech mein toot jaaye (mid-request break), toh application
 *     ko khud: timeouts, retries, idempotency, aur duplicate handling chahiye
 *
 * TCP Connection Setup (3-Way Handshake):
 *   1. Client -> Server: SYN (seq=x)
 *   2. Server -> Client: SYN-ACK (seq=y, ack=x+1)
 *   3. Client -> Server: ACK (seq=x+1, ack=y+1)
 *   -> TCP connection established!
 *
 * Port Ranges:
 *   ┌──────────────────┬───────────────────────────┬────────────────────────────────────────────────┐
 *   │ Range            │ Type                      │ Examples                                       │
 *   ├──────────────────┼───────────────────────────┼────────────────────────────────────────────────┤
 *   │ 0-1023           │ Well-known                │ HTTP (80), HTTPS (443), SSH (22)               │
 *   ├──────────────────┼───────────────────────────┼────────────────────────────────────────────────┤
 *   │ 1024-49151       │ Registered                │ PostgreSQL (5432), MySQL (3306), custom        │
 *   │                  │                           │ services                                       │
 *   ├──────────────────┼───────────────────────────┼────────────────────────────────────────────────┤
 *   │ 49152-65535      │ Dynamic / Ephemeral       │ Commonly assigned to client-side connections   │
 *   └──────────────────┴───────────────────────────┴────────────────────────────────────────────────┘
 *
 * Layer 4 Load Balancer:
 *   - Bahut saare production load balancers Layer 4 pe kaam karte hain
 *   - Yeh TCP ya UDP flows forward karta hai
 *   - Lekin isko HTTP routes, headers, cookies, ya request bodies samajh
 *     nahi aati
 *   - Fayda: FAST hai aur general hai
 *   - Nuksan: Application behaviour se aware nahi hai
 *
 * ============================================================================
 *  LAYER 5: SESSION (Conversation Manager)
 * ============================================================================
 *
 * Kaam: Communication sessions ko start karna, chalu rakhna, resume karna,
 * aur band karna
 *
 * Modern internet stacks mein usually SESSION layer ka alag implementation
 * nahi hota
 * Session behaviour usually handle hota hai:
 *   - Application protocols se
 *   - Libraries se
 *   - Frameworks se
 *   - Infrastructure se
 *
 * Lekin yeh concept useful hai kyunki production systems SESSION-LIKE STATE
 * se bharre hain:
 *   - TLS session resume kiya ja sakta hai handshake cost kam karne ke liye
 *   - HTTP client keep-alive ya connection pools se connections reuse kar
 *     sakta hai
 *   - gRPC stream ek long-running operation ke liye khula reh sakta hai
 *   - WebSocket connection ek live client session represent kar sakta hai
 *   - Resumable upload har chunk ke baad progress checkpoint kar sakta hai
 *
 * Session Lifecycle (Diagram):
 *   Establish Session -> Maintain Session -> (Recover after interruption
 *   -> Checkpoint/Resume) ya -> Terminate Session
 *
 * Session Concerns Examples:
 *   ┌─────────────────────┬──────────────────────────────────────────────┐
 *   │ Example             │ Session Concern                              │
 *   ├─────────────────────┼──────────────────────────────────────────────┤
 *   │ TLS                 │ Session resume aur key lifecycle             │
 *   │ WebSocket           │ Long-lived two-way connection                │
 *   │ gRPC streaming      │ Stream lifecycle aur cancellation            │
 *   │ SIP                 │ Voice aur video session setup                │
 *   │ Resumable uploads   │ Checkpoints aur disconnect ke baad recovery  │
 *   └─────────────────────┴──────────────────────────────────────────────┘
 *
 * System Design ke liye IMPORTANT:
 *   - Sawaal yeh hai ki REQUESTS ya CONNECTIONS ke beech kya STATE exist karti
 *     hai
 *   - Aur agar woh state kho jaye toh kya hota hai
 *   - Layer 5 ka logic physically kahan rehta hai usse kam farq padta hai
 *   - State ki LIFECYCLE zyada important hai
 *
 * ============================================================================
 *  LAYER 6: PRESENTATION (Format & Security)
 * ============================================================================
 *
 * Kaam: Data ko aise format mein represent karna ki applications usse samajh
 * paayein (bhejne se pehle ya receive karne ke baad)
 *
 * Yeh include karta hai:
 *   - Encoding (kaise bytes banate hain)
 *   - Serialization (structured data -> bytes)
 *   - Compression (data chhota karna)
 *   - Encryption (data ko lock karna - suraksha)
 *
 * Real Systems mein yeh concerns usually application libraries ya protocol
 * stacks ke andar implement hoti hain, alag layer mein nahi
 *
 * Functions aur Examples:
 *   ┌─────────────────────┬──────────────────────────────────────────────┐
 *   │ Function            │ Examples                                     │
 *   ├─────────────────────┼──────────────────────────────────────────────┤
 *   │ Serialization       │ JSON, Protocol Buffers, Avro, MessagePack   │
 *   │ Text encoding       │ UTF-8                                        │
 *   │ Compression         │ gzip, Brotli, zstd                           │
 *   │ Encryption          │ TLS 1.2, TLS 1.3                             │
 *   │ Media encoding      │ JPEG, PNG, AV1, H.264                        │
 *   └─────────────────────┴──────────────────────────────────────────────┘
 *
 * Data ka Flow (Presentation Work):
 *   Application Data (Objects/Text/Images) -> Serialize (JSON/Protobuf)
 *   -> Compress (gzip/Brotli/zstd) -> Encrypt (TLS) -> Bytes for Transport
 *
 * IMPORTANT - Presentation choices ke REAL consequences hote hain:
 *   - Verbose JSON payload mobile networks pe latency dominate kar sakta hai
 *   - Galat compression setting bandwidth bacha sakta hai lekin CPU burn
 *     karta hai
 *   - TLS misconfiguration clients ko break kar sakta hai ya security weak
 *     kar sakta hai
 *   - Schema change data corrupt kar sakta hai agar producers aur consumers
 *     carefully roll out nahi karte
 *
 * TLS ke baare mein NOTE:
 *   - Yeh modern protocol hai, SSL obsolete hai
 *   - SSL ko new systems ke liye use NAHI karna chahiye
 *
 * ============================================================================
 *  LAYER 7: APPLICATION (User Level - Sabse Upar)
 * ============================================================================
 *
 * Kaam: Application protocols define karte hain ki data ka KYA MATLAB hai
 *
 * Yeh woh layer hai jisse aapka software DIRECTLY baat karta hai:
 *   - HTTP requests (web browsing, API calls)
 *   - DNS queries (website ka naam resolve karna)
 *   - SMTP messages (email bhejna)
 *   - SSH sessions (remote access)
 *   - gRPC calls (service-to-service communication)
 *   - Kafka protocol requests (messaging)
 *   - Database wire protocols (PostgreSQL, MySQL)
 *
 * Layer 7 protocol surface cover karti hai jo applications communication ke
 * liye use karte hain
 * Yeh alag hai un applications ke BUSINESS LOGIC se
 *
 * Common Protocols aur Ports:
 *   ┌──────────────┬──────────────┬──────────────────────────────────────────────┐
 *   │ Protocol     │ Common Port  │ Purpose (Kaam)                               │
 *   ├──────────────┼──────────────┼──────────────────────────────────────────────┤
 *   │ HTTP         │ 80           │ Web aur API traffic (bina TLS)               │
 *   │ HTTPS        │ 443          │ HTTP over TLS; HTTP/2 aur HTTP/3 bhi         │
 *   │ SSH          │ 22           │ Secure remote access                         │
 *   │ SMTP         │ 25           │ Server-to-server email transfer              │
 *   │ DNS          │ 53           │ Domain name resolution                       │
 *   │ PostgreSQL   │ 5432         │ PostgreSQL database connections              │
 *   │ MySQL        │ 3306         │ MySQL database connections                   │
 *   └──────────────┴──────────────┴──────────────────────────────────────────────┘
 *
 * Layer 7 pe kaam karne wale components:
 *   - API Gateways
 *   - Reverse Proxies
 *   - WAFs (Web Application Firewalls)
 *   - Service Meshes
 *
 * Yeh kya karte hain:
 *   - Hostname ya path se route karte hain
 *   - Authentication enforce karte hain
 *   - Headers inspect karte hain
 *   - Rate limits lagate hain
 *   - TLS terminate karte hain
 *   - Selected requests retry karte hain
 *   - Application-level metrics emit karte hain
 *
 * AI Product ke liye Example:
 *   - Layer 7 ek HTTPS endpoint ho sakta hai chat completions ke liye
 *   - Ya gRPC interface ek inference service ke liye
 *   - Ya streaming response Server-Sent Events pe
 *   - Ya database protocol call ek vector store ke liye
 *   - Neeche ke network layers phir bhi matter karte hain, lekin Layer 7
 *     application ke liye CONTRACT define karti hai
 *
 * ============================================================================
 *  ENCAPSULATION AND DECAPSULATION (Data ka Safar)
 * ============================================================================
 *
 * ENCAPSULATION (Data bhejte waqt):
 *   Jab application data bhejta hai, har neeche ki layer apna WRAPPER
 *   (header) lagati hai - isko "encapsulation" kehte hain
 *
 *   HTTPS Request ka flow (Client se API service tak):
 *     1. APPLICATION LAYER: HTTP request banata hai
 *     2. PRESENTATION LAYER: Data format, compress, aur encrypt karta hai
 *     3. TRANSPORT LAYER: TCP header add karta hai -> TCP segment banata hai
 *        (HTTP/3 mein QUIC UDP pe chalta hai)
 *     4. NETWORK LAYER: IP header add karta hai -> IP packet banata hai
 *     5. DATA LINK LAYER: Ethernet ya WiFi header aur trailer add karta hai
 *        -> Frame banata hai
 *     6. PHYSICAL LAYER: Bits ko electrical, optical, ya radio signals mein
 *        bhejta hai
 *
 * DECAPSULATION (Data receive karte waqt):
 *   ULTA hota hai. Har layer apna wrapper padhta hai, samajhta hai, hatata
 *   hai, aur baaki data upar bhejta hai
 *   Plain English mein: Receiver data ko ek layer ek karke UNWRAP karta hai
 *
 * Data Units ka Naam (Har layer alag naam deti hai):
 *   ┌──────────────────────────────────────┬────────────────────────────────┐
 *   │ Layer                                │ Data Unit                      │
 *   ├──────────────────────────────────────┼────────────────────────────────┤
 *   │ Application / Presentation / Session │ Data                           │
 *   │ Transport                            │ TCP segment / UDP datagram     │
 *   │ Network                              │ Packet                         │
 *   │ Data Link                            │ Frame                          │
 *   │ Physical                             │ Bits                           │
 *   └──────────────────────────────────────┴────────────────────────────────┘
 *
 * Yeh naming debugging mein BAHUT useful hai:
 *   - Packet captures, load balancer logs, TCP resets, TLS alerts, HTTP
 *     status codes, aur application exceptions - sab communication path ke
 *     alag layers describe karte hain
 *
 * ============================================================================
 *  OSI MODEL vs TCP/IP MODEL (Difference)
 * ============================================================================
 *
 * - OSI model ek REFERENCE MODEL hai (theory ke liye)
 * - TCP/IP model zyada PRACTICAL hai (yehi internet pe implement hota hai)
 *
 * TCP/IP mein usually 4 layers hoti hain:
 *   1. Application
 *   2. Transport
 *   3. Internet
 *   4. Network Access
 *
 * Mapping (OSI -> TCP/IP):
 *   ┌──────────────────────────────────────┬─────────────────┬──────────────────────────────────┐
 *   │ OSI Layers                           │ TCP/IP Layer    │ Examples                         │
 *   ├──────────────────────────────────────┼─────────────────┼──────────────────────────────────┤
 *   │ Application + Presentation + Session │ Application     │ HTTP, DNS, SSH, TLS, gRPC        │
 *   │ Transport                            │ Transport       │ TCP, UDP, QUIC behaviour over    │
 *   │                                      │                 │ UDP                              │
 *   │ Network                              │ Internet        │ IPv4, IPv6, ICMP                 │
 *   │ Data Link + Physical                 │ Network Access  │ Ethernet, WiFi, fiber, cellular  │
 *   └──────────────────────────────────────┴─────────────────┴──────────────────────────────────┘
 *
 * NOTE: Yeh mapping APPROXIMATE hai. Real protocols hamesha clean layering
 * mein fit nahi hote.
 *
 * ============================================================================
 *  LAYER 5-7 MISCELLANEOUS NOTES
 * ============================================================================
 *
 * Upper layers cover karte hain:
 *   - Session lifecycle aur recovery (Layer 5)
 *   - Data format, compression, serialization, aur encryption (Layer 6)
 *   - Application protocols aur request behaviour (Layer 7)
 *
 * Model ka sabse useful HABIT hai PRECISION (accuracy):
 *   Jab distributed system fail ho, toh LAYER KA NAAM lo aur EVIDENCE dhoondho
 *   - Yeh vague network complaint ko FIXABLE problem mein badal deta hai
 *
 * ============================================================================
 *  HOW ENGINEERS USE OSI IN PRACTICE (Practical Use)
 * ============================================================================
 *
 * OSI model sabse zyada useful hai:
 *   1. Failures diagnose karte waqt
 *   2. Infrastructure choose karte waqt
 *
 * --- TROUBLESHOOTING BY LAYER ---
 *
 * Agar client service tak nahi pahunch pa raha, toh CONCRETE SIGNALS se kaam
 * lo, guess mat karo:
 *
 *   ┌──────────────────────────┬──────────────────────────┬────────────────────────────────────────────┐
 *   │ Symptom                  │ Likely Layer to Check    │ Examples                                   │
 *   │ (Problem)                │ (Konsi Layer Dekhein)    │                                            │
 *   ├──────────────────────────┼──────────────────────────┼────────────────────────────────────────────┤
 *   │ Link down, no carrier,   │ Layer 1                  │ Cable, transceiver, WiFi signal, radio     │
 *   │ weak signal              │                          │ interference                               │
 *   ├──────────────────────────┼──────────────────────────┼────────────────────────────────────────────┤
 *   │ Host cannot reach local  │ Layer 2                  │ VLAN, switch port, MAC table, ARP or       │
 *   │ gateway                  │                          │ Neighbor Discovery                         │
 *   ├──────────────────────────┼──────────────────────────┼────────────────────────────────────────────┤
 *   │ Traffic cannot cross     │ Layer 3                  │ Route tables, security groups, NACLs,      │
 *   │ networks                 │                          │ BGP, ICMP, MTU                             │
 *   ├──────────────────────────┼──────────────────────────┼────────────────────────────────────────────┤
 *   │ Connection refused or    │ Layer 4                  │ Port binding, firewall, load balancer      │
 *   │ timed out                │                          │ listener, TCP reset                        │
 *   ├──────────────────────────┼──────────────────────────┼────────────────────────────────────────────┤
 *   │ TLS handshake fails      │ Layer 6 / Application    │ Certificate, SNI, TLS settings, protocol   │
 *   │                          │ boundary                 │ version, mTLS                              │
 *   ├──────────────────────────┼──────────────────────────┼────────────────────────────────────────────┤
 *   │ HTTP returns 401, 404,   │ Layer 7                  │ Auth, routing, rate limits, internal       │
 *   │ 429, or 503              │                          │ service health, application capacity       │
 *   ├──────────────────────────┼──────────────────────────┼────────────────────────────────────────────┤
 *   │ Request starts but       │ Multiple layers          │ Congestion, connection pool exhaustion,    │
 *   │ stalls under load        │                          │ head-of-line blocking, backend saturation  │
 *   └──────────────────────────┴──────────────────────────┴────────────────────────────────────────────┘
 *
 * IMPORTANT:
 *   - Yeh layered thinking EXPENSIVE WRONG FIXES se bachata hai
 *   - Team ko retry logic rewrite nahi karna chahiye pehle yeh confirm kiye
 *     bina ki load balancer idle connections close kar raha hai
 *   - Database queries tune nahi karna chahiye pehle yeh confirm kiye bina ki
 *     DNS, routing, aur TLS healthy hain
 *
 * --- INFRASTRUCTURE CHOICES ---
 *
 * Layer 5 is purpose also affects design choices:
 *
 *   - LAYER 4 LOAD BALANCER: Jab fast TCP/UDP forwarding chahiye aur request
 *     inspection ki zaroorat nahi hai
 *
 *   - LAYER 7 LOAD BALANCER ya REVERSE PROXY: Jab routing hostnames, paths,
 *     headers, authentication, ya request-level policy pe depend karti hai
 *
 *   - SERVICE MESH: Usually Layer 4 connection handling + Layer 7 policy,
 *     telemetry, retries, aur mTLS combine karta hai
 *
 *   - CDN (Content Delivery Network): Layer 7 pe heavily kaam karta hai lekin
 *     DNS, routing, TLS, caching, aur edge network placement pe depend karta
 *     hai
 *
 *   - DATABASE CONNECTION POOL: Ek application concern hai, lekin failures
 *     aksar Layer 4 connection exhaustion ya timeout ke roop mein surface
 *     hoti hain
 *
 * --- PRACTICAL EXAMPLE: MODEL INFERENCE API ---
 *
 * Ek AI model inference API ka flow (step by step):
 *
 *   1. DNS resolve hota hai "api.example.com" ko kisi edge ya load balancer pe
 *   2. Client ek TCP ya QUIC connection establish karta hai
 *   3. TLS server ko authenticate karta hai aur traffic encrypt karta hai
 *   4. HTTP request ko ek inference service pe route karta hai
 *   5. Service HTTP chunking, Server-Sent Events, WebSockets, ya gRPC
 *      streaming ke through tokens wapas stream kar sakta hai
 *   6. Backend services vector databases, object storage, queues, aur
 *      telemetry systems ko apne protocols se call kar sakte hain
 *
 * Har step alag tarike se fail ho sakta hai
 * OSI model clean way deta hai un failures ke baare mein sochne ke liye bina
 * yeh pretend kiye ki stack simple hai jabki hai nahi
 *
 * ============================================================================
 *  SUMMARY (Saransh)
 * ============================================================================
 *
 * - OSI model networked systems ke liye ek PRACTICAL VOCABULARY hai
 * - Yehi ki production stacks usually TCP/IP-based hote hain aur unse 7
 *   layers se cleanly map nahi hote
 *
 * - Iski VALUE hai CONCERNS ko SEPARATE karne mein:
 *
 *   Layer 1: Signals aur media cover karta hai
 *     - Kaam: bits move karna over physical medium
 *
 *   Layer 2: Local frames aur MAC-level delivery cover karta hai
 *     - Kaam: local network ke andar frames deliver karna
 *
 *   Layer 3: IP addressing aur routing cover karta hai
 *     - Kaam: packets ko cross-network route karna
 *
 *   Layer 4: Ports, TCP, UDP, aur process-to-process delivery cover karta hai
 *     - Kaam: data ko do programs ke beech deliver karna reliably ya fast
 *
 *   Layer 5: Session lifecycle aur recovery cover karta hai
 *     - Kaam: connection start, maintain, resume, terminate karna
 *
 *   Layer 6: Data format, compression, serialization, encryption cover karta hai
 *     - Kaam: data ko readable aur secure format mein convert karna
 *
 *   Layer 7: Application protocols aur request behaviour cover karta hai
 *     - Kaam: data ka meaning define karna (HTTP, DNS, SMTP, etc.)
 *
 * MODEL KA SABSE USEFUL HABIT: PRECISION
 *   Jab distributed system fail ho:
 *     - Layer ka naam lo (identify which layer)
 *     - Evidence dhoondho (look for proof)
 *   Yeh vague "network is broken" complaint ko FIXABLE PROBLEM mein badal deta
 *   hai jise aap actually solve kar sakte ho
 *
 * ============================================================================
 **/
```
