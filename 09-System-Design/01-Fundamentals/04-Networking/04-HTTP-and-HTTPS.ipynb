```python
/**
 * ============================================================================
 *  HTTP / HTTPS - Complete Simple Notes (Hinglish)
 *  Beginners ke liye - Bilkul simple bhasha mein
 * ============================================================================
 *
 * ============================================================================
 *  1. HTTP KYA HAI? (Sabse Pehla Sawaal)
 * ============================================================================
 *
 * Aaj kal har modern system internet pe KUCH NA KUCH HTTP use karta hai
 *
 * HTTP ka full form: "Hypertext Transfer Protocol"
 * - HTTP ek REQUEST-RESPONSE PROTOCOL hai
 * - Matlab: Ek taraf se sawaal jaata hai (request), doosri taraf se jawaab
 *   aata hai (response)
 * - Yeh websites ke peeche kaam karta hai
 * - Public APIs, mobile backends, aur bahut saari internal service-to-service
 *   calls mein use hota hai
 *
 * HTTPS KYA HAI:
 * - HTTPS = HTTP + TLS protection
 * - TLS ka full form: "Transport Layer Security"
 * - TLS connection ko ENCRYPT (lock) karta hai
 * - Aur client ko check karne deta hai ki woh SAHI server se baat kar raha hai
 * - Production mein HTTPS NORMAL DEFAULT hai (har jagah yahi use hota hai)
 *
 * PLAIN HTTP (bina lock wala) kab use hota hai:
 *   - Local development mein (apne computer pe coding)
 *   - Tightly controlled internal health checks mein
 *   - Redirects mein (jo client ko HTTPS version pe le jaate hain)
 * - Koi bhi cheez jo USER-FACING hai ya SECURITY-SENSITIVE hai, use HTTPS
 *   use karna chahiye
 *
 * HTTP ke IDEAS sab versions mein SAME rehte hain:
 *   - Methods (kaam ka tarika)
 *   - Status codes (jawaab ka code)
 *   - Headers (extra jaankari)
 *   - Bodies (data)
 *
 * SIRF NEECHE KA TRANSPORT BADALTA HAI:
 *   - HTTP/1.1 aur HTTP/2 usually TCP ke upar chalte hain
 *   - HTTP/3 QUIC ke upar chalta hai, jo UDP use karta hai
 *
 * ============================================================================
 *  2. HTTP REQUEST AND RESPONSE (Request Aur Response)
 * ============================================================================
 *
 * Ek typical HTTP exchange mein 3 PARTS hote hain:
 *
 *   1. REQUEST LINE (ya naye HTTP versions mein equivalent fields)
 *   2. HEADERS (extra jaankari)
 *   3. OPTIONAL BODY (data, agar zaroorat ho)
 *
 * ----------------------------------------------------------------------------
 *  REQUEST EXAMPLE (Client kya bhejta hai):
 * ----------------------------------------------------------------------------
 *   GET /v1/models?limit=20 HTTP/1.1
 *   Host: api.example.com
 *   Accept: application/json
 *   Authorization: Bearer <token>
 *   User-Agent: example-client/1.4
 *
 * ----------------------------------------------------------------------------
 *  RESPONSE EXAMPLE (Server kya bhejta hai):
 * ----------------------------------------------------------------------------
 *   HTTP/1.1 200 OK
 *   Content-Type: application/json
 *   Cache-Control: private, max-age=60
 *   Date: Mon, 25 May 2026 10:00:00 GMT
 *
 *   {
 *     "data": [
 *       { "id": "model-small", "status": "available" }
 *     ]
 *   }
 *
 * IMPORTANT PARTS KO SAMJHO:
 *   - REQUEST TARGET: Resource identify karta hai (jaise /v1/models)
 *   - HEADERS: Extra jaankari le jaate hain (content type, authorization token)
 *   - BODY: Data carry karta hai jab method allow karta hai
 *     (jaise POST mein JSON)
 *
 * NOTE: HTTP/2 aur HTTP/3 WIRE pe exactly yeh TEXT format NAHI bhejte
 * Lekin SAME CONCEPTS rehte hain:
 *   - Method, scheme, host, path, headers, status, aur body
 *
 * ============================================================================
 *  HTTP DEFINES KYA KARTA HAI (HTTP Kya Meaning Deta Hai)
 * ============================================================================
 *
 * HTTP yeh sab define karta hai:
 *
 *   1. METHODS jaise GET, POST, PUT, aur DELETE
 *
 *   2. STATUS CODES jaise 200, 404, 429, aur 503
 *
 *   3. HEADERS jaise Content-Type, Authorization, Cache-Control, aur ETag
 *
 *   4. MESSAGE BODIES jaise HTML, JSON, images, protobuf, ya event streams
 *
 *   5. CACHING, conditional requests, redirects, content negotiation, aur
 *      authentication hooks
 *
 * ============================================================================
 *  HTTP KAISE KAAM KARTA HAI (Step by Step)
 * ============================================================================
 *
 * Ek HTTP request ka flow:
 *
 *   1. Client hostname ko DNS se resolve karta hai
 *
 *   2. Client ek transport connection open karta hai ya REUSE karta hai
 *
 *   3. HTTPS ke liye, client aur server TLS HANDSHAKE karte hain
 *
 *   4. Client ek HTTP request bhejta hai
 *
 *   5. Server request ko application code ya kisi internal service pe route
 *      karta hai
 *
 *   6. Server ek HTTP response bhejta hai
 *
 *   7. Connection BAAD KE REQUESTS ke liye REUSE ho sakti hai
 *
 * Diagram flow:
 *   Client -> DNS Resolver: "Resolve api.example.com"
 *   DNS Resolver -> Client: IP address
 *   Client -> Server: TCP connect
 *   Client -> Server: TLS handshake (HTTPS)
 *   Client -> Server: HTTP request
 *   Server: Route to application
 *   Server -> Client: HTTP response
 *   -> "Connection may be reused for later requests"
 *
 * ============================================================================
 *  YEH SIMPLE FLOW KE PEECHE BAHUT SAARE COMPONENTS HAIN
 * ============================================================================
 *
 * Production mein yeh sab beech mein hote hain:
 *   - CDNs (Content Delivery Networks)
 *   - WAFs (Web Application Firewalls)
 *   - API gateways
 *   - Reverse proxies
 *   - Service meshes
 *   - Load balancers
 *   - Sidecars
 *   - Application servers
 *
 * ============================================================================
 *  TLS TERMINATION (Traffic Kab Unlock Hota Hai)
 * ============================================================================
 *
 * Yeh components HTTP headers ko TABHI read ya change kar sakte hain:
 *   - Jab traffic DECRYPT ho chuka ho
 *   - Us POINT ko kehte hain "TLS TERMINATION"
 *
 * Us POINT se PEHLE, HTTPS traffic WIRE PE PROTECTED rehta hai
 * (koi beech mein padh nahi sakta)
 *
 * ============================================================================
 *  STATELESS PROTOCOL, STATEFUL SYSTEMS
 * ============================================================================
 *
 * HTTP PROTOCOL LEVEL pe STATELESS hai:
 *   - Matlab har request ko ITNI JAANKARI khud carry karni chahiye
 *     ki server use samajh sake
 *   - HTTP khud YEH NAHI MANTA ki server ko pichhle request ka koi
 *     hidden session yaad hai
 *
 * LEKIN - Yeh matlab NAHI hai ki web systems stateless hain!
 *   - Real systems state rakhte hain:
 *     * Cookies
 *     * Session stores
 *     * OAuth access tokens
 *     * JWTs
 *     * CSRF tokens
 *     * Server-side caches
 *     * Databases
 *   - Woh STICKY LOAD BALANCING ya CONNECTION POOLS bhi use kar sakte hain
 *     taaki requests useful state ke paas rahein
 *
 * IMPORTANT DESIGN SAWAAL (Simple hai):
 *   - STATE KAHAN REHTA HAI?
 *   - Aur KYA HOTA HAI agar:
 *     * Request RETRY ho jaaye
 *     * Request DOOSRE SERVER pe route ho jaaye
 *     * Request CLIENT KE TIMEOUT ke BAAD finish ho
 *
 * ============================================================================
 *  4. METHODS AND STATUS CODES (Kaam Aur Jawaab Codes)
 * ============================================================================
 *
 * HTTP METHODS batate hain ki KYA TYPE KA OPERATION maanga ja raha hai
 * - Yeh clients, proxies, gateways, aur retry logic sabko batata hai
 *
 *   ┌──────────┬──────────────────────────────────┬──────────┬─────────────────┐
 *   │ Method   │ Common Use                       │ Safe?    │ Idempotent?     │
 *   ├──────────┼──────────────────────────────────┼──────────┼─────────────────┤
 *   │ GET      │ Ek resource read karna           │ Yes      │ Yes             │
 *   ├──────────┼──────────────────────────────────┼──────────┼─────────────────┤
 *   │ HEAD     │ Sirf response headers read karna │ Yes      │ Yes             │
 *   ├──────────┼──────────────────────────────────┼──────────┼─────────────────┤
 *   │ POST     │ Resource create karna ya command │ No       │ Not guaranteed  │
 *   │          │ start karna                      │          │                 │
 *   ├──────────┼──────────────────────────────────┼──────────┼─────────────────┤
 *   │ PUT      │ Known URL pe resource replace ya │ No       │ Yes             │
 *   │          │ create karna                     │          │                 │
 *   ├──────────┼──────────────────────────────────┼──────────┼─────────────────┤
 *   │ PATCH    │ Resource ko partially update     │ No       │ Not guaranteed  │
 *   │          │ karna                            │          │                 │
 *   ├──────────┼──────────────────────────────────┼──────────┼─────────────────┤
 *   │ DELETE   │ Resource delete karna            │ No       │ Yes             │
 *   └──────────┴──────────────────────────────────┴──────────┴─────────────────┘
 *
 * "SAFE" ka matlab:
 *   - Client READ kar raha hai, server state CHANGE NAHI kar raha
 *
 * "IDEMPOTENT" ka matlab:
 *   - Ek hi request KO BAAR BAAR bhejne se WOHI effect hoga
 *     jo ek baar bhejne se hota
 *
 * ============================================================================
 *  IDEMPOTENCY KYUN IMPORTANT HAI
 * ============================================================================
 *
 * Idempotency TRIVIAL (chhoti baat) nahi hai
 * - Yeh decide karti hai ki clients SAFELY RETRY kar sakte hain ya nahi
 * - Jab:
 *     * Timeouts ho
 *     * Connection resets ho
 *     * Load balancer failures ho
 *
 * PRACTICAL RULE:
 *   - POST /payments ko usually IDEMPOTENCY KEY chahiye
 *   - GET /orders/123 ko state CHANGE nahi karna chahiye
 *
 * ============================================================================
 *  STATUS CODES (Jawaab Codes)
 * ============================================================================
 *
 * Status codes itne SPECIFIC hone chahiye ki clients aur operators ACTION le
 * sakein:
 *
 *   ┌──────────┬───────────────────────────────┬──────────────────────────────────────┐
 *   │ Range    │ Meaning                       │ Examples                             │
 *   ├──────────┼───────────────────────────────┼──────────────────────────────────────┤
 *   │ 1xx      │ Informational                 │ 100 Continue, 103 Early Hints        │
 *   ├──────────┼───────────────────────────────┼──────────────────────────────────────┤
 *   │ 2xx      │ Success (Kaam safal)          │ 200 OK, 201 Created, 204 No Content  │
 *   ├──────────┼───────────────────────────────┼──────────────────────────────────────┤
 *   │ 3xx      │ Redirect ya alternate         │ 301 Moved Permanently, 302 Found,    │
 *   │          │ location                      │ 304 Not Modified                     │
 *   ├──────────┼───────────────────────────────┼──────────────────────────────────────┤
 *   │ 4xx      │ Client-side problem           │ 400 Bad Request, 401 Unauthorized,   │
 *   │          │ (client ki galti)             │ 403 Forbidden, 404 Not Found,        │
 *   │          │                               │ 409 Conflict, 429 Too Many Requests  │
 *   ├──────────┼───────────────────────────────┼──────────────────────────────────────┤
 *   │ 5xx      │ Server-side ya internal       │ 500 Internal Server Error,           │
 *   │          │ dependency problem            │ 502 Bad Gateway, 503 Service         │
 *   │          │ (server ki problem)           │ Unavailable, 504 Gateway Timeout     │
 *   └──────────┴───────────────────────────────┴──────────────────────────────────────┘
 *
 * IMPORTANT - Status codes ko API CONTRACT ke hisaab se use karo:
 *   - Ek VAGUE 500 validation errors ke liye clients ko GUESS karne pe majboor
 *     karta hai
 *   - Ek 200 with an error object caches, metrics, SDKs, aur retry policies ko
 *     CONFUSE karta hai
 *
 * ============================================================================
 *  5. CACHING AND CONDITIONAL REQUESTS (Caching Kaise Kaam Karta Hai)
 * ============================================================================
 *
 * HTTP ke paas MATURE CACHING RULES hain
 * Agar achhe se use kiya jaaye toh CACHING:
 *   - Systems ko TEZ banata hai
 *   - Origin servers pe load kam karta hai
 *   - Cloud cost kam karta hai
 *   - Partial failures ke dauraan users ko KAAM KARNE deta hai
 *
 * ============================================================================
 *  IMPORTANT CACHING HEADERS
 * ============================================================================
 *
 *   ┌────────────────────┬─────────────────────────────────────────────────────┐
 *   │ Header             │ Purpose                                             │
 *   ├────────────────────┼─────────────────────────────────────────────────────┤
 *   │ Cache-Control      │ Define karta hai ki KAUN cache kar sakta hai aur    │
 *   │                    │ KITNI DER ke liye                                   │
 *   ├────────────────────┼─────────────────────────────────────────────────────┤
 *   │ ETag               │ Cached response ke liye VERSION-LIKE value          │
 *   ├────────────────────┼─────────────────────────────────────────────────────┤
 *   │ If-None-Match      │ Client poochta hai ki ETag abhi bhi CURRENT hai ya  │
 *   │                    │ nahi                                                │
 *   ├────────────────────┼─────────────────────────────────────────────────────┤
 *   │ Last-Modified      │ Timestamp jo check karta hai ki content BADLA hai   │
 *   │                    │ ya nahi                                             │
 *   ├────────────────────┼─────────────────────────────────────────────────────┤
 *   │ Vary               │ Caches ko batata hai ki KAUNSE request headers      │
 *   │                    │ response ko affect karte hain                        │
 *   ├────────────────────┼─────────────────────────────────────────────────────┤
 *   │ Content-Encoding   │ Compression indicate karta hai jaise gzip, br, ya   │
 *   │                    │ zstd                                                │
 *   └────────────────────┴─────────────────────────────────────────────────────┘
 *
 * EXAMPLE - Client unchanged response download karne se BACH SAKTA HAI:
 *
 *   GET /assets/app.js HTTP/1.1
 *   If-None-Match: "a9f31"
 *
 *   HTTP/1.1 304 Not Modified
 *   ETag: "a9f31"
 *
 * ============================================================================
 *  CACHING SIRF BROWSERS KE LIYE NAHI HAI
 * ============================================================================
 *
 * Yeh sab CORRECT CACHE HEADERS se benefit lete hain:
 *   - CDNs
 *   - API gateways
 *   - Package registries
 *   - Feature stores
 *   - Model metadata endpoints
 *   - Documentation sites
 *
 * ============================================================================
 *  PERSONALIZED YA SENSITIVE RESPONSES KE LIYE
 * ============================================================================
 *
 * - EXPLICIT raho (saaf saaf batao)
 * - RESTRICTIVE SETTINGS use karo jahan appropriate ho:
 *     Cache-Control: private
 *     Cache-Control: no-store
 *
 * ============================================================================
 *  6. WHAT HTTPS ADDS (HTTPS Kya Extra Deta Hai)
 * ============================================================================
 *
 * HTTPS = HTTP over TLS
 * - TLS MODERN SECURITY PROTOCOL hai
 * - SSL PURANI TERMINOLOGY hai aur NEW SYSTEMS ke liye use NAHI honi chahiye
 *
 * ============================================================================
 *  HTTPS KE 3 MAIN PROTECTIONS (Suraksha)
 * ============================================================================
 *
 *   1. CONFIDENTIALITY (Guptta):
 *      - Beech mein khade log ya systems PROTECTED HTTP DATA padh NAHI sakte
 *
 *   2. INTEGRITY (Akhandta):
 *      - Beech mein khade log ya systems PROTECTED TRAFFIC change NAHI kar
 *        sakte bina detect hue
 *
 *   3. SERVER AUTHENTICATION (Server ki pehchaan):
 *      - Client check kar sakta hai ki server us HOSTNAME ko use karne ka
 *        ALLOWED hai ya nahi
 *
 * ============================================================================
 *  HTTPS KYA SOLVE NAHI KARTA (Yeh Dhyan Rakho!)
 * ============================================================================
 *
 *   1. Yeh USER ko authenticate NAHI karta
 *      - Jab tak application khud authentication add na kare
 *
 *   2. Yeh ek VULNERABLE API ko SAFE NAHI banata
 *
 *   3. Yeh DESTINATION IP ADDRESS ko hide NAHI karta
 *
 *   4. Yeh hostname ko DNS ke through expose kar sakta hai
 *      - Aur deployment ke hisaab se SNI ke through bhi
 *
 *   5. Yeh ek COMPROMISED ENDPOINT ko data padhne se ROKEGA NAHI
 *
 * ============================================================================
 *  SNI (Server Name Indication)
 * ============================================================================
 *
 * SNI ek TLS FIELD hai:
 *   - Jo client ko batane deta hai ki woh KAUNSA HOSTNAME chahta hai
 *   - Yeh ENCRYPTED HTTP request bhejne se PEHLE hota hai
 *
 * Modern ENCRYPTED-CLIENT-HELLO work kuch deployments mein iska ZYADA
 * hissa hide kar sakta hai
 * Lekin yeh MAT MANO ki HTTPS HAR ROUTING HINT ko hide karta hai
 *
 * ============================================================================
 *  HTTPS BASELINE EXPECTATION HAI
 * ============================================================================
 *
 *   - Browsers PLAIN HTTP pe WARNING dete hain
 *   - Bahut saari platform features SECURE CONTEXT require karti hain
 *   - Production APIs ko HTTPS ko PEHLI DESIGN REVIEW se hi ASSUME karna
 *     chahiye
 *
 * ============================================================================
 *  7. HOW TLS WORKS (TLS Kaise Kaam Karta Hai)
 * ============================================================================
 *
 * HTTPS mein secure connection ek TLS HANDSHAKE se shuru hota hai
 *
 * HANDSHAKE ka matlab:
 *   - Woh SETUP CONVERSATION jahan client aur server agree karte hain
 *     ki connection ko KAISE PROTECT kiya jayega
 *
 * ============================================================================
 *  MODERN TLS 1.3 HANDSHAKE (Step by Step)
 * ============================================================================
 *
 *   1. CLIENTHELLO:
 *      - Client batata hai ki woh KAUNSE TLS VERSIONS aur ENCRYPTION OPTIONS
 *        support karta hai
 *      - SNI aur ALPN jaisi jaankari bhi bhejta hai
 *
 *   2. SERVERHELLO:
 *      - Server is CONNECTION ke liye SHARED SETTINGS choose karta hai
 *
 *   3. CERTIFICATE:
 *      - Server certificates bhejta hai jo PROVE karte hain ki woh us
 *        HOSTNAME ko serve karne ka ALLOWED hai
 *
 *   4. CERTIFICATE VERIFICATION:
 *      - Client check karta hai:
 *        * Hostname
 *        * Expiration date
 *        * Signatures
 *        * Trusted certificate authority chain
 *
 *   5. KEY SETUP:
 *      - DONO sides SHARED ENCRYPTION KEYS banate hain
 *      - BINA final keys network pe bheje
 *
 *   6. ENCRYPTED HTTP:
 *      - HTTP requests aur responses ab ENCRYPTED CONNECTION se flow karte hain
 *
 * Handshake diagram flow:
 *   Client -> Server: ClientHello (supported TLS options, SNI, ALPN)
 *   Server -> Client: ServerHello (chosen TLS options)
 *   Server -> Client: Certificate + signature
 *   Client: "Verify certificate chain and hostname"
 *   Both: "Both sides create shared encryption keys"
 *   Client -> Server: Encrypted HTTP request
 *   Server -> Client: Encrypted HTTP response
 *
 * ============================================================================
 *  IMPORTANT NOTE ABOUT OLD TLS EXPLANATIONS
 * ============================================================================
 *
 * Purani TLS explanations aksar batati hain ki client ek "PRE-MASTER SECRET"
 * ko server ki PUBLIC KEY se encrypt karta hai
 * - Yeh NORMAL TLS 1.3 MODEL NAHI hai
 *
 * MODERN TLS TEMPORARY KEY EXCHANGE use karta hai:
 *   - Commonly ECDHE
 *   - Practical fayda: FORWARD SECRECY
 *   - Matlab: Agar baad mein server ki LONG-TERM PRIVATE KEY chori ho jaaye
 *   - Toh purani RECORDED CONNECTIONS abhi bhi DECRYPT karna MUSHKIL hona
 *     chahiye
 *
 * ============================================================================
 *  ALPN (Application-Layer Protocol Negotiation)
 * ============================================================================
 *
 * TLS ALPN bhi use karta hai:
 *   - Taaki client aur server APPLICATION PROTOCOL pe agree kar sakein
 *   - Jaise http/1.1 ya h2
 *
 * HTTP/3 QUIC use karta hai:
 *   - Jahan TLS 1.3 QUIC handshake mein BUILT-IN hai
 *
 * ============================================================================
 *  0-RTT (Zero Round Trip)
 * ============================================================================
 *
 * TLS 1.3 aur QUIC REPEAT CONNECTIONS ke liye 0-RTT data support kar sakte hain
 *
 * Yeh kya karta hai:
 *   - Client ko BAHUT JALDI kuch data bhejne deta hai
 *   - Isse LATENCY KAM ho sakti hai
 *
 * TRADEOFF (Nuksan):
 *   - REPLAY RISK hai
 *   - Ek attacker us EARLY DATA ko REPLAY kar sakta hai
 *
 * BEST PRACTICE:
 *   - 0-RTT sirf UN operations ke liye use karo jo REPEAT karne ke liye SAFE hain
 *   - Jaise IDEMPOTENT READS
 *
 * ============================================================================
 *  8. HTTP vs HTTPS (Direct Comparison)
 * ============================================================================
 *
 *   ┌─────────────────────┬──────────────────────────┬─────────────────────────────────────┐
 *   │ Feature             │ HTTP                     │ HTTPS                               │
 *   ├─────────────────────┼──────────────────────────┼─────────────────────────────────────┤
 *   │ Protection          │ Plaintext (bina lock)    │ Encrypted and integrity-protected   │
 *   │                     │                          │ with TLS                            │
 *   ├─────────────────────┼──────────────────────────┼─────────────────────────────────────┤
 *   │ Common Port         │ 80                       │ 443                                 │
 *   ├─────────────────────┼──────────────────────────┼─────────────────────────────────────┤
 *   │ Server              │ None by default          │ Certificate-based hostname          │
 *   │ Authentication      │                          │ verification                        │
 *   ├─────────────────────┼──────────────────────────┼─────────────────────────────────────┤
 *   │ Tamper Resistance   │ None                     │ Protected by TLS                    │
 *   ├─────────────────────┼──────────────────────────┼─────────────────────────────────────┤
 *   │ Browser Treatment   │ Many contexts mein       │ Most production web features ke     │
 *   │                     │ insecure mark hota hai   │ liye required                       │
 *   ├─────────────────────┼──────────────────────────┼─────────────────────────────────────┤
 *   │ Production Use      │ Redirects, local         │ Web apps, APIs, mobile backends,    │
 *   │                     │ development, tightly     │ aur service endpoints ke liye       │
 *   │                     │ controlled internal      │ default                             │
 *   │                     │ cases                    │                                     │
 *   └─────────────────────┴──────────────────────────┴─────────────────────────────────────┘
 *
 * ============================================================================
 *  PLAIN HTTP KO PERFORMANCE KE LIYE MAT CHUNO
 * ============================================================================
 *
 * TLS overhead usually CHHOTA hota hai INKE MUQABLE:
 *   - Application work
 *   - Database calls
 *   - Network distance
 *   - Payload size
 *
 * HTTPS AB PRACTICAL DEFAULT BAN CHUKA HAI kyunki:
 *   - Connection reuse
 *   - TLS 1.3
 *   - HTTP/2
 *   - HTTP/3
 *   - Session resumption
 *   - Hardware acceleration
 *
 * ============================================================================
 *  9. HTTP VERSIONS (HTTP Ke Versions)
 * ============================================================================
 *
 * HTTP evolve hua hai bina apna CORE REQUEST-RESPONSE MODEL badle
 *
 * ============================================================================
 *  HTTP/1.1
 * ============================================================================
 *
 * HTTP/1.1 ne yeh add kiya:
 *   - REUSABLE CONNECTIONS ko default banaya
 *   - Hostnames ke around important behaviour
 *   - Caching
 *   - Content negotiation
 *   - Range requests
 *   - Transfer encodings
 *
 * OPERATIONALLY:
 *   - HTTP/1.1 SIMPLE hai aur WIDELY SUPPORTED hai
 *
 * WEAKNESS: CONCURRENCY (ek saath kaam)
 *   - Ek single connection responses ko ORDER mein handle karta hai
 *   - Matlab ek dheemi response us connection pe baad ki responses ko
 *     HOLD kar sakta hai
 *   - Clients aksar same origin ke liye MULTIPLE CONNECTIONS open karke
 *     is problem se bachte hain
 *
 * HTTP PIPELINING:
 *   - Specification mein maujood hai
 *   - Lekin safely deploy karna MUSHKIL tha
 *   - Kabhi normal browser behaviour NAHI bana
 *
 * ============================================================================
 *  HTTP/2
 * ============================================================================
 *
 * HTTP/2 SAME HTTP MEANING rakhta hai
 * Lekin DATA BHEJNE KA TARIKA BADALTA hai:
 *   - Text messages ki jagah BINARY FRAMES use karta hai
 *
 * HTTP/2 KYA ADD KARTA HAI:
 *   1. MULTIPLE STREAMS over ONE TCP connection
 *   2. HEADER COMPRESSION with HPACK
 *   3. STREAM PRIORITIES (though real-world support VARIED hai)
 *   4. BETTER CONNECTION REUSE than HTTP/1.1
 *
 * HTTP/2 HEAD-OF-LINE BLOCKING ko HTTP LAYER pe KAM karta hai
 *   - Kyunki multiple requests EK connection pe active ho sakte hain
 *
 * LEKIN - Yeh TCP-LEVEL HEAD-OF-LINE BLOCKING HATAATA NAHI:
 *   - Agar ek TCP segment kho jaaye
 *   - Toh us TCP connection pe SARI STREAMS WAIT kar sakti hain
 *     jab tak missing bytes recover na hon
 *
 * SIMPLE BHASHA MEIN:
 *   - Ek missing TCP piece UNRELATED HTTP/2 streams ko WAIT karwa sakta hai
 *   - Kyunki sab EK HI TCP connection share karte hain
 *
 * HTTP/2 SERVER PUSH:
 *   - Original design ka hissa tha
 *   - Lekin achhe se use karna MUSHKIL sabit hua
 *   - Major browsers se REMOVE ya DISABLE kar diya gaya
 *   - IMPORTANT: Server push ke around NEW SYSTEMS DESIGN MAT KARO
 *
 * ============================================================================
 *  HTTP/3
 * ============================================================================
 *
 * HTTP/3 SAME HTTP MEANING rakhta hai
 * Lekin TCP ki jagah QUIC ke upar chalta hai
 *
 * QUIC UDP ke upar chalta hai aur yeh sab include karta hai:
 *   - TLS 1.3
 *   - Multiple streams
 *   - Loss recovery
 *   - Flow control
 *   - Congestion control
 *   - Connection migration
 *
 * CONNECTION MIGRATION ka matlab:
 *   - Ek connection kuch NETWORK CHANGES survive kar sakta hai
 *   - Jaise mobile client WiFi se cellular pe move kare
 *
 * ============================================================================
 *  HTTP/3 VERSION COMPARISON TABLE
 * ============================================================================
 *
 *   ┌──────────┬────────────────┬──────────────────┬──────────────────────────┬──────────────────────────┐
 *   │ Version  │ Transport      │ Wire Format      │ Main Benefit             │ Main Caveat              │
 *   ├──────────┼────────────────┼──────────────────┼──────────────────────────┼──────────────────────────┤
 *   │ HTTP/1.1 │ TCP            │ Textual messages │ Universal support,       │ Limited concurrency per  │
 *   │          │                │                  │ simple debugging         │ connection               │
 *   ├──────────┼────────────────┼──────────────────┼──────────────────────────┼──────────────────────────┤
 *   │ HTTP/2   │ TCP            │ Binary frames    │ Multiple streams and     │ Still affected by        │
 *   │          │                │                  │ header compression       │ TCP-level head-of-line   │
 *   │          │                │                  │                          │ blocking                 │
 *   ├──────────┼────────────────┼──────────────────┼──────────────────────────┼──────────────────────────┤
 *   │ HTTP/3   │ QUIC over UDP  │ Binary HTTP/3    │ Stream-level recovery    │ UDP reachability and     │
 *   │          │                │ frames over QUIC │ and connection migration │ operational complexity   │
 *   └──────────┴────────────────┴──────────────────┴──────────────────────────┴──────────────────────────┘
 *
 * ============================================================================
 *  HTTP/3 KYA HELP KARTA HAI
 * ============================================================================
 *
 *   1. Streams ke beech TCP-LEVEL HEAD-OF-LINE BLOCKING ko KAM karta hai
 *
 *   2. Kuch cases mein FASTER CONNECTION SETUP
 *
 *   3. Mobile clients NETWORKS CHANGE karein toh BETTER BEHAVIOUR
 *
 *   4. FASTER TRANSPORT CHANGES kyunki QUIC operating system TCP stack ke
 *      bahar evolve ho sakta hai
 *
 * ============================================================================
 *  HTTP/3 AUTOMATICALLY FASTER NAHI HAI
 * ============================================================================
 *
 * Har workload ke liye HTTP/3 automatically faster NAHI hai:
 *   - Kuch networks UDP ko BLOCK ya DEGRADE karte hain
 *
 * Production systems ko chahiye:
 *   - HTTP/2 ya HTTP/1.1 pe FALLBACK
 *   - Aur visibility into:
 *     * QUIC handshake failures
 *     * HTTP/3 adoption
 *     * Performance by client network
 *
 * ============================================================================
 *  10. HTTP IN DISTRIBUTED SYSTEMS
 * ============================================================================
 *
 * HTTP shuru karna EASY hai aur GALAT USE karna bhi EASY hai
 * - Production systems ko CLEAR RULES chahiye FAILURE BEHAVIOUR ke liye
 *
 * ============================================================================
 *  TIMEOUTS (Samay Seema)
 * ============================================================================
 *
 * Har HTTP client ko REQUEST ke HAR STAGE ke liye TIME-OUT SET karna chahiye:
 *   1. DNS resolution
 *   2. Connection establishment
 *   3. TCP setup
 *   4. TLS handshake
 *   5. Request writing
 *   6. Response headers ke liye waiting
 *   7. Response body ke liye waiting
 *
 * Aur yeh bhi SET karo:
 *   - Poore call ke liye ek OVERALL DEADLINE
 *   - Pool mein baithi unused connections ke liye ek IDLE TIMEOUT
 *
 * IMPORTANT WARNING:
 *   - Bahut saari libraries ke DEFAULTS production ke liye UNSAFE hain
 *   - Ek MISSING TIMEOUT ek dheemi dependency ko:
 *     * Exhausted threads mein badal sakta hai
 *     * Stuck connection pools mein badal sakta hai
 *     * Ek BADE outage mein badal sakta hai
 *
 * ============================================================================
 *  RETRIES (Phir Se Koshish)
 * ============================================================================
 *
 * Retries ko yeh RESPECT karna chahiye:
 *   - HTTP METHOD BEHAVIOUR
 *   - Application IDEMPOTENCY
 *
 * RULES:
 *   - GET ko RETRY karna usually SAFE hai
 *   - POST ko RETRY karna DUPLICATES create kar sakta hai
 *     - Jab tak API IDEMPOTENCY KEYS support na kare
 *
 * TIMEOUT KE BAAD RETRY AMBIGUOUS hai:
 *   - Kyunki server ne request PEHLE SE PROCESS kar di ho sakti hai
 *   - Bahut AGGRESSIVE retry ek outage ko WORSE bhi bana sakta hai
 *
 * BEST PRACTICE:
 *   - BOUNDED RETRIES use karo (limited)
 *   - BACKOFF (dheere dheere gap badhao)
 *   - JITTER (thodi randomness - taaki saare clients ek saath retry na karein)
 *   - DEADLINES
 *   - Clear RETRY BUDGETS
 *
 * ============================================================================
 *  STREAMING (Data Dheere Dheere Aana)
 * ============================================================================
 *
 * HTTP sirf chhote JSON responses ke liye NAHI hai
 * Streaming modern systems mein COMMON hai:
 *
 *   - SERVER-SENT EVENTS: Token streaming ke liye
 *   - CHUNKED HTTP RESPONSES: Incremental output ke liye
 *   - WEBSOCKETS: Two-way communication ke liye
 *   - gRPC STREAMING over HTTP/2
 *   - HTTP/3 STREAMS over QUIC
 *
 * AI PRODUCTS ke liye STREAMING kya badalta hai:
 *   - User experience badalta hai
 *   - Infrastructure behaviour badalta hai
 *   - Aapko chahiye:
 *     * Cancellation (roknay ka tareeka)
 *     * Idle timeouts
 *     * Backpressure
 *     * Partial failure handling
 *     * Metrics for TIME TO FIRST TOKEN aur STREAM COMPLETION
 *
 * BACKPRESSURE ka matlab:
 *   - Sender ko SLOW karna jab receiver ya network RAKH NAHI pa raha ho
 *
 * ============================================================================
 *  PROXIES AND HEADERS (Proxy Aur Headers)
 * ============================================================================
 *
 * Zyada tar production HTTP requests PROXIES ya LOAD BALANCERS se guzarte hain
 * - Applications ko FORWARDED REQUEST INFORMATION carefully handle karni
 *   chahiye:
 *
 *   1. Host
 *   2. X-Forwarded-For
 *   3. X-Forwarded-Proto
 *   4. Forwarded
 *   5. X-Request-ID ya traceparent
 *   6. Authorization
 *
 * SECURITY RULE:
 *   - SIRF apne CONTROLLED infrastructure se forwarding headers TRUST karo
 *   - PUBLIC clients yeh headers FAKE kar sakte hain
 *   - Jab tak edge proxy unhe REMOVE ya REWRITE na kare
 *
 * ============================================================================
 *  OBSERVABILITY (Nazar Rakhna)
 * ============================================================================
 *
 * HTTP EXCELLENT OPERATIONAL SIGNALS deta hai:
 *
 *   1. Request rate
 *   2. Latency percentiles
 *   3. Status code distribution
 *   4. Retry rate
 *   5. Payload size
 *   6. TLS handshake failures
 *   7. Internal service error rate
 *   8. Cache hit ratio
 *   9. Time to first byte
 *   10. Stream duration
 *
 * IN SIGNALS KO INKE HISAB SE BREAK KARO:
 *   - Route
 *   - Method
 *   - Status class
 *   - Client
 *   - Region
 *   - Internal service
 *
 * IMPORTANT:
 *   - AVERAGES un failures ko HIDE kar dete hain jo users ACTUALLY MEHFOOS
 *     karte hain
 *
 * ============================================================================
 *  SUMMARY (Saransh)
 * ============================================================================
 *
 * 1. HTTP WEB KA REQUEST-RESPONSE LANGUAGE hai:
 *    - Methods, resources, headers, status codes, caching, aur bodies
 *    - HTTPS = HTTP jo TLS se PROTECTED hai
 *    - TLS add karta hai: ENCRYPTION, TAMPER DETECTION, aur SERVER IDENTITY CHECKS
 *    - HTTPS APPLICATION SECURITY ko REPLACE NAHI karta
 *
 * 2. PRACTICAL BASELINE:
 *    - Sare production traffic ke liye HTTPS use karo
 *    - METHOD BEHAVIOUR ko API CONTRACT ka hissa samjho
 *    - Precise status codes aur cache headers use karo
 *
 * 3. MAJOR HTTP VERSIONS SAME CORE IDEAS rakhte hain
 *    - Lekin NEECHE bahut alag hote hain:
 *      * HTTP/2 REDUCES karta hai lekin HEAD-OF-LINE BLOCKING ko ELIMINATE
 *        NAHI karta
 *      * Kyunki TCP abhi bhi connection pe HAR STREAM ko STALL kar sakta hai
 *    - HTTP/2 SERVER PUSH design around karne layak NAHI hai
 *
 * 4. HTTP/3 over QUIC HELP karta hai jahan iski properties FIT karti hain
 *    - Jab tak FALLBACK aur MONITORING jagah pe rahe
 *
 * 5. HTTP SIMPLE lagta hai kyunki tooling FAMILIAR hai
 *    - Scale pe, hard parts KISI BHI DISTRIBUTED SYSTEM jaise hi hain:
 *      * Latency
 *      * Retries
 *      * Overload
 *      * State
 *      * Security
 *      * Compatibility
 *      * Failure visibility
 *    - Timeouts, retries, deadlines, aur idempotency rules DELIBERATELY SET karo
 *
 * ============================================================================
 **/
```
