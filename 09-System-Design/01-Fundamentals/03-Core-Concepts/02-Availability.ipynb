```python
/**
 * ============================================================
 * AVAILABILITY — SYSTEM DESIGN
 * ============================================================
 *
 * Core idea:
 *
 *     Component Fails
 *          ↓
 *     Detect Failure
 *          ↓
 *     Failover / Remove Failed Component
 *          ↓
 *     Keep Serving Users
 *
 *
 * Availability is about:
 *
 *     "Can users reach and use the system when they need it?"
 *
 *
 * Scalability:
 *
 *     Can the system handle more load?
 *
 * Availability:
 *
 *     Can users still use the system when something fails?
 *
 * Reliability:
 *
 *     Does the system continue behaving correctly over time?
 *
 *
 * Example:
 *
 *     Store can process 10,000 orders/minute
 *     -> Scalability
 *
 *     One server crashes but customers can still checkout
 *     -> Availability
 *
 *     Order total is always calculated correctly
 *     -> Reliability
 *
 *
 * ============================================================
 * 1. THE AVAILABILITY PROBLEM
 * ============================================================
 *
 * Initial architecture:
 *
 *     Customers
 *         |
 *         v
 *     App Server
 *         |
 *         v
 *     Database
 *
 *
 * Problem:
 *
 *     App Server crashes
 *
 *     Database is still healthy
 *
 *     BUT
 *
 *     Customers cannot use the application.
 *
 *
 * Why?
 *
 *     The application server was a
 *     Single Point of Failure (SPOF).
 *
 *
 * Availability requires:
 *
 *     Another way to serve requests
 *     when one component fails.
 *
 *
 * ============================================================
 * 2. AVAILABILITY vs SCALABILITY vs RELIABILITY
 * ============================================================
 *
 *
 * SCALABILITY
 *
 * Question:
 *
 *     Can the system handle more load?
 *
 * Example:
 *
 *     Handle 10x more orders during a sale.
 *
 *
 * AVAILABILITY
 *
 * Question:
 *
 *     Can users reach and use the system?
 *
 * Example:
 *
 *     Customers can checkout when one server crashes.
 *
 *
 * RELIABILITY
 *
 * Question:
 *
 *     Does the system behave correctly over time?
 *
 * Example:
 *
 *     Every order total is calculated correctly.
 *
 *
 * IMPORTANT:
 *
 *     A system can be:
 *
 *     Available but not reliable.
 *
 * Example:
 *
 *     Store accepts orders
 *     BUT
 *     Calculates the wrong order total.
 *
 *
 * ============================================================
 * 3. MEASURING AVAILABILITY
 * ============================================================
 *
 * Basic formula:
 *
 *
 *     Availability =
 *
 *         Uptime
 *         -------------------------
 *         Uptime + Downtime
 *
 *
 * Example:
 *
 *     Uptime   = 364 days
 *     Downtime = 1 day
 *
 *
 *     Availability =
 *
 *         364 / 365
 *
 *         = 99.73%
 *
 *
 * IMPORTANT:
 *
 * Do not look only at one availability number
 * for the entire application.
 *
 * Measure important operations separately.
 *
 *
 * Example:
 *
 *     Browse Products -> 99.99%
 *     Add to Cart     -> 99.95%
 *     Checkout        -> 99.5%
 *
 *
 * A whole-store average may hide a serious
 * checkout outage.
 *
 *
 * Therefore:
 *
 *     Measure availability
 *     per important operation
 *     over a defined time window.
 *
 *
 * ============================================================
 * 4. THE NINES OF AVAILABILITY
 * ============================================================
 *
 * Availability is commonly described using "nines".
 *
 *
 * 99%:
 *
 *     ~3.65 days downtime/year
 *
 *
 * 99.9%:
 *
 *     ~8.76 hours/year
 *
 *
 * 99.99%:
 *
 *     ~52.6 minutes/year
 *
 *
 * 99.999%:
 *
 *     ~5.26 minutes/year
 *
 *
 * 99.9999%:
 *
 *     ~31.5 seconds/year
 *
 *
 * IMPORTANT:
 *
 * Every additional nine reduces allowed downtime
 * by roughly 10x.
 *
 *
 * Example:
 *
 *     99.9%
 *        ↓
 *     99.99%
 *
 *     Allowed downtime becomes roughly 1/10.
 *
 *
 * ============================================================
 * 5. HIGHER AVAILABILITY = HIGHER COST
 * ============================================================
 *
 * Higher availability usually requires:
 *
 *     Redundancy
 *     Spare capacity
 *     Automation
 *     Monitoring
 *     Health checks
 *     Tested recovery
 *     Faster failover
 *     Multiple locations
 *
 *
 * Example:
 *
 *     99.9%
 *
 *     -> Second server
 *     -> Basic monitoring
 *
 *
 *     99.99%
 *
 *     -> Automatic failover
 *     -> Health checks
 *     -> On-call rotation
 *     -> Tested recovery
 *
 *
 *     99.999%
 *
 *     -> Redundancy at every layer
 *     -> Multiple locations
 *     -> Automated recovery
 *     -> Minimal/no human intervention
 *
 *
 * IMPORTANT:
 *
 * Availability is a business decision as well as
 * a technical decision.
 *
 * More redundancy costs more money and adds
 * operational complexity.
 *
 *
 * ============================================================
 * 6. COMPONENTS IN SERIES
 * ============================================================
 *
 * Series dependency:
 *
 * Every component must be available for the request
 * to succeed.
 *
 *
 *     Customer
 *         |
 *         v
 *     Web Server
 *       99.9%
 *         |
 *         v
 *     App Server
 *       99.9%
 *         |
 *         v
 *     Database
 *       99.9%
 *
 *
 * If ANY required component fails:
 *
 *     Request fails.
 *
 *
 * Formula:
 *
 *
 *     Overall Availability
 *
 *       = A1 × A2 × A3
 *
 *
 * Example:
 *
 *     99.9% × 99.9% × 99.9%
 *
 *     = 99.7% approximately
 *
 *
 * With more required components:
 *
 *     Overall availability decreases.
 *
 *
 * IMPORTANT:
 *
 * Every component added to the request path
 * introduces another possible failure point.
 *
 *
 * ============================================================
 * 7. COMPONENTS IN PARALLEL
 * ============================================================
 *
 * Parallel components provide redundancy.
 *
 *
 * Example:
 *
 *              Load Balancer
 *                /       \
 *               v         v
 *           App Server  App Server
 *              99.9%       99.9%
 *
 *
 * Either server can handle the request.
 *
 * The system fails only when:
 *
 *     BOTH servers fail.
 *
 *
 * If each server has:
 *
 *     Availability = 99.9%
 *
 * Then:
 *
 *     Failure probability = 0.1%
 *
 *
 * Both fail:
 *
 *     0.1% × 0.1%
 *
 *     = 0.0001%
 *
 *
 * Therefore:
 *
 *     Availability =
 *
 *     100% - 0.0001%
 *
 *     = 99.9999%
 *
 *
 * IMPORTANT:
 *
 * This is a theoretical calculation based on
 * independent failures.
 *
 *
 * ============================================================
 * 8. SERIES vs PARALLEL
 * ============================================================
 *
 *
 * SERIES:
 *
 *     Request
 *        |
 *        v
 *      A
 *        |
 *        v
 *      B
 *        |
 *        v
 *      C
 *
 *
 * Any failure:
 *
 *     A fails -> System fails
 *     B fails -> System fails
 *     C fails -> System fails
 *
 *
 * Combined availability:
 *
 *     LOWER
 *
 *
 * ------------------------------------------------------------
 *
 * PARALLEL:
 *
 *            A
 *           / \
 * Request  /   \
 *         /     \
 *        B       C
 *
 *
 * System can continue if:
 *
 *     B works
 *     OR
 *     C works
 *
 *
 * Combined availability:
 *
 *     HIGHER
 *
 *
 * CORE IDEA:
 *
 *     Series:
 *     Any failure can break the path.
 *
 *     Parallel:
 *     Redundant components can survive
 *     individual failures.
 *
 *
 * ============================================================
 * 9. INDEPENDENCE MATTERS
 * ============================================================
 *
 * Parallel servers only improve availability if
 * their failures are reasonably independent.
 *
 *
 * Two servers can still fail together if they share:
 *
 *     Same power source
 *     Same rack
 *     Same network
 *     Same switch
 *     Same DNS
 *     Same faulty deployment
 *     Same software bug
 *
 *
 * Example:
 *
 *     Bad Deployment
 *          |
 *      +---+---+
 *      |       |
 *      v       v
 *     App1    App2
 *     DOWN    DOWN
 *
 *
 * Both servers are redundant,
 * but the shared failure removed both.
 *
 *
 * Common shared failure categories:
 *
 *
 * HARDWARE
 *
 *     Drive
 *     Server
 *     Switch
 *     Power supply
 *
 *
 * SOFTWARE
 *
 *     Bug
 *     Memory leak
 *     Deadlock
 *     Cascading failure
 *
 *
 * NETWORK
 *
 *     Packet loss
 *     Latency spike
 *     Network partition
 *     DNS failure
 *
 *
 * HUMAN ERROR
 *
 *     Bad configuration
 *     Bad deployment
 *     Accidental deletion
 *
 *
 * ============================================================
 * 10. REDUNDANCY
 * ============================================================
 *
 * Redundancy means:
 *
 *     Provide another way to handle requests
 *     when one component fails.
 *
 *
 * Example:
 *
 *     App Server 1
 *     App Server 2
 *
 *
 * If App Server 1 disappears:
 *
 *     App Server 2
 *     continues serving traffic.
 *
 *
 * At large scale:
 *
 *     Hardware failures are normal events.
 *
 * Therefore:
 *
 *     Design assuming individual machines
 *     can fail at any time.
 *
 *
 * ============================================================
 * 11. ACTIVE-PASSIVE REDUNDANCY
 * ============================================================
 *
 * One component is active.
 *
 * Another component waits as standby.
 *
 *
 * Example:
 *
 *     App Servers
 *          |
 *          v
 *     Primary Database
 *          |
 *      Replication
 *          |
 *          v
 *     Standby Database
 *
 *
 * Primary:
 *
 *     Active
 *     Handles reads/writes
 *
 *
 * Standby:
 *
 *     Passive
 *     Waits for failure
 *
 *
 * If primary fails:
 *
 *     1. Monitor detects failure
 *
 *     2. Standby is promoted
 *
 *     3. Application connections are redirected
 *
 *     4. Writes resume
 *
 *
 * This process is:
 *
 *     FAILOVER
 *
 *
 * ============================================================
 * 12. ACTIVE-PASSIVE — PROS & CONS
 * ============================================================
 *
 * PROS:
 *
 *     + One clear writer
 *     + Easier data coordination
 *     + Simple architecture
 *     + Standby does not need production traffic
 *
 *
 * CONS:
 *
 *     - Failover takes time
 *     - Standby may be untested
 *     - Split-brain risk
 *
 *
 * Split-brain:
 *
 *     Monitor incorrectly promotes standby
 *     while old primary is still alive.
 *
 *
 * Result:
 *
 *     Primary  -> accepts writes
 *     Standby  -> accepts writes
 *
 *     Two writers may create conflicting data.
 *
 *
 * ============================================================
 * 13. ACTIVE-PASSIVE FAILOVER FLOW
 * ============================================================
 *
 *     Primary Database
 *            |
 *         FAILURE
 *            ↓
 *     Monitor detects failure
 *            ↓
 *     Promote Standby
 *            ↓
 *     Redirect App Connections
 *            ↓
 *     Writes Resume
 *
 *
 * Failover is NOT instant.
 *
 * Time includes:
 *
 *     Detection
 *     Promotion
 *     Reconnection
 *
 *
 * Therefore:
 *
 *     Active-passive may cause
 *     a short interruption.
 *
 *
 * ============================================================
 * 14. COLD / WARM / HOT STANDBY
 * ============================================================
 *
 *
 * COLD STANDBY
 *
 *     Not running
 *     May require data restoration
 *
 * Failover:
 *
 *     Minutes
 *
 * Cost:
 *
 *     Lowest
 *
 *
 * WARM STANDBY
 *
 *     Running
 *     Configured
 *     Not serving production traffic
 *
 * Failover:
 *
 *     Seconds to minutes
 *
 * Cost:
 *
 *     Medium
 *
 *
 * HOT STANDBY
 *
 *     Running
 *     Data synchronized
 *     Ready to serve
 *
 * Failover:
 *
 *     Seconds
 *
 * Cost:
 *
 *     Highest
 *
 *
 * Trade-off:
 *
 *     Faster recovery
 *          =
 *     More infrastructure + higher cost
 *
 *
 * IMPORTANT:
 *
 * A standby should be tested under realistic
 * production load.
 *
 * A backup that exists but cannot handle production
 * traffic is not useful during a real outage.
 *
 *
 * ============================================================
 * 15. ACTIVE-ACTIVE REDUNDANCY
 * ============================================================
 *
 * Multiple servers actively serve traffic.
 *
 *
 *     Customers
 *         |
 *         v
 *     Load Balancer
 *        /     \
 *       v       v
 *    App 1    App 2
 *    Active   Active
 *
 *
 * Both servers serve real traffic.
 *
 *
 * If App 1 fails:
 *
 *     Load Balancer
 *          |
 *          v
 *        App 2
 *
 *
 * No traditional standby promotion is required.
 *
 *
 * ============================================================
 * 16. ACTIVE-ACTIVE — REQUIREMENTS
 * ============================================================
 *
 * Surviving servers need spare capacity.
 *
 *
 * Example:
 *
 *     App 1 -> 50% load
 *     App 2 -> 50% load
 *
 *
 * App 1 fails:
 *
 *     App 2 -> 100% load
 *
 *
 * If App 2 was already near its limit:
 *
 *     App 2 may also fail.
 *
 *
 * Therefore:
 *
 *     Keep enough spare capacity
 *     to absorb the failed server's traffic.
 *
 *
 * ============================================================
 * 17. STATELESS SERVERS + ACTIVE-ACTIVE
 * ============================================================
 *
 * Active-active works well with stateless application servers.
 *
 *
 * Bad design:
 *
 *     App 1
 *       |
 *     Local Session
 *
 *     App 2
 *       |
 *     Different Local Session
 *
 *
 * Any request can go to either server,
 * so state should not be trapped inside one server.
 *
 *
 * Better:
 *
 *     App 1 ----\
 *                \
 *                 Shared Store
 *                /
 *     App 2 ----/
 *
 *
 * Shared state:
 *
 *     Sessions
 *     Carts
 *
 *
 * Benefits:
 *
 *     Any healthy server can handle
 *     the next request.
 *
 *
 * ============================================================
 * 18. ACTIVE-ACTIVE — PROS & CONS
 * ============================================================
 *
 * PROS:
 *
 *     + No failover delay
 *     + Every node is tested under real load
 *     + Better hardware utilization
 *
 *
 * CONS:
 *
 *     - Requires spare capacity
 *     - Requires shared state
 *     - Harder for databases
 *
 *
 * Why databases are harder:
 *
 *     Multiple active writers
 *     must coordinate changes.
 *
 *
 * ============================================================
 * 19. HEALTH CHECKS
 * ============================================================
 *
 * Load balancer uses health checks to determine
 * whether a server should receive traffic.
 *
 *
 * Example:
 *
 *     Customer
 *         |
 *         v
 *     Load Balancer
 *        /       \
 *       v         v
 *    App 1      App 2
 *   Healthy    Failed
 *
 *
 * Load balancer:
 *
 *     Sends traffic -> App 1
 *     Stops traffic -> App 2
 *
 *
 * Health checks usually run periodically.
 *
 * Therefore:
 *
 *     Failure detection is not instant.
 *
 *
 * Some requests may fail:
 *
 *     Before detection
 *     During detection
 *     While requests are already in progress
 *
 *
 * ============================================================
 * 20. GOOD HEALTH CHECK
 * ============================================================
 *
 * A good health check should verify that the server
 * can actually handle requests.
 *
 *
 * Not enough:
 *
 *     "Is the process running?"
 *
 *
 * Better:
 *
 *     "Can this server actually serve traffic?"
 *
 *
 * Example:
 *
 *     Process -> Running
 *     Database connection -> Working
 *     Critical dependency -> Working
 *     Request path -> Healthy
 *
 *
 * ============================================================
 * 21. LOAD BALANCER IS ALSO A FAILURE POINT
 * ============================================================
 *
 * Architecture:
 *
 *     Customers
 *         |
 *         v
 *     Load Balancer
 *       /      \
 *      v        v
 *    App 1    App 2
 *
 *
 * Problem:
 *
 * If the Load Balancer fails:
 *
 *     App 1 -> Healthy
 *     App 2 -> Healthy
 *
 * BUT
 *
 *     Customers cannot reach them.
 *
 *
 * Therefore:
 *
 *     Load Balancer itself needs redundancy.
 *
 *
 * Managed cloud load balancers generally provide
 * redundant infrastructure.
 *
 *
 * Self-managed approach:
 *
 *     Load Balancer 1 -> Active
 *     Load Balancer 2 -> Standby
 *
 *     Both share a virtual IP.
 *
 *
 * If active LB fails:
 *
 *     Standby takes over.
 *
 *
 * ============================================================
 * 22. SPREADING ACROSS LOCATIONS
 * ============================================================
 *
 * Having multiple servers is not enough if
 * all servers are in the same facility.
 *
 *
 * Shared location risks:
 *
 *     Power outage
 *     Network outage
 *     Facility failure
 *
 *
 * Solution:
 *
 *     Spread infrastructure across locations.
 *
 *
 * Levels:
 *
 *     Availability Zones
 *          ↓
 *     Regions
 *          ↓
 *     Multiple Cloud Providers
 *
 *
 * ============================================================
 * 23. AVAILABILITY ZONES
 * ============================================================
 *
 * Availability Zones:
 *
 *     Separate data centers
 *     within the same cloud region.
 *
 *
 * Protect against:
 *
 *     Single data-center failure
 *     Power failure
 *     Network failure
 *
 *
 * Typical latency:
 *
 *     Low
 *     Usually a few milliseconds
 *
 *
 * Example:
 *
 *     Region
 *       |
 *       +---- AZ 1
 *       |
 *       +---- AZ 2
 *       |
 *       +---- AZ 3
 *
 *
 * Application can continue if one AZ fails,
 * assuming capacity exists in the remaining AZs.
 *
 *
 * ============================================================
 * 24. MULTI-REGION
 * ============================================================
 *
 * Regions are geographically separated.
 *
 *
 * Example:
 *
 *     US East
 *         |
 *         | Replication
 *         |
 *     Europe
 *
 *
 * Protects against:
 *
 *     Regional outage
 *     Regional disaster
 *
 *
 * Trade-off:
 *
 *     Higher network latency
 *
 *     Data replication becomes more complicated.
 *
 *
 * ============================================================
 * 25. MULTIPLE CLOUD PROVIDERS
 * ============================================================
 *
 * Example:
 *
 *     AWS
 *       +
 *     Google Cloud
 *
 *
 * Protects against:
 *
 *     Provider-wide outage
 *
 *
 * But introduces:
 *
 *     More operational complexity
 *     Different tooling
 *     Different networking
 *     Different failure behavior
 *
 *
 * IMPORTANT:
 *
 * More redundancy is not automatically better.
 *
 * Choose redundancy based on:
 *
 *     Failure scenarios
 *     Cost
 *     Complexity
 *     Business requirements
 *
 *
 * ============================================================
 * 26. AVAILABILITY LEVELS
 * ============================================================
 *
 *
 * AVAILABILITY ZONE:
 *
 *     Same region
 *     Lower latency
 *     Protects against one data-center failure
 *
 *
 * REGION:
 *
 *     Different geographic area
 *     Higher latency
 *     Protects against regional outage
 *
 *
 * MULTIPLE CLOUD PROVIDERS:
 *
 *     Different infrastructure providers
 *     Protects against provider-wide outage
 *
 *
 * Trade-off:
 *
 *     More separation
 *          =
 *     More isolation
 *
 *     BUT
 *
 *     More separation
 *          =
 *     More latency + complexity
 *
 *
 * ============================================================
 * 27. SYNCHRONOUS vs ASYNCHRONOUS REPLICATION
 * ============================================================
 *
 *
 * SYNCHRONOUS:
 *
 *     Write
 *       |
 *       +------> Primary
 *       |
 *       +------> Remote Replica
 *                     |
 *                  Confirm
 *                     |
 *     Write succeeds
 *
 *
 * Write succeeds only after remote replica confirms.
 *
 *
 * Benefit:
 *
 *     No data loss on failover
 *     according to the source's model.
 *
 *
 * Cost:
 *
 *     Every write waits for the remote
 *     network round trip.
 *
 *
 * ------------------------------------------------------------
 *
 * ASYNCHRONOUS:
 *
 *     Write
 *       |
 *       v
 *     Primary
 *       |
 *     Confirm immediately
 *       |
 *       v
 *     Remote Replica
 *
 *
 * Remote replica catches up later.
 *
 *
 * Benefit:
 *
 *     No additional remote round-trip latency
 *     for confirming the write.
 *
 *
 * Risk:
 *
 *     Recent writes may not yet exist
 *     on the remote replica.
 *
 *
 * Therefore:
 *
 *     Regional failure may cause loss of
 *     the latest few seconds/minutes of changes,
 *     depending on replication lag.
 *
 *
 * ============================================================
 * 28. REDUNDANCY IS NOT FREE
 * ============================================================
 *
 * Every additional:
 *
 *     Server
 *     Standby
 *     Replica
 *     Availability Zone
 *     Region
 *     Cloud Provider
 *
 * adds:
 *
 *     Cost
 *     Operational complexity
 *     Monitoring requirements
 *     Failure scenarios
 *
 *
 * Question to ask:
 *
 *     "Which failures do we actually need to survive?"
 *
 *
 * Then choose the appropriate redundancy level.
 *
 *
 * ============================================================
 * 29. QUEUES FOR OVERLOAD
 * ============================================================
 *
 * Failures can happen because of overload,
 * not just broken hardware.
 *
 *
 * Example:
 *
 * Big sale:
 *
 *     Order traffic increases
 *          ↓
 *     Confirmation emails increase
 *          ↓
 *     Email provider becomes slow
 *          ↓
 *     Checkout waits for email
 *          ↓
 *     Checkout becomes slow
 *
 *
 * Better architecture:
 *
 *     Customer
 *        |
 *        v
 *     App Server
 *        |
 *        +------> Database
 *        |
 *        +------> Durable Queue
 *                         |
 *                    +----+----+
 *                    |         |
 *                    v         v
 *                 Worker 1  Worker 2
 *                    |
 *                    v
 *              Email Provider
 *
 *
 * Request flow:
 *
 *     1. Save order
 *     2. Enqueue email job
 *     3. Confirm order
 *     4. Worker sends email later
 *
 *
 * User does NOT wait for email delivery.
 *
 *
 * ============================================================
 * 30. DURABLE QUEUE
 * ============================================================
 *
 * Queue stores the background job until a worker
 * confirms that it is completed.
 *
 *
 * If Worker 1 crashes:
 *
 *     Job is not necessarily lost.
 *
 * Another worker can retry it.
 *
 *
 * Flow:
 *
 *     Queue
 *       |
 *       v
 *     Worker 1
 *       |
 *     CRASH
 *
 *       ↓
 *
 *     Worker 2
 *       |
 *       v
 *     Process Job
 *
 *
 * This makes background processing more resilient.
 *
 *
 * ============================================================
 * 31. CASCADING FAILURE
 * ============================================================
 *
 * Cascading failure:
 *
 * One slow/failing dependency causes problems
 * to spread through the system.
 *
 *
 * Example:
 *
 *     Recommendation Service
 *             |
 *          becomes slow
 *             |
 *             v
 *       App requests wait
 *             |
 *             v
 *       Threads exhausted
 *             |
 *        +----+----+
 *        |         |
 *        v         v
 *    Product     Checkout
 *      Slow        Slow
 *
 *
 * The recommendation service may not be part of
 * checkout's actual business requirement,
 * but its slowdown can still affect checkout.
 *
 *
 * ============================================================
 * 32. TIMEOUTS
 * ============================================================
 *
 * Timeout:
 *
 * Stop waiting for a dependency after a fixed
 * amount of time.
 *
 *
 * Example:
 *
 *     App
 *      |
 *      | Request recommendation
 *      |
 *      v
 *     Recommendation Service
 *
 *
 * Timeout:
 *
 *     200 ms
 *
 *
 * If response takes longer:
 *
 *     Stop waiting.
 *
 *
 * Benefit:
 *
 *     Prevent requests from waiting indefinitely.
 *
 *
 * ============================================================
 * 33. CIRCUIT BREAKER
 * ============================================================
 *
 * Circuit breaker prevents repeated calls to a
 * dependency that is failing.
 *
 *
 * Three states:
 *
 *     CLOSED
 *     OPEN
 *     HALF-OPEN
 *
 *
 * ------------------------------------------------------------
 * CLOSED
 * ------------------------------------------------------------
 *
 * Normal operation.
 *
 * Requests are allowed through.
 *
 *
 *     App
 *       |
 *       v
 *     Dependency
 *
 *
 * If failures exceed threshold:
 *
 *     CLOSED
 *         ↓
 *       OPEN
 *
 *
 * ------------------------------------------------------------
 * OPEN
 * ------------------------------------------------------------
 *
 * Dependency is considered unhealthy.
 *
 * Calls are rejected immediately.
 *
 * No waiting for the slow/failing dependency.
 *
 *
 * After cooldown:
 *
 *     OPEN
 *       ↓
 *     HALF-OPEN
 *
 *
 * ------------------------------------------------------------
 * HALF-OPEN
 * ------------------------------------------------------------
 *
 * Send a small number of test requests.
 *
 *
 * If test succeeds:
 *
 *     HALF-OPEN
 *         ↓
 *       CLOSED
 *
 *
 * If test fails:
 *
 *     HALF-OPEN
 *         ↓
 *       OPEN
 *
 *
 * ============================================================
 * 34. CIRCUIT BREAKER FLOW
 * ============================================================
 *
 *
 *             CLOSED
 *                |
 *       failures exceed threshold
 *                |
 *                v
 *              OPEN
 *                |
 *          cooldown expires
 *                |
 *                v
 *           HALF-OPEN
 *             /     \
 *            /       \
 *       success      failure
 *          |            |
 *          v            v
 *       CLOSED        OPEN
 *
 *
 * Main purpose:
 *
 *     Contain dependency failure.
 *
 *
 * ============================================================
 * 35. FALLBACK
 * ============================================================
 *
 * Fallback:
 *
 * Return a reduced but still usable response
 * when an optional dependency is unavailable.
 *
 *
 * Example:
 *
 *     Product Page
 *          |
 *          +----> Recommendation Service
 *
 *
 * Recommendation service fails.
 *
 *
 * Without fallback:
 *
 *     Product page fails.
 *
 *
 * With fallback:
 *
 *     Product page
 *     WITHOUT recommendations
 *
 *
 * Customer can still:
 *
 *     Browse product
 *     Place order
 *
 *
 * IMPORTANT:
 *
 * Optional features should not take down
 * critical functionality.
 *
 *
 * ============================================================
 * 36. TIMEOUT + CIRCUIT BREAKER + FALLBACK
 * ============================================================
 *
 * Example:
 *
 *     App Server
 *          |
 *     Timeout
 *          |
 *     Circuit Breaker
 *          |
 *          v
 *     Recommendation Service
 *
 *
 * If service is healthy:
 *
 *     Request succeeds.
 *
 *
 * If service becomes slow:
 *
 *     Timeout prevents indefinite waiting.
 *
 *
 * If failures continue:
 *
 *     Circuit opens.
 *
 *
 *     Request
 *        |
 *        v
 *     Circuit Breaker
 *        |
 *      OPEN
 *        |
 *        v
 *     Fallback
 *        |
 *        v
 *     Product Page
 *     without recommendations
 *
 *
 * Result:
 *
 *     Recommendation failure
 *          DOES NOT
 *     take down the product page.
 *
 *
 * ============================================================
 * 37. DESIGN FOR FAILURE
 * ============================================================
 *
 * The main availability mindset:
 *
 *
 *     "Assume components will fail."
 *
 *
 * Then ask:
 *
 *     What happens if:
 *
 *     App server fails?
 *     Database fails?
 *     Load balancer fails?
 *     Availability Zone fails?
 *     Region fails?
 *     Dependency becomes slow?
 *     Worker crashes?
 *     Network fails?
 *     Bad deployment happens?
 *
 *
 * For each failure:
 *
 *     Detect
 *       ↓
 *     Recover / Redirect
 *       ↓
 *     Continue serving users
 *
 *
 * ============================================================
 * 38. AVAILABILITY DESIGN CHECKLIST
 * ============================================================
 *
 *
 * APPLICATION SERVERS:
 *
 *     -> Multiple instances
 *     -> Load balancer
 *     -> Stateless architecture
 *     -> Health checks
 *     -> Spare capacity
 *
 *
 * DATABASE:
 *
 *     -> Primary + standby
 *     -> Replication
 *     -> Failover
 *     -> Test standby under realistic load
 *
 *
 * LOAD BALANCER:
 *
 *     -> Health checks
 *     -> Remove unhealthy nodes
 *     -> Redundant load balancer
 *
 *
 * STATE:
 *
 *     -> Shared session/cart store
 *     -> Do not depend on one app server's local state
 *
 *
 * LOCATIONS:
 *
 *     -> Multiple availability zones
 *     -> Multi-region when required
 *
 *
 * BACKGROUND WORK:
 *
 *     -> Durable queue
 *     -> Multiple workers
 *     -> Retry failed jobs
 *
 *
 * DEPENDENCIES:
 *
 *     -> Timeouts
 *     -> Circuit breakers
 *     -> Fallbacks
 *
 *
 * MONITORING:
 *
 *     -> Measure availability
 *     -> Measure per important operation
 *     -> Monitor failures
 *     -> Monitor recovery
 *
 *
 * ============================================================
 * 39. FAILURE SCENARIO CHEAT SHEET
 * ============================================================
 *
 *
 * APP SERVER FAILS
 *
 *     -> Load balancer removes it
 *     -> Another active server handles requests
 *
 *
 * DATABASE PRIMARY FAILS
 *
 *     -> Monitor detects failure
 *     -> Promote standby
 *     -> Redirect connections
 *
 *
 * LOAD BALANCER FAILS
 *
 *     -> Use redundant load balancers
 *
 *
 * ONE DATA CENTER FAILS
 *
 *     -> Spread across availability zones
 *
 *
 * REGION FAILS
 *
 *     -> Multi-region deployment
 *
 *
 * BACKGROUND WORK SPIKES
 *
 *     -> Durable queue
 *     -> Scale workers
 *
 *
 * WORKER FAILS
 *
 *     -> Queue retains job
 *     -> Another worker retries
 *
 *
 * DEPENDENCY BECOMES SLOW
 *
 *     -> Timeout
 *
 *
 * DEPENDENCY KEEPS FAILING
 *
 *     -> Circuit breaker
 *
 *
 * OPTIONAL DEPENDENCY FAILS
 *
 *     -> Fallback
 *
 *
 * ============================================================
 * 40. FINAL AVAILABILITY ARCHITECTURE
 * ============================================================
 *
 *
 *                         CUSTOMERS
 *                             |
 *                             v
 *                        GLOBAL DNS
 *                         /       \
 *                        v         v
 *                  LOAD BALANCER  LOAD BALANCER
 *                    REGION 1      REGION 2
 *                       |
 *                +------+------+
 *                |             |
 *                v             v
 *              APP 1         APP 2
 *             Active        Active
 *                |             |
 *                +------+------+
 *                       |
 *                 SHARED STORE
 *                Sessions/Carts
 *
 *
 * Database:
 *
 *             PRIMARY DATABASE
 *                    |
 *                Replication
 *                    |
 *                    v
 *             STANDBY DATABASE
 *
 *
 * Background:
 *
 *             APP
 *              |
 *              v
 *        DURABLE QUEUE
 *          /       \
 *         v         v
 *      WORKER 1  WORKER 2
 *
 *
 * Optional Dependency:
 *
 *          APP
 *           |
 *       Timeout
 *           |
 *    Circuit Breaker
 *           |
 *           v
 *    Recommendation
 *       Service
 *           |
 *        Failure
 *           |
 *           v
 *       Fallback
 *           |
 *           v
 *    Product Page
 *    Without Recommendations
 *
 *
 * ============================================================
 * 41. QUICK REVISION
 * ============================================================
 *
 * Availability
 *     -> Ability of users to access and use the system.
 *
 * Availability Formula
 *     -> Uptime / (Uptime + Downtime)
 *
 * Nines
 *     -> Higher percentage = less allowed downtime.
 *
 * Series
 *     -> Every component must work.
 *
 * Parallel
 *     -> Any redundant component can work.
 *
 * Redundancy
 *     -> Provide another way to serve requests.
 *
 * Active-Passive
 *     -> One active + one standby.
 *
 * Failover
 *     -> Promote standby and redirect traffic.
 *
 * Cold Standby
 *     -> Not running; cheapest; slowest recovery.
 *
 * Warm Standby
 *     -> Running but not serving production traffic.
 *
 * Hot Standby
 *     -> Running, synchronized, ready to serve.
 *
 * Active-Active
 *     -> Multiple nodes serve traffic simultaneously.
 *
 * Health Check
 *     -> Determines whether a server can receive traffic.
 *
 * Availability Zone
 *     -> Protects against single data-center failures.
 *
 * Multi-Region
 *     -> Protects against regional failures.
 *
 * Synchronous Replication
 *     -> Wait for remote replica before confirming write.
 *
 * Asynchronous Replication
 *     -> Confirm write immediately; replica catches up later.
 *
 * Queue
 *     -> Separates background work from user requests.
 *
 * Cascading Failure
 *     -> Failure/slowdown spreads through dependencies.
 *
 * Timeout
 *     -> Stop waiting after a fixed duration.
 *
 * Circuit Breaker
 *     -> Stop repeatedly calling a failing dependency.
 *
 * Fallback
 *     -> Provide a reduced but usable response.
 *
 *
 * ============================================================
 * 42. MOST IMPORTANT TRADE-OFFS
 * ============================================================
 *
 *
 * ACTIVE-PASSIVE
 *
 * Simple coordination
 *       vs
 * Failover delay
 *
 *
 * ACTIVE-ACTIVE
 *
 * No failover delay
 *       vs
 * Requires spare capacity + shared state
 *
 *
 * COLD STANDBY
 *
 * Low cost
 *       vs
 * Slow recovery
 *
 *
 * HOT STANDBY
 *
 * Fast recovery
 *       vs
 * High cost
 *
 *
 * AVAILABILITY ZONES
 *
 * Low latency
 *       vs
 * Does not protect against regional failure
 *
 *
 * MULTI-REGION
 *
 * Greater isolation
 *       vs
 * Higher latency + complexity
 *
 *
 * SYNCHRONOUS REPLICATION
 *
 * No recent data loss
 *       vs
 * Higher write latency
 *
 *
 * ASYNCHRONOUS REPLICATION
 *
 * Lower write latency
 *       vs
 * Possible recent data loss during failover
 *
 *
 * QUEUE
 *
 * Decouples background work
 *       vs
 * Work becomes asynchronous
 *
 *
 * CIRCUIT BREAKER
 *
 * Contains dependency failures
 *       vs
 * Dependency calls may temporarily be rejected
 *
 *
 * FALLBACK
 *
 * Keeps core functionality available
 *       vs
 * Response may have reduced functionality
 *
 *
 * ============================================================
 * 43. INTERVIEW MINDSET
 * ============================================================
 *
 * When designing for availability, don't simply say:
 *
 *     "Use multiple servers."
 *
 * Explain:
 *
 *     1. What can fail?
 *
 *     2. Is it a single point of failure?
 *
 *     3. How do we detect the failure?
 *
 *     4. How do we redirect/recover?
 *
 *     5. Can the remaining infrastructure handle the load?
 *
 *     6. What shared dependency can still fail?
 *
 *     7. Do we need another failure domain?
 *
 *
 * Example:
 *
 *     App Server fails
 *
 *         ↓
 *
 *     Health check detects failure
 *
 *         ↓
 *
 *     Load balancer removes server
 *
 *         ↓
 *
 *     Other active server handles traffic
 *
 *         ↓
 *
 *     Ensure remaining server has spare capacity
 *
 *
 * ============================================================
 * CORE AVAILABILITY PRINCIPLE
 * ============================================================
 *
 *
 *              COMPONENT FAILS
 *                     ↓
 *              DETECT FAILURE
 *                     ↓
 *            FAILOVER / ISOLATE
 *                     ↓
 *             KEEP USER PATH
 *                 AVAILABLE
 *
 *
 * The goal is NOT:
 *
 *     "Nothing will ever fail."
 *
 *
 * The goal is:
 *
 *     "Components will fail,
 *      but the system should continue
 *      providing a usable service."
 *
 *
 * Final mindset:
 *
 *     Measure important operations
 *             ↓
 *     Remove single points of failure
 *             ↓
 *     Add redundancy
 *             ↓
 *     Make failures independent
 *             ↓
 *     Detect unhealthy components
 *             ↓
 *     Fail over safely
 *             ↓
 *     Isolate slow dependencies
 *             ↓
 *     Keep critical user paths working
 *
 * ============================================================
 */
```

```python
/**
 * AVAILABILITY — SYSTEM DESIGN
 *
 * Core idea:
 * > Component Fails 
 * > Detect Failure 
 * > Failover/Remove Failed Component
 * > Keep Serving Users
 *
 * 
 * Availability: "Can users reach and use the system when they need it?"
 * - Scalability : Can the system handle more load?
 * - Availability: Can users still use the system when something fails?
 * - Reliability : Does the system continue behaving correctly over time?
 *
 * Example:
 * - Store processes 10,000 orders/minute                -> Scalability
 * - One server crashes but customers can still checkout -> Availability
 * - Order total is always calculated correctly          -> Reliability
 */

/**
 * 1. THE AVAILABILITY PROBLEM
 *    Initial architecture: Customers -> App Server -> Database
 *
 *    Problem:
 *    - App Server crashes, Database is still healthy,
 *    - BUT customers cannot use the application.
 *    - Why? The app server was a Single Point of Failure (SPOF).
 *
 *    So, Availability requires another way to serve requests when one component fails.
 */

/**
 * 2. AVAILABILITY vs SCALABILITY vs RELIABILITY
 *
 *    a. SCALABILITY  : Can the system handle more load?
 *                      e.g. handle 10x more orders during a sale.
 * 
 *    b. AVAILABILITY : Can users reach and use the system?
 *                      e.g. customers can checkout when one server crashes.
 * 
 *    c. RELIABILITY  : Does the system behave correctly over time?
 *                      e.g. every order total is calculated correctly.
 *
 *    IMPORTANT: 
 *    - A system can be available but not reliable.
 *    - Example: store accepts orders BUT calculates the wrong total.
 */

/**
 * 3. MEASURING AVAILABILITY
 *    Availability = Uptime / (Uptime + Downtime)
 *
 *    Example:
 *      Uptime = 364 days, Downtime = 1 day
 *      Availability = 364 / 365 = 99.73%
 *
 *    IMPORTANT:
 *      Don't look at one availability number for the whole application.
 *      Measure important operations separately:
 *        Browse Products -> 99.99%
 *        Add to Cart     -> 99.95%
 *        Checkout        -> 99.5%
 *      A whole-store average may hide a serious checkout outage.
 *
 *    Therefore: measure availability per important operation
 *    over a defined time window.
 */

/**
 * 4. THE NINES OF AVAILABILITY
 *      99%      -> ~3.65 days downtime/year
 *      99.9%    -> ~8.76 hours/year
 *      99.99%   -> ~52.6 minutes/year
 *      99.999%  -> ~5.26 minutes/year
 *      99.9999% -> ~31.5 seconds/year
 *
 *    IMPORTANT: every additional nine cuts allowed downtime by roughly 10x.
 *    Example: 99.9% -> 99.99% means allowed downtime becomes roughly 1/10.
 */

/**
 * 5. HIGHER AVAILABILITY = HIGHER COST
 *    Usually requires: redundancy, spare capacity, automation, monitoring,
 *    health checks, tested recovery, faster failover, multiple locations.
 *
 *    99.9%   -> Second server, basic monitoring
 *    99.99%  -> Automatic failover, health checks, on-call rotation, tested recovery
 *    99.999% -> Redundancy at every layer, multiple locations,
 *               automated recovery, minimal/no human intervention
 *
 *    IMPORTANT: availability is a business decision as well as a technical one.
 *    More redundancy costs more money and adds operational complexity.
 */

/**
 * 6. COMPONENTS IN SERIES
 *    Every component must be available for the request to succeed.
 *
 *      Customer -> Web Server (99.9%) -> App Server (99.9%) -> Database (99.9%)
 *
 *    If ANY required component fails, the request fails.
 *
 *    Overall Availability = A1 x A2 x A3
 *    Example: 99.9% x 99.9% x 99.9% = ~99.7%
 *
 *    More required components -> lower overall availability.
 *    IMPORTANT: every component added to the request path adds another
 *    possible failure point.
 */

/**
 * 7. COMPONENTS IN PARALLEL
 *    Parallel components provide redundancy.
 *
 *              Load Balancer
 *                /       \
 *               v         v
 *          App Server  App Server
 *             99.9%       99.9%
 *
 *    Either server can handle the request.
 *    System fails only when BOTH servers fail.
 *
 *    Each server: 99.9% available -> 0.1% failure probability
 *    Both fail  : 0.1% x 0.1% = 0.0001%
 *    Availability = 100% - 0.0001% = 99.9999%
 *
 *    IMPORTANT: theoretical calculation, assumes independent failures.
 */

/**
 * 8. SERIES vs PARALLEL
 *
 *    SERIES: Request -> A -> B -> C
 *      A, B or C fails -> system fails.
 *      Combined availability: LOWER.
 *
 *    PARALLEL:        B
 *                    /
 *             Request
 *                    \
 *                     C
 *      System continues if B works OR C works.
 *      Combined availability: HIGHER.
 *
 *    CORE IDEA:
 *      Series   : any failure can break the path.
 *      Parallel : redundant components survive individual failures.
 */

/**
 * 9. INDEPENDENCE MATTERS
 *    Parallel servers improve availability only if failures are reasonably independent.
 *
 *    Two servers can still fail together if they share:
 *      same power source, rack, network, switch, DNS,
 *      faulty deployment, software bug.
 *
 *    Example:
 *      Bad Deployment --> App1 DOWN + App2 DOWN
 *      Both servers are redundant, but the shared failure removed both.
 *
 *    Common shared-failure categories:
 *      HARDWARE    : drive, server, switch, power supply
 *      SOFTWARE    : bug, memory leak, deadlock, cascading failure
 *      NETWORK     : packet loss, latency spike, network partition, DNS failure
 *      HUMAN ERROR : bad configuration, bad deployment, accidental deletion
 */

/**
 * 10. REDUNDANCY
 *     Provide another way to handle requests when one component fails.
 *
 *     Example: App Server 1 + App Server 2.
 *     If App Server 1 disappears, App Server 2 keeps serving traffic.
 *
 *     At large scale hardware failures are normal events.
 *     Design assuming individual machines can fail at any time.
 */

/**
 * 11. ACTIVE-PASSIVE REDUNDANCY
 *     One component is active, another waits as standby.
 *
 *       App Servers --> Primary Database --(replication)--> Standby Database
 *
 *     Primary : Active, handles reads/writes.
 *     Standby : Passive, waits for failure.
 *
 *     If primary fails:
 *       1. Monitor detects failure
 *       2. Standby is promoted
 *       3. Application connections are redirected
 *       4. Writes resume
 *
 *     This process is FAILOVER.
 */

/**
 * 12. ACTIVE-PASSIVE — PROS & CONS
 *     PROS:
 *       + One clear writer
 *       + Easier data coordination
 *       + Simple architecture
 *       + Standby doesn't need production traffic
 *
 *     CONS:
 *       - Failover takes time
 *       - Standby may be untested
 *       - Split-brain risk
 *
 *     Split-brain:
 *       Monitor wrongly promotes standby while old primary is still alive.
 *       Result: Primary and Standby both accept writes
 *       -> two writers may create conflicting data.
 */

/**
 * 13. ACTIVE-PASSIVE FAILOVER FLOW
 *       Primary Database FAILURE
 *         -> Monitor detects failure
 *         -> Promote Standby
 *         -> Redirect App Connections
 *         -> Writes Resume
 *
 *     Failover is NOT instant. Time includes: detection, promotion, reconnection.
 *     So active-passive may cause a short interruption.
 */

/**
 * 14. COLD / WARM / HOT STANDBY
 *
 *     COLD STANDBY
 *       Not running, may need data restoration.
 *       Failover: minutes | Cost: lowest
 *
 *     WARM STANDBY
 *       Running, configured, not serving production traffic.
 *       Failover: seconds to minutes | Cost: medium
 *
 *     HOT STANDBY
 *       Running, data synchronized, ready to serve.
 *       Failover: seconds | Cost: highest
 *
 *     Trade-off: faster recovery = more infrastructure + higher cost.
 *
 *     IMPORTANT: test the standby under realistic production load.
 *     A backup that can't handle production traffic is useless in a real outage.
 */

/**
 * 15. ACTIVE-ACTIVE REDUNDANCY
 *     Multiple servers actively serve traffic.
 *
 *       Customers --> Load Balancer --> App 1 (Active) / App 2 (Active)
 *
 *     If App 1 fails: Load Balancer --> App 2.
 *     No traditional standby promotion needed.
 */

/**
 * 16. ACTIVE-ACTIVE — REQUIREMENTS
 *     Surviving servers need spare capacity.
 *
 *     Example:
 *       App 1 -> 50% load, App 2 -> 50% load
 *       App 1 fails -> App 2 -> 100% load
 *       If App 2 was already near its limit, App 2 may also fail.
 *
 *     Therefore: keep enough spare capacity to absorb the failed server's traffic.
 */

/**
 * 17. STATELESS SERVERS + ACTIVE-ACTIVE
 *     Active-active works well with stateless application servers.
 *
 *     Bad design:
 *       App 1 -> Local Session | App 2 -> Different Local Session
 *       Any request can go to either server, so state must not be trapped
 *       inside one server.
 *
 *     Better:
 *       App 1 --\
 *                +--> Shared Store (Sessions, Carts)
 *       App 2 --/
 *
 *     Benefit: any healthy server can handle the next request.
 */

/**
 * 18. ACTIVE-ACTIVE — PROS & CONS
 *     PROS:
 *       + No failover delay
 *       + Every node is tested under real load
 *       + Better hardware utilization
 *
 *     CONS:
 *       - Requires spare capacity
 *       - Requires shared state
 *       - Harder for databases (multiple active writers must coordinate changes)
 */

/**
 * 19. HEALTH CHECKS
 *     Load balancer uses health checks to decide whether a server gets traffic.
 *
 *       Customer --> Load Balancer --> App 1 (Healthy)
 *                                  \-> App 2 (Failed)
 *
 *     Load balancer sends traffic to App 1, stops traffic to App 2.
 *
 *     Health checks run periodically, so failure detection is not instant.
 *     Some requests may fail:
 *       - before detection
 *       - during detection
 *       - while requests are already in progress
 */

/**
 * 20. GOOD HEALTH CHECK
 *     Verify that the server can actually handle requests.
 *
 *     Not enough: "Is the process running?"
 *     Better    : "Can this server actually serve traffic?"
 *
 *     Example checks:
 *       Process -> Running
 *       Database connection -> Working
 *       Critical dependency -> Working
 *       Request path -> Healthy
 */

/**
 * 21. LOAD BALANCER IS ALSO A FAILURE POINT
 *       Customers --> Load Balancer --> App 1 / App 2
 *
 *     If the Load Balancer fails: App 1 and App 2 are healthy
 *     BUT customers cannot reach them.
 *
 *     Therefore the Load Balancer itself needs redundancy.
 *
 *     Managed cloud load balancers generally provide redundant infrastructure.
 *
 *     Self-managed approach:
 *       Load Balancer 1 -> Active
 *       Load Balancer 2 -> Standby
 *       Both share a virtual IP.
 *       If active LB fails, standby takes over.
 */

/**
 * 22. SPREADING ACROSS LOCATIONS
 *     Multiple servers aren't enough if all are in the same facility.
 *
 *     Shared location risks: power outage, network outage, facility failure.
 *     Solution: spread infrastructure across locations.
 *
 *     Levels:
 *       Availability Zones -> Regions -> Multiple Cloud Providers
 */

/**
 * 23. AVAILABILITY ZONES
 *     Separate data centers within the same cloud region.
 *
 *     Protect against: single data-center failure, power failure, network failure.
 *     Typical latency: low, usually a few milliseconds.
 *
 *       Region --+-- AZ 1
 *                +-- AZ 2
 *                +-- AZ 3
 *
 *     App continues if one AZ fails, assuming the remaining AZs have capacity.
 */

/**
 * 24. MULTI-REGION
 *     Regions are geographically separated.
 *
 *       US East <--(replication)--> Europe
 *
 *     Protects against: regional outage, regional disaster.
 *
 *     Trade-off: higher network latency, more complicated data replication.
 */

/**
 * 25. MULTIPLE CLOUD PROVIDERS
 *     Example: AWS + Google Cloud
 *
 *     Protects against: provider-wide outage.
 *     But adds: operational complexity, different tooling,
 *     different networking, different failure behavior.
 *
 *     IMPORTANT: more redundancy is not automatically better.
 *     Choose based on failure scenarios, cost, complexity, business requirements.
 */

/**
 * 26. AVAILABILITY LEVELS
 *
 *     AVAILABILITY ZONE : same region, lower latency,
 *                         protects against one data-center failure.
 *     REGION            : different geographic area, higher latency,
 *                         protects against regional outage.
 *     MULTIPLE CLOUDS   : different infrastructure providers,
 *                         protects against provider-wide outage.
 *
 *     Trade-off: more separation = more isolation,
 *                BUT also more latency + complexity.
 */

/**
 * 27. SYNCHRONOUS vs ASYNCHRONOUS REPLICATION
 *
 *     SYNCHRONOUS:
 *       Write --> Primary + Remote Replica --> Replica confirms --> Write succeeds
 *       Write succeeds only after the remote replica confirms.
 *       Benefit : no data loss on failover.
 *       Cost    : every write waits for the remote network round trip.
 *
 *     ASYNCHRONOUS:
 *       Write --> Primary --> Confirm immediately --> Remote Replica catches up later
 *       Benefit : no extra remote round-trip latency for confirming the write.
 *       Risk    : recent writes may not yet exist on the remote replica.
 *       So a regional failure may lose the latest few seconds/minutes of
 *       changes, depending on replication lag.
 */

/**
 * 28. REDUNDANCY IS NOT FREE
 *     Every additional server, standby, replica, availability zone, region
 *     or cloud provider adds:
 *       cost, operational complexity, monitoring requirements, failure scenarios.
 *
 *     Ask: "Which failures do we actually need to survive?"
 *     Then choose the appropriate redundancy level.
 */

/**
 * 29. QUEUES FOR OVERLOAD
 *     Failures can come from overload, not just broken hardware.
 *
 *     Example (big sale):
 *       Order traffic increases -> Confirmation emails increase
 *       -> Email provider becomes slow -> Checkout waits for email
 *       -> Checkout becomes slow
 *
 *     Better architecture:
 *       Customer --> App Server --+--> Database
 *                                 +--> Durable Queue --> Worker 1 / Worker 2
 *                                                              |
 *                                                              v
 *                                                       Email Provider
 *
 *     Request flow:
 *       1. Save order
 *       2. Enqueue email job
 *       3. Confirm order
 *       4. Worker sends email later
 *
 *     User does NOT wait for email delivery.
 */

/**
 * 30. DURABLE QUEUE
 *     Queue stores the background job until a worker confirms completion.
 *
 *     If Worker 1 crashes, the job is not necessarily lost.
 *     Another worker can retry it.
 *
 *       Queue -> Worker 1 -> CRASH -> Worker 2 -> Process Job
 *
 *     This makes background processing more resilient.
 */

/**
 * 31. CASCADING FAILURE
 *     One slow/failing dependency causes problems to spread through the system.
 *
 *     Example:
 *       Recommendation Service becomes slow
 *         -> App requests wait
 *         -> Threads exhausted
 *         -> Product pages slow AND Checkout slow
 *
 *     The recommendation service may not be part of checkout's business
 *     requirement, but its slowdown can still affect checkout.
 */

/**
 * 32. TIMEOUTS
 *     Stop waiting for a dependency after a fixed amount of time.
 *
 *     Example:
 *       App --> request recommendation --> Recommendation Service
 *       Timeout: 200 ms. If the response takes longer, stop waiting.
 *
 *     Benefit: prevents requests from waiting indefinitely.
 */

/**
 * 33. CIRCUIT BREAKER
 *     Prevents repeated calls to a dependency that is failing.
 *     Three states: CLOSED, OPEN, HALF-OPEN
 *
 *     CLOSED
 *       Normal operation, requests allowed through (App -> Dependency).
 *       If failures exceed threshold: CLOSED -> OPEN.
 *
 *     OPEN
 *       Dependency considered unhealthy.
 *       Calls rejected immediately, no waiting for the slow/failing dependency.
 *       After cooldown: OPEN -> HALF-OPEN.
 *
 *     HALF-OPEN
 *       Send a small number of test requests.
 *       Test succeeds: HALF-OPEN -> CLOSED.
 *       Test fails   : HALF-OPEN -> OPEN.
 */

/**
 * 34. CIRCUIT BREAKER FLOW
 *
 *              CLOSED
 *                |
 *       failures exceed threshold
 *                v
 *              OPEN
 *                |
 *          cooldown expires
 *                v
 *           HALF-OPEN
 *            /       \
 *        success    failure
 *           v          v
 *        CLOSED      OPEN
 *
 *     Main purpose: contain dependency failure.
 */

/**
 * 35. FALLBACK
 *     Return a reduced but still usable response when an optional
 *     dependency is unavailable.
 *
 *     Example: Product Page -> Recommendation Service fails.
 *       Without fallback: product page fails.
 *       With fallback   : product page WITHOUT recommendations.
 *
 *     Customer can still browse the product and place an order.
 *
 *     IMPORTANT: optional features should not take down critical functionality.
 */

/**
 * 36. TIMEOUT + CIRCUIT BREAKER + FALLBACK
 *
 *       App Server -> Timeout -> Circuit Breaker -> Recommendation Service
 *
 *     Service healthy        : request succeeds.
 *     Service becomes slow   : timeout prevents indefinite waiting.
 *     Failures continue      : circuit opens.
 *
 *       Request -> Circuit Breaker (OPEN) -> Fallback
 *               -> Product Page without recommendations
 *
 *     Result: recommendation failure DOES NOT take down the product page.
 */

/**
 * 37. DESIGN FOR FAILURE
 *     Main availability mindset: "Assume components will fail."
 *
 *     Then ask what happens if:
 *       app server fails? database fails? load balancer fails?
 *       availability zone fails? region fails? dependency becomes slow?
 *       worker crashes? network fails? bad deployment happens?
 *
 *     For each failure:
 *       Detect -> Recover / Redirect -> Continue serving users
 */

/**
 * 38. AVAILABILITY DESIGN CHECKLIST
 *
 *     APPLICATION SERVERS
 *       -> Multiple instances, load balancer, stateless architecture,
 *          health checks, spare capacity
 *     DATABASE
 *       -> Primary + standby, replication, failover,
 *          test standby under realistic load
 *     LOAD BALANCER
 *       -> Health checks, remove unhealthy nodes, redundant load balancer
 *     STATE
 *       -> Shared session/cart store, no dependence on one app server's local state
 *     LOCATIONS
 *       -> Multiple availability zones, multi-region when required
 *     BACKGROUND WORK
 *       -> Durable queue, multiple workers, retry failed jobs
 *     DEPENDENCIES
 *       -> Timeouts, circuit breakers, fallbacks
 *     MONITORING
 *       -> Measure availability per important operation,
 *          monitor failures, monitor recovery
 */

/**
 * 39. FAILURE SCENARIO CHEAT SHEET
 *
 *     App server fails           -> LB removes it; another active server handles requests
 *     Database primary fails     -> Monitor detects; promote standby; redirect connections
 *     Load balancer fails        -> Use redundant load balancers
 *     One data center fails      -> Spread across availability zones
 *     Region fails               -> Multi-region deployment
 *     Background work spikes     -> Durable queue; scale workers
 *     Worker fails               -> Queue retains job; another worker retries
 *     Dependency becomes slow    -> Timeout
 *     Dependency keeps failing   -> Circuit breaker
 *     Optional dependency fails  -> Fallback
 */

/**
 * 40. FINAL AVAILABILITY ARCHITECTURE
 *
 *                      CUSTOMERS
 *                          |
 *                          v
 *                     GLOBAL DNS
 *                      /       \
 *                     v         v
 *            LOAD BALANCER   LOAD BALANCER
 *               REGION 1        REGION 2
 *                  |
 *           +------+------+
 *           |             |
 *           v             v
 *         APP 1         APP 2
 *        Active        Active
 *           |             |
 *           +------+------+
 *                  |
 *            SHARED STORE
 *           Sessions/Carts
 *
 *     Database:
 *       PRIMARY DATABASE --(replication)--> STANDBY DATABASE
 *
 *     Background:
 *       APP --> DURABLE QUEUE --> WORKER 1 / WORKER 2
 *
 *     Optional Dependency:
 *       APP -> Timeout -> Circuit Breaker -> Recommendation Service
 *                                                  | (failure)
 *                                                  v
 *                                              Fallback
 *                                                  v
 *                                  Product Page without Recommendations
 */

/**
 * 41. QUICK REVISION
 *
 *   Availability             -> Ability of users to access and use the system.
 *   Availability Formula     -> Uptime / (Uptime + Downtime)
 *   Nines                    -> Higher percentage = less allowed downtime.
 *   Series                   -> Every component must work.
 *   Parallel                 -> Any redundant component can work.
 *   Redundancy               -> Another way to serve requests.
 *   Active-Passive           -> One active + one standby.
 *   Failover                 -> Promote standby and redirect traffic.
 *   Cold Standby             -> Not running; cheapest; slowest recovery.
 *   Warm Standby             -> Running but not serving production traffic.
 *   Hot Standby              -> Running, synchronized, ready to serve.
 *   Active-Active            -> Multiple nodes serve traffic simultaneously.
 *   Health Check             -> Decides whether a server can receive traffic.
 *   Availability Zone        -> Protects against single data-center failures.
 *   Multi-Region             -> Protects against regional failures.
 *   Synchronous Replication  -> Wait for remote replica before confirming write.
 *   Asynchronous Replication -> Confirm write immediately; replica catches up later.
 *   Queue                    -> Separates background work from user requests.
 *   Cascading Failure        -> Failure/slowdown spreads through dependencies.
 *   Timeout                  -> Stop waiting after a fixed duration.
 *   Circuit Breaker          -> Stop repeatedly calling a failing dependency.
 *   Fallback                 -> Provide a reduced but usable response.
 */

/**
 * 42. MOST IMPORTANT TRADE-OFFS
 *
 *   ACTIVE-PASSIVE          : Simple coordination
 *                             vs Failover delay
 *   ACTIVE-ACTIVE           : No failover delay
 *                             vs Requires spare capacity + shared state
 *   COLD STANDBY            : Low cost
 *                             vs Slow recovery
 *   HOT STANDBY             : Fast recovery
 *                             vs High cost
 *   AVAILABILITY ZONES      : Low latency
 *                             vs Doesn't protect against regional failure
 *   MULTI-REGION            : Greater isolation
 *                             vs Higher latency + complexity
 *   SYNCHRONOUS REPLICATION : No recent data loss
 *                             vs Higher write latency
 *   ASYNC REPLICATION       : Lower write latency
 *                             vs Possible recent data loss during failover
 *   QUEUE                   : Decouples background work
 *                             vs Work becomes asynchronous
 *   CIRCUIT BREAKER         : Contains dependency failures
 *                             vs Dependency calls may be temporarily rejected
 *   FALLBACK                : Keeps core functionality available
 *                             vs Response may have reduced functionality
 */

/**
 * 43. INTERVIEW MINDSET
 *     Don't simply say "Use multiple servers." Explain:
 *       1. What can fail?
 *       2. Is it a single point of failure?
 *       3. How do we detect the failure?
 *       4. How do we redirect/recover?
 *       5. Can the remaining infrastructure handle the load?
 *       6. What shared dependency can still fail?
 *       7. Do we need another failure domain?
 *
 *     Example:
 *       App Server fails
 *         -> Health check detects failure
 *         -> Load balancer removes server
 *         -> Other active server handles traffic
 *         -> Ensure remaining server has spare capacity
 */

/**
 * CORE AVAILABILITY PRINCIPLE
 *
 *   COMPONENT FAILS -> DETECT FAILURE -> FAILOVER / ISOLATE
 *   -> KEEP USER PATH AVAILABLE
 *
 *   The goal is NOT: "Nothing will ever fail."
 *   The goal is:     "Components will fail, but the system should continue
 *                     providing a usable service."
 *
 *   Final mindset:
 *     Measure important operations
 *       -> Remove single points of failure
 *       -> Add redundancy
 *       -> Make failures independent
 *       -> Detect unhealthy components
 *       -> Fail over safely
 *       -> Isolate slow dependencies
 *       -> Keep critical user paths working
 */
```
