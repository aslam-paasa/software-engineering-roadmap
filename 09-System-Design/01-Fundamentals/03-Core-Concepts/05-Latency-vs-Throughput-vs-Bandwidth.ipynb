```python
/**
 * ============================================================
 * LATENCY vs THROUGHPUT vs BANDWIDTH
 * ============================================================
 *
 * System Design mein performance discuss karte time
 * 3 important terms baar-baar aate hain:
 *
 *     1. Latency
 *     2. Throughput
 *     3. Bandwidth
 *
 *
 * Ye teenon same cheez nahi hain.
 *
 *
 * Simple meaning:
 *
 *     Latency   -> Ek request ko complete hone mein kitna time?
 *
 *     Throughput -> Ek second mein kitna kaam complete?
 *
 *     Bandwidth -> Maximum kitna data transfer ho sakta hai?
 *
 *
 * ============================================================
 * 1. HIGHWAY ANALOGY
 * ============================================================
 *
 * In teen concepts ko highway se easily samajh sakte hain.
 *
 *
 *     EXIT 1
 *       |
 *       |
 *       |==============================|
 *       |==============================|
 *       |==============================|
 *       |
 *       |
 *     EXIT 10
 *
 *
 * BANDWIDTH:
 *
 *     Highway par kitni lanes hain.
 *
 *     More lanes
 *         ↓
 *     More cars can travel simultaneously.
 *
 *
 *     Bandwidth = Road capacity
 *
 *
 * THROUGHPUT:
 *
 *     Actually kitni cars per hour
 *     Exit 10 tak pahunch rahi hain.
 *
 *
 *     Throughput = Actual work completed
 *
 *
 * LATENCY:
 *
 *     Ek individual car ko
 *     Exit 1 -> Exit 10 jaane mein
 *     kitna time laga.
 *
 *
 *     Latency = Time taken by one request
 *
 *
 * ============================================================
 * 2. HIGH BANDWIDTH != HIGH THROUGHPUT
 * ============================================================
 *
 * Example:
 *
 *     Highway:
 *
 *         4 lanes
 *
 *     => High bandwidth
 *
 *
 *     But accident ho gaya:
 *
 *         Traffic blocked
 *
 *     => Low throughput
 *
 *
 *     Aur individual car ko:
 *
 *         2 hours
 *
 *     lag rahe hain.
 *
 *     => High latency
 *
 *
 * Therefore:
 *
 *     High Bandwidth
 *         does NOT guarantee
 *     High Throughput
 *
 *
 *     High Bandwidth
 *         does NOT guarantee
 *     Low Latency
 *
 *
 * ============================================================
 * 3. LATENCY
 * ============================================================
 *
 * Latency =
 *
 *     Ek request ko source se destination
 *     aur response ke saath wapas aane mein
 *     kitna time laga.
 *
 *
 * Networking mein often:
 *
 *     RTT = Round Trip Time
 *
 *
 *     Client
 *       |
 *       | Request
 *       v
 *     Server
 *       |
 *       | Response
 *       v
 *     Client
 *
 *
 *     Latency = Total Round Trip Time
 *
 *
 * Example:
 *
 *     Request sent:
 *
 *         t = 0 ms
 *
 *     Response received:
 *
 *         t = 150 ms
 *
 *
 *     Latency = 150 ms
 *
 *
 * ============================================================
 * 4. LATENCY KE COMPONENTS
 * ============================================================
 *
 * Total latency ek single delay nahi hota.
 *
 * Multiple delays milkar total latency banate hain.
 *
 *
 *     Total Latency
 *          |
 *          +----------------------+
 *          |                      |
 *          v                      v
 *     Network delays          Processing
 *
 *
 * Main components:
 *
 *
 * 1. PROPAGATION DELAY
 *
 *     Signal ko physical medium mein
 *     travel karne mein kitna time laga.
 *
 *
 *     More geographic distance
 *             ↓
 *     More propagation delay
 *
 *
 *     Example:
 *
 *     India ---------------- USA
 *
 *     Long distance
 *          ↓
 *     Higher propagation delay
 *
 *
 * ------------------------------------------------------------
 *
 * 2. TRANSMISSION DELAY
 *
 *     Data ke bits ko wire par push
 *     karne mein kitna time laga.
 *
 *
 *     Depends on:
 *
 *         Packet size
 *         Link bandwidth
 *
 *
 * ------------------------------------------------------------
 *
 * 3. PROCESSING DELAY
 *
 *     Routers, load balancers,
 *     servers etc. request ko process
 *     karne mein jo time lete hain.
 *
 *
 * ------------------------------------------------------------
 *
 * 4. QUEUING DELAY
 *
 *     Jab component busy ho,
 *     request queue/buffer mein wait karti hai.
 *
 *
 *     High traffic
 *         ↓
 *     Queue grows
 *         ↓
 *     Waiting time increases
 *         ↓
 *     Latency increases
 *
 *
 * ============================================================
 * 5. LATENCY KO KAISE MEASURE KARTE HAIN?
 * ============================================================
 *
 * Average latency alone misleading ho sakti hai.
 *
 * Isliye percentiles use karte hain.
 *
 *
 * p50:
 *
 *     50% requests isse faster hain.
 *
 *     Also called:
 *
 *         Median
 *
 *
 * p95:
 *
 *     95% requests isse faster hain.
 *
 *
 * p99:
 *
 *     99% requests isse faster hain.
 *
 *
 * p99.9:
 *
 *     99.9% requests isse faster hain.
 *
 *
 * ============================================================
 * 6. WHY NOT JUST AVERAGE LATENCY?
 * ============================================================
 *
 * Example:
 *
 *     Average latency = 10 ms
 *
 *     Looks very good.
 *
 *     But:
 *
 *     p99 = 500 ms
 *
 *
 * Means:
 *
 *     1% users ko request complete hone mein
 *     500 ms lag rahe hain.
 *
 *
 * Therefore:
 *
 *     Average can hide slow requests.
 *
 *
 * Interview mein:
 *
 *     "Latency should be analyzed using
 *      p50, p95 and p99 rather than
 *      only average latency."
 *
 *
 * ============================================================
 * 7. FACTORS AFFECTING LATENCY
 * ============================================================
 *
 *
 * Geographic Distance
 *     ↓
 * More distance = more propagation delay
 *
 *
 * Network Congestion
 *     ↓
 * More queueing
 *
 *
 * Server Load
 *     ↓
 * More processing time
 *
 *
 * Database Query
 *     ↓
 * Slow query = higher latency
 *
 *
 * DNS Resolution
 *     ↓
 * Extra lookup on cold request
 *
 *
 * TLS Handshake
 *     ↓
 * Additional round trips
 *
 *
 * ============================================================
 * 8. HOW TO REDUCE LATENCY?
 * ============================================================
 *
 *
 * 1. CDN
 *
 *     User
 *       |
 *       v
 *     Nearby CDN
 *       |
 *       v
 *     Content
 *
 *
 *     User ke closer content serve karo.
 *
 *
 * ------------------------------------------------------------
 *
 * 2. CACHING
 *
 *     Without cache:
 *
 *     Client
 *       |
 *       v
 *     App
 *       |
 *       v
 *     Database
 *
 *
 *     With cache:
 *
 *     Client
 *       |
 *       v
 *     Cache
 *
 *
 *     Database round trip avoid ho sakta hai.
 *
 *
 * ------------------------------------------------------------
 *
 * 3. CONNECTION POOLING
 *
 *     Har request ke liye new connection
 *     create karne ki need reduce hoti hai.
 *
 *
 * ------------------------------------------------------------
 *
 * 4. DATABASE OPTIMIZATION
 *
 *     Add indexes
 *     Optimize queries
 *
 *
 * ------------------------------------------------------------
 *
 * 5. GEOGRAPHIC DISTRIBUTION
 *
 *     Servers ko users ke closer deploy karo.
 *
 *
 * ------------------------------------------------------------
 *
 * 6. PROTOCOL OPTIMIZATION
 *
 *     HTTP/2
 *     HTTP/3
 *     QUIC
 *
 *
 * Goal:
 *
 *     Fewer / faster network interactions.
 *
 *
 * ============================================================
 * 9. THROUGHPUT
 * ============================================================
 *
 * Throughput =
 *
 *     Unit time mein kitna actual work
 *     complete hua.
 *
 *
 * Simple:
 *
 *     "System ek second mein kitna kaam kar raha hai?"
 *
 *
 * Common units:
 *
 *     RPS = Requests Per Second
 *
 *     TPS = Transactions Per Second
 *
 *
 * Example:
 *
 *     Server handles:
 *
 *         1,000 requests/second
 *
 *
 *     Throughput = 1000 RPS
 *
 *
 * ============================================================
 * 10. BANDWIDTH vs THROUGHPUT
 * ============================================================
 *
 * VERY IMPORTANT:
 *
 *
 * BANDWIDTH:
 *
 *     Maximum theoretical capacity.
 *
 *
 * THROUGHPUT:
 *
 *     Actual achieved transfer/work rate.
 *
 *
 * Example:
 *
 *     Network bandwidth:
 *
 *         1 Gbps
 *
 *
 *     Actual throughput:
 *
 *         600 Mbps
 *
 *
 * So:
 *
 *     Bandwidth = 1 Gbps
 *
 *     Throughput = 600 Mbps
 *
 *
 * Throughput generally bandwidth se lower hota hai
 * because of:
 *
 *     Protocol overhead
 *     Congestion
 *     Packet loss
 *     Processing limitations
 *     Inefficient resource utilization
 *
 *
 * ============================================================
 * 11. BANDWIDTH
 * ============================================================
 *
 * Bandwidth =
 *
 *     Maximum rate at which data
 *     can be transferred.
 *
 *
 * It measures:
 *
 *     CAPACITY
 *
 *
 * Common units:
 *
 *     Kbps
 *     Mbps
 *     Gbps
 *
 *
 * Example:
 *
 *     1 Gbps network link
 *
 *     Means theoretical maximum
 *     transfer capacity = 1 Gbps.
 *
 *
 * ============================================================
 * 12. TYPES OF BANDWIDTH
 * ============================================================
 *
 *
 * NETWORK BANDWIDTH
 *
 *     Network link ki capacity.
 *
 *     Example:
 *
 *         1 Gbps Ethernet
 *
 *
 * MEMORY BANDWIDTH
 *
 *     RAM ke saath data transfer rate.
 *
 *
 *     Example:
 *
 *         DDR4 ~ 25 GB/s
 *
 *
 * DISK BANDWIDTH
 *
 *     Storage read/write speed.
 *
 *
 *     Example:
 *
 *         SSD ~ 500 MB/s
 *
 *
 * BUS BANDWIDTH
 *
 *     Internal hardware data transfer rate.
 *
 *
 *     Example:
 *
 *         PCIe
 *
 *
 * ============================================================
 * 13. THROUGHPUT KA BOTTLENECK
 * ============================================================
 *
 * Very important system-design concept:
 *
 *
 *     System throughput ko
 *     bottleneck determine karta hai.
 *
 *
 * Simple rule:
 *
 *     System is only as fast as
 *     its slowest component.
 *
 *
 * Example:
 *
 *
 *     Client
 *       |
 *       v
 *     Server
 *       |
 *       v
 *     Database
 *
 *
 * Server:
 *
 *     10,000 RPS
 *
 * Database:
 *
 *     2,000 RPS
 *
 *
 * Overall system:
 *
 *     ~2,000 RPS
 *
 *
 * Because:
 *
 *     Database = Bottleneck
 *
 *
 * ============================================================
 * 14. HOW TO IMPROVE THROUGHPUT?
 * ============================================================
 *
 *
 * 1. HORIZONTAL SCALING
 *
 *     More servers add karo.
 *
 *
 *     LB
 *      |
 *    +---+---+---+
 *    |   |   |   |
 *   S1  S2  S3  S4
 *
 *
 *     Work distribute hota hai.
 *
 *
 * ------------------------------------------------------------
 *
 * 2. VERTICAL SCALING
 *
 *     Existing server ko more:
 *
 *         CPU
 *         Memory
 *
 *     provide karo.
 *
 *
 * ------------------------------------------------------------
 *
 * 3. ASYNC PROCESSING
 *
 *     Slow operation ko synchronous
 *     request path se hatao.
 *
 *
 *     Request
 *        |
 *        v
 *     API
 *        |
 *        v
 *     Queue
 *        |
 *        v
 *     Worker
 *
 *
 * User ko worker complete hone ka wait
 * nahi karna pad sakta.
 *
 *
 * ------------------------------------------------------------
 *
 * 4. BATCHING
 *
 *     Individual operations:
 *
 *         A
 *         B
 *         C
 *         D
 *
 *
 *     Instead:
 *
 *         [A B C D]
 *
 *     ek saath process karo.
 *
 *
 * ------------------------------------------------------------
 *
 * 5. CACHING
 *
 *     Repeated work avoid karo.
 *
 *
 * ------------------------------------------------------------
 *
 * 6. CONNECTION POOLING
 *
 *     Expensive connections ko reuse karo.
 *
 *
 * ------------------------------------------------------------
 *
 * 7. LOAD BALANCING
 *
 *     Work ko servers mein evenly distribute karo.
 *
 *
 * ============================================================
 * 15. THROUGHPUT LIMITERS
 * ============================================================
 *
 * Throughput ko different bottlenecks limit kar sakte hain.
 *
 *
 * CPU
 *
 *     Compute-heavy tasks.
 *
 *
 * MEMORY
 *
 *     Large working sets.
 *
 *
 * I/O
 *
 *     Disk / Network slow.
 *
 *
 * CONNECTIONS
 *
 *     Connection limit reached.
 *
 *
 * CONTENTION
 *
 *     Locks/shared resources par
 *     multiple workers compete kar rahe hain.
 *
 *
 * ============================================================
 * 16. LATENCY vs THROUGHPUT
 * ============================================================
 *
 * Dono related hain,
 * but same nahi hain.
 *
 *
 * LATENCY:
 *
 *     One request ka time.
 *
 *
 * THROUGHPUT:
 *
 *     Time period mein completed work.
 *
 *
 * Example:
 *
 *     Request latency = 100 ms
 *
 *     But multiple workers concurrently
 *     requests process kar sakte hain.
 *
 *
 * Therefore:
 *
 *     Low latency does not automatically mean
 *     high overall throughput.
 *
 *
 * ============================================================
 * 17. THROUGHPUT AND LATENCY RELATION
 * ============================================================
 *
 * PDF mein single-worker example:
 *
 *
 *     Throughput = 1 / Latency
 *
 *
 * Example:
 *
 *     Latency = 10 ms
 *
 *     10 ms = 0.01 second
 *
 *
 *     Throughput =
 *
 *         1 / 0.01
 *
 *         = 100 requests/second
 *
 *
 * IMPORTANT:
 *
 *     Ye simple relationship
 *     single-threaded / single-worker
 *     situation ke context mein samjho.
 *
 *
 * ============================================================
 * 18. MULTI-WORKER SYSTEM
 * ============================================================
 *
 * Multiple concurrent workers hone par:
 *
 *
 *     Throughput =
 *
 *         Concurrent Workers / Latency
 *
 *
 * Example:
 *
 *     Latency = 10 ms
 *
 *     Workers = 50
 *
 *
 *     10 ms = 0.01 sec
 *
 *
 *     Throughput =
 *
 *         50 / 0.01
 *
 *         = 5,000 requests/second
 *
 *
 * Isliye concurrency throughput increase
 * kar sakti hai, jab tak koi other bottleneck
 * hit na ho.
 *
 *
 * ============================================================
 * 19. LITTLE'S LAW
 * ============================================================
 *
 * Important system-design relationship:
 *
 *
 *     Concurrency =
 *
 *         Throughput × Latency
 *
 *
 * Therefore:
 *
 *
 *     To increase throughput:
 *
 *         Increase concurrency
 *
 *         OR
 *
 *         Reduce per-request latency
 *
 *
 * BUT:
 *
 *     Concurrency blindly increase karna
 *     useful nahi hai.
 *
 *     CPU, DB, network, connection limits etc.
 *     bottleneck ban sakte hain.
 *
 *
 * ============================================================
 * 20. LATENCY vs THROUGHPUT TRADE-OFF
 * ============================================================
 *
 * Kabhi-kabhi throughput increase karne ke liye
 * batching / queueing / pipelining use karte hain.
 *
 *
 * Result:
 *
 *     Throughput ↑
 *
 *     But individual request latency ↑
 *
 *
 * Example:
 *
 *     1 item at a time
 *
 *         Low waiting
 *         Lower batching efficiency
 *
 *
 *     100 items together
 *
 *         Better throughput
 *         But item may wait for batch
 *
 *
 * Therefore:
 *
 *     More throughput can sometimes
 *     increase latency.
 *
 *
 * ============================================================
 * 21. BANDWIDTH-DELAY PRODUCT (BDP)
 * ============================================================
 *
 * BDP connects:
 *
 *     Bandwidth
 *          +
 *     Latency
 *
 *
 * Formula:
 *
 *
 *     BDP = Bandwidth × Latency
 *
 *
 * Meaning:
 *
 *     Network mein ek time par
 *     kitna data "in flight" ho sakta hai.
 *
 *
 * ============================================================
 * 22. BDP EXAMPLE
 * ============================================================
 *
 * Bandwidth:
 *
 *     1 Gbps
 *
 *
 * Convert:
 *
 *     1 Gbps = 125 MB/s
 *
 *
 * Latency:
 *
 *     100 ms
 *
 *     = 0.1 sec
 *
 *
 * Therefore:
 *
 *     BDP =
 *
 *         125 MB/s × 0.1 s
 *
 *         = 12.5 MB
 *
 *
 * Meaning:
 *
 *     Around 12.5 MB data
 *     network pipe mein simultaneously
 *     in-flight ho sakta hai.
 *
 *
 * ============================================================
 * 23. TCP WINDOW vs BDP
 * ============================================================
 *
 * Agar TCP window:
 *
 *     BDP se smaller hai
 *
 * then:
 *
 *     Available network bandwidth
 *     fully utilize nahi ho payegi.
 *
 *
 * Example:
 *
 *     BDP = 12.5 MB
 *
 *     TCP window = 5 MB
 *
 *
 * Then:
 *
 *     Network capacity available hai
 *
 *     BUT
 *
 *     TCP window us capacity ko
 *     fully utilize nahi kar pa rahi.
 *
 *
 * ============================================================
 * 24. LATENCY vs THROUGHPUT vs BANDWIDTH
 * ============================================================
 *
 *
 * +-------------+------------------------+----------------+
 * | Metric      | Meaning                | Unit           |
 * +-------------+------------------------+----------------+
 * | Latency     | Time per request       | ms             |
 * | Throughput  | Work per unit time     | RPS/TPS/MB/s   |
 * | Bandwidth   | Maximum data capacity  | Mbps/Gbps      |
 * +-------------+------------------------+----------------+
 *
 *
 * LATENCY:
 *
 *     "Kitna time?"
 *
 *
 * THROUGHPUT:
 *
 *     "Kitna kaam per second?"
 *
 *
 * BANDWIDTH:
 *
 *     "Maximum kitna data transfer?"
 *
 *
 * ============================================================
 * 25. WHAT TO OPTIMIZE?
 * ============================================================
 *
 *
 * LATENCY HIGH:
 *
 *     Use:
 *
 *     CDN
 *     Caching
 *     Geographic placement
 *     Fewer round trips
 *     Connection pooling
 *     DB optimization
 *     HTTP/2 / HTTP/3
 *
 *
 * ------------------------------------------------------------
 *
 * THROUGHPUT LOW:
 *
 *     Use:
 *
 *     Horizontal scaling
 *     Concurrency
 *     Async processing
 *     Batching
 *     Caching
 *     Connection pooling
 *     Load balancing
 *
 *
 * ------------------------------------------------------------
 *
 * BANDWIDTH LOW:
 *
 *     Use:
 *
 *     Better network link
 *     Compression
 *     Parallel connections
 *
 *
 * ============================================================
 * 26. PERFORMANCE DEBUGGING MINDSET
 * ============================================================
 *
 * DON'T SAY:
 *
 *     "System slow hai, server bada kar do."
 *
 *
 * FIRST FIND:
 *
 *     Actual bottleneck kya hai?
 *
 *
 * Possible bottleneck:
 *
 *     Network
 *     CPU
 *     Memory
 *     Database
 *     Disk
 *     Connection pool
 *     Queue
 *     Lock/contention
 *
 *
 * Then optimize that component.
 *
 *
 * Example:
 *
 *     DB is bottleneck
 *
 *     Increasing network bandwidth
 *     may not improve throughput.
 *
 *
 * Therefore:
 *
 *     FIND BOTTLENECK FIRST.
 *
 *
 * ============================================================
 * 27. INTERVIEW QUESTIONS
 * ============================================================
 *
 *
 * Q1. What is latency?
 *
 * Answer:
 *
 *     Latency is the time taken by a request
 *     to travel from source to destination
 *     and receive a response.
 *
 *     Networking context mein ise often
 *     Round Trip Time (RTT) kehte hain.
 *
 *
 * ------------------------------------------------------------
 *
 * Q2. What is throughput?
 *
 * Answer:
 *
 *     Throughput is the amount of work completed
 *     per unit of time.
 *
 *     Web systems mein commonly:
 *
 *         RPS
 *         TPS
 *
 *     use hota hai.
 *
 *
 * ------------------------------------------------------------
 *
 * Q3. What is bandwidth?
 *
 * Answer:
 *
 *     Bandwidth is the maximum rate at which
 *     data can theoretically be transferred.
 *
 *     Common units:
 *
 *         Mbps
 *         Gbps
 *
 *
 * ------------------------------------------------------------
 *
 * Q4. What is the difference between
 *    bandwidth and throughput?
 *
 * Answer:
 *
 *     Bandwidth = maximum possible capacity.
 *
 *     Throughput = actual achieved rate.
 *
 *
 *     Example:
 *
 *         Bandwidth = 1 Gbps
 *         Throughput = 600 Mbps
 *
 *
 *     Throughput usually lower hota hai because
 *     of overhead, congestion and processing limits.
 *
 *
 * ------------------------------------------------------------
 *
 * Q5. Can a system have high bandwidth
 *    but low throughput?
 *
 * Answer:
 *
 *     Yes.
 *
 *     Example:
 *
 *         1 Gbps network
 *
 *     But:
 *
 *         congestion
 *         packet loss
 *         processing bottleneck
 *
 *     ki wajah se actual throughput low ho sakta hai.
 *
 *
 * ------------------------------------------------------------
 *
 * Q6. Can a system have high bandwidth
 *    but high latency?
 *
 * Answer:
 *
 *     Yes.
 *
 *     Highway analogy:
 *
 *         Many lanes
 *             +
 *         Long distance
 *
 *     High capacity ho sakti hai,
 *     but individual journey slow ho sakti hai.
 *
 *
 * ------------------------------------------------------------
 *
 * Q7. Why are p95 and p99 important?
 *
 * Answer:
 *
 *     Average latency slow requests ko hide
 *     kar sakti hai.
 *
 *     p95/p99 tail latency ko show karte hain.
 *
 *
 *     Example:
 *
 *         Average = 10 ms
 *         p99 = 500 ms
 *
 *     Means small percentage of users
 *     very slow experience face kar rahe hain.
 *
 *
 * ------------------------------------------------------------
 *
 * Q8. What are the components of latency?
 *
 * Answer:
 *
 *     1. Propagation delay
 *     2. Transmission delay
 *     3. Processing delay
 *     4. Queuing delay
 *
 *
 * ------------------------------------------------------------
 *
 * Q9. How would you reduce latency?
 *
 * Answer:
 *
 *     Depends on bottleneck.
 *
 *     Common techniques:
 *
 *         CDN
 *         Caching
 *         Connection pooling
 *         DB optimization
 *         Geographic distribution
 *         Fewer network round trips
 *         HTTP/2 / HTTP/3
 *
 *
 * ------------------------------------------------------------
 *
 * Q10. How would you improve throughput?
 *
 * Answer:
 *
 *     First identify bottleneck.
 *
 *     Then consider:
 *
 *         Horizontal scaling
 *         Vertical scaling
 *         Async processing
 *         Batching
 *         Caching
 *         Connection pooling
 *         Load balancing
 *
 *
 * ------------------------------------------------------------
 *
 * Q11. What determines maximum throughput?
 *
 * Answer:
 *
 *     The bottleneck / slowest component
 *     in the processing path.
 *
 *
 *     Example:
 *
 *         API -> DB
 *
 *         API = 10K RPS
 *         DB  = 2K RPS
 *
 *         System throughput ≈ 2K RPS
 *
 *
 * ------------------------------------------------------------
 *
 * Q12. What is BDP?
 *
 * Answer:
 *
 *     Bandwidth-Delay Product.
 *
 *
 *     Formula:
 *
 *         BDP = Bandwidth × Latency
 *
 *
 *     It represents approximately how much
 *     data can be in flight in the network.
 *
 *
 * ------------------------------------------------------------
 *
 * Q13. What happens if TCP window
 *     is smaller than BDP?
 *
 * Answer:
 *
 *     Available bandwidth fully utilize
 *     nahi ho payegi.
 *
 *     TCP window becomes a limiting factor.
 *
 *
 * ------------------------------------------------------------
 *
 * Q14. What is the relationship between
 *     latency and throughput?
 *
 * Answer:
 *
 *     They measure different things,
 *     but are mathematically related under
 *     concurrency assumptions.
 *
 *
 *     Single worker example:
 *
 *         Throughput = 1 / Latency
 *
 *
 *     With concurrent workers:
 *
 *         Throughput =
 *         Concurrent Workers / Latency
 *
 *
 *     More generally, Little's Law:
 *
 *         Concurrency =
 *         Throughput × Latency
 *
 *
 * ------------------------------------------------------------
 *
 * Q15. Can increasing throughput increase latency?
 *
 * Answer:
 *
 *     Yes.
 *
 *     Batching, queueing and pipelining can
 *     increase throughput but may add waiting
 *     time to individual requests.
 *
 *
 * ============================================================
 * 28. BEGINNER QUICK REVISION
 * ============================================================
 *
 *
 * LATENCY
 *     =
 *     One request ko complete hone mein time.
 *
 *     Unit:
 *         milliseconds (ms)
 *
 *
 * THROUGHput
 *     =
 *     Ek unit time mein actual completed work.
 *
 *     Unit:
 *         RPS
 *         TPS
 *         MB/s
 *
 *
 * BANDWIDTH
 *     =
 *     Maximum data transfer capacity.
 *
 *     Unit:
 *         Mbps
 *         Gbps
 *
 *
 * ============================================================
 * 29. EASY MEMORY TRICK
 * ============================================================
 *
 *
 * Imagine WATER PIPE:
 *
 *
 *     SOURCE
 *       |
 *       |==============================|
 *       |==============================|
 *       |==============================|
 *       |
 *       v
 *     DESTINATION
 *
 *
 * BANDWIDTH:
 *
 *     Pipe kitni wide hai?
 *
 *
 * THROUGHPUT:
 *
 *     Actually kitna water per second
 *     flow ho raha hai?
 *
 *
 * LATENCY:
 *
 *     Water ko source se destination
 *     pahunchne mein kitna time lagta hai?
 *
 *
 * ============================================================
 * 30. GOLDEN RULE
 * ============================================================
 *
 *
 *     LATENCY
 *         ↓
 *     "How long?"
 *
 *
 *     THROUGHPUT
 *         ↓
 *     "How much work per second?"
 *
 *
 *     BANDWIDTH
 *         ↓
 *     "How much maximum capacity?"
 *
 *
 * ============================================================
 * FINAL SYSTEM DESIGN MINDSET
 * ============================================================
 *
 *
 * User says:
 *
 *     "Application slow hai."
 *
 * Don't immediately scale everything.
 *
 *
 * Ask:
 *
 *     Is latency high?
 *          |
 *          v
 *     Find delay
 *
 *
 *     Is throughput low?
 *          |
 *          v
 *     Find bottleneck
 *
 *
 *     Is bandwidth insufficient?
 *          |
 *          v
 *     Increase transfer capacity
 *
 *
 *     FIRST:
 *
 *         Identify the metric.
 *
 *     THEN:
 *
 *         Identify the bottleneck.
 *
 *     THEN:
 *
 *         Optimize the correct component.
 *
 *
 * ============================================================
 * ONE-LINE INTERVIEW ANSWER
 * ============================================================
 *
 * "Latency measures how long one request takes,
 * throughput measures how much work the system
 * completes per unit time, and bandwidth represents
 * the maximum data-transfer capacity. They are related,
 * but optimizing one does not automatically optimize
 * the others."
 *
 * ============================================================
 */
```

```python
/**
 * LATENCY vs THROUGHPUT vs BANDWIDTH
 *
 * System Design mein performance discuss karte time 3 terms baar-baar aate hain:
 *   1. Latency  2. Throughput  3. Bandwidth
 * Ye teenon same cheez nahi hain.
 *
 * Simple meaning:
 *   Latency    -> Ek request ko complete hone mein kitna time?
 *   Throughput -> Ek second mein kitna kaam complete?
 *   Bandwidth  -> Maximum kitna data transfer ho sakta hai?
 */

/**
 * 1. HIGHWAY ANALOGY
 *
 *      EXIT 1
 *        |
 *        |==============================|
 *        |==============================|
 *        |==============================|
 *        |
 *      EXIT 10
 *
 *    BANDWIDTH : highway par kitni lanes hain.
 *                More lanes -> more cars can travel simultaneously.
 *                Bandwidth = Road capacity
 *    THROUGHPUT: actually kitni cars per hour Exit 10 tak pahunch rahi hain.
 *                Throughput = Actual work completed
 *    LATENCY   : ek individual car ko Exit 1 -> Exit 10 jaane mein kitna time laga.
 *                Latency = Time taken by one request
 */

/**
 * 2. HIGH BANDWIDTH != HIGH THROUGHPUT
 *    Example:
 *      Highway: 4 lanes                       => High bandwidth
 *      Accident ho gaya, traffic blocked      => Low throughput
 *      Individual car ko 2 hours lag rahe hain => High latency
 *
 *    Therefore:
 *      High Bandwidth does NOT guarantee High Throughput.
 *      High Bandwidth does NOT guarantee Low Latency.
 */

/**
 * 3. LATENCY
 *    Ek request ko source se destination aur response ke saath wapas aane
 *    mein kitna time laga.
 *
 *    Networking mein often RTT = Round Trip Time:
 *      Client --Request--> Server --Response--> Client
 *      Latency = Total Round Trip Time
 *
 *    Example: request sent at t = 0 ms, response received at t = 150 ms
 *      Latency = 150 ms
 */

/**
 * 4. LATENCY KE COMPONENTS
 *    Total latency ek single delay nahi hota, multiple delays milkar banate hain.
 *
 *    1. PROPAGATION DELAY
 *       Signal ko physical medium mein travel karne mein kitna time laga.
 *       More geographic distance -> more propagation delay.
 *       Example: India <----> USA = long distance = higher propagation delay.
 *
 *    2. TRANSMISSION DELAY
 *       Data ke bits ko wire par push karne mein kitna time laga.
 *       Depends on: packet size, link bandwidth.
 *
 *    3. PROCESSING DELAY
 *       Routers, load balancers, servers etc. ko request process karne mein
 *       jo time lagta hai.
 *
 *    4. QUEUING DELAY
 *       Jab component busy ho, request queue/buffer mein wait karti hai.
 *       High traffic -> queue grows -> waiting time increases -> latency increases.
 */

/**
 * 5. LATENCY KO KAISE MEASURE KARTE HAIN?
 *    Average latency alone misleading ho sakti hai, isliye percentiles use karte hain.
 *
 *      p50   : 50% requests isse faster hain (Median).
 *      p95   : 95% requests isse faster hain.
 *      p99   : 99% requests isse faster hain.
 *      p99.9 : 99.9% requests isse faster hain.
 */

/**
 * 6. WHY NOT JUST AVERAGE LATENCY?
 *    Example: average latency = 10 ms (looks very good), but p99 = 500 ms.
 *    Means 1% users ko request complete hone mein 500 ms lag rahe hain.
 *
 *    Average can hide slow requests.
 *
 *    Interview mein:
 *      "Latency should be analyzed using p50, p95 and p99
 *       rather than only average latency."
 */

/**
 * 7. FACTORS AFFECTING LATENCY
 *      Geographic Distance -> more distance = more propagation delay
 *      Network Congestion  -> more queueing
 *      Server Load         -> more processing time
 *      Database Query      -> slow query = higher latency
 *      DNS Resolution      -> extra lookup on cold request
 *      TLS Handshake       -> additional round trips
 */

/**
 * 8. HOW TO REDUCE LATENCY?
 *    1. CDN
 *       User --> Nearby CDN --> Content. User ke closer content serve karo.
 *
 *    2. CACHING
 *       Without cache: Client -> App -> Database
 *       With cache   : Client -> Cache
 *       Database round trip avoid ho sakta hai.
 *
 *    3. CONNECTION POOLING
 *       Har request ke liye new connection banane ki need reduce hoti hai.
 *
 *    4. DATABASE OPTIMIZATION
 *       Add indexes, optimize queries.
 *
 *    5. GEOGRAPHIC DISTRIBUTION
 *       Servers ko users ke closer deploy karo.
 *
 *    6. PROTOCOL OPTIMIZATION
 *       HTTP/2, HTTP/3, QUIC.
 *
 *    Goal: fewer / faster network interactions.
 */

/**
 * 9. THROUGHPUT
 *    Unit time mein kitna actual work complete hua.
 *    Simple: "System ek second mein kitna kaam kar raha hai?"
 *
 *    Common units:
 *      RPS = Requests Per Second
 *      TPS = Transactions Per Second
 *
 *    Example: server handles 1,000 requests/second -> Throughput = 1000 RPS
 */

/**
 * 10. BANDWIDTH vs THROUGHPUT   (VERY IMPORTANT)
 *     BANDWIDTH : maximum theoretical capacity.
 *     THROUGHPUT: actual achieved transfer/work rate.
 *
 *     Example:
 *       Network bandwidth = 1 Gbps
 *       Actual throughput = 600 Mbps
 *
 *     Throughput generally bandwidth se lower hota hai because of:
 *       protocol overhead, congestion, packet loss,
 *       processing limitations, inefficient resource utilization.
 */

/**
 * 11. BANDWIDTH
 *     Maximum rate at which data can be transferred. Measures CAPACITY.
 *     Common units: Kbps, Mbps, Gbps
 *
 *     Example: 1 Gbps link -> theoretical maximum transfer capacity = 1 Gbps.
 */

/**
 * 12. TYPES OF BANDWIDTH
 *       NETWORK BANDWIDTH : network link ki capacity.      e.g. 1 Gbps Ethernet
 *       MEMORY BANDWIDTH  : RAM ke saath data transfer rate. e.g. DDR4 ~ 25 GB/s
 *       DISK BANDWIDTH    : storage read/write speed.       e.g. SSD ~ 500 MB/s
 *       BUS BANDWIDTH     : internal hardware data transfer rate. e.g. PCIe
 */

/**
 * 13. THROUGHPUT KA BOTTLENECK
 *     System throughput ko bottleneck determine karta hai.
 *     Rule: system is only as fast as its slowest component.
 *
 *     Example: Client -> Server -> Database
 *       Server   : 10,000 RPS
 *       Database :  2,000 RPS
 *       Overall system: ~2,000 RPS, because Database = Bottleneck.
 */

/**
 * 14. HOW TO IMPROVE THROUGHPUT?
 *     1. HORIZONTAL SCALING
 *        More servers add karo: LB --> S1 / S2 / S3 / S4. Work distribute hota hai.
 *
 *     2. VERTICAL SCALING
 *        Existing server ko more CPU / memory do.
 *
 *     3. ASYNC PROCESSING
 *        Slow operation ko synchronous request path se hatao:
 *        Request -> API -> Queue -> Worker
 *        User ko worker complete hone ka wait nahi karna pad sakta.
 *
 *     4. BATCHING
 *        Individual operations A, B, C, D ke bajay [A B C D] ek saath process karo.
 *
 *     5. CACHING
 *        Repeated work avoid karo.
 *
 *     6. CONNECTION POOLING
 *        Expensive connections ko reuse karo.
 *
 *     7. LOAD BALANCING
 *        Work ko servers mein evenly distribute karo.
 */

/**
 * 15. THROUGHPUT LIMITERS
 *     Throughput ko different bottlenecks limit kar sakte hain:
 *       CPU         : compute-heavy tasks.
 *       MEMORY      : large working sets.
 *       I/O         : disk / network slow.
 *       CONNECTIONS : connection limit reached.
 *       CONTENTION  : locks/shared resources par multiple workers compete kar rahe hain.
 */

/**
 * 16. LATENCY vs THROUGHPUT
 *     Dono related hain, but same nahi.
 *       LATENCY   : one request ka time.
 *       THROUGHPUT: time period mein completed work.
 *
 *     Example: request latency = 100 ms, but multiple workers concurrently
 *     requests process kar sakte hain.
 *
 *     Low latency does not automatically mean high overall throughput.
 */

/**
 * 17. THROUGHPUT AND LATENCY RELATION
 *     Single-worker example:
 *       Throughput = 1 / Latency
 *
 *     Example: Latency = 10 ms = 0.01 second
 *       Throughput = 1 / 0.01 = 100 requests/second
 *
 *     IMPORTANT: ye simple relationship single-threaded / single-worker
 *     situation ke context mein samjho.
 */

/**
 * 18. MULTI-WORKER SYSTEM
 *     Multiple concurrent workers hone par:
 *       Throughput = Concurrent Workers / Latency
 *
 *     Example: Latency = 10 ms (0.01 sec), Workers = 50
 *       Throughput = 50 / 0.01 = 5,000 requests/second
 *
 *     Isliye concurrency throughput badha sakti hai, jab tak koi
 *     other bottleneck hit na ho.
 */

/**
 * 19. LITTLE'S LAW
 *       Concurrency = Throughput x Latency
 *
 *     To increase throughput:
 *       Increase concurrency  OR  Reduce per-request latency.
 *
 *     BUT concurrency blindly badhana useful nahi hai.
 *     CPU, DB, network, connection limits etc. bottleneck ban sakte hain.
 */

/**
 * 20. LATENCY vs THROUGHPUT TRADE-OFF
 *     Kabhi-kabhi throughput badhane ke liye batching / queueing / pipelining
 *     use karte hain.
 *     Result: Throughput up, BUT individual request latency up.
 *
 *     Example:
 *       1 item at a time : low waiting, lower batching efficiency.
 *       100 items together: better throughput, but item may wait for batch.
 *
 *     More throughput can sometimes increase latency.
 */

/**
 * 21. BANDWIDTH-DELAY PRODUCT (BDP)
 *     BDP connects Bandwidth + Latency.
 *
 *       BDP = Bandwidth x Latency
 *
 *     Meaning: network mein ek time par kitna data "in flight" ho sakta hai.
 */

/**
 * 22. BDP EXAMPLE
 *     Bandwidth: 1 Gbps = 125 MB/s
 *     Latency  : 100 ms = 0.1 sec
 *
 *     BDP = 125 MB/s x 0.1 s = 12.5 MB
 *
 *     Around 12.5 MB data network pipe mein simultaneously in-flight ho sakta hai.
 */

/**
 * 23. TCP WINDOW vs BDP
 *     Agar TCP window BDP se smaller hai, to available network bandwidth
 *     fully utilize nahi ho payegi.
 *
 *     Example: BDP = 12.5 MB, TCP window = 5 MB
 *     Network capacity available hai BUT TCP window us capacity ko
 *     fully utilize nahi kar pa rahi.
 */

/**
 * 24. LATENCY vs THROUGHPUT vs BANDWIDTH
 *
 *     +------------+-----------------------+--------------+
 *     | Metric     | Meaning               | Unit         |
 *     +------------+-----------------------+--------------+
 *     | Latency    | Time per request      | ms           |
 *     | Throughput | Work per unit time    | RPS/TPS/MB/s |
 *     | Bandwidth  | Maximum data capacity | Mbps/Gbps    |
 *     +------------+-----------------------+--------------+
 *
 *     LATENCY   : "Kitna time?"
 *     THROUGHPUT: "Kitna kaam per second?"
 *     BANDWIDTH : "Maximum kitna data transfer?"
 */

/**
 * 25. WHAT TO OPTIMIZE?
 *
 *     LATENCY HIGH:
 *       CDN, caching, geographic placement, fewer round trips,
 *       connection pooling, DB optimization, HTTP/2 / HTTP/3
 *
 *     THROUGHPUT LOW:
 *       Horizontal scaling, concurrency, async processing, batching,
 *       caching, connection pooling, load balancing
 *
 *     BANDWIDTH LOW:
 *       Better network link, compression, parallel connections
 */

/**
 * 26. PERFORMANCE DEBUGGING MINDSET
 *     DON'T SAY: "System slow hai, server bada kar do."
 *     FIRST FIND: actual bottleneck kya hai?
 *
 *     Possible bottlenecks: network, CPU, memory, database, disk,
 *     connection pool, queue, lock/contention.
 *
 *     Then optimize that component.
 *     Example: DB is the bottleneck -> increasing network bandwidth
 *     may not improve throughput.
 *
 *     FIND BOTTLENECK FIRST.
 */

/**
 * 27. INTERVIEW QUESTIONS
 *
 * Q1. What is latency?
 * A : Time taken by a request to travel from source to destination and
 *     receive a response. Networking context mein ise often Round Trip
 *     Time (RTT) kehte hain.
 *
 * Q2. What is throughput?
 * A : Amount of work completed per unit of time. Web systems mein commonly
 *     RPS / TPS use hota hai.
 *
 * Q3. What is bandwidth?
 * A : Maximum rate at which data can theoretically be transferred.
 *     Common units: Mbps, Gbps.
 *
 * Q4. Difference between bandwidth and throughput?
 * A : Bandwidth = maximum possible capacity. Throughput = actual achieved rate.
 *     Example: Bandwidth = 1 Gbps, Throughput = 600 Mbps.
 *     Throughput usually lower hota hai because of overhead, congestion
 *     and processing limits.
 *
 * Q5. Can a system have high bandwidth but low throughput?
 * A : Yes. Example: 1 Gbps network, but congestion, packet loss or a
 *     processing bottleneck makes actual throughput low.
 *
 * Q6. Can a system have high bandwidth but high latency?
 * A : Yes. Highway analogy: many lanes + long distance = high capacity,
 *     but individual journey slow.
 *
 * Q7. Why are p95 and p99 important?
 * A : Average latency slow requests ko hide kar sakti hai; p95/p99 tail
 *     latency dikhate hain. Example: average = 10 ms, p99 = 500 ms means a
 *     small percentage of users face a very slow experience.
 *
 * Q8. What are the components of latency?
 * A : 1. Propagation delay  2. Transmission delay
 *     3. Processing delay   4. Queuing delay
 *
 * Q9. How would you reduce latency?
 * A : Depends on bottleneck. Common: CDN, caching, connection pooling,
 *     DB optimization, geographic distribution, fewer network round trips,
 *     HTTP/2 / HTTP/3.
 *
 * Q10. How would you improve throughput?
 * A : First identify bottleneck. Then consider: horizontal scaling,
 *     vertical scaling, async processing, batching, caching,
 *     connection pooling, load balancing.
 *
 * Q11. What determines maximum throughput?
 * A : The bottleneck / slowest component in the processing path.
 *     Example: API = 10K RPS, DB = 2K RPS -> system throughput ~ 2K RPS.
 *
 * Q12. What is BDP?
 * A : Bandwidth-Delay Product. BDP = Bandwidth x Latency.
 *     Approximately how much data can be in flight in the network.
 *
 * Q13. What happens if TCP window is smaller than BDP?
 * A : Available bandwidth fully utilize nahi hoti. TCP window becomes
 *     the limiting factor.
 *
 * Q14. Relationship between latency and throughput?
 * A : Different things, but mathematically related under concurrency.
 *       Single worker     : Throughput = 1 / Latency
 *       Concurrent workers: Throughput = Concurrent Workers / Latency
 *       Little's Law      : Concurrency = Throughput x Latency
 *
 * Q15. Can increasing throughput increase latency?
 * A : Yes. Batching, queueing and pipelining can increase throughput but
 *     may add waiting time to individual requests.
 */

/**
 * 28. BEGINNER QUICK REVISION
 *
 *   LATENCY    = One request ko complete hone mein time.
 *                Unit: milliseconds (ms)
 *   THROUGHPUT = Ek unit time mein actual completed work.
 *                Unit: RPS, TPS, MB/s
 *   BANDWIDTH  = Maximum data transfer capacity.
 *                Unit: Mbps, Gbps
 */

/**
 * 29. EASY MEMORY TRICK — WATER PIPE
 *
 *      SOURCE
 *        |==============================|
 *        |==============================|
 *        |==============================|
 *      DESTINATION
 *
 *   BANDWIDTH : pipe kitni wide hai?
 *   THROUGHPUT: actually kitna water per second flow ho raha hai?
 *   LATENCY   : water ko source se destination pahunchne mein kitna time lagta hai?
 */

/**
 * 30. GOLDEN RULE
 *
 *   LATENCY    -> "How long?"
 *   THROUGHPUT -> "How much work per second?"
 *   BANDWIDTH  -> "How much maximum capacity?"
 */

/**
 * FINAL SYSTEM DESIGN MINDSET
 *   User says: "Application slow hai." Don't immediately scale everything.
 *
 *   Ask:
 *     Is latency high?              -> Find delay
 *     Is throughput low?            -> Find bottleneck
 *     Is bandwidth insufficient?    -> Increase transfer capacity
 *
 *   FIRST : Identify the metric.
 *   THEN  : Identify the bottleneck.
 *   THEN  : Optimize the correct component.
 */

/**
 * ONE-LINE INTERVIEW ANSWER
 *   "Latency measures how long one request takes, throughput measures how
 *    much work the system completes per unit time, and bandwidth represents
 *    the maximum data-transfer capacity. They are related, but optimizing
 *    one does not automatically optimize the others."
 */
```
