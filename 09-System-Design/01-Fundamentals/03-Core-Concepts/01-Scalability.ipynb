```python
/**
 * ============================================================
 * SCALABILITY — SYSTEM DESIGN
 * ============================================================
 *
 * Core idea:
 *
 *     Measure Load
 *          ↓
 *     Find Bottleneck
 *          ↓
 *     Scale / Optimize Bottleneck
 *          ↓
 *     Measure Again
 *          ↓
 *     Repeat
 *
 * Scalability means the system can handle increasing load
 * by adding resources while continuing to meet its
 * performance requirements.
 *
 *
 * ============================================================
 * 1. THE SCALABILITY PROBLEM
 * ============================================================
 *
 * Initially:
 *
 *     Users
 *       |
 *       v
 *     Application
 *       |
 *       v
 *     Database
 *
 * Everything may run on one server.
 *
 * As users increase:
 *
 *     More requests
 *     More data
 *     More processing
 *
 * Eventually:
 *
 *     CPU -> High
 *     Memory -> High
 *     Requests -> Waiting
 *     Latency -> High
 *     Timeouts -> Increase
 *
 *
 * Example:
 *
 *     Current:
 *     1,000 requests/sec
 *     50 ms response time
 *
 *     Goal:
 *     5,000 requests/sec
 *     50 ms response time
 *
 * The solution is not simply "add more servers".
 *
 * First find what is actually limiting the system.
 *
 *
 * ============================================================
 * 2. WHAT DOES SCALABILITY MEAN?
 * ============================================================
 *
 * Scalability:
 *
 *     Ability of a system to handle more load
 *     by adding resources.
 *
 *
 * Two common approaches:
 *
 *     Vertical Scaling
 *          +
 *     Horizontal Scaling
 *
 *
 * Important:
 *
 * Adding resources does NOT automatically mean capacity
 * increases by the same amount.
 *
 * Example:
 *
 * Adding a second application server does not help if
 * every request is waiting on one slow database.
 *
 * Therefore:
 *
 *     Find the bottleneck first.
 *
 *
 * ============================================================
 * 3. MEASURING THE LOAD
 * ============================================================
 *
 * Before scaling, understand what is increasing.
 *
 *
 * Important metrics:
 *
 *
 * 1. REQUESTS PER SECOND (RPS)
 *
 * How frequently requests arrive.
 *
 * Example:
 *
 *     1,000 requests/sec
 *
 *
 * 2. CONCURRENT USERS
 *
 * Number of users active at the same time.
 *
 * Example:
 *
 *     20,000 users online
 *
 *
 * 3. DATA VOLUME
 *
 * Amount of data stored.
 *
 * Example:
 *
 *     2 TB of profiles, posts and images
 *
 *
 * IMPORTANT:
 *
 * Registered users != Concurrent users
 *
 * Example:
 *
 *     1 million registered users
 *
 * does NOT mean:
 *
 *     1 million users sending requests simultaneously.
 *
 *
 * ============================================================
 * 4. REQUEST COMPLEXITY MATTERS
 * ============================================================
 *
 * Request count alone is not enough.
 *
 *
 * Example:
 *
 * Read profile:
 *
 *     One indexed lookup
 *     Small response
 *     Low cost
 *
 *
 * Load feed:
 *
 *     Multiple lookups
 *     Sorting
 *     Data aggregation
 *     Medium cost
 *
 *
 * Upload video:
 *
 *     Large file
 *     Transcoding
 *     Heavy CPU
 *     Large storage
 *     High cost
 *
 *
 * Therefore:
 *
 *     1,000 profile reads/sec
 *
 * is NOT equivalent to:
 *
 *     1,000 video uploads/sec
 *
 *
 * Always understand:
 *
 *     How many requests?
 *           +
 *     What does each request do?
 *
 *
 * ============================================================
 * 5. FINDING THE BOTTLENECK
 * ============================================================
 *
 * Bottleneck:
 *
 * The component that limits how much work the entire
 * system can handle.
 *
 *
 * Common things to inspect:
 *
 *     CPU
 *     Memory
 *     Disk
 *     Database Queries
 *     Network
 *
 *
 * ------------------------------------------------------------
 * CPU BOTTLENECK
 * ------------------------------------------------------------
 *
 * Symptom:
 *
 *     CPU -> 100%
 *
 * Likely cause:
 *
 *     Application is doing more computation than
 *     available CPU cores can handle.
 *
 *
 * ------------------------------------------------------------
 * MEMORY BOTTLENECK
 * ------------------------------------------------------------
 *
 * Symptoms:
 *
 *     Swapping
 *     Out-of-memory errors
 *
 * Likely cause:
 *
 *     Working data no longer fits in RAM.
 *
 *
 * ------------------------------------------------------------
 * DISK BOTTLENECK
 * ------------------------------------------------------------
 *
 * Symptom:
 *
 *     High I/O wait
 *
 * Likely cause:
 *
 *     Database/application is waiting on slow storage.
 *
 *
 * ------------------------------------------------------------
 * DATABASE BOTTLENECK
 * ------------------------------------------------------------
 *
 * Symptom:
 *
 *     A few queries take most of the time.
 *
 * Likely causes:
 *
 *     Missing index
 *     Inefficient query
 *
 *
 * IMPORTANT:
 *
 * Adding application servers will NOT fix a slow
 * database query.
 *
 *
 * ============================================================
 * 6. BOTTLENECK → SCALING LOOP
 * ============================================================
 *
 * Scalability is an iterative process.
 *
 *
 *     Measure Load
 *          ↓
 *     Find Bottleneck
 *          ↓
 *     Make One Change
 *          ↓
 *     Measure Again
 *          ↓
 *     New Bottleneck?
 *          ↓
 *         YES
 *          ↓
 *     Repeat
 *
 *
 * Example:
 *
 * Application CPU becomes the bottleneck.
 *
 *     ↓
 *
 * Add application servers.
 *
 *     ↓
 *
 * Database now becomes the bottleneck.
 *
 *     ↓
 *
 * Scale / optimize database.
 *
 *     ↓
 *
 * Cache becomes necessary.
 *
 *     ↓
 *
 * Continue measuring.
 *
 *
 * Key idea:
 *
 * Removing one bottleneck usually exposes another bottleneck.
 *
 *
 * ============================================================
 * 7. VERTICAL SCALING — SCALE UP
 * ============================================================
 *
 * Vertical scaling means making one machine more powerful.
 *
 *
 * Example:
 *
 * Current:
 *
 *     4 CPU cores
 *     16 GB RAM
 *
 *          ↓ Scale Up
 *
 * New:
 *
 *     8 CPU cores
 *     32 GB RAM
 *
 *
 * Can also increase:
 *
 *     CPU
 *     RAM
 *     Disk performance
 *     Network capacity
 *
 *
 * Choose the upgrade based on the actual bottleneck.
 *
 *
 * CPU bottleneck:
 *
 *     Add CPU
 *
 * Memory bottleneck:
 *
 *     Add RAM
 *
 * Disk bottleneck:
 *
 *     Faster storage
 *
 * Network bottleneck:
 *
 *     Faster network interface
 *
 *
 * IMPORTANT:
 *
 * More CPU does not help if the application is waiting
 * on a slow disk.
 *
 *
 * ============================================================
 * 8. VERTICAL SCALING — PROS & CONS
 * ============================================================
 *
 * PROS:
 *
 *     + Simple
 *     + Usually few/no code changes
 *     + No additional network communication
 *     + Existing architecture can remain mostly unchanged
 *
 *
 * CONS:
 *
 *     - Hardware ceiling
 *     - Expensive at higher sizes
 *     - Upgrade may require downtime/restart
 *     - Single point of failure
 *
 *
 * Example:
 *
 *     One Server
 *
 *     If it fails:
 *
 *         Application DOWN
 *
 *
 * ============================================================
 * 9. SEPARATING THE DATABASE
 * ============================================================
 *
 * Initially:
 *
 *     Application + Database
 *              |
 *           One Server
 *
 *
 * Problem:
 *
 * Application and database compete for:
 *
 *     CPU
 *     Memory
 *     Disk
 *
 *
 * Solution:
 *
 *     Application Server
 *          |
 *          v
 *     Database Server
 *
 *
 * Now resources can be sized independently.
 *
 *
 * Application Server:
 *
 *     More CPU
 *
 *
 * Database Server:
 *
 *     More memory
 *     Faster disk
 *
 *
 * IMPORTANT:
 *
 * This is still vertical scaling.
 *
 * We are scaling each machine independently.
 *
 *
 * ============================================================
 * 10. HORIZONTAL SCALING — SCALE OUT
 * ============================================================
 *
 * Horizontal scaling means adding more machines.
 *
 *
 * Instead of:
 *
 *     One Application Server
 *
 *
 * We use:
 *
 *              Load Balancer
 *              /    |    \
 *             ↓     ↓     ↓
 *           App1   App2   App3
 *
 *
 * The load balancer distributes requests between servers.
 *
 *
 * When traffic increases:
 *
 *     Add more servers.
 *
 *
 * When traffic decreases:
 *
 *     Remove servers.
 *
 *
 * This can be automated using auto-scaling.
 *
 *
 * ============================================================
 * 11. LOAD BALANCER
 * ============================================================
 *
 * Load balancer distributes incoming traffic across
 * multiple application servers.
 *
 *
 *     Users
 *       |
 *       v
 *   Load Balancer
 *      / | \
 *     v  v  v
 *   App1 App2 App3
 *
 *
 * Example:
 *
 *     600 requests/sec
 *
 *     App1 -> 200
 *     App2 -> 200
 *     App3 -> 200
 *
 *
 * If one server becomes unhealthy:
 *
 *     App1 -> DOWN
 *
 * Load balancer stops sending traffic to App1.
 *
 * Remaining servers continue serving traffic if they
 * have enough capacity.
 *
 *
 * ============================================================
 * 12. AUTO-SCALING
 * ============================================================
 *
 * Auto-scaling automatically adjusts the number of
 * application servers based on demand.
 *
 *
 * Traffic increases:
 *
 *     2 servers
 *         ↓
 *     3 servers
 *         ↓
 *     4 servers
 *
 *
 * Traffic decreases:
 *
 *     4 servers
 *         ↓
 *     3 servers
 *         ↓
 *     2 servers
 *
 *
 * Goal:
 *
 *     Match infrastructure capacity with workload.
 *
 *
 * ============================================================
 * 13. THE SESSION PROBLEM
 * ============================================================
 *
 * Horizontal scaling creates a problem if session data
 * is stored in local server memory.
 *
 *
 * Example:
 *
 * User logs in:
 *
 *     User
 *       |
 *       v
 *   Load Balancer
 *       |
 *       v
 *    Server 1
 *
 * Server 1 stores:
 *
 *     session = logged_in
 *
 *
 * Next request:
 *
 *     User
 *       |
 *       v
 *   Load Balancer
 *       |
 *       v
 *    Server 2
 *
 * Server 2 does not have Server 1's session.
 *
 *
 * Result:
 *
 *     User may appear logged out.
 *
 *
 * ============================================================
 * 14. STICKY SESSIONS
 * ============================================================
 *
 * Sticky session:
 *
 * Load balancer keeps routing the same user
 * to the same application server.
 *
 *
 *     User
 *       |
 *       v
 *   Load Balancer
 *       |
 *       v
 *    Server 1
 *
 * Next request:
 *
 *     User
 *       |
 *       v
 *   Load Balancer
 *       |
 *       v
 *    Server 1
 *
 *
 * Advantage:
 *
 *     Local session continues to work.
 *
 *
 * Problems:
 *
 *     - Less flexible traffic distribution
 *     - Some servers may get more users
 *     - Server failure loses local sessions
 *
 *
 * Better approach:
 *
 *     Make application servers stateless.
 *
 *
 * ============================================================
 * 15. STATELESS APPLICATION SERVERS
 * ============================================================
 *
 * Move shared state outside application servers.
 *
 *
 * Example:
 *
 *     User
 *       |
 *       v
 *   Load Balancer
 *       |
 *     +---+---+
 *     |   |   |
 *     v   v   v
 *    App1 App2 App3
 *      \   |   /
 *       \  |  /
 *        Redis
 *        Sessions
 *
 *
 * Now any server can process any request.
 *
 *
 * Files should also be stored outside local servers:
 *
 *     App Servers
 *          |
 *          v
 *         S3
 *     Uploaded Files
 *
 *
 * IMPORTANT:
 *
 * Stateless does NOT mean the system has no state.
 *
 * It means:
 *
 *     Application servers do not own
 *     the state individually.
 *
 *
 * State lives in shared external storage.
 *
 *
 * Benefits:
 *
 *     + Easy to add servers
 *     + Easy to remove servers
 *     + Easy to replace failed servers
 *     + Any healthy server can handle requests
 *
 *
 * ============================================================
 * 16. VERTICAL vs HORIZONTAL SCALING
 * ============================================================
 *
 *
 * VERTICAL:
 *
 *     Make one machine more powerful.
 *
 *
 * HORIZONTAL:
 *
 *     Add more machines behind a load balancer.
 *
 *
 * ------------------------------------------------------------
 * VERTICAL
 * ------------------------------------------------------------
 *
 * Code changes:
 *     Usually none
 *
 * Limit:
 *     Largest available machine
 *
 * Failure:
 *     No automatic replacement
 *
 * Cost:
 *     Grows steeply at high-end
 *
 * Complexity:
 *     Low
 *
 *
 * ------------------------------------------------------------
 * HORIZONTAL
 * ------------------------------------------------------------
 *
 * Code changes:
 *     Servers should be stateless
 *
 * Limit:
 *     Can keep adding servers as shared components allow
 *
 * Failure:
 *     Load balancer routes around failed server
 *
 * Cost:
 *     Roughly linear per server
 *
 * Complexity:
 *     Higher
 *
 *     More machines
 *     More network communication
 *     Shared data management
 *
 *
 * ============================================================
 * 17. CACHING
 * ============================================================
 *
 * Problem:
 *
 * Multiple users repeatedly request the same data.
 *
 *
 * Example:
 *
 *     Popular User Profile
 *
 * Every request:
 *
 *     App -> Database
 *
 * This creates unnecessary database work.
 *
 *
 * Solution:
 *
 *     App -> Cache
 *              |
 *              +--> HIT -> Return data
 *              |
 *              +--> MISS
 *                    |
 *                    v
 *                 Database
 *                    |
 *                    v
 *                 Cache
 *
 *
 * This is called:
 *
 *     Cache-Aside Pattern
 *
 *
 * ============================================================
 * 18. CACHE-ASIDE PATTERN
 * ============================================================
 *
 * Step 1:
 *
 *     Application checks cache.
 *
 *
 * Step 2:
 *
 *     Cache HIT
 *
 *     Return cached data.
 *
 *
 * Step 3:
 *
 *     Cache MISS
 *
 *     Read from database.
 *
 *
 * Step 4:
 *
 *     Store database result in cache.
 *
 *
 * Step 5:
 *
 *     Return result.
 *
 *
 * Flow:
 *
 *     App
 *      |
 *      v
 *    Cache
 *      |
 *      +---- HIT ----> Return
 *      |
 *      +---- MISS
 *             |
 *             v
 *          Database
 *             |
 *             v
 *           Cache
 *
 *
 * Benefits:
 *
 *     + Faster repeated reads
 *     + Less database load
 *
 *
 * ============================================================
 * 19. CACHE INVALIDATION
 * ============================================================
 *
 * Problem:
 *
 * Database data changes.
 *
 * Cached value may now be outdated.
 *
 *
 * Common approaches:
 *
 *
 * 1. DELETE CACHE ENTRY
 *
 * When data changes:
 *
 *     Update DB
 *        +
 *     Delete cache
 *
 * Next request:
 *
 *     Cache MISS
 *         ↓
 *     Read DB
 *         ↓
 *     Refill cache
 *
 *
 * 2. WRITE TO DATABASE + CACHE
 *
 * Update both:
 *
 *     Database
 *        +
 *     Cache
 *
 * Advantage:
 *
 *     Cache stays warm.
 *
 * Trade-off:
 *
 *     Application must keep two places synchronized.
 *
 *
 * 3. TTL
 *
 * Cache entry expires after a configured time.
 *
 * Trade-off:
 *
 *     Users may see stale data until expiration.
 *
 *
 * ============================================================
 * 20. SCALING THE CACHE
 * ============================================================
 *
 * One cache server may eventually run out of:
 *
 *     Memory
 *     Processing capacity
 *
 *
 * Solution:
 *
 *     Multiple cache nodes.
 *
 *
 * Example:
 *
 *     App
 *      |
 *      +----> Cache Node 1
 *      |
 *      +----> Cache Node 2
 *      |
 *      +----> Cache Node 3
 *
 *
 * A key determines which cache node stores the value.
 *
 *
 * Example:
 *
 *     hash(key) % 3
 *
 *
 *     Result 0 -> Cache Node 1
 *     Result 1 -> Cache Node 2
 *     Result 2 -> Cache Node 3
 *
 *
 * Benefits:
 *
 *     + More total cache capacity
 *     + Multiple nodes can process requests simultaneously
 *
 *
 * ============================================================
 * 21. READ REPLICAS
 * ============================================================
 *
 * Problem:
 *
 * Even after caching, many requests still reach the database.
 *
 * Examples:
 *
 *     Fresh feed
 *     Rarely viewed profile
 *     Search
 *
 *
 * Solution:
 *
 *     Read Replicas
 *
 *
 * Architecture:
 *
 *                 Writes
 *                   |
 *                   v
 *                Primary
 *                /     \
 *               v       v
 *          Replica 1  Replica 2
 *               ^       ^
 *               |       |
 *             Reads   Reads
 *
 *
 * Primary:
 *
 *     Handles writes.
 *
 * Replicas:
 *
 *     Handle reads.
 *
 *
 * Benefits:
 *
 *     + Increased read capacity
 *     + Read traffic can be distributed
 *
 *
 * IMPORTANT:
 *
 * Replicas do NOT increase primary write capacity.
 *
 * Every write still goes to the primary.
 *
 *
 * ============================================================
 * 22. REPLICATION LAG
 * ============================================================
 *
 * Primary and replicas may not be perfectly synchronized
 * at every moment.
 *
 *
 * Example:
 *
 *     Write
 *       |
 *       v
 *     Primary
 *       |
 *       | replication
 *       v
 *     Replica
 *
 *
 * There can be a small delay before the replica receives
 * the latest change.
 *
 *
 * Example:
 *
 * User updates profile:
 *
 *     Write -> Primary
 *
 * Immediately reads:
 *
 *     Read -> Replica
 *
 * Replica may still contain old data.
 *
 *
 * Solution for reads requiring latest data:
 *
 *     Read from Primary.
 *
 *
 * ============================================================
 * 23. SHARDING
 * ============================================================
 *
 * Problem:
 *
 * Replicas do not solve everything.
 *
 * Every replica contains the complete dataset.
 *
 * Every write still goes to the primary.
 *
 *
 * If:
 *
 *     Data size becomes too large
 *
 * OR
 *
 *     Write volume becomes too high
 *
 * Then use:
 *
 *     Sharding
 *
 *
 * Sharding:
 *
 * Divide data across multiple databases.
 *
 *
 * Example:
 *
 *     App Servers
 *          |
 *          v
 *     Pick Shard
 *       from User ID
 *       /    |    \
 *      v     v     v
 *   Shard1 Shard2 Shard3
 *
 *
 * ============================================================
 * 24. SHARD KEY
 * ============================================================
 *
 * Shard key:
 *
 * Value used to determine which shard stores a record.
 *
 *
 * Example:
 *
 *     user_id
 *
 *
 *     User 1 -> Shard 1
 *     User 2 -> Shard 1
 *     ...
 *     User 1M -> Shard 1
 *
 *     User 1M+ -> Shard 2
 *
 *
 * Example from the system:
 *
 *     Shard 1
 *     Users 1 - 1M
 *
 *     Shard 2
 *     Users 1M - 2M
 *
 *     Shard 3
 *     Users 2M+
 *
 *
 * Different shards can process writes simultaneously.
 *
 *
 * Benefits:
 *
 *     + Distributes storage
 *     + Distributes write traffic
 *     + Multiple shards can process work in parallel
 *
 *
 * ============================================================
 * 25. SHARDING TRADE-OFF
 * ============================================================
 *
 * Single-user query:
 *
 *     Load user profile
 *
 * Can go to:
 *
 *     One shard
 *
 *
 * Cross-user query:
 *
 *     Search all posts
 *
 * May need:
 *
 *     Shard 1
 *     Shard 2
 *     Shard 3
 *
 * Then:
 *
 *     Merge results
 *
 *
 * Therefore shard-key selection is extremely important.
 *
 *
 * Good shard key:
 *
 *     Keeps common queries inside one shard.
 *
 *
 * Bad shard key:
 *
 *     Causes many queries to touch multiple shards.
 *
 *
 * ============================================================
 * 26. QUEUES & BACKGROUND WORK
 * ============================================================
 *
 * Not every task needs to finish before responding
 * to the user.
 *
 *
 * Example:
 *
 * User publishes post.
 *
 * Required:
 *
 *     Save post
 *
 * Additional work:
 *
 *     Notify followers
 *
 *
 * Bad approach:
 *
 *     User
 *       |
 *       v
 *     Save Post
 *       |
 *       v
 *     Notify 1,000,000 followers
 *       |
 *       v
 *     Response
 *
 * User waits for all notifications.
 *
 *
 * Better:
 *
 *     User
 *       |
 *       v
 *     App Server
 *       |
 *       +----> Database
 *       |
 *       +----> Queue
 *                  |
 *               +--+--+
 *               v     v
 *            Worker1 Worker2
 *               |
 *               v
 *          Notifications
 *
 *
 * Flow:
 *
 *     1. Save post
 *     2. Enqueue notification job
 *     3. Respond to user
 *     4. Workers process notifications
 *
 *
 * Benefits:
 *
 *     + Faster user response
 *     + Background work is separated
 *     + Traffic spikes can be absorbed by queue
 *
 *
 * If backlog grows:
 *
 *     Add more workers
 *
 * provided the work can safely be processed in parallel.
 *
 *
 * IMPORTANT:
 *
 * Queue does NOT reduce the total amount of work.
 *
 * It removes that work from the user's request path.
 *
 *
 * ============================================================
 * 27. VERIFYING A SCALABILITY CHANGE
 * ============================================================
 *
 * After every scaling change:
 *
 *     Test with realistic workload.
 *
 *
 * Compare:
 *
 *     Before
 *        vs
 *     After
 *
 *
 * Important measurements:
 *
 *
 * 1. REQUESTS COMPLETED PER SECOND
 *
 * Example:
 *
 *     Before -> 1,000 RPS
 *     After  -> 5,000 RPS
 *
 *
 * Tells us:
 *
 *     Did actual capacity increase?
 *
 *
 * 2. MEDIAN RESPONSE TIME
 *
 * Example:
 *
 *     Before -> 50 ms
 *     After  -> 45 ms
 *
 * Represents a typical request.
 *
 *
 * 3. P99 RESPONSE TIME
 *
 * Example:
 *
 *     Before -> 500 ms
 *     After  -> 120 ms
 *
 * Represents the slowest 1% of requests.
 *
 *
 * 4. FAILED REQUESTS
 *
 * Example:
 *
 *     Before -> 2%
 *     After  -> 0.1%
 *
 *
 * Tells us whether the system is dropping work.
 *
 *
 * ============================================================
 * 28. WHY P95 / P99 MATTER
 * ============================================================
 *
 * Average latency can hide slow requests.
 *
 *
 * Example:
 *
 *     Average = 60 ms
 *
 * Sounds good.
 *
 * But some requests may take:
 *
 *     2 seconds
 *
 *
 * Percentiles show the experience of slower requests.
 *
 *
 * P95:
 *
 *     95% of requests are at or below this latency.
 *
 *
 * P99:
 *
 *     99% of requests are at or below this latency.
 *
 *
 * For system design:
 *
 *     Don't look only at average latency.
 *
 * Check:
 *
 *     p50
 *     p95
 *     p99
 *
 *
 * ============================================================
 * 29. COST + CORRECTNESS
 * ============================================================
 *
 * A scaling solution is not successful just because
 * it becomes faster.
 *
 *
 * Also check:
 *
 *     Cost
 *     Data correctness
 *     Failed requests
 *
 *
 * Example:
 *
 * A system:
 *
 *     Responds faster
 *
 * BUT
 *
 *     Loses writes
 *
 * Then the system has NOT actually improved.
 *
 *
 * Therefore:
 *
 *     Performance
 *         +
 *     Reliability
 *         +
 *     Correctness
 *         +
 *     Cost
 *
 * must all be considered.
 *
 *
 * ============================================================
 * 30. THERE IS NO FIXED SCALING THRESHOLD
 * ============================================================
 *
 * There is no rule like:
 *
 *     "At 1 million users, use sharding."
 *
 *
 * Different applications have different workloads.
 *
 *
 * Application A:
 *
 *     Large data per user
 *     -> May need sharding early.
 *
 *
 * Application B:
 *
 *     Small data per user
 *     Low write frequency
 *     -> May use one database for years.
 *
 *
 * Therefore:
 *
 *     Scale based on workload and bottleneck,
 *     NOT based on a fixed user count.
 *
 *
 * ============================================================
 * STEP-BY-STEP SCALABILITY JOURNEY
 * ============================================================
 *
 *
 * START:
 *
 *     Users
 *       |
 *       v
 *     Application + DB
 *       |
 *       v
 *     One Server
 *
 *
 * STEP 1:
 *     CPU / Memory becomes bottleneck
 *
 *     -> Vertical Scaling
 *
 *
 * STEP 2:
 *     Application + DB compete for resources
 *
 *     -> Separate Database
 *
 *
 * STEP 3:
 *     One application server reaches its limit
 *
 *     -> Horizontal Scaling
 *     -> Load Balancer
 *     -> Multiple App Servers
 *
 *
 * STEP 4:
 *     Local sessions prevent free traffic distribution
 *
 *     -> Stateless App Servers
 *     -> Redis for Sessions
 *     -> S3 for Files
 *
 *
 * STEP 5:
 *     Repeated reads overload database
 *
 *     -> Cache
 *     -> Cache-Aside
 *
 *
 * STEP 6:
 *     Cache itself becomes too large
 *
 *     -> Multiple Cache Nodes
 *
 *
 * STEP 7:
 *     Database read volume becomes too high
 *
 *     -> Read Replicas
 *
 *
 * STEP 8:
 *     Database write volume / data size becomes too large
 *
 *     -> Sharding
 *
 *
 * STEP 9:
 *     Slow background work increases request latency
 *
 *     -> Queue
 *     -> Background Workers
 *
 *
 * ============================================================
 * FINAL SCALABLE ARCHITECTURE
 * ============================================================
 *
 *
 *                         USERS
 *                           |
 *                           v
 *                    LOAD BALANCER
 *                           |
 *             +-------------+-------------+
 *             |             |             |
 *             v             v             v
 *          APP 1          APP 2         APP 3
 *             |             |             |
 *             +-------------+-------------+
 *                           |
 *              +------------+------------+
 *              |            |            |
 *              v            v            v
 *            CACHE        REDIS          S3
 *                         Sessions       Files
 *              |
 *              v
 *       DATABASE LAYER
 *              |
 *       +------+------+
 *       |             |
 *       v             v
 *    Primary       Replicas
 *       |
 *       v
 *    Shards
 *
 *
 * Background Work:
 *
 *     App
 *      |
 *      v
 *    Queue
 *      |
 *    +---+---+
 *    |       |
 *    v       v
 * Worker 1 Worker 2
 *
 *
 * ============================================================
 * QUICK REVISION
 * ============================================================
 *
 * Scalability
 *     -> Handle more load by adding resources.
 *
 * Bottleneck
 *     -> Component limiting system capacity.
 *
 * Vertical Scaling
 *     -> Bigger machine.
 *
 * Horizontal Scaling
 *     -> More machines.
 *
 * Load Balancer
 *     -> Distributes traffic.
 *
 * Auto Scaling
 *     -> Automatically adds/removes servers based on demand.
 *
 * Stateless Servers
 *     -> Shared state lives outside individual app servers.
 *
 * Redis
 *     -> Shared session/state store.
 *
 * S3
 *     -> Shared storage for uploaded files.
 *
 * Cache
 *     -> Avoid repeated expensive reads.
 *
 * Cache-Aside
 *     -> Cache first -> DB on miss -> Store in cache.
 *
 * Read Replica
 *     -> Scale database reads.
 *
 * Replication Lag
 *     -> Replica may temporarily contain stale data.
 *
 * Sharding
 *     -> Split data across databases.
 *
 * Shard Key
 *     -> Determines where data belongs.
 *
 * Queue
 *     -> Move slow/background work out of request path.
 *
 * Worker
 *     -> Processes queued background jobs.
 *
 * P95/P99
 *     -> Measure slow-request experience.
 *
 *
 * ============================================================
 * MOST IMPORTANT TRADE-OFFS
 * ============================================================
 *
 *
 * VERTICAL SCALING
 *
 * Simpler
 *     vs
 * Hardware ceiling + single point of failure
 *
 *
 * HORIZONTAL SCALING
 *
 * More capacity + availability
 *     vs
 * More infrastructure + network complexity
 *
 *
 * STICKY SESSION
 *
 * Simple local sessions
 *     vs
 * Uneven traffic + failure loses sessions
 *
 *
 * STATELESS SERVERS
 *
 * Easy scaling
 *     vs
 * Requires external/shared state storage
 *
 *
 * CACHE
 *
 * Faster reads + less DB load
 *     vs
 * Cache invalidation + stale data
 *
 *
 * READ REPLICAS
 *
 * Higher read capacity
 *     vs
 * Replication lag
 *
 *
 * SHARDING
 *
 * Higher write/storage capacity
 *     vs
 * Cross-shard queries + operational complexity
 *
 *
 * QUEUES
 *
 * Faster user response + traffic buffering
 *     vs
 * Asynchronous processing + eventual completion
 *
 *
 * ============================================================
 * INTERVIEW CHEAT SHEET
 * ============================================================
 *
 *
 * If CPU is high:
 *
 *     -> More CPU
 *     -> Vertical scaling
 *     -> Or horizontal scaling if application can scale out
 *
 *
 * If memory is full:
 *
 *     -> More RAM
 *     -> Check memory usage / caching
 *
 *
 * If database query is slow:
 *
 *     -> Check query
 *     -> Add index / optimize query
 *     -> Don't blindly add app servers
 *
 *
 * If application server reaches capacity:
 *
 *     -> Load Balancer
 *     -> Multiple stateless servers
 *
 *
 * If sessions are stored locally:
 *
 *     -> Redis / shared session store
 *     -> Make app servers stateless
 *
 *
 * If same data is repeatedly requested:
 *
 *     -> Cache
 *     -> Cache-aside
 *
 *
 * If cache becomes too large:
 *
 *     -> Multiple cache nodes
 *     -> Distribute keys
 *
 *
 * If database reads are too high:
 *
 *     -> Read replicas
 *
 *
 * If database writes/data size are too high:
 *
 *     -> Sharding
 *
 *
 * If request contains slow background work:
 *
 *     -> Queue
 *     -> Background workers
 *
 *
 * After every scaling change:
 *
 *     -> Measure RPS
 *     -> Measure p50/p95/p99
 *     -> Check failed requests
 *     -> Check cost
 *     -> Check correctness
 *
 *
 * ============================================================
 * CORE SYSTEM DESIGN MINDSET
 * ============================================================
 *
 * Never start with:
 *
 *     "Let's add Kubernetes."
 *
 *     "Let's add Kafka."
 *
 *     "Let's add Redis."
 *
 * Instead:
 *
 *     1. Understand the workload
 *
 *     2. Measure the load
 *
 *     3. Find the bottleneck
 *
 *     4. Choose the smallest change that addresses it
 *
 *     5. Measure again
 *
 *     6. Identify the next bottleneck
 *
 *     7. Repeat
 *
 *
 * The core scalability principle:
 *
 *     FIND WHAT BREAKS NEXT
 *              ↓
 *     SCALE THAT COMPONENT
 *              ↓
 *     MEASURE THE RESULT
 *              ↓
 *     REPEAT
 *
 * ============================================================
 */
```

```python
/**
 * SCALABILITY — SYSTEM DESIGN
 *
 * Core idea:
 * > Measure Load 
 * > Find Bottleneck 
 * > Scale/Optimize Bottleneck
 * > Measure Again 
 * > Repeat
 *
 * Scalability = the system can handle increasing load by adding resources
 * while still meeting its performance requirements.
 */

/**
 * 1. THE SCALABILITY PROBLEM
 *    Initially: Users -> Application -> Database (all on one server).
 *
 *    As users increase: more requests, more data, more processing.
 *    Eventually:
 *      CPU -> High | Memory -> High | Requests -> Waiting
 *      Latency -> High | Timeouts -> Increase
 *
 *    Example:
 *      Current : 1,000 requests/sec, 50 ms response time
 *      Goal    : 5,000 requests/sec, 50 ms response time
 *
 *    The solution is not simply "add more servers".
 *    First find what is actually limiting the system.
 */

/**
 * 2. WHAT DOES SCALABILITY MEAN?
 *    - Ability of a system to handle more load by adding resources.
 *    - Two common approaches: 
 *      a. Vertical Scaling 
 *      b. Horizontal Scaling
 *
 *    IMPORTANT:
 *    - Adding resources does NOT automatically increase capacity by the same amount.
 *    - Example: 
 *      - A second app server doesn't help if every request waits on one slow database.
 *      - Therefore: find the bottleneck first.
 */

/**
 * 3. MEASURING THE LOAD
 *    - Before scaling, understand what is increasing.
 *      1. REQUESTS PER SECOND (RPS)  - how often requests arrive.     e.g. 1,000 req/sec
 *      2. CONCURRENT USERS           - users active at the same time. e.g. 20,000 online
 *      3. DATA VOLUME                - amount of data stored.         e.g. 2 TB of profiles, posts, images
 *
 *    - IMPORTANT:
 *      - Registered users != Concurrent users
 *      - 1 million registered users does NOT mean 1 million users sending requests
 *        simultaneously.
 */

/**
 * 4. REQUEST COMPLEXITY MATTERS
 *    Request count alone is not enough.
 *
 *    Read profile : one indexed lookup, small response           -> low cost
 *    Load feed    : multiple lookups, sorting, aggregation       -> medium cost
 *    Upload video : large file, transcoding, heavy CPU, storage  -> high cost
 *
 *    1,000 profile reads/sec  !=  1,000 video uploads/sec
 *
 *    Always understand: How many requests? + What does each request do?
 */

/**
 * 5. FINDING THE BOTTLENECK
 *    Bottleneck = the component that limits how much work the whole system can handle.
 *    Inspect: CPU, Memory, Disk, Database Queries, Network
 *
 *    CPU BOTTLENECK
 *      Symptom : CPU -> 100%
 *      Cause   : App does more computation than available cores can handle.
 *
 *    MEMORY BOTTLENECK
 *      Symptoms: Swapping, out-of-memory errors
 *      Cause   : Working data no longer fits in RAM.
 *
 *    DISK BOTTLENECK
 *      Symptom : High I/O wait
 *      Cause   : DB/app is waiting on slow storage.
 *
 *    DATABASE BOTTLENECK
 *      Symptom : A few queries take most of the time.
 *      Causes  : Missing index, inefficient query.
 *
 *    IMPORTANT: Adding app servers will NOT fix a slow database query.
 */

/**
 * 6. BOTTLENECK -> SCALING LOOP
 *    Scalability is iterative:
 *      Measure Load -> Find Bottleneck -> Make One Change
 *      -> Measure Again -> New Bottleneck? -> YES -> Repeat
 *
 *    Example:
 *      App CPU is the bottleneck      -> Add app servers
 *      Database now the bottleneck    -> Scale / optimize database
 *      Cache becomes necessary        -> Continue measuring
 *
 *    Key idea: removing one bottleneck usually exposes another.
 */

/**
 * 7. VERTICAL SCALING — SCALE UP
 *    Make one machine more powerful.
 *
 *    Example:
 *      Current : 4 CPU cores, 16 GB RAM
 *      Scale Up: 8 CPU cores, 32 GB RAM
 *
 *    Can increase: CPU, RAM, disk performance, network capacity.
 *    Choose the upgrade based on the actual bottleneck:
 *      CPU bottleneck     -> Add CPU
 *      Memory bottleneck  -> Add RAM
 *      Disk bottleneck    -> Faster storage
 *      Network bottleneck -> Faster network interface
 *
 *    IMPORTANT: More CPU doesn't help if the app is waiting on a slow disk.
 */

/**
 * 8. VERTICAL SCALING — PROS & CONS
 *    PROS:
 *      + Simple
 *      + Usually few/no code changes
 *      + No additional network communication
 *      + Existing architecture stays mostly unchanged
 *
 *    CONS:
 *      - Hardware ceiling
 *      - Expensive at higher sizes
 *      - Upgrade may need downtime/restart
 *      - Single point of failure (one server fails -> application DOWN)
 */

/**
 * 9. SEPARATING THE DATABASE
 *    Initially: Application + Database on one server.
 *
 *    Problem: both compete for CPU, memory, disk.
 *
 *    Solution: Application Server -> Database Server
 *    Now resources can be sized independently:
 *      Application Server : more CPU
 *      Database Server    : more memory, faster disk
 *
 *    IMPORTANT: This is still vertical scaling
 *    (each machine is scaled independently).
 */

/**
 * 10. HORIZONTAL SCALING — SCALE OUT
 *     Add more machines instead of one bigger one.
 *
 *     Instead of one app server:
 *
 *               Load Balancer
 *               /    |    \
 *              v     v     v
 *            App1   App2   App3
 *
 *     Load balancer distributes requests between servers.
 *     Traffic increases -> add servers. Traffic decreases -> remove servers.
 *     This can be automated using auto-scaling.
 */

/**
 * 11. LOAD BALANCER
 *     Distributes incoming traffic across multiple app servers.
 *
 *       Users --> Load Balancer --> App1 / App2 / App3
 *
 *     Example: 600 req/sec -> App1: 200, App2: 200, App3: 200
 *
 *     If a server becomes unhealthy (App1 DOWN):
 *       - Load balancer stops sending traffic to App1.
 *       - Remaining servers keep serving if they have enough capacity.
 */

/**
 * 12. AUTO-SCALING
 *     Automatically adjusts the number of app servers based on demand.
 *
 *       Traffic up   : 2 servers -> 3 servers -> 4 servers
 *       Traffic down : 4 servers -> 3 servers -> 2 servers
 *
 *     Goal: match infrastructure capacity with workload.
 */

/**
 * 13. THE SESSION PROBLEM
 *     Horizontal scaling breaks if session data lives in local server memory.
 *
 *     User logs in:
 *       User --> Load Balancer --> Server 1
 *       Server 1 stores: session = logged_in
 *
 *     Next request:
 *       User --> Load Balancer --> Server 2
 *       Server 2 doesn't have Server 1's session.
 *
 *     Result: user may appear logged out.
 */

/**
 * 14. STICKY SESSIONS
 *     Load balancer keeps routing the same user to the same app server.
 *
 *       User --> Load Balancer --> Server 1   (every request)
 *
 *     Advantage:
 *       Local session continues to work.
 *
 *     Problems:
 *       - Less flexible traffic distribution
 *       - Some servers may get more users
 *       - Server failure loses local sessions
 *
 *     Better approach: make application servers stateless.
 */

/**
 * 15. STATELESS APPLICATION SERVERS
 *     Move shared state outside application servers.
 *
 *       User --> Load Balancer --> App1 / App2 / App3
 *                                      |
 *                                    Redis (Sessions)
 *
 *     Now any server can process any request.
 *
 *     Files also go outside local servers:
 *       App Servers --> S3 (uploaded files)
 *
 *     IMPORTANT:
 *       Stateless does NOT mean the system has no state.
 *       It means app servers don't individually own the state;
 *       state lives in shared external storage.
 *
 *     Benefits:
 *       + Easy to add servers
 *       + Easy to remove servers
 *       + Easy to replace failed servers
 *       + Any healthy server can handle requests
 */

/**
 * 16. VERTICAL vs HORIZONTAL SCALING
 *
 *     VERTICAL: make one machine more powerful.
 *       Code changes : usually none
 *       Limit        : largest available machine
 *       Failure      : no automatic replacement
 *       Cost         : grows steeply at high end
 *       Complexity   : low
 *
 *     HORIZONTAL: add more machines behind a load balancer.
 *       Code changes : servers should be stateless
 *       Limit        : keep adding servers, as shared components allow
 *       Failure      : load balancer routes around failed server
 *       Cost         : roughly linear per server
 *       Complexity   : higher (more machines, more network communication,
 *                      shared data management)
 */

/**
 * 17. CACHING
 *     Problem: many users repeatedly request the same data
 *     (e.g. a popular user profile). Every request goes App -> Database,
 *     creating unnecessary database work.
 *
 *     Solution:
 *       App --> Cache
 *                 |
 *                 +-- HIT  --> Return data
 *                 +-- MISS --> Database --> store in Cache
 *
 *     This is called the Cache-Aside Pattern.
 */

/**
 * 18. CACHE-ASIDE PATTERN
 *     1. Application checks cache.
 *     2. Cache HIT  -> return cached data.
 *     3. Cache MISS -> read from database.
 *     4. Store database result in cache.
 *     5. Return result.
 *
 *     Benefits:
 *       + Faster repeated reads
 *       + Less database load
 */

/**
 * 19. CACHE INVALIDATION
 *     Problem: DB data changes, so the cached value may be outdated.
 *
 *     1. DELETE CACHE ENTRY
 *        Update DB + delete cache.
 *        Next request: cache MISS -> read DB -> refill cache.
 *
 *     2. WRITE TO DATABASE + CACHE
 *        Update both.
 *        Advantage : cache stays warm.
 *        Trade-off : app must keep two places synchronized.
 *
 *     3. TTL
 *        Cache entry expires after a configured time.
 *        Trade-off : users may see stale data until expiration.
 */

/**
 * 20. SCALING THE CACHE
 *     One cache server may run out of memory or processing capacity.
 *     Solution: multiple cache nodes.
 *
 *       App --> Cache Node 1 / Cache Node 2 / Cache Node 3
 *
 *     A key decides which node stores the value:
 *       hash(key) % 3
 *         0 -> Cache Node 1
 *         1 -> Cache Node 2
 *         2 -> Cache Node 3
 *
 *     Benefits:
 *       + More total cache capacity
 *       + Multiple nodes process requests simultaneously
 */

/**
 * 21. READ REPLICAS
 *     Problem: even after caching, many requests still reach the database
 *     (fresh feed, rarely viewed profile, search).
 *
 *     Solution: Read Replicas
 *
 *                 Writes
 *                   |
 *                   v
 *                Primary
 *                /     \
 *               v       v
 *          Replica 1  Replica 2
 *               ^       ^
 *             Reads   Reads
 *
 *     Primary  : handles writes.
 *     Replicas : handle reads.
 *
 *     Benefits:
 *       + Increased read capacity
 *       + Read traffic can be distributed
 *
 *     IMPORTANT: Replicas do NOT increase primary write capacity.
 *     Every write still goes to the primary.
 */

/**
 * 22. REPLICATION LAG
 *     Primary and replicas may not be perfectly in sync at every moment.
 *
 *       Write --> Primary --(replication)--> Replica
 *
 *     There can be a small delay before the replica gets the latest change.
 *
 *     Example:
 *       User updates profile -> Write goes to Primary
 *       User immediately reads -> Read goes to Replica
 *       Replica may still hold old data.
 *
 *     Solution for reads that need the latest data: read from Primary.
 */

/**
 * 23. SHARDING
 *     Problem: replicas don't solve everything.
 *       - Every replica contains the complete dataset.
 *       - Every write still goes to the primary.
 *
 *     If data size becomes too large OR write volume becomes too high,
 *     use Sharding: divide data across multiple databases.
 *
 *       App Servers --> Pick Shard from User ID
 *                         /     |     \
 *                        v      v      v
 *                    Shard1  Shard2  Shard3
 */

/**
 * 24. SHARD KEY
 *     Value used to decide which shard stores a record. Example: user_id
 *
 *       Shard 1 : Users 1 - 1M
 *       Shard 2 : Users 1M - 2M
 *       Shard 3 : Users 2M+
 *
 *     Different shards can process writes simultaneously.
 *
 *     Benefits:
 *       + Distributes storage
 *       + Distributes write traffic
 *       + Multiple shards work in parallel
 */

/**
 * 25. SHARDING TRADE-OFF
 *     Single-user query (e.g. load user profile) -> goes to ONE shard.
 *
 *     Cross-user query (e.g. search all posts) -> may hit Shard 1, 2, 3,
 *     then results must be merged.
 *
 *     So shard-key selection is extremely important.
 *       Good shard key : keeps common queries inside one shard.
 *       Bad shard key  : many queries touch multiple shards.
 */

/**
 * 26. QUEUES & BACKGROUND WORK
 *     Not every task must finish before responding to the user.
 *
 *     Example: user publishes a post.
 *       Required        : save post
 *       Additional work : notify followers
 *
 *     Bad approach:
 *       User -> Save Post -> Notify 1,000,000 followers -> Response
 *       (user waits for all notifications)
 *
 *     Better:
 *       User --> App Server --+--> Database
 *                             +--> Queue --> Worker1 / Worker2 --> Notifications
 *
 *     Flow:
 *       1. Save post
 *       2. Enqueue notification job
 *       3. Respond to user
 *       4. Workers process notifications
 *
 *     Benefits:
 *       + Faster user response
 *       + Background work is separated
 *       + Queue absorbs traffic spikes
 *
 *     If backlog grows: add more workers (if work can safely run in parallel).
 *
 *     IMPORTANT: A queue does NOT reduce total work.
 *     It removes that work from the user's request path.
 */

/**
 * 27. VERIFYING A SCALABILITY CHANGE
 *     After every change, test with a realistic workload and compare Before vs After.
 *
 *     1. REQUESTS COMPLETED PER SECOND
 *        Before 1,000 RPS -> After 5,000 RPS
 *        Tells us: did actual capacity increase?
 *
 *     2. MEDIAN RESPONSE TIME
 *        Before 50 ms -> After 45 ms
 *        Represents a typical request.
 *
 *     3. P99 RESPONSE TIME
 *        Before 500 ms -> After 120 ms
 *        Represents the slowest 1% of requests.
 *
 *     4. FAILED REQUESTS
 *        Before 2% -> After 0.1%
 *        Tells us whether the system is dropping work.
 */

/**
 * 28. WHY P95 / P99 MATTER
 *     Average latency can hide slow requests.
 *
 *     Example: average = 60 ms sounds good,
 *     but some requests may take 2 seconds.
 *
 *     Percentiles show the experience of slower requests:
 *       P95 : 95% of requests are at or below this latency.
 *       P99 : 99% of requests are at or below this latency.
 *
 *     Don't look only at average. Check p50, p95, p99.
 */

/**
 * 29. COST + CORRECTNESS
 *     A scaling solution isn't successful just because it's faster.
 *     Also check: cost, data correctness, failed requests.
 *
 *     Example: a system that responds faster BUT loses writes
 *     has NOT actually improved.
 *
 *     Consider together: Performance + Reliability + Correctness + Cost
 */

/**
 * 30. THERE IS NO FIXED SCALING THRESHOLD
 *     No rule like "At 1 million users, use sharding."
 *
 *     Application A: large data per user -> may need sharding early.
 *     Application B: small data per user, low write frequency
 *                    -> may use one database for years.
 *
 *     Scale based on workload and bottleneck, NOT a fixed user count.
 */

/**
 * STEP-BY-STEP SCALABILITY JOURNEY
 *
 *   START: Users -> Application + DB -> One Server
 *
 *   STEP 1: CPU / Memory becomes bottleneck
 *           -> Vertical Scaling
 *   STEP 2: Application + DB compete for resources
 *           -> Separate Database
 *   STEP 3: One app server reaches its limit
 *           -> Horizontal Scaling (Load Balancer + Multiple App Servers)
 *   STEP 4: Local sessions prevent free traffic distribution
 *           -> Stateless App Servers (Redis for sessions, S3 for files)
 *   STEP 5: Repeated reads overload database
 *           -> Cache (Cache-Aside)
 *   STEP 6: Cache itself becomes too large
 *           -> Multiple Cache Nodes
 *   STEP 7: Database read volume too high
 *           -> Read Replicas
 *   STEP 8: Database write volume / data size too large
 *           -> Sharding
 *   STEP 9: Slow background work increases request latency
 *           -> Queue + Background Workers
 */

/**
 * FINAL SCALABLE ARCHITECTURE
 *
 *                       USERS
 *                         |
 *                         v
 *                  LOAD BALANCER
 *                         |
 *           +-------------+-------------+
 *           |             |             |
 *           v             v             v
 *        APP 1          APP 2         APP 3
 *           |             |             |
 *           +-------------+-------------+
 *                         |
 *            +------------+------------+
 *            |            |            |
 *            v            v            v
 *          CACHE        REDIS          S3
 *                       Sessions       Files
 *            |
 *            v
 *     DATABASE LAYER
 *            |
 *     +------+------+
 *     |             |
 *     v             v
 *  Primary       Replicas
 *     |
 *     v
 *  Shards
 *
 *   Background Work:
 *     App --> Queue --> Worker 1 / Worker 2
 */

/**
 * QUICK REVISION
 *
 *   Scalability        -> Handle more load by adding resources.
 *   Bottleneck         -> Component limiting system capacity.
 *   Vertical Scaling   -> Bigger machine.
 *   Horizontal Scaling -> More machines.
 *   Load Balancer      -> Distributes traffic.
 *   Auto Scaling       -> Auto adds/removes servers based on demand.
 *   Stateless Servers  -> Shared state lives outside individual app servers.
 *   Redis              -> Shared session/state store.
 *   S3                 -> Shared storage for uploaded files.
 *   Cache              -> Avoid repeated expensive reads.
 *   Cache-Aside        -> Cache first -> DB on miss -> Store in cache.
 *   Read Replica       -> Scale database reads.
 *   Replication Lag    -> Replica may temporarily hold stale data.
 *   Sharding           -> Split data across databases.
 *   Shard Key          -> Determines where data belongs.
 *   Queue              -> Move slow/background work out of request path.
 *   Worker             -> Processes queued background jobs.
 *   P95/P99            -> Measure slow-request experience.
 */

/**
 * MOST IMPORTANT TRADE-OFFS
 *
 *   VERTICAL SCALING   : Simpler
 *                        vs Hardware ceiling + single point of failure
 *   HORIZONTAL SCALING : More capacity + availability
 *                        vs More infrastructure + network complexity
 *   STICKY SESSION     : Simple local sessions
 *                        vs Uneven traffic + failure loses sessions
 *   STATELESS SERVERS  : Easy scaling
 *                        vs Requires external/shared state storage
 *   CACHE              : Faster reads + less DB load
 *                        vs Cache invalidation + stale data
 *   READ REPLICAS      : Higher read capacity
 *                        vs Replication lag
 *   SHARDING           : Higher write/storage capacity
 *                        vs Cross-shard queries + operational complexity
 *   QUEUES             : Faster user response + traffic buffering
 *                        vs Asynchronous processing + eventual completion
 */

/**
 * INTERVIEW CHEAT SHEET
 *
 *   CPU is high
 *     -> More CPU / vertical scaling, or horizontal scaling if app can scale out
 *   Memory is full
 *     -> More RAM; check memory usage / caching
 *   Database query is slow
 *     -> Check query; add index / optimize; don't blindly add app servers
 *   App server reaches capacity
 *     -> Load balancer + multiple stateless servers
 *   Sessions stored locally
 *     -> Redis / shared session store; make app servers stateless
 *   Same data repeatedly requested
 *     -> Cache (cache-aside)
 *   Cache becomes too large
 *     -> Multiple cache nodes; distribute keys
 *   Database reads too high
 *     -> Read replicas
 *   Database writes / data size too high
 *     -> Sharding
 *   Request contains slow background work
 *     -> Queue + background workers
 *
 *   After every scaling change:
 *     -> Measure RPS
 *     -> Measure p50 / p95 / p99
 *     -> Check failed requests
 *     -> Check cost
 *     -> Check correctness
 */

/**
 * CORE SYSTEM DESIGN MINDSET
 *
 *   Never start with: "Let's add Kubernetes." / "Let's add Kafka." / "Let's add Redis."
 *
 *   Instead:
 *     1. Understand the workload
 *     2. Measure the load
 *     3. Find the bottleneck
 *     4. Choose the smallest change that addresses it
 *     5. Measure again
 *     6. Identify the next bottleneck
 *     7. Repeat
 *
 *   Core scalability principle:
 *     FIND WHAT BREAKS NEXT -> SCALE THAT COMPONENT
 *     -> MEASURE THE RESULT -> REPEAT
 */
```
