```python
/**
 * ============================================================
 * CAP THEOREM
 * ============================================================
 *
 * CAP Theorem distributed systems ka ek important concept hai.
 *
 * Ye mainly batata hai ki:
 *
 *     Jab distributed system ke servers/nodes
 *     ek dusre se communicate nahi kar paate,
 *
 *     tab system ko CONSISTENCY aur AVAILABILITY
 *     ke beech choice karni padti hai.
 *
 *
 * CAP =
 *
 *     C -> Consistency
 *     A -> Availability
 *     P -> Partition Tolerance
 *
 *
 * IMPORTANT:
 *
 *     CAP ka real trade-off network partition ke
 *     time par dikhta hai.
 *
 *
 * ============================================================
 * 1. SIMPLE REAL-WORLD EXAMPLE
 * ============================================================
 *
 * Suppose hamara online store hai.
 *
 * Data ki copies do regions mein hain:
 *
 *
 *     +----------------+          +----------------+
 *     |   Region A     |          |   Region B     |
 *     |                |          |                |
 *     | Price = $50    | <------> | Price = $50    |
 *     |                | Replicate|                |
 *     +----------------+          +----------------+
 *
 *
 * Administrator price change karta hai:
 *
 *     $50 -> $60
 *
 *
 * Region A:
 *
 *     Price = $60
 *
 *
 * Lekin update Region B tak pahunchne se pehle
 * network connection break ho jata hai.
 *
 *
 *     Region A                  Region B
 *     Price = $60               Price = $50
 *          |                         |
 *          +---------- X ------------+
 *                  Network Down
 *
 *
 * Ab Region B ke paas old value hai:
 *
 *     Price = $50
 *
 *
 * Customer Region B se poochta hai:
 *
 *     "Current price kya hai?"
 *
 *
 * Region B ke paas sirf:
 *
 *     $50
 *
 * hai aur Region A se contact nahi ho sakta.
 *
 *
 * Ab Region B ke paas 2 choices hain.
 *
 *
 * ============================================================
 * 2. CHOICE #1 -> ANSWER LOCALLY
 * ============================================================
 *
 * Region B bol sakta hai:
 *
 *     Price = $50
 *
 *
 * Request successfully complete ho gayi.
 *
 *
 * Benefit:
 *
 *     Availability maintained
 *
 *
 * Problem:
 *
 *     Actual latest value $60 hai.
 *
 *     Customer ko old value $50 mili.
 *
 *
 * Therefore:
 *
 *     AVAILABLE
 *     but
 *     NOT CONSISTENT
 *
 *
 * ============================================================
 * 3. CHOICE #2 -> REFUSE / WAIT
 * ============================================================
 *
 * Region B bol sakta hai:
 *
 *     "Mujhe latest value confirm nahi hai."
 *
 *
 * Therefore:
 *
 *     Error
 *
 * or
 *
 *     Wait until Region A se communication
 *     restore ho.
 *
 *
 * Benefit:
 *
 *     Stale value nahi milegi.
 *
 * Therefore:
 *
 *     CONSISTENT
 *     but
 *     NOT AVAILABLE
 *
 *
 * CAP ka main trade-off yahi hai.
 *
 *
 * ============================================================
 * 4. CAP KA MAIN IDEA
 * ============================================================
 *
 * Network partition ke time:
 *
 *
 *     CONSISTENCY
 *          vs
 *     AVAILABILITY
 *
 *
 * Dono ko simultaneously guarantee nahi kar
 * sakte for affected requests.
 *
 *
 * Because:
 *
 *     Region B ko woh update pata hi nahi hai
 *     jo Region A mein hua hai.
 *
 *
 * Jab tak communication restore nahi hota,
 * koi protocol magically missing information
 * discover nahi kar sakta.
 *
 *
 * ============================================================
 * 5. C = CONSISTENCY
 * ============================================================
 *
 * CAP mein Consistency ka meaning:
 *
 *     Distributed system aise behave kare
 *     jaise ek hi latest copy of data ho.
 *
 *
 * Example:
 *
 *     Old price:
 *
 *         $50
 *
 *     Update:
 *
 *         $50 -> $60
 *
 *
 * Once update successfully complete ho gaya:
 *
 *     Future read
 *          |
 *          v
 *        $60
 *
 *
 * It should NOT return:
 *
 *        $50
 *
 *
 * because $50 stale value hai.
 *
 *
 * Simple words:
 *
 *     "Successful write ke baad
 *      kisi bhi valid read ko old value
 *      nahi milni chahiye."
 *
 *
 * ============================================================
 * 6. CONSISTENCY MAINTAIN KAISE KAR SAKTE HAIN?
 * ============================================================
 *
 * Suppose:
 *
 *     Node 1
 *     Node 2
 *
 *
 * Client:
 *
 *     Write X = 5
 *
 *
 * Flow:
 *
 *     Client
 *       |
 *       | Write X=5
 *       v
 *     Node 1
 *       |
 *       | Replicate X=5
 *       v
 *     Node 2
 *       |
 *       | ACK
 *       v
 *     Node 1
 *       |
 *       v
 *     Client -> OK
 *
 *
 * Ab dono nodes ke paas:
 *
 *     X = 5
 *
 *
 * Therefore later read:
 *
 *     Node 1 -> 5
 *     Node 2 -> 5
 *
 *
 * Consistency maintained.
 *
 *
 * IMPORTANT:
 *
 * Har replica ko exactly same instant par update
 * karna zaroori nahi hai.
 *
 * But read ko stale data return nahi karna chahiye
 * agar system strong consistency guarantee kar raha hai.
 *
 *
 * ============================================================
 * 7. CAP CONSISTENCY vs ACID CONSISTENCY
 * ============================================================
 *
 * Ye dono same nahi hain.
 *
 *
 * ACID Consistency:
 *
 *     Database ke rules/constraints valid
 *     rehne chahiye.
 *
 *
 * Example:
 *
 *     Product price negative nahi ho sakta.
 *
 *
 * CAP Consistency:
 *
 *     Replicated data ko aise behave karna chahiye
 *     jaise ek single latest copy ho.
 *
 *
 * Example:
 *
 *     Write:
 *
 *         Price = $60
 *
 *
 *     Later read:
 *
 *         Must see $60
 *
 *
 * Difference:
 *
 *
 *     ACID Consistency
 *          |
 *          v
 *     Database rules / constraints
 *
 *
 *     CAP Consistency
 *          |
 *          v
 *     Replicas ka latest consistent view
 *
 *
 * ============================================================
 * 8. A = AVAILABILITY
 * ============================================================
 *
 * CAP mein Availability ka meaning:
 *
 *     Healthy node ko bheji gayi request
 *     eventually valid response return kare.
 *
 *
 * Example:
 *
 *     Region B healthy hai.
 *
 *     Region A se communication broken hai.
 *
 *
 * Agar Region B local data return karta hai:
 *
 *     Price = $50
 *
 *
 * Then:
 *
 *     Request successful
 *     -> Available
 *
 *
 * But value stale ho sakti hai.
 *
 *
 * Agar Region B return karta hai:
 *
 *     503 Service Unavailable
 *
 *
 * Then CAP ke availability definition ke according
 * woh request available nahi mani jaati.
 *
 *
 * ============================================================
 * 9. CAP AVAILABILITY != FAST RESPONSE
 * ============================================================
 *
 * Important interview point:
 *
 * CAP Availability ka matlab:
 *
 *     "Response fast hona chahiye"
 *
 * nahi hai.
 *
 *
 * CAP availability ka meaning:
 *
 *     Request eventually valid response de.
 *
 *
 * Example:
 *
 *     Response 5 seconds baad aaya
 *
 *     Still available ho sakta hai.
 *
 *
 * But:
 *
 *     Healthy node ne error return kar diya
 *     because coordination possible nahi thi
 *
 *     -> CAP availability satisfy nahi hui.
 *
 *
 * Also:
 *
 *     99.99% uptime
 *
 * aur
 *
 *     CAP Availability
 *
 * same concept nahi hain.
 *
 * CAP availability specifically partition ke time
 * individual requests ke behavior ko describe karti hai.
 *
 *
 * ============================================================
 * 10. P = PARTITION TOLERANCE
 * ============================================================
 *
 * Partition ka matlab:
 *
 *     Nodes individually alive hain,
 *     but ek dusre se communicate nahi kar pa rahe.
 *
 *
 * Example:
 *
 *
 *     Region A                    Region B
 *
 *     Node 1                     Node 3
 *     Node 2                     Node 4
 *       |                          |
 *       +----------- X ------------+
 *               Network
 *              Partition
 *
 *
 * Important:
 *
 *     Nodes down nahi hain.
 *
 *     Problem network communication mein hai.
 *
 *
 * ============================================================
 * 11. PARTITION KYUN HO SAKTA HAI?
 * ============================================================
 *
 * Examples:
 *
 *     Broken network link
 *
 *     Routing problem
 *
 *     Firewall misconfiguration
 *
 *     Overloaded network device
 *
 *     Packet loss / delay
 *
 *
 * Important:
 *
 *     Servers healthy ho sakte hain,
 *     but communication path broken ho sakta hai.
 *
 *
 * ============================================================
 * 12. PARTITION TOLERANCE KA ACTUAL MEANING
 * ============================================================
 *
 * Partition tolerance ka matlab:
 *
 *     System ko network communication failure
 *     ke situation ke liye design kiya gaya hai.
 *
 *
 * BUT:
 *
 *     Partition tolerance ka matlab
 *     ye nahi hai ki every request successful
 *     hogi.
 *
 *
 * Partition ke baad system ko decide karna padega:
 *
 *
 *     Local data se serve karo
 *
 *             OR
 *
 *     Coordination hone tak wait/reject karo
 *
 *
 * ============================================================
 * 13. "PICK ANY TWO" EXPLANATION
 * ============================================================
 *
 * CAP ko often bola jata hai:
 *
 *     "Pick any 2 out of C, A, P"
 *
 *
 * But ye explanation oversimplified hai.
 *
 *
 * Better understanding:
 *
 *     Normal network condition mein:
 *
 *         C + A
 *
 *         dono provide karna possible ho sakta hai.
 *
 *
 *     Real trade-off:
 *
 *         Network partition ke time
 *
 *         C vs A
 *
 *
 * because:
 *
 *     P ko distributed system mein ignore karna
 *     realistic nahi hai.
 *
 *
 * ============================================================
 * 14. WHY P IS IMPORTANT?
 * ============================================================
 *
 * Distributed system:
 *
 *     Node A
 *       |
 *       X
 *       |
 *     Node B
 *
 *
 * Network failure ko completely prevent
 * karna possible nahi hai.
 *
 *
 * Isliye practical distributed system ko
 * partition ke case mein behavior define karna
 * padta hai.
 *
 *
 * ============================================================
 * 15. CP SYSTEM
 * ============================================================
 *
 * CP =
 *
 *     C -> Consistency
 *     P -> Partition Tolerance
 *
 *
 * Meaning:
 *
 *     Partition ke time consistency preserve karo.
 *
 *
 * Agar system latest state confirm nahi kar sakta:
 *
 *     WAIT
 *
 * or
 *
 *     REJECT
 *
 *
 * rather than returning stale data.
 *
 *
 * ============================================================
 * 16. CP EXAMPLE: LAST ITEM IN STOCK
 * ============================================================
 *
 * Suppose:
 *
 *     Stock = 1
 *
 *
 * Two regions:
 *
 *     Region A
 *     Region B
 *
 *
 * Network partition:
 *
 *
 *     Region A       X       Region B
 *
 *
 * Customer 1:
 *
 *     Reserve item in A
 *
 *
 * Customer 2:
 *
 *     Reserve item in B
 *
 *
 * Dono regions locally dekh sakti hain:
 *
 *     Stock = 1
 *
 *
 * Agar dono independently confirm kar dein:
 *
 *     Customer 1 -> SUCCESS
 *     Customer 2 -> SUCCESS
 *
 *
 * Result:
 *
 *     Same 1 item
 *     sold twice.
 *
 *
 * BAD!
 *
 *
 * CP system kya karega?
 *
 *     Reservation ko leader / coordinating
 *     replica group ke through process karega.
 *
 *
 * Region jo authority tak reach nahi kar sakta:
 *
 *     Reject
 *
 * or
 *
 *     Wait
 *
 *
 * Result:
 *
 *     Availability sacrifice
 *
 *     But consistency preserved.
 *
 *
 * ============================================================
 * 17. CP FLOW
 * ============================================================
 *
 *
 * Customer
 *    |
 *    v
 * Region A
 *    |
 *    v
 * Inventory Leader
 *    |
 *    v
 * Stock: 1 -> 0
 *    |
 *    v
 * Confirm
 *
 *
 * Meanwhile:
 *
 * Customer 2
 *    |
 *    v
 * Region B
 *    |
 *    X
 * Cannot reach leader
 *    |
 *    v
 * Reject / Wait
 *
 *
 * Therefore:
 *
 *     Same item cannot be sold twice.
 *
 *
 * ============================================================
 * 18. CP READS
 * ============================================================
 *
 * CP ka rule writes tak limited nahi hai.
 *
 * Read ke time bhi:
 *
 *     Agar replica latest state prove nahi kar sakti
 *
 * then:
 *
 *     Reject / wait
 *
 *
 * Instead of:
 *
 *     Guess karke old data return karna.
 *
 *
 * ============================================================
 * 19. AP SYSTEM
 * ============================================================
 *
 * AP =
 *
 *     A -> Availability
 *     P -> Partition Tolerance
 *
 *
 * Meaning:
 *
 *     Network partition ke time bhi
 *     local requests ko continue karne do.
 *
 *
 * Each reachable region:
 *
 *     Local update accept kar sakta hai.
 *
 *
 * Replication:
 *
 *     Later
 *
 *
 * Communication restore hone ke baad:
 *
 *     Data merge / reconcile
 *
 *
 * ============================================================
 * 20. AP EXAMPLE: SHOPPING CART
 * ============================================================
 *
 * Customer ke paas:
 *
 *     Phone
 *     Laptop
 *
 *
 * Phone connected to:
 *
 *     Region A
 *
 *
 * Laptop connected to:
 *
 *     Region B
 *
 *
 * Network partition:
 *
 *
 *     Region A       X       Region B
 *
 *
 * Phone:
 *
 *     Add Book
 *
 *
 * Region A:
 *
 *     Cart = Book
 *
 *
 * Laptop:
 *
 *     Add Notebook
 *
 *
 * Region B:
 *
 *     Cart = Notebook
 *
 *
 * Dono updates:
 *
 *     ACCEPTED
 *
 *
 * Because AP availability maintain kar raha hai.
 *
 *
 * ============================================================
 * 21. NETWORK RESTORE HONE KE BAAD
 * ============================================================
 *
 *
 * Region A:
 *
 *     Cart = Book
 *
 *
 * Region B:
 *
 *     Cart = Notebook
 *
 *
 * Communication restore:
 *
 *          |
 *          v
 *        MERGE
 *          |
 *          v
 *
 *     Cart = Book + Notebook
 *
 *
 * Final:
 *
 *     Region A -> Book + Notebook
 *     Region B -> Book + Notebook
 *
 *
 * This is reconciliation.
 *
 *
 * ============================================================
 * 22. AP KA TRADE-OFF
 * ============================================================
 *
 * AP system:
 *
 *     Availability preserve karta hai.
 *
 *
 * But temporary:
 *
 *     Data drift
 *
 * ho sakta hai.
 *
 *
 * Example:
 *
 *     Region A:
 *
 *         Book
 *
 *     Region B:
 *
 *         Notebook
 *
 *
 * Partition ke time views different hain.
 *
 *
 * Later:
 *
 *     Merge
 *
 *
 * ============================================================
 * 23. AP CONFLICTS
 * ============================================================
 *
 * Har update easily merge nahi hota.
 *
 *
 * Example:
 *
 *     Region A:
 *
 *         Product description = "Red Shirt"
 *
 *
 *     Region B:
 *
 *         Product description = "Blue Shirt"
 *
 *
 * Dono same field ko different value
 * mein update kar rahe hain.
 *
 *
 * Ab system ko decide karna padega:
 *
 *     Which version wins?
 *
 *
 * Possible application-level rules:
 *
 *     Last write wins
 *     Version based rule
 *     Custom merge rule
 *
 *
 * Important:
 *
 *     Merge rule system/application design ka part hai.
 *
 *
 * ============================================================
 * 24. LAST ITEM STOCK MEIN AP KYUN DANGEROUS?
 * ============================================================
 *
 * Shopping cart:
 *
 *     Book + Notebook
 *
 * easily merge ho sakta hai.
 *
 *
 * But inventory:
 *
 *     Stock = 1
 *
 *
 * Region A:
 *
 *     Reserve -> SUCCESS
 *
 *
 * Region B:
 *
 *     Reserve -> SUCCESS
 *
 *
 * Final:
 *
 *     1 item
 *     2 customers
 *
 *
 * Cannot safely merge.
 *
 *
 * Therefore:
 *
 *     Critical inventory reservation
 *     should usually require coordination.
 *
 *
 * ============================================================
 * 25. CP vs AP
 * ============================================================
 *
 *
 * CP:
 *
 *     Partition
 *        |
 *        v
 *     Cannot coordinate?
 *        |
 *        +----> WAIT / REJECT
 *
 *     Goal:
 *
 *         Consistency
 *
 *
 * ------------------------------------------------------------
 *
 * AP:
 *
 *     Partition
 *        |
 *        v
 *     Cannot coordinate?
 *        |
 *        +----> Serve locally
 *
 *     Later:
 *
 *         Merge / Reconcile
 *
 *     Goal:
 *
 *         Availability
 *
 *
 * ============================================================
 * 26. OPERATION-WISE CAP
 * ============================================================
 *
 * Very important:
 *
 *     CP/AP labels ko entire database par blindly
 *     apply karna useful nahi hai.
 *
 *
 * Different operations ko different behavior
 * ki requirement ho sakti hai.
 *
 *
 * Example:
 *
 *
 * Product description:
 *
 *     Slightly stale data
 *     acceptable
 *
 *     -> Can keep serving
 *
 *
 * Catalog browsing:
 *
 *     Small delay in updates acceptable
 *
 *     -> Can keep serving
 *
 *
 * Add item to cart:
 *
 *     Can accept and merge later
 *
 *
 * Reserve last item:
 *
 *     Must coordinate
 *
 *     -> Stop / wait
 *
 *
 * API key validity:
 *
 *     Revoked key should not continue working
 *
 *     -> Need latest state
 *
 *
 * ============================================================
 * 27. CA = CONSISTENCY + AVAILABILITY
 * ============================================================
 *
 * CA means:
 *
 *     C -> Consistency
 *     A -> Availability
 *
 *
 * Example:
 *
 *     Single-node database
 *
 *
 * Because:
 *
 *     No replicas
 *     No cross-node network coordination
 *
 *
 * Therefore healthy state mein:
 *
 *     Consistency + Availability
 *
 * dono provide kiye ja sakte hain.
 *
 *
 * ============================================================
 * 28. WHY CA IS NOT VERY USEFUL FOR
 *     DISTRIBUTED SYSTEMS
 * ============================================================
 *
 * Distributed system mein:
 *
 *     Network partition possible hai.
 *
 *
 * Once:
 *
 *     Node A X Node B
 *
 *
 * system ko decide karna hi padega:
 *
 *     C?
 *
 *     OR
 *
 *     A?
 *
 *
 * Agar design partition ka behavior define hi nahi
 * karta, to usne CAP trade-off escape nahi kiya.
 *
 *
 * Isliye practical distributed system ke context mein:
 *
 *     "CA"
 *
 * useful classification nahi hai
 * jab partitions possible hain.
 *
 *
 * ============================================================
 * 29. PACELC THEOREM
 * ============================================================
 *
 * CAP mainly partition ke time ka trade-off batata hai.
 *
 *
 * PACELC is idea ko extend karta hai.
 *
 *
 * P = Partition
 *
 * A = Availability
 *
 * C = Consistency
 *
 * E = Else
 *
 * L = Latency
 *
 * C = Consistency
 *
 *
 * Rule:
 *
 *
 *     If Partition:
 *
 *         Availability
 *              vs
 *         Consistency
 *
 *
 *     Else:
 *
 *         Latency
 *              vs
 *         Consistency
 *
 *
 * ============================================================
 * 30. PACELC SIMPLE EXAMPLE
 * ============================================================
 *
 * Network healthy hai.
 *
 *
 * Customer Region B mein hai.
 *
 *
 * Option 1:
 *
 *     Nearby replica se read
 *
 *         |
 *         v
 *     Fast
 *
 *         BUT
 *
 *     Replica slightly behind ho sakti hai.
 *
 *
 * Option 2:
 *
 *     Remote leader se read
 *
 *         |
 *         v
 *     Slower
 *
 *         BUT
 *
 *     Latest value milne ki stronger guarantee.
 *
 *
 * Therefore:
 *
 *     Faster response
 *          vs
 *     Stronger consistency
 *
 *
 * ============================================================
 * 31. WHY STRONGER CONSISTENCY CAN BE SLOWER
 * ============================================================
 *
 * Strong consistency ke liye:
 *
 *     More coordination
 *
 * required ho sakti hai.
 *
 *
 * Example:
 *
 *     Client
 *        |
 *        v
 *     Region B
 *        |
 *        v
 *     Leader in Region A
 *        |
 *        v
 *     Latest value
 *
 *
 * Cross-region communication:
 *
 *     Extra network latency
 *
 *
 * Therefore:
 *
 *     Stronger consistency
 *          |
 *          v
 *     More coordination
 *          |
 *          v
 *     Higher latency
 *
 *
 * ============================================================
 * 32. CAP vs PACELC
 * ============================================================
 *
 *
 * CAP:
 *
 *     Partition ke time:
 *
 *         C vs A
 *
 *
 * PACELC:
 *
 *     Partition:
 *
 *         C vs A
 *
 *
 *     Else / Normal operation:
 *
 *         C vs L
 *
 *
 * ============================================================
 * 33. QUICK COMPARISON
 * ============================================================
 *
 *
 * CP
 *
 *     Partition
 *        |
 *        v
 *     Preserve consistency
 *        |
 *        v
 *     Reject / Wait
 *
 *
 * AP
 *
 *     Partition
 *        |
 *        v
 *     Preserve availability
 *        |
 *        v
 *     Serve local data
 *        |
 *        v
 *     Reconcile later
 *
 *
 * CA
 *
 *     No partition assumption
 *        |
 *        v
 *     Consistency + Availability
 *
 *
 * PACELC
 *
 *     Partition
 *        |
 *        v
 *     C vs A
 *
 *     No Partition
 *        |
 *        v
 *     C vs L
 *
 *
 * ============================================================
 * 34. REAL-LIFE OPERATION MAPPING
 * ============================================================
 *
 *
 * Operation:
 *
 *     Product browsing
 *
 * Choice:
 *
 *     AP-like
 *
 * Reason:
 *
 *     Slightly stale catalog acceptable.
 *
 *
 * ------------------------------------------------------------
 *
 * Operation:
 *
 *     Shopping cart
 *
 * Choice:
 *
 *     Availability-friendly
 *
 * Reason:
 *
 *     Updates can potentially merge later.
 *
 *
 * ------------------------------------------------------------
 *
 * Operation:
 *
 *     Last item reservation
 *
 * Choice:
 *
 *     CP-like
 *
 * Reason:
 *
 *     Same item ko two customers ko sell nahi
 *     karna chahiye.
 *
 *
 * ------------------------------------------------------------
 *
 * Operation:
 *
 *     API key validation
 *
 * Choice:
 *
 *     Consistency-focused
 *
 * Reason:
 *
 *     Revoked key ko partition ke wajah se
 *     valid nahi rehna chahiye.
 *
 *
 * ============================================================
 * 35. INTERVIEW QUESTIONS
 * ============================================================
 *
 *
 * Q1. What is CAP theorem?
 *
 * Answer:
 *
 *     CAP theorem distributed system mein
 *     network partition ke time consistency aur
 *     availability ke trade-off ko describe karta hai.
 *
 *
 * ------------------------------------------------------------
 *
 * Q2. What does C mean?
 *
 * Answer:
 *
 *     Consistency means replicated data
 *     single latest up-to-date copy ki tarah
 *     behave kare.
 *
 *
 * ------------------------------------------------------------
 *
 * Q3. What does A mean?
 *
 * Answer:
 *
 *     Availability means healthy node ko
 *     request eventually valid response de.
 *
 *
 * ------------------------------------------------------------
 *
 * Q4. What does P mean?
 *
 * Answer:
 *
 *     Partition tolerance means system ko
 *     handle karna aata hai jab nodes
 *     ek dusre se communicate nahi kar paate.
 *
 *
 * ------------------------------------------------------------
 *
 * Q5. What is a network partition?
 *
 * Answer:
 *
 *     Nodes alive hain but communication
 *     between them fail ho gaya hai.
 *
 *
 * ------------------------------------------------------------
 *
 * Q6. Why can't we guarantee both C and A
 *     during a partition?
 *
 * Answer:
 *
 *     Because disconnected node ke paas
 *     latest update ki information nahi hoti.
 *
 *     Agar local old value return kare:
 *
 *         Availability maintained
 *         Consistency lost
 *
 *     Agar answer refuse kare:
 *
 *         Consistency maintained
 *         Availability lost
 *
 *
 * ------------------------------------------------------------
 *
 * Q7. What is CP?
 *
 * Answer:
 *
 *     Partition ke time consistency preserve karna,
 *     even if affected requests ko wait/reject
 *     karna pade.
 *
 *
 * ------------------------------------------------------------
 *
 * Q8. What is AP?
 *
 * Answer:
 *
 *     Partition ke time requests ko local data
 *     se serve karna aur later reconciliation
 *     karna.
 *
 *
 * ------------------------------------------------------------
 *
 * Q9. Give a CP use case.
 *
 * Answer:
 *
 *     Last item inventory reservation.
 *
 *     Same item ko do regions simultaneously
 *     sell nahi kar sakte.
 *
 *
 * ------------------------------------------------------------
 *
 * Q10. Give an AP use case.
 *
 * Answer:
 *
 *     Shopping cart updates jahan independent
 *     additions later merge kiye ja sakte hain.
 *
 *
 * ------------------------------------------------------------
 *
 * Q11. Is CAP availability same as 99.99% uptime?
 *
 * Answer:
 *
 *     No.
 *
 *     CAP availability partition ke time
 *     individual requests ke response behavior
 *     ko describe karti hai.
 *
 *
 * ------------------------------------------------------------
 *
 * Q12. Is CAP consistency same as ACID consistency?
 *
 * Answer:
 *
 *     No.
 *
 *     ACID consistency database constraints/rules
 *     ke baare mein hai.
 *
 *     CAP consistency replicated data ke
 *     latest consistent view ke baare mein hai.
 *
 *
 * ------------------------------------------------------------
 *
 * Q13. Why is "pick any two" an oversimplification?
 *
 * Answer:
 *
 *     Real trade-off mainly network partition
 *     ke time appear hota hai.
 *
 *     Healthy network mein system C + A dono
 *     provide kar sakta hai.
 *
 *
 * ------------------------------------------------------------
 *
 * Q14. What is PACELC?
 *
 * Answer:
 *
 *     PACELC CAP ko extend karta hai:
 *
 *         Partition -> A vs C
 *
 *         Else -> L vs C
 *
 *
 * ------------------------------------------------------------
 *
 * Q15. Why can stronger consistency increase latency?
 *
 * Answer:
 *
 *     Strong consistency ke liye additional
 *     coordination/replica agreement required
 *     ho sakta hai.
 *
 *     Cross-region coordination network latency
 *     increase kar sakti hai.
 *
 *
 * ------------------------------------------------------------
 *
 * Q16. Can one application use both CP and AP
 *     behavior?
 *
 * Answer:
 *
 *     Yes.
 *
 *     Different operations ki requirements
 *     different ho sakti hain.
 *
 *     Example:
 *
 *         Catalog browsing -> availability-friendly
 *
 *         Last-item reservation -> consistency-focused
 *
 *
 * ============================================================
 * 36. MOST IMPORTANT MEMORY TRICK
 * ============================================================
 *
 *
 * NETWORK HEALTHY:
 *
 *     C + A
 *
 *     Often possible.
 *
 *
 * NETWORK PARTITION:
 *
 *     C <-----> A
 *
 *     Trade-off appears.
 *
 *
 * CP:
 *
 *     "Latest data chahiye,
 *      warna wait/reject."
 *
 *
 * AP:
 *
 *     "Service chalu rakho,
 *      baad mein data reconcile kar lenge."
 *
 *
 * PACELC:
 *
 *     "Partition mein C vs A,
 *      normal condition mein C vs L."
 *
 *
 * ============================================================
 * 37. ONE-LINE INTERVIEW ANSWER
 * ============================================================
 *
 * "CAP theorem says that during a network partition
 * in a distributed system, we cannot guarantee both
 * strong consistency and availability for every
 * affected request, so the system must choose how
 * each operation behaves."
 *
 * ============================================================
 */
```

```python
/**
 * CAP THEOREM
 *
 * Distributed systems ka ek important concept.
 *
 * Jab distributed system ke servers/nodes ek dusre se communicate nahi kar
 * paate, tab system ko CONSISTENCY aur AVAILABILITY ke beech choice karni padti hai.
 *
 * CAP = C (Consistency) + A (Availability) + P (Partition Tolerance)
 *
 * IMPORTANT: CAP ka real trade-off network partition ke time par dikhta hai.
 */

/**
 * 1. SIMPLE REAL-WORLD EXAMPLE
 *    Online store, data ki copies do regions mein:
 *
 *      +--------------+   Replicate   +--------------+
 *      |  Region A    | <-----------> |  Region B    |
 *      |  Price = $50 |               |  Price = $50 |
 *      +--------------+               +--------------+
 *
 *    Admin price change karta hai: $50 -> $60.
 *    Region A: Price = $60.
 *    Lekin update Region B tak pahunchne se pehle network connection break:
 *
 *      Region A (Price = $60) ---- X ---- Region B (Price = $50)
 *                              Network Down
 *
 *    Customer Region B se poochta hai: "Current price kya hai?"
 *    Region B ke paas sirf $50 hai aur Region A se contact nahi ho sakta.
 *    Ab Region B ke paas 2 choices hain.
 */

/**
 * 2. CHOICE #1 -> ANSWER LOCALLY
 *    Region B bol sakta hai: Price = $50. Request successfully complete.
 *
 *    Benefit : Availability maintained.
 *    Problem : actual latest value $60 hai, customer ko old value $50 mili.
 *
 *    Therefore: AVAILABLE but NOT CONSISTENT.
 */

/**
 * 3. CHOICE #2 -> REFUSE / WAIT
 *    Region B bol sakta hai: "Mujhe latest value confirm nahi hai."
 *    Error return kare, ya Region A se communication restore hone tak wait kare.
 *
 *    Benefit: stale value nahi milegi.
 *    Therefore: CONSISTENT but NOT AVAILABLE.
 *
 *    CAP ka main trade-off yahi hai.
 */

/**
 * 4. CAP KA MAIN IDEA
 *    Network partition ke time: CONSISTENCY vs AVAILABILITY.
 *    Dono ko simultaneously guarantee nahi kar sakte for affected requests.
 *
 *    Because Region B ko woh update pata hi nahi hai jo Region A mein hua.
 *    Jab tak communication restore nahi hota, koi protocol magically missing
 *    information discover nahi kar sakta.
 */

/**
 * 5. C = CONSISTENCY
 *    Distributed system aise behave kare jaise ek hi latest copy of data ho.
 *
 *    Example: old price $50, update $50 -> $60.
 *    Update successfully complete hone ke baad future read -> $60.
 *    It should NOT return $50 (stale value).
 *
 *    Simple words: "Successful write ke baad kisi bhi valid read ko
 *    old value nahi milni chahiye."
 */

/**
 * 6. CONSISTENCY MAINTAIN KAISE KAR SAKTE HAIN?
 *    Nodes: Node 1, Node 2. Client: Write X = 5.
 *
 *      Client --Write X=5--> Node 1 --Replicate X=5--> Node 2
 *      Node 2 --ACK--> Node 1 --> Client: OK
 *
 *    Ab dono nodes ke paas X = 5. Later read: Node 1 -> 5, Node 2 -> 5.
 *    Consistency maintained.
 *
 *    IMPORTANT: har replica ko exactly same instant par update karna zaroori
 *    nahi, but strong consistency guarantee hai to read ko stale data
 *    return nahi karna chahiye.
 */

/**
 * 7. CAP CONSISTENCY vs ACID CONSISTENCY
 *    Ye dono same nahi hain.
 *
 *    ACID Consistency: database ke rules/constraints valid rehne chahiye.
 *      Example: product price negative nahi ho sakta.
 *
 *    CAP Consistency: replicated data ko aise behave karna chahiye jaise
 *    ek single latest copy ho.
 *      Example: write Price = $60 -> later read must see $60.
 *
 *    ACID Consistency -> database rules / constraints
 *    CAP Consistency  -> replicas ka latest consistent view
 */

/**
 * 8. A = AVAILABILITY
 *    Healthy node ko bheji gayi request eventually valid response return kare.
 *
 *    Example: Region B healthy hai, Region A se communication broken.
 *      Region B local data return kare (Price = $50) -> request successful
 *      -> Available (but value stale ho sakti hai).
 *      Region B 503 Service Unavailable return kare -> CAP availability
 *      definition ke according woh request available nahi mani jaati.
 */

/**
 * 9. CAP AVAILABILITY != FAST RESPONSE
 *    Important interview point: CAP Availability ka matlab
 *    "Response fast hona chahiye" nahi hai.
 *
 *    CAP availability = request eventually valid response de.
 *
 *    Example:
 *      Response 5 seconds baad aaya -> still available ho sakta hai.
 *      Healthy node ne error return kiya kyunki coordination possible nahi thi
 *      -> CAP availability satisfy nahi hui.
 *
 *    Also: 99.99% uptime aur CAP Availability same concept nahi hain.
 *    CAP availability specifically partition ke time individual requests
 *    ke behavior ko describe karti hai.
 */

/**
 * 10. P = PARTITION TOLERANCE
 *     Partition = nodes individually alive hain, but ek dusre se
 *     communicate nahi kar pa rahe.
 *
 *       Region A (Node 1, Node 2) ---- X ---- Region B (Node 3, Node 4)
 *                              Network Partition
 *
 *     Important: nodes down nahi hain, problem network communication mein hai.
 */

/**
 * 11. PARTITION KYUN HO SAKTA HAI?
 *     Examples: broken network link, routing problem, firewall
 *     misconfiguration, overloaded network device, packet loss / delay.
 *
 *     Servers healthy ho sakte hain, but communication path broken ho sakta hai.
 */

/**
 * 12. PARTITION TOLERANCE KA ACTUAL MEANING
 *     System ko network communication failure ke situation ke liye design kiya gaya hai.
 *
 *     BUT: iska matlab ye nahi ki every request successful hogi.
 *
 *     Partition ke baad system ko decide karna padega:
 *       Local data se serve karo  OR  Coordination hone tak wait/reject karo.
 */

/**
 * 13. "PICK ANY TWO" EXPLANATION
 *     CAP ko often bola jata hai: "Pick any 2 out of C, A, P".
 *     But ye explanation oversimplified hai.
 *
 *     Better understanding:
 *       Normal network condition mein C + A dono provide karna possible ho sakta hai.
 *       Real trade-off: network partition ke time C vs A.
 *
 *     Because P ko distributed system mein ignore karna realistic nahi hai.
 */

/**
 * 14. WHY P IS IMPORTANT?
 *     Node A ---- X ---- Node B
 *
 *     Network failure ko completely prevent karna possible nahi hai.
 *     Isliye practical distributed system ko partition ke case mein
 *     behavior define karna padta hai.
 */

/**
 * 15. CP SYSTEM
 *     CP = Consistency + Partition Tolerance.
 *     Partition ke time consistency preserve karo.
 *
 *     Agar system latest state confirm nahi kar sakta:
 *       WAIT or REJECT, rather than returning stale data.
 */

/**
 * 16. CP EXAMPLE: LAST ITEM IN STOCK
 *     Stock = 1, two regions (A, B), network partition: Region A -X- Region B.
 *
 *     Customer 1 reserves item in A, Customer 2 reserves item in B.
 *     Dono regions locally dekh sakti hain Stock = 1.
 *     Agar dono independently confirm kar dein:
 *       Customer 1 -> SUCCESS, Customer 2 -> SUCCESS
 *       -> same 1 item sold twice. BAD!
 *
 *     CP system kya karega?
 *       Reservation ko leader / coordinating replica group ke through
 *       process karega. Jo region authority tak reach nahi kar sakta:
 *       Reject or Wait.
 *
 *     Result: availability sacrifice, but consistency preserved.
 */

/**
 * 17. CP FLOW
 *
 *     Customer 1 -> Region A -> Inventory Leader -> Stock: 1 -> 0 -> Confirm
 *
 *     Meanwhile:
 *     Customer 2 -> Region B -X-> cannot reach leader -> Reject / Wait
 *
 *     Therefore: same item cannot be sold twice.
 */

/**
 * 18. CP READS
 *     CP ka rule writes tak limited nahi hai.
 *     Read ke time bhi: agar replica latest state prove nahi kar sakti
 *     -> Reject / wait, instead of guess karke old data return karna.
 */

/**
 * 19. AP SYSTEM
 *     AP = Availability + Partition Tolerance.
 *     Network partition ke time bhi local requests ko continue karne do.
 *
 *     Each reachable region local update accept kar sakta hai.
 *     Replication: later.
 *     Communication restore hone ke baad: data merge / reconcile.
 */

/**
 * 20. AP EXAMPLE: SHOPPING CART
 *     Customer ke paas Phone (-> Region A) aur Laptop (-> Region B).
 *     Network partition: Region A -X- Region B.
 *
 *     Phone  : Add Book     -> Region A: Cart = Book
 *     Laptop : Add Notebook -> Region B: Cart = Notebook
 *
 *     Dono updates ACCEPTED, because AP availability maintain kar raha hai.
 */

/**
 * 21. NETWORK RESTORE HONE KE BAAD
 *     Region A: Cart = Book | Region B: Cart = Notebook
 *     Communication restore -> MERGE -> Cart = Book + Notebook
 *
 *     Final: Region A -> Book + Notebook, Region B -> Book + Notebook.
 *     This is reconciliation.
 */

/**
 * 22. AP KA TRADE-OFF
 *     AP system availability preserve karta hai.
 *     But temporary data drift ho sakta hai.
 *
 *     Example: Region A: Book, Region B: Notebook.
 *     Partition ke time views different hain. Later: Merge.
 */

/**
 * 23. AP CONFLICTS
 *     Har update easily merge nahi hota.
 *
 *     Example:
 *       Region A: Product description = "Red Shirt"
 *       Region B: Product description = "Blue Shirt"
 *     Dono same field ko different value mein update kar rahe hain.
 *
 *     System ko decide karna padega: which version wins?
 *     Possible application-level rules: last write wins, version based rule,
 *     custom merge rule.
 *
 *     Merge rule system/application design ka part hai.
 */

/**
 * 24. LAST ITEM STOCK MEIN AP KYUN DANGEROUS?
 *     Shopping cart: Book + Notebook easily merge ho sakta hai.
 *
 *     But inventory: Stock = 1
 *       Region A: Reserve -> SUCCESS
 *       Region B: Reserve -> SUCCESS
 *       Final: 1 item, 2 customers. Cannot safely merge.
 *
 *     Therefore critical inventory reservation should usually require coordination.
 */

/**
 * 25. CP vs AP
 *
 *     CP: Partition -> cannot coordinate? -> WAIT / REJECT.   Goal: Consistency.
 *
 *     AP: Partition -> cannot coordinate? -> Serve locally.
 *         Later: Merge / Reconcile.                           Goal: Availability.
 */

/**
 * 26. OPERATION-WISE CAP
 *     CP/AP labels ko entire database par blindly apply karna useful nahi.
 *     Different operations ko different behavior ki requirement ho sakti hai.
 *
 *     Product description : slightly stale data acceptable -> keep serving.
 *     Catalog browsing    : small delay in updates acceptable -> keep serving.
 *     Add item to cart    : can accept and merge later.
 *     Reserve last item   : must coordinate -> stop / wait.
 *     API key validity    : revoked key should not continue working
 *                           -> need latest state.
 */

/**
 * 27. CA = CONSISTENCY + AVAILABILITY
 *     Example: single-node database.
 *     Because no replicas and no cross-node network coordination,
 *     healthy state mein Consistency + Availability dono provide ho sakte hain.
 */

/**
 * 28. WHY CA IS NOT VERY USEFUL FOR DISTRIBUTED SYSTEMS
 *     Distributed system mein network partition possible hai.
 *     Once Node A -X- Node B, system ko decide karna hi padega: C ya A?
 *
 *     Agar design partition ka behavior define hi nahi karta, to usne
 *     CAP trade-off escape nahi kiya.
 *
 *     Isliye jab partitions possible hain, "CA" useful classification nahi hai.
 */

/**
 * 29. PACELC THEOREM
 *     CAP mainly partition ke time ka trade-off batata hai.
 *     PACELC is idea ko extend karta hai.
 *
 *     P = Partition, A = Availability, C = Consistency,
 *     E = Else, L = Latency, C = Consistency
 *
 *     If Partition: Availability vs Consistency.
 *     Else        : Latency vs Consistency.
 */

/**
 * 30. PACELC SIMPLE EXAMPLE
 *     Network healthy hai, customer Region B mein hai.
 *
 *     Option 1: nearby replica se read
 *       -> Fast, BUT replica slightly behind ho sakti hai.
 *     Option 2: remote leader se read
 *       -> Slower, BUT latest value milne ki stronger guarantee.
 *
 *     Faster response vs Stronger consistency.
 */

/**
 * 31. WHY STRONGER CONSISTENCY CAN BE SLOWER
 *     Strong consistency ke liye more coordination required ho sakti hai.
 *
 *       Client -> Region B -> Leader in Region A -> Latest value
 *
 *     Cross-region communication = extra network latency.
 *
 *     Stronger consistency -> more coordination -> higher latency.
 */

/**
 * 32. CAP vs PACELC
 *     CAP    : partition ke time C vs A.
 *     PACELC : partition mein C vs A; else / normal operation mein C vs L.
 */

/**
 * 33. QUICK COMPARISON
 *
 *     CP     : Partition -> preserve consistency -> Reject / Wait
 *     AP     : Partition -> preserve availability -> Serve local data
 *              -> Reconcile later
 *     CA     : No partition assumption -> Consistency + Availability
 *     PACELC : Partition -> C vs A | No Partition -> C vs L
 */

/**
 * 34. REAL-LIFE OPERATION MAPPING
 *
 *     Product browsing      : AP-like. Slightly stale catalog acceptable.
 *     Shopping cart         : Availability-friendly. Updates can merge later.
 *     Last item reservation : CP-like. Same item do customers ko sell nahi karna chahiye.
 *     API key validation    : Consistency-focused. Revoked key partition ki
 *                             wajah se valid nahi rehna chahiye.
 */

/**
 * 35. INTERVIEW QUESTIONS
 *
 * Q1. What is CAP theorem?
 * A : Distributed system mein network partition ke time consistency aur
 *     availability ke trade-off ko describe karta hai.
 *
 * Q2. What does C mean?
 * A : Replicated data single latest up-to-date copy ki tarah behave kare.
 *
 * Q3. What does A mean?
 * A : Healthy node ko request eventually valid response de.
 *
 * Q4. What does P mean?
 * A : System ko handle karna aata hai jab nodes ek dusre se communicate
 *     nahi kar paate.
 *
 * Q5. What is a network partition?
 * A : Nodes alive hain but communication between them fail ho gaya hai.
 *
 * Q6. Why can't we guarantee both C and A during a partition?
 * A : Disconnected node ke paas latest update ki information nahi hoti.
 *     Local old value return kare -> availability maintained, consistency lost.
 *     Answer refuse kare          -> consistency maintained, availability lost.
 *
 * Q7. What is CP?
 * A : Partition ke time consistency preserve karna, even if affected
 *     requests ko wait/reject karna pade.
 *
 * Q8. What is AP?
 * A : Partition ke time requests ko local data se serve karna aur later
 *     reconciliation karna.
 *
 * Q9. Give a CP use case.
 * A : Last item inventory reservation. Same item ko do regions
 *     simultaneously sell nahi kar sakte.
 *
 * Q10. Give an AP use case.
 * A : Shopping cart updates jahan independent additions later merge ho sakte hain.
 *
 * Q11. Is CAP availability same as 99.99% uptime?
 * A : No. CAP availability partition ke time individual requests ke
 *     response behavior ko describe karti hai.
 *
 * Q12. Is CAP consistency same as ACID consistency?
 * A : No. ACID consistency database constraints/rules ke baare mein hai.
 *     CAP consistency replicated data ke latest consistent view ke baare mein hai.
 *
 * Q13. Why is "pick any two" an oversimplification?
 * A : Real trade-off mainly network partition ke time appear hota hai.
 *     Healthy network mein system C + A dono provide kar sakta hai.
 *
 * Q14. What is PACELC?
 * A : CAP ka extension: Partition -> A vs C, Else -> L vs C.
 *
 * Q15. Why can stronger consistency increase latency?
 * A : Strong consistency ke liye additional coordination/replica agreement
 *     chahiye. Cross-region coordination network latency badha sakti hai.
 *
 * Q16. Can one application use both CP and AP behavior?
 * A : Yes. Different operations ki requirements different ho sakti hain.
 *     Example: catalog browsing -> availability-friendly;
 *     last-item reservation -> consistency-focused.
 */

/**
 * 36. MOST IMPORTANT MEMORY TRICK
 *
 *     NETWORK HEALTHY  : C + A often possible.
 *     NETWORK PARTITION: C <-----> A, trade-off appears.
 *
 *     CP     : "Latest data chahiye, warna wait/reject."
 *     AP     : "Service chalu rakho, baad mein data reconcile kar lenge."
 *     PACELC : "Partition mein C vs A, normal condition mein C vs L."
 */

/**
 * 37. ONE-LINE INTERVIEW ANSWER
 *     "CAP theorem says that during a network partition in a distributed
 *      system, we cannot guarantee both strong consistency and availability
 *      for every affected request, so the system must choose how each
 *      operation behaves."
 */
```
