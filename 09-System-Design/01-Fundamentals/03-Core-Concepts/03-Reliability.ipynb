```python
/**
 * RELIABILITY — SYSTEM DESIGN
 * - Core idea:
 *   - Availability: "Is the system responding?"
 *   - Reliability : "Is the system responding CORRECTLY?"
 *
 * - A system can be available but unreliable.
 * - Example: Payment request succeeds > Customer is charged twice
 * - System was:
 *   - Available     -> Yes
 *   - Reliable      -> No
 *
 *
 * Another example:
 *
 *     Messaging app is available
 *     BUT messages arrive out of order.
 *
 *     Available     -> Yes
 *     Reliable      -> No
 *
 *
 * Reliability means:
 *
 *     System performs its intended function
 *     correctly and consistently over time.
 *
 *
 * ============================================================
 * 1. WHAT IS RELIABILITY?
 * ============================================================
 *
 * Reliability is the probability that a system will perform
 * its intended function correctly over a given period of time
 * under specified conditions.
 *
 *
 * Three important parts:
 *
 *
 * CORRECTLY
 *
 *     System must produce the correct result.
 *
 *
 * OVER A GIVEN PERIOD
 *
 *     Reliability is measured over time,
 *     not at one particular moment.
 *
 *
 * UNDER SPECIFIED CONDITIONS
 *
 *     We define what normal operating conditions are.
 *
 *
 * Simple definition:
 *
 *     Reliability =
 *     "Does the system do what it is supposed to do?"
 *
 *
 * ============================================================
 * 2. RELIABILITY vs AVAILABILITY
 * ============================================================
 *
 *
 * AVAILABILITY
 *
 * Question:
 *
 *     "Is the system responding?"
 *
 * Example:
 *
 *     HTTP 200 response received.
 *
 *
 * RELIABILITY
 *
 * Question:
 *
 *     "Is the response correct?"
 *
 * Example:
 *
 *     Account balance returned is accurate.
 *
 *
 * IMPORTANT:
 *
 *     Available != Reliable
 *
 *
 * Example:
 *
 *     Payment API:
 *
 *     Request succeeds
 *          ↓
 *     Wrong amount charged
 *
 *     Available -> YES
 *     Reliable  -> NO
 *
 *
 * ============================================================
 * 3. RELIABILITY vs FAULT TOLERANCE
 * ============================================================
 *
 * FAULT TOLERANCE:
 *
 *     "Can the system continue working
 *      when a component fails?"
 *
 *
 * Example:
 *
 *     Database Replica 1
 *         ↓
 *        FAIL
 *         ↓
 *     Database Replica 2
 *         ↓
 *     System continues
 *
 *
 * Reliability:
 *
 *     Is the system producing correct results?
 *
 *
 * Fault Tolerance:
 *
 *     Can the system continue operating
 *     despite component failures?
 *
 *
 * ============================================================
 * 4. RELIABILITY vs DURABILITY
 * ============================================================
 *
 * DURABILITY:
 *
 *     "Is data preserved even when failures happen?"
 *
 *
 * Example:
 *
 *     Database writes data
 *          ↓
 *     Disk fails
 *          ↓
 *     Data still exists on another copy
 *
 *
 *     Durability -> YES
 *
 *
 * IMPORTANT:
 *
 * A system can be:
 *
 *     Fault-tolerant
 *     BUT
 *     Not durable
 *
 *
 * Example:
 *
 *     Database continues operating after failover
 *     BUT some writes are lost.
 *
 *
 * ============================================================
 * 5. FOUR RELATED CONCEPTS
 * ============================================================
 *
 *
 * Availability
 *     -> Is the system responding?
 *
 *
 * Reliability
 *     -> Is the response correct?
 *
 *
 * Fault Tolerance
 *     -> Does it keep working when components fail?
 *
 *
 * Durability
 *     -> Is data preserved despite failures?
 *
 *
 * Remember:
 *
 *     Availability = UP
 *     Reliability  = CORRECT
 *     Fault Tolerance = SURVIVES FAILURE
 *     Durability = DATA SURVIVES
 *
 *
 * ============================================================
 * 6. THESE PROPERTIES ARE INDEPENDENT
 * ============================================================
 *
 * You can optimize one property
 * while accidentally hurting another.
 *
 *
 * Example:
 *
 *     Aggressive caching
 *          ↓
 *     Better availability
 *          ↓
 *     But stale data may be returned
 *          ↓
 *     Reliability decreases
 *
 *
 * Therefore:
 *
 *     Improving availability does not automatically
 *     improve reliability.
 *
 *
 * ============================================================
 * 7. MEASURING RELIABILITY
 * ============================================================
 *
 * Important reliability metrics:
 *
 *     1. MTBF
 *     2. MTTR
 *     3. Error Rate
 *     4. Data Correctness
 *
 *
 * ============================================================
 * 8. MTBF — MEAN TIME BETWEEN FAILURES
 * ============================================================
 *
 * MTBF tells us:
 *
 *     "How frequently does the system fail?"
 *
 *
 * Formula:
 *
 *
 *     MTBF =
 *
 *     Total Operating Time
 *     --------------------
 *     Number of Failures
 *
 *
 * Higher MTBF:
 *
 *     -> Failures happen less frequently.
 *
 *
 * Example:
 *
 *     System operated:
 *
 *         10,000 hours
 *
 *     Number of failures:
 *
 *         5
 *
 *
 *     MTBF =
 *
 *         10,000 / 5
 *
 *         = 2,000 hours
 *
 *
 * Meaning:
 *
 *     On average, one failure occurs
 *     approximately every 2,000 hours.
 *
 *
 * 2,000 hours ≈ 83 days.
 *
 *
 * ============================================================
 * 9. MTBF AT LARGE SCALE
 * ============================================================
 *
 * Suppose:
 *
 *     One server:
 *
 *         MTBF = 2,000 hours
 *
 *
 * Now suppose we have:
 *
 *     100 servers
 *
 *
 * More machines means more opportunities
 * for something to fail.
 *
 *
 * Therefore:
 *
 *     Large fleets experience failures
 *     much more frequently.
 *
 *
 * IMPORTANT:
 *
 * At small scale:
 *
 *     Server failure may feel rare.
 *
 * At large scale:
 *
 *     Hardware failures become normal events.
 *
 *
 * ============================================================
 * 10. MTTR — MEAN TIME TO RECOVERY
 * ============================================================
 *
 * MTTR tells us:
 *
 *     "How quickly can we recover after a failure?"
 *
 *
 * Formula:
 *
 *
 *     MTTR =
 *
 *     Total Downtime
 *     --------------
 *     Number of Failures
 *
 *
 * Lower MTTR:
 *
 *     -> Faster recovery
 *     -> Less downtime
 *
 *
 * Example:
 *
 *     Number of failures = 5
 *
 *     Total repair time = 10 hours
 *
 *
 *     MTTR =
 *
 *         10 / 5
 *
 *         = 2 hours
 *
 *
 * Average recovery time:
 *
 *     2 hours per failure.
 *
 *
 * ============================================================
 * 11. WHAT DOES MTTR INCLUDE?
 * ============================================================
 *
 * MTTR is not just repair time.
 *
 * It includes:
 *
 *     1. Detection time
 *     2. Diagnosis time
 *     3. Repair time
 *     4. Verification time
 *
 *
 * Failure:
 *
 *     System fails
 *        ↓
 *     Detect
 *        ↓
 *     Understand problem
 *        ↓
 *     Fix problem
 *        ↓
 *     Verify system
 *        ↓
 *     Recover
 *
 *
 * IMPORTANT:
 *
 * If failures cannot be completely prevented,
 * recovering quickly is extremely important.
 *
 *
 * ============================================================
 * 12. MTBF vs MTTR
 * ============================================================
 *
 *
 * MTBF:
 *
 *     How often do we fail?
 *
 *
 * MTTR:
 *
 *     How quickly do we recover?
 *
 *
 * Example:
 *
 *     System A:
 *
 *     MTBF = 1 year
 *     MTTR = 10 hours
 *
 *
 * System B:
 *
 *     MTBF = 1 year
 *     MTTR = 10 minutes
 *
 *
 * Both fail equally often.
 *
 * But System B recovers much faster.
 *
 *
 * KEY IDEA:
 *
 *     Reliability is not only about
 *     preventing failures.
 *
 *     Recovery speed also matters.
 *
 *
 * ============================================================
 * 13. ERROR RATE
 * ============================================================
 *
 * Error rate tells us:
 *
 *     "What percentage of requests fail?"
 *
 *
 * Formula:
 *
 *
 *     Error Rate =
 *
 *     Failed Requests
 *     ---------------
 *     Total Requests
 *
 *     × 100%
 *
 *
 * Example:
 *
 *     Total requests = 100,000
 *
 *     Failed requests = 100
 *
 *
 *     Error Rate =
 *
 *         100 / 100,000 × 100
 *
 *         = 0.1%
 *
 *
 * ============================================================
 * 14. ERROR RATE TARGETS
 * ============================================================
 *
 *
 * Critical systems:
 *
 *     < 0.01%
 *
 *     Approximately:
 *
 *     1 failure / 10,000 requests
 *
 *
 * Standard systems:
 *
 *     < 0.1%
 *
 *     Approximately:
 *
 *     1 failure / 1,000 requests
 *
 *
 * Tolerant systems:
 *
 *     < 1%
 *
 *     Approximately:
 *
 *     1 failure / 100 requests
 *
 *
 * IMPORTANT:
 *
 * Error rate tells us about failed requests.
 *
 * It does NOT tell us whether successful
 * responses contain correct data.
 *
 *
 * ============================================================
 * 15. DATA CORRECTNESS
 * ============================================================
 *
 * Data correctness measures:
 *
 *     "How many responses contain correct data?"
 *
 *
 * Formula:
 *
 *
 *     Correctness =
 *
 *     Correct Responses
 *     -----------------
 *     Total Responses
 *
 *     × 100%
 *
 *
 * Example:
 *
 *     100,000 responses
 *
 *     99,000 correct
 *
 *
 *     Correctness =
 *
 *         99%
 *
 *
 * ============================================================
 * 16. WHY DATA CORRECTNESS MATTERS
 * ============================================================
 *
 * A system can have:
 *
 *     99.99% availability
 *
 *     0.01% error rate
 *
 * BUT
 *
 *     1% of successful responses
 *     contain incorrect data.
 *
 *
 * Users received a response.
 *
 * But the response was wrong.
 *
 *
 * Therefore:
 *
 *     System is available
 *     BUT
 *     system is not reliable.
 *
 *
 * KEY IDEA:
 *
 *     "Success response" does not always mean
 *     "correct response."
 *
 *
 * ============================================================
 * 17. WHY SYSTEMS BECOME UNRELIABLE
 * ============================================================
 *
 * Major causes:
 *
 *     1. Hardware failures
 *     2. Software bugs
 *     3. Configuration errors
 *     4. Human errors
 *     5. Overload
 *     6. Cascading failures
 *
 *
 * ============================================================
 * 18. HARDWARE FAILURES
 * ============================================================
 *
 * Physical components can fail over time.
 *
 *
 * Examples:
 *
 *     Disk bad sectors
 *     Memory corruption
 *     CPU overheating
 *     Network card failure
 *     Power supply failure
 *
 *
 * At large scale:
 *
 *     Hardware failures become routine.
 *
 *
 * Therefore:
 *
 *     System should assume that
 *     individual hardware components
 *     can fail.
 *
 *
 * ============================================================
 * 19. SOFTWARE BUGS
 * ============================================================
 *
 * Software bugs are a major source
 * of reliability problems.
 *
 *
 * Hardware failures:
 *
 *     Usually random/independent.
 *
 *
 * Software bugs:
 *
 *     Often systematic.
 *
 *
 * Example:
 *
 *     Buggy code path
 *          ↓
 *     Every request using that path
 *          ↓
 *     Same incorrect result
 *
 *
 * Most dangerous bugs:
 *
 *     Bugs that do NOT crash the system
 *     but silently produce wrong results.
 *
 *
 * Example:
 *
 *     Payment:
 *
 *     Expected:
 *         $100
 *
 *     Actual:
 *         $90
 *
 * System remains online,
 * but the result is incorrect.
 *
 *
 * ============================================================
 * 20. CONFIGURATION ERRORS
 * ============================================================
 *
 * Configuration mistakes can cause major failures.
 *
 *
 * Example:
 *
 *     Incorrect configuration
 *          ↓
 *     Service behaves incorrectly
 *          ↓
 *     Entire service may fail
 *
 *
 * Important lesson:
 *
 *     Small configuration mistakes
 *     can have very large consequences.
 *
 *
 * ============================================================
 * 21. HUMAN ERROR
 * ============================================================
 *
 * Common examples:
 *
 *     Wrong command
 *     Untested deployment
 *     Wrong configuration
 *     Incorrect diagnosis
 *
 *
 * Production environments are especially risky
 * because engineers may be working under pressure.
 *
 *
 * Therefore reliable systems should reduce
 * the impact of human mistakes using:
 *
 *     Automation
 *     Testing
 *     Monitoring
 *     Safe deployment processes
 *
 *
 * ============================================================
 * 22. OVERLOAD + CASCADING FAILURE
 * ============================================================
 *
 * A system can work perfectly under normal load
 * but fail under heavy load.
 *
 *
 * Example:
 *
 *     Traffic increases
 *          ↓
 *     Component becomes slow
 *          ↓
 *     Requests start waiting
 *          ↓
 *     Timeouts occur
 *          ↓
 *     Requests retry
 *          ↓
 *     More load
 *          ↓
 *     System becomes even slower
 *          ↓
 *     Failure spreads
 *
 *
 * This is a cascading failure.
 *
 *
 * IMPORTANT:
 *
 * Retries can sometimes make an overload
 * problem worse.
 *
 *
 * ============================================================
 * 23. KEY PRINCIPLES OF RELIABLE SYSTEMS
 * ============================================================
 *
 * The chapter focuses on:
 *
 *     1. Redundancy
 *     2. Failover mechanisms
 *     3. Load balancing
 *     4. Monitoring and alerting
 *     5. Graceful degradation
 *
 *
 * ============================================================
 * 24. REDUNDANCY
 * ============================================================
 *
 * Redundancy means:
 *
 *     Have backup components ready
 *     to take over when one fails.
 *
 *
 * Examples:
 *
 *     Multiple application servers
 *     Duplicate network paths
 *     Backup databases
 *
 *
 * Example:
 *
 *     Users
 *        |
 *        v
 *     Load Balancer
 *       / | \
 *      /  |  \
 *     v   v   v
 *    App App App
 *     1   2   3
 *
 *
 * If App Server 1 fails:
 *
 *     Load Balancer
 *     sends traffic to:
 *
 *         App 2
 *         App 3
 *
 *
 * ============================================================
 * 25. FAILOVER MECHANISM
 * ============================================================
 *
 * Failover means:
 *
 *     Automatically switching to a redundant
 *     or standby component after failure.
 *
 *
 * Example:
 *
 *     Primary DB
 *         |
 *       FAIL
 *         ↓
 *     Standby DB
 *         |
 *      becomes
 *       Primary
 *
 *
 * Goal:
 *
 *     Continue operation
 *     with minimal disruption.
 *
 *
 * ============================================================
 * 26. LOAD BALANCING
 * ============================================================
 *
 * Load balancer distributes traffic
 * across multiple servers.
 *
 *
 *     Users
 *       |
 *       v
 *     Load Balancer
 *      /   |   \
 *     v    v    v
 *    S1   S2   S3
 *
 *
 * Benefits:
 *
 *     1. Distributes traffic
 *     2. Improves performance
 *     3. Prevents one server from
 *        becoming a single point of failure
 *
 *
 * If S1 fails:
 *
 *     Load Balancer
 *     routes traffic to S2/S3.
 *
 *
 * ============================================================
 * 27. MONITORING + ALERTING
 * ============================================================
 *
 * A reliable system needs continuous monitoring.
 *
 *
 * Monitor:
 *
 *     System health
 *     Performance
 *     Errors
 *     Failures
 *     Resource usage
 *
 *
 * Alerting:
 *
 *     Notify engineers when problems occur.
 *
 *
 * Flow:
 *
 *     System
 *        ↓
 *     Monitoring
 *        ↓
 *     Detect abnormal behavior
 *        ↓
 *     Alert
 *        ↓
 *     Engineer responds
 *
 *
 * Goal:
 *
 *     Detect problems before they become
 *     major incidents.
 *
 *
 * ============================================================
 * 28. GRACEFUL DEGRADATION
 * ============================================================
 *
 * Graceful degradation means:
 *
 *     When some components fail,
 *     keep core functionality working.
 *
 *
 * Instead of:
 *
 *     Everything fails
 *
 *
 * We do:
 *
 *     Full Service
 *          ↓
 *     Partial Service
 *          ↓
 *     Core Only
 *          ↓
 *     Emergency Mode
 *
 *
 * ============================================================
 * 29. E-COMMERCE DEGRADATION EXAMPLE
 * ============================================================
 *
 *
 * FULL SERVICE
 *
 *     Personalized recommendations
 *     Real-time inventory
 *     Multiple payment options
 *
 *
 *        ↓ Load increases / failures
 *
 *
 * PARTIAL SERVICE
 *
 *     Generic recommendations
 *     Cached inventory
 *     Primary payment options
 *
 *
 *        ↓ More failures
 *
 *
 * CORE ONLY
 *
 *     Browse products
 *     Checkout
 *     Basic payment
 *
 *
 *        ↓ Critical failure
 *
 *
 * EMERGENCY MODE
 *
 *     Cached product pages
 *     Accept orders for later processing
 *
 *
 * IMPORTANT:
 *
 * A reduced service is often better than
 * complete system failure.
 *
 *
 * ============================================================
 * 30. CIRCUIT BREAKER
 * ============================================================
 *
 * Circuit breaker is especially useful
 * in microservices.
 *
 *
 * Problem:
 *
 *     Service A
 *        |
 *        v
 *     Service B
 *        |
 *     B becomes slow
 *        ↓
 *     A waits
 *        ↓
 *     More requests wait
 *        ↓
 *     Resources exhausted
 *        ↓
 *     Failure spreads
 *
 *
 * Circuit breaker prevents this cascade.
 *
 *
 * It:
 *
 *     Detects repeated failures
 *          ↓
 *     Temporarily stops calls
 *          ↓
 *     Gives dependency time to recover
 *
 *
 * ============================================================
 * 31. CIRCUIT BREAKER STATES
 * ============================================================
 *
 * Three states:
 *
 *     CLOSED
 *     OPEN
 *     HALF-OPEN
 *
 *
 * ============================================================
 * 32. CLOSED STATE
 * ============================================================
 *
 * Normal operation.
 *
 *
 *     Request
 *        |
 *        v
 *     Circuit
 *        |
 *        v
 *     Dependency
 *
 *
 * Requests are allowed through.
 *
 * Failures are counted.
 *
 *
 * Example:
 *
 *     5 failures in 30 seconds
 *
 *     Failure threshold exceeded
 *
 *             ↓
 *
 *     CLOSED -> OPEN
 *
 *
 * ============================================================
 * 33. OPEN STATE
 * ============================================================
 *
 * Dependency is considered unhealthy.
 *
 *
 * Requests do NOT call the dependency.
 *
 *
 * Instead:
 *
 *     Request
 *        |
 *        v
 *     Circuit Breaker
 *        |
 *       OPEN
 *        |
 *        v
 *     Fail immediately / fallback
 *
 *
 * Why?
 *
 *     Prevent wasting resources
 *     waiting for a broken dependency.
 *
 *
 * Also:
 *
 *     Gives dependency time to recover.
 *
 *
 * ============================================================
 * 34. HALF-OPEN STATE
 * ============================================================
 *
 * After a timeout:
 *
 *     OPEN
 *       ↓
 *     HALF-OPEN
 *
 *
 * Only a limited number of test requests
 * are allowed through.
 *
 *
 * If tests succeed:
 *
 *     HALF-OPEN
 *         ↓
 *       CLOSED
 *
 *
 * If tests fail:
 *
 *     HALF-OPEN
 *         ↓
 *        OPEN
 *
 *
 * ============================================================
 * 35. CIRCUIT BREAKER FLOW
 * ============================================================
 *
 *
 *                 CLOSED
 *                    |
 *          failure threshold
 *               exceeded
 *                    |
 *                    v
 *                  OPEN
 *                    |
 *              timeout expires
 *                    |
 *                    v
 *               HALF-OPEN
 *                 /     \
 *                /       \
 *           success       failure
 *              |             |
 *              v             v
 *           CLOSED          OPEN
 *
 *
 * Example:
 *
 *     5 failures in 30 seconds
 *          ↓
 *     Circuit opens
 *          ↓
 *     Wait 30 seconds
 *          ↓
 *     Send test request
 *          ↓
 *     Success -> Close circuit
 *     Failure -> Open circuit again
 *
 *
 * ============================================================
 * 36. IDEMPOTENCY
 * ============================================================
 *
 * Idempotency is one of the most important
 * reliability concepts.
 *
 *
 * Problem:
 *
 *     Network failure makes it unclear
 *     whether a request succeeded.
 *
 *
 * Example:
 *
 *     Client
 *       |
 *       | Transfer $100
 *       v
 *     Server
 *       |
 *       v
 *     Database
 *
 * Server processes transfer:
 *
 *     $100 transferred
 *
 * But response is lost.
 *
 *
 * Client thinks:
 *
 *     "Maybe request failed."
 *
 *
 * Client retries:
 *
 *     Transfer $100
 *
 *
 * Without idempotency:
 *
 *     $100 + $100
 *
 *     = $200 transferred
 *
 *
 * This is dangerous for:
 *
 *     Payments
 *     Money transfers
 *     Orders
 *     Other important operations
 *
 *
 * ============================================================
 * 37. WHAT IS AN IDEMPOTENT OPERATION?
 * ============================================================
 *
 * Idempotent operation:
 *
 *     Executing the same operation multiple times
 *     produces the same final result as executing
 *     it once.
 *
 *
 * Example:
 *
 *     Set balance = $100
 *
 * Execute once:
 *
 *     $100
 *
 * Execute again:
 *
 *     $100
 *
 * Execute again:
 *
 *     $100
 *
 *
 * Final result remains:
 *
 *     $100
 *
 *
 * ============================================================
 * 38. IDEMPOTENCY KEY
 * ============================================================
 *
 * Client sends:
 *
 *     Transfer $100
 *
 *     Idempotency-Key:
 *
 *         abc123
 *
 *
 * Server receives request.
 *
 *
 * First request:
 *
 *     Check key:
 *         abc123 does not exist
 *
 *     Execute transfer
 *
 *     Store:
 *
 *         abc123 -> result
 *
 *
 * Response is lost.
 *
 *
 * Client retries:
 *
 *     Transfer $100
 *
 *     Idempotency-Key:
 *
 *         abc123
 *
 *
 * Server checks:
 *
 *     abc123 already exists.
 *
 *
 * Therefore:
 *
 *     DO NOT execute transfer again.
 *
 *     Return stored result.
 *
 *
 * Result:
 *
 *     Only $100 transferred.
 *
 *     NOT $200.
 *
 *
 * ============================================================
 * 39. IDEMPOTENCY FLOW
 * ============================================================
 *
 *
 * CLIENT
 *    |
 *    | Transfer $100
 *    | ID = abc123
 *    v
 * SERVER
 *    |
 *    | Check abc123
 *    |
 *    | Doesn't exist
 *    v
 * DATABASE
 *    |
 *    | Store transfer
 *    | Store idempotency key
 *    v
 * SUCCESS
 *
 *
 * Response gets lost.
 *
 *
 * CLIENT
 *    |
 *    | Retry Transfer $100
 *    | ID = abc123
 *    v
 * SERVER
 *    |
 *    | Check abc123
 *    |
 *    | Already exists
 *    v
 * Return stored result
 *
 *
 * Transfer happens only once.
 *
 *
 * ============================================================
 * 40. WHERE IDEMPOTENCY IS IMPORTANT
 * ============================================================
 *
 * Especially important for:
 *
 *     Payment systems
 *     Money transfers
 *     Order creation
 *     Financial transactions
 *
 *
 * The source specifically notes that payment processors
 * such as Stripe and PayPal use idempotency keys for
 * money-moving operations.
 *
 *
 * Main problem solved:
 *
 *     Network retry
 *          +
 *     Non-idempotent operation
 *          =
 *     Duplicate effect
 *
 *
 * Idempotency prevents this.
 *
 *
 * ============================================================
 * 41. RELIABILITY TECHNIQUE SUMMARY
 * ============================================================
 *
 *
 * REDUNDANCY
 *
 *     Problem:
 *         Component failure
 *
 *     Solution:
 *         Backup components
 *
 *
 * FAILOVER
 *
 *     Problem:
 *         Active component fails
 *
 *     Solution:
 *         Switch to redundant component
 *
 *
 * LOAD BALANCING
 *
 *     Problem:
 *         Single server overload/failure
 *
 *     Solution:
 *         Distribute traffic
 *
 *
 * MONITORING
 *
 *     Problem:
 *         Failure goes unnoticed
 *
 *     Solution:
 *         Continuously monitor system
 *
 *
 * ALERTING
 *
 *     Problem:
 *         Engineers don't know about failure
 *
 *     Solution:
 *         Notify engineers
 *
 *
 * GRACEFUL DEGRADATION
 *
 *     Problem:
 *         Optional feature failure
 *
 *     Solution:
 *         Keep core functionality working
 *
 *
 * CIRCUIT BREAKER
 *
 *     Problem:
 *         Failing dependency causes cascade
 *
 *     Solution:
 *         Stop calling failing dependency
 *
 *
 * IDEMPOTENCY
 *
 *     Problem:
 *         Retry causes duplicate operation
 *
 *     Solution:
 *         Use idempotency key
 *
 *
 * ============================================================
 * 42. RELIABILITY vs AVAILABILITY — INTERVIEW EXAMPLE
 * ============================================================
 *
 *
 * Imagine a payment service.
 *
 *
 * CASE 1:
 *
 *     API is down.
 *
 *     User cannot make payment.
 *
 *     Availability problem.
 *
 *
 * CASE 2:
 *
 *     API responds successfully
 *     but charges customer twice.
 *
 *     Reliability problem.
 *
 *
 * CASE 3:
 *
 *     Database primary fails
 *     but replica takes over.
 *
 *     Fault tolerance.
 *
 *
 * CASE 4:
 *
 *     Database fails
 *     but replicated data survives.
 *
 *     Durability.
 *
 *
 * ============================================================
 * 43. RELIABILITY MINDSET
 * ============================================================
 *
 * When designing any feature,
 * ask these questions:
 *
 *
 *     1. What if the dependency fails?
 *
 *     2. What if the dependency becomes slow?
 *
 *     3. What if the response is incorrect?
 *
 *     4. What if the request succeeds
 *        but the response is lost?
 *
 *     5. What happens when we retry?
 *
 *     6. Can the operation happen twice?
 *
 *     7. Can we safely fail over?
 *
 *     8. Can we continue with reduced functionality?
 *
 *     9. How will we detect the problem?
 *
 *     10. Can the system recover automatically?
 *
 *
 * ============================================================
 * 44. RELIABILITY FAILURE FLOW
 * ============================================================
 *
 *
 *              FAILURE
 *                 |
 *        +--------+--------+
 *        |        |        |
 *      Slow     Wrong    Down
 *        |       Data      |
 *        |        |        |
 *        v        v        v
 *     Timeout  Validate  Failover
 *        |        |        |
 *        v        v        v
 *     Circuit  Correct   Redundant
 *     Breaker  / Reject  Component
 *        |                 |
 *        +--------+--------+
 *                 |
 *                 v
 *          Graceful Recovery
 *
 *
 * ============================================================
 * 45. FINAL RELIABILITY CHECKLIST
 * ============================================================
 *
 *
 * BEFORE DESIGNING:
 *
 *     -> Define what "correct" means.
 *     -> Identify critical operations.
 *     -> Identify possible failure modes.
 *
 *
 * DURING DESIGN:
 *
 *     -> Add redundancy.
 *     -> Add failover.
 *     -> Use load balancing.
 *     -> Replicate important data.
 *     -> Add monitoring.
 *     -> Add alerting.
 *     -> Design graceful degradation.
 *     -> Use circuit breakers.
 *     -> Make important operations idempotent.
 *
 *
 * AFTER DESIGN:
 *
 *     -> Measure MTBF.
 *     -> Measure MTTR.
 *     -> Measure error rate.
 *     -> Measure data correctness.
 *     -> Test failover.
 *     -> Test recovery.
 *
 *
 * IMPORTANT:
 *
 * An untested failover mechanism can create
 * a false sense of safety.
 *
 *
 * ============================================================
 * 46. QUICK REVISION
 * ============================================================
 *
 * Reliability
 *     -> System performs its intended function correctly
 *        over time.
 *
 * Availability
 *     -> System is responding.
 *
 * Fault Tolerance
 *     -> System continues working when components fail.
 *
 * Durability
 *     -> Data survives failures.
 *
 *
 * MTBF
 *     -> How often the system fails.
 *
 *     MTBF =
 *     Operating Time / Number of Failures
 *
 *
 * MTTR
 *     -> How quickly the system recovers.
 *
 *     MTTR =
 *     Downtime / Number of Failures
 *
 *
 * Error Rate
 *     -> Percentage of failed requests.
 *
 *
 * Data Correctness
 *     -> Percentage of responses containing correct data.
 *
 *
 * Redundancy
 *     -> Backup components.
 *
 *
 * Failover
 *     -> Switch to backup component.
 *
 *
 * Load Balancing
 *     -> Distribute traffic across servers.
 *
 *
 * Graceful Degradation
 *     -> Keep core functionality working
 *        when optional parts fail.
 *
 *
 * Circuit Breaker
 *     -> Stop calling a failing dependency.
 *
 *
 * Idempotency
 *     -> Repeating the same operation does not
 *        create duplicate effects.
 *
 *
 * Idempotency Key
 *     -> Unique key used to recognize retries
 *        and prevent duplicate operations.
 *
 *
 * ============================================================
 * 47. MOST IMPORTANT INTERVIEW TAKEAWAYS
 * ============================================================
 *
 *
 * 1.
 *     Availability asks:
 *
 *         "Is it responding?"
 *
 *     Reliability asks:
 *
 *         "Is it responding correctly?"
 *
 *
 * 2.
 *     MTBF tells:
 *
 *         How frequently failures happen.
 *
 *
 *     MTTR tells:
 *
 *         How quickly we recover.
 *
 *
 * 3.
 *     Don't only measure whether requests succeed.
 *
 *     Also measure:
 *
 *         "Was the returned data correct?"
 *
 *
 * 4.
 *     Hardware, software bugs, configuration,
 *     human mistakes, and overload can all
 *     cause reliability problems.
 *
 *
 * 5.
 *     Redundancy + failover
 *     keeps systems running when components fail.
 *
 *
 * 6.
 *     Graceful degradation:
 *
 *         Optional feature fails
 *              ↓
 *         Core feature continues.
 *
 *
 * 7.
 *     Circuit breaker:
 *
 *         Dependency failing
 *              ↓
 *         Stop calling it
 *              ↓
 *         Prevent cascading failure.
 *
 *
 * 8.
 *     Idempotency:
 *
 *         Retry request
 *              ↓
 *         Same idempotency key
 *              ↓
 *         Detect duplicate
 *              ↓
 *         Don't execute twice.
 *
 *
 * ============================================================
 * CORE RELIABILITY PRINCIPLE
 * ============================================================
 *
 *
 *       EXPECT FAILURE
 *            ↓
 *       DETECT FAILURE
 *            ↓
 *       CONTAIN FAILURE
 *            ↓
 *       RECOVER SAFELY
 *            ↓
 *       VERIFY CORRECTNESS
 *            ↓
 *       CONTINUE SERVING USERS
 *
 *
 * The ultimate question in system design:
 *
 *     "If a dependency returns wrong data,
 *      becomes slow, or fails completely,
 *      what does the user see?
 *
 *      And can the system recover on its own?"
 *
 *
 * ============================================================
 */
```

```python
/**
 * RELIABILITY — SYSTEM DESIGN
 *
 * Core idea:
 *   Availability: "Is the system responding?"
 *   Reliability : "Is the system responding CORRECTLY?"
 *
 * A system can be available but unreliable.
 *
 * Example 1: Payment request succeeds, but customer is charged twice.
 *   Available -> Yes | Reliable -> No
 *
 * Example 2: Messaging app is available, but messages arrive out of order.
 *   Available -> Yes | Reliable -> No
 *
 * Reliability = system performs its intended function correctly
 * and consistently over time.
 */

/**
 * 1. WHAT IS RELIABILITY?
 *    The probability that a system will perform its intended function
 *    correctly over a given period of time under specified conditions.
 *
 *    Three parts:
 *      CORRECTLY                 : system must produce the correct result.
 *      OVER A GIVEN PERIOD       : measured over time, not at one moment.
 *      UNDER SPECIFIED CONDITIONS: we define what normal operating conditions are.
 *
 *    Simple definition: "Does the system do what it is supposed to do?"
 */

/**
 * 2. RELIABILITY vs AVAILABILITY
 *
 *    AVAILABILITY: "Is the system responding?"
 *      e.g. HTTP 200 response received.
 *
 *    RELIABILITY: "Is the response correct?"
 *      e.g. account balance returned is accurate.
 *
 *    IMPORTANT: Available != Reliable
 *      Payment API: request succeeds -> wrong amount charged
 *      Available -> YES | Reliable -> NO
 */

/**
 * 3. RELIABILITY vs FAULT TOLERANCE
 *    FAULT TOLERANCE: "Can the system continue working when a component fails?"
 *      Database Replica 1 FAILS -> Replica 2 takes over -> system continues.
 *
 *    Reliability     : is the system producing correct results?
 *    Fault Tolerance : can the system keep operating despite component failures?
 */

/**
 * 4. RELIABILITY vs DURABILITY
 *    DURABILITY: "Is data preserved even when failures happen?"
 *      Database writes data -> disk fails -> data still exists on another copy.
 *      Durability -> YES
 *
 *    IMPORTANT: a system can be fault-tolerant BUT not durable.
 *    Example: database keeps operating after failover BUT some writes are lost.
 */

/**
 * 5. FOUR RELATED CONCEPTS
 *
 *    Availability    -> Is the system responding?             (UP)
 *    Reliability     -> Is the response correct?              (CORRECT)
 *    Fault Tolerance -> Does it keep working when components fail? (SURVIVES FAILURE)
 *    Durability      -> Is data preserved despite failures?   (DATA SURVIVES)
 */

/**
 * 6. THESE PROPERTIES ARE INDEPENDENT
 *    You can optimize one property while accidentally hurting another.
 *
 *    Example:
 *      Aggressive caching -> better availability
 *      -> but stale data may be returned -> reliability decreases
 *
 *    Improving availability does not automatically improve reliability.
 */

/**
 * 7. MEASURING RELIABILITY
 *    Important metrics:
 *      1. MTBF
 *      2. MTTR
 *      3. Error Rate
 *      4. Data Correctness
 */

/**
 * 8. MTBF — MEAN TIME BETWEEN FAILURES
 *    "How frequently does the system fail?"
 *
 *    MTBF = Total Operating Time / Number of Failures
 *    Higher MTBF -> failures happen less frequently.
 *
 *    Example:
 *      Operated 10,000 hours, 5 failures
 *      MTBF = 10,000 / 5 = 2,000 hours (~83 days)
 *      On average, one failure roughly every 2,000 hours.
 */

/**
 * 9. MTBF AT LARGE SCALE
 *    One server: MTBF = 2,000 hours.
 *    With 100 servers there are more opportunities for something to fail,
 *    so large fleets experience failures much more frequently.
 *
 *    IMPORTANT:
 *      Small scale : server failure may feel rare.
 *      Large scale : hardware failures become normal events.
 */

/**
 * 10. MTTR — MEAN TIME TO RECOVERY
 *     "How quickly can we recover after a failure?"
 *
 *     MTTR = Total Downtime / Number of Failures
 *     Lower MTTR -> faster recovery, less downtime.
 *
 *     Example:
 *       5 failures, total repair time 10 hours
 *       MTTR = 10 / 5 = 2 hours per failure.
 */

/**
 * 11. WHAT DOES MTTR INCLUDE?
 *     MTTR is not just repair time. It includes:
 *       1. Detection time
 *       2. Diagnosis time
 *       3. Repair time
 *       4. Verification time
 *
 *     System fails -> Detect -> Understand problem -> Fix -> Verify -> Recover
 *
 *     IMPORTANT: if failures can't be completely prevented,
 *     recovering quickly is extremely important.
 */

/**
 * 12. MTBF vs MTTR
 *     MTBF: how often do we fail?
 *     MTTR: how quickly do we recover?
 *
 *     System A: MTBF = 1 year, MTTR = 10 hours
 *     System B: MTBF = 1 year, MTTR = 10 minutes
 *     Both fail equally often, but System B recovers much faster.
 *
 *     KEY IDEA: reliability is not only about preventing failures;
 *     recovery speed also matters.
 */

/**
 * 13. ERROR RATE
 *     "What percentage of requests fail?"
 *
 *     Error Rate = (Failed Requests / Total Requests) x 100%
 *
 *     Example:
 *       Total = 100,000, Failed = 100
 *       Error Rate = 100 / 100,000 x 100 = 0.1%
 */

/**
 * 14. ERROR RATE TARGETS
 *     Critical systems : < 0.01%  (~1 failure / 10,000 requests)
 *     Standard systems : < 0.1%   (~1 failure / 1,000 requests)
 *     Tolerant systems : < 1%     (~1 failure / 100 requests)
 *
 *     IMPORTANT: error rate tells us about failed requests.
 *     It does NOT tell us whether successful responses contain correct data.
 */

/**
 * 15. DATA CORRECTNESS
 *     "How many responses contain correct data?"
 *
 *     Correctness = (Correct Responses / Total Responses) x 100%
 *
 *     Example: 100,000 responses, 99,000 correct -> Correctness = 99%
 */

/**
 * 16. WHY DATA CORRECTNESS MATTERS
 *     A system can have 99.99% availability and 0.01% error rate,
 *     BUT 1% of successful responses contain incorrect data.
 *
 *     Users got a response, but it was wrong.
 *     System is available BUT not reliable.
 *
 *     KEY IDEA: "Success response" does not always mean "correct response."
 */

/**
 * 17. WHY SYSTEMS BECOME UNRELIABLE
 *     Major causes:
 *       1. Hardware failures
 *       2. Software bugs
 *       3. Configuration errors
 *       4. Human errors
 *       5. Overload
 *       6. Cascading failures
 */

/**
 * 18. HARDWARE FAILURES
 *     Physical components can fail over time.
 *     Examples: disk bad sectors, memory corruption, CPU overheating,
 *     network card failure, power supply failure.
 *
 *     At large scale hardware failures become routine, so the system should
 *     assume individual hardware components can fail.
 */

/**
 * 19. SOFTWARE BUGS
 *     A major source of reliability problems.
 *
 *     Hardware failures : usually random/independent.
 *     Software bugs     : often systematic.
 *
 *     Buggy code path -> every request using it -> same incorrect result
 *
 *     Most dangerous bugs: those that do NOT crash the system
 *     but silently produce wrong results.
 *
 *     Example: payment expected $100, actual $90.
 *     System stays online, but the result is incorrect.
 */

/**
 * 20. CONFIGURATION ERRORS
 *     Incorrect configuration -> service behaves incorrectly
 *     -> entire service may fail.
 *
 *     Lesson: small configuration mistakes can have very large consequences.
 */

/**
 * 21. HUMAN ERROR
 *     Common examples: wrong command, untested deployment,
 *     wrong configuration, incorrect diagnosis.
 *
 *     Production is especially risky because engineers may be working
 *     under pressure.
 *
 *     Reduce the impact of human mistakes using:
 *     automation, testing, monitoring, safe deployment processes.
 */

/**
 * 22. OVERLOAD + CASCADING FAILURE
 *     A system can work perfectly under normal load but fail under heavy load.
 *
 *       Traffic increases -> component becomes slow -> requests wait
 *       -> timeouts occur -> requests retry -> more load
 *       -> system becomes even slower -> failure spreads
 *
 *     This is a cascading failure.
 *
 *     IMPORTANT: retries can sometimes make an overload problem worse.
 */

/**
 * 23. KEY PRINCIPLES OF RELIABLE SYSTEMS
 *       1. Redundancy
 *       2. Failover mechanisms
 *       3. Load balancing
 *       4. Monitoring and alerting
 *       5. Graceful degradation
 */

/**
 * 24. REDUNDANCY
 *     Have backup components ready to take over when one fails.
 *     Examples: multiple app servers, duplicate network paths, backup databases.
 *
 *       Users --> Load Balancer --> App 1 / App 2 / App 3
 *
 *     If App 1 fails, load balancer sends traffic to App 2 and App 3.
 */

/**
 * 25. FAILOVER MECHANISM
 *     Automatically switch to a redundant or standby component after failure.
 *
 *       Primary DB FAILS -> Standby DB becomes Primary
 *
 *     Goal: continue operation with minimal disruption.
 */

/**
 * 26. LOAD BALANCING
 *     Distributes traffic across multiple servers.
 *
 *       Users --> Load Balancer --> S1 / S2 / S3
 *
 *     Benefits:
 *       1. Distributes traffic
 *       2. Improves performance
 *       3. Prevents one server from becoming a single point of failure
 *
 *     If S1 fails, load balancer routes traffic to S2/S3.
 */

/**
 * 27. MONITORING + ALERTING
 *     A reliable system needs continuous monitoring.
 *
 *     Monitor : system health, performance, errors, failures, resource usage.
 *     Alerting: notify engineers when problems occur.
 *
 *       System -> Monitoring -> Detect abnormal behavior
 *              -> Alert -> Engineer responds
 *
 *     Goal: detect problems before they become major incidents.
 */

/**
 * 28. GRACEFUL DEGRADATION
 *     When some components fail, keep core functionality working.
 *
 *     Instead of "everything fails":
 *       Full Service -> Partial Service -> Core Only -> Emergency Mode
 */

/**
 * 29. E-COMMERCE DEGRADATION EXAMPLE
 *
 *     FULL SERVICE
 *       Personalized recommendations, real-time inventory,
 *       multiple payment options
 *         | load increases / failures
 *         v
 *     PARTIAL SERVICE
 *       Generic recommendations, cached inventory, primary payment options
 *         | more failures
 *         v
 *     CORE ONLY
 *       Browse products, checkout, basic payment
 *         | critical failure
 *         v
 *     EMERGENCY MODE
 *       Cached product pages, accept orders for later processing
 *
 *     IMPORTANT: a reduced service is often better than complete failure.
 */

/**
 * 30. CIRCUIT BREAKER
 *     Especially useful in microservices.
 *
 *     Problem:
 *       Service A -> Service B; B becomes slow
 *       -> A waits -> more requests wait -> resources exhausted
 *       -> failure spreads
 *
 *     Circuit breaker prevents this cascade. It:
 *       detects repeated failures -> temporarily stops calls
 *       -> gives the dependency time to recover
 */

/**
 * 31. CIRCUIT BREAKER STATES
 *     Three states: CLOSED, OPEN, HALF-OPEN
 *
 *     CLOSED
 *       Normal operation: Request -> Circuit -> Dependency.
 *       Requests allowed through, failures are counted.
 *       Example: 5 failures in 30 seconds -> threshold exceeded
 *       -> CLOSED -> OPEN.
 *
 *     OPEN
 *       Dependency considered unhealthy.
 *       Requests do NOT call the dependency:
 *         Request -> Circuit Breaker (OPEN) -> fail immediately / fallback
 *       Why? Avoid wasting resources waiting for a broken dependency,
 *       and give the dependency time to recover.
 *
 *     HALF-OPEN
 *       After a timeout: OPEN -> HALF-OPEN.
 *       Only a limited number of test requests are allowed through.
 *       Tests succeed -> CLOSED.
 *       Tests fail    -> OPEN.
 */

/**
 * 32. CIRCUIT BREAKER FLOW
 *
 *                CLOSED
 *                   |
 *         failure threshold exceeded
 *                   v
 *                 OPEN
 *                   |
 *            timeout expires
 *                   v
 *              HALF-OPEN
 *               /       \
 *          success    failure
 *             v          v
 *          CLOSED      OPEN
 *
 *     Example:
 *       5 failures in 30 seconds -> circuit opens
 *       -> wait 30 seconds -> send test request
 *       -> Success: close circuit | Failure: open circuit again
 */

/**
 * 33. IDEMPOTENCY
 *     One of the most important reliability concepts.
 *
 *     Problem: network failure makes it unclear whether a request succeeded.
 *
 *     Example:
 *       Client --(Transfer $100)--> Server --> Database
 *       Server processes the transfer ($100 transferred), but the response is lost.
 *       Client thinks "maybe the request failed" and retries: Transfer $100.
 *
 *     Without idempotency: $100 + $100 = $200 transferred.
 *
 *     Dangerous for: payments, money transfers, orders, other important operations.
 */

/**
 * 34. WHAT IS AN IDEMPOTENT OPERATION?
 *     Executing the same operation multiple times produces the same
 *     final result as executing it once.
 *
 *     Example: "Set balance = $100"
 *       Execute once: $100 | again: $100 | again: $100
 *       Final result remains $100.
 */

/**
 * 35. IDEMPOTENCY KEY
 *     Client sends: Transfer $100, Idempotency-Key: abc123
 *
 *     First request:
 *       Check key: abc123 does not exist
 *       -> execute transfer -> store abc123 -> result
 *       Response is lost.
 *
 *     Client retries: Transfer $100, Idempotency-Key: abc123
 *       Server checks: abc123 already exists
 *       -> DO NOT execute transfer again; return stored result.
 *
 *     Result: only $100 transferred, NOT $200.
 */

/**
 * 36. IDEMPOTENCY FLOW
 *
 *     CLIENT --(Transfer $100, ID = abc123)--> SERVER
 *       SERVER checks abc123 -> doesn't exist
 *       -> DATABASE: store transfer + store idempotency key -> SUCCESS
 *
 *     Response gets lost.
 *
 *     CLIENT --(Retry Transfer $100, ID = abc123)--> SERVER
 *       SERVER checks abc123 -> already exists
 *       -> return stored result
 *
 *     Transfer happens only once.
 */

/**
 * 37. WHERE IDEMPOTENCY IS IMPORTANT
 *     Especially for: payment systems, money transfers, order creation,
 *     financial transactions.
 *
 *     Payment processors such as Stripe and PayPal use idempotency keys
 *     for money-moving operations.
 *
 *     Main problem solved:
 *       Network retry + Non-idempotent operation = Duplicate effect
 *     Idempotency prevents this.
 */

/**
 * 38. RELIABILITY TECHNIQUE SUMMARY
 *
 *     REDUNDANCY           : Component failure              -> Backup components
 *     FAILOVER             : Active component fails         -> Switch to redundant component
 *     LOAD BALANCING       : Single server overload/failure -> Distribute traffic
 *     MONITORING           : Failure goes unnoticed         -> Continuously monitor system
 *     ALERTING             : Engineers don't know           -> Notify engineers
 *     GRACEFUL DEGRADATION : Optional feature failure       -> Keep core functionality working
 *     CIRCUIT BREAKER      : Failing dependency cascades    -> Stop calling failing dependency
 *     IDEMPOTENCY          : Retry causes duplicate op      -> Use idempotency key
 */

/**
 * 39. RELIABILITY vs AVAILABILITY — INTERVIEW EXAMPLE (payment service)
 *
 *     CASE 1: API is down, user can't pay.
 *             -> Availability problem.
 *     CASE 2: API responds successfully but charges customer twice.
 *             -> Reliability problem.
 *     CASE 3: Database primary fails but replica takes over.
 *             -> Fault tolerance.
 *     CASE 4: Database fails but replicated data survives.
 *             -> Durability.
 */

/**
 * 40. RELIABILITY MINDSET
 *     When designing any feature, ask:
 *       1.  What if the dependency fails?
 *       2.  What if the dependency becomes slow?
 *       3.  What if the response is incorrect?
 *       4.  What if the request succeeds but the response is lost?
 *       5.  What happens when we retry?
 *       6.  Can the operation happen twice?
 *       7.  Can we safely fail over?
 *       8.  Can we continue with reduced functionality?
 *       9.  How will we detect the problem?
 *       10. Can the system recover automatically?
 */

/**
 * 41. RELIABILITY FAILURE FLOW
 *
 *              FAILURE
 *                 |
 *        +--------+--------+
 *        |        |        |
 *      Slow     Wrong     Down
 *        |       Data       |
 *        v        v         v
 *     Timeout  Validate   Failover
 *        |        |         |
 *        v        v         v
 *     Circuit   Correct   Redundant
 *     Breaker   / Reject  Component
 *        |                  |
 *        +--------+---------+
 *                 v
 *          Graceful Recovery
 */

/**
 * 42. FINAL RELIABILITY CHECKLIST
 *
 *     BEFORE DESIGNING:
 *       -> Define what "correct" means.
 *       -> Identify critical operations.
 *       -> Identify possible failure modes.
 *
 *     DURING DESIGN:
 *       -> Add redundancy and failover.
 *       -> Use load balancing.
 *       -> Replicate important data.
 *       -> Add monitoring and alerting.
 *       -> Design graceful degradation.
 *       -> Use circuit breakers.
 *       -> Make important operations idempotent.
 *
 *     AFTER DESIGN:
 *       -> Measure MTBF, MTTR, error rate, data correctness.
 *       -> Test failover.
 *       -> Test recovery.
 *
 *     IMPORTANT: an untested failover mechanism can create
 *     a false sense of safety.
 */

/**
 * 43. QUICK REVISION
 *
 *   Reliability      -> System performs its intended function correctly over time.
 *   Availability     -> System is responding.
 *   Fault Tolerance  -> System continues working when components fail.
 *   Durability       -> Data survives failures.
 *   MTBF             -> How often the system fails.
 *                       = Operating Time / Number of Failures
 *   MTTR             -> How quickly the system recovers.
 *                       = Downtime / Number of Failures
 *   Error Rate       -> Percentage of failed requests.
 *   Data Correctness -> Percentage of responses containing correct data.
 *   Redundancy       -> Backup components.
 *   Failover         -> Switch to backup component.
 *   Load Balancing   -> Distribute traffic across servers.
 *   Graceful Degradation -> Keep core functionality working when optional parts fail.
 *   Circuit Breaker  -> Stop calling a failing dependency.
 *   Idempotency      -> Repeating the same operation doesn't create duplicate effects.
 *   Idempotency Key  -> Unique key used to recognize retries and prevent duplicates.
 */

/**
 * 44. MOST IMPORTANT INTERVIEW TAKEAWAYS
 *
 *   1. Availability asks "Is it responding?"
 *      Reliability asks  "Is it responding correctly?"
 *
 *   2. MTBF tells how frequently failures happen.
 *      MTTR tells how quickly we recover.
 *
 *   3. Don't only measure whether requests succeed.
 *      Also measure: "Was the returned data correct?"
 *
 *   4. Hardware, software bugs, configuration, human mistakes and overload
 *      can all cause reliability problems.
 *
 *   5. Redundancy + failover keep systems running when components fail.
 *
 *   6. Graceful degradation:
 *      Optional feature fails -> core feature continues.
 *
 *   7. Circuit breaker:
 *      Dependency failing -> stop calling it -> prevent cascading failure.
 *
 *   8. Idempotency:
 *      Retry request -> same idempotency key -> detect duplicate
 *      -> don't execute twice.
 */

/**
 * CORE RELIABILITY PRINCIPLE
 *
 *   EXPECT FAILURE -> DETECT FAILURE -> CONTAIN FAILURE
 *   -> RECOVER SAFELY -> VERIFY CORRECTNESS -> CONTINUE SERVING USERS
 *
 *   The ultimate question in system design:
 *     "If a dependency returns wrong data, becomes slow, or fails completely,
 *      what does the user see? And can the system recover on its own?"
 */
```
