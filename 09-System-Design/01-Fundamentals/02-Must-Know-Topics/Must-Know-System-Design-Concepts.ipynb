```python
/**
 * ============================================================
 * 30 MUST-KNOW SYSTEM DESIGN CONCEPTS
 * ============================================================
 *
 * Goal:
 * Understand the core building blocks used repeatedly in
 * real-world system design and system design interviews.
 *
 * The concepts are divided into 6 groups:
 *
 * 1. Networking Foundations
 * 2. APIs & Communication
 * 3. Data Storage
 * 4. Scaling
 * 5. Distributed Systems
 * 6. Architecture Patterns
 *
 *
 * ============================================================
 * GROUP 1: NETWORKING FOUNDATIONS
 * ============================================================
 *
 *
 * 1. CLIENT-SERVER MODEL
 * ============================================================
 *
 * Basic communication model:
 *
 *     Client
 *       |
 *       | HTTP Request
 *       v
 *     Server
 *       |
 *       | HTTP Response
 *       v
 *     Client
 *
 * Client:
 * - Browser
 * - Mobile App
 * - Another Server
 *
 * Server:
 * - Receives requests
 * - Processes business logic
 * - Sends responses
 *
 * Why useful?
 *
 * Client and server are separated, so they can be developed
 * and scaled independently.
 *
 * Example:
 *
 *     Web App
 *        |
 *     Mobile App
 *        |
 *        v
 *     Backend
 *
 * Same backend can serve multiple clients.
 *
 *
 * Interview Point:
 *
 * Client-server separation allows us to change or scale clients
 * without necessarily changing the backend.
 *
 *
 * ============================================================
 * 2. IP ADDRESS
 * ============================================================
 *
 * IP address identifies a machine on a network.
 *
 * Example:
 *
 * IPv4:
 *     192.168.1.1
 *
 * IPv6:
 *     2001:0db8:85a3::8a2e:0370:7334
 *
 * Basic flow:
 *
 *     Client
 *       |
 *       v
 *     Router / Gateway
 *       |
 *       v
 *     Internet
 *       |
 *       v
 *     Server IP
 *
 * System Design relevance:
 *
 * Servers, load balancers and database nodes communicate
 * using network addresses.
 *
 *
 * ============================================================
 * 3. DNS
 * ============================================================
 *
 * DNS = Domain Name System.
 *
 * Purpose:
 *
 *     Domain Name
 *          |
 *          v
 *        DNS
 *          |
 *          v
 *     IP Address
 *
 * Example:
 *
 *     google.com
 *          |
 *          v
 *     142.250.80.46
 *
 * DNS resolution roughly works like:
 *
 * Browser Cache
 *      |
 *      v
 * DNS Resolver
 *      |
 *      v
 * Root Server
 *      |
 *      v
 * TLD Server (.com)
 *      |
 *      v
 * Authoritative Name Server
 *      |
 *      v
 * IP Address
 *
 * Results are cached to avoid repeating the complete lookup.
 *
 * DNS can also help with:
 *
 * - Load balancing
 * - Failover
 *
 *
 * Interview Point:
 *
 * DNS translates human-readable domain names into IP addresses.
 *
 *
 * ============================================================
 * 4. FORWARD PROXY vs REVERSE PROXY
 * ============================================================
 *
 *
 * FORWARD PROXY
 *
 * Sits in front of clients.
 *
 *     Client
 *        |
 *        v
 *   Forward Proxy
 *        |
 *        v
 *     Internet
 *        |
 *        v
 *     Server
 *
 * Server sees the proxy instead of the original client.
 *
 * Uses:
 *
 * - Privacy
 * - Content filtering
 * - Corporate network control
 * - Geo restriction handling
 *
 *
 * REVERSE PROXY
 *
 * Sits in front of servers.
 *
 *     Client
 *        |
 *        v
 *   Reverse Proxy
 *      /   \
 *     v     v
 * Server  Server
 *
 * Client thinks it is communicating with one server,
 * while the reverse proxy can route traffic to many servers.
 *
 * Common examples:
 *
 * - Nginx
 * - HAProxy
 *
 * Reverse proxy can handle:
 *
 * - SSL termination
 * - Caching
 * - Compression
 * - Load balancing
 *
 *
 * Easy way to remember:
 *
 * Forward Proxy:
 *     protects / represents CLIENT
 *
 * Reverse Proxy:
 *     protects / represents SERVER
 *
 *
 * ============================================================
 * 5. LATENCY
 * ============================================================
 *
 * Latency = time taken for data to travel from point A to B.
 *
 * Usually measured in milliseconds (ms).
 *
 * Latency can come from:
 *
 * - Network distance
 * - Serialization
 * - Server processing
 * - Queueing
 *
 *
 * Example:
 *
 *     User in India
 *          |
 *          | long distance
 *          v
 *     Server in USA
 *
 * More physical distance generally means more network latency.
 *
 *
 * Ways to reduce latency:
 *
 * 1. CDN
 * 2. Caching
 * 3. Regional deployments
 * 4. Keep servers closer to users
 *
 *
 * Interview Point:
 *
 * Latency is an important non-functional requirement.
 *
 *
 * ============================================================
 * 6. HTTP / HTTPS
 * ============================================================
 *
 * HTTP:
 *
 * Protocol used by clients and servers to communicate.
 *
 * HTTPS:
 *
 * HTTP + TLS encryption.
 *
 * Basic HTTP request:
 *
 *     Client
 *       |
 *       | GET /users
 *       v
 *     Server
 *       |
 *       | 200 OK + JSON
 *       v
 *     Client
 *
 *
 * Important HTTP methods:
 *
 * GET
 * POST
 * PUT
 * PATCH
 * DELETE
 *
 *
 * Important status codes:
 *
 * 200 -> OK
 * 201 -> Created
 * 404 -> Not Found
 * 500 -> Server Error
 *
 *
 * HTTP is stateless:
 *
 * Each request is independent.
 *
 * Server does not automatically remember previous requests.
 *
 * State can be maintained using:
 *
 * - Cookies
 * - Tokens
 * - Sessions
 *
 *
 * HTTPS adds TLS:
 *
 * ClientHello
 *      |
 *      v
 * ServerHello + Certificate
 *      |
 *      v
 * Key Exchange
 *      |
 *      v
 * Encrypted Communication
 *
 *
 * ============================================================
 * GROUP 2: APIs & COMMUNICATION
 * ============================================================
 *
 *
 * 7. API
 * ============================================================
 *
 * API = Application Programming Interface.
 *
 * API is a contract between two pieces of software.
 *
 * It defines:
 *
 * - What can be requested
 * - How to request it
 * - What response will be returned
 *
 *
 * Example:
 *
 *     Client
 *       |
 *       | JSON Request
 *       v
 *     API
 *       |
 *       v
 *     Business Logic
 *       |
 *       v
 *     Database
 *
 *
 * Think of API as a restaurant menu:
 *
 * Menu       -> API documentation
 * Order      -> Request
 * Food       -> Response
 * Kitchen    -> Backend
 *
 *
 * ============================================================
 * 8. REST API
 * ============================================================
 *
 * REST = Representational State Transfer.
 *
 * REST treats data as resources and uses standard HTTP methods.
 *
 *
 * Example:
 *
 * GET /users
 *     -> Get all users
 *
 * GET /users/123
 *     -> Get user 123
 *
 * POST /users
 *     -> Create user
 *
 * PUT /users/123
 *     -> Update user
 *
 * DELETE /users/123
 *     -> Delete user
 *
 *
 * REST principles from the source:
 *
 * - Stateless
 * - Cacheable
 * - Uniform interface
 *
 *
 * REST naturally maps to CRUD:
 *
 * Create -> POST
 * Read   -> GET
 * Update -> PUT / PATCH
 * Delete -> DELETE
 *
 *
 * Interview Point:
 *
 * REST is usually a safe default for general-purpose APIs.
 *
 *
 * ============================================================
 * 9. GRAPHQL
 * ============================================================
 *
 * GraphQL allows clients to request exactly the data they need.
 *
 *
 * REST:
 *
 *     GET /users/123
 *     GET /users/123/posts
 *
 * Potentially multiple requests.
 *
 *
 * GraphQL:
 *
 *     One query
 *        |
 *        v
 *     User + Posts
 *
 *
 * Example:
 *
 * query {
 *   user(id: 1) {
 *     name
 *     posts {
 *       title
 *     }
 *   }
 * }
 *
 *
 * Advantages:
 *
 * - Client gets exactly required fields
 * - Can combine data from multiple resources
 * - Useful for flexible data requirements
 *
 *
 * Trade-offs:
 *
 * - More server-side complexity
 * - Resolvers required
 * - Caching can be harder
 * - Deeply nested queries can cause performance issues
 *
 *
 * Interview Point:
 *
 * REST is generally simpler.
 *
 * GraphQL is useful when clients need flexible data fetching.
 *
 *
 * ============================================================
 * 10. WEBSOCKETS
 * ============================================================
 *
 * HTTP:
 *
 *     Client -> Request
 *     Server -> Response
 *
 * WebSocket:
 *
 *     Client <=================> Server
 *             Persistent
 *             Connection
 *
 * WebSocket starts as an HTTP request and then upgrades
 * to a persistent bidirectional connection.
 *
 *
 * Both client and server can send messages at any time.
 *
 *
 * Useful for:
 *
 * - Chat applications
 * - Live sports scores
 * - Collaborative editors
 * - Real-time notifications
 * - Typing indicators
 *
 *
 * Main advantage:
 *
 * Real-time bidirectional communication.
 *
 *
 * Main disadvantage:
 *
 * Connections are stateful.
 *
 * If a server goes down:
 *
 *     Its WebSocket connections are lost.
 *
 * Scaling therefore becomes more difficult.
 *
 *
 * ============================================================
 * 11. WEBHOOKS
 * ============================================================
 *
 * Webhook = server sends a notification when an event happens.
 *
 *
 * Traditional polling:
 *
 *     Client -> "Anything changed?"
 *     Client -> "Anything changed?"
 *     Client -> "Anything changed?"
 *
 *
 * Webhook:
 *
 *     Event Happens
 *          |
 *          v
 *     External Server
 *          |
 *          | POST /webhook
 *          v
 *     Your Application
 *
 *
 * Example:
 *
 * Stripe payment completed
 *          |
 *          v
 * Stripe sends webhook
 *          |
 *          v
 * POST /payment-webhook
 *
 *
 * Common use cases:
 *
 * - Payment notifications
 * - GitHub events
 * - Message delivery status
 *
 *
 * Important reliability concerns:
 *
 * - Retry logic
 * - Event logging
 * - Idempotency
 *
 *
 * Interview Point:
 *
 * Use webhooks when another system needs to notify your
 * application asynchronously about an event.
 *
 *
 * ============================================================
 * GROUP 3: DATA STORAGE
 * ============================================================
 *
 *
 * 12. DATABASES
 * ============================================================
 *
 * Database = organized collection of data used for:
 *
 * - Storage
 * - Retrieval
 * - Manipulation
 *
 *
 * Common database types:
 *
 *
 * Relational:
 *     PostgreSQL
 *     MySQL
 *
 * Document:
 *     MongoDB
 *
 * Key-Value:
 *     Redis
 *     DynamoDB
 *
 * Graph:
 *     Neo4j
 *
 *
 * Choose database based on:
 *
 *     Data Model
 *          +
 *     Query Pattern
 *          +
 *     Scalability Requirement
 *
 * There is no universally best database.
 *
 *
 * ============================================================
 * 13. SQL vs NoSQL
 * ============================================================
 *
 *
 * SQL
 *
 * - Structured schema
 * - ACID transactions
 * - Complex joins
 * - Strong consistency use cases
 *
 * Examples:
 *
 * PostgreSQL
 * MySQL
 *
 *
 * NoSQL
 *
 * - Flexible schema
 * - Horizontal scalability
 * - Useful for high write throughput
 * - Useful for naturally unstructured/evolving data
 *
 * Examples:
 *
 * MongoDB
 * Cassandra
 * DynamoDB
 *
 *
 * Choose SQL when:
 *
 * - Strong consistency is important
 * - Complex joins are required
 * - Schema is well-defined
 * - ACID transactions are required
 *
 *
 * Choose NoSQL when:
 *
 * - Horizontal scaling is important
 * - Schema changes frequently
 * - High write throughput is required
 * - Data is naturally unstructured
 *
 *
 * Important:
 *
 * Real production systems can use BOTH.
 *
 * Example:
 *
 *     PostgreSQL
 *         |
 *         +--> Orders
 *
 *     MongoDB
 *         |
 *         +--> Product Catalog
 *
 *
 * Interview Point:
 *
 * Never say:
 *
 *     "SQL is better."
 *
 * or
 *
 *     "NoSQL is better."
 *
 * Explain why a particular storage model fits the
 * particular workload.
 *
 *
 * ============================================================
 * 14. DATABASE INDEXING
 * ============================================================
 *
 * Without index:
 *
 *     Search row
 *         |
 *         v
 * Scan every row
 *
 * With index:
 *
 *     Search value
 *         |
 *         v
 *     Index
 *         |
 *         v
 *     Required row
 *
 *
 * Index makes lookups faster.
 *
 * Most databases commonly use B-tree indexes.
 *
 * B-tree lookup:
 *
 *     O(log n)
 *
 *
 * Example:
 *
 *     CREATE INDEX on email
 *
 * Database maintains an index that helps locate rows
 * efficiently.
 *
 *
 * Trade-off:
 *
 * Faster reads
 *      BUT
 * Slower writes
 *
 * Because INSERT/UPDATE may also need to update indexes.
 *
 * Indexes also consume storage.
 *
 *
 * Good candidates:
 *
 * - WHERE columns
 * - JOIN columns
 *
 *
 * Don't index everything.
 *
 *
 * ============================================================
 * 15. VERTICAL PARTITIONING
 * ============================================================
 *
 * Vertical partitioning splits a wide table by columns.
 *
 *
 * Before:
 *
 * Users
 * ------------------------------
 * id
 * name
 * email
 * password_hash
 * avatar
 * bio
 * billing_address
 * card_number
 * login_count
 *
 *
 * After:
 *
 * Profile
 * ----------------
 * id
 * name
 * email
 * avatar
 * bio
 *
 * Auth
 * ----------------
 * id
 * password_hash
 * login_count
 *
 * Billing
 * ----------------
 * id
 * billing_address
 * card_number
 *
 *
 * Main idea:
 *
 * Split data based on access pattern.
 *
 *
 * Benefits:
 *
 * - Fewer bytes read
 * - Faster queries
 * - Different access controls
 * - Partitions can be optimized independently
 *
 *
 * Use when:
 *
 * A table has many columns and different parts are accessed
 * in different contexts.
 *
 *
 * ============================================================
 * 16. CACHING
 * ============================================================
 *
 * Cache stores frequently accessed data in a fast layer,
 * usually memory.
 *
 *
 * Main goal:
 *
 *     Reduce database reads
 *     +
 *     Improve read latency
 *
 *
 * Common technologies:
 *
 * - Redis
 * - Memcached
 *
 *
 * CACHE-ASIDE / LAZY LOADING
 *
 *     Client
 *       |
 *       v
 *     App Server
 *       |
 *       v
 *     Check Cache
 *       |
 *       +---- HIT ----> Return data
 *       |
 *       +---- MISS
 *               |
 *               v
 *            Database
 *               |
 *               v
 *            Store in Cache
 *               |
 *               v
 *            Return data
 *
 *
 * Main challenge:
 *
 * Cache invalidation.
 *
 *
 * Common strategies:
 *
 * TTL
 *     Data automatically expires.
 *
 * Write-through
 *     Update cache and database together.
 *
 * Write-behind
 *     Update cache first, database later.
 *
 *
 * ============================================================
 * 17. DENORMALIZATION
 * ============================================================
 *
 * Normalization:
 *
 * Data is split into separate tables to reduce duplication.
 *
 *
 * Problem:
 *
 * A read may require many JOINs.
 *
 *
 * Denormalization:
 *
 * Deliberately duplicate some data to reduce JOINs.
 *
 *
 * Example:
 *
 * Normalized:
 *
 * Orders
 *    |
 *    +----> Users
 *    |
 *    +----> Products
 *
 *
 * Denormalized:
 *
 * Orders
 * -------------------------
 * order_id
 * user_id
 * user_name
 * product_id
 * product_title
 * price
 *
 *
 * Benefit:
 *
 * Faster reads.
 *
 * Cost:
 *
 * Duplicate data.
 *
 * If user_name changes, multiple records may need updating.
 *
 *
 * Use when:
 *
 * - System is read-heavy
 * - Reads require expensive joins
 *
 *
 * Important:
 *
 * Whenever using denormalization, explain how redundant data
 * will remain consistent.
 *
 *
 * ============================================================
 * 18. BLOB STORAGE
 * ============================================================
 *
 * BLOB = Binary Large Object.
 *
 * Examples:
 *
 * - Images
 * - Videos
 * - PDFs
 * - Backups
 *
 *
 * Don't store large files directly inside the main database
 * when specialized blob storage is more appropriate.
 *
 *
 * Typical architecture:
 *
 *     User
 *       |
 *       v
 *     App Server
 *       |
 *       v
 *     Blob Storage
 *       |
 *       | URL
 *       v
 *     Database
 *
 *
 * Database stores:
 *
 *     file_url
 *
 * Instead of:
 *
 *     10 MB image data
 *
 *
 * Common example:
 *
 *     Amazon S3
 *
 *
 * Typical flow:
 *
 * 1. Upload file
 * 2. Blob storage returns URL
 * 3. Store URL in database
 * 4. CDN can serve the file
 *
 *
 * Useful for:
 *
 * - Media-heavy systems
 * - Document systems
 * - Backup systems
 *
 *
 * ============================================================
 * GROUP 4: SCALING
 * ============================================================
 *
 *
 * 19. VERTICAL SCALING
 * ============================================================
 *
 * Scale Up = make one machine more powerful.
 *
 *
 * Example:
 *
 * Small:
 *     2 CPU
 *     4 GB RAM
 *
 *       ↓
 *
 * Medium:
 *     8 CPU
 *     32 GB RAM
 *
 *       ↓
 *
 * Large:
 *     64 CPU
 *     256 GB RAM
 *
 *
 * Advantages:
 *
 * - Simple
 * - Usually no major architecture changes
 * - No need to distribute work across machines
 *
 *
 * Problems:
 *
 * - Hardware ceiling
 * - Cost increases
 * - Single point of failure
 *
 *
 * ============================================================
 * 20. HORIZONTAL SCALING
 * ============================================================
 *
 * Scale Out = add more machines.
 *
 *
 *     Load Balancer
 *        / | \
 *       /  |  \
 *      v   v   v
 *    S1   S2   S3
 *
 *
 * Instead of one powerful server,
 * use multiple servers.
 *
 *
 * Advantages:
 *
 * - More capacity
 * - Can continue adding servers
 * - Better availability
 *
 *
 * Requirements:
 *
 * Application should preferably be stateless.
 *
 * Shared state should live in external/shared storage such as
 * Redis rather than local server memory.
 *
 *
 * Usually requires:
 *
 * - Load balancer
 * - Stateless servers
 * - Database scaling strategy
 *
 *
 * Database may require:
 *
 * - Replication
 * - Sharding
 *
 *
 * ============================================================
 * 21. LOAD BALANCERS
 * ============================================================
 *
 * Load balancer distributes incoming requests across servers.
 *
 *
 *     Clients
 *        |
 *        v
 *   Load Balancer
 *      / | \
 *     v  v  v
 *    S1 S2 S3
 *
 *
 * Common algorithms:
 *
 * 1. Round Robin
 *
 * Request 1 -> Server 1
 * Request 2 -> Server 2
 * Request 3 -> Server 3
 *
 *
 * 2. Least Connections
 *
 * Send request to the server currently handling
 * the fewest connections.
 *
 *
 * 3. Weighted
 *
 * More powerful servers receive more traffic.
 *
 *
 * Health Checks:
 *
 *     Server 1 -> Healthy
 *     Server 2 -> Healthy
 *     Server 3 -> DOWN
 *
 * Traffic is routed away from unhealthy servers.
 *
 *
 * Main benefits:
 *
 * - Scalability
 * - Availability
 *
 *
 * ============================================================
 * 22. REPLICATION
 * ============================================================
 *
 * Replication = copying data across multiple database servers.
 *
 *
 * Common model:
 *
 *              Primary
 *             /       \
 *            /         \
 *           v           v
 *       Replica 1   Replica 2
 *
 *
 * Primary:
 *
 * - Writes
 * - Can also serve reads
 *
 * Replicas:
 *
 * - Usually read-only
 * - Receive replicated data
 *
 *
 * Main benefit:
 *
 * Read scaling.
 *
 *
 * Example:
 *
 *     App
 *      |
 *      +---- WRITE ----> Primary
 *      |
 *      +---- READ -----> Replica 1
 *      |
 *      +---- READ -----> Replica 2
 *
 *
 * If a replica fails:
 *
 *     Redirect reads to another replica.
 *
 * If primary fails:
 *
 *     A replica can potentially be promoted.
 *
 *
 * Trade-off:
 *
 * Replication lag.
 *
 * Data written to primary may take some time to appear
 * on replicas.
 *
 *
 * Therefore:
 *
 * Eventually:
 *
 *     Primary = latest
 *     Replica = possibly slightly stale
 *
 *
 * For critical reads such as account balance,
 * reading from the primary may be necessary.
 *
 *
 * ============================================================
 * 23. SHARDING
 * ============================================================
 *
 * Sharding splits data across multiple database nodes.
 *
 *
 * Replication:
 *
 *     Every node has same data.
 *
 *
 * Sharding:
 *
 *     Each node has different portion of data.
 *
 *
 * Example:
 *
 *     user_id
 *        |
 *        v
 *    Shard Router
 *        |
 *     +--+--+
 *     |  |  |
 *     v  v  v
 *    S1 S2 S3
 *
 *
 * Example ranges:
 *
 * Shard 1:
 *     user_id 1 - 1M
 *
 * Shard 2:
 *     user_id 1M - 2M
 *
 * Shard 3:
 *     user_id 2M - 3M
 *
 *
 * Common approaches:
 *
 * 1. Hash-based sharding
 *
 *     hash(key) % N
 *
 *
 * 2. Range-based sharding
 *
 *     Consecutive key ranges
 *
 *
 * Important:
 *
 * Shard key selection is critical.
 *
 * Bad shard key
 *      |
 *      v
 * Hot Shard
 *      |
 *      v
 * One node receives excessive traffic.
 *
 *
 * Challenges:
 *
 * - Cross-shard queries
 * - Rebalancing
 * - Operational complexity
 *
 *
 * ============================================================
 * GROUP 5: DISTRIBUTED SYSTEMS
 * ============================================================
 *
 *
 * 24. CAP THEOREM
 * ============================================================
 *
 * CAP:
 *
 * C = Consistency
 * A = Availability
 * P = Partition Tolerance
 *
 *
 * Consistency:
 *
 * Every read gets the latest write.
 *
 *
 * Availability:
 *
 * Every request gets a response.
 *
 *
 * Partition Tolerance:
 *
 * System continues working despite network communication
 * failures between nodes.
 *
 *
 * In a distributed system, network partitions are considered
 * unavoidable.
 *
 * Therefore the practical choice is commonly:
 *
 *     CP
 *     OR
 *     AP
 *
 *
 * CP:
 *
 *     Consistency
 *          +
 *     Partition Tolerance
 *
 * System may reject/delay requests if it cannot guarantee
 * up-to-date data.
 *
 * Useful when stale data causes serious problems.
 *
 * Examples mentioned in the source:
 *
 * - MongoDB
 * - HBase
 *
 *
 * AP:
 *
 *     Availability
 *          +
 *     Partition Tolerance
 *
 * System continues responding even if data may be slightly stale.
 *
 * Examples mentioned in the source:
 *
 * - Cassandra
 * - DynamoDB
 *
 *
 * Interview Point:
 *
 * Don't simply say:
 *
 *     "CAP means choose any two."
 *
 * Explain that network partitions force a trade-off between
 * consistency and availability in a partitioned distributed system.
 *
 *
 * ============================================================
 * 25. CDN
 * ============================================================
 *
 * CDN = Content Delivery Network.
 *
 * A globally distributed network of servers that caches and
 * serves content close to users.
 *
 *
 * Without CDN:
 *
 *     User
 *       |
 *       v
 *   Origin Server
 *
 *
 * With CDN:
 *
 *     User
 *       |
 *       v
 *   Nearest Edge Node
 *       |
 *       v
 *   Origin Server (on cache miss)
 *
 *
 * Common content:
 *
 * - Images
 * - CSS
 * - JavaScript
 * - Videos
 *
 *
 * Benefits:
 *
 * - Lower latency
 * - Lower origin-server load
 * - Better availability
 *
 *
 * Common providers mentioned:
 *
 * - AWS CloudFront
 * - Cloudflare
 * - Akamai
 *
 *
 * Typical flow:
 *
 * First request:
 *
 *     User
 *       |
 *       v
 *     CDN Miss
 *       |
 *       v
 *     Origin
 *
 *
 * Later requests:
 *
 *     User
 *       |
 *       v
 *     CDN HIT
 *       |
 *       v
 *     Content
 *
 *
 * ============================================================
 * 26. IDEMPOTENCY
 * ============================================================
 *
 * Idempotent operation:
 *
 * Executing it multiple times produces the same effective result
 * as executing it once.
 *
 *
 * Why is it important?
 *
 * In distributed systems:
 *
 *     Request
 *       |
 *       v
 *     Server processes request
 *       |
 *       X
 *     Response lost
 *       |
 *       v
 *     Client retries
 *
 * The server may actually have already completed the operation.
 *
 * Without idempotency:
 *
 *     Payment #1
 *     Payment #2
 *
 * Same request could cause duplicate side effects.
 *
 *
 * Naturally idempotent examples from the source:
 *
 * GET
 * DELETE
 *
 * POST is not naturally idempotent.
 *
 *
 * Solution:
 *
 *     Idempotency-Key
 *
 *
 * Example:
 *
 * POST /payments
 *
 * Idempotency-Key: abc-123
 *
 *
 * First request:
 *
 *     Process payment
 *     Store key abc-123
 *     Return result
 *
 *
 * Retry:
 *
 *     Same key abc-123
 *          |
 *          v
 *     Key already exists
 *          |
 *          v
 *     Return stored result
 *
 * No duplicate payment.
 *
 *
 * Important use cases:
 *
 * - Payments
 * - Order creation
 * - Any operation where duplicate execution is harmful
 *
 *
 * ============================================================
 * GROUP 6: ARCHITECTURE PATTERNS
 * ============================================================
 *
 *
 * 27. MICROSERVICES
 * ============================================================
 *
 * As applications grow, a single monolithic codebase can become
 * harder to maintain, deploy and scale.
 *
 *
 * MONOLITH:
 *
 *     Single Application
 *       |
 *       +-- Users
 *       +-- Orders
 *       +-- Payments
 *       +-- Inventory
 *
 *     Single Codebase
 *     Single DB
 *
 *
 * MICROSERVICES:
 *
 *     Users Service ---- Users DB
 *
 *     Orders Service --- Orders DB
 *
 *     Payments Service - Payments DB
 *
 *     Inventory Service - Inventory DB
 *
 *
 * Each service:
 *
 * - Owns a business capability
 * - Can be developed independently
 * - Can be deployed independently
 * - Can be scaled independently
 *
 *
 * Example:
 *
 * Payments traffic increases.
 *
 * Instead of scaling the entire application:
 *
 *     Scale Payments Service only.
 *
 *
 * Benefits:
 *
 * - Independent deployment
 * - Independent scaling
 * - Team ownership
 * - Different technologies can be used
 *
 *
 * Downsides:
 *
 * - Network calls instead of function calls
 * - Distributed-system complexity
 * - Data consistency challenges
 * - More monitoring/deployment overhead
 * - Debugging becomes harder
 *
 *
 * Interview Point:
 *
 * Microservices are not automatically better than monoliths.
 * They introduce benefits as well as significant operational
 * complexity.
 *
 *
 * ============================================================
 * 28. MESSAGE QUEUES
 * ============================================================
 *
 * Synchronous communication:
 *
 *     Service A
 *        |
 *        | Request
 *        v
 *     Service B
 *        |
 *        | Response
 *        v
 *     Service A
 *
 *
 * Problem:
 *
 * If Service B is slow or down,
 * Service A is affected.
 *
 *
 * Message Queue:
 *
 *     Producer
 *        |
 *        v
 *      Queue
 *        |
 *     +--+----------+
 *     |      |      |
 *     v      v      v
 *   Email Inventory Analytics
 *   Service Service   Service
 *
 *
 * Producer sends a message and continues.
 *
 * Consumers process the message when ready.
 *
 *
 * Benefits:
 *
 * 1. Decoupling
 *
 * Services don't need direct knowledge of each other.
 *
 * 2. Buffering
 *
 * Absorbs traffic spikes.
 *
 * 3. Reliability
 *
 * Messages can remain available if consumers temporarily crash.
 *
 * 4. Scalability
 *
 * Add more consumers to process messages faster.
 *
 *
 * Technologies mentioned:
 *
 * Kafka
 *     High throughput / event streaming
 *
 * RabbitMQ
 *     Traditional message broker
 *
 * SQS
 *     Managed AWS queue
 *
 *
 * ============================================================
 * 29. RATE LIMITING
 * ============================================================
 *
 * Rate limiting controls how many requests a client can make
 * within a given time window.
 *
 *
 * Example:
 *
 *     User
 *       |
 *       v
 *   Rate Limiter
 *       |
 *       +---- Under limit ----> Allow
 *       |
 *       +---- Over limit -----> 429
 *
 *
 * HTTP response:
 *
 *     429 Too Many Requests
 *
 *
 * Why use rate limiting?
 *
 * - Prevent abuse
 * - Protect backend
 * - Protect expensive operations
 * - Ensure fair usage
 * - Prevent one client from overwhelming the system
 *
 *
 * Common algorithms:
 *
 *
 * 1. TOKEN BUCKET
 *
 * Tokens refill at a fixed rate.
 *
 * Each request consumes a token.
 *
 *
 * 2. SLIDING WINDOW
 *
 * Counts requests in a rolling time window.
 *
 *
 * 3. FIXED WINDOW
 *
 * Counts requests in fixed time intervals.
 *
 *
 * Typical architecture:
 *
 *     Client
 *       |
 *       v
 *   API Gateway
 *       |
 *       v
 *   Rate Limiter
 *       |
 *       v
 *     Redis
 *
 *
 * Redis can be used as a fast store for request counters.
 *
 *
 * Example:
 *
 * Free user:
 *     100 req/min
 *
 * Paid user:
 *     10,000 req/min
 *
 *
 * ============================================================
 * 30. API GATEWAY
 * ============================================================
 *
 * API Gateway = single entry point for client requests.
 *
 *
 * Without API Gateway:
 *
 * Client
 *   |
 *   +----> User Service
 *   |
 *   +----> Order Service
 *   |
 *   +----> Payment Service
 *   |
 *   +----> Search Service
 *
 *
 * Client must know many service addresses.
 *
 *
 * With API Gateway:
 *
 * Client
 *   |
 *   v
 * API Gateway
 *   |
 *   +----> User Service
 *   |
 *   +----> Order Service
 *   |
 *   +----> Payment Service
 *   |
 *   +----> Search Service
 *
 *
 * Gateway can centralize:
 *
 * - Routing
 * - Authentication
 * - Rate limiting
 * - Request transformation
 * - Response transformation
 * - Response aggregation
 * - Caching
 * - Circuit breaking
 *
 *
 * Example:
 *
 *     GET /users
 *          |
 *          v
 *     API Gateway
 *          |
 *          v
 *     Users Service
 *
 *
 * Client only needs to know:
 *
 *     One gateway URL
 *
 * Instead of:
 *
 *     20 different service URLs
 *
 *
 * Popular options mentioned:
 *
 * - Kong
 * - AWS API Gateway
 * - Nginx
 *
 *
 * ============================================================
 * 30 CONCEPTS - QUICK REVISION
 * ============================================================
 *
 * 1.  Client-Server
 *     -> Client requests, server responds.
 *
 * 2.  IP Address
 *     -> Identifies machines on a network.
 *
 * 3.  DNS
 *     -> Domain name -> IP address.
 *
 * 4.  Proxy
 *     -> Forward proxy represents clients;
 *        reverse proxy represents servers.
 *
 * 5.  Latency
 *     -> Time taken for data to travel/process.
 *
 * 6.  HTTP/HTTPS
 *     -> Web communication; HTTPS adds TLS encryption.
 *
 * 7.  API
 *     -> Contract between software components.
 *
 * 8.  REST
 *     -> Resource-based, stateless API style.
 *
 * 9.  GraphQL
 *     -> Client requests exactly the required data.
 *
 * 10. WebSockets
 *     -> Persistent bidirectional real-time communication.
 *
 * 11. Webhooks
 *     -> Server notifies another system when an event occurs.
 *
 * 12. Databases
 *     -> Persistent structured/unstructured data storage.
 *
 * 13. SQL vs NoSQL
 *     -> Choose based on data model, consistency and workload.
 *
 * 14. Indexing
 *     -> Faster reads/search at the cost of write overhead.
 *
 * 15. Vertical Partitioning
 *     -> Split columns based on access patterns.
 *
 * 16. Caching
 *     -> Store frequently accessed data in a fast layer.
 *
 * 17. Denormalization
 *     -> Duplicate data to reduce expensive joins.
 *
 * 18. Blob Storage
 *     -> Store large files such as images/videos/documents.
 *
 * 19. Vertical Scaling
 *     -> Make one machine more powerful.
 *
 * 20. Horizontal Scaling
 *     -> Add more machines.
 *
 * 21. Load Balancer
 *     -> Distributes traffic across servers.
 *
 * 22. Replication
 *     -> Copy data across database nodes.
 *
 * 23. Sharding
 *     -> Split data across database nodes.
 *
 * 24. CAP Theorem
 *     -> Consistency, Availability, Partition Tolerance.
 *
 * 25. CDN
 *     -> Serve content from locations closer to users.
 *
 * 26. Idempotency
 *     -> Retries should not create duplicate side effects.
 *
 * 27. Microservices
 *     -> Independently deployable business services.
 *
 * 28. Message Queues
 *     -> Decouple async producers and consumers.
 *
 * 29. Rate Limiting
 *     -> Control request frequency.
 *
 * 30. API Gateway
 *     -> Single entry point for client requests.
 *
 *
 * ============================================================
 * SYSTEM DESIGN CONNECTION MAP
 * ============================================================
 *
 *
 *                 CLIENT
 *                    |
 *                    v
 *                  DNS
 *                    |
 *                    v
 *              Load Balancer
 *                    |
 *                    v
 *              API Gateway
 *                    |
 *          +---------+---------+
 *          |                   |
 *          v                   v
 *      Services            WebSocket
 *          |
 *          +-------------------------+
 *          |                         |
 *          v                         v
 *       Cache                    Message Queue
 *          |                         |
 *          v                         v
 *       Database                  Workers
 *          |
 *      +---+---+
 *      |       |
 *      v       v
 * Replication Sharding
 *
 * Large files:
 *
 *      Application
 *          |
 *          v
 *      Blob Storage
 *          |
 *          v
 *         CDN
 *          |
 *          v
 *        Users
 *
 *
 * ============================================================
 * MOST IMPORTANT TRADE-OFFS
 * ============================================================
 *
 *
 * CACHE
 *
 * Faster reads
 *      vs
 * Cache invalidation
 *
 *
 * INDEX
 *
 * Faster reads
 *      vs
 * More storage + slower writes
 *
 *
 * DENORMALIZATION
 *
 * Faster reads
 *      vs
 * Data duplication / consistency complexity
 *
 *
 * VERTICAL SCALING
 *
 * Simplicity
 *      vs
 * Hardware ceiling + single point of failure
 *
 *
 * HORIZONTAL SCALING
 *
 * More capacity
 *      vs
 * More distributed-system complexity
 *
 *
 * REPLICATION
 *
 * Better read scaling + availability
 *      vs
 * Replication lag
 *
 *
 * SHARDING
 *
 * More storage/write capacity
 *      vs
 * Cross-shard queries + rebalancing complexity
 *
 *
 * MICROSERVICES
 *
 * Independent deployment/scaling
 *      vs
 * Network + operational complexity
 *
 *
 * MESSAGE QUEUE
 *
 * Decoupling + buffering
 *      vs
 * Asynchronous/eventual processing
 *
 *
 * GRAPHQL
 *
 * Flexible data fetching
 *      vs
 * More server complexity + harder caching
 *
 *
 * WEBSOCKETS
 *
 * Real-time communication
 *      vs
 * Stateful connections + harder scaling
 *
 *
 * ============================================================
 * SYSTEM DESIGN INTERVIEW CHEAT SHEET
 * ============================================================
 *
 *
 * If traffic is increasing:
 *
 *     Horizontal Scaling
 *     +
 *     Load Balancer
 *
 *
 * If reads are increasing:
 *
 *     Cache
 *     +
 *     Database Replicas
 *
 *
 * If writes/data are too large for one DB:
 *
 *     Sharding
 *
 *
 * If files are large:
 *
 *     Blob Storage
 *     +
 *     CDN
 *
 *
 * If work can happen asynchronously:
 *
 *     Message Queue
 *     +
 *     Worker
 *
 *
 * If clients need real-time updates:
 *
 *     WebSockets
 *
 *
 * If another system needs to notify your application:
 *
 *     Webhook
 *
 *
 * If duplicate requests are dangerous:
 *
 *     Idempotency Key
 *
 *
 * If API is being abused:
 *
 *     Rate Limiting
 *
 *
 * If there are many microservices:
 *
 *     API Gateway
 *
 *
 * If users are geographically distributed:
 *
 *     CDN
 *     +
 *     Regional Deployment
 *
 *
 * If database reads are slow:
 *
 *     Index
 *     +
 *     Cache
 *     +
 *     Read Replicas
 *
 *
 * If JOINs are expensive:
 *
 *     Denormalization
 *
 *
 * ============================================================
 * FINAL INTERVIEW MINDSET
 * ============================================================
 *
 * Don't memorize:
 *
 *     "Redis = cache"
 *     "Kafka = queue"
 *     "MongoDB = NoSQL"
 *
 * Instead think:
 *
 *     REQUIREMENT
 *          |
 *          v
 *       PROBLEM
 *          |
 *          v
 *     DESIGN OPTION
 *          |
 *          v
 *      TRADE-OFF
 *          |
 *          v
 *     FINAL DECISION
 *
 *
 * Example:
 *
 * Requirement:
 *     Millions of repeated reads.
 *
 * Problem:
 *     Database is overloaded.
 *
 * Solution:
 *     Add cache.
 *
 * Trade-off:
 *     Cache invalidation / stale data.
 *
 *
 * This reasoning pattern is more important than simply
 * memorizing technologies.
 *
 * ============================================================
 */
```
