```python
/**
 * OVERVIEW:
 * - Jab hum koi system design karte hain, tab sirf ye important nahi hota ki
 *   system "kya karta hai", balki ye bhi important hota hai ki system
 *   "kitni achhi tarah kaam karta hai".
 * 
 * Non-Functional System Characteristics system ki quality ko describe karti
 * hain, jaise:
 *
 *    1. Availability    -> System kitne time accessible rehta hai?
 *    2. Reliability     -> System kitni consistently correct results deta hai?
 *    3. Scalability     -> System increasing load ko kitni achhi tarah handle karta hai?
 *    4. Maintainability -> System ko modify, debug aur extend karna kitna easy hai?
 *    5. Fault Tolerance -> Components fail hone ke baad bhi system kaam kar sakta hai?
*/

/**
 * Non-Functional System Characteristics:
 * 
 * 1. AVAILABILITY:
 *    - Percentage of time some service/infrastructure is accessible to clients.
 * 
 *    How to improve availability?
 *    a. Replication         : Multiple instances of servers/databases.
 *    b. Load Balancers      : Redirect traffic away from unhealthy nodes. 
 *    c. Failover Mechanisms : Automatically switch to backup systems when primary fails.
 *    d. Graceful Degradation: Some features can temp. fail without total outage
 *    e. CDNs                : Server cached data when backend is unreachable
 * 
 * 2. RELIABILITY:
 *    - Reliability measures how consistently the system performs correctly under
 *      expected conditions.
 *    - A reliable system doesn't just stay up - it gives correct results every time. 
 * 
 *    How to improve reliability?
 *    a. Data durability            : backups, write-ahead logs, or replications
 *    b. Idempotency                : retries don't produce duplicates
 *    c. Monitoring & alerting      : detect silent failures early
 *    d. Testing & Choas Engineering: simulate and harden against failures
 *    e. Consistency Checks         : periodic verification of data integrity
 * 
 *    Availability vs Reliability
 *    - A system can be available but reliable
 *    - Ex: It's up but returns wrong data. 
 *    - Reliability builds trust, but just uptime.
 * 
 * 3. SCALABILITY:
 *    - Scalability is a system's ability to handle increased load without performance
 *      degradation.
 *    - Can it grow seamlessly as traffic, data, or users increase?
 * 
 *    Types of Scalability:
 *    a. Vertical Scaling   (Scale Up)  : Add more CPU/RAM to a single machine. 
 *    b. Horizontal Scaling (Scale Out) : Add more machines and distribute load.
 * 
 *    How to achieve Scalability?
 *    a. Load Balancing                  : distribute requests across servers
 *    b. Database Partitioning (sharding): distribute data
 *    c. Asynchronous Processing (Queues): handle bursts gracefully
 *    d. Stateless Services              : allow easy horizontal scaling
 * 
 * 4. MAINTAINABILITY:
 *    - Maintainability means the system is easy to modify, monitor, debug, and extend
 *      over time. 
 *    - Engineers can add features, fix bugs, or scale parts of the system without
 *      major rewrites.
 * 
 *    How to improve Maintainability?
 *    a. Modular Architecture: Microservices, or well-defined APIs
 *    b. Code Readability and Documentation
 *    c. Automated Testing and CI/CD Pipelines
 *    d. Observability - logging, metrics, tracing
 *    e. Configuration Management - tools like Ansible, Terraform, or Kubernetes.
 * 
 * 5. FAULT TOLERANCE:
 *    - Fault tolerance is the ability of a system to continue operating properly
 *      even when one or more of its components fail.
 *    - It's not about avoiding failure - it's about surviving it.
 * 
 *    Techniques for Fault Tolerance:
 *    a. Redundancy           : Multiple instances of critical components
 *    b. Leader Election      : Automatic failover for distributed nodes
 *    c. Retry mechanism with backoffs
 *    d. Circuit Breaker      : Stop cascading failures
 *    e. Graceful Degradation : limit feature loss, not total downtime
 * 
 *    Example
 *    If one database replica goes down:
 *    - A fault-tolerant system re-routes traffic to another replica.
 *    - A non fault-tolerant system crashes or returns 500 errors.  
*/
```

```python
/**
 * WHAT IS A DISTRIBUTED SYSTEM?
 * > A Distributed System is a collection of multiple computers (nodes) that work
 *   together as a single system.
 * > Although the application runs on multiple servers, it appears as one system
 *   to the user.
 * > Example: Instead of running an application on one server, we run it on multiple
 *   servers connected through a network.
 * 
 * WHY DO WE NEED DISTRIBUTED SYSTEMS?
 * > A single server works well for small applications, but as the number of users
 *   and requests increases, it becomes difficult to handle everything on one
 *   machine.
 * > Problems with a Single Server:
 *   - Limited CPU, Memory, and Storage.
 *   - Cannot handle millions of users efficiently.
 *   - If the server crashes, the entire application becomes unavailable.
 *   - Difficult to scale as traffic grows.
 * > To solve these problems, we distribute the application across multiple
 *   servers.
 *
 * Benefits:
 * - Better Scalability
 * - Higher Availability
 * - Improved Performance
 * - Better Fault Tolerance
 *
 * > However, distributing data across multiple servers introduces new challenges:
 *   - Network failures
 *   - Communication delays
 *   - Data synchronization
 *   - Network partitions
 * > This is where CAP Theorem becomes important.
*/
```

```python
/**
 * CAP THEOREM
 * > CAP Theorem is a Distributed System Design Principle.
 * > It explains the trade-offs in a distributed system during a network partition 
 *   (communication failure between nodes).
 * > CAP Theorem states that during a network partition, a distributed system cannot 
 *   guarantee both Consistency and Availability at the same time.
*/

/**
 * Understanding CAP Theorem: 
 * > Consider a banking application running on two servers:
 *
 *      Server A  <------------->  Server B
 *
 * > Normally, both servers communicate with each other and keep their data synchronized.
 * > Example:
 *   - A user transfers ₹5,000 from Account A to Account B.
 *   - Server A processes the transaction.
 *   - Server A immediately synchronizes the updated balance with Server B.
 *   - Now both servers contain the latest account balance.
 *   - Since both servers have the same data:
 *     - Every user gets the latest balance.
 *     - Every request receives a successful response.
 *   - Everything works perfectly.
 *
 *
 * WHAT HAPPENS IF THE NETWORK FAILS?
 * > Suddenly, the network connection between the servers is lost.
 *
 *      Server A      X      Server B
 *
 * > Both servers are still running, but they can no longer communicate with each other.
 * > This communication failure is called a Network Partition.
 *
 * 1. PARTITION TOLERANCE (P)
 *    - Partition Tolerance means the distributed system continues running even
 *      though some servers cannot communicate with each other.
 *    - In real-world distributed systems, network failures such as lost packets
 *      or connection drops between data centers, are unavoidable.
 *      Therefore, Partition Tolerance is usually considered mandatory.
 *    - Once a partition happens, a new problem appears...
 *
 * 2. CONSISTENCY (C)
 *    - Suppose the ₹5,000 transfer was processed only on Server A before the
 *      partition occurred.
 *    - Now another user checks the account balance, but their request reaches
 *      Server B.
 *    - Server B still has the old balance because it never received the update.
 *
 *    If we want to guarantee Consistency:
 *    - Server B must NOT return the outdated balance.
 *    - Instead, it should return an error or wait until it can verify the latest
 *      data.
 *   
 *    Consistency means:
 *    - Every successful read returns the most recent successful write
 *      (or an error if it cannot guarantee correctness).
 *
 * 3. AVAILABILITY (A)
 *    - Now suppose we choose Availability instead.
 *    - The user's request still reaches Server B.
 *    - Even though Server B has outdated data, it immediately returns the old
 *      balance instead of returning an error.
 *    - The response is available, but it may not contain the latest data.
 *
 *    Availability means:
 *    - Every request receives a response, even if that response contains temporarily 
 *      stale/old data.
 *
 * THE TRADE-OFF:
 * > During a Network Partition:
 *   a. If you choose Consistency (PC):
 *      - Users always get correct data.
 *      - Some requests may fail or wait.
 * 
 *   b. If you choose Availability (PA):
 *      - Every request gets a response.
 *      - Some responses may contain old data.
 * 
 *   c. This is exactly what the CAP Theorem says:
 *      - During a Network Partition, you cannot guarantee both Consistency and 
 *        Availability at the same time.
 *      - You must choose one of them.
 */

/**
 * CONSISTENCY:
 * > Consistency ensures that every user sees the correct and latest data.
 * > Whenever data is updated, all users should eventually see that update based
 *   on the consistency model used by the system.
 * > The main goal of consistency is to keep data synchronized across all nodes
 *   in a distributed system.
 * 
 * Example: Banking Transaction
 * - Your account balance is ₹10,000.
 * - You transfer ₹5,000.
 * - The updated balance becomes ₹5,000.
 * - If you immediately check your balance from another server, that server
 *   should also return ₹5,000.
 * - Returning the old balance (₹10,000) would be stale (outdated) data.
 *
 * 
 * TYPES OF CONSISTENCY:
 * > Consistency is not always implemented in the same way.
 * > Different distributed systems provide different levels of consistency based
 *   on their requirements.
 * > The two most common types are:
 *   - Strong Consistency
 *   - Eventual Consistency
 *
 * 1. STRONG CONSISTENCY:
 *
 *    What is it?
 *    - Every read always returns the latest successfully written data.
 *    - Once a write operation is complete, every server immediately sees the
 *      updated value.
 *    - No user can read stale (outdated) data.
 *
 *    How does it work?
 *    - Before confirming a write operation, the system makes sure that all
 *      required nodes have agreed on and stored the latest data.
 *    - Only after this synchronization is complete does the system consider the
 *      write successful.
 *    - This is usually achieved using distributed consensus protocols.
 *
 *    Example: Banking Transaction
 *    - Your account balance is ₹10,000.
 *    - You transfer ₹5,000.
 *    - The system updates every required server before confirming the transfer.
 *    - If you immediately check your balance from any server, it will always show
 *      ₹5,000.
 *    - No server is allowed to return the old balance.
 *
 *    Pros
 *    - Always returns the latest data.
 *    - Prevents stale reads.
 *    - Maintains high data accuracy.
 *    - Ideal for critical business operations.
 *
 *    Cons
 *    - Higher latency because servers must synchronize first.
 *    - Lower availability during network failures.
 *    - More complex to implement.
 *   
 *    Best For
 *    - Banking Systems
 *    - Payment Systems
 *    - Stock Trading
 *    - Inventory Management
 *    - Financial Applications
 * 
 * 
 * 2. EVENTUAL CONSISTENCY
 *
 *    What is it?
 *    - All nodes will eventually have the same data, but not immediately.
 *    - Right after a write operation, some servers may still have old (stale)
 *      data for a short period of time.
 *    - Once the update has propagated to every server, all nodes become
 *      consistent.
 *   
 *    How does it work?
 *    - The system first accepts the write request and responds to the user.
 *    - The updated data is then replicated to other servers asynchronously
 *      (in the background).
 *    - During this synchronization period, some servers may temporarily return
 *      stale data.
 *   
 *    Example: Social Media Post
 *    - You upload a new profile picture.
 *    - The server handling your request immediately stores the new picture and
 *      responds with "Profile Updated Successfully."
 *    - Another user viewing your profile from a different server may still see
 *      the old profile picture for a few seconds.
 *    - After the update is replicated to all servers, everyone sees the new
 *      profile picture.
 *   
 *    Pros
 *    - High Availability.
 *    - Better Performance.
 *    - Lower Latency.
 *    - Scales well for large distributed systems.
 *   
 *    Cons
 *    - Users may temporarily see stale (outdated) data.
 *    - Different users may see different data for a short period.
 *    - Not suitable for applications that require immediate accuracy.
 *   
 *    Best For
 *    - Social Media
 *    - DNS
 *    - Product Catalogs
 *    - News Feeds
 *    - Video Streaming Platforms
 *    - Large Distributed Databases
 */


/**
 * AVAILABILITY
 * - Availability means a system is able to respond to user requests.
 * - The goal is to keep the application accessible even when some components
 *   fail.
 * - A highly available system minimizes downtime and continues serving users
 *   whenever possible.
 * 
 * Example: Online Banking
 * - A user wants to check their account balance.
 * - Even if one of the banking servers crashes, the application should still
 *   respond to the user's request.
 * - The response may come from another healthy server instead of the failed one.
 * 
 * HOW TO ACHIEVE HIGH AVAILABILITY
 * > High Availability is achieved using multiple techniques such as:
 *   - Replication
 *   - Redundant Servers
 *   - Load Balancing
 *   - Automatic Failover
 *   - Fault Tolerance
 *   - Health Monitoring
 *
 *
 * 1. HIGH AVAILABILITY (HA)
 *
 *    What is it?
 *    - High Availability (HA) is a system design approach that keeps an application
 *      running even if one or more servers fail.
 *    - The goal is to minimize downtime and keep the service accessible to users.
 *   
 *    How does it work?
 *    - Multiple servers run the same application.
 *    - A Load Balancer distributes requests among them.
 *    - If one server fails, traffic is automatically redirected to healthy
 *      servers.
 *   
 *    Example: Online Banking
 *    - The banking application is running on three servers.
 *    - One server suddenly crashes.
 *    - The Load Balancer detects the failure and sends new requests to the
 *      remaining two servers.
 *    - Users can still log in, check their balance, and perform transactions.
 *   
 *    Pros
 *    - Minimal downtime.
 *    - Better user experience.
 *    - Handles server failures automatically.
 *    - Supports millions of users.
 *   
 *    Cons
 *    - Higher infrastructure cost.
 *    - More complex architecture.
 *    - Requires monitoring and failover mechanisms.
 *   
 *    Best For
 *    - Banking Applications
 *    - E-commerce Platforms
 *    - Cloud Services
 *    - Healthcare Systems
 *    - Online Gaming
 * 
 * 
 * 2. FAULT TOLERANCE
 *
 *    What is it?
 *    - Fault Tolerance is the ability of a system to continue operating even when
 *      one or more components fail.
 *    - Unlike High Availability, which may briefly switch to another server,
 *      Fault Tolerance aims to continue running with little or no interruption.
 *   
 *    How does it work?
 *    - The system uses redundant hardware, software, or network paths.
 *    - If one component fails, another immediately takes over with little or no
 *      interruption.
 *   
 *    Example
 *    a. Airplane
 *       - An airplane has multiple engines.
 *       - If one engine fails, the airplane can continue flying safely.
 *    b. Distributed System
 *       - A server crashes while processing requests.
 *       - Another server immediately takes over without interrupting users.
 *   
 *    Pros
 *    - Continuous operation during failures.
 *    - Higher reliability.
 *    - Reduced service interruption.
 *   
 *    Cons
 *    - Expensive to build.
 *    - More complex system design.
 *    - Requires redundant resources.
 *   
 *    Best For
 *    - Banking
 *    - Aviation
 *    - Healthcare
 *    - Cloud Infrastructure
 *    - Critical Enterprise Systems
 */
```

```python

/**
 * Domain Name System (DNS):
 * - In this session, we'll deep dive into:
 *   - Domain Name System (DNS)
 *   - Load Balancer
 * - These two are the first points of contact when a client connects to any large-scale
 *   system. 
 * - We'll see how they route, distribute, and scale traffic across the internet.
 * 
 * What happens whe you type a URL into your browser?
 * - Bob enters a URL into the browser
 * - Browser looks up IP in cache (DNS Cache)
 * - Browser looks up IP using recursive DNS Lookup
 *   - DNS Resolver
 *   - Recursive lookup
 *   - DNS Server
 * - Browser establishes TCP connection with the web server
 * - Browser sends HTTP request to the web server
 * - Server sends back HTTP response
 * - Brower renders HTTP content
*/


/**
 * DOMAIN NAME SYSTEM (DNS)
 * > In this section, we'll learn about:
 *   - Domain Name System (DNS)
 *   - Load Balancer
 * > These are usually the first components a client interacts with when accessing
 *   a large-scale distributed system.
 * > They help users find the correct server, route requests, distribute traffic,
 *   and improve the scalability and availability of applications.
 *
 * WHY DO WE NEED DNS?
 * - Every server on the Internet has a unique IP address, such as: 142.250.183.78
 * - Computers communicate using IP addresses.
 * - However, remembering IP addresses for every website is difficult.
 *
 * - Instead, we use easy-to-remember domain names such as:
 *   - google.com
 *   - amazon.com
 *   - youtube.com
 *
 * - DNS (Domain Name System) acts like the Internet's phonebook.
 * - It translates a domain name into its corresponding IP address so that the
 * - browser knows which server to connect to.
 *
 * DNS Server Types:
 * - There are 4 DNS Servers involved in loading a webpage:
 *   1. Recursive Resolvers
 *   2. Root Name Servers
 *   3. TLD Name Servers
 *   4. Authoritative Name Servers
 *
 * How do they work together?
 * - Suppose the browser wants the IP address of: www.google.com
 * - The browser first sends the request to the Recursive Resolver.
 * - The Recursive Resolver now becomes responsible for finding the IP address.
 * - Since it doesn't know the answer, it starts asking other DNS Servers.
 *
 *
 * 1. Recursive Resolver:
 *    - The Recursive Resolver is the first DNS server contacted by the browser.
 *    - It acts like a middleman between the browser and the DNS system.
 *    - Its responsibility is to find the correct IP address and return it to the browser.
 *    - It checks whether it already has the answer in its DNS Cache.
 *      - If found, it immediately returns the IP address.
 *      - Otherwise, it sends a request to the Root Name Server.
 *
 *
 * 2. Root Name Server:
 *    - The Root Name Server is the starting point of the DNS hierarchy.
 *    - Root Server don't store specific domain records like google.com's IP
 *      [It does NOT know the IP address of every website.]
 *    - Instead, it points you to the correct Top-Level Domain (TLD) server.
 *    - It tells the Recursive Resolver which TLD Name Server to contact.
 *    - Example:
 *      - Browser asks for: www.google.com
 *      - The Root Server replies:
 *        "I don't know the IP address, but the '.com' TLD server does."
 *
 *
 * 3. TLD Name Server:
 *    - A TLD Server manages domains belonging to a specific top-level domain.
 *    - Examples:
 *      a. Generic     : .com, .org, .net, .edu 
 *      b. Country Code: .in, .uk, .ca 
 *      c. New TLDs    : .app, .tech, .ai
 *    - It doesn't know the final IP address either.
 *    - Instead, it knows which Authoritative Name Server manages that domain.
 *    - It tells the Recursive Resolver which Authoritative Name Server to contact.
 *    - Example:
 *      - The .com TLD Server replies:
 *      - "I don't know Google's IP address, but Google's Authoritative Name Server does."
 *
 *
 * 4. Authoritative Name Server:
 *    - The Authoritative Name Server stores the actual DNS records for a domain, such as:
 *      a. A record  -> IP address
 *      b. MX record -> mail server
 *      c. CNAME     -> alias
 *      d. NS record -> delegated name servers
 *    - It is the final and authoritative source of truth for that domain.
 *    - The Recursive Resolver sends the domain name to the Authoritative Name Server.
 *    - It looks up the requested DNS record and returns the correct IP address to the
 *      Recursive Resolver.
 *    - Example:
 *      google.com -> A = 142.250.183.78
 *
 *    The Recursive Resolver then:
 *    - Stores the result in its DNS Cache.
 *    - Returns the IP address to the browser.
 *
 *    Finally, the browser uses that IP address to establish a connection with the
 *    web server.
 * 
 * 
 * DNS Caching:
 * - Temporary storage of DNS records to speed up domain resolution.
 * - Reduces repeated DNS lookups -> faster load times, lower bandwidth & CPU usage.
 * - DNS data is cached in multiple layers, each using a TTL (Time-to-Live) value.
 * 
 * Where does DNS Caching Occur?
 * - Browser Cache
 * - Operating System (OS) cache
 * - Recursive Cache (ISP or public DNS like Cloudflare DNS or Google DNS)
 * 
 * 
 * 1. Browser DNS Caching:
 *    - Modern browsers store DNS records locally for quick retrieval.
 *    - Browser cache is the first location checked during a DNS lookup.
 *    - Minimizes network hops -> fastest possible resolution.
 *    - Example: In chrome: view DNS cache at - chrome://net-internals/#dns
 * 
 * 2. OS Level DNS Caching:
 *    - OS (Stub Resolver) Caching
 *    - The OS runs a stub resolver that handles DNS requests from applications.
 *    - Checks its own DNS cache before sending queries out of the network.
 *    - If not cached -> sends a recursive query to the ISP's resolver.
 * 
 * 3. Recursive Resolver Caching:
 *    - ISP/Public Recursive Resolvers store DNS records to reuse across many users.
 *    - Caching Scenarios:
 *      - Has authoritative NS records -> direct query those authoritative servers
 *        (skip root + TLD)
 *      - Has TLD NS records    -> Query TLD servers (skip root)
 *      - Has no cached records -> Query root > TLD > authoritative (full resolution)
 * 
 * Note:
 * 1. Name Servers:
 *    - DNS Servers that respond to user's queries are called name servers.
 * 
 * 2. Resource records:
 *    - The DNS database stores domain name to IP address mappings in the form of 
 *      resource records (RR).
*/


/**
 * WHAT HAPPENS WHEN YOU TYPE A URL INTO YOUR BROWSER?
 * - User enters: https://www.google.com
 *
 *
 * Step 1: Browser checks the DNS Cache
 * - The browser first checks its DNS Cache.
 * - DNS Cache is temporary storage that stores recently resolved domain names
 *   and their IP addresses in Browser Cache and Operating System (OS) cache which 
 *   reduces repeated DNS lookups -> faster load times, lower bandwidth & CPU usage.
 * - Example: google.com → 142.250.xxx.xxx
 * - If the IP address is found, the browser immediately uses it.
 * - Otherwise, it performs a DNS lookup.
 *
 * Step 2: Browser contacts a DNS Resolver
 * - The browser sends the request to a DNS Resolver.
 * - A DNS Resolver is a special DNS server (usually provided by your ISP,
 *   Google DNS, or Cloudflare DNS).
 * - Its responsibility is to find the IP address on behalf of the browser.
 * - Browser asks: "What is the IP address of www.google.com?"
 * - The browser now waits while the DNS Resolver searches for the answer.
 *
 *
 * Step 3: Recursive DNS Lookup
 * - The DNS Resolver now performs a Recursive DNS Lookup.
 * - Recursive means the Resolver is responsible for finding the final answer.
 * - If the Resolver already has the answer in its own DNS Cache like our local ISP 
 *   or public DNS like Cloudflare/Google DNS, it immediately returns the IP address 
 *   to the browser.
 * - Otherwise, it starts contacting other DNS Servers.
 *
 *
 * 3.1 Root Name Server
 * - The Recursive Resolver first contacts a Root Name Server.
 * - The Root Name Server does NOT know the IP address of every website.
 * - Instead, it knows which Top-Level Domain (TLD) Name Server should be
 *   contacted.
 *
 * - Example:
 *   - Resolver asks: "What is the IP address of www.google.com?"
 *   - Root Server replies: "I don't know Google's IP address, but ask the .com 
 *     TLD Name Server."
 *
 *
 * 3.2 TLD (Top-Level Domain) Name Server
 * - The Recursive Resolver now contacts the .com TLD Name Server.
 * - TLD Servers manage domains belonging to a specific top-level domain.
 * - Examples:
 *   - .com
 *   - .org
 *   - .net
 *   - .in
 *   - .edu
 * - The TLD Server doesn't know the final IP address either.
 * - Instead, it knows which Authoritative Name Server manages google.com.
 *
 * - Example: TLD Server replies:
 *   "I don't know Google's IP address, but ask Google's Authoritative Name Server."
 *
 *
 * 3.3 Authoritative Name Server
 * - The Recursive Resolver now contacts Google's Authoritative Name Server.
 * - This server stores the actual DNS records for google.com.
 * - It is the final source of truth for that domain.
 * - Example:
 *   - google.com -> 142.250.183.78
 *   - The Authoritative Name Server returns the IP address to the Recursive
 *     Resolver.
 *
 *
 * 3.4 Return Result
 * - The Recursive Resolver stores the IP address in its DNS Cache.
 * - It returns the IP address to the browser.
 * - The browser also stores the IP address in its own DNS Cache for future
 *   requests.
 *
 *
 * Step 4: Establish TCP Connection
 * - Now the browser knows the server's IP address.
 * - It establishes a TCP (Transmission Control Protocol) connection with the
 *   web server.
 * - TCP creates a reliable communication channel, ensuring data is delivered
 *   completely, in the correct order, and without duplication.
 *
 *
 * Step 5: Browser Sends an HTTP Request
 * - Once the TCP connection is established, the browser sends an HTTP/HTTPS
 *   request.
 * - Examples of requested resources:
 *   - HTML page
 *   - CSS file
 *   - JavaScript file
 *   - Image
 *   - API data
 *
 * Step 6: Server Sends an HTTP Response
 * - The server processes the request and sends back an HTTP Response.
 * - The response usually contains:
 *   - Status Code (200, 404, 500, etc.)
 *   - Headers
 *   - Response Body (HTML, JSON, Images, etc.)
 *
 * Step 7: Browser Renders the Web Page
 * - The browser parses the HTML.
 * - Downloads CSS, JavaScript, fonts, and images.
 * - Executes JavaScript.
 * - Applies CSS styles.
 * - Finally, renders the complete web page on the user's screen.
 */



/**
 * LOAD BALANCER:
 * - Imagine millions of users opening www.google.com at the same time.
 * - If every request is sent to a single server:
 *   - The server may become overloaded.
 *   - Response time will increase.
 *   - Eventually, the server may crash.
 * - To solve this problem, modern applications use multiple servers instead of relying
 *   on a single server.
 * - The next question is:
 *   - "How do we decide which server should handle each user's request?"
 *   - This is where a Load Balancer comes in.
 *
 *
 * WHAT IS A LOAD BALANCER?
 * - A Load Balancer is a component that sits in front of a group of servers.
 * - Its job is to receive incoming client requests and distribute them fairly
 *   among the available servers.
 * - Instead of one server handling all requests, the work is shared across many
 *   servers.
 *
 * Example:
 *
 * Client Requests
 *        │
 *        ▼
 *   Load Balancer
 *      ├────────► Server 1
 *      ├────────► Server 2
 *      ├────────► Server 3
 *      └────────► Server 4
 *
 * - Millions of requests can arrive every second in a large-scale system.
 * - Thousands (or even hundreds of thousands) of servers work together, and the
 *   Load Balancer decides which server should handle each request.
 *
 *
 * WHY DO WE NEED A LOAD BALANCER?
 * A Load Balancer provides several important benefits.
 *
 * 1. Scalability
 *    - As the number of users grows, we can simply add more servers.
 *    - The Load Balancer automatically starts sending requests to the new servers.
 *    - This allows the application to handle increasing traffic without changing
 *      the client.
 *
 * 2. Availability
 *    - Sometimes a server may crash or become unavailable.
 *    - The Load Balancer detects unhealthy servers using health checks.
 *    - Instead of sending requests to the failed server, it redirects traffic to
 *      healthy servers.
 *    - As a result, the application remains available to users.
 *
 * 3. Performance
 *    - The Load Balancer can send requests to servers with less load.
 *    - This reduces response time and improves the user experience.
 *    - No single server becomes overloaded while others remain idle.
 *
 *
 * WHERE CAN WE PLACE LOAD BALANCERS?
 * - Load Balancers can be placed at different layers of a distributed system.
 *
 *
 * 1. Between Client and Web Server
 *    - Distributes incoming user requests among multiple web servers.
 * 
 *      Client
 *         │
 *         ▼
 *     Load Balancer
 *         │
 *         ▼
 *    Multiple Web Servers
 *
 * 2. Between Web Server and Application Server
 *    - Distributes business logic requests among application servers.
 * 
 *      Web Servers
 *           │
 *           ▼
 *      Load Balancer
 *           │
 *           ▼
 *      Multiple Application Servers
 *
 * 3. Between Application Server and Database Server
 *    - Distributes database requests among database replicas or clusters.
 * 
 *      Application Servers
 *           │
 *           ▼
 *      Load Balancer
 *           │
 *           ▼
 *      Multiple Database Servers
 *
 * SUMMARY
 * - DNS answers: "Which application should I connect to?"
 * - Load Balancer answers: "Which server should handle this request?"
 * - Together they ensure that requests are routed efficiently and that the
 *   application remains scalable, highly available, and performant.
 */

/**
 * Load Balancing Algorithms:
 * - A Load Balancing Algorithm is a set of rules that determines how incoming
 *   requests are distributed among the available servers.
 * - Broadly, there are two categories:
 *   1. Global Load Balancer
 *   2. Local Load Balancer
 *      - Static Load Balancing
 *      - Dynamic Load Balancing
 * 
 * Global Load Balancing:
 * - Distribute traffic across servers or data centers located in multiple geographic
 *   regions.
 *
 * Local Load Balancer:
 * 
 * 1. Static Loading Balancing
 *    - In Static Load Balancing, the decision is made using predefined rules.
 *    - The Load Balancer does NOT consider:
 *      - Current CPU usage
 *      - Memory usage
 *      - Number of active connections
 *      - Current server health
 *    - It simply follows a fixed algorithm to distribute requests.
 *    - Static algorithms are simple, fast, and have very little overhead.
 * 
 *    Common Static Load Balancing Algorithms:
 *    - Round Robin
 *    - Weighted Round Robin
 *    - IP Hash
 * 
 *    Round Robin:
 *    > Round Robin distributes requests one by one in a circular order.
 *    > Example: 
 *      - Servers: A → B → C
 *      - Requests:
 *        - Request 1 → Server A
 *        - Request 2 → Server B
 *        - Request 3 → Server C
 *        - Request 4 → Server A
 *        - Request 5 → Server B
 *        - Request 6 → Server C
 *      - After the last server, it starts again from the first server.
 *    > Advantages
 *      - Very simple to implement.
 *      - Fair distribution when all servers have similar capacity.
 *      - Minimal processing overhead.
 *    > Disadvantages
 *      - Assumes every server has the same processing power.
 *      - Doesn't consider CPU, memory, or current load.
 *      - Doesn't know whether a server is busy or slow.
 *      - May overload weaker servers.
 *    > Best for:
 *      - Homogenous Servers
 *      - Small to medium applications
 * 
 *    Weighted Round Robin:
 *    > Not every server has the same hardware.
 *    > Example:
 *      - Server A → 32 CPU Cores
 *      - Server B → 16 CPU Cores
 *      - Server C → 8 CPU Cores
 *    > Giving every server the same number of requests is unfair.
 *    > Each server is assigned a weight based on its capacity.
 *    > Servers with higher weights receive more requests.
 *    > Example:
 *      - Server A (Weight = 4)
 *      - Server B (Weight = 2)
 *      - Server C (Weight = 1)
 *    > Request Distribution:
 *      - A → A → A → A
 *      - B → B
 *      - C
 *      - Then repeat.
 *    > Advantages
 *      - Better utilization of server resources.
 *      - Prevents smaller servers from becoming overloaded.
 *      - Easy to adjust when server hardware changes.
 *    > Disadvantages
 *      - Still ignores the current runtime load.
 *      - A powerful server may already be busy but continues receiving more requests.
 *    > Best For
 *      - Heterogeneous server environments.
 *
 * 
 *    IP Hash Load Balancing:
 *    > Some applications require a user to continue interacting with the same server.
 *    > Example:
 *      - Shopping Cart
 *      - User Login Session
 *    > If every request goes to a different server, the user's session may be lost.
 *    > The Load Balancer creates a hash using the client's IP address.
 *    > The generated hash determines which backend server will handle the request.
 *    > Since the client's IP usually remains the same, future requests from that
 *      client are routed to the same server.
 * 
 *    Example
 *    - Every future request from 192.168.1.10 will also be sent to Server B.
 *    - Client IP: 192.168.1.10
 *                      │
 *                      ▼
 *                  Hash Function
 *                      │
 *                      ▼
 *                   Server B
 *   
 *    Advantages
 *    - Maintains session consistency.
 *    - Predictable routing.
 *    - Useful for stateful applications.
 *    
 *    Disadvantages
 *    - Can produce uneven load distribution.
 *    - Large organizations behind the same public IP may overload one server.
 *    - Adding or removing servers changes the hash mapping, potentially breaking
 *      existing sessions.
 *
 *
 *    Best for:
 *    - Shopping Carts
 *    - User login Sessions
 *    - Stateful Applications
 * 
 * 
 * 2. Dynamic Load Balancing:
 *    - Decisions are made at runtime, based on current system state (load, latency,
 *      health)
 *    
 *    How it works?
 *    - Continuously monitors servers
 *    - Routes work to the least-loaded or healthiest resource
 * 
 *    Common Strategies:
 *    - Least connections
 *    - Least response time
 *    - Weighted load balancing
 *    - Resource-based Scheduling
 * 
 *    Least Connections:
 *    - When a new request arrives, the load balancer inspects all servers and forwards
 *      the request to the one with the lowest connection count.
 *    - Directs traffic away from busy servers, stopping any single node from becoming
 *      bottleneck.
 *    - Ideal for applications where sessions or requests take different amounts of
 *      time (e.g., streaming, database queries).
 * 
 *    Least Response Time:
 *    - The load balancer continuously checks the number of active connections and the
 *      average time it takes for each server to respond (Time-to-First-Byte: TTFB)
 *    - Incoming requests are sent to the server that responds fastest.
 *    - If a server gets busy or slow, its score increases, and traffic is shifted to
 *      other, faster servers.
 *    - Resource-Intensive: Requires continuous monitory and calculation, unlike simpler
 *      methods.
 *    - Ideal for: Real-time apps, gaming, VoIP, and e-commerce where speed is critical.
 * 
 *    Weighted Load Balancing:
 *    - Weighted least connection gives administrators the ability to assign different
 *      weights to each server, assuming that some servers can handle more connections
 *      than others.
 *    - Weighted response time averages the response time of each server, and combines
 *      that the number of connections each server has open to determine where to
 *      send traffic.
 * 
 *    Resource-based Scheduling:
 *    - Small monitoring programs(agents) run on each server, constantly gathering data
 *      on resource availability (CPU, RAM, active connections)
 *    - When a new request arrives, the load balancer checks the agents and directs the
 *      traffic to the server with the most free resources.
 *    - Ideal for:
 *      - Applications with varied workloads, where some requests are resource-intensive
 *        and others are light.
 *      - Real-time server health monitoring for smarter routing.
*/ 
```

```python
/**
 * Types of Databases:
 * 1. Relational Databases
 *    - Data stored in relational databases has prior structure
 *    - Organizes data into one or more relations (also called tables), with a 
 *      unique key for each tuple (instance)
 * 2. Non-Relational Databases
 *    - Manages data in more flexible ways
 *    - Suitable for handling large volumes of unstructured, semi-structured, or
 *      rapidly changing data.
 * 
 * Key Characteristics:
 * a. Schema-less or Flexible Schema:
 *    - Each record (or document) can have a different structure.
 * b. Scalability: 
 *    - Add more servers to handle more data easily.
 * c. High performance for specific use cases:
 *    - Optimized for read-heavy or write-heavy operations, depending on type.
*/

/**
 * Why Choose Relational Database?
 * 1. Reduced Redundancy (Normalization)
 *    - Reduces duplicate data and inconsistent entries.
 *    - Keeps the database clean, reliable, and easier to maintain.
 * 2. Concurrecy
 *    - Multiple users can read and write data at the same time safely.
 *    - Relational databases use transactions and ACID properties to prevent conflicts.
 * 3. Integration
 *    - Allow different applications to share and access data easily.
 *    - Example: Aggregating sales, inventory, and HR data from multiple apps into
 *      a central database.
 * 4. Backup and Disaster Recovery
 *    - Maintain data consistency at all times.
 *    - Support backup, restore, and mirroring, so data is safe even during failures.
 * 
 * Drawback:
 * - A single aggregated value in the view is composed of several rows and tables in
 *   the relational database.
*/

/**
 * Why Choose NoSQL Database?
 * - Designed for wide varietry of data models
 * - Handles large volumes of semi structured and unstructured data.
 * - Optimized for low latency and flexible schemas
 * - This can be achieved by relaxing consistency restrictions of other databases.
 * 
 * Characteristics of NoSQL Databases:
 * 1. Simple Design
 *    - Data stored in single documents
 *    - Fewer joins, less code
 * 2. Horizontal Scaling
 *    - Runs on large clusters
 *    - Easy scale-out with more nodes
 * 3. High Availability
 *    - Built-in replication
 *    - No downtime or node failure
 * 4. Flexible Data Model
 *    - Supports unstructured & semi structured data
 *    - Schema-less (JSON, XML, BSON)
 * 5. Low Latency
 *    - Optimized for fast reads/writes 
 *    - Suitable for real-time applications
 * 6. Cost Effective
 *    - Open-source options
 *    - Runs on commodity hardware
*/


/**
 * When to use SQL Database?
 * 1. Use Relational Databases when:
 *    - Data is highly structured (tables, rows, columns)
 *    - Strong ACID guarantees are required
 *    - Data integrity and consistency are critical
 *    - Complex Joins and Relationships exist
 *    - Schema is stable or changes infrequently
 *    - Business logic relies on transactions
 * 2. Typical Use Cases:
 *    - Banking & financial systems
 *    - Payment processing
 *    - Inventory & order management
 *    - ERP/CRM systems
 * 
 * When to use NoSQL Database?
 * 1. Use NoSQL Databases when:
 *    - Data is semi-structured or unstructured
 *    - Schema changes frequently
 *    - Application requires horizontal scalability
 *    - Low latency is more important than strict consistency
 *    - Handling large volumes of data
 *    - High availability is required
 * 2. Typical Use Cases:
 *    - Social media platforms
 *    - Real-time analytics
 *    - Logging & monitoring systems
 *    - IoT & event-driven systems
*/

/**
 * Database Replication (Master-Slave Model)
 * a. Master DB:
 *    - Handles all write operations
 *    - Insert, Update, Delete
 * b. Slave DBs
 *    - Handle read operations
 *    - Receive replicated data from master
 * 
 * - Used in read-heavy systems
 * - Typically more slaves than masters
 * - Separate reads and writes to improve scalability
 * 
 * Why use Database Replication?
 * a. Better Performance
 *    - Reads distributed across slaves
 *    - Parallel query processing
 * b. High Availability
 *    - System continues even if a DB fails
 * c. Reliability
 *    - Data replicated across multiple nodes
 *    - Reduced risk of data loss
 * 
 * Failure Handling:
 * a. Slave Failure:
 *    - Redirect reads to other slaves or master
 *    - Replace failed slave
 * b. Master failure:
 *    - Promote a slave to master
 *    - Recover missing data if needed
*/

/**
 * Database Replication System Flow:
 *  
 *          +------------------------+  www.mysite.com
 *          |    +----+    +----+    | ----------------> +-----+
 *          |    |    |    |    |    |                   | DNS |
 *          |    +----+    +----+    | <---------------- +-----+
 *          |      PC      Mobile    |     IP Address
 *          +------------------------+  
 *                 |         |
 *  www.mysite.com |         | api.mysite.com
 *                 |         |
 *                 +---------+
 *                      |
 *                      |
 *                      V
 *                  +-------+
 *                  |       | Load Balancer
 *                  +-------+
 *                      |
 *                      V
 *          +------------------------+
 *          |    +----+    +----+    |
 *          |    |    |    |    |    | Web tier
 *          |    +----+    +----+    | 
 *          |   Server-1  Server-2   |
 *          +------------------------+
 *                  |         |
 *                  |         |
 *         Write    |         |  Read
 *            +-----+---------+----+
 *            |                    |
 *            V                    V
 *     +-----------------------------------+
 *     |    +----+  Replication  +----+    |
 *     |    |    |<------------->|    |    | Data Tier
 *     |    +----+               +----+    |
 *     |   Master DB            Slave DB   |
 *     +-----------------------------------+
 * 
 * - A user gets the IP address of the load balancer from DNS.
 * - A user connected the load balancer with this IP address. 
 * - The HTTP request is routed to either Server 1 or Server 2.
 * - A web server reads user data from a slave database.
 * - A web server routes any data-modifying operations to the master database.
 *   This includes write, update and delete operations.
*/
```

```python
/**
 * Cache:
 * a. What is a Cache?
 *    - Temporary storage for frequently accessed or expensive data
 *    - Stored in memory for fast access
 *    - Reduces repeated database calls
 * b. Cache Tier
 *    - Separate layer between web server and database
 *    - Much faster than db
 *    - Can be scaled independently
 * c. Benefits:
 *    - Faster response time
 *    - Reduced database load
 *    - Better system performance
 * 
 * 
 *            1. If data exists in cache,
 *               read data from cache
 *  +-----+ <------------------------------+ +-----+                                     +-----+
 *  |     |                                  |     | <---------------------------------- |     |
 *  +-----+ <------------------------------- +-----+  2.1 If data doesn't exist in cache +-----+
 * Web Server 2.2 Return data to web server   Cache         save data to cache              DB
*/ 

/**
 * Cache Design Considerations:
 * 1. When to use Cache
 *    - Read-heavy data
 *    - Infrequently updated data
 *    - Not suitable for persistent storage (memory is volatile)
 * 2. Key Considerations:
 *    a. Expiration Policy
 *       - Avoid too short (DB overload)
 *       - Avoid too long (stale data)
 *    b. Consistency
 *       - Cache & DB updates not atomic
 *       - Harder across multiple regions
 *    c. Failure Handling
 *       - Avoid Single Point of Failure (SPOF)
 *       - Use multiple cache servers
 *       - Overprovision memory
 *    d. Eviction Policy
 *       - LRU (most common)
 *       - LFU, FIFO (based on use case)
*/

/**
 * System Design:
 * 
 *  
 *          www.mysite.com    +------------------------+
 * +-----+ ---------------->  |    +----+    +----+    |
 * | DNS |                    |    |    |    |    |    |
 * +-----+ <----------------  |    +----+    +----+    |
 *             IP Address     |      PC      Mobile    |
 *                            +------------------------+  
 *                                   |         |
 *                    www.mysite.com |         | api.mysite.com
 *                                   |         |
 *                                   +---------+
 *                                        |
 *                                        |
 *                                        V
 *                                    +-------+
 *                                    |       | Load Balancer
 *                                    +-------+
 *                                        |
 *                                        V
 *                            +------------------------+
 *                            |    +----+    +----+    |
 *                            |    |    |    |    |    | Web tier
 *                            |    +----+    +----+    | 
 *                            |   Server-1  Server-2   |
 *                            +------------------------+
 *                                    |         |
 *                                    |         |
 *                           Write    |    Read |           Cache
 *                     +--------------------+-------------------+
 *                     |                    |                   |
 *                     V                    V                   V
 *              +-----------------------------------+         +-----+
 *              |    +----+  Replication  +----+    |         |     |
 *              |    |    |<------------->|    |    |         |     |
 *              |    +----+               +----+    |         +-----+
 *              |   Master DB            Slave DB   |          Cache
 *              +-----------------------------------+
 *                            Data Tier
*/


/**
 * Content Delivery Network (CDN):
 * 1. What is CDN?
 *    - A network of geographically distributed servers.
 *    - Used to deliver static content: images, videos, CSS, JS files
 * 2. Why CDN?
 *    - Serve content from the closest server to the user
 *    - Reduces latency and page load time
*/

/**
 * System Design:
 * 
 *  
 *          www.mysite.com    +------------------------+
 * +-----+ ---------------->  |    +----+    +----+    |         +-----+
 * | DNS |                    |    |    |    |    |    |-------> |     |
 * +-----+ <----------------  |    +----+    +----+    |         +-----+
 *             IP Address     |      PC      Mobile    |           CDN
 *                            +------------------------+  
 *                                   |         |
 *                    www.mysite.com |         | api.mysite.com
 *                                   |         |
 *                                   +---------+
 *                                        |
 *                                        |
 *                                        V
 *                                    +-------+
 *                                    |       | Load Balancer
 *                                    +-------+
 *                                        |
 *                                        V
 *                            +------------------------+
 *                            |    +----+    +----+    |
 *                            |    |    |    |    |    | Web tier
 *                            |    +----+    +----+    | 
 *                            |   Server-1  Server-2   |
 *                            +------------------------+
 *                                    |         |
 *                                    |         |
 *                           Write    |         | Read         Cache
 *                     +--------------------+-------------------+
 *                     |                    |                   |
 *                     V                    V                   V
 *              +-----------------------------------+         +-----+
 *              |    +----+  Replication  +----+    |         |     |
 *              |    |    |<------------->|    |    |         |     |
 *              |    +----+               +----+    |         +-----+
 *              |   Master DB            Slave DB   |          Cache
 *              +-----------------------------------+
 *                            Data Tier
*/


/**
 * Stateful vs Stateless Web Tier:
 * 
 * 1. Stateful Web Tier:
 *    - Web Server stores client state
 *      - Session data
 *      - User-specific information
 *    - Each user must be routed to the same server
 *    - Requires sticky sessions at load balancer
 * 
 *    Problems:
 *    - Hard to scale horizontally
 *    - Difficult to handle server failures
 *    - Tight coupling between user and server
 * 
 *         User-A                      User-B                      User-C
 *         +-----+                     +-----+                     +-----+             
 *         |     |                     |     |                     |     |             
 *         +-----+                     +-----+                     +-----+             
 *            | http request              | http request              | http request   
 *            V                           V                           V                 
 *         +-----+                     +-----+                     +-----+             
 *         |     |                     |     |                     |     |             
 *         +-----+                     +-----+                     +-----+             
 *         Server-1                    Server-2                    Server-3             
 *   - Session data to User-A    - Session data to User-B    - Session data to User-C   
 *   - Profile Image for User-A  - Profile Image for User-B  - Profile Image for User-C 
 * 
 * 2. Stateless Web Tier:
 *    - Web Servers do not store session state
 *    - State is stored in shared persistent storage
 *      - Relational DB
 *      - NoSQL
 *      - Redis/Memcached 
 *    - How it works
 *      - Any request can go to any web server
 *      - Web Server fetches state from shared storage
 *      - Servers are interchangeable
 *    - Result: Simpler, Scalable, and more robust system
 * 
 *         User-A                      User-B                      User-C
 *         +-----+                     +-----+                     +-----+             
 *         |     |                     |     |                     |     |             
 *         +-----+                     +-----+                     +-----+             
 *            |                           |                           |
 *            | http request              | http request              | http request   
 *            |                           |                           |
 *            +---------------------------+---------------------------+     
 *                                        |
 *                                        V
 *                                     +-----+ 
 *                                     |     | 
 *                                     +-----+
 *                                Multiple Web Servers
 *                                        |
 *                                        | fetch state
 *                                        V
 *                                     +-----+
 *                                     |     | 
 *                                     +-----+
 *                                  Shared Storage
*/

/**
 * System Design:
 *  
 *          www.mysite.com    +------------------------+
 * +-----+ ---------------->  |    +----+    +----+    |         +-----+
 * | DNS |                    |    |    |    |    |    |-------> |     |
 * +-----+ <----------------  |    +----+    +----+    |         +-----+
 *             IP Address     |      PC      Mobile    |           CDN
 *                            +------------------------+  
 *                                   |         |
 *                    www.mysite.com |         | api.mysite.com
 *                                   |         |
 *                                   +---------+
 *                                        |
 *                                        |
 *                                        V
 *                                    +-------+
 *                                    |       | Load Balancer
 *                                    +-------+
 *                                        |
 *                                        V
 *                 +--------------------------------------------+
 *                 |    +----+    +----+    +----+    +----+    |
 *                 |    |    |    |    |    |    |    |    |    | Auto Scale
 *                 |    +----+    +----+    +----+    +----+    | 
 *                 |   Server-1  Server-2  Server-3  Server-4   |
 *                 +--------------------------------------------+
 *                                        |      
 *                                        |      
 *        Write                Read       V        Cache               NoSQL
 *          +--------------------+-------------------+-------------------+
 *          |                    |                   |                   |
 *          V                    V                   V                   V
 *   +-----------------------------------+         +-----+            +-----+
 *   |    +----+  Replication  +----+    |         |     |            |     |
 *   |    |    |<------------->|    |    |         |     |            |     |
 *   |    +----+               +----+    |         +-----+            +-----+
 *   |   Master DB            Slave DB   |          Cache
 *   +-----------------------------------+
 * 
 * Note: This architecture is only for one data center. What if our user's are distributed
 *        geographically? In that case, we might need different data center. 
*/


/**
 * Data Center:
 * a. Problem:
 *    - Website grows globally
 *    - Users access from different geographical regions
 * b. Solution:
 *    - Deploy multiple data centers across regions 
 *    - Improve:
 *      - Availability
 *      - Fault Tolerance
 *      - User Experience (low latency)
 * c. Key Concept: GeoDNS
 *    - DNS resolves domain to nearest data center
 *    - Traffic split across regions (e.g., US-East/US-West)
 * d. Traffic Redirection:
 *    - Use GeoDNS to route users correctly
 *    - Dynamically shift traffic during failures
 * e. Data Synchronization:
 *    - Data may differ across regions
 *    - Use cross-data center replication
 * f. Testing & Deployment
 *    - Must test applications from multiple regions
 *    - Use automated deployment tools
 *    - Ensure consistency across data centers
 * 
 *          www.mysite.com    +------------------------+
 * +-----+ ---------------->  |    +----+    +----+    |         +-----+
 * | DNS |                    |    |    |    |    |    |-------> |     |
 * +-----+ <----------------  |    +----+    +----+    |         +-----+
 *             IP Address     |      PC      Mobile    |           CDN
 *                            +------------------------+  
 *                                   |         |
 *                    www.mysite.com |         | api.mysite.com
 *                                   |         |
 *                                   +---------+
 *                                        |
 *                                        |
 *                                        V
 *                                    +-------+
 *                                    |       | Load Balancer
 *                                    +-------+
 *                    Geo-routed          |        Geo-routed
 *             +--------------------------+--------------------------+
 *             |                                                     |
 * DC1 US-East V                                         DC2 US-West V  
 * +-----------------------+                             +-----------------------+
 * | +----+ +----+  +----+ |                             | +----+ +----+  +----+ | 
 * | | WS | | WS |  | WS | |                             | | WS | | WS |  | WS | | 
 * | +----+ +----+  +----+ |                             | +----+ +----+  +----+ | 
 * |   |       |       |   |                             |   |       |       |   |
 * |   +-------+-------+---|--------------+--------------|---+-------+-------+   |
 * |           |           |              |              |           |           |
 * |   +---------------+   |              V              |   +---------------+   |
 * |   |               |   |           +-----+           |   |               |   |
 * | Databases      Caches |           |     |           | Databases      Caches |
 * +-----------------------+           +-----+           +-----------------------+
 *                                      NoSQL
*/


/**
 * Database Scaling:
 * a. Why Database Scaling?
 *    - Growing users & data volume
 *    - Single database becomes a bottleneck
 * b. Two Approaches:
 *    - Vertical Scaling (Scale Up)
 *    - Horizontal Scaling (Scale Out/Sharding)
 * 
 * Vertical vs Horizontal Scaling:
 * a. Vertical Scaling (Scale Up)
 *    - Add more CPU, RAM, Disk to one server
 *    - Simple to implement
 *    - Ex: Single powerful DB handling large traffic
 *    - Drawbacks:
 *      - Hardware limits
 *      - Single Point of Failure
 *      - Very Expensive
 *      - Limited long-term scalability
 * 
 * b. Horizontal Scaling (Sharding)
 *    - Add more database servers
 *    - Split data across multiple machines
 *    - Each shard stores a subset of data
 *    - Benefits:
 *      - Better scalability
 *      - Fault isolation
 *      - Handles large datasets
*/

/**
 * Database Sharding:
 * - Database split into shards
 * - Each shard has the same schema
 * - Data distributed using a sharding key
 * - Ex: user_id % 4
 * - Sharding key must evenly distribute data
 * 
 *           +-------------+
 *           | user_id % 4 |
 *           +-------------+
 *                  |
 *                  V
 *      +-------+-------+-------+
 *      |       |       |       |
 *    DB-0    DB-1    DB-2    DB-3
 * 
 * Database Sharding Techniques:
 * 1. Key-Based Sharding
 * 2. Range-Based Sharding
 * 3. Directory-Based Sharding
 * 4. Consistent Hashing
 * 
 * Challenges:
 * a. Resharding:
 *    - Needed when data grows or shards become imbalanced
 *    - Consistent Hashing helps
 * b. Hotspot/Celebrity Problem
 *    - Certain keys receive excessive traffic
 *    - May require dedicated shards
 * c. Joins & De-normalization
 *    - Joins across shards are difficult
 *    - Often solved by de-normalizing data
*/
```
