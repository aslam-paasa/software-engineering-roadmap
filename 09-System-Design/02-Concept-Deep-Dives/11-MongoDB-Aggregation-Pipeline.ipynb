```python
/**
 * ============================================================
 * Types of Databases
 * ============================================================
 *
 *   There are two broad categories of database management systems.
 *   Understanding when to use which is one of the most important
 *   decisions in system design.
 *
 *
 *   VISUAL — Two worlds of databases:
 *
 *    ┌─────────────────────────────────────────────────────────────┐
 *    │                    DATABASES                                 │
 *    └──────────────────────┬──────────────────────────────────────┘
 *                           │
 *               ┌───────────┴───────────┐
 *               │                       │
 *               ▼                       ▼
 *    ┌─────────────────┐      ┌──────────────────────┐
 *    │    Relational   │      │   Non-Relational      │
 *    │    (RDBMS)      │      │   (NoSQL)             │
 *    │                 │      │                       │
 *    │  Tables + Rows  │      │  Documents, Graphs,   │
 *    │  Fixed Schema   │      │  Key-Value, Columns   │
 *    │  SQL language   │      │  Flexible schema      │
 *    │  Strong ACID    │      │  Horizontal scaling   │
 *    │                 │      │                       │
 *    │  MySQL          │      │  MongoDB              │
 *    │  PostgreSQL     │      │  DynamoDB             │
 *    │  Oracle         │      │  Redis                │
 *    │  SQL Server     │      │  Cassandra            │
 *    └─────────────────┘      └──────────────────────┘
 *
 *
 * ───────────────────────────────────────────────────────
 * RDBMS — Relational Database Management System
 * ───────────────────────────────────────────────────────
 *
 *   - Stores data in tables (rows and columns).
 *   - Every row in a table has the same structure (fixed schema).
 *   - Tables can be related to each other using foreign keys.
 *   - Uses SQL (Structured Query Language) to query data.
 *   - Strong ACID guarantees out of the box.
 *   - Easy to query with JOINs, aggregations, and filters.
 *   - Harder to scale horizontally — designed for vertical scaling.
 *
 *
 *   VISUAL — Relational data in tables:
 *
 *    users table:                    orders table:
 *    ┌────┬──────────┬───────┐       ┌────┬─────────┬────────┐
 *    │ id │ name     │ email │       │ id │ user_id │ amount │
 *    ├────┼──────────┼───────┤       ├────┼─────────┼────────┤
 *    │  1 │ Hitesh   │ h@... │       │  1 │    1    │  ₹500  │
 *    │  2 │ Piyush   │ p@... │       │  2 │    1    │  ₹200  │
 *    │  3 │ Aman     │ a@... │       │  3 │    2    │  ₹800  │
 *    └────┴──────────┴───────┘       └────┴─────────┴────────┘
 *                │                              │
 *                └──────── JOIN on id ──────────┘
 *                          (relationship)
 *
 *   SELECT u.name, o.amount FROM users u
 *   JOIN orders o ON u.id = o.user_id
 *   WHERE u.id = 1
 *
 *   Result: Hitesh | ₹500
 *           Hitesh | ₹200
 *
 *
 * ============================================================
 * Non-Relational Databases (NoSQL)
 * ============================================================
 *
 *   - NoSQL does NOT mean "No SQL" — it means "Not Only SQL."
 *   - These databases do not store data in tables.
 *   - They are designed for huge scale, flexible data, and
 *     fast horizontal scaling.
 *   - Schema is flexible — different records can have
 *     completely different structures.
 *   - Trade-off: usually weaker JOINs and consistency,
 *     but much better at scale and speed.
 *
 *
 *   NoSQL comes in several types:
 *
 *     +--------------------+--------------------+--------------------+
 *     |  Type              |  Stores data as    |  Examples          |
 *     +--------------------+--------------------+--------------------+
 *     |  Document Store    |  JSON/BSON docs    |  MongoDB, CouchDB  |
 *     |  Key-Value Store   |  Key → Value pairs |  Redis, DynamoDB   |
 *     |  Column Store      |  Column families   |  Cassandra, HBase  |
 *     |  Graph Database    |  Nodes + Edges     |  Neo4j, Neptune    |
 *     +--------------------+--------------------+--------------------+
 *
 *
 * ───────────────────────────────────────────────────────
 * TYPE 1 — Document Store (MongoDB, CouchDB)
 * ───────────────────────────────────────────────────────
 *
 *   What is it?
 *   - Instead of storing data in rows, you store it as documents.
 *   - A document is a JSON or BSON object — it can have any
 *     structure you want.
 *   - Each document can have completely different fields.
 *   - Documents are grouped in Collections (like tables in SQL).
 *
 *
 *   VISUAL — Table vs Document Store:
 *
 *    Relational (rigid, all rows same structure):
 *    ┌────┬──────────┬─────┬────────┬──────────┐
 *    │ id │ name     │ age │ city   │ linkedin │
 *    ├────┼──────────┼─────┼────────┼──────────┤
 *    │  1 │ Hitesh   │  28 │ Mumbai │ NULL     │ ← linkedin not set
 *    │  2 │ Piyush   │  25 │ Delhi  │ p/piyush │
 *    └────┴──────────┴─────┴────────┴──────────┘
 *
 *    Document Store (flexible, each doc different):
 *
 *    Document 1:                  Document 2:
 *    {                            {
 *      _id: "001",                  _id: "002",
 *      name: "Hitesh",              name: "Piyush",
 *      age: 28,                     age: 25,
 *      city: "Mumbai",              city: "Delhi",
 *      courses: [                   linkedin: "p/piyush",
 *        "JS", "Node"               skills: ["React","AWS"]
 *      ]                          }
 *    }
 *
 *    Document 1 has courses array.
 *    Document 2 has linkedin and skills.
 *    No NULL columns, no wasted space ✓
 *
 *
 *   The No JOIN problem:
 *   - In SQL, you JOIN two tables to combine related data.
 *   - MongoDB does not have native JOINs like SQL.
 *   - Instead, you either:
 *     a. Embed related data inside the document (denormalise)
 *     b. Use Aggregation Pipelines with $lookup (like a JOIN)
 *
 *
 *   Option A — Embedding (denormalisation):
 *
 *    Instead of separate users and orders tables,
 *    embed orders inside the user document:
 *
 *    {
 *      _id: "001",
 *      name: "Hitesh",
 *      orders: [
 *        { orderId: "o1", amount: 500, status: "paid" },
 *        { orderId: "o2", amount: 200, status: "pending" }
 *      ]
 *    }
 *
 *    Pros: one read gets everything — very fast ✓
 *    Cons: orders array can grow unbounded,
 *          same order data repeated if shared across users ✗
 *
 *
 *   Option B — Aggregation Pipelines (the MongoDB way):
 *   - An Aggregation Pipeline is a chain of stages.
 *   - Each stage transforms the documents passing through it.
 *   - The output of one stage becomes the input of the next.
 *   - Think of it as a assembly line for your data.
 *
 *
 *   VISUAL — Aggregation Pipeline concept:
 *
 *    Raw Documents
 *         │
 *         ▼
 *    ┌──────────┐     ┌──────────┐     ┌──────────┐     ┌──────────┐
 *    │  Stage 1 │────►│  Stage 2 │────►│  Stage 3 │────►│  Result  │
 *    │ $match   │     │ $group   │     │  $sort   │     │          │
 *    │ (filter) │     │(aggregate│     │  (order) │     │          │
 *    └──────────┘     └──────────┘     └──────────┘     └──────────┘
 *
 *    Each stage receives documents, transforms them,
 *    and passes the result to the next stage.
 *
 *
 * ───────────────────────────────────────────────────────
 * AGGREGATION PIPELINE STAGES — Explained with Examples
 * ───────────────────────────────────────────────────────
 *
 *   Sample data — orders collection:
 *
 *    { _id:1, userId:"u1", product:"JS Course",  amount:500,  status:"paid",    city:"Mumbai" }
 *    { _id:2, userId:"u1", product:"Node Course", amount:800,  status:"paid",    city:"Mumbai" }
 *    { _id:3, userId:"u2", product:"JS Course",   amount:500,  status:"pending", city:"Delhi"  }
 *    { _id:4, userId:"u3", product:"React Course",amount:1200, status:"paid",    city:"Delhi"  }
 *    { _id:5, userId:"u2", product:"AWS Course",  amount:1500, status:"paid",    city:"Delhi"  }
 *
 *
 * ── Stage 1: $match (like WHERE in SQL) ──────────────────
 *
 *   - Filters documents — only documents matching the
 *     condition pass through to the next stage.
 *   - Always put $match as early as possible — reduces the
 *     number of documents flowing through the pipeline.
 *
 *   Example: get only paid orders
 *
 *   db.orders.aggregate([
 *     { $match: { status: "paid" } }
 *   ])
 *
 *   Input (5 docs):
 *   { _id:1, status:"paid"    ... }  ✓ passes
 *   { _id:2, status:"paid"    ... }  ✓ passes
 *   { _id:3, status:"pending" ... }  ✗ filtered out
 *   { _id:4, status:"paid"    ... }  ✓ passes
 *   { _id:5, status:"paid"    ... }  ✓ passes
 *
 *   Output (4 docs — pending removed):
 *   { _id:1, userId:"u1", amount:500,  city:"Mumbai" }
 *   { _id:2, userId:"u1", amount:800,  city:"Mumbai" }
 *   { _id:4, userId:"u3", amount:1200, city:"Delhi"  }
 *   { _id:5, userId:"u2", amount:1500, city:"Delhi"  }
 *
 *
 *   VISUAL — $match as a filter gate:
 *
 *    [All Orders]  ──►  [$match: status="paid"]  ──►  [Paid Orders Only]
 *         5 docs              filters 1                     4 docs
 *
 *
 * ── Stage 2: $project (like SELECT in SQL) ───────────────
 *
 *   - Controls which fields appear in the output.
 *   - 1 = include, 0 = exclude.
 *   - Can also create new computed fields.
 *
 *   Example: show only userId and amount, hide everything else
 *
 *   db.orders.aggregate([
 *     { $match: { status: "paid" } },
 *     { $project: { userId: 1, amount: 1, _id: 0 } }
 *   ])
 *
 *   Input:
 *   { _id:1, userId:"u1", product:"JS Course", amount:500, city:"Mumbai" }
 *
 *   Output:
 *   { userId:"u1", amount:500 }
 *   (only userId and amount — rest removed)
 *
 *
 *   Example: add a computed field (discounted price)
 *
 *   { $project: {
 *       userId: 1,
 *       amount: 1,
 *       discounted: { $multiply: ["$amount", 0.9] }  // 10% off
 *   }}
 *
 *   Output:
 *   { userId:"u1", amount:500, discounted:450 }
 *
 *
 * ── Stage 3: $group (like GROUP BY + SUM in SQL) ──────────
 *
 *   - Groups documents by a field and performs aggregations
 *     like sum, count, average, min, max.
 *   - _id in $group defines the grouping key.
 *
 *   Example: total revenue per city
 *
 *   db.orders.aggregate([
 *     { $match: { status: "paid" } },
 *     { $group: {
 *         _id: "$city",
 *         totalRevenue: { $sum: "$amount" },
 *         orderCount:   { $sum: 1 }
 *     }}
 *   ])
 *
 *   Input (4 paid docs):
 *   { city:"Mumbai", amount:500  }
 *   { city:"Mumbai", amount:800  }
 *   { city:"Delhi",  amount:1200 }
 *   { city:"Delhi",  amount:1500 }
 *
 *   Output (grouped by city):
 *   { _id:"Mumbai", totalRevenue:1300, orderCount:2 }
 *   { _id:"Delhi",  totalRevenue:2700, orderCount:2 }
 *
 *
 *   VISUAL — $group collapsing multiple docs into one:
 *
 *    ┌─────────────────────────────────────────────────────────────┐
 *    │  Before $group (4 docs):                                    │
 *    │  Mumbai:500  Mumbai:800  Delhi:1200  Delhi:1500             │
 *    └──────────────────────────────┬──────────────────────────────┘
 *                                   │ $group by city
 *                                   ▼
 *    ┌─────────────────────────────────────────────────────────────┐
 *    │  After $group (2 docs):                                     │
 *    │  { Mumbai: total=1300, count=2 }                            │
 *    │  { Delhi:  total=2700, count=2 }                            │
 *    └─────────────────────────────────────────────────────────────┘
 *
 *
 * ── Stage 4: $sort (like ORDER BY in SQL) ─────────────────
 *
 *   - Sorts documents by one or more fields.
 *   - 1 = ascending, -1 = descending.
 *
 *   Example: show cities sorted by highest revenue first
 *
 *   db.orders.aggregate([
 *     { $match: { status: "paid" } },
 *     { $group: {
 *         _id: "$city",
 *         totalRevenue: { $sum: "$amount" }
 *     }},
 *     { $sort: { totalRevenue: -1 } }  // descending
 *   ])
 *
 *   Input:
 *   { _id:"Mumbai", totalRevenue:1300 }
 *   { _id:"Delhi",  totalRevenue:2700 }
 *
 *   Output (sorted highest first):
 *   { _id:"Delhi",  totalRevenue:2700 }  ← first
 *   { _id:"Mumbai", totalRevenue:1300 }  ← second
 *
 *
 * ── Stage 5: $limit and $skip (like LIMIT OFFSET in SQL) ──
 *
 *   - $limit: take only first N documents.
 *   - $skip: skip first N documents.
 *   - Used together for pagination.
 *
 *   Example: get top 3 orders by amount (leaderboard)
 *
 *   db.orders.aggregate([
 *     { $sort:  { amount: -1 } },
 *     { $limit: 3 }
 *   ])
 *
 *   Input (5 docs sorted):
 *   amount: 1500, 1200, 800, 500, 500
 *
 *   Output (top 3 only):
 *   amount: 1500
 *   amount: 1200
 *   amount: 800
 *
 *
 *   Pagination example (page 2, 10 items per page):
 *
 *   db.orders.aggregate([
 *     { $sort:  { createdAt: -1 } },
 *     { $skip:  10 },   // skip page 1 (first 10)
 *     { $limit: 10 }    // take page 2 (next 10)
 *   ])
 *
 *
 * ── Stage 6: $lookup (JOIN between collections) ───────────
 *
 *   - $lookup performs a LEFT JOIN between two collections.
 *   - It is the MongoDB answer to the "no JOIN" problem.
 *   - Slower than embedding for reads, but more flexible
 *     when data changes frequently.
 *
 *   Two collections:
 *
 *   users collection:
 *   { _id:"u1", name:"Hitesh",  email:"h@..." }
 *   { _id:"u2", name:"Piyush",  email:"p@..." }
 *
 *   orders collection:
 *   { _id:1, userId:"u1", amount:500  }
 *   { _id:2, userId:"u1", amount:800  }
 *   { _id:3, userId:"u2", amount:1500 }
 *
 *   Example: get each order with full user details attached
 *
 *   db.orders.aggregate([
 *     { $lookup: {
 *         from:         "users",     // the other collection
 *         localField:   "userId",    // field in orders
 *         foreignField: "_id",       // field in users
 *         as:           "userDetails" // output field name
 *     }}
 *   ])
 *
 *   Output:
 *   {
 *     _id: 1,
 *     userId: "u1",
 *     amount: 500,
 *     userDetails: [
 *       { _id:"u1", name:"Hitesh", email:"h@..." }
 *     ]
 *   }
 *
 *
 *   VISUAL — $lookup joining two collections:
 *
 *    orders collection:         users collection:
 *    ┌───────────────────┐      ┌──────────────────┐
 *    │ _id:1, userId:"u1"│      │ _id:"u1"         │
 *    │ amount: 500       │      │ name: "Hitesh"   │
 *    └────────┬──────────┘      └─────────┬────────┘
 *             │                           │
 *             └─────── $lookup ───────────┘
 *                      joins on userId = _id
 *             │
 *             ▼
 *    ┌─────────────────────────────────────┐
 *    │ _id:1, amount:500,                  │
 *    │ userDetails: [{ name:"Hitesh" }]    │
 *    └─────────────────────────────────────┘
 *
 *
 * ── Stage 7: $unwind (flatten arrays) ────────────────────
 *
 *   - $unwind deconstructs an array field.
 *   - Each element of the array becomes its own document.
 *   - Used when you need to work with individual array items.
 *
 *   Example: a user has multiple courses, unwind to one per doc
 *
 *   Input:
 *   {
 *     userId: "u1",
 *     name: "Hitesh",
 *     courses: ["JS", "Node", "AWS"]
 *   }
 *
 *   { $unwind: "$courses" }
 *
 *   Output (3 separate docs):
 *   { userId:"u1", name:"Hitesh", courses:"JS"   }
 *   { userId:"u1", name:"Hitesh", courses:"Node" }
 *   { userId:"u1", name:"Hitesh", courses:"AWS"  }
 *
 *
 *   VISUAL — $unwind exploding array into rows:
 *
 *    Before $unwind:          After $unwind:
 *    ┌──────────────────┐     ┌─────────────────┐
 *    │ name: "Hitesh"   │     │ name:"Hitesh"   │
 *    │ courses:         │     │ courses: "JS"   │
 *    │  ["JS","Node",   │────►├─────────────────┤
 *    │   "AWS"]         │     │ name:"Hitesh"   │
 *    │  (1 document)    │     │ courses: "Node" │
 *    └──────────────────┘     ├─────────────────┤
 *                             │ name:"Hitesh"   │
 *                             │ courses: "AWS"  │
 *                             │  (3 documents)  │
 *                             └─────────────────┘
 *
 *
 * ── Stage 8: $addFields (add new fields) ─────────────────
 *
 *   - Adds new fields to each document without removing
 *     existing ones.
 *   - Good for computed values you want to carry forward.
 *
 *   Example: add a taxAmount field (18% GST)
 *
 *   { $addFields: {
 *       taxAmount:   { $multiply: ["$amount", 0.18] },
 *       totalWithTax: { $multiply: ["$amount", 1.18] }
 *   }}
 *
 *   Input:  { userId:"u1", amount:500 }
 *   Output: { userId:"u1", amount:500, taxAmount:90, totalWithTax:590 }
 *
 *
 * ── Stage 9: $count (count documents) ────────────────────
 *
 *   - Counts the number of documents reaching this stage.
 *   - Returns a single document with the count.
 *
 *   Example: how many paid orders do we have?
 *
 *   db.orders.aggregate([
 *     { $match: { status: "paid" } },
 *     { $count: "totalPaidOrders" }
 *   ])
 *
 *   Output:
 *   { totalPaidOrders: 4 }
 *
 *
 * ───────────────────────────────────────────────────────
 * COMPLETE PIPELINE EXAMPLE — Real World Use Case
 * ───────────────────────────────────────────────────────
 *
 *   Business question:
 *   "For each city, what is the total revenue from paid orders,
 *    and who are the top customers? Show top 2 cities only."
 *
 *   db.orders.aggregate([
 *
 *     // Stage 1: only paid orders
 *     { $match: { status: "paid" } },
 *
 *     // Stage 2: join with users to get names
 *     { $lookup: {
 *         from: "users", localField: "userId",
 *         foreignField: "_id", as: "user"
 *     }},
 *
 *     // Stage 3: flatten the user array
 *     { $unwind: "$user" },
 *
 *     // Stage 4: group by city, sum revenue, collect users
 *     { $group: {
 *         _id: "$city",
 *         totalRevenue: { $sum: "$amount" },
 *         customers:    { $addToSet: "$user.name" }
 *     }},
 *
 *     // Stage 5: sort by highest revenue
 *     { $sort: { totalRevenue: -1 } },
 *
 *     // Stage 6: top 2 cities only
 *     { $limit: 2 },
 *
 *     // Stage 7: clean up output fields
 *     { $project: {
 *         city: "$_id",
 *         totalRevenue: 1,
 *         customers: 1,
 *         _id: 0
 *     }}
 *   ])
 *
 *   Output:
 *   {
 *     city: "Delhi",
 *     totalRevenue: 2700,
 *     customers: ["Piyush", "Aman"]
 *   },
 *   {
 *     city: "Mumbai",
 *     totalRevenue: 1300,
 *     customers: ["Hitesh"]
 *   }
 *
 *
 *   VISUAL — Pipeline stages flowing data:
 *
 *    [5 raw orders]
 *         │
 *         ▼  $match: status="paid"
 *    [4 paid orders]
 *         │
 *         ▼  $lookup: join users
 *    [4 orders + user details embedded]
 *         │
 *         ▼  $unwind: flatten user array
 *    [4 flat documents]
 *         │
 *         ▼  $group: by city, sum amount
 *    [2 city summaries]
 *         │
 *         ▼  $sort: totalRevenue descending
 *    [2 city summaries — Delhi first]
 *         │
 *         ▼  $limit: 2
 *    [2 city summaries — unchanged]
 *         │
 *         ▼  $project: rename/clean fields
 *    [Final result — 2 clean documents] ✓
 *
 *
 * ───────────────────────────────────────────────────────
 * TYPE 2 — Key-Value Store (Redis, DynamoDB)
 * ───────────────────────────────────────────────────────
 *
 *   What is it?
 *   - The simplest NoSQL type.
 *   - Stores data as key → value pairs.
 *   - Like a giant dictionary / hashmap.
 *   - Blazing fast reads and writes — O(1) lookup.
 *   - No schema, no relationships, no complex queries.
 *
 *
 *   VISUAL — Key-Value Store:
 *
 *    Key                    Value
 *    ─────────────────────────────────────────────
 *    "user:001:name"     →  "Hitesh"
 *    "user:001:session"  →  "abc123xyz"
 *    "user:002:cart"     →  '["item1","item2"]'
 *    "page:home:cache"   →  "<html>...</html>"
 *    "rate:user:001"     →  "5"  (requests today)
 *    ─────────────────────────────────────────────
 *    Any key → Any value. Nothing else.
 *
 *
 *   Best use cases:
 *   - Session storage (user login tokens)
 *   - Caching (cache expensive DB query results)
 *   - Rate limiting (count API calls per user)
 *   - Leaderboards (Redis sorted sets)
 *   - Real-time counters (likes, views)
 *
 *
 * ───────────────────────────────────────────────────────
 * TYPE 3 — Column Store (Cassandra, HBase)
 * ───────────────────────────────────────────────────────
 *
 *   What is it?
 *   - Stores data in column families instead of rows.
 *   - Optimised for queries that read a few columns across
 *     millions of rows — like analytics and time-series data.
 *   - Massive horizontal scaling — petabytes of data.
 *   - Built for write-heavy workloads.
 *
 *
 *   VISUAL — Row store vs Column store:
 *
 *    Row Store (SQL):          Column Store (Cassandra):
 *    Stored by row:            Stored by column:
 *
 *    [Hitesh|28|Mumbai]        [Hitesh, Piyush, Aman]  ← name col
 *    [Piyush|25|Delhi ]        [28, 25, 30]            ← age col
 *    [Aman  |30|Pune  ]        [Mumbai, Delhi, Pune]   ← city col
 *
 *    Reading age of all        Reading age of all
 *    users: must load          users: reads only
 *    entire row for each ✗     the age column ✓
 *                              (very fast for analytics)
 *
 *   Best use cases:
 *   - Time-series data (IoT sensor readings, logs)
 *   - Analytics (read one column across billions of rows)
 *   - Event tracking (clickstream, user behaviour)
 *
 *
 * ───────────────────────────────────────────────────────
 * TYPE 4 — Graph Database (Neo4j, Amazon Neptune)
 * ───────────────────────────────────────────────────────
 *
 *   What is it?
 *   - Stores data as nodes (entities) and edges (relationships).
 *   - Designed for deeply connected data where relationships
 *     are as important as the data itself.
 *   - Traversing relationships is extremely fast — no JOINs needed.
 *
 *
 *   VISUAL — Graph Database:
 *
 *    (Hitesh) ──FOLLOWS──► (Piyush)
 *        │                    │
 *        │ ENROLLED_IN        │ ENROLLED_IN
 *        ▼                    ▼
 *    (JS Course) ◄──────── (Node Course)
 *        │
 *        └──TAUGHT_BY──► (Instructor: Aman)
 *
 *    Query: "Who are the instructors of courses Hitesh enrolled in?"
 *    → Follow: Hitesh → ENROLLED_IN → course → TAUGHT_BY → instructor
 *    → One traversal, no JOINs ✓
 *
 *   Best use cases:
 *   - Social networks (friend of friend recommendations)
 *   - Fraud detection (unusual transaction chains)
 *   - Recommendation engines
 *   - Knowledge graphs
 *
 *
 * ───────────────────────────────────────────────────────
 * RDBMS vs NoSQL — When to Use Which
 * ───────────────────────────────────────────────────────
 *
 *     +------------------------------+------------+------------------+
 *     |  Situation                   |  Use RDBMS |  Use NoSQL       |
 *     +------------------------------+------------+------------------+
 *     |  Structured, fixed schema    |  ✓         |                  |
 *     |  Complex JOINs needed        |  ✓         |                  |
 *     |  Strong ACID required        |  ✓         |                  |
 *     |  Financial / transactional   |  ✓         |                  |
 *     |  Flexible / changing schema  |            |  ✓ Document      |
 *     |  Massive horizontal scale    |            |  ✓ Any NoSQL     |
 *     |  Caching / sessions          |            |  ✓ Key-Value     |
 *     |  Analytics on huge datasets  |            |  ✓ Column Store  |
 *     |  Deeply connected data       |            |  ✓ Graph DB      |
 *     |  Semi-structured data        |            |  ✓ Document      |
 *     +------------------------------+------------+------------------+
 *
 *
 *   Summary — all pipeline stages at a glance:
 *
 *     +-------------+----------------------------+-------------------+
 *     |  Stage      |  What it does              |  SQL equivalent   |
 *     +-------------+----------------------------+-------------------+
 *     |  $match     |  Filter documents          |  WHERE            |
 *     |  $project   |  Choose/compute fields     |  SELECT           |
 *     |  $group     |  Group + aggregate         |  GROUP BY + SUM   |
 *     |  $sort      |  Order results             |  ORDER BY         |
 *     |  $limit     |  Take first N docs         |  LIMIT            |
 *     |  $skip      |  Skip first N docs         |  OFFSET           |
 *     |  $lookup    |  Join another collection   |  LEFT JOIN        |
 *     |  $unwind    |  Flatten arrays            |  (no equivalent)  |
 *     |  $addFields |  Add computed fields       |  computed columns |
 *     |  $count     |  Count documents           |  COUNT(*)         |
 *     +-------------+----------------------------+-------------------+
 */
```
