```python
/**
 * Architecting AI-Powered Systems:
 * =================================
 * > AI naya protein jaisa ho gaya hai—har jagah milta hai, lekin samajh nahi aata ki 
 *   iska exactly kya use karein.
 *
 * > AI System Architecture:
 *   - RAG Pipelines
 *   - Large Context Management Strategies
 *   - Vector databases alongside traditional application databases
 *   - Workflow Engines – AI applications ke liye queue system
 *   - Real-time AI applications, including voice-based AI systems
 */


/**
 * TRADITIONAL APPLICATIONS:
 * =============================
 *
 * - Ye wo applications hain jahan AI use nahi hota. Inmein ek user hota hai
 *   jo server ke saath REST API endpoints ke through interact karta hai.
 *   Server multiple databases se communicate karta hai aur response return karta hai.
 *
 * - Traditional applications mein backend aur database ko independently scale kiya jata hai.
 * - REST API endpoints ke upar ek UI layer hoti hai, kyunki user directly APIs
 *   se interact nahi karta. UI layer hi APIs ke saath interact karti hai.
 *
 * ARCHITECTURE:
 *
 *   [User]
 *     |
 *     | (UI Layer — user directly API se baat nahi karta)
 *     v
 *   [UI Layer]
 *     |
 *     | REST API endpoints
 *     v
 *   [Server / Backend]
 *     |
 *     | (independently scale hota hai)
 *     v
 *   [Database-1] [Database-2] ...
 *
 *
 * LIMITATION:
 * - Traditional applications mein NLP (Natural Language Processing) nahi hota.
 *
 * EXAMPLE — Instagram:
 *
 *   GET /posts              → saari posts return karega
 *   GET /comments?postId=1  → us post ke comments return karega
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Data fetch karne ka sirf ek tarika hota hai —           │
 *   │  fixed endpoints ke through.                             │
 *   │                                                          │
 *   │  "Hey, give me all comments of postId 1" — yeh           │
 *   │  Natural Language hai. Traditional apps ise              │
 *   │  samajh nahi sakte.                                      │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * AI APPLICATIONS (INTRODUCTION):
 * ==================================
 *
 * - AI ke aane se NLP (Natural Language Processing) common ho gaya hai.
 * - Ab hum systems se natural language mein baat kar sakte hain.
 *
 * - Hum input natural language mein dete hain aur output bhi
 *   natural language mein milta hai.
 *
 *   AI = LLM (Large Language Model)
 *
 *   - Pre-trained data par kaam karta hai
 *   - Input: Natural Language
 *   - Output: Natural Language (next token predict karta hai)
 *
 * FLOW:
 *
 *   [User Input (NL)] ──> [LLM] ──> [Output (NL)]
 */


/**
 * TRADITIONAL vs AI:
 * ======================
 *
 * - NLP ki wajah se log traditional tareeke se kaam karna
 *   kam karne lage hain.
 *
 * EXAMPLES:
 *   - Google par search kyun karein jab GPT se seedha answer mil sakta hai?
 *   - Saare comments kyun dekhein jab AI sirf positive comments filter karke de sakta hai?
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  PEHLE (Traditional):                                    │
 *   │  "Mujhe postId 1 ke comments chahiye"                    │
 *   │   → GET /comments?postId=1                               │
 *   │                                                          │
 *   │  AB (AI):                                                │
 *   │  "Hey, give me all positive comments of postId 1"        │
 *   │   → AI samjhega, filter karega, aur result dega ✓✓✓     │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * SAHI APPROACH — AI + TRADITIONAL APP:
 * =========================================
 *
 * - AI standalone ek complete feature nahi hai.
 * - AI tab powerful hota hai jab ise existing application ke saath 
 *   integrate kiya jaye.
 *
 * APPROACH:
 *
 *   - Database ka READ-ONLY access AI ko do
 *   - User natural language mein query kare
 *   - AI DB se data fetch karke result return kare
 *
 * ARCHITECTURE:
 *
 *   [User] ──NL Query──> [AI] ──DB Query──> [Database]
 *                         |                     |
 *                         |<────── Result ───────┘
 *                         |
 *                         v
 *                   [NL Response] ──> [User] ✓
 *
 * - Yahi aaj industry mein AI ka real use case hai.
 */


/**
 * TYPES OF APPLICATIONS (CURRENT WORLD):
 * =========================================
 *
 *   TYPE 1 — Pure AI Apps:
 *   -----------------------
 *   - Claude, GPT, Midjourney, Sora, etc.
 *   - Text, image, aur video generation ke liye use hote hain.
 *   - Yeh already badi companies build kar chuki hain.
 *   - Naye players ke liye yahan compete karna mushkil hai.
 *
 *
 *   TYPE 2 — Existing App + AI Integration:
 *   ---------------------------------------
 *   - Tumhare paas pehle se ek app hai
 *   - Uske andar AI integrate karo
 *   - Yahi se system design start hota hai
 *   - Real-world opportunity yahan hai
 */


/**
 * SYSTEM DESIGN PERSPECTIVE
 * ==========================
 *
 * TRADITIONAL APP:
 *
 *   [User] → [UI] → [Backend] → [Database]
 *
 *   Simple flow — fixed endpoints, fixed responses.
 *   User jo maange woh exactly wahi milta hai, zyada nahi kam nahi.
 *
 *
 * AI-INTEGRATED APP:
 *
 *   [User] → [UI] → [Backend] → [Database]
 *                        |
 *                        | (READ-only access diya AI ko)
 *                        v
 *                      [AI / LLM]
 *                        |
 *                        | NL query → DB query → NL response
 *                        v
 *                   [User ko Natural Language mein answer] ✓
 *
 *   Ab user fixed endpoints yaad karne ki zaroorat nahi —
 *   seedha puchh lo jo chahiye.
 */


/**
 * AI KI DO BADI PROBLEMS
 * ========================
 *
 * AI integrate karna sunne mein simple lagta hai — lekin do
 * fundamental problems hain jo system design ko directly affect
 * karti hain.
 *
 *
 * PROBLEM 1 — CONTEXT WINDOW (Limited Memory):
 * ──────────────────────────────────────────────
 *
 * LLM ek baar mein sirf limited data process kar sakta hai —
 * ise CONTEXT WINDOW kehte hain. Sochlo jaise ek insaan jo ek
 * baar mein sirf 10 pages padh sakta hai, 1000 nahi.
 *
 *   Context window = 200 comments
 *   Total comments  = 3,000,000
 *
 *   200 se zyada daalo → LLM fail  
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Real Example:                                           │
 *   │                                                          │
 *   │  40 min ki class complete hui.                           │
 *   │  Instructor ne kaha — "Students ke liye ek AI agent      │
 *   │  do jo doubts solve kare."                               │
 *   │                                                          │
 *   │  Class mein 55,000 students hain.                        │
 *   │                                                          │
 *   │  Scenario 1 — Sab "hi" bhi bhejein:                      │
 *   │  55,000 messages → 55,000 LLM calls → BIG BILL           │
 *   │                                                          │
 *   │  Scenario 2 — Har student ke 2-3 doubts:                 │
 *   │  55,000 × 3 = 165,000 LLM calls → HUGE BILL              │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * PROBLEM 2 — COST (Paisa lagta hai):
 * ─────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Traditional System scale nahi hua → System down         │
 *   │  (Downtime hoti hai — fix kar lo)                        │
 *   │                                                          │
 *   │  AI System scale nahi kiya → BAHUT BADA BILL             │
 *   │  (Paisa chala jaata hai — recover karna mushkil)         │
 *   │                                                          │
 *   │  Isliye AI ko blindly integrate nahi kar sakte —         │
 *   │  proper system design zaroori hai.                       │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * CONTEXT PROBLEM SOLVE KARNA:
 * ============================
 *
 * Problem: 200-250 page PDF hai, users QnA karna chahte hain.
 * Poori book ek saath LLM mein kaise daalen?
 *
 *
 * APPROACH 1 — NAIVE (Seedha galat):
 * ─────────────────────────────────────
 *
 *   Poori book ek saath LLM mein daalo.
 *
 *   - Context window exceed → LLM fail
 *   - Poori book ka cost = bahut zyada bill
 *   - Agar 1000 users hain to 1000 × poori book = massive cost
 *
 *
 * APPROACH 2 — SMART (Sirf relevant part do):
 * ──────────────────────────────────────────────
 *
 *   "Poori book kyun daalen jab sirf ek page ki zaroorat hai?"
 *
 *   STEP 1: User query + "Sirf page number identify karo" → LLM
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  User: "What is TCP/IP?"                                 │
 *   │  Hum LLM ko bolte hain:                                  │
 *   │  "Is book mein TCP/IP kis page pe hai? Sirf page number  │
 *   │   batao."                                                │
 *   │                                                          │
 *   │  LLM: "Page 20"                                          │
 *   │                                                          │
 *   │  Sirf ek choti si query → Cheap LLM call                 │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   STEP 2: Code se woh page read karo (free hai!)
 *
 *   const pageContent = fs.readFileSync(`page_20.txt`)
 *   // ← koi LLM call nahi, koi cost nahi
 *
 *   STEP 3: Original query + Page 20 content → LLM
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  "User ne puchha: What is TCP/IP?                        │
 *   │   Context: [Page 20 ka content]                          │
 *   │   Answer karo."                                          │
 *   │                                                          │
 *   │  - Better answer                                         │
 *   │  - Low cost — sirf 2 small LLM calls                     │
 *   │  - Context window safe                                   │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * WHAT IF KEYWORD MULTIPLE PAGES PE HAI?
 * ────────────────────────────────────────
 *
 *   User: "Where is networking mentioned in this PDF?"
 *
 *   Poori book scan karni padegi — page by page.
 *   But AI se scanning free nahi hai — to kya karein?
 *
 *   STEP 1: LLM se related keywords nikaalo (cheap, ek call)
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Query → LLM:                                            │
 *   │  "Networking se related saare keywords do"               │
 *   │                                                          │
 *   │  LLM → [networking, LAN, MAN, WAN, TCP, IP, ...]         │
 *   │                                                          │
 *   │  - Sirf ek cheap call                                    │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   STEP 2: Code se regex search karo (bilkul free!)
 *
 *   const foundPages = []
 *   for (const page of allPages) {
 *     if (regex(keywords, page)) {
 *       foundPages.push(page)  // [5, 9, 10]
 *     }
 *   }
 *   // ← koi LLM call nahi, koi cost nahi
 *
 *   STEP 3: Sirf relevant pages + original query → LLM
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  3 pages ka content + user query → LLM                   │
 *   │                                                          │
 *   │  - Accurate answer + page numbers                        │
 *   │  - Minimum cost                                          │
 *   │  - Context window safe                                   │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * VISUAL FLOW:
 * ─────────────
 *
 *   [User Query]
 *        |
 *        v
 *   [LLM — Keywords extract] ← cheap call
 *        |
 *        v
 *   [Code — Regex scan all pages] ← FREE
 *        |
 *        v
 *   [Found Pages: 5, 9, 10]
 *        |
 *        v
 *   [3 Pages Content + Original Query → LLM] ← small context
 *        |
 *        v
 *   [Accurate Answer] ✓
*/


/**
 *
 * LEKIN EK PROBLEM ABHI BHI HAI:
 * ─────────────────────────────────
 *
 *   Yeh approach ek PDF ke liye theek hai.
 *   Lekin agar MULTIPLE PDFs hain? 100 files? 1000 files?
 *
 *   - Regex scan 1000 PDFs ke across bahut slow hoga
 *   - Keywords manually manage karna impossible
 *   - Yeh approach scale nahi hoti
 *
 *   Iska solution hai — RAG (Retrieval Augmented Generation)
 */


/**
 * RAG — RETRIEVAL AUGMENTED GENERATION
 * =======================================
 *
 * Jab data source bahut bada ho (multiple files, databases,
 * websites) aur hum uske upar generation karna chahein —
 * tab RAG use hota hai.
 *
 * RAG ki core idea simple hai:
 * "Poora data LLM ko mat do — pehle relevant part dhundho,
 *  phir sirf woh do."
 *
 *
 * RAG PIPELINE — 2 PHASES:
 * ─────────────────────────
 *
 *   Phase A: INDEXING   → Data ek baar process karo
 *   Phase B: QUERY      → Har user request pe smart retrieval
 */


/**
 * PHASE A — INDEXING (Ek baar ka kaam)
 * =======================================
 *
 * Sochlo jaise library mein books ka catalogue banana —
 * ek baar banao, phir baar baar use karo.
 *
 *
 * STEP 1 — Documents Load karo:
 * ────────────────────────────────
 *
 *   Source kuch bhi ho sakta hai:
 *   PDF files, Websites, Databases, Word documents, etc.
 *
 *
 * STEP 2 — Chunks Banao:
 * ────────────────────────
 *
 *   Poori book ek saath process nahi kar sakte —
 *   isko chhote pieces (chunks) mein todo.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Chunking strategies (developer decide karta hai):       │
 *   │                                                          │
 *   │  a. Page by page     → Har page ek chunk                 │
 *   │  b. Paragraph by paragraph → Har para ek chunk           │
 *   │  c. Fixed size       → Har 500 characters ek chunk       │
 *   │                                                          │
 *   │  Koi ek best strategy nahi hai — depends on use case     │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  200 page book → 200 chunks (page by page)               │
 *   │                                                          │
 *   │  Chunk-1: "Chapter 1 — Introduction to Networks..."      │
 *   │  Chunk-2: "Chapter 2 — OSI Model..."                     │
 *   │  Chunk-3: "Chapter 3 — TCP/IP Protocol..."               │
 *   │  ...                                                     │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * STEP 3 — Vector Embeddings Banao:
 * ───────────────────────────────────
 *
 *   Yeh sabse important step hai. Har chunk ko NUMBERS mein
 *   convert karo — ise VECTOR EMBEDDING kehte hain.
 *
 *   Kyun numbers mein convert karein?
 *   Kyunki computers text se zyada fast numbers compare kar
 *   sakte hain — aur similar meaning wale texts ke numbers
 *   bhi similar hote hain.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Text → Embedding Model → Vector (Numbers ka array)      │
 *   │                                                          │
 *   │  "TCP/IP is a protocol" → [0.23, 0.87, 0.12, 0.45, ...]  │
 *   │  "Networking basics"    → [0.21, 0.85, 0.14, 0.43, ...]  │
 *   │                           ↑                              │
 *   │  Similar topics → Similar numbers (close vectors) ✓✓✓   │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Special embedding models use hote hain (cheap):
 *   - OpenAI: text-embedding-3
 *   - Google: text-embedding-004
 *   - Hugging Face: free open source models
 *
 *
 * STEP 4 — Vector Database mein Store karo:
 * ────────────────────────────────────────────
 *
 *   Har chunk ke liye 3 cheezein store karo:
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Entry in Vector DB:                                     │
 *   │                                                          │
 *   │  a. Vector    → [0.23, 0.87, 0.12, ...]  ← numbers       │
 *   │  b. Content   → "TCP/IP is a protocol..." ← original text│
 *   │  c. Metadata  → { pageNo: 20, chunkNo: 5 } ← extra info  │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Popular Vector Databases:
 *   Pinecone, Qdrant, Weaviate, pgvector, Chroma
 *
 *
 * INDEXING COMPLETE VISUAL:
 * ──────────────────────────
 *
 *   [200 Page PDF]
 *         |
 *         v
 *   [Chunking — 200 chunks banao]
 *         |
 *         v
 *   [Embedding Model] → har chunk → vector
 *         |
 *         v
 *   [Vector Database]
 *    ┌─────────────────────────────────────────┐
 *    │  chunk-1: [0.23, 0.87...] "Chapter 1..."│
 *    │  chunk-2: [0.45, 0.12...] "OSI Model..."│
 *    │  chunk-3: [0.21, 0.85...] "TCP/IP..."   │
 *    │  ...200 entries total...                │
 *    └─────────────────────────────────────────┘
 *    Indexing done — ek baar ka kaam!
 */


/**
 * PHASE B — QUERY (Har user request pe)
 * ========================================
 *
 * Ab user kuch bhi puchhe — hum smartly relevant chunks
 * dhundhte hain aur sirf woh LLM ko dete hain.
 *
 *
 * STEP 1 — User Query aati hai:
 * ────────────────────────────────
 *
 *   User: "What is networking and on which page it is mentioned?"
 *
 *
 * STEP 2 — Query ko bhi Vector mein convert karo:
 * ──────────────────────────────────────────────────
 *
 *   SAME embedding model use karo jo indexing mein use kiya tha.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  "What is networking?" → [0.22, 0.84, 0.13, ...]         │
 *   │                                                          │
 *   │  Notice: Yeh vector chunk-3 ke vector se similar hai!    │
 *   │  chunk-3 tha "TCP/IP..." → [0.21, 0.85, 0.14, ...]       │
 *   │                                                          │
 *   │  Similar topic → Similar numbers → Similarity match ✓✓✓ │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * STEP 3 — Vector DB mein Similarity Search karo:
 * ──────────────────────────────────────────────────
 *
 *   "Mere paas query ka vector hai — isse similar vectors
 *    dhundho Vector DB mein."
 *
 *   Vector DB → Top-K most similar chunks return karta hai
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Query Vector: [0.22, 0.84, 0.13, ...]                   │
 *   │                                                          │
 *   │  Most Similar:                                           │
 *   │  1. chunk-3 (page 20): "TCP/IP..." similarity: 0.95      │
 *   │  2. chunk-7 (page 45): "LAN/WAN..." similarity: 0.87     │
 *   │  3. chunk-2 (page 10): "OSI Model..." similarity: 0.82   │
 *   │                                                          │
 *   │  Semantic search — exact words match nahi chahiye        │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * STEP 4 — Relevant Chunks + Query → LLM:
 * ──────────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  LLM ko dete hain:                                       │
 *   │  1. Original user query                                  │
 *   │  2. Top 3 relevant chunks (small context)                │
 *   │  3. Metadata (page numbers)                              │
 *   │                                                          │
 *   │  LLM → Accurate answer with page references ✓✓✓         │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * COMPLETE RAG FLOW VISUALLY:
 * ────────────────────────────
 *
 *   INDEXING (ek baar):
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  [PDF] → [Chunks] → [Embedding Model] → [Vector DB] ✓✓✓ │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   QUERY (har request pe):
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  [User Query]                                            │
 *   │       |                                                  │
 *   │       v                                                  │
 *   │  [Same Embedding Model] → Query Vector                   │
 *   │       |                                                  │
 *   │       v                                                  │
 *   │  [Vector DB Similarity Search] → Top-K Relevant Chunks   │
 *   │       |                                                  │
 *   │       v                                                  │
 *   │  [LLM] ← Chunks + Original Query                         │
 *   │       |                                                  │
 *   │       v                                                  │
 *   │  [Final Accurate Answer] ✓✓✓                            │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Note: Google NotebookLM ek RAG application hai.
*/

/**
 * RAG vs OLD APPROACH — KYA FARK PADAA:
 * ─────────────────────────────────────────
 *
 *   ┌──────────────────────┬──────────────────┬──────────────────┐
 *   │                      │ Old Approach     │ RAG              │
 *   ├──────────────────────┼──────────────────┼──────────────────┤
 *   │ Multiple PDFs        │  Regex slow      │  Vector search   │
 *   ├──────────────────────┼──────────────────┼──────────────────┤
 *   │ Scaling              │  Hard            │  Easy            │
 *   ├──────────────────────┼──────────────────┼──────────────────┤
 *   │ Exact word match     │  Works           │  Semantic match  │
 *   │ zaroorat             │                  │    (better)      │
 *   ├──────────────────────┼──────────────────┼──────────────────┤
 *   │ Cost                 │ Medium           │  Low             │
 *   └──────────────────────┴──────────────────┴──────────────────┘
 */


/**
 * What is the success metric of RAG?
 * ==================================
 * > Every RAG follows indexing > retrieval > generation.
 * > So how do we compare different RAG systems? ACCURACY!
 * 
 * > Speed and Cost are important, but if accuracy is poor, the system
 *   fails
 * 
 * > Ideal RAG:
 *   - High Accuracy
 *   - Low Cost
 *   - Low Latency
 * 
 * > If the speed is fast and cost is low but accuracy is wrong
 *   (2 + 2 = 5), the system is useless.
*/

/**
 * What controls accuracy?
 * =======================
 * > Many people think:
 *   - Better model?
 *   - Better system prompt?
 * > These help, but they do not solve the real issue.
 * 
 * > The most important factor is: Context Quality
 * > If the Context is wrong, output will be wrong, and this leads to
 *   the core RAG problem: GIGO (Garbage in, garbage out)
*/


/**
 * RAG KI 3 BADI PROBLEMS
 * ========================
 *
 *
 * PROBLEM 1 — POOR SOURCE FILES:
 * ─────────────────────────────────
 *
 *   Agar original documents hi bekar hain:
 *   - Bad formatting (tables, images jo text mein nahi aate)
 *   - Broken sentences (PDF se extract karte waqt toot gayi)
 *   - Missing sections (incomplete documents)
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Poor Document                                           │
 *   │       |                                                  │
 *   │       v                                                  │
 *   │  Weak Embeddings  X                                      │
 *   │       |                                                  │
 *   │       v                                                  │
 *   │  Poor Retrieval   X                                      │
 *   │       |                                                  │
 *   │       v                                                  │
 *   │  Wrong Output     X                                      │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Solution: Document preprocessing — pehle clean karo,
 *   phir index karo.
 *
 *
 * PROBLEM 2 — POOR USER QUERY:
 * ──────────────────────────────
 *
 *   User ki query directly RAG ko milti hai — agar query
 *   bekar hai to retrieval bhi bekar hogi.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Query problems:                                         │
 *   │                                                          │
 *   │  Spelling mistakes   → "how diabetis cure"               │
 *   │  Missing terms       → Technical terms nahi pata user ko │
 *   │  Missing keywords    → Jo actually relevant hai woh      │
 *   │                        query mein hai hi nahi            │
 *   │  Unordered keywords  → Wrong emphasis                    │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  "how diabetis cure"                                     │
 *   │         |                                                │
 *   │         v                                                │
 *   │  Vector search → No good match (typo ki wajah se)   X    │
 *   │         |                                                │
 *   │         v                                                │
 *   │  Retrieval poor → Generation poor   X                    │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * PROBLEM 3 — ONE-SHOT RETRIEVAL:
 * ──────────────────────────────────
 *
 *   Basic RAG sirf ek baar retrieve karta hai — no second chances.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Retrieval galat hua                                     │
 *   │         |                                                │
 *   │         v                                                │
 *   │  Koi correction nahi                   X                 │
 *   │  Koi retry nahi                        X                 │
 *   │  Koi refinement nahi                   X                 │
 *   │  Blindly top-K results trust karta hai X                 │
 *   │         |                                                │
 *   │         v                                                │
 *   │  Wrong answer generate hoti hai        X                 │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * ADVANCED RAG PATTERNS — PROBLEMS KA SOLUTION
 * ===============================================
 *
 * In teeno problems ko solve karne ke liye advanced patterns
 * hain. Har pattern ek specific problem target karta hai.
 *
 *   ┌──────────────────────┬──────────────────────────────────┐
 *   │ Pattern              │ Kya solve karta hai              │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ 1. Query Rewriting   │ Poor user query (typos, unclear) │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ 2. Query Expansion   │ Missing keywords, broad search   │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ 3. Subquery RAG      │ Complex multi-part questions     │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ 4. HyDE              │ Vague/unclear queries            │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ 5. Self-RAG          │ One-shot retrieval problem       │
 *   └──────────────────────┴──────────────────────────────────┘
 *
 * (Har pattern detail mein aage cover kiya gaya hai.)
 */


/**
 * Understanding Corrective RAG:
 * - Corrective RAG is an advanced RAG pattern that improves user queries.
 * - It uses two main approaches:
 *   1. Query Rewriting: Corrects and enhances queries
 *   2. Query Expansion: Broadens queries for better results
 *   3. Subquery RAG: Breaks down complex queries into smaller subqueries
 * 
 * - Main Goals:                        Benefits:
 *   - Improve accuracy                 - Better search results
 *   - Better understand user intent    - Reduced errors & more relevant responses
 *   - Improve retrieval quality        - More relevant responses
 * 
 * - Note: We can use any of these approaches or a combination of them.
*/

/**
 * 1. Query Rewriting:
 *    - Whenever a user query comes in, we cannot depend directly on it. We 
 *      always use a mini or nano model for auto-correction and improvement 
 *      of the query through:
 *      a. Fixing Typos
 *      b. Adding more context
 *    - When we create vector embeddings for this improved query, it will
 *      attract more relevant chunks from the vector store. Then we can
 *      use a chatbot to generate the response.
 *    - However, this approach has tradeoffs as it increases both cost and
 *      processing time.
 *                                                                Chunk Store                     
 *      +--------+         +--------+        +-----------+       +-----------+       +-----------+         +--------+
 *      |        |-------->|        |------->|           |------>|           |-------|           |--------+|        |-----> Answer
 *      +--------+         +--------+        +-----------+       +-----------+       +-----------+         +--------+ 
 *      User Query       Query Rewriting      Embeddings         Vector Search      Retrieved Chunks        Chat/LLM
 * 
 *    - While adding more GPUs can improve speed, it significantly increases
 *      infrastructure costs.
 *    - Self-hosting models can help optimize both speed and costs, but it
 *      introduces scalability challenges when user traffic increases.
*/

/**
 * 2. Query Expansion:
 *    - Query expansion is a technique that builds upon query rewriting by
 *      focusing on broadening the search scope through related terms.
 *    
 *    Key Components (Similar to Query Rewriting):
 *    a. Auto-correction & Typo Fixing
 *       - First, we fix spelling mistakes or small errors.
 *       - This ensures the query is accurate before we try expanding it.
 *       - Example: "javasript tutorl" -> "javascript tutorial"
 *    
 *    b. Context Enhancement (Broaden the Query)
 *       - Generate multiple variations (chunks) of the query.
 *       - Add:
 *         - Synonyms (car -> automobile)
 *         - Related words (AI -> Artificial Intelligence, machine learning)
 *         - Domain-specific terms (medicine: heart attack -> myocardial infraction)
 *       - This way, the search covers a wider set of possibilities.
 *
 *    c. Quality Control with a Mini Model (Relevance Check)
 *       - Since expansion generates many query variations, not all will be
 *         useful.
 *       - Use a small/mini model as a judge:
 *         1. Pass the original query + retrieved chunks to the mini model
 *         2. Ask the model to tell us:
 *            - How good are these retrieved chunks compared to the original
 *              query?
 *            - How many are actually relevant? 
 *         3. The model scores each chunk for semantic match & relevance.
 *       - Keep only the most relevant chunks, discard the noisy ones.
 *       - Add these refined chunks back into the context and re-run the
 *         pipeline.
 *       - Repeat until you get high-quality, relevant
 * 
 * Note: Now result accuracy is very high, but this is a very expensive 
 *       process and is not recommended for production environments.
 *
 *    B. Architecture Diagram:
 *                                                                   +-------> Chunk-1 +------+
 *                                                                   |                        |   All Chunks + User Query
 *      +-----------+         +-----------+       +-----------+      +-------> Chunk-2 +------+       +----------+
 *      |           |-------->|           |------>|           |------|                        |------>|Judgement |------> Most relevant chunks
 *      +-----------+         +-----------+       +-----------+      +-------> Chunk-3 +------+       +----------+
 *       User Query          Query Rewriting        Embedding        |                        |        Mini Model 
 *                                  ^                                +-------> Chunk-4 +------+            |
 *                                  |                                                                      |
 *                                  |                                                                      |
 *                                  +----------------------------------------------------------------------+
 *                                     Re-write Query based on the Judgement & check more chunks created
 *                                                                    |
 *                                                                    +-------> Chunk-5 +------+
 *                                                                    |                        |
 *                                                                    +-------> Chunk-6 +------+
 *                                                                Adding this into context and re-run the pipeline
 * 
 * 
 *    C. Pipeline Flow:
 *       1. User submits initial query
 *       2. Query undergoes pre-processing and expansion
 *       3. System generates embeddings for expanded query
 *       4. Vector search retrieves multiple chunk sets
 *       5. Quality control model evaluates chunks
 *       6. If needed, query is refined and process repeats
 *       7. Final set of most relevant chunks is selected
 *
 *    Tradeoffs & Considerations:
 *    - Increased Processing: Multiple embeddings per query
 *    - Higher Latency      : More vector searches required
 *    - Cost Impact         : Additional API calls and compute resources
 *    - Precision vs Recall : Broader search may reduce accuracy
 *    
 *    Best Practices:
 *    - Use semantic expansion over simple synonym matching
 *    - Implement relevance scoring to filter expanded results
 *    - Consider caching common query expansions
 *    - Monitor and tune expansion breadth based on results
*/

/**
 * 3. Subquery RAG:
 *    - Subquery RAG is a technique where instead of treating the user query
 *      as one big question, we split it into smaller subqueries.
 *    - Each subquery is run through the retrieval system to fetch focused
 *      chunks, and then the results are merged to form a more accurate
 *      final answer.
 * 
 *    Key Components:
 *    A. Query Decomposition (Breaking into smaller queries):
 *       - The original query is analyzed and broken into multiple smaller
 *         pipelines.
 *       - Example: 
 *         User Query : What are the causes, symptoms and treatment of diabetes?
 *         Sub-Queries: a. What are the causes of diabetes?
 *                      b. What are the symptoms of diabetes?
 *                      c. What are the treatmesnts of diabetes?
 *       - This ensures that each aspect of the query is properly covered.
 * 
 *    B. Independent Retrieval (Chunk fetching per subquery):
 *       - Each subquery is run independently against the vector database 
 *         (retriever).
 *       - This helps avoid missing information that might be skipped if we
 *         only retrieved with the big original query.
 *       - Ensures focused and higher recall chunks for each part.
 * 
 *    C. Aggregation & Fusion:
 *       - After retrieving results for all subqueries, we combine the chunks.
 *       - Techniques used:
 *         a. Merging      : Simply combining chunks.
 *         b. Reranking    : Using a mini model to score and rank the chunks 
 *                           by relevance (remove irrelevant chunks).
 *         c. Deduplication: Removing overlapping or repetitive chunks.
 *       - This ensures that the final answer is accurate and relevant.
 * 
 *    D. Answer Composition:
 *       - The model now has a comprehensive context covering all parts of 
 *         the query.
 *       - It merges the subquery answers into a single coherent response.
 *       - Example final answer Covers:
 *         > causes, symptoms, and treatments all in one structured response.
 *
 *    Architecture Diagram:
 *                                                                                       +-------> Subquery-A +------+
 *                                                                                       |                           |   All Subqueries + Results
 *      +-----------+         +---------+         +-----------+       +-----------+      +-------> Subquery-B +------+       +----------+
 *      |           |-------->|         |-------->|           |------>|           |------|                           |------>|Combine & |------> Generate Final Answer
 *      +-----------+         +---------+         +-----------+       +-----------+      +-------> Subquery-C +------+       |Reranking |
 *       User Query        Query Translation    Query Decomposer      Embedding          |                           |       +----------+
 *                                                 (LLM/Rules)                           +-------> Subquery-D +------+    Fuse All retrieved chunks &
 *                                                                                                       Rerank chunks
 *                                                                                                       Deduplication
 *
 *    Pipeline Flow:
 *    1. Complex user query is received
 *    2. Query decomposer breaks it into simpler subqueries
 *    3. Each subquery gets processed independently
 *    4. Vector search runs for each subquery
 *    5. Results are combined and deduplicated
 *    6. Final context is built from merged results
 *    7. LLM generates answer using combined context
 *
 *    Tradeoffs & Considerations:
 *    - Higher Latency : Multiple parallel searches needed
 *    - Increased Costs: More embedding operations
 *    - Better Coverage: Handles complex queries better
 *    - Result Quality : May need careful deduplication
 *    
 *    Best Practices:
 *    - Use intelligent query decomposition
 *    - Implement efficient deduplication
 *    - Consider parallel processing
 *    - Balance subquery count vs performance
*/


/**
 * 4. Hypothetical Document Generation (HyDE):
 *    - HyDE stands for Hypothetical Document Embeddings
 *    - It uses the LLM's knowledge to increase the accuracy of the search
 *      by generating a hypothetical answer before searching.
 *    - Instead of directly searching with the user's question, it first
 *      creates a detailed hypothetical answer and then searches using that.
 *    - Core Concept:
 *      "Rather than searching with the question, imagine what an ideal answer
 *       would look like, and search with that instead!"
 * 
 *    Key Components & Process Flow:
 *    A. Initial User Query:
 *       - Example: "What is Node.js used for?"
 *       - This is the starting point of the HyDE process
 *   
 *    B. Generate a Hypothetical Answer:
 *       - Instead of searching with just the question, we ask the LLM to
 *         write a fake but realistic answer.
 *       - Example of Hypothetical Answer:
 *         "Node.js is a runtime environment for executing JavaScript code
 *          outside of a browser. It's used for building server-side
 *          applications and APIs."
 *       - This doesn't come from your database yet - it's just generated
 *         by the model!
 *   
 *    C. Embedding Generation:
 *       - The hypothetical answer is converted into vector embeddings
 *       - Why? Because embeddings captures:
 *         - Semantic meaning
 *         - Technical context
 *         - Relationship between concepts
 *         (what the answer is about, not just the words)
 *         
 * 
 *    D. Retrieve Similar Documents
 *       - Use the embedding of the hypothetical answer to search your vector
 *         database.
 *       - The database will return the most relevant chunks from the Node.js
 *         PDF.
 *       - Example retrieved chunks:
 *         "Node.js provides event-driven architecture....."
 *         "It is commonly used for REST APIs and real-time apps like chats..."
 * 
 *    E. Generate the Final Answer:
 *       - Now, feed these things to the LLM:
 *         1. The original user query
 *         2. Retrieved relevant chunks
 *         3. The hypothetical answer context
 *       - Produces a comprehensive final response:
 *         "Node.js is a powerful runtime used for developing scalable network
 *          applications, REST APIs, and real-time services like chat applications
 *          and streaming platforms. Its non-blocking, event-driven architecture
 *          makes it ideal for handling concurrent operations."
 *   
 * 
 * 
 *    Architecture Diagram:                                         Hypothetical Answer Embedding
 *                                                                           +----------+                                    +----Chunk-1----+
 *                                                                  +------->|          |--------------+                     |               |
 *                                                                  |        +----------+              |                     +----Chunk-2----+
 *    +-----------+         +-----------+        +-----------+      |                                  +--------------+      |               |           +--------------+
 *    |           |-------->|           |------->|           |------+                                  |              |------+----Chunk-3----+---------->| Final Answer |
 *    +-----------+         +-----------+        +-----------+      |                                  +--------------+      |               |           +--------------+
 *    User Query          Query Translation       Claude LLM        |        +----------+              | Vector Search       +----Chunk-4----+            Generated using
 *                                              Create 100 words    +------->|          |--------------+                     |               |            retrieved chunks
 *                                              answer using LLM             +----------+                                    +----Chunk-5----+            & original query
 *                                              based on query           User Query Embedding                                                 
 *                                                                                                                            
 * 
 *    Why HyDE is powerful?
 *    We use HyDE when the question lacks specificity or lacks easily
 *    identifiable elements to derive an answer from a given context.
 * 
 *    Drawbacks:
 *    But there is a drawback of this approach, it may not consistently
 *    produce good results.
*/

/**
 * 5. Self-RAG (Most Advanced Pattern)
 *    > Self-RAG introduces self-evaluation and adaptive retrieval.
 *    > Core Idea:
 *      [Retrieve > Evaluate > Retry(if needed) > Generate > Done]
 *        - Do I need retrieval?
 *        - Is retrieved context enough?
 *        - Should I retrieve again?
 * 
 *    > High-Level Flow:
 *
 *                    +----------------+
 *                    |   User Query   |
 *                    +----------------+
 *                             |
 *                             v
 *                    +----------------+
 *                    |  LLM (Decide)  |
 *                    | Need Retrieval?|
 *                    +----------------+
 *                         |        |
 *                  (No)   |        |  (Yes)
 *                         |        v
 *                         |   +----------------+
 *                         |   |   Embedding    |
 *                         |   +----------------+
 *                         |            |
 *                         |            v
 *                         |   +----------------+
 *                         |   | Vector Search  |
 *                         |   +----------------+
 *                         |            |
 *                         |            v
 *                         |   +----------------+
 *                         |   | Retrieved Docs |
 *                         |   +----------------+
 *                         |            |
 *                         |            v
 *                         |   +------------------------+
 *                         |   | LLM Evaluates Context  |
 *                         |   | Is it sufficient?      |
 *                         |   +------------------------+
 *                         |        |            |
 *                         |   (No) |            | (Yes)
 *                         |        v            |
 *                         |   +------------------------+
 *                         |   | Refine Query / Retry   |
 *                         |   +------------------------+
 *                         |        |
 *                         |        +---------> (Back to Embedding)
 *                         |
 *                         v
 *                 +----------------+
 *                 | Final Answer   |
 *                 +----------------+
 *
 *    > Step-by-Step Pipeline:
 *      1. User submits query
 *      2. LLM decides:
 *           - Can I answer directly?
 *           - Do I need external knowledge?
 *      3. If retrieval is required:
 *           - Generate embedding
 *           - Perform vector search
 *           - Retrieve chunks
 *      4. LLM evaluates retrieved chunks:
 *           - Are they relevant?
 *           - Is information complete?
 *      5. If not sufficient:
 *           - Refine query
 *           - Retrieve again
 *      6. Once context is sufficient:
 *           - Generate final answer
*/

/**
 * MEMORY IN AI AGENTS
 * ====================
 *
 * Abhi tak humne RAG Pipeline mein query optimization seekhi —
 * Query Rewriting, Expansion, Subquery, HyDE, Self-RAG.
 *
 * Ab next problem: AI ko memory kaise dein?
 *
 *
 * PROBLEM — AI STATELESS HOTA HAI:
 * ──────────────────────────────────
 *
 * Jab hum AI se baat karte hain, hum basically API calls karte
 * hain. Lekin yeh calls STATELESS hain.
 *
 * Stateless matlab — har call independent hai, AI ko pichli
 * baat yaad nahi rehti.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Turn 1: "My name is Piyush"                             │
 *   │  AI: "Nice to meet you Piyush!"                          │
 *   │                                                          │
 *   │  Turn 2: "What is my name?"                              │
 *   │  AI: "I don't know your name." ← yaad nahi raha   X      │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * BASIC SOLUTION — CONVERSATION HISTORY BHEJO:
 * ──────────────────────────────────────────────
 *
 *   Har naye call mein poori conversation history saath bhejo.
 *   AI usse "context" ki tarah padh kar respond karta hai.
 *
 *   messages = []
 *
 *   Turn 1:
 *   messages.push({ role: "user", content: "My name is Piyush" })
 *   → LLM ko messages array bhejo ✓
 *
 *   Turn 2:
 *   messages.push({ role: "assistant", content: "Nice to meet you!" })
 *   messages.push({ role: "user", content: "What is my name?" })
 *   → Poori history saath bhejo → AI ko yaad rehta hai ✓
 *
 *
 * LEKIN BADI PROBLEM — CONTEXT WINDOW LIMIT:
 * ────────────────────────────────────────────
 *
 *   Context window 1M tokens bhi ho — jaise jaise conversation
 *   badhti hai, window bhar jaati hai.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Context Window: [────────────────────────] 1M tokens    │
 *   │                                                          │
 *   │  Conversation shuru:                                     │
 *   │  [msg1][msg2][msg3] ← bahut jagah hai                    │
 *   │                                                          │
 *   │  Conversation lamba ho gaya:                             │
 *   │  [msg500]...[msg999][msg1000] ← Window almost full!      │
 *   │                                                          │
 *   │  Sliding window move karta hai:                          │
 *   │  [msg2][msg3]...[msg1001] ← msg1 DISCARD ho gaya  X      │
 *   │                                                          │
 *   │  Ab puchho: "What is my name?"                           │
 *   │  AI: "I don't know." ← "My name is Piyush" discard       │
 *   │  ho chuka tha                                            │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * MEMORY KA SOLUTION — HUMAN BRAIN SE SEEKHO
 * =============================================
 *
 * Human brain bhi sab kuch yaad nahi rakhta — sirf important
 * cheezein yaad rakhta hai, baaki bhool jaata hai.
 *
 * Isi concept pe AI memory design karte hain.
 *
 *
 * 2 TYPES OF MEMORY:
 * ───────────────────
 *
 *   ┌─────────────────────┬──────────────────────────────────────┐
 *   │ SHORT TERM MEMORY   │ LONG TERM MEMORY                     │
 *   │ (STM)               │ (LTM)                                │
 *   ├─────────────────────┼──────────────────────────────────────┤
 *   │ Word-to-word        │ Sirf important facts yaad rakhta     │
 *   │ conversations yaad  │ hai — details bhool jaata hai        │
 *   │ rehti hain          │                                      │
 *   ├─────────────────────┼──────────────────────────────────────┤
 *   │ Temporary — jab tak │ Permanent — session end hone ke      │
 *   │ session chal raha   │ baad bhi rehta hai                   │
 *   │ hai                 │                                      │
 *   ├─────────────────────┼──────────────────────────────────────┤
 *   │ Tab tak valid:      │ Examples:                            │
 *   │ - User online hai   │ - "My name is Piyush"                │
 *   │ - Session active    │ - "Talk to me in hinglish"           │
 *   │ - Page refresh      │ - "I prefer bullet points"           │
 *   │   tak               │                                      │
 *   │ - Browser tab       │                                      │
 *   │   close tak         │                                      │
 *   └─────────────────────┴──────────────────────────────────────┘
 *
 *
 * REAL EXAMPLE — KYUN DONO CHAHIYE:
 * ────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  User: "My name is Piyush" ← LTM mein store karo         │
 *   │  User: "Talk to me in hinglish" ← LTM mein store karo    │
 *   │  User: "How are you?" ← STM (current chat)               │
 *   │  User: "What is 2+2?" ← STM (current chat)               │
 *   │                                                          │
 *   │  Next Session mein bhi:                                  │
 *   │  AI ko pata hoga — "Piyush hai, hinglish mein baat karo" │
 *   │  Bina user ke dobara bataye ✓✓✓                         │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * CONTEXT WINDOW PROBLEM KA SOLUTION:
 * ──────────────────────────────────────
 *
 *   Har new session mein:
 *
 *   [LTM facts inject karo] + [Current STM messages]
 *         |                          |
 *         v                          v
 *   "Name: Piyush              [Turn 1 message]
 *    Lang: Hinglish"           [Turn 2 message]
 *         |                          |
 *         └──────────┬───────────────┘
 *                    v
 *               [LLM call] ✓
 *
 *   Ab 1M messages ke baad bhi puchho "What is my name?" →
 *   LTM se context inject hoga → AI answer dega ✓
 */


/**
 * LONG TERM MEMORY — 3 TYPES
 * ============================
 *
 * LTM ko 3 parts mein divide karte hain — har ek alag type
 * ka data store karta hai.
 *
 *
 * TYPE 1 — FACTUAL MEMORY:
 * ─────────────────────────
 *
 *   User ke baare mein permanent facts — user se user alag hoti hai.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Examples:                                               │
 *   │  - "My name is Piyush"                                   │
 *   │  - "Birthday: 1st Jan"                                   │
 *   │  - "My favorite color is blue"                           │
 *   │  - "Talk to me in hinglish"                              │
 *   │                                                          │
 *   │  Storage: Database mein structured format mein           │
 *   │  name: "Piyush"                                          │
 *   │  language: "hinglish"                                    │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * TYPE 2 — EPISODIC MEMORY:
 * ──────────────────────────
 *
 *   Past interactions ke important events ya patterns — user
 *   specific, lekin time ke saath purane events forget ho jaate hain.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Examples:                                               │
 *   │  - "Last time we talked about movies, Alex got very      │
 *   │     excited about sci-fi"                                │
 *   │  - "Alex was frustrated yesterday when internet          │
 *   │     was slow"                                            │
 *   │  - "Piyush ne Sunday live stream schedule ki thi"        │
 *   │     ← Sunday ke baad yeh relevant nahi rahega            │
 *   │                                                          │
 *   │  Storage: Vector embeddings — similarity search se       │
 *   │  relevant episodes fetch karte hain                      │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Real life analogy:
 *   Agar Sunday night ko live stream hai to wo Sunday tak
 *   yaad rahegi — baad mein brain automatically bhool jaata hai
 *   kyunki event khatam ho gaya. Yahi LTSM (Long Term Short Memory)
 *   hai — important tha lekin ab relevant nahi.
 *
 *
 * TYPE 3 — SEMANTIC MEMORY:
 * ──────────────────────────
 *
 *   General knowledge jo platform ke saare users ke liye same
 *   hoti hai — kisi specific user se related nahi.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Examples:                                               │
 *   │  - "Company ka hackathon prize $1000 hai"                │
 *   │  - "Support office Sundays ko closed hai"                │
 *   │  - "Current promo code: SUMMER20"                        │
 *   │                                                          │
 *   │  Note: Yeh change ho sakti hai future mein —             │
 *   │  "SUMMER20" expire hoga, prize change ho sakta hai       │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * TEENO KA COMPARISON:
 * ─────────────────────
 *
 *   ┌────────────────┬───────────────┬────────────────────────────┐
 *   │ Memory Type    │ User Specific │ Example                    │
 *   ├────────────────┼───────────────┼────────────────────────────┤
 *   │ Factual        │   Yes         │ "Name: Piyush"             │
 *   ├────────────────┼───────────────┼────────────────────────────┤
 *   │ Episodic       │   Yes         │ "Piyush excited about AI   │
 *   │                │               │  last week"                │
 *   ├────────────────┼───────────────┼────────────────────────────┤
 *   │ Semantic       │   No          │ "Hackathon prize: $1000"   │
 *   │                │ (sabke liye)  │ (sabke liye same)          │
 *   └────────────────┴───────────────┴────────────────────────────┘
 */


/**
 * MEMORY ARCHITECTURE — HOW IT WORKS
 * =====================================
 *
 *
 * PART 1 — STM IMPLEMENTATION:
 * ──────────────────────────────
 *
 *   STM ka kaam hai current conversation ka context maintain
 *   karna — jab tak user active hai tab tak.
 *  
 *   Last 50 messages context window mein rakhte hain:
 *   25 AI messages + 25 User messages
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Redis mein store karo (fast, in-memory):                │
 *   │                                                          │
 *   │  redis.lpush("chat:piyush", message)  ← new message add  │
 *   │  redis.ltrim("chat:piyush", 0, 49)    ← sirf last 50     │
 *   │                                                          │
 *   │  CRON Job:                                               │
 *   │  Har kuch minutes mein Redis se messages uthao aur       │
 *   │  MongoDB mein vector embeddings banake store karo        │
 *   │  (50 messages ka ek batch)                               │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  STM FLOW:                                               │
 *   │                                                          │
 *   │  [New Message]                                           │
 *   │       |                                                  │
 *   │       v                                                  │
 *   │  [Redis List] ← last 50 messages                         │
 *   │       |                                                  │
 *   │       | (CRON Job — batch sync)                          │
 *   │       v                                                  │
 *   │  [MongoDB — Vector Embeddings] ← permanent storage       │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * PART 2 — LTM CREATION PROCESS:
 * ────────────────────────────────
 *
 *   User chat kar raha hai → side by side messages enqueue
 *   hote hain → memory build hoti jaati hai.
 *
 *   STEP 1: Message aaya → Queue mein daalo
 *
 *   User: "My name is Piyush"
 *          |
 *          v
 *   [Message Queue]
 *
 *
 *   STEP 2: Mini LLM decide kare — STM ya LTM?
 *   (Few Shot Prompting use karke)
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Few Shot Examples jo Mini LLM ko dete hain:             │
 *   │                                                          │
 *   │  "Hi" → STM                                              │
 *   │  "My name is Piyush" → LTM                               │
 *   │  "I prefer hinglish" → LTM                               │
 *   │  "How are you?" → STM                                    │
 *   │  "What is 2+2?" → STM                                    │
 *   │  ...50-60 examples...                                    │
 *   │                                                          │
 *   │  Ab Mini LLM predict kar sakta hai ✓✓✓                  │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   [Message] → [Mini LLM]
 *                    |
 *           ┌────────┴────────┐
 *           v                 v
 *          STM               LTM
 *           |                 |
 *        Discard X          STEP 3 ↓
 *
 *
 *   STEP 3 (If LTM): Kaun sa type hai — Factual, Episodic, Semantic?
 *   (Fir se Mini LLM + 50-60 few shot examples)
 *
 *   [LTM Message] → [Mini LLM]
 *                        |
 *          ┌─────────────┼─────────────┐
 *          v             v             v
 *       Factual       Episodic      Semantic
 *          |             |             |
 *        STEP 4        STEP 5        STEP 6
 *
 *
 *   STEP 4 (Factual): Facts extract karke DB mein store karo
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Raw message: "Hey Agent, My name is Piyush"             │
 *   │         |                                                │
 *   │         v  (Mini LLM extract karta hai)                  │
 *   │  DB Storage: { name: "Piyush" }  ← clean fact ✓✓✓       │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 *   STEP 5 (Episodic): Vector embeddings banao aur store karo
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  "Piyush was excited about AI last week"                 │
 *   │         |                                                │
 *   │         v                                                │
 *   │  Vector embedding banao → Vector DB mein store karo      │
 *   │  (Similarity search se future mein fetch karenge)        │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 *   STEP 6 (Semantic): Platform-wide facts DB mein store karo
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  "Hackathon prize is $1000"                              │
 *   │         |                                                │
 *   │         v                                                │
 *   │  Shared DB mein store karo (sabke liye same)             │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * COMPLETE LTM CREATION FLOW:
 * ────────────────────────────
 *
 *   [User Message]
 *         |
 *         v
 *   [Mini LLM — STM ya LTM?]
 *         |              |
 *       STM             LTM
 *         |              |
 *      Discard      [Mini LLM — Factual/Episodic/Semantic?]
 *                         |            |              |
 *                      Factual      Episodic       Semantic
 *                         |            |              |
 *                   [Mini LLM     [Vector DB     [Shared DB
 *                    extract]      store]          store]
 *                         |
 *                   [DB mein
 *                    clean fact]
 */


/**
 * PART 3 — MEMORY FETCHING
 * ==========================
 *
 * Memory create karna kaafi nahi — sahi waqt pe sahi memory
 * fetch karna bhi zaroori hai.
 *
 *
 * FETCHING STRATEGIES:
 * ─────────────────────
 *
 *   ┌────────────────┬──────────────────────────────────────────┐
 *   │ Memory Type    │ Fetching Strategy                        │
 *   ├────────────────┼──────────────────────────────────────────┤
 *   │ Factual        │ ALWAYS available — har request ke saath  │
 *   │                │ inject karo (user ka naam, language,     │
 *   │                │ preferences, etc.)                       │
 *   ├────────────────┼──────────────────────────────────────────┤
 *   │ Episodic       │ ON-DEMAND — user ne past event ke baare  │
 *   │                │ mein puchha → query ki vector embedding  │
 *   │                │ banao → Vector DB similarity search →    │
 *   │                │ relevant episodes fetch karo             │
 *   ├────────────────┼──────────────────────────────────────────┤
 *   │ Semantic       │ ALWAYS available — platform-wide info    │
 *   │                │ har session mein inject karo             │
 *   └────────────────┴──────────────────────────────────────────┘
 *
 *
 * NEW SESSION MEIN MEMORY INJECT KARO:
 * ──────────────────────────────────────
 *
 *   [New Chat Session starts]
 *         |
 *         v
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  System Prompt mein inject karo:                         │
 *   │                                                          │
 *   │  Factual: "User name: Piyush, Language: Hinglish"        │
 *   │  Semantic: "Hackathon prize: $1000, Office: Closed Sun"  │
 *   │                                                          │
 *   │  + Current STM (last 50 messages)                        │
 *   │  + On-demand Episodic (agar relevant ho)                 │
 *   └──────────────────────────────────────────────────────────┘
 *         |
 *         v
 *   [LLM] → User ko lagta hai AI ko sab yaad hai ✓
 *
 *
 * EPISODIC — ON-DEMAND FETCHING FLOW:
 * ─────────────────────────────────────
 *
 *   User: "Remember last time we talked about that movie?"
 *         |
 *         v
 *   [Query ki vector embedding banao]
 *         |
 *         v
 *   [Vector DB — Similarity Search]
 *         |
 *         v
 *   [Relevant Episodic Memory fetch karo]
 *   "Alex was excited about sci-fi last week"
 *         |
 *         v
 *   [LLM ko context ke saath do] → Accurate response ✓
 */


/**
 * COMPLETE MEMORY ARCHITECTURE — EK NAZAR MEIN
 * ===============================================
 *
 *   USER CHATTING
 *         |
 *    ┌────┴────────────────────────────────┐
 *    │                                     │
 *    v                                     v
 *  [STM]                                [LTM Creation]
 *  Redis — last 50 msgs              Mini LLM classify karta hai
 *    |                                     |
 *    | CRON Job                     ┌──────┼──────┐
 *    v                              v      v      v
 *  [MongoDB                     Factual Episodic Semantic
 *   Vector Embeddings]              |      |      |
 *                                  DB   VectorDB  DB
 *
 *
 *   NEW SESSION START
 *         |
 *         v
 *   ┌─────────────────────────────────────────────┐
 *   │  Inject:                                    │
 *   │  - Factual Memory (hamesha)                 │
 *   │  - Semantic Memory (hamesha)                │
 *   │  - STM — last 50 msgs                       │
 *   │  - Episodic — on-demand (query se match)    │
 *   └─────────────────────────────────────────────┘
 *         |
 *         v
 *   [LLM] → Context-aware response ✓
 *
 *
 * BEST APPROACHES FOR LTM:
 * ─────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Few Shot Prompting  ← STM/LTM classify karne ke liye    │
 *   │         +                                                │
 *   │  Vector Similarity Search ← Episodic fetch karne ke liye │
 *   │         +                                                │
 *   │  Graphs ← Complex relationships track karne ke liye      │
 *   └──────────────────────────────────────────────────────────┘
 */

/**
 * MEMORY SUMMARY — STM vs LTM:
 * ──────────────────────────────
 *
 *   ┌──────────────────────┬──────────────────┬──────────────────┐
 *   │                      │ STM              │ LTM              │
 *   ├──────────────────────┼──────────────────┼──────────────────┤
 *   │ Storage              │ Redis            │ DB + Vector DB   │
 *   ├──────────────────────┼──────────────────┼──────────────────┤
 *   │ Content              │ Word-to-word     │ Important facts  │
 *   │                      │ conversation     │ only             │
 *   ├──────────────────────┼──────────────────┼──────────────────┤
 *   │ Duration             │ Session tak      │ Permanent        │
 *   ├──────────────────────┼──────────────────┼──────────────────┤
 *   │ Size                 │ Last 50 messages │ Grows slowly     │
 *   ├──────────────────────┼──────────────────┼──────────────────┤
 *   │ Fetching             │ Hamesha          │ Factual: Hamesha │
 *   │                      │                  │ Episodic: On-    │
 *   │                      │                  │ demand           │
 *   └──────────────────────┴──────────────────┴──────────────────┘
 */


/**
 * WHY WE NEED AGENTS (AFTER RAG):
 * 
 * --------------------------------------------------
 * 1. RAG KYA KARTA HAI (LIMITATIONS)
 * --------------------------------------------------
 *
 * RAG : Retrieval + Generation
 * FLOW: User Query > Vector DB Search > Relevant Context > LLM > Answer
 *
 *
 * RAG KI CAPABILITIES:
 * ---------------------
 * ✔ External data fetch kar sakta hai
 * ✔ Context-aware answer de sakta hai
 * ✔ Hallucination reduce karta hai
 *
 *
 * RAG KI LIMITATIONS:
 * --------------------
 * ✖ Sirf "READ" kar sakta hai (read-only system)
 * ✖ Real-world action perform nahi kar sakta
 * ✖ Multi-step problems solve nahi kar sakta
 * ✖ Decision making limited hai
 * ✖ Tool use nahi karta
 *
 *
 * CORE INSIGHT:
 * --------------
 * - RAG   = "Smart Reader"
 * - Agent = "Smart Worker"
 *
 *
 * --------------------------------------------------
 * 2. REAL-WORLD PROBLEMS RAG SE SOLVE KYU NAHI HOTE?
 * --------------------------------------------------
 *
 * Example Query:
 *   "Mera last order cancel karo aur refund status batao"
 *
 * Isme kya chahiye?
 *
 *   Step 1: User identify karo
 *   Step 2: Order DB se fetch karo
 *   Step 3: Order cancel karo (action)
 *   Step 4: Refund status check karo (API call)
 *   Step 5: Final response do
 *
 *
 * RAG kya karega?
 * ----------------
 *   → Sirf bata dega "order cancel kaise hota hai"
 *   → Actual cancel nahi karega 
 *
 *
 * VISUAL:
 * --------
 *
 *   [User Query]
 *        |
 *        v
 *      [RAG]
 *        |
 *        v
 *  "Yeh steps follow karo"
 *
 *
 * REAL NEED:
 * ------------
 *
 *                     [User Query]
 *                          |
 *                          v
 *                      [System]
 *                          |
 *   ┌───────────────┬──────────────┬──────────────┐
 *   │               │              │              │
 *   v               v              v              v
 * [DB]          [Cancel API]   [Refund API]     [LLM]
 *   │               │              │              │
 *   +───────────────+───────┬──────+──────────────+
 *                           |
 *                           v
 *                  ✅ Real Action + Answer
 *
 *
 * Solution: AGENTS + TOOLS
 * a. RAG: 
 *    - "Mujhe batao kya hai"
 *    - Knowledge Layer
 * b. AGENT: 
 *    - "Mujhe batao + karke dikhao"
 *    - Execution Layer
 */


/**
 * AGENTS + TOOLS:
 * ===============
 * 
 * Ab tak humne cover kiya:
 *   - RAG (retrieve + generate)
 *   - Memory (STM + LTM)
 *
 * Next level: AGENTS + TOOLS (Real-world execution layer)
 *
 *
 * ============================================================
 * 1. AGENT KYA HOTA HAI? (DEEP UNDERSTANDING)
 * ============================================================
 *
 * - Agent ek intelligent orchestrator hota hai jo:
 *   → Input ko samajhta hai
 *   → Plan banata hai
 *   → Tools select karta hai
 *   → Multi-step execution karta hai
 *   → Final answer generate karta hai
 *
 * - Important:
 *   LLM khud "think" karta hai but "act" nahi karta
 *   Agent = Think + Act + Observe + Repeat
 *
 *
 * CORE DIFFERENCE:
 * ------------------
 *
 *   SIMPLE LLM:
 *   ------------
 *   Input → Output
 *
 *   AGENT SYSTEM:
 *   ---------------
 *   Input → Reason → Decide → Tool Call → Observe → Iterate → Output
 *
 *
 * ============================================================
 * 2. AGENT CORE LOOP (VERY IMPORTANT)
 * ============================================================
 *
 *   [User Query]
 *         |
 *         v
 *   [Agent / LLM]
 *         |
 *   (Reasoning Step)
 *   "Mujhe kya karna chahiye?"
 *         |
 *         v
 *   [Action Selection]
 *         |
 *         v
 *   [Tool Execution]
 *         |
 *         v
 *   [Observation]
 *         |
 *   "Kya goal achieve hua?"
 *         |
 *     YES / NO
 *         |
 *   (Repeat if NO)
 *         |
 *         v
 *   [Final Answer]
 *
 *
 * ============================================================
 * 3. TOOLS KYA HOTE HAIN?
 * ============================================================
 *
 * - Tools external capabilities hote hain jo LLM ko empower karte hain.
 *
 * - Without tools:
 *   ❌ Real-time data access nahi
 *   ❌ File read nahi
 *   ❌ DB query nahi
 *
 * - With tools:
 *   ✅ APIs call
 *   ✅ DB access
 *   ✅ File system access
 *   ✅ Calculations
 *
 *
 * TOOL TYPES:
 * ------------
 *
 * 1. Retrieval Tools:
 *    - Vector DB search (RAG)
 *
 * 2. Data Tools:
 *    - Database queries
 *
 * 3. Action Tools:
 *    - Send email
 *    - Place order
 *
 * 4. Utility Tools:
 *    - Calculator
 *    - Code execution
 *
 *
 * ============================================================
 * 4. AGENT ARCHITECTURE (HIGH LEVEL)
 * ============================================================
 *
 *   [User]
 *     |
 *     v
 *   [Agent Controller]
 *     |
 *   ┌───────────────┬──────────────────┬──────────────────┐
 *   │               │                  │                  │
 *   v               v                  v                  v
 * [LLM]          [Tools]           [Memory]            [RAG]
 * (Reasoning)    (Execution)       (STM + LTM)         (Context)
 *
 *     |
 *     v
 *   [Final Response]
 *
 *
 * ============================================================
 * 5. BASIC AGENT LOOP
 * ============================================================
 */

function agentLoop(userQuery) {

    let state = {
        query: userQuery,
        steps: [],
        done: false
    };

    while (!state.done) {

        /* STEP 1: Reasoning */
        let decision = LLM(`
      User Query: ${state.query}
      Previous Steps: ${state.steps}

      Decide:
      - What should I do next?
      - Which tool should I use?
      - Or can I answer directly?
    `);

        if (decision.type === "FINAL_ANSWER") {
            state.done = true;
            return decision.answer;
        }

        /* STEP 2: Tool Selection */
        let tool = decision.tool;

        /* STEP 3: Tool Execution */
        let result = tool.execute(decision.input);

        /* STEP 4: Observation Store */
        state.steps.push({
            tool: tool.name,
            output: result
        });
    }
}


/**
 * ===================
 * 6. TOOL DEFINITION 
 * ===================
 */

const tools = {
    readFile: {
        name: "readFile",
        execute: (filePath) => {
            return fs.readFileSync(filePath, "utf-8");
        }
    },

    searchVectorDB: {
        name: "vectorSearch",
        execute: (queryEmbedding) => {
            return vectorDB.similaritySearch(queryEmbedding);
        }
    },

    callAPI: {
        name: "apiCall",
        execute: (endpoint) => {
            return fetch(endpoint);
        }
    }
};


/**
 * ===============
 * 7. AGENT + RAG 
 * ===============
 */

function ragAgent(query) {

    /* STEP 1: Convert query to embedding */
    let embedding = embeddingModel(query);

    /* STEP 2: Retrieve relevant chunks */
    let chunks = vectorDB.search(embedding);

    /* STEP 3: Generate answer */
    let answer = LLM(`
    Question: ${query}
    Context: ${chunks}
    Answer:
  `);

    return answer;
}


/**
 * ============================================================
 * 8. AGENT + MEMORY + TOOLS (FULL SYSTEM)
 * ============================================================
 */

function smartAgent(userQuery, userId) {

    // STEP 1: Fetch memory
    let factualMemory = DB.getUserFacts(userId);
    let stm = Redis.getLastMessages(userId);

    // STEP 2: Decide if episodic needed
    let episodic = vectorDB.search(embeddingModel(userQuery));

    // STEP 3: Combine context
    let context = {
        factual: factualMemory,
        recent: stm,
        episodic: episodic
    };

    // STEP 4: Agent reasoning
    let decision = LLM(`
    Query: ${userQuery}
    Context: ${JSON.stringify(context)}

    Decide:
    - Answer directly?
    - Call tool?
  `);

    if (decision.tool) {
        let result = tools[decision.tool].execute(decision.input);

        return LLM(`
      Query: ${userQuery}
      Tool Result: ${result}
      Final Answer:
    `);
    }

    return decision.answer;
}


/**
 * ============================================================
 * 9. REAL-WORLD FLOW (END-TO-END)
 * ============================================================
 *
 * Example:
 * User: "Mere last order ka status kya hai?"
 *
 * FLOW:
 *
 * [User Query]
 *     |
 *     v
 *   [Agent]
 *      |
 *      V
 *   Step 1: Identify intent → "Order status"
 *   Step 2: Fetch userId (LTM)
 *   Step 3: Call Order API (Tool)
 *   Step 4: Receive response
 *   Step 5: Convert to natural language
 *
 *
 * ============================================================
 * 10. ADVANCED AGENT PATTERNS (THEORY)
 * ============================================================
 *
 * 1. ReAct (Reason + Act):
 *    - Alternate between thinking and acting
 *
 * 2. Toolformer:
 *    - Model learns when to use tools
 *
 * 3. Planner-Executor:
 *    - Step 1: Plan banata hai
 *    - Step 2: Execute karta hai
 *
 * 4. Multi-Agent Systems:
 *    - Multiple agents collaborate
 *    - Example:
 *        - Research Agent
 *        - Coding Agent
 *        - Review Agent
 *
*/



/**
 * ╔══════════════════════════════════════════════════════════════╗
 * ║         TEXT-BASED CHATBOT (WITHOUT STREAMING)               ║
 * ╚══════════════════════════════════════════════════════════════╝
 *
 * Abhi tak humne seekha (Phase-I Recap):
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Traditional App  → Fixed endpoints, no NLP              │
 *   │  AI + Traditional → NL queries, DB read access           │
 *   └──────────────────────────────────────────────────────────┘
 *
 * Ab Phase-II mein:
 *   → Inhi concepts ko use karke ek actual chatbot banate hain
 *   → Pehle BINA streaming ke — kyunki yeh base hai
 *   → Isko samjhe bina streaming samajhna mushkil hoga
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 1 — KYA HAI NON-STREAMING CHATBOT?
 * ══════════════════════════════════════════════════════════════
 *
 * Sochlo tum ek dukaan par gaye.
 * Tumne order diya → waiter andar gaya → khana bana → POORA
 * khana ek baar mein table pe rakh diya → tum khao.
 *
 * Yahi NON-STREAMING chatbot hai:
 *
 *   User query bhejta hai
 *      ↓
 *   Server LLM se POORA answer generate karta hai
 *      ↓
 *   Poora answer ek baar mein user ko bhejta hai
 *
 * Streaming mein (jo hum baad mein seekhenge):
 *   Waiter ek ek dish laata — jaise jaise banti jaati hai.
 *
 *
 * SIMPLEST FLOW:
 * ───────────────
 *
 *   [User]  ──POST /chat──> [Backend]  ──query──> [LLM]
 *   [User]  <──full answer─ [Backend]  <──answer─ [LLM]
 *
 * Note: Yeh synchronous hai — matlab ek kaam hota hai,
 *       uske khatam hone ka wait karo, phir agla kaam.
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 2 — HIGH LEVEL ARCHITECTURE
 * ══════════════════════════════════════════════════════════════
 *
 *   ┌──────────────┐     POST /chat      ┌──────────────────┐
 *   │   CLIENT     │ ──────────────────> │   BACKEND        │
 *   │  (Browser)   │                     │  (Node/Express)  │
 *   └──────────────┘                     └────────┬─────────┘
 *                                                 │
 *                                           LLM API call
 *                                                 │
 *                                                 v
 *                                         ┌──────────────────┐
 *                                         │    LLM           │
 *                                         │ (Claude/GPT)     │
 *                                         └────────┬─────────┘
 *                                                  │
 *                                        Full answer generate
 *                                                  │
 *   ┌──────────────┐    JSON response   ┌──────────┴─────────┐
 *   │   CLIENT     │ <────────────────  │   BACKEND          │
 *   │  (Browser)   │                    └────────────────────┘
 *   └──────────────┘
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 3 — CONVERSATION HISTORY (CHATBOT KI MEMORY)
 * ══════════════════════════════════════════════════════════════
 *
 *   LLM STATELESS hota hai — har call independent hai.
 *   "My name is Piyush" → next call pe bhool jaata hai.
 *
 * Chatbot mein memory simulate karne ke liye hum CONVERSATION
 * HISTORY bhejte hain — har call mein poori history.
 *
 *
 * CONVERSATION HISTORY KYA HOTI HAI?
 * ─────────────────────────────────────
 *
 *   messages = [
 *     { role: "user",      content: "What is TCP?" },
 *     { role: "assistant", content: "TCP stands for..." },
 *     { role: "user",      content: "What port does it use?" },
 *     { role: "assistant", content: "TCP uses port..." },
 *     { role: "user",      content: "Compare it with UDP" },  ← new
 *   ]
 *
 *   Yeh poora array LLM ko bhejte hain — tab LLM
 *   pehle ki baat "samjhta" hai aur context ke saath
 *   answer deta hai.
 *
 *
 * PROBLEM — CONTEXT WINDOW PHIR SE:
 * ────────────────────────────────────
 *
 *   Jaise jaise conversation lamba hota hai:
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Turn 1:   [msg1]                    → 100 tokens        │
 *   │  Turn 10:  [msg1...msg10]            → 1,000 tokens      │
 *   │  Turn 100: [msg1...msg100]           → 10,000 tokens     │
 *   │  Turn 500: [msg1...msg500]           → 50,000 tokens     │
 *   │                                                          │
 *   │  Context window: 200,000 tokens → eventually FULL!       │
 *   │  Cost bhi ↑ kyunki tokens ↑                              │
 *   └──────────────────────────────────────────────────────────┘
 *
 * SOLUTION (Basic non-streaming chatbot mein):
 *
 *   Option 1: Last N messages hi bhejo (sliding window)
 *             e.g., sirf last 20 messages
 *
 *   Option 2: Summary banana (Phase-I ki LTM concept)
 *             Purani history summarize karo → compress karo
 *
 *   Option 3: Hybrid — recent messages + LTM summary inject
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Non-streaming basic chatbot mein:                       │
 *   │  Option 1 (sliding window) sabse simple aur common hai   │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 4 — REQUEST FLOW (STEP BY STEP, DETAILED)
 * ══════════════════════════════════════════════════════════════
 *
 *
 * STEP 1: User message type karta hai
 * ────────────────────────────────────
 *
 *   User UI mein likhta hai: "Compare TCP and UDP"
 *   "Send" button dabata hai.
 *   Ek conversationId bhi hoti hai — is chat session ki ID.
 *
 *
 * STEP 2: Client → Backend POST request
 * ──────────────────────────────────────
 *
 *   POST /chat
 *   Content-Type: application/json
 *
 *   {
 *     "conversationId": "conv_abc123",
 *     "message": "Compare TCP and UDP"
 *   }
 *
 *   Note: conversationId se backend samjhega ki yeh
 *   KAUN SI conversation hai → history fetch kare.
 *
 *
 * STEP 3: Backend — History fetch karo
 * ──────────────────────────────────────
 *
 *   Backend DB/cache se is conversationId ki
 *   pichli messages fetch karta hai.
 *
 *   fetchedHistory = [
 *     { role: "user",      content: "What is TCP?" },
 *     { role: "assistant", content: "TCP stands for..." },
 *   ]
 *
 *   Phir SLIDING WINDOW apply karo:
 *   Agar history bahut lambi hai → sirf last 20 messages lo
 *
 *
 * STEP 4: LLM ko poora context do
 * ─────────────────────────────────
 *
 *   LLM ko yeh saab bhejte hain:
 *
 *   System Prompt (hamesha):
 *     "You are a helpful networking assistant.
 *      Answer based on provided context only."
 *
 *   Messages Array (history + new message):
 *     [
 *       { role: "user",      content: "What is TCP?" },
 *       { role: "assistant", content: "TCP stands for..." },
 *       { role: "user",      content: "Compare TCP and UDP"
 *                                      + "\nContext: [RAG chunks]" }
 *     ]
 *
 *   LLM → POORA answer generate karta hai
 *
 *   ⏳ Yahan user WAIT karta hai — kuch dikhta nahi
 *      (Yahi non-streaming ka drawback hai)
 *
 *
 * STEP 5: Answer save karo → Client ko bhejo
 * ─────────────────────────────────────────────
 *
 *   a. New user message + AI answer → DB mein save karo
 *      (Taaki next request pe history mein aaye)
 *
 *   b. Client ko full JSON response bhejo:
 *
 *      {
 *        "conversationId": "conv_abc123",
 *        "answer": "TCP (Transmission Control Protocol) and UDP...",
 *        "tokensUsed": 342
 *      }
 *
 *
 * STEP 6: Client — UI update karo
 * ─────────────────────────────────
 *
 *   JSON response aaya → answer UI mein show karo.
 *   Poora text ek dum appear hota hai (no typing effect).
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 5 — SYSTEM BEHAVIOR — DEEP THEORY
 * ══════════════════════════════════════════════════════════════
 *
 *
 * 1. SYNCHRONOUS EXECUTION KYA HOTA HAI?
 * ─────────────────────────────────────────
 *
 *   Synchronous = ek kaam hota hai → khatam hone ka wait → agla kaam
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Timeline:                                               │
 *   │                                                          │
 *   │  t=0s  → User request bhejta hai                         │
 *   │  t=0s  → Backend: History fetch karo   ← wait            │
 *   │  t=0.1s→ Backend: RAG embedding        ← wait            │
 *   │  t=0.3s→ Backend: Vector search        ← wait            │
 *   │  t=0.5s→ Backend: LLM call starts      ← wait            │
 *   │  t=8s  → LLM finishes generation                         │
 *   │  t=8s  → Backend: DB mein save         ← wait            │
 *   │  t=8.1s→ Client ko response            ← user dekhta hai │
 *   │                                                          │
 *   │  User 8.1 seconds tak KUCH NAHI dekhta   ← BAD UX        │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * 2. HTTP CONNECTION LIFECYCLE:
 * ──────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  [Client connects]                                       │
 *   │         ↓                                                │
 *   │  [Request sent]                                          │
 *   │         ↓                                                │
 *   │  [Connection OPEN — waiting]                             │
 *   │         ↓                                                │
 *   │  [Server processes everything]                           │
 *   │         ↓                                                │
 *   │  [Response sent — single payload]                        │
 *   │         ↓                                                │
 *   │  [Connection CLOSED]  ← ek request = ek response = close │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Non-streaming = Connection sirf response ke waqt open hoti hai
 *                   Khatam hote hi band ho jaati hai
 *
 *
 * 3. SERVER LOAD — MULTIPLE USERS KA EFFECT:
 * ────────────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  1 user    → 1 LLM call → 8 seconds hold                 │
 *   │  10 users  → 10 concurrent calls → server busy           │
 *   │  100 users → 100 concurrent LLM calls                    │
 *   │                                                          │
 *   │  Problem: Node.js event loop ek thread pe kaam karta hai │
 *   │  Agar LLM call synchronously block kare → ek user        │
 *   │  baki sabko block kar deta                               │
 *   │                                                          │
 *   │  Solution: async/await use karo → LLM wait mein          │
 *   │  event loop aur requests handle karta rehta hai ✓        │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * 4. COST ANALYSIS — NON-STREAMING VS STREAMING:
 * ─────────────────────────────────────────────────
 *
 *   LLM cost tokens pe based hoti hai — streaming ya non-streaming
 *   se cost SAME hoti hai. Fark sirf UX ka hai.
 *
 *   ┌──────────────────────┬─────────────────────────────────┐
 *   │                      │ Non-Streaming                   │
 *   ├──────────────────────┼─────────────────────────────────┤
 *   │ Input tokens         │ (history + RAG + query)         │
 *   │                      │ = 500 tokens → $0.003           │
 *   ├──────────────────────┼─────────────────────────────────┤
 *   │ Output tokens        │ Full answer                     │
 *   │                      │ = 200 tokens → $0.006           │
 *   ├──────────────────────┼─────────────────────────────────┤
 *   │ Total per request    │ ~$0.009                         │
 *   ├──────────────────────┼─────────────────────────────────┤
 *   │ 10,000 users/day     │ ~$90/day                        │
 *   └──────────────────────┴─────────────────────────────────┘
 *
 *   Sliding window kyun zaroori hai:
 *   Bina window ke → history 500 messages → 5000 input tokens
 *   → cost 10x badh jaati hai
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 6 — COMPLETE FLOW DIAGRAM (EK NAZAR MEIN)
 * ══════════════════════════════════════════════════════════════
 *
 *
 *   USER KI SIDE:
 *   ─────────────
 *   "Compare TCP and UDP"
 *         │
 *         │ POST /chat { conversationId, message }
 *         ▼
 *
 *   BACKEND KI SIDE:
 *   ─────────────────
 *   ┌─────────────────────────────────────────────────────┐
 *   │                                                     │
 *   │  1. History Fetch                                   │
 *   │     DB → last 20 messages                           │
 *   │                                                     │
 *   │  2. RAG (if enabled)                                │
 *   │     Query → Embedding → Vector Search → Top 3 Chunks│
 *   │                                                     │
 *   │  3. Prompt Assembly                                 │
 *   │     System Prompt + History + RAG Context + Query   │
 *   │                                                     │
 *   │  4. LLM Call (BLOCKING)                             │
 *   │     ⏳ 5-15 seconds wait...                         │
 *   │     Full answer aaya                                │
 *   │                                                     │
 *   │  5. DB Save                                         │
 *   │     User message + AI answer → messages table       │
 *   │                                                     │
 *   │  6. Response Return                                 │
 *   │     { answer, tokensUsed } → client                 │
 *   │                                                     │
 *   └─────────────────────────────────────────────────────┘
 *         │
 *         │ JSON { answer: "TCP and UDP differ..." }
 *         ▼
 *
 *   USER DEKHTA HAI:
 *   ─────────────────
 *   "TCP and UDP differ in that TCP is connection-oriented..."
 *   (Poora text ek dum appear hota hai)
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 7 — LIMITATIONS (WHY STREAMING ZAROOORI HAI)
 * ══════════════════════════════════════════════════════════════
 *
 *
 * LIMITATION 1 — BAD PERCEIVED PERFORMANCE:
 * ───────────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Actual time:    8 seconds                               │
 *   │  Perceived time: 8 seconds (kuch nahi dikh raha)         │
 *   │                                                          │
 *   │  Streaming mein:                                         │
 *   │  Actual time:    8 seconds                               │
 *   │  Perceived time: "Fast!" (instantly words aane lagte)    │
 *   │                                                          │
 *   │  Same speed, different feel — yahi UX ka magic hai       │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * LIMITATION 2 — TIMEOUT RISK:
 * ──────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  HTTP default timeout: 30-60 seconds (load balancer)     │
 *   │  LLM long response: 45 seconds                           │
 *   │                                                          │
 *   │  Result: Request timeout → error → response lost   X     │
 *   │                                                          │
 *   │  Streaming mein: Connection active rehti hai → no timeout│
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * LIMITATION 3 — NO EARLY EXIT:
 * ───────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  LLM galat direction mein ja raha hai → user dekhna      │
 *   │  chahta hai aur rok dena chahta hai                      │
 *   │                                                          │
 *   │  Non-streaming mein: IMPOSSIBLE — poora generate hoga    │
 *   │  Streaming mein: User "Stop" dabaye → generation ruk     │
 *   │  jaye (API stream cancel karo)                           │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * LIMITATION 4 — WASTED COMPUTE:
 * ────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  User ne request ki → phir tab close kar diya            │
 *   │                                                          │
 *   │  Non-streaming: LLM POORA generate karta rehta hai       │
 *   │  (User chala gaya, paise ja rahe hain)           X       │
 *   │                                                          │
 *   │  Streaming: User disconnect → stream cancel → LLM ruk    │
 *   │  jaata hai → cost bachti hai                    ✓✓✓     │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 8 — KAB USE KAREIN NON-STREAMING?
 * ══════════════════════════════════════════════════════════════
 *
 *   ┌──────────────────────────────┬──────────────────────────┐
 *   │ Use Case                     │ Reason                   │
 *   ├──────────────────────────────┼──────────────────────────┤
 *   │ Admin / Internal tools       │ UX utna important nahi   │
 *   │                              │ developers use karte hain│
 *   ├──────────────────────────────┼──────────────────────────┤
 *   │ Batch processing             │ Ek baar mein 1000        │
 *   │ (emails, reports generate)   │ documents process karo   │
 *   │                              │ no user waiting          │
 *   ├──────────────────────────────┼──────────────────────────┤
 *   │ Short responses expected     │ "What is 2+2?" →         │
 *   │                              │ "4" — instant anyway     │
 *   ├──────────────────────────────┼──────────────────────────┤
 *   │ API integrations             │ Machine-to-machine       │
 *   │ (B2B, microservices)         │ no human waiting         │
 *   ├──────────────────────────────┼──────────────────────────┤
 *   │ Classification tasks         │ "Is this spam?" →        │
 *   │                              │ Yes/No — quick           │
 *   └──────────────────────────────┴──────────────────────────┘
 *
 *   ✗ Use mat karo jab:
 *   - End user facing product hai (ChatGPT jaise)
 *   - Long responses expected hain (essays, code)
 *   - Real-time feel important hai
 */


/**
 * ╔══════════════════════════════════════════════════════════════╗
 * ║         TEXT-BASED CHATBOT (WITH STREAMING)                  ║
 * ╚══════════════════════════════════════════════════════════════╝
 * 
 * Phase-II mein kya badlega:
 *   → Conversation history (array) — SAME rahega
 *   → LLM call — SAME rahega
 *   → Sirf DELIVERY ka tarika badlega
 *   → Token generate hote hi user tak pohonchega — word by word
 *
 * Note: Is phase mein RAG aur Memory nahi hai.
 *       Sirf streaming pe focus hai. Clean aur isolated concept.
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 1 — FIRST PRINCIPLES: LLM ANDAR SE KAISE KAAM KARTA HAI
 * ══════════════════════════════════════════════════════════════
 *
 * Yeh samajhna zaroori hai — tabhi streaming ka need samjhega.
 *
 *
 * LLM TOKEN-BY-TOKEN GENERATE KARTA HAI:
 * ────────────────────────────────────────
 *
 *   LLM ek baar mein POORA answer nahi banata.
 *   Yeh bar bar ek probability distribution calculate karta hai:
 *   "Agli position pe kaun sa token aana chahiye?"
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Input: "What is TCP?"                                   │
 *   │                                                          │
 *   │  t=10ms → "TCP"       (1st token predict)                │
 *   │  t=20ms → " is"       (2nd token predict)                │
 *   │  t=30ms → " a"        (3rd token predict)                │
 *   │  t=40ms → " reliable" (4th token predict)                │
 *   │  t=50ms → " protocol" (5th token predict)                │
 *   │  ...                                                     │
 *   │  t=800ms→ [END]       (generation complete)              │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Har token ~10-20ms mein generate hota hai.
 *   200 token response = 2-4 seconds generation time.
 *
 *
 * MISMATCH — GENERATION vs DELIVERY:
 * ─────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  LLM andar se:  incremental  (token by token)            │
 *   │  Phase-I mein:  batched      (sab kuch ek saath)         │
 *   │                                                          │
 *   │  Yeh MISMATCH hai — LLM ready hai tokens bhejne ko       │
 *   │  lekin Phase-I mein hum unhe rok ke rakhte the!          │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * STREAMING KA SOLUTION:
 * ───────────────────────
 *
 *   "Token generate hua → turant client ko bhejo"
 *   "Wait mat karo poora response aane ka"
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Generation aur delivery ek saath chalti hai             │
 *   │                                                          │
 *   │  LLM: "TCP"       → turant client: "TCP"                 │
 *   │  LLM: " is"       → turant client: "TCP is"              │
 *   │  LLM: " a"        → turant client: "TCP is a"            │
 *   │  LLM: " reliable" → turant client: "TCP is a reliable"   │
 *   │  ...                                                     │
 *   │                                                          │
 *   │  User pehla word dekhta hai 10ms mein — not 800ms!  ✓   │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * METRICS IMPACT (ENGINEERING TERMS):
 * ─────────────────────────────────────
 *
 *   ┌────────────────────┬──────────────┬──────────────────────┐
 *   │ Metric             │ Non-Streaming│ Streaming            │
 *   ├────────────────────┼──────────────┼──────────────────────┤
 *   │ TTFB               │ 8-15 seconds │ ~50-200ms     ✓✓✓   │
 *   │ (Time to First     │              │                      │
 *   │  Byte/Token)       │              │                      │
 *   ├────────────────────┼──────────────┼──────────────────────┤
 *   │ Perceived Latency  │ HIGH         │ LOW           ✓✓✓   │
 *   │ (User ka feel)     │ (frustrating)│ (responsive)         │
 *   ├────────────────────┼──────────────┼──────────────────────┤
 *   │ Actual Latency     │ Same         │ Same                 │
 *   │ (Total time)       │              │ (cost same hai)      │
 *   ├────────────────────┼──────────────┼──────────────────────┤
 *   │ Timeout Risk       │ HIGH         │ ZERO          ✓✓✓   │
 *   ├────────────────────┼──────────────┼──────────────────────┤
 *   │ Early Cancel       │ NOT possible │ Possible      ✓✓✓   │
 *   └────────────────────┴──────────────┴──────────────────────┘
 *
 *   Key insight: Total time same hai, cost same hai.
 *   Sirf USER EXPERIENCE aur reliability improve hoti hai.
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 2 — HTTP KI LIMITATION AUR DO SOLUTIONS
 * ══════════════════════════════════════════════════════════════
 *
 * Traditional HTTP kya problem create karta hai?
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  HTTP = Request → Response → Connection Close            │
 *   │                                                          │
 *   │  Server KHUD SE client ko data push nahi kar sakta       │
 *   │  Har data piece ke liye naya request chahiye             │
 *   │  Ek open request mein partial data bheja nahi ja sakta   │
 *   └──────────────────────────────────────────────────────────┘
 *
 * Streaming ke liye 2 solutions exist karte hain:
 *
 *   ┌──────────────────┬───────────────────────────────────────┐
 *   │ Solution         │ Concept                               │
 *   ├──────────────────┼───────────────────────────────────────┤
 *   │ SSE              │ HTTP hi use karo, lekin connection    │
 *   │ (Server-Sent     │ open rakho — server baar baar         │
 *   │  Events)         │ data push karta rahe                  │
 *   ├──────────────────┼───────────────────────────────────────┤
 *   │ WebSocket        │ HTTP se bilkul alag protocol          │
 *   │                  │ Ek baar connect karo — dono sides     │
 *   │                  │ freely data bhej sakti hain           │
 *   └──────────────────┴───────────────────────────────────────┘
 *
 * Dono alag hain — alag use cases ke liye.
 * Dono detail mein cover karenge.
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 3 — SSE (SERVER-SENT EVENTS) — DEEP DIVE
 * ══════════════════════════════════════════════════════════════
 *
 *
 * SSE KYA HAI?
 * ─────────────
 *
 *   SSE ek HTTP-based technique hai jisme:
 *   - Client ek baar connection open karta hai
 *   - Server us connection pe baar baar events/data push karta hai
 *   - Connection tab tak open rehti hai jab tak server "done" na kare
 *   - Sirf SERVER → CLIENT direction (one-way)
 *
 *   Sochlo jaise ek radio broadcast:
 *   Tumne radio on kiya (connection open).
 *   Radio station ek ke baad ek songs bajata rehta hai (server pushes).
 *   Tum sirf sun sakte ho — radio station ko respond nahi kar sakte.
 *   Jab program khatam → radio band (connection close).
 *
 *
 * SSE KA DATA FORMAT:
 * ────────────────────
 *
 *   SSE ka ek specific text format hota hai:
 *
 *   data: {json payload here}\n\n
 *
 *   Rules:
 *   - Har event "data: " se shuru hota hai
 *   - Event ke end mein TWO newlines (\n\n) — yeh ZAROORI hai
 *   - \n\n = event ka delimiter (yahi batata hai "ek event khatam")
 *   - JSON andar ho sakta hai — flexible
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Wire pe actual data kuch aisa dikhta hai:               │
 *   │                                                          │
 *   │  data: {"token":"TCP"}\n\n                               │
 *   │  data: {"token":" is"}\n\n                               │
 *   │  data: {"token":" a"}\n\n                                │
 *   │  data: {"token":" reliable"}\n\n                         │
 *   │  data: {"done":true}\n\n                                 │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * SSE LIFECYCLE — STEP BY STEP:
 * ───────────────────────────────
 *
 *   STEP 1: Client HTTP request bhejta hai (special headers ke saath)
 *   STEP 2: Server response headers set karta hai (SSE mode)
 *   STEP 3: Connection OPEN rehti hai — normal HTTP ki tarah close nahi
 *   STEP 4: Server baar baar res.write() se events push karta hai
 *   STEP 5: Client onmessage handler mein events receive karta hai
 *   STEP 6: Server res.end() call karta hai → connection close
 *
 *
 * SSE SYSTEM FLOW DIAGRAM:
 * ─────────────────────────
 *
 *   [Browser]                         [Express Server]           [LLM]
 *      │                                     │                     │
 *      │── POST /chat (msg + history) ──────>│                     │
 *      │                                     │── stream() ────────>│
 *      │   Connection OPEN rehti hai         │                     │
 *      │                                     │<── token: "TCP" ────│
 *      │<── data: {"token":"TCP"}\n\n ────── │                     │
 *      │   UI: "TCP"                         │<── token: " is" ────│
 *      │<── data: {"token":" is"}\n\n ────── │                     │
 *      │   UI: "TCP is"                      │<── token: " a" ─────│
 *      │<── data: {"token":" a"}\n\n ─────── │                     │
 *      │   UI: "TCP is a"                    │        ...          │
 *      │           ...                       │<── [STREAM END] ────│
 *      │<── data: {"done":true}\n\n ─────────│                     │
 *      │   Connection CLOSE                  │                     │
 *
 *
 * SSE MEIN CONVERSATION HISTORY KAISE BHEJEIN?
 * ──────────────────────────────────────────────
 *
 *   Phase-I mein POST body mein history + message bhejte the.
 *   SSE mein bhi wahi — POST body ke saath.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  IMPORTANT: Browser ka EventSource sirf GET support      │
 *   │  karta hai — POST nahi.                                  │
 *   │                                                          │
 *   │  Solution: EventSource ki jagah fetch() use karo         │
 *   │  fetch() POST + streaming dono support karta hai         │
 *   │  (ReadableStream se response consume karo)               │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Client bhejega:
 *   POST /chat
 *   {
 *     "message": "What is TCP?",
 *     "history": [
 *       { "role": "user",      "content": "Hello" },
 *       { "role": "assistant", "content": "Hi! How can I help?" }
 *     ]
 *   }
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 4 — SSE CLIENT PSEUDOCODE (BROWSER SIDE)
 * ══════════════════════════════════════════════════════════════
 *
 * Client ki responsibility:
 * 1. POST request bhejo (message + history)
 * 2. Response ko stream ki tarah padhte raho
 * 3. Har token aate hi UI mein append karo
 * 4. "done" aane pe history update karo
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 5 — SSE MEIN EDGE CASES (PRODUCTION MEIN IMPORTANT)
 * ══════════════════════════════════════════════════════════════
 *
 *
 * EDGE CASE 1 — BUFFER FRAGMENTATION:
 * ─────────────────────────────────────
 *
 *   Network kabhi kabhi ek SSE event ko 2 TCP packets mein
 *   split kar deta hai:
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Expected: data: {"token":"TCP"}\n\n                     │
 *   │                                                          │
 *   │  Actually arrive karta hai:                              │
 *   │  Chunk 1: data: {"tok                                    │
 *   │  Chunk 2: en":"TCP"}\n\n                                 │
 *   │                                                          │
 *   │  Agar buffer nahi rakha → JSON.parse fail → crash   ✗✗  │
 *   │  Buffer ke saath → dono chunks join → parse OK      ✓✓✓ │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Solution: Buffer += chunk → \n\n se split → last piece buffer mein raho
 *
 *
 * EDGE CASE 2 — NETWORK DISCONNECT:
 * ────────────────────────────────────
 *
 *   User ka internet beech mein chala gaya → SSE connection toot gayi.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  SSE ka built-in feature: AUTO RECONNECT                 │
 *   │                                                          │
 *   │  Browser automatically reconnect karta hai!              │
 *   │  (EventSource use kiya ho tab)                           │
 *   │                                                          │
 *   │  fetch() ke saath: manually handle karna padega          │
 *   │  try-catch + retry logic likhna padega                   │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * EDGE CASE 3 — USER "STOP" BUTTON:
 * ────────────────────────────────────
 *
 *   User generation beech mein rokna chahta hai.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Client side:                                            │
 *   │  AbortController use karo:                               │
 *   │                                                          │
 *   │  const controller = new AbortController()                │
 *   │  fetch('/chat', { signal: controller.signal, ... })      │
 *   │                                                          │
 *   │  Stop button click:                                      │
 *   │  controller.abort() → fetch cancel → connection close    │
 *   │                                                          │
 *   │  Server side:                                            │
 *   │  req.on('close') → LLM stream cancel → tokens stop       │
 *   │  Cost bachti hai ✓✓✓                                    │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 7 — WEBSOCKET — DEEP DIVE
 * ══════════════════════════════════════════════════════════════
 *
 *
 * WEBSOCKET KYA HAI?
 * ───────────────────
 *
 *   WebSocket ek alag protocol hai (HTTP nahi).
 *   Isme ek baar "handshake" hota hai, phir ek
 *   PERSISTENT BIDIRECTIONAL channel khul jaata hai.
 *
 *   Dono sides — client aur server — KABHI BHI data bhej sakte hain.
 *   Koi request-response cycle nahi — pure event-driven.
 *
 *   Sochlo jaise ek Whatsapp call:
 *   Ek baar connect → dono baat kar sakte hain freely
 *   Koi "aap pehle bolo, main baad mein" rule nahi
 *
 *
 * WEBSOCKET HANDSHAKE — HTTP SE UPGRADE:
 * ────────────────────────────────────────
 *
 *   WebSocket shuru hota hai ek special HTTP request se:
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Client bhejta hai:                                      │
 *   │                                                          │
 *   │  GET /ws HTTP/1.1                                        │
 *   │  Host: server.com                                        │
 *   │  Upgrade: websocket     ← "WebSocket mein switch karo"   │
 *   │  Connection: Upgrade                                     │
 *   │  Sec-WebSocket-Key: dGhlIHNhbXBsZQ==  ← random key       │
 *   │  Sec-WebSocket-Version: 13                               │
 *   │                                                          │
 *   │  Server respond karta hai:                               │
 *   │                                                          │
 *   │  HTTP/1.1 101 Switching Protocols                        │
 *   │  Upgrade: websocket     ← "Theek hai, switch kar rahe"   │
 *   │  Connection: Upgrade                                     │
 *   │  Sec-WebSocket-Accept: s3pPLMBiTxaQ9kYGzzhZRbK+xOo=      │
 *   │                                                          │
 *   │  Ab HTTP band — WebSocket protocol shuru ✓✓✓            │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * WEBSOCKET vs SSE — KAB KYA CHOOSE KAREIN:
 * ────────────────────────────────────────────
 *
 *   ┌────────────────────────┬───────────────┬───────────────────┐
 *   │ Feature                │ SSE           │ WebSocket         │
 *   ├────────────────────────┼───────────────┼───────────────────┤
 *   │ Direction              │ Server→Client │ Both ways         │
 *   │                        │ (one-way)     │ (two-way)         │
 *   ├────────────────────────┼───────────────┼───────────────────┤
 *   │ Protocol               │ HTTP          │ ws:// (alag)      │
 *   ├────────────────────────┼───────────────┼───────────────────┤
 *   │ Implementation         │ Simple        │ Thoda complex     │
 *   ├────────────────────────┼───────────────┼───────────────────┤
 *   │ Proxy/Firewall         │ Friendly      │ Kabhi blocked     │
 *   ├────────────────────────┼───────────────┼───────────────────┤
 *   │ Auto-reconnect         │ Built-in      │ Manual likhna     │
 *   ├────────────────────────┼───────────────┼───────────────────┤
 *   │ Per message overhead   │ Zyada         │ Bahut kam         │
 *   │                        │ (HTTP headers)│                   │
 *   ├────────────────────────┼───────────────┼───────────────────┤
 *   │ Server memory          │ Kam           │ Zyada             │
 *   │                        │               │ (persistent conn) │
 *   ├────────────────────────┼───────────────┼───────────────────┤
 *   │ Best for               │ AI streaming  │ Live chat, games, │
 *   │                        │ Notifications │ Collab tools      │
 *   └────────────────────────┴───────────────┴───────────────────┘
 *
 *   AI Chatbot ke liye: SSE preferred hai
 *   Real-time multiplayer / collab: WebSocket preferred hai
 *
 *
 * WEBSOCKET SYSTEM FLOW:
 * ───────────────────────
 *
 *   [Browser]                        [WebSocket Server]          [LLM]
 *      │                                    │                      │
 *      │── HTTP Upgrade Request ───────────>│                      │
 *      │<── 101 Switching Protocols ────────│                      │
 *      │                                    │                      │
 *      │    [PERSISTENT CHANNEL OPEN]       │                      │
 *      │                                    │                      │
 *      │── {type:'chat', msg, history} ────>│                      │
 *      │                                    │── stream() ─────────>│
 *      │                                    │<── token:"TCP" ───── │
 *      │<── {type:'token', token:'TCP'} ────│                      │
 *      │   UI: "TCP"                        │<── token:" is" ───── │
 *      │<── {type:'token', token:' is'} ────│                      │
 *      │   UI: "TCP is"                     │        ...           │
 *      │           ...                      │<── [STREAM END] ──── │
 *      │<── {type:'done'} ──────────────────│                      │
 *      │                                    │                      │
 *      │── {type:'chat', msg2, history} ───>│  ← bina naya        │
 *      │   (SAME connection pe next message)│    connection khole  │
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 11 — COMPLETE FLOW COMPARISON (EK NAZAR MEIN)
 * ══════════════════════════════════════════════════════════════
 *
 *
 * PHASE-I (Non-Streaming) vs PHASE-II (Streaming):
 * ──────────────────────────────────────────────────
 *
 *   ┌──────────────────────────────┬──────────────────────────────┐
 *   │ Phase-I                      │ Phase-II                     │
 *   ├──────────────────────────────┼──────────────────────────────┤
 *   │ POST /chat                   │ POST /chat (SSE)             │
 *   │ JSON body                    │ JSON body (same!)            │
 *   ├──────────────────────────────┼──────────────────────────────┤
 *   │ messages.create() — blocking │ messages.stream() — async    │
 *   ├──────────────────────────────┼──────────────────────────────┤
 *   │ res.json({answer}) — ek baar │ res.write() — baar baar      │
 *   ├──────────────────────────────┼──────────────────────────────┤
 *   │ await response.json()        │ reader.read() loop           │
 *   ├──────────────────────────────┼──────────────────────────────┤
 *   │ UI.textContent = answer      │ UI.textContent += token      │
 *   ├──────────────────────────────┼──────────────────────────────┤
 *   │ User 8s blank dekhta hai     │ User 50ms mein pehla word ✓  │
 *   └──────────────────────────────┴──────────────────────────────┘
 *
 *   Jo SAME raha:
 *   → Conversation history (array)
 *   → Sliding window (last 20)
 *   → LLM model, system prompt
 *   → Request body structure
 *
 *   Jo BADLA:
 *   → LLM call method (.stream vs .create)
 *   → Response delivery (chunks vs full)
 *   → Client reading (ReadableStream vs res.json)
 *   → UX (progressive vs all-at-once)
 */


/**
 * ══════════════════════════════════════════════════════════════
 * FINAL SUMMARY — PHASE-II
 * ══════════════════════════════════════════════════════════════
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  STREAMING KYA HAI:                                      │
 *   │  Token generate hua → turant client ko bhejo             │
 *   │  Connection tab tak open rakho jab tak sab tokens na     │
 *   │  aa jaayein                                              │
 *   │                                                          │
 *   │  DO TARIKE:                                              │
 *   │  SSE       → HTTP based, one-way, simple, AI ke liye ✓✓✓│
 *   │  WebSocket → alag protocol, two-way, complex apps ke liye│
 *   │                                                          │
 *   │  PHASE-I SE KYA BADLA:                                   │
 *   │  Sirf delivery layer — baaki sab same                    │
 *   │  Same history, same LLM, same logic                      │
 *   │                                                          │
 *   │  NEXT STEP — PHASE-III:                                  │
 *   │  Is streaming chatbot mein RAG add karenge               │
 *   │  User document upload kare → chatbot us pe               │
 *   │  stream karke answer de                                  │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * EVOLUTION PATH:
 * ────────────────
 *
 *   Phase-I  → Non-streaming chatbot (array history) ✅
 *   Phase-II → Streaming chatbot (SSE + WS)          ✅ ← ABHI YAHAN
 *   Phase-III→ Streaming chatbot + RAG               ⏳
 *   Phase-IV → Streaming chatbot + Memory            ⏳
 *   Phase-V  → Agentic chatbot (Tools)               ⏳
 */


/**
 * ╔══════════════════════════════════════════════════════════════╗
 * ║ CHATBOT + RAG INTEGRATION + STREAMING                        ║
 * ╚══════════════════════════════════════════════════════════════╝
 * 
 * Abhi tak humne seekha (Quick Recap):
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Phase-I  → Non-Streaming Chatbot                        │
 *   │             User query → History (array) → LLM →         │
 *   │             Poora answer ek baar mein                    │
 *   │                                                          │
 *   │  Phase-II → Streaming Chatbot                            │
 *   │             Same logic + SSE/WebSocket se                │
 *   │             word-by-word delivery                        │
 *   │                                                          │
 *   │  Dono mein ek badi problem thi:                          │
 *   │  LLM sirf apni TRAINING DATA se answer karta tha         │
 *   │  → Private data nahi pata                    ✗✗         │
 *   │  → Updated info nahi pata                    ✗✗         │
 *   │  → Hallucination ka risk                     ✗✗         │
 *   └──────────────────────────────────────────────────────────┘
 *
 * Phase-III mein kya add hoga:
 *   → RAG layer add karenge — streaming ke saath
 *   → User apne documents upload kare → chatbot us pe answer de
 *   → Har answer grounded hoga real data mein — hallucination kam
 *   → Streaming same rahega (Phase-II ka code reuse hoga)
 *
 * Note: Is phase mein Memory nahi hai.
 *       RAG + Streaming ka combination — clean aur isolated.
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 1 — FIRST PRINCIPLES: LLM KI LIMITATION KYA HAI?
 * ══════════════════════════════════════════════════════════════
 *
 *
 * LLM EK STATIC KNOWLEDGE BASE HAI:
 * ────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  LLM = ek insaan jo bahut padha-likha hai                │
 *   │        lekin 2024 ke baad ki koi bhi khabar nahi         │
 *   │        aur tumhari company ke internal documents         │
 *   │        kabhi padhe hi nahi                               │
 *   │                                                          │
 *   │  Woh apni memory se answer karta hai — accurate nahi     │
 *   │  hoga agar question naye ya private data pe ho           │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   3 Core Limitations:
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  1. TRAINING CUTOFF                                      │
 *   │     "What is the latest iPhone model?" →                 │
 *   │     LLM 2024 ke baad ka model nahi jaanta                │
 *   │                                                          │
 *   │  2. PRIVATE DATA                                         │
 *   │     "What does our refund policy say?" →                 │
 *   │     LLM tumhari company ka PDF kabhi nahi padha          │
 *   │                                                          │
 *   │  3. HALLUCINATION                                        │
 *   │     LLM confident lagti answer deta hai                  │
 *   │     jo bilkul galat hoti hai — kyunki woh                │
 *   │     "guess" karta hai jab nahi pata                      │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * RAG KA SOLUTION:
 * ─────────────────
 *
 *   "LLM se pehle, relevant documents dhundho.
 *    Phir woh documents LLM ko do.
 *    Ab LLM apni training se nahi — TUMHARE DATA se answer dega."
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Bina RAG: "Tumhari refund policy kya hai?"              │
 *   │  LLM: "Generally, refunds are processed in 30 days..."   │
 *   │  ← Hallucination — LLM ne guess kiya                 ✗✗ │
 *   │                                                          │
 *   │  RAG ke saath:                                           │
 *   │  1. "refund policy" → Vector DB mein search              │
 *   │  2. Relevant chunk mila: "Our policy: 7-day full refund" │
 *   │  3. LLM ko diya → "Based on the document: 7-day..."  ✓✓✓│
 *   └──────────────────────────────────────────────────────────┘
 *
 *   RAG = Retrieve + Augment + Generate
 *   Retrieve  → relevant documents dhundho
 *   Augment   → LLM prompt mein inject karo
 *   Generate  → LLM grounded answer deta hai
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 2 — RAG KE DO PHASES (BIG PICTURE)
 * ══════════════════════════════════════════════════════════════
 *
 * RAG do alag alag phases mein kaam karta hai:
 *
 *   ┌──────────────────────┬──────────────────────────────────┐
 *   │ Phase A: INDEXING    │ Phase B: QUERY                   │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Ek baar ka kaam      │ Har user request pe hota hai     │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Documents process    │ User query aati hai              │
 *   │ karo → store karo    │ → relevant chunks dhundho        │
 *   │                      │ → LLM ko do → answer aata hai    │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Offline / background │ Online / real-time               │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Cost: ek baar        │ Cost: har request pe             │
 *   └──────────────────────┴──────────────────────────────────┘
 *
 *
 * ANALOGY:
 * ─────────
 *
 *   Sochlo library mein ek librarian hai.
 *
 *   INDEXING = Library mein books arrange karna, catalogue banana
 *   (Ek baar kiya — phir baar baar use karo)
 *
 *   QUERY = Tum aaye → librarian ne relevant books dhundhe → diye
 *   (Har baar hota hai — fast aur targeted)
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 3 — PHASE A: INDEXING PIPELINE (DEEP DIVE)
 * ══════════════════════════════════════════════════════════════
 *
 *
 * INDEXING KYA HOTA HAI?
 * ───────────────────────
 *
 *   Documents ko is tarah process karo ki baad mein
 *   fast aur accurate search ho sake.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  [Raw Documents]                                         │
 *   │       ↓                                                  │
 *   │  [Step 1: Load & Clean]                                  │
 *   │       ↓                                                  │
 *   │  [Step 2: Chunking]                                      │
 *   │       ↓                                                  │
 *   │  [Step 3: Embedding]                                     │
 *   │       ↓                                                  │
 *   │  [Step 4: Vector DB mein Store]                          │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * STEP 1 — DOCUMENTS LOAD AUR CLEAN KARO:
 * ─────────────────────────────────────────
 *
 *   Sources kuch bhi ho sakti hain:
 *   PDF files, Word docs, Websites, Databases, APIs, etc.
 *
 *   Cleaning kyun?
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Raw PDF se extract kiya text:                           │
 *   │  "TCP  is a  proto-\ncol used for..."  ← messy           │
 *   │                                                          │
 *   │  Extra spaces, broken words, headers/footers, page       │
 *   │  numbers — yeh sab embedding quality kharab karte hain   │
 *   │                                                          │
 *   │  Clean text → better embeddings → better retrieval ✓✓✓  │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * STEP 2 — CHUNKING (CRITICAL DECISION):
 * ──────────────────────────────────────
 *
 *   Poora document ek saath store nahi karte — kyunki:
 *   a. Ek document bahut bada ho sakta hai
 *   b. LLM ko sirf relevant PART chahiye, poora nahi
 *   c. Embedding ek bade text ki specific meaning capture
 *      nahi kar sakti
 *
 *   Chunking strategies:
 *
 *   ┌──────────────────────┬──────────────────────────────────┐
 *   │ Strategy             │ Kab use karein                   │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Fixed size           │ Simple, general purpose          │
 *   │ (500 chars each)     │ Overlap rakhna zaroori           │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Paragraph-based      │ Natural text (articles, docs)    │
 *   │                      │ Context preserved hota hai       │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Page-based           │ PDFs — structure preserve        │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Semantic chunking     │ AI se detect karo kahan topic   │
 *   │                      │ badla → wahan cut karo           │
 *   │                      │ Best quality, highest cost       │
 *   └──────────────────────┴──────────────────────────────────┘
 *
 *   OVERLAP kyun?
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Chunk-1: "...TCP ensures reliable delivery. It uses"    │
 *   │  Chunk-2: "three-way handshake to establish..."          │
 *   │                                                          │
 *   │  "three-way handshake" sirf Chunk-2 mein hai             │
 *   │  Context split ho gaya — retrieval galat ho sakti hai    │
 *   │                                                          │
 *   │  WITH OVERLAP (100 char overlap):                        │
 *   │  Chunk-1: "...TCP ensures reliable delivery. It uses"    │
 *   │  Chunk-2: "It uses three-way handshake to establish..."  │
 *   │           ↑ Overlap — context preserved ✓                │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Chunk size tradeoff:
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  SMALL chunks (100-200 chars):                           │
 *   │  + Precise retrieval                                     │
 *   │  - Context loss — sentence incomplete ho sakti hai       │
 *   │                                                          │
 *   │  LARGE chunks (1000-2000 chars):                         │
 *   │  + Rich context                                          │
 *   │  - Noisy — irrelevant info bhi aa jaati hai              │
 *   │  - Higher LLM cost (zyada tokens)                        │
 *   │                                                          │
 *   │  Sweet spot: 300-500 chars with 50-100 char overlap      │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * STEP 3 — EMBEDDING GENERATION:
 * ────────────────────────────────
 *
 *   Har chunk ko numbers ke array mein convert karo.
 *   Yeh numbers chunk ki "meaning" represent karte hain.
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Text → Embedding Model → Vector                         │
 *   │                                                          │
 *   │  "TCP is a reliable protocol"                            │
 *   │         ↓ Embedding Model                                │
 *   │  [0.23, 0.87, 0.12, 0.45, 0.91, ...]  ← 1536 numbers     │
 *   │                                                          │
 *   │  "TCP ek reliable protocol hai"                          │
 *   │         ↓ Embedding Model                                │
 *   │  [0.22, 0.85, 0.13, 0.44, 0.90, ...]  ← similar! ✓✓✓    │
 *   │                                                          │
 *   │  Same meaning → similar numbers → similarity search      │
 *   │  kaam karta hai even across languages ✓✓✓               │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Popular embedding models:
 *   ┌────────────────────────┬───────────────────────────────┐
 *   │ Model                  │ Notes                         │
 *   ├────────────────────────┼───────────────────────────────┤
 *   │ text-embedding-3-small │ OpenAI, cheap, fast           │
 *   │ text-embedding-3-large │ OpenAI, better quality        │
 *   │ text-embedding-004     │ Google, strong multilingual   │
 *   │ nomic-embed-text       │ Open source, self-hostable    │
 *   └────────────────────────┴───────────────────────────────┘
 *
 *   IMPORTANT: Indexing aur Query mein SAME embedding model
 *   use karna ZAROORI hai — warna vectors incompatible honge.
 *
 *
 * STEP 4 — VECTOR DATABASE MEIN STORE:
 * ──────────────────────────────────────
 *
 *   Har chunk ke liye 3 cheezein store karo:
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  {                                                       │
 *   │    id:       "chunk_042",                                │
 *   │    vector:   [0.23, 0.87, 0.12, ...], ← search ke liye   │
 *   │    content:  "TCP is a reliable...",  ← LLM ko dena      │
 *   │    metadata: {                        ← filtering ke liye│
 *   │      docId:    "networking-guide.pdf",                   │
 *   │      pageNo:   20,                                       │
 *   │      chunkNo:  5,                                        │
 *   │      docTitle: "Networking Guide"                        │
 *   │    }                                                     │
 *   │  }                                                       │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Popular Vector Databases:
 *   ┌──────────────┬──────────────────────────────────────────┐
 *   │ DB           │ Best For                                 │
 *   ├──────────────┼──────────────────────────────────────────┤
 *   │ Pinecone     │ Managed, production, easy setup          │
 *   │ Qdrant       │ Open source, self-host, fast             │
 *   │ pgvector     │ Already using Postgres? Add vector!      │
 *   │ Chroma       │ Local dev, quick prototyping             │
 *   │ Weaviate     │ Hybrid search built-in                   │
 *   └──────────────┴──────────────────────────────────────────┘
 *
 *
 * COMPLETE INDEXING FLOW DIAGRAM:
 * ────────────────────────────────
 *
 *   [PDF / Docs / DB]
 *          │
 *          ▼ Load + Clean
 *   ["TCP is a protocol used for...  (raw text)"]
 *          │
 *          ▼ Chunk (500 chars, 100 overlap)
 *   [Chunk-1: "TCP is a protocol..."]
 *   [Chunk-2: "...protocol used for reliable..."]
 *   [Chunk-3: "...reliable data delivery..."]
 *          │
 *          ▼ Embedding Model (same model hamesha)
 *   [Vector-1: [0.23, 0.87, ...]]
 *   [Vector-2: [0.45, 0.12, ...]]
 *   [Vector-3: [0.21, 0.85, ...]]
 *          │
 *          ▼ Store
 *   [Vector DB]
 *    ┌─────────────────────────────────────────────────┐
 *    │  chunk-1: [0.23, 0.87...] "TCP is a protocol..."│
 *    │  chunk-2: [0.45, 0.12...] "...reliable..."      │
 *    │  chunk-3: [0.21, 0.85...] "...data delivery..." │
 *    └─────────────────────────────────────────────────┘
 *    Indexing DONE — ek baar ka kaam ✓
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 4 — PHASE B: QUERY PIPELINE (DEEP DIVE)
 * ══════════════════════════════════════════════════════════════
 *
 * Har user request pe yeh hota hai — fast hona chahiye.
 *
 *
 * QUERY PIPELINE STEPS:
 * ──────────────────────
 *
 *   STEP 1: User query aati hai
 *   STEP 2: Query ka embedding banao (SAME model jo indexing mein use kiya)
 *   STEP 3: Vector DB mein similarity search karo
 *   STEP 4: Top-K relevant chunks milte hain
 *   STEP 5: Conversation history + chunks + query → LLM
 *   STEP 6: LLM grounded answer generate karta hai (streaming mein)
 *
 *
 * SIMILARITY SEARCH KYA HOTA HAI?
 * ─────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────────────┐
 *   │  Query: "What is TCP?"                                           │
 *   │  Query vector: [0.22, 0.84, 0.13, ...]                           │
 *   │                                                                  │
 *   │  Vector DB mein stored vectors ke saath compare karo:            │
 *   │                                                                  │
 *   │  chunk-1 vector: [0.23, 0.87, 0.12, ...] → similarity: 0.97 ✓✓✓ |
 *   │  chunk-2 vector: [0.45, 0.12, 0.78, ...] → similarity: 0.34      │
 *   │  chunk-3 vector: [0.21, 0.85, 0.14, ...] → similarity: 0.95 ✓✓✓ │
 *   │  chunk-4 vector: [0.88, 0.23, 0.45, ...] → similarity: 0.12      │
 *   │                                                                  │
 *   │  Top-K=2: chunk-1, chunk-3 return karo ✓✓✓                      │ 
 *   │                                                                  │
 *   │  SEMANTIC match — exact words nahi chahiye                       │
 *   │  "TCP protocol" aur "TCP networking" → similar vectors           │
 *   └──────────────────────────────────────────────────────────────────┘
 *
 *   Cosine Similarity:
 *   Vectors ke beech angle measure karta hai.
 *   1.0 = identical meaning, 0.0 = no relation, -1.0 = opposite
 *
 *
 * QUERY PIPELINE FLOW DIAGRAM:
 * ─────────────────────────────
 *
 *   [User]: "What is TCP and why is it reliable?"
 *         │
 *         ▼ Embedding (same model)
 *   Query Vector: [0.22, 0.84, 0.13, ...]
 *         │
 *         ▼ Similarity Search (Vector DB)
 *   Top-3 Chunks:
 *     Chunk-1 (similarity: 0.97): "TCP is a connection-oriented protocol..."
 *     Chunk-3 (similarity: 0.95): "TCP's reliability comes from ACK signals..."
 *     Chunk-7 (similarity: 0.89): "Three-way handshake establishes connection..."
 *         │
 *         ▼ Prompt Assembly
 *   System: "Answer based on context only."
 *   History: [prev messages]
 *   Context: [Chunk-1 + Chunk-3 + Chunk-7]
 *   User: "What is TCP and why is it reliable?"
 *         │
 *         ▼ LLM (Streaming)
 *   "TCP (Transmission Control Protocol) is a..."  → word by word → Client ✓
 *
 *
 * TOP-K KYA HONA CHAHIYE?
 * ────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  K=1: Bahut precise, lekin info miss ho sakti hai        │
 *   │  K=3: Sweet spot for most use cases ✓✓✓                 │
 *   │  K=5: Richer context, thoda zyada noise                  │
 *   │  K=10: Context window bhar sakta hai, cost badh jaati    │
 *   │                                                          │
 *   │  Rule of thumb: K=3 se shuru karo, tune karo             │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 5 — COMPLETE SYSTEM ARCHITECTURE
 * ══════════════════════════════════════════════════════════════
 *
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │               CLIENT                                     │
 *   │  - Chat UI (Phase-II jaisi)                              │
 *   │  - Stream listener (SSE/WebSocket)                       │
 *   │  - Document upload button (NEW)                          │
 *   └──────────────────┬───────────────────────────────────────┘
 *                      │
 *              ┌───────┴────────┐
 *              │                │
 *         POST /upload      POST /chat
 *         (indexing)        (query + stream)
 *              │                │
 *              ▼                ▼
 *   ┌──────────────────────────────────────────────────────────┐
 *   │                    BACKEND                               │
 *   │                                                          │
 *   │  /upload route:              /chat route:                │
 *   │  1. File receive karo        1. SSE headers set karo     │
 *   │  2. Text extract karo        2. History prepare karo     │
 *   │  3. Chunk karo               3. Query embed karo         │
 *   │  4. Embed karo               4. Vector search karo       │
 *   │  5. Vector DB mein store     5. Chunks + History → LLM   │
 *   │                              6. Stream tokens → client   │
 *   └──────┬───────────────────────────────┬───────────────────┘
 *          │                               │
 *          ▼                               ▼
 *   ┌──────────────┐               ┌──────────────────┐
 *   │  Vector DB   │               │    LLM           │
 *   │  (Qdrant /   │               │  (Claude/GPT)    │
 *   │   pgvector)  │               └──────────────────┘
 *   └──────────────┘
 */


/**
 * ╔══════════════════════════════════════════════════════════════╗
 * ║               AGENTIC CHATBOT (TOOLS)                        ║
 * ║    (TOOL USE + FUNCTION CALLING + AUTONOMOUS ACTIONS)        ║
 * ╚══════════════════════════════════════════════════════════════╝
 *
 * Abhi tak humne seekha (Quick Recap):
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Phase-I   → Non-Streaming Chatbot (array history)       │
 *   │  Phase-II  → Streaming Chatbot (SSE/WebSocket)           │
 *   │  Phase-III → RAG Chatbot (documents se grounded answers) │
 *   │                                                          │
 *   │  Teeno mein ek common limitation thi:                    │
 *   │  LLM sirf TEXT generate kar sakta tha                    │
 *   │  → Koi action nahi le sakta tha          ✗✗             │
 *   │  → Real world se interact nahi kar sakta ✗✗             │
 *   │  → Calculator, API, DB — kuch nahi       ✗✗             │
 *   └──────────────────────────────────────────────────────────┘
 *
 * Phase-IV mein kya add hoga:
 *   → LLM ko "Tools" denge — jaise ek insaan ko calculator dena
 *   → LLM khud decide karega: "Mujhe tool use karna chahiye ya nahi?"
 *   → Tool call karega → result milega → answer mein use karega
 *   → Yahi "Agentic" behavior hai — autonomous decision + action
 *
 * Note: Is phase mein Memory nahi hai (Phase-V mein aayega).
 *       Pure agentic tool use pe focus.
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 1 — FIRST PRINCIPLES: AGENT KYA HOTA HAI?
 * ══════════════════════════════════════════════════════════════
 *
 *
 * CHATBOT vs AGENT — FUNDAMENTAL DIFFERENCE:
 * ────────────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  CHATBOT (Phase I-III):                                  │
 *   │  Input → Think → Output (text)                           │
 *   │                                                          │
 *   │  Sirf bolta hai — kuch karta nahi                        │
 *   │                                                          │
 *   │  User: "AAPL ka current price kya hai?"                  │
 *   │  Chatbot: "Mujhe real-time data nahi pata" ← useless  ✗✗│
 *   └──────────────────────────────────────────────────────────┘
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  AGENT (Phase IV):                                       │
 *   │  Input → Think → Act → Observe → Think → Output (text)   │
 *   │                                                          │
 *   │  Sochta hai AUR karta bhi hai                            │
 *   │                                                          │
 *   │  User: "AAPL ka current price kya hai?"                  │
 *   │  Agent: [stock_price tool call karta hai]                │
 *   │         [Result: $189.43]                                │
 *   │         "AAPL abhi $189.43 pe trade ho raha hai" ✓✓✓    │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * AGENT = LLM + TOOLS + LOOP:
 * ─────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  LLM    = Brain (reasoning, decision making)             │
 *   │  Tools  = Hands (actions, real world interaction)        │
 *   │  Loop   = Observe result → think again → act again       │
 *   │                                                          │
 *   │  Ek baar act karke ruk nahi jaata —                      │
 *   │  Result observe karta hai → phir decide karta hai        │
 *   │  "Kya aur kuch karna chahiye?"                           │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * REAL WORLD ANALOGY:
 * ────────────────────
 *
 *   Sochlo ek nayi city mein aaye ho.
 *   Tum (LLM) sochte ho: "Mujhe hotel dhundhna hai"
 *
 *   Chatbot (without tools):
 *   "Generally, hotels booking.com pe milte hain" ← generic
 *
 *   Agent (with tools):
 *   1. search_hotels("budget hotel, New York, tonight") → call
 *   2. Result: 3 hotels mile
 *   3. get_price(hotel_id_1) → $89/night
 *   4. check_reviews(hotel_id_1) → 4.2 stars
 *   5. "Grand Inn available hai $89/night pe, 4.2 stars — book karoon?" ✓
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 2 — TOOL USE KA CORE CONCEPT (FUNCTION CALLING)
 * ══════════════════════════════════════════════════════════════
 *
 *
 * TOOL KYA HOTA HAI?
 * ───────────────────
 *
 *   Tool = ek function jo LLM call kar sakta hai.
 *
 *   Tum LLM ko batate ho:
 *   "Yeh yeh functions available hain — agar zaroorat ho toh use karo"
 *
 *   LLM khud decide karta hai:
 *   "Is question ke liye mujhe tool chahiye ya nahi?"
 *   "Kaun sa tool chahiye?"
 *   "Kya arguments doon?"
 *
 *
 * TOOL DEFINITION KYA HOTI HAI?
 * ───────────────────────────────
 *
 *   Har tool ke liye hum LLM ko 3 cheezein batate hain:
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  1. NAME        → tool ka naam (LLM is naam se call kare)│
 *   │  2. DESCRIPTION → kab use karna hai (LLM yahi padh ke   │
 *   │                   decide karta hai)                      │
 *   │  3. PARAMETERS  → kya input chahiye (type, required)     │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Example tool definition:
 *
 *   {
 *     name: "get_stock_price",
 *     description: "Get the current stock price for a given ticker symbol.
 *                   Use this when user asks about stock prices or market data.",
 *     input_schema: {
 *       type: "object",
 *       properties: {
 *         ticker: {
 *           type: "string",
 *           description: "Stock ticker symbol e.g. AAPL, GOOGL, TSLA"
 *         }
 *       },
 *       required: ["ticker"]
 *     }
 *   }
 *
 *
 * IMPORTANT — LLM ACTUAL CODE NAHI CHALATA:
 * ───────────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  LLM sirf yeh batata hai:                                │
 *   │  "Main chahta hoon get_stock_price({ticker: 'AAPL'})"    │
 *   │                                                          │
 *   │  TUMHARA CODE actual function run karta hai!             │
 *   │                                                          │
 *   │  LLM → "get_stock_price call karo" → Tumhara backend     │
 *   │  → actual API call → result → wapas LLM ko do            │
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Yeh security ke liye bhi achha hai:
 *   LLM ko direct system access nahi — tumhara code middleman hai
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 3 — AGENTIC LOOP (THE CORE PATTERN)
 * ══════════════════════════════════════════════════════════════
 *
 *
 * AGENTIC LOOP KYA HOTA HAI?
 * ────────────────────────────
 *
 *   Ek simple chatbot mein: User → LLM → Answer. Done.
 *   Ek agent mein: yeh ek LOOP hai jab tak kaam khatam na ho.
 *
 *
 * LOOP KA FLOW:
 * ─────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │                                                          │
 *   │  [User Query]                                            │
 *   │       │                                                  │
 *   │       ▼                                                  │
 *   │  ┌─────────────────────────────────┐                     │
 *   │  │         LLM THINKS              │                     │
 *   │  │  "Mujhe kya karna chahiye?"     │                     │
 *   │  └──────────────┬──────────────────┘                     │
 *   │                 │                                        │
 *   │        ┌────────┴────────┐                               │
 *   │        │                 │                               │
 *   │   [Tool Call?]     [Final Answer?]                       │
 *   │        │                 │                               │
 *   │        ▼                 ▼                               │
 *   │  [Execute Tool]    [Return to User] ← LOOP END           │
 *   │        │                                                 │
 *   │        ▼                                                 │
 *   │  [Tool Result]                                           │
 *   │        │                                                 │
 *   │        ▼                                                 │
 *   │  [LLM gets result → thinks again] ← LOOP CONTINUES       │
 *   │        │                                                 │
 *   │  (back to top)                                           │
 *   │                                                          │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * MULTI-STEP EXAMPLE:
 * ────────────────────
 *
 *   User: "AAPL aur GOOGL mein se kaun sa better investment hai?"
 *
 *   Loop Iteration 1:
 *   LLM thinks → "Mujhe dono ke prices chahiye"
 *   Tool call → get_stock_price("AAPL") → $189.43
 *
 *   Loop Iteration 2:
 *   LLM thinks → "AAPL mila, ab GOOGL chahiye"
 *   Tool call → get_stock_price("GOOGL") → $141.20
 *
 *   Loop Iteration 3:
 *   LLM thinks → "Dono prices hain, ab P/E ratio chahiye"
 *   Tool call → get_fundamentals("AAPL") → {pe: 28.5, eps: 6.43}
 *
 *   Loop Iteration 4:
 *   LLM thinks → "Ab answer de sakta hoon"
 *   Final answer → "AAPL $189 pe trade ho raha hai, P/E 28.5..."
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Ek sawaal → 3 tool calls → 1 comprehensive answer       │
 *   │  User ne kuch nahi kiya — agent ne khud plan banaya ✓   │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * STOP CONDITIONS (Loop kab khatam hoga?):
 * ──────────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  1. LLM ne final text answer diya (tool call nahi)       │
 *   │  2. Max iterations reach ho gayi (infinite loop prevent) │
 *   │  3. Tool execution mein error aaya → graceful stop       │
 *   │  4. User ne cancel kiya                                  │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 4 — ANTHROPIC TOOL USE API (ACTUAL FORMAT)
 * ══════════════════════════════════════════════════════════════
 *
 *
 * ANTHROPIC KA TOOL USE PROTOCOL:
 * ─────────────────────────────────
 *
 *   Step 1: Tools define karo → LLM ko bhejo (tools array mein)
 *   Step 2: LLM response mein "tool_use" block aata hai
 *   Step 3: Tum actual function run karo
 *   Step 4: Result "tool_result" ke roop mein wapas bhejo
 *   Step 5: LLM final answer deta hai
 *
 *
 * LLM RESPONSE TYPES:
 * ────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Type 1: stop_reason = "end_turn"                        │
 *   │  → LLM ne final answer diya — loop khatam                │
 *   │                                                          │
 *   │  Type 2: stop_reason = "tool_use"                        │
 *   │  → LLM tool call karna chahta hai — loop continue        │
 *   └──────────────────────────────────────────────────────────┘
 *
 *
 * MESSAGES ARRAY KA EVOLUTION (LOOP MEIN):
 * ──────────────────────────────────────────
 *
 *   Iteration 1 — LLM ko bheja:
 *   messages = [
 *     { role: "user", content: "AAPL ka price?" }
 *   ]
 *
 *   LLM response (tool call):
 *   { role: "assistant", content: [
 *     { type: "tool_use", id: "tool_1", name: "get_stock_price",
 *       input: { ticker: "AAPL" } }
 *   ]}
 *
 *   Iteration 2 — Tool result ke saath LLM ko bheja:
 *   messages = [
 *     { role: "user",      content: "AAPL ka price?" },
 *     { role: "assistant", content: [tool_use block] },  ← LLM ka response
 *     { role: "user",      content: [               ← tool result
 *       { type: "tool_result", tool_use_id: "tool_1",
 *         content: "$189.43" }
 *     ]}
 *   ]
 *
 *   LLM final response:
 *   { role: "assistant", content: [
 *     { type: "text", text: "AAPL abhi $189.43 pe trade ho raha hai." }
 *   ]}
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  IMPORTANT: Tool result bhi "user" role mein jaata hai   │
 *   │  Yeh Anthropic ka protocol hai — confusing lagta hai     │
 *   │  lekin yahi sahi format hai                              │
 *   └──────────────────────────────────────────────────────────┘
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 5 — TOOLS KE TYPES (PRACTICAL EXAMPLES)
 * ══════════════════════════════════════════════════════════════
 *
 *
 *   ┌──────────────────────┬──────────────────────────────────┐
 *   │ Tool Type            │ Examples                         │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Data Fetching        │ Stock prices, Weather, News,     │
 *   │                      │ Currency rates, Sports scores    │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Computation          │ Calculator, Unit converter,      │
 *   │                      │ Date/time calculations           │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Database Operations  │ Search DB, Insert record,        │
 *   │                      │ Update user data                 │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ External APIs        │ Send email, Create calendar      │
 *   │                      │ event, Post to Slack             │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ File Operations      │ Read file, Write file,           │
 *   │                      │ Search in codebase               │
 *   ├──────────────────────┼──────────────────────────────────┤
 *   │ Web Search           │ Search Google, Fetch webpage,    │
 *   │                      │ Scrape data                      │
 *   └──────────────────────┴──────────────────────────────────┘
 *
 *
 * HUM IS PHASE MEIN YEH TOOLS BANAYENGE:
 * ────────────────────────────────────────
 *
 *   1. calculator     → Math expressions solve karo
 *   2. get_weather    → Kisi city ka weather fetch karo
 *   3. search_web     → Web search karo
 *   4. get_time       → Current date/time do
 *
 *   Yeh 4 tools ek strong foundation hain — concept clear karte hain.
 *   Production mein 20-30 tools ho sakte hain.
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 6 — COMPLETE SYSTEM ARCHITECTURE
 * ══════════════════════════════════════════════════════════════
 *
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │                      CLIENT                              │
 *   │  - Chat UI (Phase-II/III jaisi)                          │
 *   │  - Stream listener (SSE)                                 │
 *   │  - Tool execution indicator (NEW) "🔧 Searching web..."  │
 *   └──────────────────────┬───────────────────────────────────┘
 *                          │ POST /chat
 *                          ▼
 *   ┌──────────────────────────────────────────────────────────┐
 *   │                    BACKEND                               │
 *   │                                                          │
 *   │  1. History prepare karo                                 │
 *   │  2. Tools define karo                                    │
 *   │  3. AGENTIC LOOP shuru karo:                             │
 *   │     a. LLM call karo (tools ke saath)                    │
 *   │     b. Tool call aaya? → Execute karo                    │
 *   │     c. Result ko messages mein add karo                  │
 *   │     d. Loop back (a pe)                                  │
 *   │     e. Final answer aaya? → Stream karo → Client         │
 *   │                                                          │
 *   └──────────────────────┬───────────────────────────────────┘
 *                          │
 *            ┌─────────────┼─────────────┐
 *            ▼             ▼             ▼
 *       [Calculator]  [Weather API]  [Web Search]
 *       (local fn)    (external)     (external)
 *
 *
 * KEY DESIGN DECISION — STREAMING + TOOLS:
 * ──────────────────────────────────────────
 *
 *   ┌──────────────────────────────────────────────────────────┐
 *   │  Tool calls ke waqt STREAMING nahi hoti                  │
 *   │  (LLM tool call decide karta hai — text nahi generate)   │
 *   │                                                          │
 *   │  Streaming tab hoti hai jab LLM FINAL TEXT deta hai      │
 *   │                                                          │
 *   │  Flow:                                                   │
 *   │  Tool call 1 → execute → [non-streaming, instant]        │
 *   │  Tool call 2 → execute → [non-streaming, instant]        │
 *   │  Final answer → STREAMING → client word by word      ✓✓✓│
 *   └──────────────────────────────────────────────────────────┘
 *
 *   Client ko SSE events se batate hain kya ho raha hai:
 *   { type: "tool_start", tool: "get_weather", input: {...} }
 *   { type: "tool_done",  tool: "get_weather", result: "25°C" }
 *   { type: "token",      token: "Mumbai" }
 *   { type: "token",      token: " mein" }
 *   { type: "done" }
 */


/**
 * ══════════════════════════════════════════════════════════════
 * SECTION 7 — TOOLS DEFINITION + IMPLEMENTATION PSEUDOCODE
 * ══════════════════════════════════════════════════════════════
 */


/**
 * STEP 1: TOOLS DEFINE KARO (LLM ke liye)
 * > Yeh array LLM ko bhejoge — woh in descriptions se samjhega kab kya use karna hai
*/

const AVAILABLE_TOOLS = [

    {
        name: "calculator",
        description: `Evaluate mathematical expressions and perform calculations.
                  Use this for any math: arithmetic, percentages, unit conversions.
                  Do NOT use for simple mental math — only complex calculations.
                  Examples: "2^10", "15% of 3500", "sqrt(144)"`,
        input_schema: {
            type: "object",
            properties: {
                expression: {
                    type: "string",
                    description: "The mathematical expression to evaluate. E.g. '2 + 2', '15% of 500'"
                }
            },
            required: ["expression"]
        }
    },

    {
        name: "get_weather",
        description: `Get current weather information for a specific city.
                  Use this when user asks about weather, temperature, or climate conditions.
                  Returns temperature, humidity, and conditions.`,
        input_schema: {
            type: "object",
            properties: {
                city: {
                    type: "string",
                    description: "City name. E.g. 'Mumbai', 'New York', 'London'"
                },
                unit: {
                    type: "string",
                    enum: ["celsius", "fahrenheit"],
                    description: "Temperature unit. Default: celsius"
                }
            },
            required: ["city"]
        }
    },

    {
        name: "search_web",
        description: `Search the web for current, real-time information.
                  Use this for: recent news, current events, live data, anything
                  that might have changed after training cutoff.
                  Do NOT use for general knowledge questions you already know.`,
        input_schema: {
            type: "object",
            properties: {
                query: {
                    type: "string",
                    description: "Search query. Be specific and concise."
                },
                max_results: {
                    type: "number",
                    description: "Number of results to return. Default: 3, Max: 5"
                }
            },
            required: ["query"]
        }
    },

    {
        name: "get_current_time",
        description: `Get the current date and time for a specific timezone.
                  Use this when user asks about current time, date, day of week.`,
        input_schema: {
            type: "object",
            properties: {
                timezone: {
                    type: "string",
                    description: "Timezone string. E.g. 'Asia/Kolkata', 'America/New_York', 'UTC'"
                }
            },
            required: ["timezone"]
        }
    }

]

/**
 * STEP 2: ACTUAL TOOL IMPLEMENTATIONS
 * > Yeh real functions hain — LLM inhe nahi chalata, TUMHARA backend chalata hai
*/

async function executeTool(toolName, toolInput) {

    switch (toolName) {

        case "calculator": {
            /* Safe math evaluation — eval() use mat karo (security risk!) */
            /* mathjs library use karo                                     */
            const math = require('mathjs')
            try {
                const result = math.evaluate(toolInput.expression)
                return { result: result.toString(), expression: toolInput.expression }
            } catch (err) {
                return { error: "Invalid expression", expression: toolInput.expression }
            }
        }

        case "get_weather": {
            /* OpenWeatherMap API (free tier available) */
            const unit = toolInput.unit || 'celsius'
            const units = unit === 'celsius' ? 'metric' : 'imperial'

            const response = await fetch(
                `https://api.openweathermap.org/data/2.5/weather` +
                `?q=${toolInput.city}&units=${units}&appid=${process.env.WEATHER_API_KEY}`
            )
            const data = await response.json()

            if (data.cod !== 200) {
                return { error: `City not found: ${toolInput.city}` }
            }

            return {
                city: data.name,
                temperature: `${data.main.temp}°${unit === 'celsius' ? 'C' : 'F'}`,
                feels_like: `${data.main.feels_like}°${unit === 'celsius' ? 'C' : 'F'}`,
                humidity: `${data.main.humidity}%`,
                conditions: data.weather[0].description,
                wind_speed: `${data.wind.speed} m/s`
            }
        }

        case "search_web": {
            /* Serper API ya Brave Search API use karo */
            const maxResults = Math.min(toolInput.max_results || 3, 5)

            const response = await fetch('https://google.serper.dev/search', {
                method: 'POST',
                headers: {
                    'X-API-KEY': process.env.SERPER_API_KEY,
                    'Content-Type': 'application/json'
                },
                body: JSON.stringify({ q: toolInput.query, num: maxResults })
            })
            const data = await response.json()

            /* Sirf relevant fields return karo — poora response bahut bada hoga */
            const results = (data.organic || []).slice(0, maxResults).map(r => ({
                title: r.title,
                snippet: r.snippet,
                link: r.link
            }))

            return { query: toolInput.query, results }
        }

        case "get_current_time": {
            const now = new Date()
            const formatter = new Intl.DateTimeFormat('en-IN', {
                timeZone: toolInput.timezone,
                dateStyle: 'full',
                timeStyle: 'long'
            })
            return {
                timezone: toolInput.timezone,
                datetime: formatter.format(now),
                iso: now.toISOString()
            }
        }

        default:
            return { error: `Unknown tool: ${toolName}` }
    }
}


/**
 * AGENTIC ROLES — LEARNING SEQUENCE
 * ==========================================================
 *
 * Goal: Simple se start karke gradually complex systems tak jaana
 *
 *
 * ============================================================
 * LEVEL 0 — BASE (NO AGENT)
 * ============================================================
 * 
 * LLM ONLY: Input → Output
 *  - No decision making
 *  - No tools
 *  - No actions
 *
 *
 * ============================================================
 * LEVEL 1 — TOOL CALLING AGENT (ENTRY POINT)
 * ============================================================
 *
 *   WHY FIRST?
 *   → Sabse simple form of "action"
 *
 *
 *   CAPABILITY:
 *   - Decide:
 *       → Answer directly
 *       → Ya tool call karo
 *
 *
 *   FLOW: Query → LLM → Tool Call → Result → Final Answer
 *
 *
 *   LEARNING FOCUS:
 *   ✔ Tool schema
 *   ✔ Structured outputs
 *   ✔ API integration
 *
 *
 * ============================================================
 * LEVEL 2 — ReAct AGENT (THINK + ACT LOOP)
 * ============================================================
 *
 *   WHY NEXT?
 *   → Ab multi-step reasoning introduce hota hai
 *
 *
 *   CAPABILITY:
 *   - Thought → Action → Observation loop
 *   - Step-by-step decision making
 *
 *
 *   FLOW: Thought → Action → Observation → Repeat
 *
 *
 *   LEARNING FOCUS:
 *   ✔ Prompt structure
 *   ✔ Loop control
 *   ✔ Reasoning traces
 *
 *
 * ============================================================
 * LEVEL 3 — PLANNER–EXECUTOR AGENT
 * ============================================================
 *
 *   WHY AFTER ReAct?
 *   → ReAct reactive hai
 *   → Yeh strategic planning introduce karta hai
 *
 *
 *   CAPABILITY:
 *   - Pehle full plan
 *   - Phir execution
 *
 *
 *   FLOW: Plan → Execute → Final Answer
 *
 *
 *   LEARNING FOCUS:
 *   ✔ Task decomposition
 *   ✔ Plan validation
 *   ✔ Sequential execution
 *
 *
 * ============================================================
 * LEVEL 4 — HYBRID AGENTS (REAL-WORLD SYSTEMS)
 * ============================================================
 *
 *   WHY IMPORTANT?
 *   → Real systems ek pattern use nahi karte
 *
 *
 *   EXAMPLES:
 *   - Planner + ReAct
 *   - Tool Calling + Memory + RAG
 *
 *
 *   LEARNING FOCUS:
 *   ✔ System composition
 *   ✔ Trade-offs
 *
 *
 * ============================================================
 * LEVEL 5 — MULTI-AGENT SYSTEM
 * ============================================================
 *
 *   WHY LAST?
 *   → Sabse complex + distributed system
 *
 *
 *   CAPABILITY:
 *   - Multiple agents
 *   - Role-based collaboration
 *
 *
 *   FLOW: Orchestrator → Agents → Collaboration → Output
 *
 *
 *   LEARNING FOCUS:
 *   ✔ Coordination
 *   ✔ Communication
 *   ✔ Parallelism
 *
 *
 * ============================================================
 * FINAL LEARNING PATH SUMMARY
 * ============================================================
 *
 *   0. LLM (No Agent)
 *   1. Tool Calling Agent        ← START HERE
 *   2. ReAct Agent
 *   3. Planner–Executor
 *   4. Hybrid Systems
 *   5. Multi-Agent Systems       ← ADVANCED
 *
 *
 * ============================================================
 * MENTAL MODEL (VERY IMPORTANT)
 * ============================================================
 * 1. Tool Calling: "Kaunsa tool use karu?"
 * 2. ReAct       : "Step by step kaise solve karu?"
 * 3. Planner     : "Pehle pura plan bana leta hu"
 * 4. Multi-Agent : "Team bana ke kaam karte hain"
 *
 *
 * ============================================================
 * FINAL INSIGHT
 * ============================================================
 *  Agent systems evolve like this:
 *  > Actions → Reasoning → Planning → Collaboration
 *  > Yehi actual industry progression hai 
 */
```
