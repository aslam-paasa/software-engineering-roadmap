```python
/**
 * Designing a Simple Todo App (1 User Case)
 * 
 * Let's start with the simplest case:
 *   - A Todo App used by only one user. 
 *   - Even for 1 user, a system needs proper structure.
 * 
 * 1. Application needs an Identity (Domain Name)
 * > When you build an application and deploy it, you give it a domain
 *   name, like: piyushgarg.online
 * > This domain acts as:
 *   - A human-friendly name
 *   - An entry point to your application
 * 
 * 2. User Access Through Browser
 * > Now imagine:
 *   - You send this domain to your friend
 *   - Your friend opens it in a browser
 * > Important Question:
 *   - Does the user know what the domain actually is?
 *   - Ans: No!
 *   - Because:
 *     - A domain is just a name
 *     - It does not directly tell where the server is
 * 
 * 3. Core Components of any System
 * > Every basic system has 2 main components:
 *   a. Client (User Side)
 *      - This is where the user interacts
 *      - Ex  : Browser, Mobile App
 *      - Role: Sends request to the server
 * 
 *   b. Server (Application Side)
 *      - This is where your application runs
 *      - Handles logic and data
 *      - Role:
 *        - Receives request
 *        - Processes it
 *        - Sends response back
 * 
 * > Important Concept: Request-Response Cycle
 *   - This communication b/w client and server works like this:
 *     a. User sends request
 *     b. Server processes it
 *     c. Server sends response
 *   - This is called Request -> Response Cycle
 * 
 * 4. Problem: How Does Client Reach Server?
 * > Now comes the core problem
 * > Your server exists somewhere on the internet and has an:
 *   - IP Address (Internet Address)
 *   - Ex: 10.34.2.1
 * 
 * > Why not use IP directly?
 *   - IP addresses are hard to remember
 *   - Humans are better with names than numbers
 *   - So, we need a better solution
 * 
 * 5. Domain Name System (DNS)
 *    > To solve this problem, we created: Domain Name System (DNS)
 *    > DNS is a system that converts domain names to IP Addresses.
 *    > Think of it like a Phonebook of the Internet.
 * 
 * 
 * 6. How Browser Finds IP froma Domain?
 *    > When you type a domain like: piyushgarg.online
 *    > Your browser does not directly know the IP address
 *    > It follows a multi-step process called DNS Resolution:
 * 
 *      a. Check Browser Cache
 *         - The browser first checks: "Have I already visited this domain before?"
 *           - If yes -> IP is already stored -> use it directly
 *           - If no  -> go to the next step
 *         - This is called DNS Cache
 * 
 *      b. Check Operating System (OS) Cache
 *         - If browser doesn't know:
 *         - It asks the OS (Windows/Linux/Mac)
 *         - OS also keeps a DNS cache
 *           - If found -> returns IP
 *           - If not   -> go further
 * 
 *      c. Check Hosts File (Local Mapping)
 *         - Before going to internet, system checks: Host file
 *         - This is a local file where you can manually map: 127.0.0.1-to-localhost
 *         - If domain is found here:
 *           - It uses this mapping directly
 *           - No internet request needed
 * 
 *      d. Ask DNS Resolver (ISP DNS Server)
 *         - If still not found: Request goes to DNS Resolver
 *           - Usually provided by your ISP (Jio, Airtel, etc.)
 *           - Or Custom (Google DNS: 8.8.8.8)
 *         - Resolver's job: Find the IP for that domain
 * 
 * 
 * Now the real internet journey begins:
 * > If resolver doesn't know the answer, it performs a recursive query:
 * 
 * 7. Root DNS Server
 *    - Resolver asks: "Where can I find .online domain info?"
 *    - Root Server replies: "Go to TLD server for .online"
 * 
 * 8. TLD Server (Top-Level Domain)
 *    - Now resolver asks .online server: "Where is piyushgarg.online?"
 *    - TLD responds: "Ask this authoritative server"
 * 
 * 9. Authoratitive DNS Server
 *    - This is the actual source of truth
 *    - It replies: "piyushgarg.online" -> 10.34.2.1
 * 
 * 10. Response Back to Browser
 *     - Resolver sends IP to OS
 *     - OS sends it to the browser
 *     - Browser finally gets IP
 * 
 * 11. Actual Request to Server
 *     - Now browser can finally send HTTP/HTTPS request to: 10.34.2.1
 *     - And server responds
*/
```

```python
/**
 * Design a Todo Application System for 1 User.
 * > Humne ek system design kr k application bana k deploy kr diya, ab
 *   humein apne application pe 1 User k liye sabse pehle kya chahiye?
 *   - Application Domain (piyushgarg.online)
 *   - Iss domain ko maine apne friend ko bheja, ab wo apne browser k
 *     andar open karega.
 *   - Kya mere friend ko koi idea hai ki "piyushgarg.online" kya hai?
 *     No!
 * > Ek system k andar 2 component hote hai: Client/User & Server/App
 *   - Inn 2 component k alawa ek simple system m koi aur component nhi
 *     hota hai.
 *     a. Client: User sends request to Application
 *     b. Server: Application responds back to the User
 * > And this is whole request-response cycle, jisme ek user ko ek rasta
 *   chahiye jisse wo server tak reach kr pae, aur server k pass reach
 *   krne ka sirf ek tarika hota hai i.e., iss server ka IP Address
 *   (Internet Address) - 10.34.2.1.
 * > But remembering IP Address is very hard because human minds number
 *   ko bahut acche se yaad nhi rkh sakta isliye humne ek aisa system
 *   banaya jiske through hm server ko ek user friendly naam de pae like
 *   alphabetic names.
 * > So, we need some kind of way jaha agar hum apne browser mai iss
 *   alphabetic names ko enter kare to hum server pe pahuch jae:
 *   Browser > piyushgarg.online > 10.34.2.1
 * > Iske liye humein ek phonebook directory jaise ek server banate hai,
 *   jisme ye phonebook mere domain ko map kr diya ek IP Address k upar
 *   and iss telephone directory ko humne internet k upar available kr
 *   diya, jisse meri first problem solve ho gyi i..e agar koi user koi
 *   domain enter krta hai to unhone browsers k andar ek special code
 *   likha jisme hum koi v domain enter krte hai to we will do a step
 *   known as DNS Resolution, means phonebook m jaa k dhundnge.
 * > DNS Resolution m humari request pehle DNS Server m jaati hai aur puchti hai
 *   "piyushgarg.com" ka IP Address batao, aur DNS Server humein IP Address
 *   return kr deta hai.
 * > Ab jab humein IP Address mil chuka hai i.e., 10.43.2.1, Can I now go to my server
 *   because I have the IP Address.
 * > Ab humari server k upar ek request aa chuki hai, server isse process karega means
 *   humara backend code execute hoga aur user ko output return kr k dega. This is
 *   called request-response cycle.
 * 
 * > Mere friend ne "piyushgarg.com" ka link utha k kisi whatsapp group m daal diya ki
 *   dekho iss bande ne ye application banaya hai. Isse ye hoga ki jaha humare server
 *   k paas ek user tha wahi 4 user ho jaenge.
 * > Now chaaro user mere server k upar request aai, to server ne one by one sabki request
 *   prcoess ki aur chaaro ko unka data mil gya.
 * 
 * > Initially hum todo application bana rhe the, that means humara server wo logo k
 *   todos ko manage krta hai. 
 * 
 *   - User jo v todoItem banaega usse hum todos list k andar push kr denge.
 *     const todos = [] => List of todos
 *     function createTodo(todoItem) { 
 *        todos.push(todoItem)
 *     }
 * 
 *   - User jab v getAllTodos() ko call karega to ye user k saare todos return kr dega
 *     function getAllTodos() {
 *        return todos;
 *     } 
 * 
 *   - And humne ek system bana diya which is fully functional and the system is working
 *     fine for one user.
 * 
 * > Ab iss user ne 10 todos bana diye, it means array k andar 10 elements hai. Suppose
 *   karo ki 1todo item 1mb storage leta hai, to humare server k andar hum kitni RAM
 *   use kr rhe hai? 10 todos for 10 mb.
 * > What if humare paas 10,000 users ho gye aur har ek ne 10 todos banae, Ab humane
 *   kitna storage use kr liya? 1Lakh MB use kr diya, jisse humari server crash kr gyi
 *   because our server is out of memory.
 * > Hum 1Million users k liye design kr rhe the lekin ye to 10k users m hi crash kr gya.
 *   Even 1000 users m v crash kr jaega i.e., 10,000 MB.
 * 
 * > Jaha humare paas 10GB ki RAM thi wahi hum 100GB ki RAM laga denge because humare
 *   users ko badhnge lekin saath hi saath existing users k todos v to badhnge, to
 *   kabhi na kabhi ye fir down jaega, fir hum 1000GB ko 2000GB kr denge, etc. Hum
 *   bas per month cost increase kr rhe hai by increasing the capacity of the server
 *   because humein ek bottleneck dikhai de rha tha.
 * > Agar system keh rhe hai ki mai crash ho rha hu because mere paas RAM nahi hai to
 *   what is the very first thing you should do?
 *   - RAM add kar do.
 * 
 * Memort k badle TradeOff:
 * a. Cost: More hardware, more money
 * b. Availability: 
 *    Ek chalti machine m RAM nhi badli jaa skti, that means jab v mera server down jaane
 *    wala hoga suppose 80% full tb hum RAM ko upgrade kr denge. That means system thodi
 *    der k liye down jaega.
 * 
 * > Vertical Scaling or Scaling up, increases system capacity by adding more resources
 *   like CPU, RAM, or storage - to a single server.
 * 
 * 
 * Problem: 
 * > Humare paas sirf ek system hai, aur light chali gyi, bug in codebase, ya koi other
 *   issue aaya to mera ye system down jisse mera poora application down ho jaega, 
 *   because of Single Points of Failure.
 * > Ab chahe hum kitna v resources add kare hum apni availability risk pe rakhe hai
 *   because of Single Point of Failure.
 * 
 * Solution: 
 * > Company ki sabse pehli problem availability ki aati hai aur uska one of the reason
 *   is Single Point of Failure.
 * > Agar humein humare system ko Single Point of Failure se bachana hai to humein ek
 *   backup server rakhna padega.
 * > Humne ek domain liya tha i.e. 'piyushgarg.online', ab hum ek aur domain le lenge
 *   i.e., 'backup.piyushgarg.online', aur dono domain to dono server pe map kr denge
 * 
 *      Main Server           Backup Server
 *     +----------+           +----------+
 *     | Server 1 |           | Server 2 |
 *     +----------+           +----------+
 *   piyushgarg.online    backup.piyushgarg.online
 * 
 * > Hum apne users ko bolnge ki piyushgarg.online wale website pe jao, hopefully wo hmara
 *   kaam kr rha hoga because usme 4GB ki RAM hai. Aur humne apne website m likh rakha
 *   tha ki agar ye kaam nhi kar rha hai to visit: backup.piyushgarg.online, jisse users
 *   idhar aa jaenge, but data to Main Server pe hi reh gya because saare todos to wahi
 *   stored tha, aur iss server m users ki todos exist hi nhi kr rha. 
 * > Humara availability ki problem fix ho gyi lekin data inconsistency issue aa gyi.
 * 
 * CAP Theorem:
 * > C - Consistency
 * > A - Availability
 * > P - Partition-Tolerance
 * 
 * > CAP Theorem kehta hai ki hum at a time sirf 2 chije achieve kr skte hai.
 * > Agar humne availability and partition-tolerance le ki to humari consistency chali gyi.
 * > Ab consistency laane k liye humein backup server hatani padegi, kyuki agar database
 *   hi nhi chal rha to hm todos nhi bana sakte aur retrieve kr skte, isliye data
 *   consistency k liye humein availability trade off krna padega.
 * 
 * Problem: Stateful Servers
 * > Humara data wo http server m hi stored tha jisse hum stateful server kehte hai.
 * > Isko stateful isliye kehte hai because there is a state, means this is something
 *   stored in the memory. And stateful servers ko scale krna is very difficult because
 *   Stateful Servers ko agar hum vertical scale kr rhe hai to downtime, single point of
 *   failure aata hai, and jab wo down jaaega to stateful k state clear ho jaegi, and 
 *   even if we do some backup, stateful servers consumes high RAM.
 * > For example, WebSocket humare servers k saath one-on-one connection maintain kr k
 *   rakhte hai, means Server k upar kuch na kuch ek state maintain hoti hai jisme
 *   stored hai ki kon sa user kon se socket se connected, that means WebSocket is a
 *   stateful Server, and we know scaling WebSocket is hard.
 * > Suppose we're dealing with every growing number of WebSocket connections as more
 *   people join and use the app. If we have just one WebSocket Server, there's a finite
 *   amount of resources you can add to it, which limits the number of concurrent 
 *   connections your system can handle. That;s why scaling a stateful server is hard.
 * 
 * Solution: Stateless Server
 * > Humein servers ko stateful banana band krna padega, and apne servers ko stateless 
 *   banana padega like HTTP Server. Why? Because stateless server m memory nahi hai
 *   and it's easy to scale, and agar stateless hai to usme multiple users rakh sakte
 *   hai aur meri consistency kharab nhi hogi.
 * > Pehle hm aur servers add nhi kar paa rhe the because there was some kind of state,
 *   means agar ek server k paas koi data hai aur dusre k paas nhi hai to consistency
 *   kharab ho gyi, because of that issue humein ek server rakhna padta tha. But agar
 *   mere servers stateless hai to user kisi v server k paas request karega to same
 *   response paega. 
 * > Ab jab saare servers same response de rhe hai to hm easily multiple server create
 *   kr skte hai aur humari consistency v kharab nhi hogi and single point of failure
 *   ka issue v resolve ho gya. And using this we have achieved both Availability and
 *   Consistency (Servers m data hi nhi hai so no inconsistency). 
 * > We also resolved the cost problem, how? Agar humare database m 10,000 users hai
 *   and currently active to 10k se kam honge, fir v humein 10,000 users k hisaab se
 *   apne server ki resource planning krni hoti hai because saare users ek saath active
 *   ho gye to kya krnge. Isliye chahe kitne v users active ho jae humein 10K users ki
 *   capacity planning krni hoti hai.
 * > It's very rare ki saare active ho jae but fir v iss issue se bachne k liye we do
 *   AutoScaling. 
 * > Maan lo humara ek server 10 users ko handle kr skta hai, to agar ek user hai to 
 *   ekhi server rakho, but jaise hi 7-8 users ho jae to ek new server launch kr denge,
 *   aur baaki users ko idhar redirect kr denge jisse downntime v nhi aaya aur scale
 *   tab kiya jab jarurat thi. 
 * > Agar 10k users m se sirf 20 active hai to hum sirf 2 servers ka cost de rhe hai, 
 *   aur users kam ho rhe hai to automatically hm servers kam v kar rhe hai jisse mera
 *   cost v bach rha hai. 
 * > So, we can do a much better resource planning with stateless servers, and this
 *   scaling is called Horizontal Scaling.
 * 
 * 
 * Problem: 
 * > Agar mere paas 5 different servers hai to 5 different addresses honge, lekin domain
 *   sirf ek hai i.e., piyushgarg.online. 
 * 
 *   +----------+ +----------+ +----------+ +----------+ +----------+
 *   | Server 1 | | Server 2 | | Server 3 | | Server 4 | | Server 5 |
 *   +----------+ +----------+ +----------+ +----------+ +----------+
 *     10.2.3.1     10.2.3.2     10.2.3.3     10.2.3.4     10.2.3.5  
 * 
 * Solution:
 * > DNS bolta hai: One domain, One Mapping. To ab mere paas 2 options hai:
 *   1. Buy multiple domains for multiple servers:
 *      - a.piyushgarg.online
 *      - b.piyushgarg.online
 *      - c.piyushgarg.online
 *      - d.piyushgarg.online
 *      - e.piyushgarg.online
 *      Ab inn domains ko ek ek kar k ek ek server se map kar denge, aur user ko keh
 *      denge ki kisi v domain ko access kr skte hai, koi na koi to chal hi rha hoga,
 *      lekin Human Psychology kehta hai hm pehle wale ko pehle try krnge and it's 
 *      very frustrating because jaise jaise users badhta jaenge hmein aur servers up
 *      kr k new domain buy kr k map krna padega. It's not a good solution.
 * 
 *   2. Mere servers aur users k bich m hm ek server baitha denge jiske ekhi kaam hoga
 *      ki user request ko kisi specific server k paas forward kar k load ko balance
 *      karna. And this server is called Load Balancer.
 * 
 *      +----------+ +----------+ +----------+ +----------+ +----------+
 *      | Server 1 | | Server 2 | | Server 3 | | Server 4 | | Server 5 |
 *      +----------+ +----------+ +----------+ +----------+ +----------+
 *        10.2.3.1     10.2.3.2     10.2.3.3     10.2.3.4     10.2.3.5 
 *            |            |            |            |            |
 *            +------------+------------+------------+------------+
 *                                      |
 *                              +---------------+
 *                              | Load Balancer | (10.2.3.0)
 *                              +---------------+
 *                                      |
 *                              +---------------+
 *                              |     Users     | (piyush.garg.online)
 *                              +---------------+
 *      - Humare DNS ki proble ye thi ki wo sirf ek domain ko ek IP se map kr skta hai,
 *        isliye hum piyushgarg.online <--> 10.2.3.0 se map kr denge. 
 *      - Ab jo v users aaenge aur DNS ko query karenge to unka request Load Balancer ko
 *        milega, ab load balancer 2 chije krega:
 *        a. Pehla tarika ye hai humne load balancer ko bola ki humein kisi ka address
 *           de aur load balancer ne humein bola ki tu iss server se baat kar, that means
 *           load balancer ne user ko uss server ka address return kr diya. Now we have
 *           to make a communication with it. But there is an additional network hop in
 *           this approach, jaha load balancer ne pehle pta lagaya ki kon sa server hai
 *           fir wapas aaya fir user ko bataya, which is not safe because user ko direct
 *           server details provide krna is not secure and scaling m v problem aaegi.
 *        b. That's why load balancer users ki request ko internally humare behalf pe 
 *           server ko forward karega and server se response le k humein forward karega.
 *           So load balancer is basically a proxy.
 * 
 *      - Jab v hum VPN use krte hai to user request VPN pe jaata hai which acts as
 *        Proxy and wo mere request ko kisi server pe forward krta hai. This is called
 *        Forward Proxy.
 *      - But jab hum user request VPN ko bhejte hai aur wo user request ko multiple
 *        servers ko forward krta hai tb isse Reverse Proxy kehte hai.
 *      - So, load balancer is a reverse proxy server.
 * 
 * > Because user request is going to the server through load balancer, ek user kabhi
 *   v publically inn servers tak nhi aa paega. That's why inn servers ko hm kabhi v 
 *   public IP Address assign krte hi nhi hai, inn servers ka sirf private IP Address 
 *   hota hai, and hm load balancer ko same network m rakh dete hai jisse ye hota hai
 *   ki load balancer is the only machine jiske paas public IP hota hai aur ye public
 *   IP internally private IPs se communication karta hai ek network k andar but no
 *   one can access our server from outside the network, called VPC.
 * 
 *  +-------------------------------------------------------------------------+
 *  |                              +----------+                               |
 *  |                              | Database |===> Database Layer            |
 *  |                              +----------+                               |
 *  |                                    |                                    |
 *  |          +------------+------------+------------+------------+          | 
 *  |          |            |            |            |            |          | 
 *  |    +----------+ +----------+ +----------+ +----------+ +----------+     |
 *  |    | Server 1 | | Server 2 | | Server 3 | | Server 4 | | Server 5 |=====|====> App Layer
 *  |    +----------+ +----------+ +----------+ +----------+ +----------+     |
 *  |          |            |            |            |            |          | 
 *  |          +------------+------------+------------+------------+          | 
 *  |                                    |                                    |
 *  |                            +---------------+                            |
 *  |                            | Load Balancer | (10.2.3.0)                 |
 *  |                            +-------+-------+                            |
 *  +------------------------------------|------------------------------------+
 *                              +--------+------+
 *                              |     Users     | (piyush.garg.online)
 *                              +---------------+
 *   
 * Problem: Agar load balancer down chala gya to Single Point of Failure aa jaega.
 * 
 * Solution:
 * > Load Balancer khud se kuch v process ni kr rha hai, to load balancer k upar kuch
 *   CPU intensive task, RAM intensive task nhi hoga. Load Balancer ka kaam hai sirf
 *   request ko forward kr dena, so load balancer expensive nhi hoga and there are less
 *   changes are going down + inke andar bahut saari micro-machines hoti hai jinko 
 *   worker threads kehte hai. But still worst case m load balancer down jaa skta hai.
 * > But this is architecture we will use 99% times.
 * 
 * Note: Database Layers ko scale krne k alag techniques hai jo hm next class m pdhnge
*/


/**
 * Caching-101:
 * > Humare paas ek user hai jo kisi na kisi backend server ko request karta hoga, aur
 *   ye server usually ek database se baat karta hoga. This is the traditional structure.
 * > Interview v yahi se start hoti hai.
 * 
 *   +------+        +---------+          +----------+
 *   | User |<------>| Backend |<-------->| Database |
 *   +------+        +---------+          +----------+
 * 
 * > Agar bookmyshow pe friday 9bje ek movie lagi hai aur sab usi movie k baare m
 *   janna chahte hai, aur sbko wahi movie dekhni hai. Agar ticket lete hai to theek hai
 *   wrna sirf read krni hai to humari caching is best strategy. It means agar ek hi
 *   request baar baar aa rhi hai to cache se puch lo, agar cache m hi request mil jaati
 *   hai to wahi se wo data read kr lega aur ye user ko reply kr dega, humein database
 *   hit krne ki jarurt hi nahi padegi.
 * 
 *   +------+        +---------+          +----------+
 *   | User |<------>| Backend |<-------->| Database |<-----+
 *   +------+        +---------+          +----------+      |
 *                        |                                 |
 *                        |               +-------+         |
 *                        +-------------->| Cache |---------+
 *                                        +-------+
 * > Ab kyuki cache sab chij memory m rakhta hai isliye ye fast hota hai, aur isi liye
 *   hm cache ko in-memory v bolte hai. And iss cache ko bahut saare naam se jaana
 *   jaata hai like:
 *   - In-memory Cache
 *   - Redis 
 *   - Upstash
 * > Aur agar cache k paas data nhi hai to wo database k paas se data laa sakta hai,
 *   pehle khud k memory m hold kr k rkhega aur fir user ko v response m dega.
 * > Iss Cache k andar jo data hoga, kabhi na kabhi to usse expire krna pdega aur
 *   iss expiration of data ko kehte hai TTL - Time to Leave.
 * 
 * Issue: The Thundering Herd Problem
 * > Humne jo architecture banaya tha wo ye tha ki pehle cache k dekho agar waha data 
 *   nhi hai tab database m jao aur waha se data laa k cache m rkho aur user k response
 *   bhejo. 
 * > Humne jab cache lagaya to usko ek TTL diya i.e., 6:15, but uske just ek second
 *   baad ek event ho gya 6:16 pe i.e., IPL match, aur uss event se bahut request aayi 
 *   humare server pe. 
 * > 6:15 m cache clear ho gya to avi usme kuch hai hi nhi, aur saari request database
 *   k paas jaegi jisse database ka load spike ho jaega aur crash kr jaega kyuki sudden
 *   traffic spike m liye humara database ready hi nahi tha.
 * > To aise issues ko rokne k liye kon kon strategies use ki jaati hai?
 * 
 * Solution-1: Jitter on TTL
 * > Humare paas jitna saara cache rakha hua hai sabka TTL ek saath mat udao, sabka
 *   time different hona chahiye.
 * 
 * Solution-2: Probability Early Computation
 * > Agar humein pta hai ki koi event hone wala hai 6:15 pe, to khud cache clear kr do
 *   6:15 se pehle.
 * 
 * Solution-3: MUTEX
 * > Agar humara cache empty hai to mai sirf ek request ko allow karunga database 
 *   tak pahuchne tak, fir wo data mere cache pe fill hoga, uske baad baaki request
 *   allow hoga
 * > Basically, MUTEX means acquiring lock.
 * 
 * Solution-4: Stale While Revalidating (Most used)
 * > Stale data means puraana data
 * > Puraane cache data ko mat udao, new cache laa k rakho, user ko puraane cache data 
 *   se reply krte raho kuch time tak jisse database overwhelm nhi hoga.
 * > This is the most used strategy by apps like Hotstar.
 * 
 * Solution-5: Cache Warming
*/

/**
 * Note: Never start designing, start questions first
*/
```
