```python
/**
 * What is Rate Limiting?
 * - Rate Limiting is a technique used in system architecture to regulate how quickly
 *   a system processes or serves incoming requests or actions. 
 * - It limits the quantity or frequency of client requests to prevent overload, 
 *   maintain stability, and ensure fair resource distribution.
 *   - Reduces the risk of resource abuse and denial-of-service (DoS) attacks, 
 *     improving performance, reliability, and security.
 *   - Used in web servers, APIs, network traffic management, and database access, 
 *     with examples like AWS API Gateway throttling, AWS WAF, Azure API Management 
 *     rate limiting and quota policies, and GCP API Gateway quota policies.
 *   - Ensures fair resource usage among multiple users and improves overall system 
 *     reliability and performance.
 * 
 * Use Cases:
 * - Rate Limiting is used in various systems to control request traffic, prevent abuse,
 *   and ensure fair and stable use of resources.
 *   1. API Rate Limiting: APIs commonly employ rate limitation to control the volume
 *      of client request, ensure fair access to resources, and prevent abuse.
 *   2. Web Server Rate Limiting: Web Servers employ rate limitation as a defense
 *      against denial-of-service attacks and to prevent server overload.
 *   3. Database Rate Limitation: To keep the database server from experiencing undue
 *      strain and to preserve database performance, rate limitation is applied to
 *      database queries. For instance, to avoid resource exhaustion and guarantee
 *      seamless functioning, an e-commerce website can restrict the quantity of
 *      database queries per user.
 *   4. Login Rate Restruction: To stop password guessing and brute-force assaults,
 *      login systems employ rate restriction. Systems can prevent unwanted access
 *      by restricting the quantity of login attempts mode by each person or IP address.
 * 
 * 
 * Types of Rate Limiting:
 * - Different approaches are used to control and regulate the number of requests
 *   a user or client can make to a system within a specified time period.
 * 
 * 1. IP-based Rate Limiting:
 *    - This technique limits the number of requests a client can make based on their
 *      IP address within a specific time period.
 *    - It is commonly used to prevent abuse like bots and denial-of-service attacks.
 *    - Example: An online retailer allows only 10 request per minute per IP address
 *      to prevent bots from scraping product data while allowing normal users to
 *      browse smoothly.
 * 
 *    Advantages:
 *    - This approach is widely used due to its simplicity and effectiveness in basic 
 *      traffic control.
 *      a. Simple to implement at both network and application levels
 *      b. Helps block excessive traffic from a single source
 * 
 *    Disadvantages:
 *    - Despite its benefits, it has some limitations in real-world scenarios.
 *      a. Can be bypassed using VPNs, proxies, or botnets
 *      b. May block legitimate users sharing the same IP (e.g., corporate networks)
 * 
 * 2. Server-Based Rate Limiting:
 *    - This technique limits the number of requests a server can handle within a 
 *      specific time period to prevent overload and maintain performance.
 *    - Example: A music streaming service allows only 100 requests per second per 
 *      server to ensure the system remains fast and responsive during peak usage.
 * 
 *    Advantages:
 *    - This approach helps maintain system stability by controlling traffic at the 
 *      server level.
 *      a. Protects servers from being overwhelmed during high traffic
 *      b. Ensures fair resource usage so no single user degrades performance
 * 
 *    Disadvantages
 *    - However, it may not be fully effective in distributed environments.
 *      a. Can be bypassed if requests are spread across multiple servers
 *      b. Legitimate users may face delays if limits are too strict or traffic is high
 * 
 * 3. Geography-Based Rate Limiting:
 *    - This technique limits requests based on the geographic location of the user’s 
 *      IP address. 
 *    - It is useful for controlling traffic from specific regions and improving 
 *      security or compliance.
 *    - Example: A social media platform limits requests from a region known for bot 
 *      activity to 10 requests per minute to reduce spam and fake accounts.
 * 
 *    Advantages
 *    - This approach is helpful for controlling region-specific traffic and improving
 *      security.
 *      a. Helps reduce malicious traffic from high-risk regions
 *      b. Assists in complying with regional laws and regulations
 * 
 *    Disadvantages
 *    - However, it may affect legitimate users and can be bypassed.
 *      a. Can be bypassed using VPNs or proxy servers
 *      b. May block genuine users traveling or using international networks
*/

/**
 * Read:
 * [https://geeksforgeeks.org/system-design/how-to-design-a-rate-limiter-api-learn-system-design/]
 * [https://www.geeksforgeeks.org/system-design/rate-limiting-in-system-design/]
 * [https://www.geeksforgeeks.org/system-design/rate-limiting-algorithms-system-design/]
*/
```

```python

/**
 * Designing a Rate Limiter:
 * - Imagine our application has a server that receives requests from many users.
 * - If one user keeps sending too many requests very quickly, our server can
 *   become overloaded.
 * - To prevent this, we use a Rate Limiter.
 *   a. Rate Limiter
 *      - A Rate Limiter is like a traffic controller for our application.
 *      - It controls how many requests a user can send within a certain time.
 *      - If the request is within the allowed limit, it is passed to the backend.
 *      - If the user exceeds the limit, the extra requests are blocked/rejected.
 *  
 *      Example:
 *      - Suppose we allow 2 posts per second.
 *      - User sends 1st post -> Allowed
 *      - User sends 2nd post -> Allowed
 *      - User sends 3rd post within the same second -> Blocked
 *  
 *   b. Examples:
 *      - 2 posts/second
 *        -> A user can create at most 2 posts every second.
 *  
 *      - 10 accounts/day per IP Address
 *        -> One IP address can create at most 10 accounts in a day.
 *  
 *      - Claim 5 rewards/week per device
 *        -> One device can claim at most 5 rewards in a week.
 *  
 *      The limit depends on what we want to protect in our application.
 *  
 *      Basic Flow:
 *  
 *      Client
 *         |
 *         v
 *      Rate Limiter
 *         |
 *         |---- Request within limit ----> Backend
 *         |
 *         |---- Request exceeds limit ---> Block/Reject
 *  
 *   c. Why needed?
 *      - Prevent DDoS / abusive traffic
 *        -> Stops a client from sending an excessive number of requests.
 *  
 *      - Reduce Cost
 *        -> Fewer unnecessary requests mean less usage of servers,
 *           databases, APIs, and other infrastructure.
 *  
 *      - Avoid Overload
 *        -> Protects our server and database from receiving too many
 *           requests at the same time.
 *  
 * In simple words:
 * A Rate Limiter is a gatekeeper that controls how frequently a client
 * can access our system.
 */
```

```python
/**
 * Step 1 - Problem ko samajhna aur Design ka Scope decide karna
 *
 * Rate Limiter design karne se pehle humein clearly samajhna hoga ki
 * hum exactly kya build kar rahe hain aur system ko kya-kya support karna hai.
 *
 *
 * Q: Hum kis type ka Rate Limiter design karenge?
 *    Client-side Rate Limiter ya Server-side API Rate Limiter?
 *
 * A: Hum Server-side API Rate Limiter design karenge.
 *
 *    Matlab Rate Limiter server side par hoga aur incoming API requests ko
 *    control karega, taaki har request directly backend tak na pahunch jaaye.
 *
 *    Client
 *       |
 *       v
 *    Rate Limiter
 *       |
 *       v
 *    Backend / API
 *
 *
 * Q: Rate Limiter kaise decide karega ki request allow karni hai
 *    ya block karni hai?
 *
 *    Kya IP Address, User ID ya kisi aur property ke basis par?
 *
 * A: Rate Limiter flexible hona chahiye, taaki hum different
 *    throttling rules support kar saken.
 *
 *    Example:
 *    - IP Address ke basis par request limit karna.
 *    - User ID ke basis par request limit karna.
 *    - API Key ke basis par request limit karna.
 *    - Device ID ke basis par request limit karna.
 *
 *    Matlab application ki requirement ke according hum decide kar sakte hain
 *    ki kis property ke basis par requests ko limit karna hai.
 *
 *
 * Q: System ka scale kya hoga?
 *    Kya ye small startup ke liye hai ya large company ke liye
 *    jahan bahut saare users hain?
 *
 * A: Hum system ko large number of requests handle karne ke liye design karenge.
 *
 *    Matlab Rate Limiter khud bottleneck nahi banna chahiye.
 *    Agar Rate Limiter hi slow ho gaya, to poori application slow ho sakti hai.
 *
 *
 * Q: Kya system distributed environment mein kaam karega?
 *
 * A: Yes.
 *
 *    Hamari application ke multiple servers ho sakte hain aur requests
 *    different servers par distribute ho sakti hain.
 *
 *       Client
 *          |
 *          v
 *       Load Balancer
 *        /    |    \
 *       v     v     v
 *    Server Server Server
 *
 *    Rate Limiter ko correctly kaam karna chahiye chahe request
 *    kisi bhi server par jaaye.
 *
 *
 * Q: Rate Limiter ko separate service banana chahiye ya
 *    application ke code ke andar implement karna chahiye?
 *
 * A: Ye ek design decision hai.
 *
 *    Hum Rate Limiter ko:
 *    - Application/backend code ke andar implement kar sakte hain.
 *    - Ek separate Rate Limiter service bana sakte hain.
 *    - API Gateway ya reverse proxy level par implement kar sakte hain.
 *
 *    Kaunsa approach choose karna hai, ye system ki requirements,
 *    scale, performance aur complexity par depend karega.
 *
 *
 * Q: Kya user ko batana hai ki uski request throttle/block ho gayi hai?
 *
 * A: Yes.
 *
 *    Agar user allowed limit se zyada requests bhejta hai,
 *    to system ko user ko clearly batana chahiye ki request reject
 *    kyun hui.
 *
 *    Example:
 *
 *    "Too many requests. Please try again later."
 *
 *
 * Simple words mein:
 *
 * Rate Limiter design karne se pehle hum 6 cheezein decide karte hain:
 *
 *    1. Kahan kaam karega?
 *       -> Server-side API Rate Limiter
 *
 *    2. Kis basis par requests limit karenge?
 *       -> IP, User ID, API Key, Device ID, etc.
 *
 *    3. Kitna traffic handle karega?
 *       -> Large number of requests
 *
 *    4. Kya multiple servers support karne hain?
 *       -> Yes, distributed environment
 *
 *    5. Kahan implement karna hai?
 *       -> Application code, separate service, ya API Gateway
 *
 *    6. Kya user ko throttling ke baare mein batana hai?
 *       -> Yes
 *
 * Ye step basically system ka scope define karta hai.
 * Scope clear hone ke baad hum Rate Limiter ki actual architecture
 * aur components design karna start kar sakte hain.
 */ 
```

```python
/**
 * Requirements:
 *
 * Rate Limiter design karte waqt humein kuch important requirements
 * ka dhyaan rakhna hoga, taaki system accurate, fast aur reliable ho.
 *
 * 1. Accurate Limiting
 *    -> Rate Limiter ko correctly count karna chahiye ki user ne
 *       kitni requests bheji hain.
 *
 *    Example:
 *    Agar limit 10 requests/minute hai, to user ko exactly
 *    allowed limit tak requests milni chahiye aur uske baad
 *    extra requests block honi chahiye.
 *
 *
 * 2. Low Latency
 *    -> Rate Limiter request ko check karne mein bahut kam time
 *       lena chahiye.
 *
 *    Rate Limiter har API request ke beech mein hota hai, isliye
 *    agar ye slow hua to poori application slow ho jayegi.
 *
 *
 * 3. Low Memory
 *    -> Rate Limiter ko request information store karne ke liye
 *       kam memory use karni chahiye.
 *
 *    Large system mein millions of users ho sakte hain, isliye
 *    har user ka bahut zyada data memory mein store karna expensive
 *    ho sakta hai.
 *
 *
 * 4. Distributed Support
 *    -> Rate Limiter ko multiple servers ke saath correctly kaam
 *       karna chahiye.
 *
 *    Example:
 *
 *       Client
 *          |
 *          v
 *      Load Balancer
 *       /    |    \
 *      v     v     v
 *    Server Server Server
 *
 *    Agar same user ki requests different servers par jaati hain,
 *    to Rate Limiter ko phir bhi correct request count maintain karna
 *    chahiye.
 *
 *
 * 5. Clear Errors
 *    -> Jab user limit exceed kare, to system ko clear error dena chahiye.
 *
 *    Example:
 *    "Too many requests. Please try again later."
 *
 *    Isse client ko samajh aata hai ki request kyun reject hui.
 *
 *
 * 6. Fault Tolerant
 *    -> Agar Rate Limiter ka koi component fail ho jaaye,
 *       to poora system crash nahi hona chahiye.
 *
 *    System ko failures handle karne aur recover karne ki ability
 *    honi chahiye.
 *
 *
 * Simple words mein:
 *
 * Rate Limiter:
 *    -> Requests ko accurately limit kare.
 *    -> Request ko quickly check kare.
 *    -> Kam memory use kare.
 *    -> Multiple servers ke saath kaam kare.
 *    -> User ko clear error de.
 *    -> Failure hone par bhi reliable rahe.
 *
 * In requirements ka goal hai ki Rate Limiter
 * accurate + fast + scalable + reliable ho.
 */
```

```python
/**
 * Step 2 - High-Level Design propose karna aur feedback lena
 * - Ab jab humne requirements samajh li hain, next step hai decide karna ki 
 *   Rate Limiter ko system mein exactly kahan place karna hai.
 *
 * Rate Limiter ko mainly do jagah rakh sakte hain:
 * 1. Client-Side Rate Limiter
 *    - Is approach mein Rate Limiter client/browser/app ke andar hota hai.
 *    - Problem:
 *      a. Unreliable
 *         -> Client hamare control mein nahi hota.
 *         -> User client-side code ko modify ya disable kar sakta hai.
 *      b. Easily Bypassed
 *         -> Malicious user Rate Limiter ko bypass karke directly API ko requests 
 *            bhej sakta hai.
 *
 *              Client-Side Rate Limiter
 *
 *       Client
 *      +------+        +----------------+        +-------------+
 *      |      |------->| Rate Limiter   |------->| API Servers |
 *      +------+        +----------------+        +-------------+
 *
 *    - Isliye client-side Rate Limiter security ke liye reliable nahi hai.
 *
 *
 * 2. Server-Side / Middleware Rate Limiter
 *    - Is approach mein Rate Limiter server side par hota hai.
 *    - Client se request pehle Rate Limiter ke through jaati hai.
 *    - Agar request allowed hai, tabhi API server ko forward hoti hai.
 *
 *    Benefits:
 *    a. Central Control
 *       -> Saari incoming requests ko ek central place par control kiya ja sakta hai.
 *    b. Protects APIs
 *       -> Excessive requests API servers tak pahunchne se pehle hi block ho jaati hain.
 *
 *       Client
 *      +------+       HTTP Request       +----------------------------+
 *      |      |------------------------->|       API Servers          |
 *      +------+                          | Rate Limiter / Middleware  |
 *                                        +----------------------------+
 *
 *
 * Request ka basic flow:
 *    Client
 *       |
 *       | HTTP Request
 *       v
 *    Rate Limiter
 *       |
 *       |--- Request allowed ---> API Server
 *       |
 *       |--- Limit exceeded ---> Reject Request
 *
 *
 * Conclusion:
 * - Client-side Rate Limiter:
 *   -> Unreliable
 *   -> Easily bypassed
 * - Server-side / Middleware Rate Limiter:
 *   -> Central control
 *   -> APIs ko protect karta hai
 *   -> Client ko bypass karna difficult hota hai
 *
 * - Isliye hamare system mein hum Server-side / Middleware Rate Limiter ko prefer 
 *   karenge.
 * - Ab next step mein hum dekhenge ki server-side Rate Limiter ko architecture mein 
 *   exactly kahan place karna hai aur kaise requests ko allow ya reject karega.
 */
```

```python
/**
 * How Rate Limiting Works?
 * - Rate Limiting ka simple kaam hai decide karna ki Client ki request allow karni hai
 *   ya block karni hai.
 * - Client se jab request aati hai, sabse pehle request Rate Limiter ke paas jaati hai.
 *
 *    +--------+              +--------------+
 *    | Client | ------------>| Rate Limiter |
 *    +--------+              +--------------+
 *                                  |
 *                                  |
 *                           Request check hoti hai
 *                                  |
 *                    +-------------+-------------+
 *                    |                           |
 *                Limit ke andar            Limit exceed
 *                    |                           |
 *                    v                           X
 *                API Server                 Request Block
 *
 * Rate Limiter check karta hai:
 * - "Kya is client ne allowed limit ke andar request bheji hai?"
 * - Agar request limit ke andar hai:
 *    -> Request ko allow karo.
 *    -> Request API Server/backend tak jaayegi.
 * - Agar request limit exceed kar chuki hai:
 *    -> Request ko reject/block karo.
 *    -> API Server tak request nahi jaayegi.
 *
 * Example:
 * - Rule: 2 requests/second
 *   - 1st request -> Allow
 *   - 2nd request -> Allow
 *   - 3rd request -> Block
 *
 * Is tarah Rate Limiter API Server ko excessive traffic se protect
 * karta hai aur ensure karta hai ki client defined limit ke andar
 * hi requests bheje.
 *
 * Simple words mein:
 * - Rate Limiter ek gatekeeper ki tarah kaam karta hai.
 * - Har request ke liye wo check karta hai:
 *   - "Allowed hai?"      -> API Server ko bhejo
 *   - "Allowed nahi hai?" -> Request ko block karo
 */
```

```python
/**
 * Server vs Gateway Rate Limiter:
 * - Ab question hai ki Rate Limiter ko exactly kahan implement karein?
 *   1. Application / Server ke andar?
 *   2. API Gateway par?
 * - Iska koi fixed answer nahi hai.
 * - Ye hamare system ke architecture, technology stack aur requirements par depend 
 *   karta hai.
 *
 * 1. Server-Side Rate Limiter
 *    - Is approach mein Rate Limiter directly application/server ke andar implement 
 *      hota hai.
 *  
 *      Client
 *         |
 *         v
 *      API Server
 *         |
 *         +--> Rate Limiter
 *         |
 *         v
 *      Application Logic
 *
 *    Benefits:
 *    a. Full Algorithm Control
 *       -> Humein Rate Limiting algorithm par complete control milta hai.
 *       -> Hum apni requirement ke according custom logic bana sakte hain.
 *    b. Language Flexibility
 *       -> Rate Limiter ko apni backend programming language mein implement kar sakte h.
 *       -> Example: Java, Node.js, Python, Go, etc.
 *
 * 2. API Gateway Rate Limiter
 *    - Is approach mein Rate Limiter API Gateway par implement hota hai.
 *
 *      Client
 *         |
 *         v
 *      API Gateway
 *         |
 *         +--> Rate Limiter
 *         |
 *         +-------------------+
 *         |         |         |
 *         v         v         v
 *      Service A  Service B  Service C
 *
 *    Benefits:
 *    a. Fits Microservices
 *       -> Microservices architecture mein bahut saare services hote hain.
 *       -> Gateway ek central entry point ki tarah kaam karta hai.
 *       -> Isliye ek hi jagah par Rate Limiter laga kar multiple services ko protect
 *          kiya ja sakta hai.
 *    b. Centralized Control
 *       -> Saari incoming API requests gateway se pass hoti hain.
 *       -> Isliye Rate Limiting ko ek central location se manage kar sakte hain.
 *
 * Practical Choice:
 * a. Agar resources limited hain:
 *    -> API Gateway par Rate Limiter rakhna better ho sakta hai.
 *    -> Ek central place par Rate Limiting manage ho jaati hai.
 * b. Agar custom requirements hain:
 *    -> Server-side Rate Limiter better ho sakta hai.
 *    -> Humein algorithm aur business logic par zyada control milta hai.
 *
 *
 * Simple words mein:
 * a. Server-Side:
 *    -> More control
 *    -> More customization
 *    -> Language flexibility
 * b. API Gateway:
 *    -> Centralized control
 *    -> Microservices ke liye useful
 *    -> Multiple APIs ko ek saath protect kar sakta hai
 *
 * Isliye Rate Limiter ki location choose karna system ke architecture aur requirements
 * par depend karta hai.
 */
```

```python
/**
 * Algorithms for Rate Limiting:
 * 1. Token Bucket
 * 2. Leaking Bucket
 * 3. Fixed window counter
 * 4. Sliding window log
 * 5. Sliding window counter
*/
```

```python
/**
 * Algorithm-1: Token Bucket
 * - Token Bucket ek widely used, simple aur reliable Rate Limiting algorithm hai.
 * - Iska basic idea samajhne ke liye imagine karo ki har client ke liye ek
 *   bucket hai jisme tokens rakhe hue hain. Bucket ki ek fixed maximum capacity
 *   hoti hai aur time ke saath us bucket mein tokens periodically add hote rehte hain.
 *
 * - Maan lo hamari bucket ki capacity 5 tokens hai aur har second 1 token add hota hai.
 * - Iska matlab bucket mein maximum 5 tokens hi ho sakte hain.
 *
 * - Ab jab client API request bhejta hai, Rate Limiter sabse pehle bucket check karta
 *   hai aur dekhta hai ki usmein koi token available hai ya nahi.
 *
 * - Agar token available hai, to Rate Limiter ek token consume karta hai aur request 
 *   ko allow karke API Server tak bhej deta hai.
 *
 *    Client
 *       |
 *       | Request
 *       v
 *    Token Bucket
 *       |
 *       |-- Token available --> 1 Token consume --> Request Allow
 *       |
 *       |-- No Token ---------> Request Block
 *
 * Example ke liye, maan lo bucket mein initially 5 tokens hain.
 * - Client pehli request bhejta hai, ek token consume hota hai aur 4 tokens bach jaate hain.
 * - Dusri request par 1 aur token consume hota hai aur 3 tokens bach jaate hain.
 * - Isi tarah 5 requests tak sabhi requests allow ho jaayengi aur bucket mein 0 tokens
 *   bachenge.
 *
 *    Request 1 -> 5 to 4 tokens -> Allow
 *    Request 2 -> 4 to 3 tokens -> Allow
 *    Request 3 -> 3 to 2 tokens -> Allow
 *    Request 4 -> 2 to 1 token  -> Allow
 *    Request 5 -> 1 to 0 tokens -> Allow
 *    Request 6 -> No token      -> Block
 *
 * - Lekin token khatam hone ka matlab ye nahi hai ki client permanently block ho gaya.
 * - Tokens time ke saath automatically refill hote rehte hain.
 *
 * - Agar refill rate 1 token/second hai, to 1 second ke baad bucket mein 1 token aa jayega.
 * - Ab agar client ek new request bhejta hai, to wo token consume hoga aur request 
 *   allow ho jaayegi. Phir next token ke liye system ko refill ka wait karna padega.
 *
 *    0 Tokens
 *       |
 *       | 1 second
 *       v
 *    1 Token
 *       |
 *       | 1 second
 *       v
 *    2 Tokens
 *
 * - Bucket ki capacity fixed hoti hai, isliye tokens unlimited nahi ho sakte.
 * - Agar bucket already full hai aur naya token add hona hai, to extra token store
 *   nahi hoga.
 *
 *    Capacity = 5
 *
 *    +-----------+
 *    |   Token   |
 *    |   Token   |
 *    |   Token   |
 *    |   Token   |
 *    |   Token   |
 *    +-----------+
 *
 * - Iska ek important benefit ye hai ki Token Bucket limited burst traffic ko allow kar
 *   sakta hai. Agar client ne kuch time tak requests nahi bheji, to tokens bucket mein
 *   collect hote rahenge. Baad mein client ek saath multiple requests bhej sakta hai,
 *   jab tak bucket mein tokens available hain.
 *
 * - Simple words mein, Token Bucket ko ek real bucket samjho jisme tokens stored hain.
 * - Har API request bucket se ek token leti hai. Agar token hai to request allow hoti 
 *   hai, aur agar token nahi hai to request block ho jaati hai. Tokens time ke saath 
 *   refill hote hain, lekin bucket ki maximum capacity fixed hoti hai.
 *
 * - Isi simple mechanism ki wajah se Token Bucket client ki request rate ko control 
 *   karta hai, limited bursts ko allow karta hai aur API Server ko excessive traffic
 *   se protect karta hai.
 */ 
```

```python
/**
 * Algorithm-2: Leaky Bucket
 * Ab Token Bucket ke baad next Rate Limiting algorithm hai Leaky Bucket.
 * Isko samajhne ke liye imagine karo ki hamare paas ek bucket hai jisme
 * incoming requests paani ki tarah continuously aa rahi hain.
 *
 * Bucket ki ek fixed capacity hoti hai aur bucket ke neeche ek chhota hole
 * hota hai jahan se requests ek fixed rate par bahar nikalti hain.
 *
 *    Incoming Requests
 *          |
 *          v
 *    +-------------+
 *    | Request     |
 *    | Request     |
 *    | Request     |
 *    | Request     |
 *    +-------------+
 *          |
 *          | Fixed Rate
 *          v
 *       API Server
 *
 * Maan lo bucket ki capacity 5 requests hai aur processing rate
 * 1 request/second hai.
 *
 * Iska matlab ek time par bucket maximum 5 pending requests store kar sakti hai,
 * aur Rate Limiter har second sirf 1 request ko API Server ki taraf bhejega.
 *
 * Ab maan lo client ek saath 5 requests bhejta hai.
 * Saari requests bucket mein temporarily store ho jaayengi.
 *
 *    Request 1
 *    Request 2
 *    Request 3
 *    Request 4
 *    Request 5
 *       |
 *       v
 *    +---------+
 *    | Bucket  |
 *    |---------|
 *    | Req 1   |
 *    | Req 2   |
 *    | Req 3   |
 *    | Req 4   |
 *    | Req 5   |
 *    +---------+
 *
 * Ab bucket fixed rate par requests release karegi:
 *
 *    1st second -> Request 1 -> API Server
 *    2nd second -> Request 2 -> API Server
 *    3rd second -> Request 3 -> API Server
 *    4th second -> Request 4 -> API Server
 *    5th second -> Request 5 -> API Server
 *
 * Yaani requests ek saath aane ke baad bhi API Server ko
 * controlled aur consistent rate par requests milengi.
 *
 * Lekin agar bucket already full hai aur uske baad ek aur request aa jaati hai,
 * to bucket mein space nahi hogi.
 *
 *    Bucket Capacity = 5
 *
 *    +---------+
 *    | Req 1   |
 *    | Req 2   |
 *    | Req 3   |
 *    | Req 4   |
 *    | Req 5   |
 *    +---------+
 *
 *    New Request -> No Space -> Reject / Drop
 *
 * Isliye Leaky Bucket ka main purpose incoming traffic ko ek
 * fixed aur smooth rate par process karna hai.
 *
 * Token Bucket mein hum tokens generate/refill karte hain aur
 * request aane par token consume hota hai.
 *
 * Leaky Bucket mein hum requests ko bucket mein queue karte hain
 * aur unhe ek fixed rate par bahar nikalte hain.
 *
 *    Token Bucket:
 *    Request -> Token check -> Allow/Block
 *
 *    Leaky Bucket:
 *    Request -> Bucket/Queue -> Fixed Rate -> API Server
 *
 * Simple words mein:
 *
 * Leaky Bucket ko ek aisi bucket samjho jisme requests paani ki tarah
 * collect hoti hain aur neeche ke chhote hole se ek fixed speed par
 * bahar nikalti hain.
 *
 * Chahe client ek saath bahut saari requests bheje, Leaky Bucket
 * API Server ko requests ek controlled rate par hi forward karega.
 *
 * Agar bucket full ho gayi aur new request ke liye space nahi hai,
 * to new request reject/drop ho jaayegi.
 *
 * Isi wajah se Leaky Bucket sudden traffic ko smooth karke
 * API Server ko overload hone se protect karta hai.
 */

/**
 * Pros:
 * 1. Memory Efficient
 * 2. Stable throughtput due to fixed rate
 * 
 * Cons:
 * 1. Burst unfriendly (recent requests can be rate limited due to old requests)
 * 2. Hard to tune
*/
```

```python
/**
 * How to Scale These Buckets?
 *
 * Ab humne Token Bucket aur Leaky Bucket samajh liya.
 * Lekin real-world system mein ek aur important question aata hai:
 *
 * "Agar millions of users hain aur multiple servers hain,
 * to in buckets ko kaise scale karenge?"
 *
 *
 * Sabse pehle single server ka case samjho.
 *
 * Agar hamare paas sirf ek server hai, to hum buckets ko server ki
 * memory mein maintain kar sakte hain:
 *
 *    Client A -> Server -> Bucket A
 *    Client B -> Server -> Bucket B
 *    Client C -> Server -> Bucket C
 *
 * Lekin production system mein multiple servers ho sakte hain:
 *
 *                             +--> Server 1
 *    Client --> Load Balancer +--> Server 2
 *                             +--> Server 3
 *
 * Ab problem ye hai ki same client ki requests different servers
 * par ja sakti hain.
 *
 * Example:
 *
 *    Request 1 -> Server 1 -> Bucket A
 *    Request 2 -> Server 2 -> Bucket A
 *
 * Agar har server apni memory mein separate Bucket A maintain karega,
 * to Server 1 aur Server 2 ko client ke bucket ka different state
 * pata ho sakta hai.
 *
 * Isse Rate Limiting incorrect ho sakti hai.
 *
 *
 * Is problem ko solve karne ke liye hum bucket state ko ek
 * Shared Store mein rakh sakte hain.
 *
 *                         +--> Server 1 --+
 *                         |               |
 *    Client --> Load Balancer +--> Server 2 --+--> Shared Store
 *                         |               |
 *                         +--> Server 3 --+
 *
 * Shared Store mein hum har client ke bucket ka state maintain kar sakte hain.
 *
 *    User A -> tokens = 3
 *    User B -> tokens = 7
 *    User C -> tokens = 1
 *
 * Rate Limiter ke liye Redis jaise fast in-memory datastore ko
 * shared state store karne ke liye use kiya ja sakta hai.
 *
 *
 * Lekin sirf shared storage use karna enough nahi hai.
 * Ek aur important problem hai: Race Condition.
 *
 * Maan lo User A ke bucket mein sirf 1 token available hai
 * aur same time par 2 requests aa jaati hain.
 *
 *    Request 1 ---> Check: 1 Token Available
 *    Request 2 ---> Check: 1 Token Available
 *
 * Agar dono requests pehle token check karein aur baad mein
 * token consume karein, to dono requests allow ho sakti hain.
 *
 * Ye incorrect hai, kyunki sirf 1 token available tha.
 *
 * Isliye bucket ka operation Atomic hona chahiye.
 *
 *    Check Token
 *        +
 *    Consume Token
 *        +
 *    Update Bucket
 *
 * Ye operations ek atomic operation ki tarah execute hone chahiye.
 *
 * Redis mein is type ke logic ko atomic operations ya Lua Script
 * ke through implement kiya ja sakta hai.
 *
 *
 * Ab ek aur problem hai:
 *
 * "Agar millions of users hain to millions of buckets kaise manage karenge?"
 *
 * Maan lo 10 million users hain:
 *
 *    10 Million Users
 *          |
 *          v
 *    10 Million Buckets
 *
 * Har bucket mein bahut zyada information store karna memory expensive
 * ho sakta hai.
 *
 * Isliye bucket state ko minimal rakhte hain.
 *
 * Example:
 *
 *    Bucket:
 *       userId
 *       tokens
 *       lastRefillTime
 *
 * Aur jo users long time se inactive hain, unke bucket records ko
 * TTL / expiration ke through automatically remove kiya ja sakta hai.
 *
 *
 * Ab agar Rate Limiter ka shared Redis bhi bahut large traffic
 * handle nahi kar pa raha ho, to Redis ko bhi horizontally scale
 * kar sakte hain.
 *
 *                         Rate Limiter
 *                              |
 *                 +------------+------------+
 *                 |            |            |
 *                 v            v            v
 *              Redis 1      Redis 2      Redis 3
 *                 |            |            |
 *              Users A-D    Users E-H    Users I-L
 *
 * Bucket state ko different Redis nodes par distribute/shard
 * kiya ja sakta hai.
 *
 * Example:
 *
 *    bucket:user:101 -> Redis 1
 *    bucket:user:202 -> Redis 2
 *    bucket:user:303 -> Redis 3
 *
 *
 * Final high-level architecture kuch aisa ho sakta hai:
 *
 *                         +----------------+
 *                         |  Load Balancer |
 *                         +-------+--------+
 *                                 |
 *                   +-------------+-------------+
 *                   |             |             |
 *                   v             v             v
 *             +-----------+ +-----------+ +-----------+
 *             | Rate      | | Rate      | | Rate      |
 *             | Limiter 1 | | Limiter 2 | | Limiter 3 |
 *             +-----+-----+ +-----+-----+ +-----+-----+
 *                   |             |             |
 *                   +-------------+-------------+
 *                                 |
 *                                 v
 *                       +-------------------+
 *                       |   Redis Cluster   |
 *                       |                   |
 *                       |   Bucket State   |
 *                       +---------+---------+
 *                                 |
 *                                 v
 *                            API Servers
 *
 *
 * Token Bucket ke case mein shared state kuch aisa ho sakta hai:
 *
 *    Bucket:
 *       capacity       = 10
 *       tokens         = 6
 *       refillRate     = 2 tokens/sec
 *       lastRefillTime = ...
 *
 * Har request par Rate Limiter:
 *
 *    1. Current time check karega.
 *    2. Calculate karega ki kitne tokens refill hone chahiye.
 *    3. Bucket update karega.
 *    4. Ek token consume karega.
 *    5. Request ko Allow ya Reject karega.
 *    6. Updated bucket state store karega.
 *
 *
 * Simple words mein:
 *
 * Rate Limiter ko horizontally scale karne ke liye multiple
 * Rate Limiter instances ko Load Balancer ke peeche rakh sakte hain.
 *
 * Bucket state ko server ki local memory mein rakhne ke bajay
 * shared low-latency datastore, jaise Redis, mein rakh sakte hain.
 *
 * Bucket updates ko atomic rakhna zaroori hai taaki race conditions
 * na aayein. Large-scale system mein bucket state ko shard bhi
 * kar sakte hain aur inactive buckets ko TTL se remove kar sakte hain.
 *
 * Interview mein simple answer:
 *
 *    "We horizontally scale Rate Limiter instances behind a Load Balancer,
 *     keep bucket state in a shared low-latency store like Redis, make
 *     bucket updates atomic, shard the state for large scale, and expire
 *     inactive bucket entries to control memory usage."
 */
```

```python
/**
 * Algorithm-3: Fixed Window Counter
 * Ab Leaky Bucket ke baad next Rate Limiting algorithm hai
 * Fixed Window Counter.
 *
 * Isko samajhne ke liye imagine karo ki hum time ko fixed windows
 * mein divide kar dete hain.
 *
 * Example:
 *    Limit = 5 requests/minute
 *
 * To hum time ko 1-minute ke windows mein divide karenge:
 *
 *    Window 1 -> 10:00:00 to 10:00:59
 *    Window 2 -> 10:01:00 to 10:01:59
 *    Window 3 -> 10:02:00 to 10:02:59
 *
 * Har window ka apna request counter hota hai.
 * Jaise hi new window start hoti hai, counter reset ho jaata hai.
 *
 * Maan lo 10:00 baje new window start hui aur limit 5 requests hai.
 * Starting mein counter 0 hai.
 *
 *    10:00:05 -> Request 1 -> Counter = 1 -> Allow
 *    10:00:10 -> Request 2 -> Counter = 2 -> Allow
 *    10:00:20 -> Request 3 -> Counter = 3 -> Allow
 *    10:00:30 -> Request 4 -> Counter = 4 -> Allow
 *    10:00:40 -> Request 5 -> Counter = 5 -> Allow
 *
 * Ab 5 requests ki limit complete ho gayi.
 * Agar 10:00:50 par ek aur request aati hai:
 *
 *    Request 6 -> Counter = 5
 *             -> Limit exceed
 *             -> Block
 *
 * Ab 10:01:00 par new window start hoti hai.
 * Counter reset ho jaata hai:
 *
 *    Counter = 0
 *
 * Ab new requests phir se allow hongi:
 *
 *    10:01:05 -> Request 1 -> Counter = 1 -> Allow
 *
 * Yaani Fixed Window Counter mein hum sirf current fixed time window
 * ke andar requests count karte hain.
 *
 *
 * Basic Flow:
 *
 *    Client
 *       |
 *       | Request
 *       v
 *    Fixed Window Counter
 *       |
 *       |-- Counter < Limit --> Counter + 1 --> Allow
 *       |
 *       |-- Counter = Limit --> Reject / Block
 *
 *
 * Example:
 *
 *    Limit = 5 requests/minute
 *
 *    +--------------------------------+
 *    | 10:00:00 - 10:00:59            |
 *    |                                |
 *    | Req 1 -> Allow                 |
 *    | Req 2 -> Allow                 |
 *    | Req 3 -> Allow                 |
 *    | Req 4 -> Allow                 |
 *    | Req 5 -> Allow                 |
 *    | Req 6 -> Block                 |
 *    +--------------------------------+
 *
 *    10:01:00
 *         |
 *         v
 *    Counter Reset -> 0
 *
 *    +--------------------------------+
 *    | 10:01:00 - 10:01:59            |
 *    |                                |
 *    | Req 1 -> Allow                 |
 *    | Req 2 -> Allow                 |
 *    | ...                            |
 *    +--------------------------------+
 *
 *
 * Is algorithm ka main benefit ye hai ki ye bahut simple aur
 * memory-efficient hai.
 *
 * Humein har request ka exact timestamp store karne ki zarurat nahi hoti.
 * Sirf current window ka counter maintain karna hota hai.
 *
 * Lekin Fixed Window Counter ki ek important limitation hai:
 * boundary ke paas traffic ka sudden burst aa sakta hai.
 *
 * Example:
 *
 *    Limit = 5 requests/minute
 *
 *    10:00:55 -> 5 requests -> Allow
 *    10:01:05 -> 5 requests -> Allow
 *
 * Sirf 10 seconds ke andar client ne total 10 requests bhej di,
 * even though limit 5 requests/minute hai.
 *
 * Ye isliye hua kyunki pehli 5 requests old window mein count hui
 * aur next 5 requests new window mein count hui.
 *
 * Is problem ko Fixed Window Boundary Problem kehte hain.
 *
 * Simple words mein:
 *
 * Fixed Window Counter mein time ko fixed intervals mein divide karte hain,
 * jaise 1 minute. Har interval mein requests ka counter maintain hota hai.
 * Limit ke andar request hai to counter increase karke request allow hoti hai.
 * Limit reach ho gayi to remaining requests block ho jaati hain.
 * Jaise hi new window start hoti hai, counter reset ho jaata hai.
 *
 * Ye approach simple, fast aur memory-efficient hai,
 * lekin window boundaries par sudden traffic burst ko perfectly control
 * nahi kar paati.
 */
```

```python
/**
 * Algorithm-4: Sliding Window Log
 * Ab Fixed Window Counter ke baad next Rate Limiting algorithm hai
 * Sliding Window Log.
 *
 * Fixed Window Counter mein problem ye thi ki time ko fixed windows mein
 * divide karne ki wajah se window boundary ke paas sudden burst aa sakta tha.
 *
 * Sliding Window Log is problem ko solve karne ke liye har request ka
 * timestamp store karta hai aur ek continuously moving time window
 * ke andar requests ko count karta hai.
 *
 * Example:
 *
 *    Limit = 5 requests/minute
 *
 * Iska matlab hai:
 *    "Kisi bhi point par pichhle 1 minute mein maximum 5 requests allowed hain."
 *
 * Ab maan lo client requests bhejta hai:
 *
 *    10:00:10 -> Request 1
 *    10:00:20 -> Request 2
 *    10:00:30 -> Request 3
 *    10:00:40 -> Request 4
 *    10:00:50 -> Request 5
 *
 * Ab 10:00:55 par ek new request aati hai.
 *
 * Rate Limiter pichhle 1 minute ko dekhega:
 *
 *    09:59:55 ---------------- 10:00:55
 *                                ^
 *                              Current
 *
 * Is window ke andar 5 requests already hain,
 * isliye 6th request block ho jaayegi.
 *
 *    Request 6 -> 5 requests already present -> Block
 *
 * Ab time aage badhta hai.
 *
 * Maan lo 10:01:11 par ek new request aati hai.
 *
 * Ab sliding window hai:
 *
 *    10:00:11 ---------------- 10:01:11
 *                                ^
 *                              Current
 *
 * 10:00:10 wali request ab window se bahar ho gayi,
 * kyunki wo 1 minute se purani hai.
 *
 * Isliye ab sirf 4 requests window ke andar hain.
 *
 *    Old Request -> Remove
 *    Current Count = 4
 *    New Request -> Allow
 *
 * Aur new request ka timestamp log mein add kar diya jaata hai.
 *
 *
 * Basic Flow:
 *
 *    Client
 *       |
 *       | Request
 *       v
 *    Sliding Window Log
 *       |
 *       |-- Old timestamps remove karo
 *       |
 *       |-- Current requests count karo
 *       |
 *       +-- Count < Limit --> Add timestamp --> Allow
 *       |
 *       +-- Count >= Limit -> Reject / Block
 *
 *
 * Is algorithm mein hum request ke timestamps ko ek log/list mein
 * maintain karte hain.
 *
 * Example:
 *
 *    Limit = 5 requests/minute
 *
 *    Timestamp Log:
 *
 *    [10:00:10,
 *     10:00:20,
 *     10:00:30,
 *     10:00:40,
 *     10:00:50]
 *
 * Jab 10:00:55 par request aati hai:
 *
 *    -> Last 1 minute ke timestamps check karo.
 *    -> 5 requests already hain.
 *    -> New request reject karo.
 *
 * Jab 10:01:11 par request aati hai:
 *
 *    -> 10:00:10 timestamp ab window se bahar hai.
 *    -> Us timestamp ko remove karo.
 *    -> Ab 4 requests remaining hain.
 *    -> New request allow karo.
 *    -> New timestamp add karo.
 *
 *
 * Fixed Window Counter se difference:
 *
 *    Fixed Window:
 *    10:00:00 ---------------- 10:00:59
 *    10:01:00 ---------------- 10:01:59
 *
 *    Window fixed hoti hai aur counter reset hota hai.
 *
 *
 *    Sliding Window Log:
 *
 *       <------ 1 minute ------>
 *                  |
 *                  v
 *    10:00:15 ---------------- 10:01:15
 *
 *    Window current time ke saath continuously move hoti hai.
 *
 *
 * Isliye Sliding Window Log fixed window ke boundary problem ko
 * kaafi accurately solve karta hai.
 *
 * Lekin iska ek drawback hai:
 *
 *    -> Har request ka timestamp store karna padta hai.
 *    -> High traffic system mein bahut saare timestamps store ho sakte hain.
 *    -> Isliye memory usage Fixed Window Counter se zyada ho sakti hai.
 *
 * Example:
 *
 *    Agar millions of users hain aur har user bahut saari requests
 *    bhej raha hai, to timestamps kaafi memory consume kar sakte hain.
 *
 *
 * Simple words mein:
 *
 * Sliding Window Log mein hum har request ka timestamp store karte hain.
 * Jab new request aati hai, hum pichhle fixed duration, jaise 1 minute,
 * ke andar aane wali requests ko count karte hain.
 *
 * Jo timestamps 1 minute se purane ho gaye hain unhe remove kar dete hain.
 * Agar current window mein requests limit se kam hain, to new request
 * allow karke uska timestamp store kar dete hain.
 * Agar limit already reach ho chuki hai, to request block kar dete hain.
 *
 * Is approach ka main benefit ye hai ki rate limiting zyada accurate hoti hai,
 * lekin iska main drawback high memory usage hai because har request ka
 * timestamp maintain karna padta hai.
 */
```

```python
/**
 * Algorithm-5: Sliding Window Counter
 * Ab Sliding Window Log ke baad next Rate Limiting algorithm hai
 * Sliding Window Counter.
 *
 * Sliding Window Log mein hum har request ka exact timestamp store karte the.
 * Problem ye hai ki high traffic system mein bahut saare timestamps store
 * karne pad sakte hain, jiski wajah se memory usage bahut badh sakti hai.
 *
 * Sliding Window Counter isi problem ko solve karta hai.
 * Ismein hum har request ka timestamp store nahi karte.
 * Instead, hum time windows ke counters maintain karte hain aur
 * current sliding window ke requests ka approximate count calculate karte hain.
 *
 *
 * Example:
 *
 *    Limit = 5 requests/minute
 *
 * Hum time ko 1-minute windows mein divide karte hain:
 *
 *    Previous Window       Current Window
 *    10:00 - 10:01         10:01 - 10:02
 *
 * Maan lo:
 *
 *    Previous Window Counter = 4
 *    Current Window Counter  = 2
 *
 * Ab current time 10:01:30 hai.
 *
 * Iska matlab current window ka 50% time complete ho chuka hai.
 * To previous window ka sirf remaining 50% portion hamari
 * sliding window mein relevant hoga.
 *
 *
 * Sliding Window:
 *
 *    Previous Window             Current Window
 *    |----------------|          |----------------|
 *    10:00          10:01       10:02
 *                     ^
 *                     |
 *                  Current
 *
 * Current time 10:01:30 hone par:
 *
 *    Previous Window ka relevant part = 50%
 *    Current Window ka relevant part  = 100%
 *
 *
 * Ab approximate request count calculate karenge:
 *
 *    Previous Count × Previous Window ka relevant portion
 *    + Current Window Count
 *
 *    = 4 × 50% + 2
 *    = 2 + 2
 *    = 4 requests
 *
 * Current sliding window mein approximately 4 requests hain.
 *
 * Limit 5 hai, isliye new request allow ho sakti hai.
 *
 *
 * Basic Flow:
 *
 *    Client
 *       |
 *       | Request
 *       v
 *    Sliding Window Counter
 *       |
 *       |-- Previous Window Counter
 *       |-- Current Window Counter
 *       |
 *       |-- Approximate count calculate
 *       |
 *       +-- Count < Limit --> Counter increase --> Allow
 *       |
 *       +-- Count >= Limit -> Reject / Block
 *
 *
 * Ab ek aur example samjho.
 *
 *    Limit = 5 requests/minute
 *
 *    Previous Window = 5 requests
 *    Current Window  = 3 requests
 *
 * Agar current window ka sirf 20% time complete hua hai,
 * to previous window ka approximately 80% portion relevant hoga.
 *
 *    Estimated Count
 *       = 5 × 80% + 3
 *       = 4 + 3
 *       = 7
 *
 * Estimated count 5 ki limit se zyada hai.
 *
 *    New Request -> Block
 *
 *
 * Is algorithm mein important baat ye hai ki hum har request ka
 * timestamp store nahi karte.
 *
 * Instead:
 *
 *    Previous Window -> Counter
 *    Current Window  -> Counter
 *
 * Bas in counters aur current time ke basis par approximate
 * request count calculate karte hain.
 *
 *
 * Fixed Window Counter se difference:
 *
 *    Fixed Window Counter:
 *
 *    10:00 ---------------- 10:01
 *    10:01 ---------------- 10:02
 *
 *    Har window independent hoti hai.
 *    Window boundary par counter reset ho jaata hai.
 *
 *
 *    Sliding Window Counter:
 *
 *    Window current time ke saath continuously move hoti hai.
 *    Previous aur current window ke counters ko combine karke
 *    current sliding window ka approximate count nikala jaata hai.
 *
 *
 * Sliding Window Log se difference:
 *
 *    Sliding Window Log:
 *       -> Har request ka timestamp store karta hai.
 *       -> Zyada accurate.
 *       -> Zyada memory use karta hai.
 *
 *    Sliding Window Counter:
 *       -> Sirf counters store karta hai.
 *       -> Memory usage bahut kam.
 *       -> Count approximate hota hai.
 *       -> Large-scale systems ke liye zyada practical ho sakta hai.
 *
 *
 * Simple words mein:
 *
 * Sliding Window Counter mein hum har request ka timestamp store nahi karte.
 * Hum previous aur current time window ke request counters maintain karte hain.
 * Jab new request aati hai, to current time ke according previous window ke
 * relevant portion aur current window ke count ko combine karke approximate
 * request count calculate karte hain.
 *
 * Agar calculated count limit ke andar hai, request allow hoti hai.
 * Agar limit exceed ho gayi hai, request block ho jaati hai.
 *
 * Isliye Sliding Window Counter ek practical balance provide karta hai:
 *
 *    Accuracy  -> Good
 *    Memory    -> Low
 *    Performance -> Good
 *
 * Yaani ye Sliding Window Log ki accuracy ke idea ko retain karta hai,
 * lekin har request ka timestamp store na karke memory usage ko reduce karta hai.
 */
```

```python
/**
 * How to Scale Fixed Window, Sliding Window Log, and Sliding Window Counter?
 *
 * Ab Token Bucket aur Leaky Bucket ke baad, humein ye samajhna hai ki
 * Fixed Window, Sliding Window Log aur Sliding Window Counter ko
 * distributed system mein kaise scale karenge.
 *
 * Basic problem same hai:
 *
 *    Client
 *       |
 *       v
 *    Load Balancer
 *      / | \
 *     v  v  v
 *   Server Server Server
 *
 * Agar har server apna alag counter ya log maintain karega,
 * to same client ka request count different servers par divide ho jayega.
 *
 * Isliye distributed Rate Limiter mein state ko shared ya
 * consistently distributed storage mein maintain karna zaroori hai.
 *
 *
 * 1. Scaling Fixed Window Counter
 *
 * Fixed Window mein humein sirf current window ka counter maintain karna hota hai.
 *
 * Example:
 *
 *    Limit = 5 requests/minute
 *
 *    User A
 *       |
 *       v
 *    Current Window
 *       |
 *       v
 *    Counter = 3
 *
 * Distributed system mein hum counter ko shared store mein rakh sakte hain.
 *
 *                         +--> Rate Limiter 1 --+
 *                         |                     |
 *    Client --> Load Balancer +--> Rate Limiter 2 --> Redis
 *                         |                     |
 *                         +--> Rate Limiter 3 --+
 *
 * Redis mein key kuch aisi ho sakti hai:
 *
 *    user:123:window:10:00 -> 3
 *
 * New request aane par:
 *
 *    -> Current window identify karo.
 *    -> Counter read/increment karo.
 *    -> Agar counter limit ke andar hai -> Allow.
 *    -> Agar limit reach ho gayi hai -> Block.
 *
 * Counter increment atomic hona chahiye.
 *
 * Example:
 *
 *    Limit = 5
 *
 *    Request 1 -> Counter 1 -> Allow
 *    Request 2 -> Counter 2 -> Allow
 *    Request 3 -> Counter 3 -> Allow
 *    Request 4 -> Counter 4 -> Allow
 *    Request 5 -> Counter 5 -> Allow
 *    Request 6 -> Limit reached -> Block
 *
 * Window expire hone ke baad old counter ko TTL ke through
 * automatically remove kiya ja sakta hai.
 *
 * Fixed Window ko scale karna relatively easy hai because
 * sirf ek small counter maintain karna hota hai.
 *
 *
 * 2. Scaling Sliding Window Log
 *
 * Sliding Window Log mein problem thodi different hai.
 *
 * Yahan hum har request ka timestamp store karte hain.
 *
 * Example:
 *
 *    User A:
 *
 *    [10:00:10,
 *     10:00:20,
 *     10:00:35,
 *     10:00:45,
 *     10:00:50]
 *
 * Distributed system mein ye timestamps shared store mein
 * maintain karne padenge.
 *
 *                         +--> Rate Limiter 1 --+
 *                         |                     |
 *    Client --> Load Balancer +--> Rate Limiter 2 --> Redis
 *                         |                     |
 *                         +--> Rate Limiter 3 --+
 *
 * Redis mein hum user ke timestamps ko efficiently maintain
 * karne ke liye sorted data structure use kar sakte hain.
 *
 * Conceptually:
 *
 *    user:123 -> [timestamp1, timestamp2, timestamp3, ...]
 *
 * Jab new request aati hai:
 *
 *    1. Current time nikalo.
 *    2. Window se bahar ke old timestamps remove karo.
 *    3. Remaining timestamps count karo.
 *    4. Agar count < limit:
 *          -> New timestamp add karo.
 *          -> Request Allow.
 *    5. Agar count >= limit:
 *          -> Request Block.
 *
 * Yahan bhi ye operations atomic hone chahiye.
 *
 * Sliding Window Log ki biggest scaling problem memory hai.
 *
 * Agar millions of users aur high traffic hai:
 *
 *    Millions of users
 *          |
 *          v
 *    Millions of timestamps
 *          |
 *          v
 *    High Memory Usage
 *
 * Isliye large-scale system mein:
 *
 *    -> Expired timestamps immediately remove karo.
 *    -> Inactive users ke data par TTL use karo.
 *    -> Users/keys ko shard karo.
 *    -> Redis Cluster jaise distributed storage use karo.
 *
 * Sliding Window Log accurate hai, lekin Fixed Window ke comparison
 * mein scale karna expensive hai because har request ka timestamp
 * maintain karna padta hai.
 *
 *
 * 3. Scaling Sliding Window Counter
 *
 * Sliding Window Counter Fixed Window aur Sliding Window Log ke
 * beech ka practical solution hai.
 *
 * Ismein har request ka timestamp store nahi karte.
 * Sirf previous aur current window ke counters maintain karte hain.
 *
 * Example:
 *
 *    Previous Window:
 *       Counter = 4
 *
 *    Current Window:
 *       Counter = 2
 *
 *    Current time ke basis par previous window ka relevant portion
 *    calculate karke approximate request count nikala jaata hai.
 *
 *
 * Distributed architecture:
 *
 *                         +--> Rate Limiter 1 --+
 *                         |                     |
 *    Client --> Load Balancer +--> Rate Limiter 2 --> Redis
 *                         |                     |
 *                         +--> Rate Limiter 3 --+
 *
 * Redis mein:
 *
 *    user:123:previous -> 4
 *    user:123:current  -> 2
 *
 * Request aane par:
 *
 *    -> Previous counter identify karo.
 *    -> Current counter identify karo.
 *    -> Current time ke according weighted count calculate karo.
 *    -> Agar count < limit:
 *          -> Current counter increment karo.
 *          -> Request Allow.
 *    -> Agar count >= limit:
 *          -> Request Block.
 *
 * Ismein sirf few counters store karne padte hain,
 * isliye memory usage Sliding Window Log se bahut kam hota hai.
 *
 *
 * Comparison:
 *
 *    Fixed Window Counter:
 *       -> Sirf current counter
 *       -> Lowest memory
 *       -> Easy to scale
 *       -> Boundary burst problem
 *
 *    Sliding Window Log:
 *       -> Har request ka timestamp
 *       -> Highest memory
 *       -> More accurate
 *       -> Large scale par expensive
 *
 *    Sliding Window Counter:
 *       -> Previous + current counters
 *       -> Low memory
 *       -> Good accuracy
 *       -> Large-scale systems ke liye practical
 *
 *
 * Common Scaling Architecture:
 *
 *                         +----------------+
 *                         | Load Balancer  |
 *                         +-------+--------+
 *                                 |
 *                   +-------------+-------------+
 *                   |             |             |
 *                   v             v             v
 *              +---------+   +---------+   +---------+
 *              | Limiter |   | Limiter |   | Limiter |
 *              |    1    |   |    2    |   |    3    |
 *              +----+----+   +----+----+   +----+----+
 *                   |             |             |
 *                   +-------------+-------------+
 *                                 |
 *                                 v
 *                        +------------------+
 *                        |   Redis Cluster  |
 *                        |                  |
 *                        | Counters / Logs  |
 *                        +------------------+
 *                                 |
 *                                 v
 *                            API Servers
 *
 *
 * Simple words mein:
 *
 * Fixed Window ko scale karna sabse easy hai because humein
 * sirf counters maintain karne hote hain.
 *
 * Sliding Window Log ko scale karna comparatively difficult hai
 * because har request ka timestamp store karna padta hai.
 * Isliye memory aur storage operations zyada hote hain.
 *
 * Sliding Window Counter ek middle ground hai.
 * Ye har request ka timestamp store nahi karta, balki few counters
 * maintain karke approximate sliding-window count calculate karta hai.
 * Isliye ye memory-efficient aur large-scale systems ke liye practical hai.
 *
 * Teeno algorithms ko distributed environment mein scale karne ke liye
 * common idea hai:
 *
 *    Multiple Rate Limiter instances
 *          +
 *    Shared / Distributed State
 *          +
 *    Atomic Updates
 *          +
 *    Sharding
 *          +
 *    TTL / Expiration
 *
 * Yaani Rate Limiter instances ko horizontally scale karo,
 * state ko shared distributed store mein rakho, concurrent updates ko
 * atomic rakho, aur large traffic ke liye state ko shard karo.
 */
```

```python
/**
 * How does Rate Limiter track individual rate limits for millions/billions of users?
 * - Ye distributed Rate Limiter ka ek very important concept hai.
 * - Maan lo hamare paas 100 million users hain aur har user ka apna rate limit hai:
 *   - User A -> 10 requests/minute
 *   - User B -> 100 requests/minute
 *   - User C -> 50 requests/minute
 *     ...
 *   - User 100M -> 20 requests/minute
 *
 * Question ye hai:
 * - "Rate Limiter ko kaise pata chalega ki kaunsi request kis user ki hai aur us 
 *    user ne kitni requests already bheji hain?"
 *
 * Step 1: Request se User ko identify karna
 * - Har request mein koi identifier hota hai jiske basis par Rate Limiter user ko 
 *   identify karta hai.
 * - Example:
 *   - Authorization: Bearer <JWT>
 *   - JWT ke andar : userId = 12345
 *   - Ya API Key ke case mein: API-Key: abc123
 *   - Ya requirement ke according:
 *     - IP Address
 *     - Device ID
 *     - API Key
 *     - User ID
 * - Rate Limiter pehle request se ek unique identifier nikalta hai.
 *
 * Step 2: Identifier ko Rate Limit Key mein convert karna
 * - Maan lo User ID = 12345 hai.
 * - Rate Limiter ek key bana sakta hai: rate:user:12345
 * - Fixed Window ke case mein: rate:user:12345:10:30
 * - Yahan "10:30" current time window ko represent karta hai.
 * - Redis mein conceptually: rate:user:12345:10:30 -> 7
 * - Matlab User 12345 ne current window mein 7 requests kar di hain.
 *
 * Step 3: Har user ke liye separate server nahi hota
 * - Ye bahut important point hai.
 * - 100 million users ka matlab ye nahi hai ki:
 *   - User A -> Server A
 *   - User B -> Server B
 *   - User C -> Server C
 * - Instead, saare users ka state ek distributed data store mein logically separate 
 *   keys ke form mein maintain hota hai.
 *   - User A -> Key A
 *   - User B -> Key B
 *   - User C -> Key C
 *     ...
 *   - User 100M -> Key 100M
 *
 * Step 4: Data ko shard kiya jaata hai
 * - Agar ek single Redis server 100 million users ka state efficiently handle nahi 
 *   kar sakta, to data multiple Redis nodes mein distribute kar diya jaata hai.
 *
 *    User ID
 *       |
 *       v
 *    Hash(userId)
 *       |
 *       +----> Redis Node 1
 *       +----> Redis Node 2
 *       +----> Redis Node 3
 *       +----> Redis Node 4
 *
 * - Example:
 *   - User 101 -> Redis 1
 *   - User 202 -> Redis 3
 *   - User 303 -> Redis 2
 *   - User 404 -> Redis 4
 * 
 * - Ye process sharding ke through automatically manage ho sakta hai.
 *
 * Step 5: Sirf active users ka state maintain karna
 * - Hum har user ka data permanently store nahi karte.
 * - Example: User ne 10 requests/minute ka limit use kiya.
 * - Jab uski window expire ho jaati hai aur user active nahi hai, to uska old 
 *   rate-limit state unnecessary hai.
 * - Isliye key par TTL/expiration laga sakte hain.
 *
 *    rate:user:12345:10:30
 *                |
 *                v
 *             Expires
 *
 * - Isse inactive users ka data automatically remove ho jaata hai.
 *
 * Step 6: Different users ke different limits
 * - Har user ka limit same hona zaroori nahi hai.
 * - Example:
 *   - Free User       : 10 requests/minute
 *   - Pro User        : 100 requests/minute
 *   - Enterprise User : 1000 requests/minute
 * - Rate Limiter request ke userId se user ka configuration retrieve karke appropriate
 *   limit apply kar sakta hai.
 *
 * Conceptually:
 *
 *    userId = 12345
 *        |
 *        v
 *    User Plan = Pro
 *        |
 *        v
 *    Limit = 100 requests/minute
 *        |
 *        v
 *    Check current counter
 *
 *
 * Step 7: Massive scale par important optimization
 * - Rate Limiter ko har request par heavy database query nahi karni chahiye.
 * - Example:
 *
 *    Request
 *       |
 *       v
 *    Rate Limiter
 *       |
 *       X
 *    PostgreSQL
 *
 * - Agar har request par PostgreSQL se: "User 12345 ne kitni requests bheji hain?"
 *   poocha jaayega, to database bottleneck ban sakta hai.
 * - Isliye rate-limit state ko low-latency in-memory distributed store mein rakhna 
 *   practical hota hai.
 *
 *
 * Final architecture:
 *
 *                       +----------------+
 *                       | Load Balancer  |
 *                       +-------+--------+
 *                               |
 *                 +-------------+-------------+
 *                 |             |             |
 *                 v             v             v
 *            Rate Limiter  Rate Limiter  Rate Limiter
 *                 |             |             |
 *                 +-------------+-------------+
 *                               |
 *                               v
 *                       +---------------+
 *                       | Redis Cluster |
 *                       +---------------+
 *                         /     |      \
 *                        /      |       \
 *                     Node 1  Node 2   Node 3
 *
 *
 * Example request:
 *
 *    User 12345
 *        |
 *        | API Request
 *        v
 *    Rate Limiter
 *        |
 *        | Extract userId = 12345
 *        |
 *        v
 *    Create Key:
 *    rate:user:12345:current-window
 *        |
 *        v
 *    Redis Cluster
 *        |
 *        | Find key's shard
 *        v
 *    Counter = 7
 *        |
 *        | Limit = 10
 *        v
 *    7 < 10
 *        |
 *        v
 *    Counter -> 8
 *        |
 *        v
 *    Request Allowed
 *
 *
 * Simple words mein:
 *
 * Rate Limiter millions of users ko individually track karta hai,
 * lekin har user ke liye separate server ya process create nahi karta.
 *
 * Har user ko ek unique key/identifier diya jaata hai.
 * Us key ke against sirf required rate-limit state, jaise counter,
 * timestamp ya tokens, store kiye jaate hain.
 *
 * Is state ko distributed in-memory store mein shard kiya jaata hai,
 * aur inactive state ko TTL ke through remove kiya jaata hai.
 *
 * Is tarah millions ya even billions of users ka rate-limit state
 * horizontally distributed infrastructure par manage kiya ja sakta hai.
 *
 * Core idea:
 *
 *    User ID
 *       |
 *       v
 *    Unique Rate-Limit Key
 *       |
 *       v
 *    Sharded Distributed Store
 *       |
 *       v
 *    Counter / Timestamp / Token State
 *
 * Yaani "individual tracking" ka matlab individual server nahi hai;
 * iska matlab hai har user ke liye ek logically separate state/key
 * maintain karna.
 */
```

```python
/**
 * Where Shall We Store Counters / Tokens?
 * - Ab humein pata hai ki Rate Limiter ko har user ke liye individual state maintain
 *   karni hai. 
 * - Next question hai: "Ye counters, tokens aur timestamps store kahan karenge?"
*/

/**
 * Maan lo User 12345 ke liye:
 * a. Token Bucket use kar rahe hain:
 * 
 *     User 12345
 *        |
 *        v
 *     Bucket State
 *        |
 *        +-- tokens = 7
 *        +-- lastRefillTime = ...
 * 
 * b. Ya Fixed Window ke case mein:
 *
 *     rate:user:12345:window
 *            |
 *            v
 *        counter = 8
 *
 * Ab is state ko store karne ke liye humare paas kuch options hain: 
 * 1. Application Server Memory
 * 2. Relational Database
 * 3. Distributed In-Memory Store
 * 4. Local Cache + Distributed Store
*/

/**
 * 1. Application Server Memory
 *
 *    Sabse simple approach ye hai ki counters, tokens ya timestamps ko
 *    directly application server ki local memory mein store kar diya jaaye.
 *
 *    Example:
 *
 *         Client
 *            |
 *            v
 *         Server
 *            |
 *            v
 *       Local Memory
 *            |
 *            +-- User A -> 5 tokens
 *            +-- User B -> 2 tokens
 *            +-- User C -> 7 tokens
 *
 *    Jab User A request bhejta hai, Rate Limiter directly server ki
 *    memory se User A ka bucket/counter check kar sakta hai.
 *
 *    Is approach ka sabse bada benefit hai speed.
 *
 *    Memory access bahut fast hota hai, isliye:
 *
 *         Request
 *            |
 *            v
 *       Rate Limiter
 *            |
 *            v
 *       Local Memory
 *
 *    State ko quickly read aur update kiya ja sakta hai.
 *
 *
 *    Lekin jaise hi hum distributed system mein multiple servers
 *    introduce karte hain, problem start hoti hai.
 *
 *         Client
 *            |
 *            v
 *       Load Balancer
 *          /     \
 *         v       v
 *     Server 1  Server 2
 *
 *
 *    Maan lo User A ki bucket mein initially 5 tokens hain.
 *
 *         Server 1
 *         User A -> 5 tokens
 *
 *    Ab User A ki next request Server 2 par chali jaati hai.
 *
 *         Server 2
 *         User A -> 5 tokens
 *
 *    Ab dono servers ke paas User A ki bucket ki separate copy hai.
 *
 *
 *         Server 1                     Server 2
 *        +----------+                 +----------+
 *        | User A   |                 | User A   |
 *        | 5 tokens |                 | 5 tokens |
 *        +----------+                 +----------+
 *
 *    Problem ye hai ki actual mein User A ke paas logically ek hi
 *    bucket hona chahiye, lekin servers apni-apni copy maintain kar rahe hain.
 *
 *    Isse Rate Limiter incorrectly allow kar sakta hai.
 *
 *    Example:
 *
 *         Server 1 -> User A -> 5 tokens
 *         Server 2 -> User A -> 5 tokens
 *
 *    User A theoretically 5 requests kar sakta tha,
 *    lekin distributed system mein wo Server 1 aur Server 2 ke
 *    different copies ka advantage lekar zyada requests bhej sakta hai.
 *
 *
 *    Challenge 1: State Synchronization
 *
 *    Multiple servers ke local memory states ko synchronize karna padega.
 *
 *         Server 1
 *             |
 *             +------+
 *                    |
 *         Server 2 ----+----> Sync State
 *                    |
 *             +------+
 *             |
 *         Server 3
 *
 *    Lekin high-scale Rate Limiter mein har request ke state ko
 *    servers ke beech continuously synchronize karna expensive
 *    aur complicated ho sakta hai.
 *
 *
 *    Challenge 2: Server Restart
 *
 *    Local memory volatile hoti hai.
 *
 *         Server
 *            |
 *            v
 *       Local Memory
 *            |
 *            X
 *       Server Restart
 *
 *    Server restart hone par memory ka rate-limit state lose ho sakta hai.
 *
 *    Example:
 *
 *         Before Restart:
 *         User A -> 8 requests
 *
 *         Server Restart
 *               |
 *               v
 *         User A -> State Lost
 *
 *    Ab Rate Limiter incorrectly samajh sakta hai ki User A ne
 *    koi request nahi bheji.
 *
 *
 *    Challenge 3: Horizontal Scaling
 *
 *    Agar traffic increase ho jaaye aur hum additional servers add karein:
 *
 *         Client
 *            |
 *            v
 *       Load Balancer
 *        /    |    \
 *       v     v     v
 *    Server  Server  Server
 *       |      |      |
 *       v      v      v
 *    Memory  Memory  Memory
 *
 *    Har server ke paas separate rate-limit state hogi.
 *
 *    Isliye servers badhane ke saath state consistency maintain karna
 *    difficult hota jaayega.
 *
 *
 *    Challenge 4: No Reliable Shared State
 *
 *    Rate Limiter ko distributed environment mein ye guarantee chahiye:
 *
 *         User A
 *            |
 *            v
 *      Same Logical State
 *            |
 *        +---+---+
 *        |       |
 *        v       v
 *     Server 1 Server 2
 *
 *    Local memory mein ye guarantee naturally nahi milti.
 *
 *
 *    Isliye local memory kab useful hai?
 *
 *    -> Single-server application.
 *    -> Development/testing.
 *    -> Small-scale system.
 *    -> Temporary caching.
 *
 *
 *    Example:
 *
 *         Client
 *            |
 *            v
 *       Single Server
 *            |
 *            v
 *       Local Memory
 *
 *    Agar sirf ek server hai, to synchronization problem nahi hai,
 *    kyunki saara rate-limit state ek hi memory mein maintain ho raha hai.
 *
 *
 *    Lekin large distributed Rate Limiter ke liye:
 *
 *         Client
 *            |
 *            v
 *       Load Balancer
 *          / | \
 *         v  v  v
 *      Server Server Server
 *          \  |  /
 *           \ | /
 *            v
 *       Shared Store
 *
 *    Shared distributed store zyada suitable hota hai, kyunki
 *    multiple Rate Limiter instances same state ko access kar sakte hain.
 *
 *
 *    Final comparison:
 *
 *         Application Server Memory
 *         ------------------------
 *         + Extremely fast
 *         + Simple implementation
 *         + No network call for state access
 *         - Not shared between servers
 *         - State synchronization problem
 *         - State lost on restart
 *         - Difficult to scale horizontally
 *
 *
 *    Simple words mein:
 *
 *    Local memory Rate Limiter ka sabse simple aur fastest option hai,
 *    lekin ye tabhi practical hai jab Rate Limiter ek single server par
 *    run kar raha ho.
 *
 *    Jaise hi application multiple servers par scale hoti hai,
 *    har server ki memory mein separate counters/tokens create ho jaate hain.
 *
 *    Isse rate-limit state inconsistent ho sakti hai.
 *
 *    Isliye large distributed systems mein counters, tokens aur timestamps
 *    ko shared distributed store mein rakhna zyada practical approach hai.
 *
 *    Core idea:
 *
 *         Single Server
 *              |
 *              v
 *         Local Memory
 *              |
 *              v
 *         Fast but not distributed
 *
 *         Multiple Servers
 *              |
 *              v
 *         Shared Distributed Store
 *              |
 *              v
 *         Consistent Rate-Limit State
 */

/**
 * 2. Relational Database
 *
 *    Counters aur tokens ko PostgreSQL ya MySQL jaise relational database
 *    mein bhi store kiya ja sakta hai.
 *
 *    Basic architecture:
 *
 *         Rate Limiter
 *              |
 *              v
 *         PostgreSQL
 *              |
 *              +-- User A -> Counter
 *              +-- User B -> Counter
 *              +-- User C -> Counter
 *
 *
 *    Maan lo User A ke liye limit hai:
 *
 *         10 requests/minute
 *
 *    Database mein hum kuch state maintain kar sakte hain:
 *
 *         User A -> currentCount = 7
 *         User B -> currentCount = 3
 *
 *    Request aane par Rate Limiter database se state read karega,
 *    limit check karega aur phir updated counter/token state database
 *    mein write karega.
 *
 *
 *    Relational Database ka benefit:
 *
 *    - Strong Consistency
 *      -> Database transactions aur consistency mechanisms provide karta hai.
 *      -> Concurrent updates ko safely handle karne ke liye
 *         transactions/locking jaise mechanisms available hote hain.
 *
 *    - Durability
 *      -> Data disk par persist hota hai.
 *      -> Database restart hone par state easily recover ki ja sakti hai.
 *
 *    - Mature Technology
 *      -> PostgreSQL/MySQL highly mature aur reliable systems hain.
 *
 *
 *    Lekin Rate Limiter ke case mein ek major problem hai:
 *
 *    Rate Limiter ko almost har API request par state read/update karni hoti hai.
 *
 *         Client Request
 *              |
 *              v
 *         Rate Limiter
 *              |
 *              +---- Read Counter
 *              |
 *              +---- Check Limit
 *              |
 *              +---- Update Counter
 *              |
 *              v
 *         API Server
 *
 *
 *    Ab maan lo application millions of requests/second handle kar rahi hai:
 *
 *         Millions of Requests
 *                |
 *                v
 *           Rate Limiter
 *                |
 *                | High-frequency
 *                | Read/Write
 *                v
 *           PostgreSQL
 *                |
 *                X
 *             Bottleneck
 *
 *
 *    Problem ye hai ki Rate Limiter ko bahut low latency chahiye.
 *    Agar har request ke liye database round trip karna pade,
 *    to latency increase hogi aur database par enormous load aa sakta hai.
 *
 *
 *    Ek aur problem hai:
 *
 *    High concurrency.
 *
 *    Maan lo same User A ki multiple requests ek hi time par aa gayi:
 *
 *         Request 1 ----+
 *         Request 2 ----+
 *         Request 3 ----+----> Rate Limiter ----> PostgreSQL
 *         Request 4 ----+
 *
 *    Sab requests same counter ko simultaneously read/update kar sakti hain.
 *
 *    Is situation mein humein:
 *
 *         -> Transactions
 *         -> Row-level locking
 *         -> Atomic updates
 *         -> Concurrency control
 *
 *    use karna pad sakta hai.
 *
 *    Ye consistency improve karta hai, lekin high traffic par
 *    database contention aur latency aur badh sakti hai.
 *
 *
 *    Example:
 *
 *         Current Counter = 9
 *         Limit = 10
 *
 *         Request A -> Read 9
 *         Request B -> Read 9
 *
 *    Agar synchronization properly handle nahi hui:
 *
 *         Request A -> 10
 *         Request B -> 10
 *
 *    Dono requests ko incorrectly allow kiya ja sakta hai.
 *
 *    Isliye database-level concurrency control important hai.
 *
 *
 *    Ab imagine karo:
 *
 *         10 Million Users
 *                |
 *                v
 *         10 Million Rate States
 *                |
 *                v
 *         PostgreSQL
 *
 *    Agar har API request ke saath ye state continuously update hogi,
 *    to relational database Rate Limiter ka bottleneck ban sakta hai.
 *
 *
 *    Isliye practical architecture mein:
 *
 *         PostgreSQL
 *              |
 *              +--> Durable business data
 *
 *         Redis / In-Memory Store
 *              |
 *              +--> High-frequency rate-limit state
 *
 *
 *    Matlab PostgreSQL ko completely useless nahi samajhna chahiye.
 *    Hum PostgreSQL mein relatively stable configuration store kar sakte hain:
 *
 *         User -> Plan -> Rate Limit
 *
 *         User A -> Free       -> 10 req/min
 *         User B -> Pro        -> 100 req/min
 *         User C -> Enterprise -> 1000 req/min
 *
 *    Lekin frequently changing state:
 *
 *         Counter
 *         Token count
 *         Last refill time
 *         Request timestamps
 *
 *    ko low-latency distributed in-memory store mein rakhna
 *    zyada practical hota hai.
 *
 *
 *    Final comparison:
 *
 *         Relational Database
 *         -------------------
 *         + Strong consistency
 *         + Durability
 *         + Transactions
 *         - Higher latency
 *         - Expensive high-frequency writes
 *         - Database contention
 *         - Difficult to scale for extremely high request rates
 *
 *
 *    Simple words mein:
 *
 *    PostgreSQL/MySQL Rate Limiter state store kar sakte hain,
 *    aur small-scale ya low-traffic systems mein ye perfectly workable
 *    bhi ho sakta hai.
 *
 *    Lekin large-scale Rate Limiter mein har request ke saath
 *    database read/write karna expensive aur slow ho sakta hai.
 *
 *    Isliye commonly stable rate-limit configuration ko relational
 *    database mein rakh sakte hain, while rapidly changing counters,
 *    tokens aur timestamps ko distributed in-memory store mein rakhna
 *    zyada suitable hota hai.
 *
 *    Core idea:
 *
 *         PostgreSQL
 *             |
 *             +--> Configuration / Durable Data
 *
 *         Redis
 *             |
 *             +--> Counter / Token / Timestamp State
 *
 *    Yaani relational database consistency aur durability ke liye strong hai,
 *    lekin ultra-high-frequency Rate Limiter state ke liye low-latency
 *    in-memory store generally better fit hota hai.
 */

/**
 * 3. Distributed In-Memory Store
 *
 * Large-scale distributed Rate Limiter mein counters, tokens aur timestamps
 * ko server ki local memory mein rakhna practical nahi hai, kyunki requests
 * multiple Rate Limiter instances par distribute ho sakti hain.
 *
 * Isliye hum ek shared, fast aur distributed in-memory datastore use kar sakte hain.
 *
 * Example: Redis
 *
 *                         +----------------+
 *                         | Rate Limimiter |
 *                         +-------+--------+
 *                                 |
 *                                 v
 *                         +---------------+
 *                         | Redis Cluster |
 *                         +-------+-------+
 *                              /  |  \
 *                             /   |   \
 *                            v    v    v
 *                         Node 1 Node 2 Node 3
 *
 *
 * Redis ka kaam:
 *
 *    -> Rate-limit state ko centrally/shared location par store karna.
 *    -> Multiple Rate Limiter instances ko same state access karne dena.
 *    -> Counters, tokens aur timestamps ko very low latency mein
 *       read/update karna.
 *
 *
 * Example 1: Fixed Window Counter
 *
 *    Key:
 *       rate:user:12345:window:10:30
 *
 *    Value:
 *       8
 *
 *    Matlab:
 *       User 12345 ne current window mein 8 requests kar di hain.
 *
 *
 * Example 2: Token Bucket
 *
 *    Key:
 *       bucket:user:12345
 *
 *    State:
 *       +-- tokens = 7
 *       +-- lastRefillTime = ...
 *       +-- capacity = 10
 *       +-- refillRate = 2/sec
 *
 *
 * Example 3: Sliding Window Log
 *
 *    Key:
 *       window:user:12345
 *
 *    Timestamps:
 *       +-- timestamp 1
 *       +-- timestamp 2
 *       +-- timestamp 3
 *       +-- timestamp 4
 *
 *
 * Lekin distributed Rate Limiter mein sirf data store karna enough nahi hai.
 * Humein concurrency, race conditions aur synchronization ko bhi handle karna hoga.
 *
 *
 * Challenge 1: Race Condition
 *
 * Maan lo User 123 ke bucket mein sirf 1 token available hai.
 * Same time par 2 requests aa jaati hain:
 *
 *    Request A --------+
 *                      |
 *    Request B --------+----> Rate Limiter
 *
 * Dono requests bucket ko check karti hain:
 *
 *    Request A -> "1 token available"
 *    Request B -> "1 token available"
 *
 * Agar dono pehle read karein aur baad mein update karein:
 *
 *    Request A -> Read 1
 *    Request B -> Read 1
 *    Request A -> Consume -> 0
 *    Request B -> Consume -> 0
 *
 * Dono requests allow ho sakti hain.
 *
 * Lekin actually sirf 1 token available tha.
 *
 * Ye Race Condition hai.
 *
 *
 * Challenge 2: Check + Update ke beech Race Condition
 *
 * Rate Limiter ka actual operation logically kuch aisa hai:
 *
 *    1. Current state read karo.
 *    2. Refill / counter calculate karo.
 *    3. Limit check karo.
 *    4. State update karo.
 *    5. Request allow/reject karo.
 *
 * Problem ye hai ki agar multiple requests same time par ye steps
 * independently perform karein, to dono same old state dekh sakti hain.
 *
 * Isliye:
 *
 *    Check + Update
 *
 * ko ek atomic operation ki tarah execute karna important hai.
 *
 *
 * Challenge 3: Atomic Operations
 *
 * Humein ensure karna hai ki:
 *
 *    Check Limit
 *         +
 *    Update Counter/Token
 *
 * ke beech koi doosri request state ko incorrectly modify na kar sake.
 *
 * Conceptually:
 *
 *    Before:
 *       tokens = 1
 *
 *    Atomic Operation:
 *       Check -> Consume -> Update
 *
 *    After:
 *       tokens = 0
 *
 * Is operation ke beech doosri request ko inconsistent state nahi milni chahiye.
 *
 * Redis atomic commands aur Lua Scripts jaise mechanisms use karke
 * multiple operations ko atomic execution mein combine kiya ja sakta hai.
 *
 *
 * Challenge 4: Synchronization Between Rate Limiter Servers
 *
 * Maan lo hamare paas multiple Rate Limiter instances hain:
 *
 *              +--> Rate Limiter 1 --+
 *              |                    |
 *    Client -->+--> Rate Limiter 2 --+--> Redis
 *              |                    |
 *              +--> Rate Limiter 3 --+
 *
 * Same user ki requests kisi bhi Rate Limiter instance par aa sakti hain.
 *
 * Isliye Rate Limiter 1 ke paas jo state hai aur Rate Limiter 2 ke paas
 * jo state hai, unhe independently maintain nahi karna chahiye.
 *
 * Shared Redis state ke through:
 *
 *    Rate Limiter 1 ----+
 *    Rate Limiter 2 ----+----> Same User State
 *    Rate Limiter 3 ----+
 *
 * Isse sabhi Rate Limiter instances ko same logical state milti hai.
 *
 *
 * Challenge 5: High Concurrency
 *
 * Large-scale system mein ek hi user ke liye bhi multiple requests
 * simultaneously aa sakti hain.
 *
 *    User 123
 *       |
 *       +--> Request 1
 *       +--> Request 2
 *       +--> Request 3
 *       +--> Request 4
 *
 * Agar har request state ko independently modify karegi,
 * to incorrect count ya token calculation ho sakta hai.
 *
 * Isliye Rate Limiter state update ko atomic aur concurrency-safe
 * banana important hai.
 *
 *
 * Challenge 6: Redis Failure
 *
 * Ab imagine karo Redis temporarily unavailable ho gaya:
 *
 *    Rate Limiter
 *         |
 *         X
 *    Redis Unavailable
 *
 * Ab Rate Limiter ko decide karna hoga ki requests ka kya karna hai.
 *
 *    Fail-Open:
 *       -> Redis unavailable hai
 *       -> Request allow karo.
 *
 *       Benefit:
 *       -> Application available rahegi.
 *
 *       Risk:
 *       -> Rate Limiting temporarily bypass ho sakti hai.
 *
 *
 *    Fail-Closed:
 *       -> Redis unavailable hai
 *       -> Request reject karo.
 *
 *       Benefit:
 *       -> Rate limit/security maintain rahegi.
 *
 *       Risk:
 *       -> Legitimate users bhi temporarily blocked ho sakte hain.
 *
 * Ye decision API ke use case aur security requirements par depend karta hai.
 *
 *
 * Challenge 7: Large Scale Synchronization
 *
 * Agar Redis Cluster mein data multiple nodes par distributed hai:
 *
 *       Redis Cluster
 *        /    |    \
 *       v     v     v
 *     Node 1 Node 2 Node 3
 *
 * To user ki rate-limit state ek appropriate shard/node par store hogi.
 *
 * Example:
 *
 *    User 12345
 *        |
 *        v
 *    Hash(userId)
 *        |
 *        v
 *    Redis Node 2
 *
 * Isse state horizontally distribute ho sakti hai aur ek single
 * Redis node bottleneck nahi banta.
 *
 *
 * Final Idea:
 *
 *    Distributed Rate Limiter
 *             |
 *             v
 *       Multiple Instances
 *             |
 *             v
 *        Redis Cluster
 *             |
 *       +-----+-----+
 *       |           |
 *    Shared      Atomic
 *    State       Updates
 *
 *
 * Simple words mein:
 *
 * Redis ka use sirf counters/tokens store karne ke liye nahi hai.
 * Distributed Rate Limiter mein sabse important problem ye ensure karna
 * hai ki multiple servers aur multiple concurrent requests same rate-limit
 * state ko safely update karein.
 *
 * Isliye humein:
 *
 *    -> Shared state
 *    -> Atomic operations
 *    -> Race-condition handling
 *    -> Concurrency control
 *    -> Multiple Rate Limiter instances ke beech synchronization
 *    -> Redis failure handling
 *    -> Sharding for horizontal scaling
 *
 * ka dhyaan rakhna padta hai.
 *
 * Core principle:
 *
 *    "Read + Decide + Update"
 *
 * ko ek safe, atomic aur concurrency-aware operation banana
 * distributed Rate Limiter ke design ka most important part hai.
 */

/**
 * 4. Local Cache + Distributed Store
 *
 *    Ab ek aur practical approach dekhte hain:
 *    Local Cache + Distributed Store.
 *
 *    Large-scale system mein hum har request ke liye har cheez ko
 *    Redis se fetch nahi karna chahte, kyunki unnecessary network
 *    calls latency aur Redis ka load dono increase kar sakte hain.
 *
 *    Isliye hum stable aur frequently accessed data ko local cache
 *    mein rakh sakte hain, while actual shared rate-limit state ko
 *    distributed store mein maintain karte hain.
 *
 *
 *    Basic Architecture:
 *
 *                         +----------------+
 *                         | Rate Limiter   |
 *                         +-------+--------+
 *                                 |
 *                    +------------+------------+
 *                    |                         |
 *                    v                         v
 *              +-----------+             +-----------+
 *              | Local     |             |   Redis   |
 *              | Cache     |             |  Cluster  |
 *              +-----------+             +-----------+
 *
 *
 *    Maan lo application mein users ke different plans hain:
 *
 *         User A -> Free       -> 10 requests/minute
 *         User B -> Pro        -> 100 requests/minute
 *         User C -> Enterprise -> 1000 requests/minute
 *
 *    User ka plan aur uska allowed rate-limit configuration relatively
 *    stable data hai.
 *
 *    Is information ko local cache mein rakha ja sakta hai:
 *
 *         Local Cache
 *              |
 *              +-- User A -> Free -> 10 req/min
 *              +-- User B -> Pro  -> 100 req/min
 *              +-- User C -> Enterprise -> 1000 req/min
 *
 *    Ab har request par Rate Limiter ko User Plan ke liye
 *    Redis/database ko query karne ki zarurat nahi padegi.
 *
 *
 *    Lekin actual dynamic rate-limit state:
 *
 *         Counter
 *         Token Count
 *         Last Refill Time
 *         Request Timestamps
 *
 *    ko shared distributed store mein rakhna safer hota hai.
 *
 *         Redis
 *           |
 *           +-- User A -> counter = 7
 *           +-- User B -> tokens = 43
 *           +-- User C -> window state = ...
 *
 *
 *    Example:
 *
 *    User A request bhejta hai:
 *
 *         Client
 *            |
 *            v
 *       Rate Limiter
 *            |
 *            +----> Local Cache
 *            |          |
 *            |          +-- User A -> Free
 *            |                    -> Limit = 10/min
 *            |
 *            +----> Redis
 *                       |
 *                       +-- User A -> Counter = 7
 *
 *    Rate Limiter ko pehle local cache se user ka configuration mil jaata hai.
 *
 *    Phir Redis se actual current rate-limit state milti hai.
 *
 *    Rate Limiter decide karta hai:
 *
 *         Current Count = 7
 *         Limit         = 10
 *
 *         7 < 10
 *            |
 *            v
 *         Request Allow
 *
 *         Counter -> 8
 *
 *
 *    Is approach ka main benefit hai:
 *
 *    1. Lower Latency
 *       -> Local cache memory mein hota hai.
 *       -> Configuration read karne ke liye network call avoid ho sakti hai.
 *
 *    2. Reduced Redis Load
 *       -> Frequently accessed static/semi-static configuration
 *          baar-baar Redis se fetch nahi karni padti.
 *
 *    3. Distributed Consistency
 *       -> Actual shared rate-limit state Redis jaise distributed
 *          store mein maintain hoti hai.
 *
 *    4. Better Scalability
 *       -> Multiple Rate Limiter instances apna local cache maintain
 *          kar sakte hain, while shared dynamic state centralized
 *          distributed store mein hoti hai.
 *
 *
 *    Lekin ek important challenge hai:
 *    Local Cache mein data stale ho sakta hai.
 *
 *    Example:
 *
 *         Redis / Database:
 *         User A -> Limit = 100 req/min
 *
 *         Local Cache:
 *         User A -> Limit = 100 req/min
 *
 *    Ab administrator User A ka plan change karke:
 *
 *         User A -> Limit = 50 req/min
 *
 *    kar deta hai.
 *
 *    Agar local cache update nahi hui, to Rate Limiter temporarily
 *    old limit = 100 use kar sakta hai.
 *
 *    Is problem ko Stale Cache kehte hain.
 *
 *
 *    Isliye local cache ke saath:
 *
 *         -> TTL
 *         -> Cache expiration
 *         -> Cache invalidation
 *         -> Refresh mechanism
 *
 *    use kiya ja sakta hai.
 *
 *    Example:
 *
 *         Local Cache
 *              |
 *              | TTL = 1 minute
 *              v
 *         Configuration expires
 *              |
 *              v
 *         Fetch latest configuration
 *
 *
 *    Ek aur important distinction:
 *
 *         Configuration
 *         -------------
 *         User Plan
 *         Rate Limit
 *         API Policy
 *
 *         -> Local Cache mein rakh sakte hain.
 *
 *
 *         Dynamic State
 *         -------------
 *         Current Counter
 *         Token Count
 *         Last Refill Time
 *         Request Timestamps
 *
 *         -> Shared Distributed Store mein rakhna better hai.
 *
 *
 *    Final Architecture:
 *
 *                              +----------------+
 *                              |     Client     |
 *                              +-------+--------+
 *                                      |
 *                                      v
 *                              +---------------+
 *                              | Rate Limiter  |
 *                              +-------+-------+
 *                                      |
 *                         +------------+------------+
 *                         |                         |
 *                         v                         v
 *                  +-------------+          +---------------+
 *                  | Local Cache |          | Redis Cluster |
 *                  +-------------+          +---------------+
 *                         |                         |
 *                         |                         |
 *                    Configuration             Dynamic State
 *                         |                         |
 *                         +------------+------------+
 *                                      |
 *                                      v
 *                                Rate Decision
 *                                      |
 *                         +------------+------------+
 *                         |                         |
 *                       Allow                     Block
 *                         |                         |
 *                         v                         v
 *                    API Server                  429
 *
 *
 *    Simple words mein:
 *
 *    Local Cache ko hum speed ke liye use karte hain aur
 *    Distributed Store ko shared consistency ke liye.
 *
 *    Local Cache mein relatively stable information jaise
 *    User Plan aur Rate Limit configuration rakh sakte hain.
 *
 *    Redis jaise shared store mein frequently changing state,
 *    jaise counters, tokens aur timestamps rakhte hain.
 *
 *    Isse humein dono benefits milte hain:
 *
 *         Local Cache       -> Fast Access
 *         Redis Cluster     -> Shared State
 *
 *    Lekin local cache stale na ho, isliye TTL aur cache
 *    invalidation/refresh strategy ka dhyaan rakhna zaroori hai.
 *
 *    Core idea:
 *
 *         Stable Configuration
 *                 |
 *                 v
 *            Local Cache
 *                 |
 *                 +--> Low Latency
 *
 *         Dynamic Rate State
 *                 |
 *                 v
 *            Redis Cluster
 *                 |
 *                 +--> Shared + Consistent State
 */


/**
 * Different Algorithms ke liye Storage:
 * 1. Fixed Window Counter
 *    -> Counter + Expiration
 *    -> Very small state
 * 2. Sliding Window Log
 *    -> Request timestamps
 *    -> More memory required
 * 3. Sliding Window Counter
 *    -> Previous + Current counters
 *    -> Small state
 * 4. Token Bucket
 *    -> Token count
 *    -> Last refill timestamp
 *    -> Small state
 * 5. Leaky Bucket
 *    -> Queue / pending request state
 *    -> Queue size aur processing state
 * 
 * Note: Lekin Rate Limiter ke high-frequency counters/tokens ko generally durable 
 *       database mein store karna avoid karte hain, kyunki humein extremely low latency
 *       aur high throughput chahiye.
*/

/**
 * Practical Choice:
 * 1. Small / Single Server: 
 *    - Local Memory
 * 2. Distributed / Large Scale:
 *    - Distributed In-Memory Store
 *    - Example: Redis
 * 3. Durable business data:
 *    - PostgreSQL / MySQL
*/

/**
 * Final Architecture:
 *
 *                         +----------------+
 *                         | Load Balancer  |
 *                         +-------+--------+
 *                                 |
 *                    +------------+------------+
 *                    |            |            |
 *                    v            v            v
 *               Rate Limiter  Rate Limiter  Rate Limiter
 *                    |            |            |
 *                    +------------+------------+
 *                                 |
 *                                 v
 *                         +---------------+
 *                         | Redis Cluster |
 *                         +---------------+
 *                           /     |      \
 *                          v      v       v
 *                       Node 1  Node 2   Node 3
*/

/**
 * Simple words mein:
 * - Rate Limiter ko bahut fast read/write karna padta hai, isliye counters aur 
 *   tokens ko low-latency in-memory distributed store mein rakhna practical hai.
 * - Single server ke simple system mein local memory kaafi ho sakti hai, lekin 
 *   distributed large-scale system mein shared distributed store use karna better hai.
 *
 * Aur ek important point:
 * - Rate Limiter ke counters/tokens ko permanently preserve karna usually zaroori nahi
 *   hota. Ye temporary state hai.
 * - Isliye expiration/TTL use karke old aur inactive state ko automatically clean kiya
 *   ja sakta hai.
 *
 * Core idea:
 * - Request
 *     |
 *     v
 * - Rate Limiter
 *     |
 *     v
 * - Fast Shared State
 *     |
 *     +--> Counter
 *     +--> Token
 *     +--> Timestamp
 *     +--> Window State
 *
 * - Yaani Rate Limiter ka actual "memory" distributed in-memory store mein hota hai, 
 *   jahan multiple Rate Limiter servers same state ko safely access kar sakte hain.
 */
```

```python
/**
 * Step 3: Rate Limiter - Overall Design
 *
 * Ab hum Rate Limiter ka complete high-level design ek single flow mein
 * samajhte hain. Is architecture ka main goal hai ki Rate Limiter
 * multiple servers par scale ho sake aur sabhi servers ek shared
 * rate-limit state ko safely maintain kar saken.
 *
 *
 *                         +----------------+
 *                         |     Client     |
 *                         +-------+--------+
 *                                 |
 *                                 | HTTP Request
 *                                 v
 *                         +----------------+
 *                         | Load Balancer  |
 *                         +-------+--------+
 *                                 |
 *                    +------------+------------+
 *                    |            |            |
 *                    v            v            v
 *              +---------+  +---------+  +---------+
 *              |  Rate   |  |  Rate   |  |  Rate   |
 *              | Limiter |  | Limiter |  | Limiter |
 *              |    1    |  |    2    |  |    3    |
 *              +----+----+  +----+----+  +----+----+
 *                   |            |            |
 *                   +------------+------------+
 *                                |
 *                                | Read / Update
 *                                v
 *                       +-------------------+
 *                       |   Redis Cluster   |
 *                       |                   |
 *                       | Counters / Tokens |
 *                       | Timestamps / State|
 *                       +---------+---------+
 *                                 |
 *                                 | Request Allowed
 *                                 v
 *                         +---------------+
 *                         |  API Servers  |
 *                         +-------+-------+
 *                                 |
 *                                 v
 *                            Application
 *
 *
 * Ab ek request ko step-by-step follow karte hain.
 *
 * Step 1:
 *
 * Client API request bhejta hai.
 *
 *    Client
 *       |
 *       | HTTP Request
 *       v
 *    Load Balancer
 *
 *
 * Step 2:
 *
 * Load Balancer request ko available Rate Limiter instances mein se
 * kisi ek instance par forward karta hai.
 *
 *    Load Balancer
 *       |
 *       +----> Rate Limiter 1
 *       |
 *       +----> Rate Limiter 2
 *       |
 *       +----> Rate Limiter 3
 *
 * Important:
 *    Request kisi bhi Rate Limiter instance par ja sakti hai.
 *    Isliye Rate Limiter state local memory mein nahi, balki
 *    shared distributed store mein maintain karna important hai.
 *
 *
 * Step 3:
 *
 * Rate Limiter request se client ko identify karta hai.
 *
 *    Request
 *       |
 *       +--> User ID
 *       +--> API Key
 *       +--> IP Address
 *       +--> Device ID
 *
 * Example:
 *
 *    User ID = 12345
 *
 * Ab Rate Limiter ko pata hai ki kis user ka rate limit check karna hai.
 *
 *
 * Step 4:
 *
 * Rate Limiter Redis Cluster se User 12345 ka current
 * rate-limit state retrieve karta hai.
 *
 *    Rate Limiter
 *         |
 *         v
 *    Redis Cluster
 *         |
 *         v
 *    User 12345 State
 *
 * State algorithm ke according different ho sakti hai.
 *
 *
 * Token Bucket:
 *
 *    User 12345
 *       |
 *       +-- tokens = 7
 *       +-- capacity = 10
 *       +-- refillRate = 2/sec
 *       +-- lastRefillTime = ...
 *
 *
 * Fixed Window:
 *
 *    User 12345
 *       |
 *       +-- window = 10:30
 *       +-- counter = 8
 *
 *
 * Sliding Window Log:
 *
 *    User 12345
 *       |
 *       +-- timestamp 1
 *       +-- timestamp 2
 *       +-- timestamp 3
 *
 *
 * Sliding Window Counter:
 *
 *    User 12345
 *       |
 *       +-- previousWindow = 4
 *       +-- currentWindow = 2
 *
 *
 * Step 5:
 *
 * Rate Limiter selected algorithm ke according request ko check karta hai.
 *
 *    Token Bucket
 *       -> Tokens refill/check
 *
 *    Fixed Window
 *       -> Current counter check
 *
 *    Sliding Window Log
 *       -> Recent timestamps check
 *
 *    Sliding Window Counter
 *       -> Previous + Current counter se estimated count
 *
 *
 * Step 6:
 *
 * Ab Rate Limiter decide karta hai:
 *
 *    "Kya request allowed hai?"
 *
 *                 +----------------+
 *                 |   Rate Limiter |
 *                 +-------+--------+
 *                         |
 *                    Check Limit
 *                         |
 *              +----------+----------+
 *              |                     |
 *              v                     v
 *        Limit Available        Limit Exceeded
 *              |                     |
 *              v                     v
 *        Update State            Reject
 *              |                     |
 *              v                     v
 *         API Server             429 Error
 *
 *
 * Agar request allowed hai:
 *
 *    Rate Limiter
 *         |
 *         v
 *    Update Counter/Token
 *         |
 *         v
 *    API Server
 *         |
 *         v
 *    Application
 *
 *
 * Agar request limit exceed kar chuki hai:
 *
 *    Rate Limiter
 *         |
 *         X
 *    Request Rejected
 *         |
 *         v
 *    HTTP 429 Too Many Requests
 *
 *
 * Step 7:
 *
 * State update atomic hona chahiye.
 *
 * Maan lo User ke paas sirf 1 token hai aur same time par
 * 2 requests aa jaati hain:
 *
 *    Request A ----+
 *                  |
 *    Request B ----+----> Rate Limiter
 *
 * Dono requests ko same time par token nahi milna chahiye.
 *
 * Isliye:
 *
 *    Check
 *      +
 *    Update
 *
 * ko atomic operation ki tarah execute karna important hai.
 *
 * Isse race condition aur incorrect counters/tokens se bach sakte hain.
 *
 *
 * Step 8:
 *
 * Ab imagine karo system mein millions of users hain.
 *
 * Hum har user ke liye separate server nahi banayenge.
 * Instead, har user ka rate-limit state ek unique key ke through
 * Redis Cluster mein logically maintain hoga.
 *
 *    User 101 -> Redis Node 1
 *    User 202 -> Redis Node 2
 *    User 303 -> Redis Node 3
 *
 * Isse rate-limit state horizontally distribute ho sakti hai.
 *
 *
 * Step 9:
 *
 * Old/inactive rate-limit state ko permanently store karne ki zarurat nahi hai.
 *
 * Isliye counters, timestamps aur temporary state par TTL/expiration
 * use kiya ja sakta hai.
 *
 *    rate:user:12345:window
 *             |
 *             | TTL expires
 *             v
 *        State Removed
 *
 * Isse Redis memory continuously grow nahi karegi.
 *
 *
 * Complete Request Flow:
 *
 *    Client
 *       |
 *       v
 *    Load Balancer
 *       |
 *       v
 *    Rate Limiter Instance
 *       |
 *       +--> Identify User
 *       |
 *       +--> Read Rate State
 *       |
 *       v
 *    Redis Cluster
 *       |
 *       +--> Check / Calculate Limit
 *       |
 *       +--> Atomic State Update
 *       |
 *       +-------------------------+
 *                                 |
 *                       +---------+---------+
 *                       |                   |
 *                     ALLOW                BLOCK
 *                       |                   |
 *                       v                   v
 *                  API Server         HTTP 429 Error
 *                       |
 *                       v
 *                  Application
 *
 *
 * Important Design Points:
 *
 *    -> Rate Limiter horizontally scalable hai.
 *
 *    -> Multiple Rate Limiter instances simultaneously run kar sakte hain.
 *
 *    -> Request kisi bhi Rate Limiter instance par aa sakti hai.
 *
 *    -> Shared rate-limit state Redis Cluster mein maintain hoti hai.
 *
 *    -> Counter/token/timestamp updates atomic hone chahiye.
 *
 *    -> Race conditions aur concurrent requests ko safely handle karna hai.
 *
 *    -> Redis state ko shard karke large traffic distribute kiya ja sakta hai.
 *
 *    -> Inactive rate-limit entries ko TTL/expiration ke through remove
 *       kiya ja sakta hai.
 *
 *    -> Rate Limiter API Server tak request pahunchne se pehle request
 *       ko check karta hai.
 *
 *    -> Allowed request hi API Server tak jaati hai.
 *
 *    -> Blocked request ko HTTP 429 Too Many Requests response milta hai.
 *
 *
 * Simple words mein:
 *
 *    Client
 *       |
 *       v
 *    Load Balancer
 *       |
 *       v
 *    Rate Limiter
 *       |
 *       v
 *    "Is request ko allow karna hai?"
 *       |
 *       +---- YES ----> Atomic State Update ---> API Server
 *       |
 *       +---- NO -----> HTTP 429
 *
 *
 * Aur agar system distributed hai:
 *
 *    Multiple Rate Limiters
 *            |
 *            v
 *       Redis Cluster
 *            |
 *            v
 *       Shared Rate State
 *
 * Yaani overall design ka core idea hai:
 *
 *    "Multiple Rate Limiter instances + Shared Distributed State
 *     + Atomic Updates + Horizontally Scalable Infrastructure"
 *
 * Isi architecture ke andar hum Token Bucket, Leaky Bucket,
 * Fixed Window, Sliding Window Log aur Sliding Window Counter
 * jaise different Rate Limiting algorithms implement kar sakte hain.
 */
```
