```python
/**
 * > What is Server?
 *   Server ek program (software) hota hai jo machine (computer) par run karta hai.
 *   Iska kaam hota hai user se aayi request ko receive karna, usse code tak bhejna,
 *   aur fir response wapas user ko dena.
 *
 * > Let's say we have a machine jiske andar ek server run kr rha hai,
 *   aur ye server ek special program hai jiska kaam hai incoming request ko
 *   process karna, isliye isse hum server bulaate hai.
 *
 *   +---------------+
 *   |               |
 *   | Server (HTTP) |---------> Code (JS/Java/Python/Go)
 *   |               |
 *   +---------------+
 *
 * > Yaha Server ek middleman ki tarah kaam karta hai:
 *   User ---> Server ---> Code ---> Server ---> User
 *
 * > Let's say user ne authentication (Login/Signup) karna hai.
 *   To kya humara ye codebase authentication handle kr skta hai? Yes!
 */
```

```python
/**
 * Design Social Media App like Twitter (Monolith):
 *
 * > Sabse pehle hum ek machine lete hain jiske andar ek server run ho raha hai,
 *   aur humein apna codebase likhna hai jo user ki requests ko handle kare.
 *   Server ka kaam hai request lena, code tak bhejna, aur response wapas dena.
 *
 * > Sabse pehla feature jo hum banate hain wo hai authentication (Login/Signup).
 *   Iske liye hum JS/Java/Python kisi bhi language m code likh sakte hain. Ab user
 *   apna account bana sakta hai aur login kar sakta hai.
 *
 * > Next step m humein User Profile feature chahiye jisme user apni details set
 *   kar sake (photo, bio, etc.). To humne Profile/User Setup ka code likha aur
 *   GitHub par push kar diya.
 *
 * > Ab tak humare system m 2 features aa chuke hain:
 *   1. Authentication
 *   2. User Profile
 *
 * > Iske baad humne ek aur important feature add kiya jise hum "FeedCode" bol rahe hain.
 *   Isme user tweets post kar sakta hai aur dusre users ke tweets dekh sakta hai.
 *   Saath hi humne ek recommendation logic bhi likha jo decide karta hai ki user ko
 *   kaunse tweets dikhne chahiye.
 *
 * > Jaise jaise system grow karta gaya, humne aur features add kiye jaise:
 *   likes/comments (posts par interaction), chat/DM (users ke beech messaging),
 *   follow/unfollow (users connect kar sake), aur spaces (live/audio features).
 *
 * > Dhyaan dene wali baat ye hai ki ye saara code ek hi jagah (single repository)
 *   m likha gaya hai. Is pure system ko hum build karke server par deploy kar dete hain
 *   jahan ye run karta hai. Is type ke system ko hum MONOLITH bolte hain jahan sab
 *   features ek hi codebase m hote hain.
 *
 * > Lekin jaise jaise users badhne lagte hain, system par load bhi badhne lagta hai.
 *   Ab humein system ko scale karna padta hai taaki wo zyada users handle kar sake.
 *
 * > Scaling do tarah ki hoti hai:
 *
 *   a. Vertical Scaling:
 *   > Ek hi machine ko aur powerful bana dete hain (zyada RAM, better CPU).
 *     Lekin iski ek limit hoti hai aur ye mehenga pad sakta hai.
 *
 *   b. Horizontal Scaling:
 *   > Ek machine ke bajaye multiple machines use karte hain aur load ko distribute
 *     kar dete hain. Isse system zyada users ko efficiently handle kar pata hai.
 * 
 * > Ab maan lo humne horizontal scaling choose kiya. Iska matlab hai ek hi machine
 *   ko bada karne ke bajaye hum multiple servers use karenge. In sab servers ko hum
 *   ek group me daal dete hain jise ASG (Auto Scaling Group) bolte hain.
 *
 * > ASG ke andar jitne bhi servers hain un sabke andar exact same codebase hota hai.
 *   Matlab har server same tarah se request handle kar sakta hai.
 *
 * > In sab servers ke aage hum ek Load Balancer lagate hain. Load Balancer ka kaam
 *   hota hai incoming requests ko alag-alag servers me distribute karna taaki
 *   kisi ek server par zyada load na aaye.
 *
 *                         +-----------------------------+
 *                         |           Database          |
 *                         +-----------------------------+
 *
 *              +---------------------------------------------------+
 *              | +------+  +------+  +------+  +------+  +------+  |
 *              | |      |  |      |  |      |  |      |  |      |  |
 *              | |      |  |      |  |      |  |      |  |      |  | ASG
 *              | |      |  |      |  |      |  |      |  |      |  |
 *              | |      |  |      |  |      |  |      |  |      |  | 
 *              | +------+  +------+  +------+  +------+  +------+  |
 *              |  Server1   Server2   Server3   Server4   Server5  |
 *              +---------------------------------------------------+
 *              
 *                         +-----------------------------+
 *                         |        Load Balancer        |
 *                         +-----------------------------+
 *
 * > Ab jab bahut saare users system ko use karte hain:
 *   - Sab requests pehle Load Balancer tak aati hain
 *   - Load Balancer unhe alag-alag servers me baant deta hai
 *
 * > Example:
 *   - User-1 ki authentication Server-1 handle karega
 *   - User-2 ki authentication Server-2 handle karega
 *   - User-3 ki request Server-3 handle karega
 *
 * > Is tarah system automatically scale karta hai kyunki load distribute ho raha hai.
 *
 * > Important baat:
 *   - Har server same code run kar raha hai
 *   - Saara code ek hi repository me hai
 *   - Ek hi jagah se build aur deploy hota hai
 *
 * > Isi architecture ko hum MONOLITH bolte hain:
 *   - Single codebase
 *   - Single deployable unit
 *
 * > Database bhi ek hi hota hai jo sab servers use karte hain.
 *   Is database ko hum aur optimize kar sakte hain using:
 *   - Sharding (data ko parts me todna)
 *   - Replication (multiple copies banana for reliability)
 */
```

```python
/**
 * Problem with Monolith Architecture:
 * > Users complain kr rhe hai ki:
 *   - Excess memory consumption
 *   - Slow response
 *   - Kabhi kabhi respond hi nahi karta
 * > Ab dekho din m signup zyada hote hai ya tweet? → Tweet!
 *   Matlab authentication ka code sirf ~10% use hota hai, jabki tweet ~90% use hota hai.
 *
 * > Problem ye hai ki hum monolith architecture pe hai jahan saara code ek hi jagah hai,
 *   to jaise hi load badhta hai, poora system ek saath scale ho jata hai.
 * > Iska matlab:
 *   - Tweet zyada ho raha hai to bhi authentication/DM/spaces sab scale honge
 *   - Jo feature use nahi ho raha uspe bhi resources waste honge
 *
 * Real issue:
 *   - Agar log bahut tweet kar rahe hain → signup nahi ho paega
 *   - Agar bahut signup ho rahe hain     → tweet slow ho jayega
 * > Kyunki server ek hi code run kar raha hai aur sab features same resources share karte hain
 * > Agar DDoS attack ho jaye (bahut saare fake signups): Poora system slow ho jayega
 * > Humne monolith isliye choose kiya tha:
 *   - Easy to maintain
 *   - Single deployment
 *   - Cheaper
 * > Lekin grow hone ke baad problems:
 *   - Large codebase
 *   - Feature level scaling issue
 *   - One bug → poora server crash
*/
```

```python
/**
 * Solution: Microservices
 * > Ab hum alag alag repositories bana dete hain:
 *   - AuthService: sirf authentication (Signin/Signup/ForgetPassword)
 *   - TweetService: sirf tweet related kaam (Create/Update/Delete/Like/Comment)
 *   - ChatService: messages bhejna
 *   - SpacesService: voice transfer (realtime)
 * 
 * > Ab codebase alag ho gaya hai to teams ko bhi easily divide kar sakte hain:
 *   - Jo auth pe kaam karega → sirf AuthService
 *   - Jo tweet pe kaam karega → sirf TweetService
 *
 * > Ab hum har service ko alag se deploy aur scale kar sakte hain:
 *   1. AuthService (kam use hota hai):
 *      - AuthService-1
 *      - AuthService-2
 *   2. TweetService (zyada use hota hai):
 *      - TweetService-1
 *      - TweetService-2
 *      - TweetService-3
 *      - TweetService-4
 *   3. ChatService:
 *      - ChatService-1
 *      - ChatService-2
 *   4. SpacesService (realtime, stateful):
 *      - SpacesService (easily horizontally scale nahi hota)
 *
 * > Ab fayda ye hai ki:
 *   - Har service apne hisaab se scale ho rahi hai
 *   - Jo feature zyada use ho raha hai usko zyada servers mil rahe hain
 *   - Sab kuch independent ho gaya hai (alag codebases)
*/
```

```python
/**
 * Problem with Microservices: 
 * > User ko kaise pta chalega ki usse kis service pe jaana hai?
 * 
 * Solution: Load Balancer
 * > Har service ke aage ek alag Load Balancer lagta hai jo us service ko expose karta hai.
 *
 *   +--------------------+   +---------------------+   +--------------------+   +----------------------+
 *   |   Load Balancer    |   |  Load Balancer      |   |  Load Balancer     |   |  Load Balancer       |
 *   |(auth.chaitweet.com)|   |(tweet.chaitweet.com)|   |(chat.chaitweet.com)|   |(spaces.chaitweet.com)|
 *   +--------------------+   +---------------------+   +--------------------+   +----------------------+
 *            |                         |                        |                         |
 *    +------------------+      +------------------+     +------------------+      +------------------+
 *    | AuthService-1    |      | Tweet-1  Tweet-2 |     | ChatService-1    |      |  SpacesService   |
 *    | AuthService-2    |      | Tweet-3  Tweet-4 |     | ChatService-2    |      |                  |
 *    +------------------+      +------------------+     +------------------+      +------------------+
 *
 *
 * > Ab har service ka apna unique entry point (Load Balancer) hai, isliye user ko pata 
 *   hona chahiye ki kis service ko call karna hai.
 * > Ab humare fontend application ko agar tweets load krne hai to wo tweet.chaitweet.com 
 *   k upar GET request karega. Similalry, agar usse authentication call krna hai to wo
 *   auth.chaitweet.com pe call krega, etc.
 * > Isse hum har services ko independently scale kr paenge lekin humari cost bahut
 *   badh jaegi lekin as our application and user grows, it's okay to scale our cost
 *   of the server.
*/
```

```python
/**
 * Problem: Different Domains
 * > Ab humare paas 4 different domains ho gye, which is not good, because agar humare
 *   paas 30 microservices hai to kya hm 30 different domains banaenge? 
 *   We can do it: 
 *   - auth.chaitweet.com
 *   - tweet.chaitweet.com
 *   - chat.chaitweet.com
 *   - spaces.chaitweet.com
 * > But this is not recommended, so are looking for some smart solution:
 *   Hum kuch aisa kr skte hai ki humare paas ek single domain ho 'chaitweet.com', agar
 *   main 'chaitweet.com/auth/*' likhu to ye request AuthService pe jaani chahiye. 
 *   Similarly, koi v request 'chaitweet.com/tweet/*' se start ho rhi hai to usse
 *   Tweet Service pe le jao.
 * > To kya hm ek single domain pe path based routing kr skte hai? Yes! This is called
 *   API Gateway.
 * > chaitweet.com is pointing to APIGateway, aur hm iss API Gateway k andar rules setup
 *   krte hai ki agar koi chij kisi 'pathname' ko hit karega to wo request uske 
 *   load balancer pe jaega, jisse hum ek single domain pe multiple services k andar 
 *   routing kr skte hai:
 *   a. chaitweet.com/auth/*   - AuthService LoadBalancer
 *   b. chaitweet.com/tweet/*  - TweetService LoadBalancer
 *   c. chaitweet.com/chat/*   - ChatService LoadBalancer
 *   d. chaitweet.com/spaces/* - SpacesService LoadBalancer
 * > Flow: Users > APIGateway > LoadBalancer > Service > Database
 * 
 *                                       +----------------------+
 *                                       |     API Gateway      |
 *                                       |   (chaitweet.com)    |
 *                                       +----------------------+
 *                                                   |
 *            ------------------------------------------------------------------------------
 *            |                         |                        |                         |
 *   +--------------------+   +---------------------+   +--------------------+   +----------------------+
 *   |   Load Balancer    |   |  Load Balancer      |   |  Load Balancer     |   |  Load Balancer       |
 *   |      (auth)        |   |     (tweet)         |   |     (chat)         |   |     (spaces)         |
 *   +--------------------+   +---------------------+   +--------------------+   +----------------------+
 *            |                         |                        |                         |
 *    +------------------+      +------------------+     +------------------+      +------------------+
 *    | AuthService-1    |      | Tweet-1  Tweet-2 |     | ChatService-1    |      |  SpacesService   |
 *    | AuthService-2    |      | Tweet-3  Tweet-4 |     | ChatService-2    |      |                  |
 *    +------------------+      +------------------+     +------------------+      +------------------+
 * 
 * Microservices + API Gateway (Benefits):
 * > Ab agar AuthService ke andar bug aa jata hai to sirf AuthService down hoga,
 *   matlab new users login/signup nahi kar paenge, lekin jo users already logged-in
 *   hain wo tweet, chat aur spaces jaise features use kar sakte hain.
 *
 * > Isi tarah agar kisi ne AuthService par DDoS attack kiya, to sirf wahi service
 *   affect hogi aur baaki services (Tweet, Chat, Spaces) normal chalti rahengi,
 *   kyunki sab services independent hain.
 *
 * > Ab jab humare paas API Gateway hai, to hum uspe Rate Limiting bhi laga sakte hain,
 *   yani kisi user ki requests ko limit kar sakte hain aur ye check service tak
 *   pahunchne se pehle hi ho jata hai, jisse unnecessary load andar nahi jata aur
 *   DDoS attack ko kaafi had tak control kiya ja sakta hai.
 */
```

```python
/**
 * Microservices + Different Databases:
 * > Ab jab humari saari services alag alag ho gayi hain, to hum databases bhi
 *   alag choose kar sakte hain apni need ke according:
 *   - AuthService  → Postgres
 *   - TweetService → MongoDB
 *   - ChatService  → ClickHouse
 *   Isse har service apne use-case ke hisaab se best database choose karti hai
 *   aur decision making fast ho jati hai.
 *
 * > Problem:
 *   User ka data AuthService me stored hai, lekin TweetService aur ChatService ko
 *   bhi user ki info chahiye hoti hai (UserId, profile, etc.). Yaha issue ye hai ki
 *   DB level JOIN possible nahi hai kyunki sab alag alag databases use kar rahe hain.
 *
 * > Solution: Application Level JOIN
 *   Is problem ka solution hai Application Level JOIN. Matlab JOIN database ke andar
 *   nahi, balki application (code) ke andar hota hai. Jaise TweetService ke paas
 *   tweet data hai jisme userId hai, to wo AuthService ko HTTP call karega aur bolega
 *   "mujhe is userId ka data de do". AuthService user ka data return karega, aur
 *   TweetService dono data ko combine karke final response bana dega.
 * > Example:
 *   - User request karta hai: "Get Tweet"
 *   - API Gateway request ko TweetService tak bhejta hai
 *   - TweetService apne DB (MongoDB) se tweet nikalta hai:
 *       { tweet: "Hello World", userId: 7 }
 *   - Ab TweetService AuthService ko call karta hai:
 *       "userId = 7 ka data bhejo"
 *   - AuthService apne DB (Postgres) se data nikalta hai:
 *       { userId: 7, name: "Mohammad" }
 *   - TweetService dono ko combine karta hai:
 *       { tweet: "Hello World", user: "Mohammad" }
 *   - Final response user ko bhej diya jata hai
 *
 * > Isi tarah ChatService bhi jab message bhejta hai ya dikhata hai, to wo AuthService
 *   se user info le sakta hai. Is process me ek extra network call (extra hop) hota hai,
 *   lekin ye microservices me normal hai.
 *
 * > Inter-Service Communication:
 *   Services aapas me communicate kar sakti hain:
 *   - HTTP / REST calls
 *   - Future me async events, queues, etc.
 *
 * > Conclusion:
 *   - DB JOIN          → possible nahi (different databases)
 *   - Application JOIN → possible (service to service calls)
 *   - Har service independent hai lekin zarurat padne par dusri service ko call karti hai
 */
```

```python
/**
 * OIDC + OAuth2.0:
 * > Application Level JOIN se humara data sharing problem solve ho gaya tha,
 *   lekin ek naya problem aaya: "User kaun hai?" (Authentication & Identity).
 * > Jab user request bhejta hai (e.g. "Post Tweet"), to TweetService ko kaise pata
 *   chalega ki ye request kis user ne bheji hai? Kya har baar AuthService ko call kare?
 *   Ye slow bhi hoga aur system par unnecessary load bhi badhega.
 *
 * > Maan lo Twitter ek naya feature launch karta hai (Groq). Ab humein naya
 *   codebase banana hai, lekin existing codebase ko touch karne ki zarurat nahi hai.
 *   Hum ek nayi service bana sakte hain.
 * > Ab sawal: kya humein authentication ka code dobara likhna padega? → No!
 *   Kyunki authentication already ek separate service hai, aur wo reusable hai.
 *   Jo user Twitter me logged-in hai, wahi user Groq me bhi logged-in maana jayega.
 * > Isi concept ko hum real world me dekhte hain:
 *   - "Sign in with Google"
 *   - YouTube, Gmail, Drive, Photos sab me same login use hota hai
 *
 * > Google ne ek centralized Auth Service bana rakhi hai:
 *
 *                     +-------------+
 *                     | Auth-Google |
 *                     +-------------+
 *                            |
 *            +----------+---------+---------+
 *            |          |         |         |
 *            V          V         V         V
 *       +---------+ +--------+ +------+ +--------+
 *       | YouTube | | Gmail  | | Maps | | Photos |
 *       +---------+ +--------+ +------+ +--------+
 *
 * > Ye saari services ek hi Auth Service use karti hain. Internally ye services
 *   authentication system se baat karne ke liye ek protocol follow karti hain
 *   jise OIDC (OpenID Connect) Protocol kehte hain.
 *
 *
 * OIDC (OpenID Connect):
 * > OIDC Protocl define karta hai ki microservices authentication system se kaise baat
 *   karengi aur user ki identity ka data kaise exchange hoga.
 * > Matlab:
 *   - Agar YouTube ko user ka data chahiye → OIDC use karega
 *   - Gmail/Drive/Photos                   → sab OIDC use karenge
 *
 *
 * Problem with External Apps:
 * > Ab maan lo tumne ek external app banaya "Teachyst" aur tum chahte ho:
 *   "Sign in with Google"
 * > Kya tum directly Google Auth Service ko call kar sakte ho? → No!
 *   Kyunki Google kisi unknown app ko direct access nahi deta (security reasons).
 *
 *
 * OAuth 2.0 (Trust Layer):
 * > Yaha OAuth2.0 ka role aata hai. Ye trust establish karta hai between:
 *   - External App (Teachyst)
 *   - Google Auth Service
 * > Developer OAuth setup karta hai jisme:
 *   - App ka naam
 *   - Kya data access karna hai (profile, email, etc.)
 *   - Permissions / consent
 * > Jab trust establish ho jata hai:
 *   - External app ab Google Auth se securely baat kar sakta hai
 *   - Aur user ki identity access kar sakta hai
 * 
 * 
 * Core Concept:
 * > OAuth2.0 ek authorization framework hai jiske upar OIDC build hota hai,
 *   jisse authentication + identity layer milti hai. Is flow me AuthService
 *   ek Identity Provider (IdP) ki tarah act karta hai.
 * > Identity Provider ka kaam hota hai:
 *   - User ko authenticate karna (verify credentials)
 *   - Ek secure token issue karna jo user ki identity represent kare
 *
 * Token Based Identity:
 * > Authentication ke baad AuthService ek token (generally JWT - JSON Web Token)
 *   generate karta hai. Ye token self-contained hota hai jisme:
 *   - userId
 *   - roles/permissions
 *   - expiry time
 *   jaise claims hote hain.
 * > Ye token digitally signed hota hai, isliye ise verify kiya ja sakta hai bina
 *   AuthService ko baar baar call kiye.
 *
 *
 * Request Lifecycle with OIDC:
 *  - User login karta hai → AuthService token deta hai
 *  - User jab bhi Tweet/Chat/Spaces use karega → request ke saath token bhejega
 *  - API Gateway ya service token verify karegi
 *  - Agar token valid hai → request allow
 *  - Agar invalid hai → reject
 *
 *          (Login)
 *   User -------------> AuthService
 *                         |
 *                         | (Verify + Generate Token)
 *                         v
 *                    [ JWT Token ]
 *                         |
 *                         v
 *   User ----(Token)---> API Gateway ----> Services
 *                                   |--> TweetService
 *                                   |--> ChatService
 *                                   |--> SpacesService
 *
 * > Important:
 *   - AuthService sirf authentication karta hai
 *   - Baaki services sirf token verify karti hain
 *   - Har baar DB check karne ki zarurat nahi hoti
 *
 * FINAL IDEA:
 * > OAuth2.0 → Trust establish karta hai + secure access enable karta hai
 * > OIDC     → OAuth2.0 ke upar identity layer provide karta hai
 */
```

```python
/**
 * Problem: Synchronous vs Asynchronous Tasks
 * > Monolith Architecture aur Microservices Architecture ke saath ek important concept 
 *   aata hai:
 *   - Synchronous Task  → ek kaam ke baad hi next kaam hoga (blocking flow)
 *   - Asynchronous Task → kaam background/parallel me ho sakta hai (non-blocking)
 *
 * > Monolith me agar hum simple Signup flow likhte hain:
 *   - user data DB me store karo
 *   - email send karo ----------------+
 *   - token/redirect profile          |
 *                                     V
 * > Yaha email wala part external services pe depend karta hai:
 *   - Gmail APIs
 *   - AWS APIs
 *   - SendGrid APIs
 *
 * > Problem kya hai?
 *   Humara system scalable hai, to agar 3M requests aati hain:
 *   - 3M users DB me store ho jayenge
 *   - 3M email requests external service ko bhej di jayengi
 * > Lekin agar us time:
 *   - External service down ho
 *   - Ya rate limit hit ho jaye (429 Too Many Requests)
 * > To:
 *   - Email step fail hoga
 *   - Kyunki flow synchronous hai → next step (redirect) bhi fail hoga
 *   - Poora request chain fail hone lagega
 * > Result:
 *   - System overload
 *   - Server crash
 *
 * > Important:
 *   - Server me koi bug nahi tha
 *   - Problem external dependency ki wajah se aayi
 *   - Synchronous flow ne poore system ko fail kar diya
 * > Even agar humne:
 *   - ASG
 *   - Load Balancer
 *   - API Gateway
 *   use kiya ho, fir bhi system crash ho sakta hai
 *   kyunki code blocking/synchronous hai aur external services pe dependent hai
 *
 *
 * Solution: Queue (Asynchronous Processing in Microservices)
 * > Hum Email ke liye ek alag EmailService bana dete hain jo independently emails
 *   send karegi aur external services (Gmail / AWS SES, etc.) se connect karegi.
 * > Ab direct HTTP call karna sahi nahi hai kyunki wo fir se synchronous ban jayega.
 *   Isliye hum QUEUE use karte hain (e.g., AWS SQS).
 * > Jab bhi koi event hota hai (signup, tweet, spaces):
 *   - Hum direct EmailService ko call nahi karte
 *   - Balki event ko queue me daal dete hain
 * > Ab millions of requests aa rahi hain from multiple services like Auth, Tweet, Spaces,
 *   aur sab queue me store ho rahi hain:
 *   - EmailQueue   (Producer) → events queue me daal raha hai
 *   - EmailService (Consumer) → apni speed se process kar raha hai
 * > Example:
 *   - EmailService 10 emails/sec process kar sakti hai
 *   - To wo har second limited messages uthayegi
 * > Isse fayda:
 *   - Rate limiting avoid hoti hai
 *   - External services overload nahi hoti
 *   - System stable rehta hai
 * > Agar EmailService down ho jaye:
 *   - Queue me messages safe rahenge
 *   - Service jab wapas aayegi → processing resume ho jayegi
 *

 *                           +----------------------+
 *                           |     API Gateway      |
 *                           +----------------------+
 *                                      |
 *                           +----------------------+
 *                           |    Load Balancer     |
 *                           +----------------------+
 *                                      |
 *        +----------------+------------------+-------------------+
 *        |                |                  |                   |
 *        V                V                  V                   V
 *
 *   +--------------+  +--------------+  +--------------+  +--------------+
 *   |Signup Service|  |Tweet Service |  | Chat Service |  |Spaces Service|
 *   +--------------+  +--------------+  +--------------+  +--------------+
 *        |                   |                 |                 |
 *        |                   |                 |                 |
 *        +---------- Push Event to Email Queue ------------------+
 *                                      |
 *                                      V
 *                           +----------------------+
 *                           |     Email Queue      |
 *                           +----------------------+
 *                                      |
 *                             (Consume at own rate)
 *                                      |
 *                                      V
 *                           +----------------------+
 *                           |     Email Service    |
 *                           +----------------------+
 *                                      |
 *                             (External API Call)
 *                                      |
 *                                      V
 *                           +----------------------+
 *                           | Gmail / AWS SES etc. |
 *                           +----------------------+
 *
 *
 * > Flow: Multiple Services → Email Queue → Email Service → External Service
 * > Yaha:
 *   - Sab services directly EmailService ko call nahi karti
 *   - Sab ek hi queue me events daalti hain
 *   - EmailService apni capacity ke hisaab se process karti hai
 */
```

```python
/**
 * Problem: Multiple Queues / Tight Coupling
 * > Ab humne Queue introduce kar diya hai aur EmailService usse consume kar rahi hai.
 *   Ab requirement aati hai ki:
 *   - Email bhejna hai
 *   - SMS bhi bhejna hai
 * 
 * > To hum kya karenge?
 *   Ek aur SMSService bana denge aur ek aur queue (SMSQueue) laga denge jo har services
 *   k incoming request ko SMSQueue & EmailQueue m daal dega:
 *   - DB me user ko store karo
 *   - email.queue.send(user-details)
 *   - sms.queue.send(user-details)
 *   - token bana do
 * 
 * > Problem ye hai ki:
 *   - Har naye feature ke liye nayi queue + nayi service add karni pad rahi hai
 *   - Same event ko multiple queues me daalna pad raha hai
 *   - Code duplication ho raha hai
 * > Example:
 *   - Agar kal Push Notification add karna ho → ek aur queue add karni padegi
 *   - Agar SMS remove karna ho → har jagah se sms.queue.remove karna padega
 * > Result:
 *   - System phir se tightly coupled ho raha hai queues ke saath
 *   - Maintainability issue aa raha hai
 * 
 *
 *                           +----------------------+
 *                           |     API Gateway      |
 *                           +----------------------+
 *                                      |
 *                           +----------------------+
 *                           |    Load Balancer     |
 *                           +----------------------+
 *                                      |
 *        +-------------------------------------------------------+
 *        |                |                  |                   |
 *        V                V                  V                   V
 *
 *   +--------------+  +--------------+  +--------------+  +--------------+
 *   |Signup Service|  |Tweet Service |  | Chat Service |  |Spaces Service|
 *   +--------------+  +--------------+  +--------------+  +--------------+
 *          |                |                  |                 |
 *          +----------------+------------------+-----------------+
 *                   |                                    |
 *          +------------------+                +------------------+
 *          |   Email Queue    |                |    SMS Queue     |
 *          +------------------+                +------------------+
 *                   |                                    |
 *          +------------------+                +------------------+
 *          |   Email Service  |                |    SMS Service   |
 *          +------------------+                +------------------+
 *                  |                                    |
 *                  V                                    V
 *          +------------------+                +------------------+
 *          | Gmail / AWS SES  |                | Twilio / SMS API |
 *          +------------------+                +------------------+
 *
 * 
 * 
 * Solution: Single Queue with Multiple Consumers (Publisher-Subscriber Model)
 * > Isko solve karne ke liye hum ek hi event ko multiple jagah bhejne ke bajaye:
 *   - Ek single queue use karte hain (e.g., UserEventQueue)
 * > Ab flow:
 *   - DB me user ko store karo
 *   - user.signup event queue me daal do
 *   - token bana do
 * > Aur:
 *   - EmailService → same queue se consume karegi
 *   - SMSService   → same queue se consume karegi
 * > Fayda:
 *   - Code duplication nahi hota
 *   - Publisher ko sirf ek jagah event bhejna hota hai
 *   - Nayi service add karni ho → bas usse queue se connect kar do
 * 
 *                           +----------------------+
 *                           |     API Gateway      |
 *                           +----------------------+
 *                                      |
 *                           +----------------------+
 *                           |    Load Balancer     |
 *                           +----------------------+
 *                                      |
 *        +-------------------------------------------------------+
 *        |                |                  |                   |
 *        V                V                  V                   V
 *   +------------+   +------------+   +------------+   +-------------+
 *   | AuthService|   |TweetService|   | ChatService|   |SpacesService|
 *   +------------+   +------------+   +------------+   +-------------+
 *           |               |                 |                 |
 *           |               |                 |                 |
 *           +------------------- Event Raise -------------------+
 *                                      |
 *                                      V
 *                           +----------------------+
 *                           |   Event Bus / Queue  |
 *                           +----------------------+
 *                             /        |        \
 *                            /         |         \
 *                           V          V          V
 *                    +-----------+ +-----------+ +-------------+
 *                    | EmailSvc  | | SMS Svc   | | WebhookSvc  |
 *                    +-----------+ +-----------+ +-------------+
 * 
 * Note: Event-Driven Architecture:
 * > Sabhi services events raise kar rahi hain aur hum un events ke basis par
 *   SMS/Email/Webhook call etc. trigger kar rahe hain, aur uske basis par
 *   services apna kaam kar rahi hain.
 * > Isse Event-Driven Architecture kaha jata hai, jisme hum ek event raise karte hain,
 *   baaki services usse khud pick kar leti hain.
 * > Hum ek central system ke through trigger karte hain aur wo system decide karta hai
 *   ki mujhe ye event kahan kahan bhejna hai.
 * > Isi process ko "Fan-Out" bola jata hai.
 */

```
