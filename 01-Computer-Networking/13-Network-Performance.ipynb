```python
/**
 * Bandwidth as maximum Link Capacity:
 * 1. Bandwidth means maximum capacity of a network link
 * 2. How much data a link can carry per second at maximum.
 * 3. First principle analogy:
 *    - Bigger pipe > more water
 *    - Higher bandwidth link > more data
 *      - 100 Mbps : Smaller pipe carries less water/data 
 *      - 1 Gbps   : Bigger pipe carries more water/data
 * 4. Full forms:
 *    - Mbps = Megabits per second
 *    - Gbps = Gigabits per second
 * 5. why it exists? 
 *    - Shows the upper limit of a connection.
 *    - Capacity is not the same as actual speed.
 * 6. Examples:
 *    - 100 Mbps = internet plan
 *    - 1 Gbps   = data center link
*/
```

```python
/**
 * Throughput as actual data transfer:
 * 1. Throughput means actual data transferred per second.
 * 2. First-Principle Analogy:
 *    - Bandwidth is what the road can support.
 *    - Throughput is how many cars actually pass.
 * 3. Bandwidth = 300 mbps, Throughput = 120 mbps
 * 4. Examples:
 *    - Internet plan   = 300 mbps
 *    - Actual download = 120 mbps
 * 5. TCP = Transmission Control Protocol (Bandwidth + Throughput)
 *    - Capacity(bandwidth)      = 300 cars/minute 
 *    - Actual Flow (throughput) = 120 cars/minute
 * 6. Why throughput is lower?
 *    - Congestion
 *    - Packet loss
 *    - Weak Wi-Fi
 *    - Slow Server
 *    - Protocol overhead
 *    - Other users sharing link
 *    - Bottleneck in path
 *    - TCP slowing down
*/
```

```python
/**
 * Latency and Round Trip Time:
 * 1. Latency means delay
 *    - Time taken for data or a request to travel from source to destination.
 * 2. Round Trip Time
 *    - Time for request to go to server and response to come back
 * 
 *      Client(Browser) <==================> Server
 *      <------Request sent + Response sent-------->
 *      <------------ Round Trip Time ------------->
 * 
 * 3. Example:
 *    a. Browser sends request
 *    b. Server starts responding after 200 milliseconds.
 *       - 0ms: Request leaves the browser
 *       - 100ms: Request reaches the server
 *       - 200ms: Response returns to the browser
 *       - RTT  : 200ms
 *    Note: Opening a webpage needs many request-response steps.
 * 
 * 4. Why latency matters?
 *    - Small requests are affected more by latency than bandwidth.
*/
```
